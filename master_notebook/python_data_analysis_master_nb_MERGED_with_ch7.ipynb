# 🧠 Part 1 — Python Foundations  
## **Chapter 1: Python Data Types**
## ------------------------------------------------------------  

This section introduces the essential data types used in Python and explains how Python classifies, stores, and manages values. Understanding data types is foundational for effective data analysis because it determines how Python interprets variables, executes operations, and handles errors. The lesson also introduces the `type()` function, type conversion, the concept of iterability, and the distinction between mutable and immutable data types.

Python data types fall into key categories:
- **NoneType**  
- **Numeric** (int, float, complex)  
- **Text** (str)  
- **Sequence** (list, tuple, range)  
- **Mapping** (dict)  
- **Set Types** (set, frozenset)  
- **Boolean** (bool)

The goal of this section is to understand what each data type represents, how to identify them, and how to convert between compatible types using built-in functions.

---

## 📘 Python Data Types: Single Value vs Multiple Value

The table below summarizes Python data types based on whether they store a **single value** or **multiple values**. This aligns with the conceptual framework introduced in the lecture notes and your handwritten outline.

| **Single Value** | **Multiple Value** |
|------------------|--------------------|
| **Numeric** <br> • `int` <br> • `float` <br> • `complex` | **Sequence Types** <br> • `list` <br> • `tuple` <br> • `range` |
| **Text** <br> • `str` | **Mapping Type** <br> • `dict` |
| **Boolean** <br> • `bool` (True, False) | **Set Types** <br> • `set` <br> • `frozenset` |
| **NoneType** <br> • `None` (placeholder for no value) | *(No additional categories)* |

---

### Notes:
- **Single-value data types** represent one atomic value, such as a number, text string, or boolean.  
- **Multiple-value data types** store collections of values. These structures allow iteration and support a wide range of data manipulation patterns used in analysis workflows.  
- `NoneType` exists as a placeholder that represents “no value.”
---

# -------------------------------------------------------------
# 🔹 Checking the data type of various Python objects
# -------------------------------------------------------------

empty_value = None
print(type(empty_value))

x_int = 5
x_float = 5.24
x_complex = 5 + 2j
print(type(x_int), type(x_float), type(x_complex))

name = "snowboard"
print(type(name))

numbers_list = [1, 3, 5, 7, 9]
items_tuple = ("snowboard", "skis")
num_range = range(1, 10, 2)
print(type(numbers_list), type(items_tuple), type(num_range))

equipment = {"snowboard": 24, "skis": 17}
print(type(equipment))

flag = True
print(type(flag))

gear_set = {"snowboard", "skis"}
frozen_gear = frozenset({"snowboard", "skis"})
print(type(gear_set), type(frozen_gear))


`````python
# -------------------------------------------------------------
# 🔹 Type Conversion (valid and invalid conversions)
# -------------------------------------------------------------

num_string = "123"
converted_int = int(num_string)
print(converted_int, type(converted_int))

integer_value = 456
int_string = str(integer_value)
print(int_string, type(int_string))

sample_list = [1, 2, 3]
list_tuple = tuple(sample_list)
print(list_tuple, type(list_tuple))

# This will raise a TypeError:
# int(sample_list)
`````

`````python
# -------------------------------------------------------------
# 🔹 Iterable vs Non-Iterable Examples
# -------------------------------------------------------------

# Strings are iterable
for character in "snowboard":
    print(character)

# Lists are iterable
for item in [1, 3, 5]:
    print(item)

# Non-iterable: int
# for i in 5:
#     print(i)    # TypeError if uncommented
`````

`````python
# -------------------------------------------------------------
# 🔹 Mutability Examples
# -------------------------------------------------------------

mutable_list = [1, 2, 3]
mutable_list.append(4)
print(mutable_list)

mutable_set = {"snowboard", "skis"}
mutable_set.add("helmet")
print(mutable_set)

mutable_dict = {"snowboard": 24}
mutable_dict["skis"] = 17
print(mutable_dict)
`````

`````python
# -------------------------------------------------------------
# 🔹 Sets automatically remove duplicates
# -------------------------------------------------------------

numbers = [1, 2, 2, 3, 3, 3]
unique_numbers = set(numbers)
print(unique_numbers)

# Duplicate values inside {} will also be removed
example = {"a", "a", "b"}
print(example)
`````

## 🧩 Syntax Table for Set and Tuple Creation

| Syntax                  | What it Creates        | Why                                                |
| ----------------------- | ---------------------- | -------------------------------------------------- |
| `{"a", "b"}`            | **set**                | curly braces + comma-separated values              |
| `("a", "b")`            | **tuple**              | parentheses + commas                               |
| `"a", "b"`              | **tuple**              | commas alone create a tuple                        |
| `frozenset({"a", "b"})` | **frozen set**         | must call `frozenset()` explicitly                 |
| `{}`                    | **dictionary**         | Python reserves `{}` for empty dict                |
| `set()`                 | **empty set**          | correct way to create an empty set                 |


## 🧩 Conversions to a Set Remove Duplicates

| Conversion      | Removes duplicates? | Explanation                                  |
| --------------- | ------------------- | -------------------------------------------- |
| `set(list)`     | ✔ Yes               | All repeated elements removed                |
| `set(tuple)`    | ✔ Yes               | Adds each unique tuple value once            |
| `set(range)`    | ✔ Yes               | Range contains no duplicates                 |
| `set("string")` | ✔ Yes               | Unique characters only                       |
| `set(dict)`     | ✔ Yes               | Converts dictionary *keys* into a set        |

**Note:** Sets may only contain unique elements.

## 🧩 Mutability Comparison (Mutable vs Immutable)

| **Category**     | **Type**        | **Mutable?** | **Example**                    |
|------------------|------------------|--------------|--------------------------------|
| Numeric          | `int`, `float`, `complex` | No | `5`, `5.24`, `5+2j`            |
| Text             | `str`            | No           | `"snowboard"`                  |

| NoneType         | `NoneType`       | No           | `None`                         |

|                  | `tuple`          | No           | `(1, 2, 3)`                    |
|                  | `range`          | No           | `range(1, 10)`                 |
| Mapping          | `dict`           | Yes          | `{"a": 1}`                     |
| Set Types        | `set`            | Yes          | `{"a", "b"}`                   |
|                  | `frozenset`      | No           | `frozenset({"a", "b"})`        |


## 🔁 Iterable vs Non-Iterable Data Types

| **Type**        | **Iterable?** | **Example**                    | **Notes**                                 |
|-----------------|----------------|--------------------------------|-------------------------------------------|
| `list`          | Yes           | `[1, 2, 3]`                    | Ordered and mutable                       |
| `tuple`         | Yes           | `(1, 2, 3)`                    | Ordered and immutable                     |
| `range`         | Yes           | `range(1, 5)`                  | Generates numbers efficiently             |
| `dict`          | Yes           | `{"a": 1}`                     | Iterates over keys by default             |
| `set`           | Yes           | `{"a", "b"}`                   | Unordered, unique values                  |
| `frozenset`     | Yes           | `frozenset({"a", "b"})`        | Immutable set                             |
| `str`           | Yes           | `"abc"`                        | Character-by-character                    |
| `int`           | No            | `5`                            | Not iterable                              |
| `float`         | No            | `3.14`                         | Not iterable                              |
| `bool`          | No            | `True`                         | Not iterable                              |
| `NoneType`      | No            | `None`                         | Not iterable                              |


## 🧠 **Chapter 1 Summary: Data Types in Python**
## **Key Takeaways:**

- Python includes several core data type categories: **numeric, text, sequence, mapping, set, boolean, and NoneType**.  
- The `type()` function identifies an object's data type and is essential for debugging when operations fail due to incompatible types.  
- Type conversion uses the format `datatype(object)` and succeeds only when the value is compatible (e.g., `"123"` → `int`, but `[1,2,3]` → `int` fails).  
- **Iterables** include strings, lists, tuples, dictionaries, ranges, sets, and frozensets—any object you can loop through.  
- **Strings are immutable yet iterable**, meaning characters cannot be changed in place but can be looped over.  
- **Mutable types** (lists, sets, dictionaries) can be changed after creation.  
  **Immutable types** (numbers, strings, tuples, booleans, frozensets, NoneType) cannot.  
- Dictionaries store **key-value pairs**, forming the backbone of JSON-style structured data widely used in analytics.  
- Sets store **unique values only**, removing duplicates automatically during creation or conversion.  
- Curly braces `{}` create:  
  - a **set** when values are comma-separated  
  - a **dictionary** when using key-value pairs  
  - an empty dictionary when used as `{}`  
- An empty set must be created with **`set()`**, since `{}` defaults to a dictionary.  
- Understanding mutability, iterability, and data type behavior is foundational for working with **Pandas, NumPy, data cleaning, merging, grouping, and transformation workflows**.


# 🧭 Chapter 2. Variables: Overview

Variables are containers that store values for later use in a Python program. They can store any data type, including numbers, strings, lists, dictionaries, and even function calls. Understanding how to assign, overwrite, delete, and track variables is essential for writing clear and maintainable code.

A variable assignment uses the format:

```
variable_name = value
```

Variable assignment allows you to label values, reuse them, and perform operations without hard coding every value. Naming conventions and clear organization help make code easier to follow. This section builds on the concepts from the Maven Analytics notes by demonstrating assignment, overwriting, deleting variables, and tracking variables in a Jupyter environment.

---

# 🧪 Naming Rules and Conventions

Variable names:

- Can contain letters, numbers, and underscores  
- Must begin with a letter or underscore  
- Cannot begin with a number  
- Cannot contain spaces or special characters  
- Cannot use reserved keywords such as `del`, `list`, or `for`

Snake case is the preferred Python naming style:

- `price_list`
- `total_revenue`
- `snowboard_inventory_2024`

---


`````python
# -----------------------------------------------------------
# 🔹 Basic Variable Assignment
# -----------------------------------------------------------

# Assigning simple values to variables
price = 5
city = "Los Angeles"
heights = [180, 173, 191]

print(price)          # prints: 5
print(city)           # prints: "Los Angeles"
print(heights)        # prints: [180, 173, 191]

# Using a variable in an operation
print(price + 1)      # prints: 6 (adds 1 to the value stored in 'price')
# adding 1 is hardcoded (e.g., set fee)
# or can it can be a variable for values that can change
`````

`````python
# -----------------------------------------------------------
# 🔹 Assigning a Function Call to a Variable
# -----------------------------------------------------------

price_list = [10.99, 12.49, 8.75]   
result = print(price_list)          
# 'result' stores the return value of print(), which is None

print(result)         # prints: None
# ⚠️ Note: Understanding which functions return values is essential (see markdown)
`````

`````python
# -----------------------------------------------------------
# 🔹 Overwriting a Variable
# -----------------------------------------------------------

# -------------------------------------------------------------
# 1️⃣ Simple overwrite: Python stores the most recent value
# -------------------------------------------------------------
price = 5
price = 6
price = 7           # last assignment "wins"
print("Simple overwrite:", price)   # 7


# -------------------------------------------------------------
# 2️⃣ Keeping both old and new values (good for testing)
# -------------------------------------------------------------
old_price = 5       # original value
price = 6           # updated value
print("Old vs new:", old_price, price)   # 5 6


# -------------------------------------------------------------
# 3️⃣ Overwrite after testing (cleanup pattern)
# -------------------------------------------------------------
price = 5
new_price = 6

# Imagine we run calculations, tests, or print checks here.
print("Before overwrite:", price, new_price)  # 5 6

# Once we are satisfied with the new value, we overwrite:
price = new_price
print("After overwrite:", price)             # 6

# Optional: update again
new_price = 7
price = new_price
print("Final price:", price)                 # 7
`````

`````python
# -----------------------------------------------------------
# 🔹 Assigning One Variable to Another
# -----------------------------------------------------------

price = 5
new_price = 6

# Assign the value of new_price to price
price = new_price

print(price)          # prints: 6
print(new_price)      # prints: 6

# Changing new_price again does not affect price
new_price = 7
print(price)          # still 6
print(new_price)      # now 7
`````

`````python
# -----------------------------------------------------------
# 🔹 Deleting a Variable with del
# -----------------------------------------------------------

value = 25
print(value)          # prints: 25

del value             # removes 'value' from memory

# The following line will raise a NameError
# print(value)
`````

`````python
# Cleanest version: show ONLY variables created by the user (not Jupyter)
user_vars = [
    var for var, val in globals().items()
    if not var.startswith("_")                    # exclude private + system
    and var not in ('In', 'Out', 'get_ipython')   # exclude Jupyter internals
    and not callable(val)                         # exclude functions
    and not isinstance(val, type(__builtins__))   # exclude modules
]
user_vars
`````

### 📝 Note: Understanding Function Calls Assigned to Variables

In this example, a call to the `print()` function is assigned to the variable `result`. The purpose of this demonstration is to show how Python handles **function return values** during variable assignment.

When Python executes:

```python
result = print(price_list)
```

two separate events occur:

1. **`print(price_list)` runs**, which displays the list in the notebook.  
2. **The return value of `print()` is stored in `result`**.  
   Since `print()` does not return meaningful data, its return value is always `None`.

As a result, printing `result` will display:

```
None
```

This example highlights an important concept:

- Some functions **display information** but **return nothing** (for example, `print()`).
- Other functions **return actual values** that can be stored, reused, or manipulated.  
  Examples include:
  - `len()` → returns the length of a sequence  
  - `sum()` → returns the total of numeric values  
  - `max()` → returns the highest value  
  - `sorted()` → returns a new sorted list  

⚠️ Note: Understanding which functions return values and which only perform actions is essential for writing clear and predictable Python code.


## 🔁 Overwriting and Updating Variables

In Python, a variable always stores the **most recent value** assigned to it.  
When you reassign a variable, the old value is replaced in memory for that name.


### 🧠 Overwriting Variables: Key Takeaways

- A variable name always refers to the **most recent value** assigned to it.  
- Reassigning a variable (for example, `price = 7`) overwrites the previous value.  
- Use an extra variable such as `old_price` or `new_price` when:
  - you want to compare before/after values, or  
  - you are testing a change and do not want to lose the original value.  
- After you are confident the new value is correct, you can overwrite the original
  variable and remove temporary variables to keep the code clean.  
- This pattern is common in data analysis when updating thresholds, prices, or
  configuration values during experimentation.


## 🧾 Tracking Variables in Jupyter with `%who` and `%whos`

IPython (and Jupyter) provide special **magic commands** that help you see which
variables are currently defined in your notebook:

- `%who`  → lists variable **names**  
- `%whos` → shows names **plus** types and extra info  

These commands only work in IPython environments such as **Jupyter, VS Code
Notebooks, and Colab**.

These commands help you check which objects exist and track their data types.

## ✅ When to Use These Tools

Use `%%who` or `%%whos` whenever you need to:

- **Check which variables currently exist** in your notebook  
- **Confirm data types** before passing values into functions  
- **Clean up your namespace** during long exploratory sessions  
- **Debug data transformations**, especially when values change unexpectedly  
- **Prepare clean, reproducible code** before committing to GitHub  
- **Avoid accidental overwriting** by ensuring variable names are unique and intentional  

These commands help you stay organized, prevent logical errors, and maintain a tidy working environment — all essential skills for professional data analysis.


## 🧠 **Chapter 2 Summary: Variables**
## **Key Takeaways**

- Variables store values in memory and allow you to reference them throughout your program.
- Assignment uses the structure `variable_name = value`, with spaces on both sides of the equal sign.
- Overwriting a variable replaces its previous value; that previous value is removed from memory unless stored elsewhere.
- Use intuitive, descriptive variable names following snake case to improve readability.
- Avoid overwriting important values during testing by using names such as `old_price` and `new_price`.
- The `del` keyword removes variables from memory, but in most cases reassigning the variable is sufficient.
- Use `%who` and `%whos` in Jupyter notebooks to track your active variables.

## **Chapter 3: Numeric Data**
## ------------------------------------------------------------  

This chapter introduces Python’s numeric data types and how to work with them in data analysis. You will learn how Python represents numbers, how to convert between types, how arithmetic operators behave, and which built-in numeric functions analysts use most frequently.

Python includes **three numeric types**:

| Numeric Type | Description | Examples |
|--------------|-------------|----------|
| **int** | Whole numbers without decimals | `3`, `42`, `-14` |
| **float** | Real numbers with decimals | `3.14`, `0.088`, `-1.20134` |
| **complex** | Numbers with real + imaginary components | `3 + 2j`, `-2j` |

For data analysis, **int** and **float** are the primary types. Complex numbers appear rarely.

---

## 🔢 Numeric Type Conversion

Python allows converting between numeric types using the standard pattern:

```
datatype(object)
```

Examples include:

- `int(3.8)` → rounds down to `3`
- `float(5)` → converts integer to `5.0`
- `float("12.5")` → succeeds because the string contains a valid number
- `int("abc")` → raises `ValueError` (letters cannot be converted)

Numeric conversion is essential when importing data from CSV files, user input, APIs, and JSON, where numbers often arrive as strings.

---

## ➗ Arithmetic Operators

Python supports all common arithmetic operators:

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division | `5 / 3` | `1.6666666667` |
| `//` | Floor division | `5 // 3` | `1` |
| `%` | Modulo (remainder) | `5 % 3` | `2` |
| `**` | Exponentiation | `5 ** 3` | `125` |

---

## 🧮 Order of Operations (PEMDAS)

Python follows the standard execution priority:

1. **P**arentheses  
2. **E**xponentiation  
3. **M**ultiplication & **D**ivision (including `//` and `%`)  
4. **A**ddition & **S**ubtraction  

Parentheses allow you to control evaluation, especially when expressions become complex.

---

## 🔧 Numeric Functions

Python provides numeric-specific helper functions:

| Function | Purpose | Example | Result |
|----------|----------|---------|--------|
| `round(n, digits)` | Round to a given number of decimal places | `round(3.14159, 2)` | `3.14` |
| `abs(n)` | Absolute value | `abs(-8)` | `8` |
| `sum(iterable)` | Add all numeric elements | `sum([2, 4, 6])` | `12` |
| `min(iterable)` | Smallest value | `min([2, 4, 6])` | `2` |
| `max(iterable)` | Largest value | `max([2, 4, 6])` | `6` |

These functions simplify numeric analysis without requiring manual loops.

---

## 📘 What You Will Learn in the Code Section

The coding portion of this chapter will show:

- How to declare and inspect numeric values  
- How type conversion works (valid and invalid cases)  
- How arithmetic operators behave  
- How Python evaluates expressions with PEMDAS  
- How to apply numeric helper functions to real data  

Code examples will follow the Python Foundations format with detailed inline comments.


`````python
## 🧪 Python Code Examples: Numeric Data

# -------------------------------------------------------------
# 🔹 Declaring numeric variables
# -------------------------------------------------------------

a_int = 10               # integer
b_float = 3.75           # floating-point number
c_complex = 2 + 5j       # complex number (real + imaginary)

print(type(a_int))       # <class 'int'>
print(type(b_float))     # <class 'float'>
print(type(c_complex))   # <class 'complex'>
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Type Conversion: float → integer
# -------------------------------------------------------------

x = int(7.89)             
print(x, type(x))         # truncates decimal, rounds down to 7, <class 'int'>
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Type Conversion: integer → float
# -------------------------------------------------------------

# Convert integer → float
y = float(12)             
print(y, type(y))         # 12.0, <class 'float'>
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Type Conversion: string → float (valid)
# -------------------------------------------------------------

# Convert string → float (valid)
price_str = "19.99"     # <class, 'str'>

price_float = float(price_str)      # string → float, as long as they contain only numeric charaters
print(price_float, type(price_float))
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Type Conversion: string → int (invalid)
# -------------------------------------------------------------

# Invalid conversion (string with letters)
number = "abc123"      # <class, 'str'>
# fixed_number = int(number)       # ❌ ValueError: could not convert string to float
# string → integer, only when they contain only numeric charaters
`````

`````python
# -------------------------------------------------------------
# 🔹 Arithmetic Operators
# -------------------------------------------------------------

x = 10
y = 3

print("Addition:", x + y)           # 13
print("Subtraction:", x - y)        # 7
print("Multiplication:", x * y)     # 30
print("Division:", x / y)           # 3.3333...
print("Floor division:", x // y)    # 3, rounds down
print("Modulo:", x % y)             # 1, returns remainder of a division (not equal to 0 is odd)
print("Exponent:", x ** y)          # 1000
`````

`````python
# -------------------------------------------------------------
# 🔹 Order of Operations (PEMDAS)
# -------------------------------------------------------------

result1 = 10 + 3 * 2         # multiplication first → 10 + 6 = 16
result2 = (10 + 3) * 2       # parentheses override → 13 * 2 = 26
result3 = 10 ** 2 * 3        # exponent first → 100 * 3 = 300

print(result1, result2, result3)
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Data Types: Arithmetic demo
# -------------------------------------------------------------

my_int = 20
my_float = 8.33

print(my_int / my_float)          # action between int and float returns float by default
print(int(my_int / my_float))     # wrap in integer function to return int

result = my_int + my_float
print(result)                     # 28.33
print(round(result, 2))           # 28.33

golf_scores = [-3, 0, 2, 30]
print(min(golf_scores))
print(max(golf_scores))
print(sum(golf_scores))
sum(golf_scores, 3)               # sum of golf_scores plus 3
`````

`````python
# 🧮 Financial Calculations Practice (Arithmetic Operations)

# -----------------------------------------------------------
# 1️⃣ Calculate the gross profit on a snowboard
#    Formula: profit = price - cost
# -----------------------------------------------------------
snowboard_price = 499.99
snowboard_cost = 199.99

gross_profit = snowboard_price - snowboard_cost      # subtraction
print("Gross profit:", gross_profit)


# -----------------------------------------------------------
# 2️⃣ Calculate the gross margin on the snowboard
#    Formula: margin = gross_profit / price
# -----------------------------------------------------------
gross_margin = gross_profit / snowboard_price         # division
print("Gross margin (rounded):", round(gross_margin, 2))


# -----------------------------------------------------------
# 3️⃣ Calculate the price needed for a 70% gross margin
#    Formula: price_needed = cost / (1 - desired_margin)
# -----------------------------------------------------------
desired_margin = 0.70

price_needed_for70 = snowboard_cost / (1.0 - desired_margin)
print("Price needed for 70% margin:", round(price_needed_for70, 2))


# -----------------------------------------------------------
# 4️⃣ Calculate sales tax for the snowboard
#    Tax rate: 8%
#    Formula: tax = price * tax_rate
# -----------------------------------------------------------
tax_rate = 0.08

sales_tax = snowboard_price * tax_rate                # multiplication
print("Sales tax (rounded):", round(sales_tax, 2))


# -----------------------------------------------------------
# 5️⃣ Calculate returns on profits invested for one year
#    Assume sale of 5 snowboards and 5% interest rate
#    Formulas:
#      amount_invested = gross_profit * quantity
#      return_after_1yr = amount_invested + (amount_invested * interest_rate)
# -----------------------------------------------------------
quantity_sold = 5
interest_rate = 0.05

amount_invested = gross_profit * quantity_sold        # multiplication
return_after_1yr = amount_invested + (amount_invested * interest_rate)

print("Amount invested:", amount_invested)
print("Value after 1 year:", return_after_1yr)
`````

`````python
# ---------------------------------------------------------------
`````
`````python
# Universal String Index Map Helper (Positive & Negative Indices)
# ---------------------------------------------------------------

# Original text:
customer_testimonial = '''I love skiing. It's my favorite hobby.
Some people say, "It's not a hobby, it's life."'''
print(repr(customer_testimonial))  # repr() returns an unambiguous string, often a valid expression

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")  # positive and negative column indices

show_index_map(customer_testimonial)
`````

`````python
# -------------------------------------------------------------
# Creating strings in Python
# -------------------------------------------------------------

# Single quotes
product_single = 'Super Snowboard'
print("Single quotes string:", product_single)

# Double quotes (functionally the same)
product_double = "Super Snowboard"
print("Double quotes string:", product_double)

# Triple quotes: useful for multi-line strings (docstrings, messages)
product_description = """\nThe Super Snowboard is our best-selling model.
It is designed for all-mountain riders and deep powder days."""

# Best practice: use end="\n\n" for adding two lines without extra spaces
print(product_description, end="\n\n")

# docstrings only work inside a function, class, or module — NOT regular variables
# print(product_description.__doc__)  # variable does not store docstrings

# Docstring example
def add(a, b):
    """Add two numbers and return the result.

    Args:
        a (int): The first number.
        b (int): The second number.

    Returns:
        int: The sum of a and b.
    """
    return a + b

print(add.__doc__)   # Safely prints the docstring
`````

`````python
# -------------------------------------------------------------
# Basic slicing examples (positive step only)
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# -------------------------------------------------------------
# 1) Slice from the beginning up to a position (stop not included)
# -------------------------------------------------------------
# Index positions: 0 1 2 3 4 5 6
# Characters:      I _ h o p e _

first_phrase = message[0:6:1]   # indices 0 through 5 (stop not included)
print(first_phrase)

first_phrase = message[0:6]     # step defaults to 1
print(first_phrase)             # Output: 'I hope'


# -------------------------------------------------------------
# 2) Slice from index 2 up to index 6 (stop not included)
# -------------------------------------------------------------
# Index positions: 2 3 4 5
# Characters:      h o p e

middle_slice = message[2:6:1]   # indices 2 through 5 (stop not included)
print(middle_slice)

middle_slice = message[2:6]
print(middle_slice)             # Output: 'hope'


# -------------------------------------------------------------
# 3) Slice from index 6 to the end
# -------------------------------------------------------------
# message[6:] means:
#   start = 6
#   stop = default (len(message))
#   step = 1

after_first_two_words = message[6:None:1]  
# Using None to explicitly slice syntax with start, stop, and step
print(after_first_two_words)

after_first_two_words = message[6:]
print(after_first_two_words)         # Output: ' it snows TONIGHT!'


# -------------------------------------------------------------
# 4) Slice with a step (every other character)
# -------------------------------------------------------------
# message[0:6:2] takes indices: 0, 2, 4

skip_chars = message[0:6:2]
print(skip_chars)               # Output: 'Ihp '
`````

`````python
# -------------------------------------------------------------
# 5) Reverse the entire string
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[::-1] uses the slicing pattern [start : stop : step]
#   - start: omitted → Python begins at the last character (because step = -1)
#   - stop: omitted → Python continues until it moves past the first character
#   - step: -1 → move backward one character at a time
#
# Together, message[::-1] produces a full reverse of the string.

reversed_message = message[::-1]
print(reversed_message)
# This is the most common Pythonic way to reverse a string.


# -------------------------------------------------------------
# 6) Slice from the end using negative indices
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[-8:-1] uses the slicing pattern [start : stop : step]
#   - start at index -8  → the 'T' in 'TONIGHT'
#   - stop at index -1   → but EXCLUDE index -1 (the '!')
#   - step defaults to 1 → moving forward left-to-right

# Negative index mapping for this string:
#    message = "I hope it snows TONIGHT!"
#                      T   O   N   I   G   H   T   !
#            indices: -8  -7  -6  -5  -4  -3  -2  -1

last_word = message[-8:-1]
print(last_word)       # Output: 'TONIGHT' (excludes the '!')


# -------------------------------------------------------------
# 7) Slice the last 8 characters including the final character
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[-8:] uses the slicing pattern [start : stop : step]
#   - start at index -8  → the first 'T' in 'TONIGHT'
#   - stop omitted       → continue to the end of the string
#   - step defaults to 1 → move forward left-to-right

# Negative index mapping for this string:
#    message = "I hope it snows TONIGHT!"
#                      T    O    N    I    G    H    T    !
#            indices: -8  -7   -6   -5   -4   -3   -2   -1

tail_slice = message[-8:]
print(tail_slice)      # Output: 'TONIGHT!'
`````

`````python
# -------------------------------------------------------------
# 6) Slice from the end using negative indices
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[-8:-1] uses the slicing pattern [start : stop : step]
#   - start at index -8  → the 'T' in 'TONIGHT'
#   - stop at index -1   → but EXCLUDE index -1 (the '!')
#   - step defaults to 1 → moving forward left-to-right

# Negative index mapping for this string:
#    message = "I hope it snows TONIGHT!"
#                      T   O   N   I   G   H   T   !
#            indices: -8  -7  -6  -5  -4  -3  -2  -1

last_word = message[-8:-1]
print(last_word)       # Output: 'TONIGHT' (excludes the '!')
`````

`````python
# ---------------------------------------------------------------
`````
`````python
# Universal String Index Map Helper (Positive & Negative Indices)
# ---------------------------------------------------------------

# Original text:
customer_testimonial = '''I love skiing. It's my favorite hobby.
Some people say, "It's not a hobby, it's life."'''
print(repr(customer_testimonial))  # repr() returns an unambiguous string, often a valid expression

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")

show_index_map(customer_testimonial)
`````

`````python
# -------------------------------------------------------------
# Measuring Length with len()
# -------------------------------------------------------------
# len() is a built-in Python function that returns:
#   • the number of characters in a string
#   • the number of items/elements in a list (or other collections)
# -------------------------------------------------------------


# -------------------------------------------------------------
# 1) len() with a STRING
# -------------------------------------------------------------
headline = "Fresh powder on the mountain today!"

# len(string) → counts every character, including:
#   spaces, punctuation, and special characters.
char_count = len(headline)

print("Original headline:", headline)
print("Character count:", char_count)     # total number of characters


# -------------------------------------------------------------
# 2) len() with a LIST
# -------------------------------------------------------------
stages = ["beginner", "enthusiast", "pro"]

# len(list) → counts number of items in the list (3 items here)
element_count = len(stages)

print("\nStages list:", stages)
print("Number of elements in list:", element_count)


# -------------------------------------------------------------
# 3) len() with SPLIT STRINGS (words)
# -------------------------------------------------------------
# headline.split() splits the string into words using spaces
word_list = headline.split()

# Now len() counts number of words in the list
word_count = len(word_list)

print("\nWord list from split():", word_list)
print("Number of words:", word_count)
`````

`````python
# -------------------------------------------------------------
# Len() Exercise: Simple Cryptography
# -------------------------------------------------------------
# Goal:
#   Use the length of different messages as "indexes" into the
#   alphabet string to build a password.
#
# Process:
#   1. Measure the length of each message using len()
#   2. Use that number as an index into the alphabet
#   3. Concatenate the resulting letters into a password
# -------------------------------------------------------------

# A reference alphabet for indexing
alphabet = "abcdefghijklmnopqrstuvwxyz"

# Messages of varying lengths
message1 = "Hello World!"
message2 = ""
message3 = "It snowed a lot today"
message4 = "eeee"
message5 = "I love snow!!"

# -------------------------------------------------------------
# Step-by-step explanation:
#
print(len(message1))  #  → 12
print(alphabet[12])  # → 'm'

print(len(message2))  # → 0
print(alphabet[0])  # → 'a'

print(len(message3))  # → 21
print(alphabet[21])  # → 'v'

print(len(message4))  # → 4
print(alphabet[4])  # → 'e'

print(len(message5))  # → 13
print(alphabet[13])  # → 'n'
#
# Final password: "maven"
# -------------------------------------------------------------

password = (
    alphabet[len(message1)] +
    alphabet[len(message2)] +
    alphabet[len(message3)] +
    alphabet[len(message4)] +
    alphabet[len(message5)]
)

print("Generated password:", password)
`````

`````python
# -------------------------------------------------------------
# String Methods: Demonstrating .strip() to Remove Extra Spaces
# -------------------------------------------------------------
# Methods are functions that apply only to a specific data type or object.
# .strip() removes leading and trailing whitespace automatically - no indexing needed!
# -------------------------------------------------------------

# Example string with intentional leading/trailing spaces
message = "  I hope it snows TONIGHT!  "

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------
# This function helps visualize how .strip() affects the positions
# of characters by printing:
#   • positive indexing (0 → len-1)
#   • negative indexing (-len → -1)
#
# Seeing the spaces in the index map makes it clear
# why .strip() removes ONLY the characters at the edges.
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula for corresponding negative index
        print(f"{i:3} {neg:4} {repr(ch)}")


# Display index map BEFORE stripping spaces
print("Before .strip():")
show_index_map(message)

# -------------------------------------------------------------
# Using .strip() to remove leading and trailing spaces
# -------------------------------------------------------------
# .strip() removes ONLY whitespace at the beginning and end.
# It does NOT remove spaces inside the string.
cleaned = message.strip()

print("\nAfter .strip():")
show_index_map(cleaned)   # show the new indexing after cleaning

# Print both versions to compare clearly
print(f"\nOriginal message: [{message}]")
print(f"Stripped message: [{cleaned}]")
`````

`````python
# -------------------------------------------------------------
`````
`````python
# String Methods: .strip(chars)
# removes only the characters listed in `chars`,
# and only from the **beginning and end**, *not the middle*
# -------------------------------------------------------------

# Example strings
message = "   I love skiing and snow.   "     # has leading + trailing spaces
punctuated = "...powder day!!!..."            # mix of periods and exclamation marks

# Helper: Show positive & negative index map for any string
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n
        print(f"{i:3} {neg:4} {repr(ch)}")
        
print("\n--- Punctuated String ---")
show_index_map(punctuated)

# Remove only periods from both ends
no_periods = punctuated.strip(".")
print("\n--- After .strip('.') ---")
show_index_map(no_periods)

# Remove periods AND exclamation marks from both ends
cleaned_all = punctuated.strip(".!")
print("\n--- After .strip('.!') (remove . and !) ---")
show_index_map(cleaned_all)

print(f"\nOriginal:   [{punctuated}]")
print(f"strip('.'): [{no_periods}]")
print(f"strip('.!'):[{cleaned_all}]")
`````

`````python
# String Methods
# -------------------------------------------------------------
`````
`````python
# .count()
# -------------------------------------------------------------
message = "I hope it snows tonight because snow is coming."

# -------------------------------------------------------------
# 1. Count single characters
# -------------------------------------------------------------
print(message.count("o"))     # count how many 'o' characters
print(message.count("i"))     # count of 'i' characters

# -------------------------------------------------------------
# 2. Count whole substrings
# -------------------------------------------------------------
print(message.count("snow"))   # number of times "snow" appears

# -------------------------------------------------------------
# 3. Count in a specific range of indexes: .count(sub, start, end)
# -------------------------------------------------------------
# Count only in the first 20 characters
print(message.count("o", 0, 20))

# Count "snow" only in the second half of the sentence
midpoint = len(message) // 2
print(message.count("snow", midpoint, len(message)))
`````

`````python
# String Methods
# -------------------------------------------------------------
# Method Chaining — perform multiple text operations in sequence
# -------------------------------------------------------------

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------

message = "  I HOPE it snows TONIGHT!  "
cleaned_message = message.strip()

def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")

show_index_map(cleaned_message)

# GOAL:
#   Extract everything AFTER "I HOPE "
#   Make the result lowercase
#   Convert to a list of words
#
# Breakdown of what happens:
#   message.strip()      → removed outer space first for cleaned_message
#   [7::]                → skipping [0:6] and slicing on [7:]
#   .lower()             → convert the remaining text to lowercase
#   .split()             → break into list of words

# processed = cleaned_message[7:].lower().split()
processed = (
    cleaned_message
    [7::]         # slice off "I HOPE "
    .lower()      # convert to lowercase
    .split()      # convert to list of words
)

print(processed)                     # ['it', 'snows', 'tonight!']
print("Word count:", len(processed)) # demonstrates chaining + len()
`````

### Understanding Why `__doc__` Works for Functions but *not* Variables

Python allows you to attach documentation (called a **docstring**) to functions, classes, and modules by placing a triple-quoted string immediately after the definition. These docstrings are accessible using the `.__doc__` attribute.

However, a **variable does not store docstrings**. If you call `.__doc__` on a string variable, Python returns the documentation for the **`str` class itself**, not the variable’s content.

This section compares how `__doc__` behaves for:
- a triple-quoted *multi-line string variable*, and  
- a properly documented function.

### 📌 Key Takeaways

- Triple-quoted strings assigned to variables are **just strings**, not docstrings.  
- Only **functions, classes, and modules** store real docstrings accessible via `.__doc__`.  
- Calling `.__doc__` on a string gives you the documentation for Python’s built-in `str` class.  
- To see a variable’s contents, print the variable directly.  
- To see a function’s docstring, print `function.__doc__`.

## 🔍 Slicing Strings in Python

Python slicing uses the pattern:

```python
string[start : stop : step]
```

### ✔️ What each argument really means

- **start** — index to begin the slice  
  - default depends on step direction  
  - for positive step, default = `0`  
- **stop** — index to stop *before*  
  - default depends on step direction  
  - for positive step, default = `len(string)`  
- **step** — how much to move between characters  
  - default = `1`  
  - must be **non-zero**

---

### 🔹 Important correction
The defaults for `start` and `stop` are **NOT always 0 and len(string)**.  
Those defaults apply **only when the step is positive**.

We will use **positive-step slicing only** in this section,  
so the defaults here behave normally.

---


---

## 🔄 Reversing Strings and Using Negative Indices

Python allows slicing with **negative indices** and **negative step values**, which gives you powerful ways to read a string from right to left.

### ✔️ Important Rules

- Negative indices count from the right  
  ```
  message = "I hope it snows TONIGHT!"
  Index:    -23 … -1
  ```
- A **negative step (-1)** means slicing moves backwards.
- When step is negative:
  - default `start` becomes the **last index**
  - default `stop` becomes **before the first character**
- Slicing still stops **before** the `stop` index — even for negative steps.

## ✂️ Cleaning Text with `.strip()` — Visualizing Leading & Trailing Whitespace

In Python, the `.strip()` method removes **only** the whitespace at the *beginning* and *end* of a string.  
It does **not** remove spaces inside the text.  

The helper function below prints a full index map (positive and negative indices) so you can *see*  
how `.strip()` changes the string's length and character positions.

```python

# -------------------------------------------------------------

    Useful for understanding slicing and how .strip() affects length.
    """

    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n   # convert positive → negative index

# -------------------------------------------------------------
```

- The index map helper makes whitespace behavior visible and intuitive.

## ✂️ Advanced Stripping Methods: `.lstrip()`, `.rstrip()`, `.strip(chars)`

Python provides three related cleaning methods for removing characters from the *edges* of a string:

| Method | Removes | Example |
|--------|----------|----------|
| `.strip()` | Both left + right whitespace (or characters) | `s.strip()` |
| `.lstrip()` | Left-side only | `s.lstrip()` |
| `.rstrip()` | Right-side only | `s.rstrip()` |
| `.strip(chars)` | Removes *only the characters listed* in `chars` from both ends | `s.strip(".!?")` |

Each version affects indexing differently because the *length of the string changes*.  
The helper below visualizes those changes.


## 3️⃣ `.strip(chars)` – Remove Specific Characters From Both Ends

`.strip(chars)` removes **only** the characters listed in `chars`,

## 🔍 Searching Within Strings Using `.find()`

The `.find()` method locates the **first occurrence** of a substring inside a string.  
It returns:

- **the index of the first match**, or  
- **–1** if the substring is not found.

Understanding `.find()` becomes much easier when combined with the index-map helper.

---

## Example String

```python
message = "I hope it snows tonight!"
```

---

## 🧭 Helper: Universal Index Map (Positive & Negative Indices)

```python
def show_index_map(s):
    """
    Display positive and negative indices for each character in s.
    Useful for understanding how .find() aligns with indexing.
    """
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n   # convert positive index → equivalent negative index
        print(f"{i:3} {neg:4} {repr(ch)}")
```

---

## 1️⃣ Find a Substring That Exists

```python
print("Index map of message:")
show_index_map(message)

first_snow_index = message.find("snow")
print("\nIndex of 'snow':", first_snow_index)
```

This tells you **where the substring begins**, so you can slice it:

```python
snow_word = message[first_snow_index : first_snow_index + len("snow")]
print("Extracted substring:", snow_word)
```

---

## 2️⃣ Find a Substring That Does *Not* Exist

```python
missing_index = message.find("rain")
print("\nIndex of 'rain':", missing_index)   # returns -1 when not found
```

Because `.find()` returns `-1`, it prevents indexing errors:

```python
if missing_index == -1:
    print("Substring 'rain' does not appear in the message.")
```

---

## 🔑 Key Takeaways

- `.find()` returns the starting index of the **first** match.
- If the substring is not found, it returns **–1**, not an error.
- Always check for `-1` before slicing based on `.find()`.
- Visualizing the string with the index map makes `.find()` easier to use in slicing and extraction tasks.
- `.find()` is frequently used in:
  - text parsing  
  - cleaning messy CSV fields  
  - locating delimiters  
  - substring extraction  
  - quick data validation  


---

## 🧮 Key Points — `.count()`

- `text.count("a")` → counts how many times the character `"a"` appears  
- `text.count("snow")` → counts how many times the substring `"snow"` appears  
- `text.count(x, start, end)` → counts only within the slice `[start:end]`  
  - `start` is **inclusive**  
  - `end` is **exclusive** (standard Python slicing rule)  
- `.count()` tells you **how many times**, *not where* the substring occurs  
- `.count()` does **not** count overlapping matches  
  - Example: `"aaa".count("aa")` → `1` (not `2`) because `"aa"` overlaps itself  

---

## 📌 When to Use `.count()`

Use `.count()` when answering questions like:

- “How many times does this character appear?”  
- “How many times is this substring repeated?”  
- “How many occurrences appear inside a specific part of the string?”  

Use `.count()` for **frequency**, not for locating positions.

For positions, prefer:

- `.find()`  
- `.rfind()`  
- `.index()`

This section summarizes the **most important string operations** in Python, grouped into five practical categories:

1. Navigating strings (length, indexing, slicing)  
2. Changing appearance (formatting and cleaning)  
3. Splitting and joining (text ↔ lists)  
4. Searching and checking (finding content)

It also includes a **universal index map helper** that prints both positive and negative indices for any string.  
This helper makes slicing and character access much easier for longer text.

## 1️⃣ Navigating Strings – Length, Indexing, and Slicing

```python
# Basic string for navigation examples
word = "Analytics"

# Length: total number of characters
print("Length of word:", len(word))      # 9

# Positive indexing: 0-based from left to right
print("word[0]:", word[0])              # 'A'
print("word[1]:", word[1])              # 'n'
print("word[8]:", word[8])              # 's' (last character)

# Negative indexing: -1 is last character, -2 is second-to-last, etc.
print("word[-1]:", word[-1])           # 's'
print("word[-2]:", word[-2])           # 'c'

# Slicing: [start:stop] includes start, excludes stop
print("word[0:4]:", word[0:4])         # 'Anal'
print("word[4:9]:", word[4:9])         # 'ytics'

# Slicing with omitted bounds
print("word[:4]:", word[:4])           # 'Anal' (from start up to index 4)
print("word[4:]:", word[4:])           # 'ytics' (from index 4 to end)

# Slicing with a step value: [start:stop:step]
print("word[0:9:2]:", word[0:9:2])     # 'Aayc' (every 2nd character)
print("word[::-1]:", word[::-1])       # 'scitylanA' (reverse string)
```

---

### Universal String Index Map Helper (Positive and Negative Indices)

This helper prints a table that shows **both positive and negative indices** for each character in a string.  
It is useful when you want to design precise slices without counting characters manually.

## 5️⃣ Combining and Formatting Strings – Building Output

These patterns are used constantly for **program output, logs, file names, and messages**.

```python
name = "Dawn"
price = 22.55
product = "Maven snowboard"

# Concatenation with +
sentence = "My " + product + " cost $" + str(price) + "!"
print(sentence)

# f-strings for cleaner formatting
sentence_fstring = f"My {product} cost ${price}!"
print(sentence_fstring)

# Repetition to create simple visual elements
print("*" * 10)       # '**********'
print("-" * 40)       # divider line
```

---

## Strings and Loops – Repeating Actions

Strings often appear inside loops, especially when you need to **repeat actions** or process characters one by one.

```python
subject = "My"
noun = "Maven snowboard"
verb = "cost"
price = 22.55

sentence = subject + " " + noun + " " + verb + " $" + str(price) + "! "
print(sentence, end="\n\n")

# Repeat as one long line using multiplication
print("Repeat 3 times as one long string:")
print(sentence * 3, end="\n\n")

# Repeat line by line using range()
print("Repeat 3 times on separate lines:")
for _ in range(3):   # range(3) → 0, 1, 2, but we do not need the index, so use "_"
    print("Repeat 3 times:", sentence, end="\n\n")
```

---

## 🔑 Key Takeaways

- Python strings are **sequences of characters**, which means indexing, slicing, and looping apply.
- You create strings with quotes or by converting other data types using `str()`.  
- Use **string arithmetic**:
  - `+` to concatenate text  
  - `*` to repeat text  
- Access single characters with **indexing**:
  - `s[0]` is the first character, `s[-1]` is the last.  
- Use **slicing** `s[start:stop:step]` to extract substrings or to reverse text with `s[::-1]`.  
- `len()` returns the number of characters in a string or the number of elements in a list.  
- Learn core **string methods**:
  - `.find()` to locate substrings; returns the index of the first match, or `-1` if not found
  - `.upper()` and `.lower()` to standardize case; convert entire strings to new casing
  - `.strip()` (and `.lstrip()`, `.rstrip()`) removes whitespace (or a specified character) on both ends
  - `.replace(old, new)` substitutes all occurrences of one substring  
  - `.split()` breaks a string into a list
  - `.join()` combines list elements into one string
  - `.count()` returns the number of non-overlapping occurrences of a substring within a string. It can optionally take start and end argument

### 🧠 The Five Categories of String Operations:

  1. Navigating strings (length, indexing, slicing)
  2. Changing appearance (case changes, trimming, replacing)
  3. Splitting and joining (text ↔ lists)
  4. Searching and checking (membership, find, count, startswith, endswith)
  5. Combining and formatting (concatenation, f-strings, repetition)
- The **index map helper** removes the need to count positions by hand and makes slicing more reliable.
- Methods like `.split()` and `.join()` are critical for data cleaning and preparing text fields for analysis.
- Searching methods (`in`, `.find()`, `.count()`, `.startswith()`, `.endswith()`) are often used for simple text filters and validations.
- You can **chain methods** to build compact data-cleaning pipelines, such as `message.strip().lower().split()`.  
- Use **f-strings** (`f"..."`) to combine variables and text, and format numbers (for example, `price:.2f`) for clean, dynamic output.  
- These core operations build the foundation for more advanced text handling in `pandas` using the `.str` accessor.

These patterns are the foundation for working with categorical fields, cleaning raw text, and building readable reporting messages in real analysis projects.

# 🐍 Chapter 5. Conditional Logic: 
## 🧩 Control Flow Essentials (if / elif / else) (Part 2)


## 2. If Statements & Core Control Flow

In Part 1 you built a strong foundation in **Boolean logic**:
comparisons, membership tests, `and` / `or` / `not`, precedence, parentheses, and short-circuiting.

Now we use those Boolean expressions to **control the flow of a program** with:

- `if`
- `elif`
- `else`

Control flow lets Python **choose** which block of code to run based on conditions.

---

### 2.1 What Is Control Flow?

Without conditions, a program runs **top to bottom** in a straight line.

With control flow, a program can:

- make decisions  
- follow different paths  
- skip or execute blocks of code based on Boolean results  

You can think of it as a series of **questions**:

> If this is true, do this.  
> Otherwise, do something else.

---

## 2.2 The `if` Statement

### Basic pattern

```python
# General syntax

if condition:
    # This indented block runs only if condition is True
    do_something()
```

Important points:

- `condition` must evaluate to a Boolean (`True` or `False`)
- A colon `:` ends the `if` line
- The following indented block is the **body** of the `if`
- Indentation in Python is not cosmetic. It defines the block.

---

### Example: Basic `if` with a Boolean expression

```python
# --- Example: check if a customer qualifies for free shipping ---

order_total = 120

# Boolean expression from Part 1: comparison operator
if order_total >= 100:
    print("Free shipping applied!")
    
print("Order processed.")
```

What happens:

- If `order_total >= 100` is True, Python prints `"Free shipping applied!"`
- `"Order processed."` prints every time, because it is **outside** the `if` block

Try changing `order_total` to 80 and re-running.

---

## 2.3 `if` + `else`: Two-Way Decision

An `if` statement can stand alone.  
`else` **cannot** stand alone. It always pairs with an `if`.

Use `else` when you want a **default path** if the condition is False.

```python
# General pattern

if condition:
    # runs when condition is True
    do_this()
else:
    # runs when condition is False
    do_that_instead()
```

---

### Example: Free shipping or standard shipping

```python
# --- Example: two-way decision with if/else ---

order_total = 75

if order_total >= 100:
    print("Free shipping applied.")
else:
    print("Standard shipping rate will be used.")

print("Order complete.")
```

- If `order_total` is 120 → prints `Free shipping applied.`
- If `order_total` is 75 → prints `Standard shipping rate will be used.`

Only **one** of the `if` or `else` blocks runs, never both.

---

### Example: Using a helper to show which branch ran

```python
# --- Helper: show which branch executed ---

def check_shipping(order_total):
    print(f"\nChecking shipping for order_total = {order_total}")
    
    if order_total >= 100:
        print("Branch: if → Free shipping")
    else:
        print("Branch: else → Standard shipping")

# Try a few values
check_shipping(120)
check_shipping(75)
check_shipping(100)
```

This example makes branching very visible.  
You can change values and see which branch is selected.

---

## 2.4 `if` + `elif` + `else`: Multi-Way Decision

Sometimes there are **more than two** meaningful cases.  
This is where `elif` (short for "else if") is useful.

```python
# General pattern

if condition_1:
    # runs if condition_1 is True
    block_1
elif condition_2:
    # runs if condition_1 is False and condition_2 is True
    block_2
elif condition_3:
    # runs if all previous conditions are False and condition_3 is True
    block_3
else:
    # runs if none of the above conditions are True
    default_block
```

Rules:

- Conditions are checked **top to bottom**
- As soon as one condition is True, Python runs that block and **skips the rest**
- `if` can exist alone  
- `if` + `else` is a two-branch decision  
- `if` + one or more `elif`s + optional `else` creates a **decision chain**

---

### Example: Customer status and discount level

We will build a simple rule:

- If status is `"VIP"` → 20% discount  
- Else if status is `"LOYAL"` → 10% discount  
- Else if status is `"NEW"` → 5% discount  
- Else → no discount

```python
# --- Example: multi-way decision with if/elif/else ---

def discount_for_status(status):
    print(f"\nChecking discount for status = {status!r}")
    
    if status == "VIP":
        print("20% discount applied.")
    elif status == "LOYAL":
        print("10% discount applied.")
    elif status == "NEW":
        print("5% discount applied.")
    else:
        print("No discount for this status.")

# Try a few different statuses
discount_for_status("VIP")
discount_for_status("LOYAL")
discount_for_status("NEW")
discount_for_status("GUEST")
```

Observe:

- Only one branch runs per call  
- The order of conditions matters

---

### Ordering matters: more specific conditions should appear first

Consider this pattern:

```python
# Example of ordering

score = 95

if score >= 60:
    print("Pass")
elif score >= 90:
    print("High Distinction")
```

For `score = 95`, the result is `"Pass"` because:

- Python checks `score >= 60` first → True  
- It runs that block and **does not** check `elif score >= 90`

Better ordering:

```python
score = 95

if score >= 90:
    print("High Distinction")
elif score >= 60:
    print("Pass")
else:
    print("Fail")
```

Principle:

- Put **stricter** or **higher priority** conditions earlier
- Put more general conditions later in the chain

---

## 2.5 Combining Boolean Logic with Control Flow

Everything from Part 1 can be used inside `if`, `elif`, and `else` conditions.

- Comparisons: `price < 100`
- Membership: `"hat" in items`
- Logical operators: `and`, `or`, `not`

### Example: Discount rule with multiple conditions

Rule:

- Apply a discount if:
  - the customer is VIP, or  
  - the order total is at least 100 and the item is in the discount list

```python
# --- Example: combining Boolean logic with control flow ---

def apply_discount(status, order_total, item):
    print(f"\nStatus={status!r}, order_total={order_total}, item={item!r}")
    
    discount_items = ["gloves", "hat"]
    
    # Boolean expression from Part 1:
    # (status == "VIP") or (order_total >= 100 and item in discount_items)
    
    if status == "VIP" or (order_total >= 100 and item in discount_items):
        print("Discount applied.")
    else:
        print("No discount applied.")

apply_discount("VIP", 50, "gloves")          # VIP always qualifies
apply_discount("REGULAR", 120, "gloves")     # Non-VIP but high total and discount item
apply_discount("REGULAR", 80, "jacket")      # Does not qualify
```

This example shows how **Boolean expressions** and **control flow** work together.

---

## 2.6 trace_decision() Helper for Control Flow
### (Advanced Helper: Traces Full If / Elif / Else Structures)

This helper wraps a function that contains your decision logic.  
It prints a clean header and lets the inner function show which branches ran.

```python
# --- Advanced Helper: trace_decision(label, func) ---

def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.
    
    label: A short description of what is being traced
    func:  A zero-argument function containing the decision logic
    """
    print("\n" + "=" * 60)
    print("DECISION TRACE:", label)
    print("=" * 60)
    func()
```

### Example: Shipping logic with rush + order total

```python
# --- Example: using trace_decision with a multi-branch decision ---

def decide_shipping(order_total, rush):
    def inner():                                       # Wrap the trace_decision logic inside an inner function
        print(f"order_total={order_total}, rush={rush}")
        
        if rush:
            print("Branch: if → Express shipping")
        elif order_total >= 100:
            print("Branch: elif → Free standard shipping")
        else:
            print("Branch: else → Paid standard shipping")
    
    trace_decision("Shipping decision", inner)

# --- Run the helper-driven trace ---
decide_shipping(150, rush=True)
decide_shipping(100, rush=False)
decide_shipping(50, rush=False)
```

This helper is excellent for **complex decision trees**.
This pattern encourages you to think about:

- which branch runs  
- which conditions are tested  
- how order affects the outcome
- **Why do we use inner()?**
→ trace_decision() expects a zero-argument function. 
   inner() captures variables (order_total, rush) and runs the logic.
- **Why do we manually pass rush=True/False?**
→ The helper should NEVER define its own inputs.
   We pass the values we want to test, so we can see different branches.

## ✅ Key Takeaways – If, Elif, Else

- `if` runs its block when the condition is **True**.  
- `else` is a fallback block that runs when the `if` condition is **False`.  
- `elif` lets you create multi-branch decisions. Conditions are checked **top to bottom**.  
- Only **one** branch in an `if` / `elif` / `else` chain runs per execution.  
- Order matters. Put **more specific** or higher-priority conditions before more general ones.  
- Boolean logic from Part 1 (`and`, `or`, `not`, comparisons, membership) fits directly inside `if` conditions.  
- Helpers like `check_shipping`, `discount_for_status`, and `trace_decision` can make the flow of decisions easier to understand and debug.

You now have the core tools for **control flow** in Python.  
Next, you will build on this with **nested if statements** and more complex decision structures.


# 🐍 Chapter 5. Conditional Logic: 
## 🧩 Nested If Statements (Part 3)
 
### Expanding control flow with deeper, multi-layered logic  

Nested `if` statements allow you to evaluate **a second condition only if a first condition is already true**.  
They add an extra “layer” of logic and are useful when your business rules depend on two steps of decision-making.

---

# 1. Why Use Nested If Statements?

Nested logic is appropriate when:

- You need to **check Condition B only when Condition A is true**
- Rules follow a **hierarchy** or **decision tree**
- The second decision would be *nonsensical* unless the first is satisfied  
  (e.g., “If item is out of stock, then check whether you want to offer a display model.”)

### 🔍 Visual Model
A nested structure behaves like this:

```
if condition_A:
    if condition_B:
        # Nested True block
    else:
        # Nested False block
else:
    # Outer False block
```

This is like saying:
> “Only if A is true should I bother checking B.”

---

# 2. Simple Example

```python
customer = "Chris"
inventory = 0

if inventory <= 0:                     # Outer If condition
    if customer == "Chris":            # Nested If condition
        print("You can have the display model, sir.")
    else:
        print("Out of stock.")         # Nested Else code
else:
    print("Item available!")
```

### Why this is nested:
- It makes sense to check the **customer’s name** ONLY after confirming that inventory is zero.
- The business rule is **hierarchical**, not parallel.

---

# 3. The Same Logic *Without* Nesting

From page 3 of the PDF, Maven shows a “without nesting” version.  
Here’s how it looks conceptually:

```
if customer == "Chris" and inventory <= 0:
    ...
```

This works… **but loses hierarchy clarity**.

Nested version makes it obvious:
> “We only ask about Chris because inventory is zero.”

Flat version makes it look like the two conditions are equally important.

Which should you use?  
➡ Use **nesting** when your logic flows like:  
**First A → then maybe check B**

➡ Use **flat logic** when your rule is truly:  
**A AND B must both be true with equal weight**

---

# 4. Real-World Example: Return Authentication

Nested ifs shine in real workflows:

```python
is_logged_in = True
is_admin = False

if is_logged_in:
    if is_admin:
        print("Welcome, admin.")
    else:
        print("Welcome, user.")
else:
    print("Please log in.")
```

This mirrors a real application:
- First confirm login  
- THEN decide admin/user behavior

---

# 5. The Helper Tools Still Work Here  
Nested structures can be confusing.  
So we continue using the tools built in Section 2.8 — especially:

- `trace_decision(label, func)`
- `trace_condition(name, value)`
- `trace_branch(label)`

Let’s load them.

```python
# === Helper Toolkit ===================================================

    print("=" * 60)

    print(f"Branch: {label}")
```

# 6. Enriched Nested-If Example (Helper Version)
  
This demonstrates both **nested structure** and **the decision path**.

### 🧪 Scenario  
Business rule:  
If inventory ≤ 0 **and** customer’s name is "Chris":  
→ offer display model  
Else:  
→ out of stock  
Otherwise keep previous inventory logic.

```
inventory = 0
customer_name = "Chris"
```

```python
def check_inventory_nested(inventory, customer_name):
```

`````python
# --- Boolean basics: True & False as values and numbers ---

is_ski_season = True
is_store_closed = False

print("is_ski_season:", is_ski_season)
print("is_store_closed:", is_store_closed)

# Booleans act like 1 and 0 in arithmetic
print("True + 1  →", True + 1)   # 1 + 1 = 2
print("False + 1 →", False + 1)  # 0 + 1 = 1


# --- Using 'not' to invert Boolean values ---

print("not True  →", not True)   # becomes False
print("not False →", not False)  # becomes True

on_sale = False
print("on_sale:", on_sale)
print("not on_sale:", not on_sale)  # flips False → True
`````

`````python
# --- Comparison Operators: Numbers ---

inventory_count = 5

print("inventory_count == 0  →", inventory_count == 0)
print("inventory_count != 0  →", inventory_count != 0)
print("inventory_count < 10  →", inventory_count < 10)
print("inventory_count > 10  →", inventory_count > 10)
print("inventory_count <= 5  →", inventory_count <= 5)
print("inventory_count >= 5  →", inventory_count >= 5)
`````

`````python
# --- Comparison Operators: Strings ---

product = "super snowboard"
greeting = "Hello"
farewell = "world!"

print("Does product match the expected value? →", product == "super snowboard")

print("Are the greeting and farewell different? →", greeting != farewell)

print("Is farewell something other than 'world!'? →", farewell != "world!")


# Useful: checking if user input matches expected formats
username_input = "Dawn"
expected_username = "dawn"

print("Is the username an exact match (case-sensitive)?",
      username_input == expected_username)

# Useful: verifying product categories
category = "Winter Sports"

print("Is the product category correct?",
      category == "Winter Sports")
`````

`````python
# --- Comparison Operators Using len() ---

# Compare lengths for password validation
user_password = "SnowDay123"
minimum_required_length = 8

print(
    "Is the password long enough?",
    len(user_password) >= minimum_required_length
)

# Ensure an error message is not empty before writing to logs
error_message = "Inventory count mismatch on SKU-4452"

print(
    "Does error_message contain text?",
    len(error_message) > 0
)

# Compare string lengths for two product descriptions
short_desc = "Ski Boots"
long_desc  = "High-performance alpine ski boots"

print(
    "Is long_desc longer than short_desc?",
    len(long_desc) > len(short_desc)
)
`````

`````python
# 🔗 Practical Chained Comparison Examples
# These examples mirror real-world analytics: pricing, scheduling, SLAs, and validation ranges.


# --- Helper function to display chained comparison results clearly ---

def show_chain(label, value):
    """
    Display a chained comparison and its Boolean result.
    Keeps output clean and helps reinforce understanding.
    """
    print(f"{label:50} → {value}")


# --- Chained comparisons (Python evaluates them left to right) ---

price = 499.99
sku_id = 4520
days_until_renewal = 45
response_time = 12   # minutes
temperature = -5     # Celsius

show_chain("Is price between $300 and $600?",
           300 < price < 600)

show_chain("Is SKU within valid range (4000–5000)?",
           4000 <= sku_id <= 5000)

show_chain("Is renewal 30–60 days away?",
           30 <= days_until_renewal <= 60)

show_chain("Is response time acceptable (10–20 min)?",
           10 <= response_time <= 20)

show_chain("Is temperature safe (−10 to +5 °C)?",
           -10 <= temperature <= 5)
`````

`````python
# --- Membership tests with strings and lists ---

#🔍 Helper Function for Clean Membership Output

def show_member(label, value):
    """
    Clean, aligned output for membership test results.
    Matches the formatting style of other helpers in this notebook.
    """
    print(f"{label:45} → {value}")

# --- Membership tests with strings ---
# Keywords `in` and `not in` are used to conduct membership tests
message = "Heavy snow expected this weekend on the mountain."

show_member("'snow' in message", "snow" in message)
show_member("'rain' in message", "rain" in message)
show_member("'snow' not in message", "snow" not in message)
show_member("'rain' not in message", "rain" not in message)

# --- Membership tests with lists ---
# Lists and other iterables work the same way
# Keywords `in` and `not in` are used to conduct membership tests

available_products = ["skis", "snowboard", "helmet", "goggles"]

show_member("'skis' in available_products", "skis" in available_products)
show_member("'boots' in available_products", "boots" in available_products)
show_member("'boots' not in available_products", "boots" not in available_products)
`````

`````python
# --- Combining conditions with 'and' & 'or' ---

inventory = 5
price = 99.0

# Both clauses must be True
print("inventory > 0 and price < 100          →", inventory > 0 and price < 100)

# At least one clause must be True
print("inventory == 0 or price < 100          →", inventory == 0 or price < 100)


# --- Additional Boolean operator examples ---

print("5 > 4 and 7 > 8                        →", 5 > 4 and 7 > 8)      # The `and` operator requires ALL statements to be true
print("5 > 4 or 7 > 8                         →", 5 > 4 or 7 > 8)       # The `or` operator requires ONE statement to be true

print("15 > 10 and 11 > 10 and 12 > 10        →",
      15 > 10 and 11 > 10 and 12 > 10)                                  # Chain together any number of boolean expressions


# Explanation (for reference)
# If the first expression in an AND chain is False,
# Python will NOT evaluate the rest (short-circuit).
# If the first expression in an OR chain is True,
# Python will NOT evaluate the rest.
`````

`````python
# --- Parentheses change the order of evaluation ---

item = "skis"
price = 120

# Example where 'or' is evaluated with parentheses
expr_1 = (item == "skis") or (price <= 100)                 # The `or` operator requires ONE statement to be true
expr_2 = (item == "snowboard") and (price <= 100)           # The `and` operator requires ALL statements to be true

print("item == 'skis' or price <= 100 →", expr_1)
print("item == 'snowboard' and price <= 100 →", expr_2)

# --- Parentheses to control the order of evaluation ---

clothing = "hat"
sale_price = 99

(sale_price >= 100 and clothing == "gloves") or (clothing == "hat")  # The `or` operator requires ONE statement to be true
`````

`````python
# --- Explicit grouping using parentheses ---

clothing = "hat"
sale_price = 99

# Intended rule:
# "Discount applies if (price <= 100 AND item is gloves) OR item is a hat"

with_parentheses = (sale_price <= 100 and clothing == "gloves") or (clothing == "hat")

print("With parentheses →", with_parentheses)


# 🔍 Explanation  
# - `sale_price <= 100` → True  
# - `clothing == "gloves"` → False  
# - `(True AND False)` → False  
# - `clothing == "hat"` → True  
# - Final: `False OR True` → **True**  
# - **Discount applies**


# ❗ Without Parentheses (Python chooses based on precedence)

# --- Same logic, but WITHOUT parentheses ---

without_parentheses = sale_price <= 100 and clothing == "gloves" or clothing == "hat"

print("Without parentheses →", without_parentheses)

#🔍 What happens?
# Python interprets it as:

(sale_price <= 100 and clothing == "gloves") or (clothing == "hat")

# Python is not short-circuiting the string check you expected
# Python is effectively using its own implicit parentheses based on precedence


# 🧪 When Results **Change** (Critical Lesson)

# Let’s change one variable and observe the difference:

# --- A setup where parentheses DO change the result ---

clothing = "jacket"
sale_price = 99

with_parentheses = (sale_price <= 100 and clothing == "gloves") or (clothing == "hat")
without_parentheses = sale_price <= 100 and clothing == "gloves" or clothing == "hat"

print("\nCase where differences appear:")
print("With parentheses →", with_parentheses)
print("Without parentheses →", without_parentheses)

# In this case the final value happens to be the same — but the logic is **ambiguous and error-prone**, especially in real systems with longer rules.
`````

`````python
# ==================================================
# 🔬 Demonstration: How Precedence Affects Results
# ==================================================

# 🧠 Boolean Operator Precedence Demo Helper

def show_precedence(label, expression, expected=None):
    """
    Prints the label, the actual evaluated result,
    and optionally the expected logical interpretation.
    Use this to visualize how Python groups Boolean expressions.
    """
    print(f"{label:45} → {expression}")
    if expected is not None:
        print(f"   Expected (manual reasoning): {expected}")
    print()  # blank line for readability
    

# AND outranks OR
show_precedence(
    "True or False and False",
    True or False and False,
    "True or (False and False) → True"
)

show_precedence(
    "False or True and False",
    False or True and False,
    "False or (True and False) → False"
)

# NOT outranks everything
show_precedence(
    "not True and False",
    not True and False,
    "(not True) and False → False"
)

show_precedence(
    "not False or False",
    not False or False,
    "(not False) or False → True"
)

# Parentheses override default order
show_precedence(
    "not False or False and False",
    not False or False and False,
    "(not False) or (False and False) → True"
)

show_precedence(
    "(not False or False) and False",
    (not False or False) and False,
    "((not False) or False) and False → False"
)
`````

`````python
# =====================================================================
# --- Demonstration: Short-Circuit using clear printed side effects ---
# =====================================================================

def fast_check():
    print("fast_check() ran → quick evaluation")  # pretend fast and simple
    return False  # Note: Returns False

def slow_check():
    print("slow_check() RAN → expensive evaluation (should NOT run if skipped)")  # pretend slow and complex
    return True   # Note: Returns True


print("\nCase 1: AND short-circuit (first condition is False)")
result_1 = fast_check() and slow_check()  # return False AND return True comparison                   
print("Final Result:", result_1)  # AND short-circuit: if A is False, Python NEVER evaluates B

# ⛳ Interpretation
# - In the `and` example, `fast_check()` is False → `slow_check()` never runs  

def fast_true():
    print("fast_true() ran → quick positive result")
    return True

print("\nCase 2: OR short-circuit (first condition is True)")
result_2 = fast_true() or slow_check()
print("Final Result:", result_2)

# ⛳ Interpretation
# - In the `or` example, `fast_true()` is True → `slow_check()` never runs
`````

`````python
# ================================================
# Short-circuit example with realistic conditions
# ================================================

def item_in_stock(inventory):
    print("item_in_stock() ran")
    return inventory > 0

def is_discount_item(item):
    print("is_discount_item() ran")
    return item in ["gloves", "hat"]

inventory = 0
item = "gloves"

print("\nReal short-circuit with AND (inventory = 0):")
result = item_in_stock(inventory) and is_discount_item(item)
print("Final:", result)

# ================================================
# Short-circuit example with realistic conditions
# ================================================

inventory = 1
item = "gloves"

def item_in_stock(inventory):
    print("item_in_stock() ran")
    return inventory > 0

def is_discount_item(item):
    print("is_discount_item() ran")
    return item in ["gloves", "hat"]

print("\nReal short-circuit with AND (inventory = 0):")
result = item_in_stock(inventory) and is_discount_item(item)
print("Final:", result)

# is_discount_item() ran — not short-circuited because inventory was > 0.
`````

`````python
# ============================================
# 🔍 trace_logic – Boolean Logic Tracing Helper
# ============================================

def trace_logic(label, func):
    """
    Run a logical expression (wrapped in a zero-argument function),
    print a clear header, and show the final Boolean result.

    Usage:
        trace_logic(
            "Description of the logic",
            lambda: some_check() and other_check() or third_check()
        )

    To see short-circuiting, each inner check should print when it runs.
    """
    print("\n" + "=" * 60)
    print("TRACE:", label)
    print("=" * 60)

    result = func()  # Evaluate the expression once
    print("Final result:", result)
    return result


# 🧵 Examples: Using `trace_logic()` with Real Checks
# --- Example condition functions with side effects ---
# Scenario 1: inventory = 0 (AND short-circuits, OR still needed)

inventory = 0
item = "gloves"
status = "VIP"

def item_in_stock(inventory):
    print(f"item_in_stock({inventory}) ran")
    return inventory > 0

def is_discount_item(item):
    print(f"is_discount_item({item!r}) ran")
    return item in ["gloves", "hat"]

def is_vip_customer(status):
    print(f"is_vip_customer({status!r}) ran")
    return status == "VIP"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)

# What you’ll see
# - In **Scenario 1** (`inventory = 0`):  
#   - `item_in_stock()` runs, returns False  
#   - `is_discount_item()` is **skipped** (AND short-circuited)  
#   - `is_vip_customer()` runs (needed for OR)


# --- Example condition functions with side effects ---
# Scenario 2: inventory > 0 and discount item (OR short-circuits)

inventory = 5
item = "gloves"
status = "regular"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)

# What you’ll see
# - In **Scenario 2** (`inventory = 5`, `item="gloves"`):  
#   - `item_in_stock()` runs, returns True  
#   - `is_discount_item()` runs, returns True  
#   - `(True and True)` becomes True  
#   - `is_vip_customer()` is **skipped** (OR short-circuited)

# `trace_logic()` gives you a clean header and final result;  
# your inner functions show **which steps actually executed**.
`````

`````python
# ============================================================
# Exercise Using `trace_decision()` and `trace_condition()`
# ============================================================

# ============================================================
# Helper Prerequisite: trace_decision(label, func)
# ============================================================

def trace_condition(name, value):
    print(f"{name:30} → {value}")


inventory = 3

if inventory <= 0:
    print("OUT OF STOCK")
elif inventory <= 5:
    print("LOW STOCK")
else:
    print("IN STOCK")


# ============================================================
# Advanced Helper 1: trace_decision(label, func)
# ============================================================

def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.

    label: A short description of what is being traced
    func:  A zero-argument function containing the decision logic
    """
    print("\n" + "=" * 60)
    print("DECISION TRACE:", label)
    print("=" * 60)
    func()


def classify_inventory(inventory):
    def inner():
        print(f"inventory = {inventory}")
        
        cond_out    = inventory <= 0
        cond_low    = 0 < inventory <= 5
        
        trace_condition("inventory <= 0",     cond_out)
        trace_condition("0 < inventory <= 5", cond_low)
        
        if cond_out:
            print("OUT OF STOCK")
        elif cond_low:
            print("LOW STOCK")
        else:
            print("IN STOCK")
            
    trace_decision("Inventory classification", inner)


# Test it
for inv in [0, 2, 6]:
    classify_inventory(inv)
`````

`````python
# ============================================================
# Exercise Using `trace_decision()` and `trace_condition()`
# ============================================================

# --- Price thresholds, descending order ---
price = 1499.99
luxury_threshold = 1000.00
expert_threshold = 500.00
intermediate_threshold = 300.00

test_prices = [1499.99, 800.00, 450.00, 250.00]

# ============================================================
# Advanced Helper 1: trace_decision(label, func)
# ============================================================

def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.

    label: A short description of what is being traced
    func:  A zero-argument function containing the decision logic
    """
    print("\n" + "=" * 60)
    print("DECISION TRACE:", label)
    print("=" * 60)
    func()

# ============================================================
# Wrap the trace_decision logic inside an inner function
# ============================================================

def classify_snowboard(price):
    def inner():                         # Wrap the trace_decision logic inside an inner function
        print(f"price = {price}")

        if price > intermediate_threshold:
            print("Branch: if → Good for intermediate users.")
        elif price > expert_threshold:
            print("Branch: elif → For experienced users.")
        elif price > luxury_threshold:
            print("Branch: elif → Luxury board (gold doesn't improve the ride!)")
        else:
            print("Branch: else → Suitable for a beginner.")

    trace_decision("Snowboard Skill Classification", inner)

# ============================================================
# --- Run the helper-driven trace ---
# ============================================================

classify_snowboard(price)


# ============================================================
# Helper 2: trace_condition(name, value)
# ============================================================

def trace_condition(name, value):
    """
    Print the name of a condition and its Boolean result.
    Useful for debugging and understanding complex rule logic.
    """
    print(f"{name:30} → {value}")


# =================================================
# 1. Buggy version: ordering is incorrect
# ❌ (demonstrates that elif blocks never run) ❌
# =================================================

def classify_snowboard_buggy(price):
    """
    Buggy snowboard classification example.

    The intent:
      - If price > luxury_threshold         → Luxury board
      - Elif price > expert_threshold       → For experienced users
      - Elif price > intermediate_threshold → For intermediate users
      - Else                                → Beginner board

    But the actual code (below) checks intermediate first, then expert, then luxury.
    Because intermediate_threshold is the lowest, the first condition catches
    almost everything, and the elif branches never run.
    """
    def inner():
        print(f"\n[BUGGY] price = {price}")
        
        # Show how the conditions evaluate
        cond_intermediate = (price > intermediate_threshold)
        cond_expert      = (price > expert_threshold)
        cond_luxury      = (price > luxury_threshold)

        trace_condition("price > intermediate_threshold", cond_intermediate)
        trace_condition("price > expert_threshold",      cond_expert)
        trace_condition("price > luxury_threshold",      cond_luxury)

        # Buggy ordering: intermediate first, then expert, then luxury
        if price > intermediate_threshold:
            print("Branch: if    → Good for intermediate users.")
        elif price > expert_threshold:
            print("Branch: elif1 → For experienced users.")
        elif price > luxury_threshold:
            print("Branch: elif2 → Luxury board (gold looks great).")
        else:
            print("Branch: else  → Suitable for a beginner.")

    trace_decision("Buggy snowboard classification", inner)


# Run buggy classification for several prices
for p in test_prices:
    classify_snowboard_buggy(p)


# ==========================================
# 2. Fixed version: ordering is corrected
#    (highest thresholds checked first)
# ==========================================

def classify_snowboard_fixed(price):
    """
    Correct snowboard classification example.

    Correct logic:
      - If price > luxury_threshold      → Luxury board
      - Elif price > expert_threshold    → For experienced users
      - Elif price > intermediate_threshold → For intermediate users
      - Else                             → Beginner board

    Conditions are checked from highest threshold to lowest.
    """
    def inner():
        print(f"\n[FIXED] price = {price}")
        
        # Boolean evaluation preview
        cond_intermediate = (price > intermediate_threshold)
        cond_expert      = (price > expert_threshold)
        cond_luxury      = (price > luxury_threshold)

        trace_condition("price > luxury_threshold",      cond_luxury)
        trace_condition("price > expert_threshold",      cond_expert)
        trace_condition("price > intermediate_threshold", cond_intermediate)

        # Correct threshold ordering: luxury first, then expert, then intermediate
        if cond_luxury:
            print("Branch: if    → Luxury board.")
        elif cond_expert:
            print("Branch: elif1 → For experienced users.")
        elif cond_intermediate:
            print("Branch: elif2 → Good for intermediate users.")
        else:
            print("Branch: else  → Suitable for a beginner.")

    trace_decision("Correct snowboard classification", inner)


# Run fixed classification for the same prices
for p in test_prices:
    classify_snowboard_fixed(p)
`````

## 1. The Boolean Data Type (`True` / `False`)

A **Boolean** is a data type with only **two possible values**:

- `True`
- `False`

In arithmetic:

- `True` behaves like `1`
- `False` behaves like `0`
- The `not` keyword **inverts** a Boolean

## 2. Comparison Operators

**Comparison operators** evaluate a relationship and return a **Boolean** (`True` / `False`):

| Operator | Meaning                    | Example        |
|---------:|----------------------------|----------------|
| `==`     | equal to                   | `price == 99`  |
| `!=`     | not equal to               | `item != 'ski'`|
| `<`      | less than                  | `inv < 10`     |
| `>`      | greater than               | `temp > 32`    |
| `<=`     | less than or equal to      | `age <= 18`    |
| `>=`     | greater than or equal to   | `score >= 80`  |

## 🔗 3. Chained Comparisons (Readable Multi-Condition Checks)

Python allows you to “chain” comparisons together, which makes range checks concise and readable.  
Chained comparisons evaluate **left to right**, and Python ensures all comparisons are True.

This is extremely common in pricing logic, validation rules, SLAs, and data quality checks.


This form reads almost like natural language, which is why chained comparisons are widely used in analytics logic, data validation, and business rule engines.


### 🔗 Chained Comparisons Helper Function (Enhanced Clarity)

## 5. Boolean Operators: `and`, `or`, `not` (+ Short-Circuiting & Precedence)

Boolean operators allow you to combine multiple logical expressions into a single condition:

- **`A and B`** → True only if **both** A and B are True  
- **`A or B`** → True if **at least one** of A or B is True  
- **`not A`** → inverts A (True ↔ False)

---

### 🔄 Short-Circuit Behavior (Efficiency Tip)

Python **stops evaluating early** when the final outcome is already determined:

- For **`A and B`**:  
  - If `A` is **False**, Python **does not evaluate B**
- For **`A or B`**:  
  - If `A` is **True**, Python **does not evaluate B**

This is why we typically place **simple or fast checks first**, and **more expensive operations last**.

---

### 🧠 Operator Precedence (Highest → Lowest)

Python always evaluates Boolean operators in this fixed order:

1. **`not`**  
2. **`and`**  
3. **`or`**

**This means:**
- `not` is always evaluated **before** `and`  
- `and` is always evaluated **before** `or`  
- This happens **even if the operators appear in a different order in the code**

Example mental model:

```
A or B and C
→ A or (B and C)
```

Python will always group `B and C` first because `and` outranks `or`.

---

### 🟦 Parentheses Always Win

Parentheses let you **override** Python’s default precedence rules and intentionally control how the logic is grouped.

Use parentheses to:
- Make your logic explicit  
- Avoid ambiguity in complex conditions  
- Ensure correctness when mixing `and` and `or`  
- Improve readability for yourself and others

Whenever your logic combines different operators, parentheses are a best practice for clarity and safety.

1. `not`
2. `and`
3. `or`

## ⚡ Short-Circuit Evaluation (Efficiency in Action)

Python **stops evaluating early** when a final result is already determined.

- For **`A and B`**:  
  If `A` is False → *Python never evaluates B*
- For **`A or B`**:  
  If `A` is True → *Python never evaluates B*

Short-circuiting is important for performance and safety.

Here is a clear demonstration using “fast” and “slow” checks.

```python
# --- Short-Circuit Demo using clear printed side effects ---

def fast_check():
    print("fast_check() ran → quick evaluation")
    return False  # pretend fast and simple

def slow_check():
    print("slow_check() RAN → expensive evaluation (should NOT run if skipped)")
    return True   # pretend slow and complex


print("\nCase 1: AND short-circuit (first condition is False)")
result_1 = fast_check() and slow_check()
print("Final Result:", result_1)


def fast_true():
    print("fast_true() ran → quick positive result")
    return True

print("\nCase 2: OR short-circuit (first condition is True)")
result_2 = fast_true() or slow_check()
print("Final Result:", result_2)
```

### ⛳ Interpretation
- In the `and` example, `fast_check()` is False → `slow_check()` never runs  
- In the `or` example, `fast_true()` is True → `slow_check()` never runs  

This mirrors real code patterns where we put:
- Cheap validation checks **first**  
- Expensive computations **second**  


## 🧪 `trace_logic()` – Visualize How Your Conditions Run

`trace_logic()` is a small helper that makes it easier to **see** how a complex Boolean expression is evaluated.

You’ll use it together with functions that print when they run (like `item_in_stock()` and `is_discount_item()`).  
This lets you watch:

- which checks run
- which checks are skipped (short-circuited)
- what the final result is

---

```python
# ============================================
# 🔍 trace_logic – Boolean Logic Tracing Helper
# ============================================

def trace_logic(label, func):
    """
    Run a logical expression (wrapped in a zero-argument function),
    print a clear header, and show the final Boolean result.

    Usage:
        trace_logic(
            "Description of the logic",
            lambda: some_check() and other_check() or third_check()
        )

    To see short-circuiting, each inner check should print when it runs.
    """
    print("\n" + "=" * 60)
    print("TRACE:", label)
    print("=" * 60)

    result = func()  # Evaluate the expression once
    print("Final result:", result)
    return result
```

---

## 🧵 Example: Using `trace_logic()` with Real Checks

```python
# --- Example condition functions with side effects ---

def item_in_stock(inventory):
    print(f"item_in_stock({inventory}) ran")
    return inventory > 0

def is_discount_item(item):
    print(f"is_discount_item({item!r}) ran")
    return item in ["gloves", "hat"]

def is_vip_customer(status):
    print(f"is_vip_customer({status!r}) ran")
    return status == "VIP"


# Scenario 1: inventory = 0 (AND short-circuits, OR still needed)
inventory = 0
item = "gloves"
status = "VIP"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)


# Scenario 2: inventory > 0 and discount item (OR short-circuits)
inventory = 5
item = "gloves"
status = "regular"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)
```

### What you’ll see

- In **Scenario 1** (`inventory = 0`):  
  - `item_in_stock()` runs, returns False  
  - `is_discount_item()` is **skipped** (AND short-circuited)  
  - `is_vip_customer()` runs (needed for OR)

- In **Scenario 2** (`inventory = 5`, `item="gloves"`):  
  - `item_in_stock()` runs, returns True  
  - `is_discount_item()` runs, returns True  
  - `(True and True)` becomes True  
  - `is_vip_customer()` is **skipped** (OR short-circuited)

`trace_logic()` gives you a clean header and final result;  
your inner functions show **which steps actually executed**.

---

## ✅ Key Takeaways – `trace_logic()`

- `trace_logic(label, func)` runs a **single Boolean expression** and shows the final result with a clear trace header.
- Wrap your expression in a `lambda` so it **is not evaluated** until `trace_logic()` calls it.
- Put `print()` statements inside your condition functions (like `item_in_stock`) to see:
  - which checks run  
  - where short-circuiting skips later checks  
- Use `trace_logic()` to:
  - experiment with **AND / OR / NOT**
  - confirm **precedence and parentheses**
  - build deep intuition about **how Python really evaluates conditions**

This tool is perfect for “x-ray vision” into your Boolean logic while you’re learning and debugging.


## 2.8 Advanced Decision-Flow Utilities  
These helper tools make it easier to **see** and **understand** which branches execute when using `if`, `elif`, and `else`.

# 🔹 1. Helper Demo (Simple, In-Function Branch Tracing)

This first example prints the branch **inside the function itself**.  
It is great for beginners and quick tests.

```python
# --- Demonstration Helper: show which branch executes ---

def check_shipping(order_total):
    """
    Simple demonstration function:
    Prints which shipping branch runs based on order_total.
    """
    print(f"\nChecking shipping for order_total = {order_total}")
    
    if order_total >= 100:
        print("Branch: if → Free shipping")
    else:
        print("Branch: else → Standard shipping")

# Try a few values
check_shipping(120)
check_shipping(75)
check_shipping(100)
```

This is a good first step, but the next helpers are more flexible and powerful.

---

# 🔹 2. trace_decision()  
### (Advanced Helper: Traces Full If / Elif / Else Structures)

This helper wraps a function that contains your decision logic.  
It prints a clean header and lets the inner function show which branches ran.

```python
# --- Advanced Helper: trace_decision(label, func) ---

def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.
    
    label: A short description of what is being traced
    func:  A zero-argument function containing the decision logic
    """
    print("\n" + "=" * 60)
    print("DECISION TRACE:", label)
    print("=" * 60)
    func()
```

### Example: Shipping logic with rush + order total

```python
# --- Example: using trace_decision with a multi-branch decision ---

def decide_shipping(order_total, rush):
    def inner():                                       # Wrap the trace_decision logic inside an inner function
        print(f"order_total={order_total}, rush={rush}")
        # if / elif / else chain here...
        if rush:
            print("Branch: if → Express shipping")
        elif order_total >= 100:
            print("Branch: elif → Free standard shipping")
        else:
            print("Branch: else → Paid standard shipping")
    
    trace_decision("Shipping decision", inner)        # label, inner function

# --- Run the helper-driven trace ---
decide_shipping(150, rush=True)
decide_shipping(100, rush=False)
decide_shipping(50, rush=False)
```

This helper is excellent for **complex decision trees**.
This pattern encourages you to think about:

- which branch runs  
- which conditions are tested  
- how order affects the outcome
- **Why do we use inner()?**
→ trace_decision() expects a zero-argument function. 
   inner() captures variables (order_total, rush) and runs the logic.
- **Why do we manually pass rush=True/False?**
→ The helper should NEVER define its own inputs.
   We pass the values we want to test, so we can see different branches.
  
---

# 🔹 4. trace_condition()  
### (Shows How Each Boolean Condition Evaluates)

This helper prints the **name** of a condition and its Boolean result.

```python
# --- Helper: trace_condition(name, value) ---

def trace_condition(name, value):
    """
    Print the name of a condition and its Boolean result.
    Useful for debugging and understanding complex rule logic.
    """
    print(f"{name:30} → {value}")
```

### Example: Debugging a discount rule

```python
# --- Example: inspecting a multi-condition discount rule ---

def debug_discount_rule(status, order_total, item):
    print(f"\nStatus={status!r}, order_total={order_total}, item={item!r}")
    
    discount_items = ["gloves", "hat"]
    
    cond_vip = (status == "VIP")
    cond_high_total = (order_total >= 100)
    cond_discount_item = (item in discount_items)
    
    # Show each condition's result
    trace_condition("status == 'VIP'", cond_vip)
    trace_condition("order_total >= 100", cond_high_total)
    trace_condition("item in discount_items", cond_discount_item)
    
    # Combine the rule
    eligible = cond_vip or (cond_high_total and cond_discount_item)
    trace_condition("Final rule result", eligible)
    
    if eligible:
        print("Decision: Discount applied.")
    else:
        print("Decision: No discount.")

debug_discount_rule("VIP", 50, "gloves")
debug_discount_rule("REGULAR", 120, "gloves")
debug_discount_rule("REGULAR", 80, "jacket")
```

# 🔹 5. trace_if()  
### (Compact Helper for Single If/Else Decisions)

This helper describes the decision, shows the condition result,  
and names the branch selected.

```python
# --- Helper: trace_if(description, condition, if_true, if_false) ---

def trace_if(description, condition, if_true, if_false):
    """
    Describe an if/else style decision in a compact, readable format.
    
    description: description of the decision being checked
    condition:   Boolean expression
    if_true:     message to print if the condition is True
    if_false:    message to print if the condition is False
    """
    print(f"\nChecking: {description}")
    print(f"Condition result: {condition}")
    
    if condition:
        print(f"Branch selected: {if_true}")
    else:
        print(f"Branch selected: {if_false}")
```

### Example: Using trace_if for two related decisions

```python
# --- Example: shipping decisions with trace_if ---

def shipping_decision(order_total, rush):

    trace_if(
        description="Rush shipping requested",
        condition=rush,
        if_true="Express shipping",
        if_false="No rush shipping"
    )
    
    trace_if(
        description="Eligible for free standard shipping",
        condition=(order_total >= 100),
        if_true="Free standard shipping",
        if_false="Standard shipping (paid)"
    )

shipping_decision(150, rush=False)
shipping_decision(40, rush=True)
shipping_decision(40, rush=False)
```

| Teach or inspect **Boolean logic**               | `trace_condition`                               |
| **Full visibility for complex decisions**        | ⭐ `trace_branch` + `trace_condition` + `trace_decision` ⭐ |

---

# --- Helper 1: trace_decision --------------------------------
def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.
    
    label: Short description of what is being traced.
    func:  Zero-argument function containing the decision logic.
    """
    print("\n" + "=" * 60)
    print(f"DECISION TRACE: {label}")
    print("=" * 60)
    func()


# --- Helper 2: trace_condition -------------------------------
def trace_condition(name, value):
    """
    Print the name of a Boolean condition and its result (True/False).
    Useful for debugging and teaching how conditions evaluate.
    
    name:  String label for the condition being checked.
    value: Boolean expression result.
    """
    print(f"{name:30} → {value}")


# --- Helper 3: trace_branch ----------------------------------
def trace_branch(label):
    """
    Print a simple message showing which branch was taken.
    Use inside if / elif / else blocks.
    
    label: Text describing the branch.
    """
    print(f"→ Branch taken: {label}")


# --- Helper 4: trace_if --------------------------------------
def trace_if(description, condition, if_true, if_false):
    """
    Compact helper for simple if/else decisions.
    
    description: Description of the decision.
    condition:   Boolean expression.
    if_true:     Message printed when condition is True.
    if_false:    Message printed when condition is False.
    """
    print(f"\nChecking: {description}")
    print(f"Condition result: {condition}")
    
    if condition:
        print(f"Branch selected: {if_true}")
    else:
        print(f"Branch selected: {if_false}")

print("🟢 Helper Toolkit Loaded: trace_decision, trace_condition, trace_branch, trace_if")



# ✅ Key Takeaways — Chapter 5: Conditional Logic

### **1. All conditional logic is built on Boolean expressions**
Everything begins with expressions that evaluate to **True** or **False**:
- Comparisons (`>`, `<`, `>=`, `<=`, `==`, `!=`)
- Membership tests (`in`, `not in`)
- Logical operators (`and`, `or`, `not`)
- Chained comparisons (`0 < x <= 5`)

Understanding how to build Boolean expressions is the foundation of all control flow.

---

### **2. Python follows strict Boolean precedence**
Python evaluates conditions in this order:

1. `not`  
2. `and`  
3. `or`  

Use parentheses when your intent does not match Python’s default order.

---

### **3. Short-circuiting affects program behavior**
- `A and B`: if A is False → Python never evaluates B  
- `A or B`: if A is True → Python never evaluates B  

Short-circuiting controls:
- performance
- side-effects (function calls inside conditions)
- rule behavior in real systems

---

### **4. `if`, `elif`, `else` form a single decision chain**
Important principles:
- Conditions are checked **top to bottom**
- Only **one branch** runs
- When a branch is True, Python **skips** all remaining branches
- Most specific conditions should appear first

Always order conditions from **most restrictive** → **least restrictive**.

---

### **5. Nesting creates multi-layered, real-world decision trees**
Use nested logic when rules follow a natural sequence:

> "If this is true, then check something else."

Nesting is ideal for:
- multi-step approvals  
- customer + inventory combinations  
- authentication + permissions  
- multi-rule eligibility checks  

---

### **6. Combine Boolean logic + control flow for expressive rule systems**
Everything from Part 1 becomes building blocks in Part 2:

- `(status == "VIP") and (order_total >= 100)`
- `"hat" in discount_items`
- `not is_blocked`
- `0 < inventory <= 5`

This enables:
- pricing systems  
- discount logic  
- safety checks  
- menu/option logic  
- workflows and gating rules  

---

### **7. Proper ordering prevents unreachable branches**
Incorrect ordering example:

```
if score >= 60:
    ...
elif score >= 90:   # never runs
```

Correct ordering:

```
if score >= 90:
    ...
elif score >= 60:
    ...
```

Always ask:
- Is this condition more general than the ones below?  
- Should this rule be evaluated earlier?  

---

### **8. Helper tools dramatically increase clarity and reduce bugs**
Your helper suite is a game-changer:

- **trace_decision()**  
  Shows a labeled, full-width decision trace

- **trace_condition()**  
  Displays Boolean results for each rule

- **trace_branch()**  
  Labels which branch executed

- **trace_if()**  
  Compact single-condition inspector

Using these consistently gives you:
- transparency  
- easier debugging  
- better intuition  
- fewer logic errors  

These helpers raise your skill far above the Maven-level exercises.

---

### **9. Maintain readable, intention-driven logic**
Best practices:
- Use descriptive variable names  
- Use parentheses for clarity  
- Keep branches small and readable  
- Avoid unnecessary deep nesting  
- Comment the *why*, not the *what*  

Readable control flow → faster debugging and fewer mistakes.

---

### **10. Conditional logic powers real-world automation**
You can now design systems for:
- pricing rules  
- discount engines  
- inventory logic  
- option/menu trees  
- quality and safety checks  
- multi-step authentication  
- eligibility workflows  

This prepares you for the next big steps:
- loops  
- functions  
- pandas filtering  
- data-cleaning decisions  
- machine-learning preprocessing  

You’ve mastered both the **syntax** and the **thinking** behind decision logic.

`````python
# 06.05 Slicing: Extracting Portions of a List
# Using the List Index Map Helper for a flat list

# Slice syntax: list[start : stop : step]

sizes = ['XS', 'S', 'M', 'L', 'XL', 'XXL']

print(sizes[:3])       # ['XS', 'S', 'M']
print(sizes[:4])       # ['XS', 'S', 'M', 'L']
print(sizes[2:5])      # ['M', 'L', 'XL']
print(sizes[::2])      # ['XS', 'M', 'XL']
print(sizes[2:4])      # ['M', 'L']
print(sizes[-3:])      # ['L', 'XL', 'XXL']
print()

# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input: validate isinstance(obj, type)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
show_list_index_map(sizes)         # flat list
print()
`````

`````python
# --------------------------------------------------------------
# 🔧 Helper: Slice Visualizer (for lists, tuples, strings, ranges)
# --------------------------------------------------------------

def visualize_slice(seq, start=None, stop=None, step=None, show_negative=True, label="seq"):
    """
    Visualize how a slice seq[start:stop:step] selects elements
    from a sequence (list, tuple, string, range, etc.).

    Parameters
    ----------
    seq : sequence
        Any indexable, sliceable Python sequence (e.g., list, tuple, str, range).
    start : int or None
        Slice start index (can be negative or None).
    stop : int or None
        Slice stop index (can be negative or None).
    step : int or None
        Slice step (can be negative or None).
    show_negative : bool, default True
        Whether to display negative indices.
    label : str, default "seq"
        Optional label to show in the output header.

    Usage
    -----
    visualize_slice([10, 20, 30, 40, 50], 1, 4, 2)
    visualize_slice(range(10), 2, 9, 3)
    visualize_slice(('a', 'b', 'c', 'd', 'e'), 0, None, 2)
    """

    # ----------------------------------------------------------
    # 1. Basic validation
    # ----------------------------------------------------------
    try:
        n = len(seq)
    except TypeError:
        raise TypeError(
            f"visualize_slice expects a sequence with len(), got {type(seq).__name__}"
        )

    slice_obj = slice(start, stop, step)

    # Normalize slice parameters using slice.indices()
    norm_start, norm_stop, norm_step = slice_obj.indices(n)

    # Compute selected indices based on normalized values
    selected_indices = list(range(norm_start, norm_stop, norm_step)) if n > 0 else []

    # ----------------------------------------------------------
    # 2. High-level summary
    # ----------------------------------------------------------
    print(f"Sequence label : {label}")
    print(f"Type           : {type(seq).__name__}")
    print(f"Length         : {n}")
    print(f"Slice requested: {label}[{start}:{stop}:{step}]")
    print(f"Normalized to  : {label}[{norm_start}:{norm_stop}:{norm_step}]")
    print()

    if n == 0:
        print("This sequence is empty. No indices to visualize.")
        return

    # ----------------------------------------------------------
    # 3. Build index and value rows
    # ----------------------------------------------------------
    # We'll give each slot a fixed width for alignment
    cell_width = 5

    # Positive indices
    idx_row = "idx : "
    for i in range(n):
        idx_row += f"{i:>{cell_width-1}} "

    # Negative indices (optional)
    if show_negative:
        neg_row = "neg : "
        for i in range(n):
            neg_row += f"{i - n:>{cell_width-1}} "
    else:
        neg_row = None

    # Values (repr truncated for readability)
    val_row = "val : "
    for i in range(n):
        text = repr(seq[i])
        if len(text) > cell_width - 1:
            text = text[:cell_width - 2] + "…"
        val_row += f"{text:>{cell_width-1}} "

    # Selection markers
    sel_row = "sel : "
    selected_set = set(selected_indices)
    for i in range(n):
        marker = "^" if i in selected_set else ""
        sel_row += f"{marker:>{cell_width-1}} "

    # ----------------------------------------------------------
    # 4. Print the visualization
    # ----------------------------------------------------------
    print(idx_row)
    if neg_row is not None:
        print(neg_row)
    print(val_row)
    print(sel_row)
    print()

    # ----------------------------------------------------------
    # 5. Show the actual slice result
    # ----------------------------------------------------------
    result = seq[slice_obj]
    print(f"Result of {label}[{start}:{stop}:{step}] → {repr(result)}")
    print()


# --------------------------------------------------------------
# Example usage (you can comment these out in your notebook):
# --------------------------------------------------------------
visualize_slice(['XS', 'S', 'M', 'L', 'XL', 'XXL'], 1, 5, 2, label="sizes")
visualize_slice(range(10, 0, -2), None, None, None, label="r")
visualize_slice(('r', 'g', 'b', 'y', 'k'), 1, 4, 1, label="colors")
`````

`````python
# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types & Nested Lists)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input: validate isinstance(obj, type)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
        # Optionally expand nested lists one level deep
        if expand_nested and isinstance(value, list):
            inner_len = len(value)
            print(f"      ↳ nested list (len={inner_len})")
            for j, inner_val in enumerate(value):
                inner_neg = j - inner_len if show_negative else None
                inner_type = type(inner_val).__name__ if show_types else ""
                inner_repr = _repr_value(inner_val)
                
                # Example format: [2][1]   -1    int  99
                nested_row = f"      [{i}][{j}] "
                if show_negative:
                    nested_row += f"{inner_neg:>4} "
                if show_types:
                    nested_row += f"{inner_type:>10} "
                nested_row += f"{inner_repr}"
                print(nested_row)


# -------------------------------------------
# ✅ Example Usage (Flat List)
# -------------------------------------------
sizes = ['XS', 'S', 'M', 'L', 'XL']
show_list_index_map(sizes)         # flat list
print()

# -------------------------------------------
# ✅ Example Usage (Nested List, Expanded)
# -------------------------------------------
orders = [
    [19.99, 5.99],
    [3.49, 9.99, 14.99],
    [60.00]
]

show_list_index_map(
    orders,                         # nested list
    show_types=True,
    show_negative=True,
    truncate=50,
)
`````

# 📘 Common List Methods — Essential & Advanced Reference

Python lists are one of the most powerful and frequently used data structures.  
These tables provide a clear, student-friendly reference that you will use constantly throughout:

- Conditional logic  
- Loops (Chapter 07)

# 🟦 Table 1 — Essential List Methods (Everyday Use)

These are the **must-know** operations used in 90% of real-world Python work.

| **Method**          | **Purpose**                                    | **Mutates list?** | **Example**                               |
|---------------------|------------------------------------------------|-------------------|-------------------------------------------|
| `lst.append(x)`     | Add item to end of list                        | Yes               | `sizes.append("XL")`                      |
| `lst.insert(i, x)`  | Insert item at index `i`                       | Yes               | `sizes.insert(0, "XS")`                   |
| `lst.extend(iter)`  | Add all elements from another iterable         | Yes               | `sizes.extend(["2XL", "3XL"])`            |
| `lst.remove(x)`     | Remove *first* matching value                  | Yes               | `sizes.remove("M")`                       |
| `lst.pop(i)`        | Remove + return item at index `i`              | Yes               | `sizes.pop(2)`                            |
| `lst.clear()`       | Remove all items                               | Yes               | `sizes.clear()`                           |
| `lst.sort()`        | Sort list in place                             | Yes               | `numbers.sort()`                          |
| `lst.reverse()`     | Reverse list in place                          | Yes               | `numbers.reverse()`                       |
| `lst.count(x)`      | Count occurrences of value                     | No                | `sizes.count("L")`                        |
| `lst.index(x)`      | Return index of first matching value           | No                | `sizes.index("M")`                        |

---

## 🔍 Notes for Essential Methods

- `append()` vs `insert()`  
  - Use `append()` unless you *truly* need a specific position  
  - `insert()` shifts the entire list (costly with large data)

- `remove()` deletes **only the first** matching item  
  - Useful to know when cleaning data

- `pop()` is the **only list method that returns a value**

- `sort()` is **in-place**  
  - Use `sorted(lst)` when you want to keep the original list intact  
  - This difference becomes crucial for data analysis

---

# 🟧 Table 2 — Advanced & Deep-Dive List Methods, Patterns, and Behaviors

These tools build **intermediate → advanced skill**, prevent major mistakes, and support:

- nested lists  
- object references  
- copying behavior  
- safe transformations  
- professional data processing patterns  

| **Method / Behavior**       | **Purpose / Explanation**                                                   | **Mutates?** | **Example**                               |
|-----------------------------|-------------------------------------------------------------------------------|--------------|--------------------------------------------|
| `lst.copy()`               | Creates a *shallow* copy (outer list only)                                   | No (returns new list) | `new_lst = lst.copy()` |
| `copy.deepcopy(lst)`       | Creates a *full deep copy*, including nested lists                           | No | `from copy import deepcopy` <br>`safe = deepcopy(lst)` |
| List slicing `lst[a:b]`    | Returns a new list — non-inclusive end index                                 | No | `subset = lst[1:4]`                         |
| Slice assignment           | Replace multiple items at once                                               | Yes | `lst[1:3] = ["A", "B"]`                     |
| Extended slice `lst[::-1]`| Reverse list without mutating original                                        | No | `rev = lst[::-1]`                           |
| Concatenation `lst + lst2`| Combine lists into a new list                                                 | No | `all_sizes = summer + winter`              |
| Repetition `lst * n`       | Repeat elements n times                                                      | No | `[1, 2] * 3 → [1,2,1,2,1,2]`                |
| Membership test `x in lst` | Check for presence                                                            | No | `'L' in sizes`                              |
| `del lst[i]`               | Remove item at index                                                         | Yes | `del lst[0]`                                |
| `del lst[a:b]`             | Delete slice                                                                 | Yes | `del lst[2:5]`                              |
| `min(lst)` / `max(lst)`    | Return smallest or largest item                                              | No | `max(prices)`                              |
| `sum(lst)`                 | Total numeric values                                                          | No | `sum(revenues)`                            |
| List comprehension         | Create transformed lists cleanly                                              | No | `[x*2 for x in nums]`                      |

---

## 🧠 Deep-Dive Notes for Advanced Behaviors

### 🔸 Shallow vs Deep Copy (Critical Concept)
- `lst.copy()` copies *only* the outer list  
- nested lists remain linked  
- modifying nested elements affects both lists  
- Use `deepcopy()` when safety is required

### 🔸 Slice Non-Inclusive Behavior
- `lst[a:b]` excludes index `b`  
- Same rule as string slicing  
- Predictable and consistent across Python

### 🔸 Reversing Safely
- `lst[::-1]` → **non-destructive**  
- `lst.reverse()` → **mutates**  
- Choose based on whether you need the original

### 🔸 Slice Assignment
Example:  
`lst[1:3] = ["A", "B", "C"]`  
- You can replace a slice with a different number of items  
- Useful for editing data in place  
- Python will automatically adjust list size

### 🔸 Membership + Condition Patterns
These feed directly into loops:
- `if item in lst:`  
- `if "XL" not in sizes:`

# 06.01 Overview

In Python, **sequences** are ordered collections of items. The three core sequence types in this chapter are:

1. **Lists**  
2. **Tuples**  
3. **Ranges**

This part focuses on **lists**, the most flexible and commonly used sequence type in Python.

According to the Maven workbook :contentReference[oaicite:1]{index=1}:  
- Lists are *iterable*  
- Lists can store many values  
- Lists can contain elements of **any data type**  
- Lists are **mutable** (you can add, remove, or change elements)

This makes lists essential for data cleaning, analytics, transformations, and representing rows of data.

---

# 06.02 Creating Lists

Lists are created using **square brackets**:

```python
# Creating a simple list
sizes = ['S', 'M', 'L', 'XL']

# Lists can hold mixed data types (not recommended for analytics work)
mixed_list = ['Alice', 42, True, 9.5]

# Creating an empty list
empty_list = []

# Creating a list from another iterable (string, range, tuple, etc.)
letters = list('abcde')       # ['a', 'b', 'c', 'd', 'e']
numbers = list(range(5))      # [0, 1, 2, 3, 4]
```

### Mini-diagram: List structure

```
sizes = ['S', 'M', 'L', 'XL']

Index:        0    1    2     3
Element:     'S'  'M'  'L'   'XL'
```

Python always starts indexing at **0**, not 1.

# 06.03 Membership Tests ("in" and "not in")

Because lists are iterable, you can test whether a value exists inside a list.

```python
sizes_in_stock = ['XS', 'S', 'M', 'L', 'XL']

print('M' in sizes_in_stock)       # True
print('XXL' in sizes_in_stock)     # False

# You can use membership tests in if/elif logic
if 'M' in sizes_in_stock:
    print("Medium is available.")
elif 'S' in sizes_in_stock:
    print("Small is available instead.")
else:
    print("Neither size is available.")
```

This concept is heavily used in:

- Filtering data  
- Validating inputs  
- Checking categories  
- Cleaning data  
- Applying business rules  

---

# 06.04 Indexing (Accessing Individual Elements)

Use square brackets with an index number to access a single item.

```python
colors = ['red', 'blue', 'green', 'yellow']

print(colors[0])    # 'red'   first element
print(colors[2])    # 'green' third element
```

### Mini-diagram

```
colors = ['red', 'blue', 'green', 'yellow']
Index:      0       1        2         3
```

### Negative indexing  
Python allows negative indexes, which count from the end:

```python
print(colors[-1])   # 'yellow'
print(colors[-2])   # 'green'
```

Negative indexing is standard in Python and very powerful for last-item access.

---

# 06.05 Slicing (Extracting Portions of a List)

Slice syntax:

```
list[start : stop : step]
```

- **start** = where the slice begins  
- **stop** = where the slice ends (non-inclusive)  
- **step** = how many to skip  

```python
sizes = ['XS', 'S', 'M', 'L', 'XL', 'XXL']

print(sizes[:3])       # ['XS', 'S', 'M']
print(sizes[2:5])      # ['M', 'L', 'XL']
print(sizes[::2])      # ['XS', 'M', 'XL']
print(sizes[-3:])      # ['L', 'XL', 'XXL']
```

### Mini-diagram: Non-inclusive stop  
For `sizes[2:5]`:

```
Index:    0    1    2     3     4     5
Value:   XS    S    M     L    XL   XXL
Slice:              [ M    L    XL ]
Stop index = 5 (excluded)
```

Common uses of slicing:

- Selecting subsets  
- Pulling rows of data  
- Extracting ranges  
- Removing headers  
- Splitting time intervals

# 06.06 Unpacking Lists

Unpacking allows you to assign list elements directly to variables.

```python
sizes = ['S', 'M', 'L']

small, medium, large = sizes

print(small)    # 'S'
print(medium)   # 'M'
print(large)    # 'L'
```

### Important rule  
> The number of variables must match the number of elements.

Otherwise, Python raises a `ValueError`.

- Your data needs to be **ordered**  
- You expect values to **change over time**  
- You need to **grow or shrink** the collection  
- You need flexibility during analysis  

Examples in real analytics:

- Storing transactions  
- Storing cleaned values  
- Holding unique categories  
- Working with rows before converting to pandas  
- Preparing data for loops

# 🧮 Universal List Index Map Helper (Professional Version – With Parameters)

This helper is the **list equivalent** of your Universal String Index Map Helper.

It is designed to:

- Show **positive and negative indices** for each element in a list  
- Optionally display **data types**  
- Optionally **truncate long values** for readability  
- Optionally **expand nested lists** and show their indices as `[outer][inner]`  
- Provide a **clear, student-friendly view** of how indexing works in real data  

You can use this to:

- Understand and debug indexing  
- Teach indexing to others  
- Explore nested list structures  
- Build confidence with list access patterns before loops and slicing-heavy work  

---

## 🔧 Python Helper: `show_list_index_map`

```python
# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types & Nested Lists)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input validation (student-friendly)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
        # Optionally expand nested lists one level deep
        if expand_nested and isinstance(value, list):
            inner_len = len(value)
            print(f"      ↳ nested list (len={inner_len})")
            for j, inner_val in enumerate(value):
                inner_neg = j - inner_len if show_negative else None
                inner_type = type(inner_val).__name__ if show_types else ""
                inner_repr = _repr_value(inner_val)
                
                # Example format: [2][1]   -1    int  99
                nested_row = f"      [{i}][{j}] "
                if show_negative:
                    nested_row += f"{inner_neg:>4} "
                if show_types:
                    nested_row += f"{inner_type:>10} "
                nested_row += f"{inner_repr}"
                print(nested_row)


# -------------------------------------------
# ✅ Example Usage (Flat List)
# -------------------------------------------
sizes = ['XS', 'S', 'M', 'L', 'XL']
show_list_index_map(sizes)

# -------------------------------------------
# ✅ Example Usage (Nested List, Expanded)
# -------------------------------------------
orders = [
    [19.99, 5.99],
    [3.49, 9.99, 14.99],
    [60.00]
]

show_list_index_map(
    orders,
    show_types=True,
    show_negative=True,
    truncate=50,
    expand_nested=True
)
```

---

## 📝 Key Takeaways for the List Index Helper

- `show_list_index_map(lst, ...)` gives you a **readable table** of:
  - positive index (`idx`)  
  - optional negative index (`neg`)  
  - optional type (`type`)  
  - value (`value`)  

- `expand_nested=True` lets you:
  - see **nested lists** clearly  
  - view indices like `[outer][inner]`  
  - understand how to access them using `lst[i][j]`  

- `truncate` helps keep very long values readable in the console.

- This helper is ideal for:
  - learning list indexing  
  - debugging index errors  
  - understanding nested structures  
  - preparing for loops, slices, and data munging

`````python
# 06.12 Adding Elements: .append() and .insert()

# Two main ways to add new elements:

# 1. append(element) - Adds to the end of the list.
# Mutates the list in place ⚠️

items = ['Hat', 'Gloves', 'Scarf']
items.append('Coat')

print(items)  # display mutated list


# 2. insert(index, element) 
# Inserts at a specific position.
# Mutates the list in place ⚠️ 

items = ['Hat', 'Gloves', 'Scarf']
items.insert(3, 'Coat')   # insert at index 3 (fourth position)

print(items, end="\n\n")   # display mutated list


# ==================================================
# 06.12b Changing List Elements (Direct Assignment)
# ==================================================
# Replace items using indexing
# Mutates the list in place ⚠️

new_items = ["Coat", "Backpack", "Scarf"]

# Replace the second item
new_items[1] = "Gloves"

print(new_items)   # display mutated list
`````

`````python
# 06.15 Sorting Lists: .sort() vs sorted()
# This is a **critical** concept in Python.

# .sort() method
# Mutates the list in place ⚠️

values = [5, 2, 8, 1]
values.sort()

print(values, end="\n\n")   # [1, 2, 5, 8]


# sorted() function - returns a new sorted list  
# Does not mutate the original

values = [5, 2, 8, 1]
new_values = sorted(values)

print(new_values, "\n")    # [1, 2, 5, 8]
print(values)              # [5, 2, 8, 1]  # original unchanged
`````

`````python
# 06.16 Reversing Lists: .reverse() and slicing

# Two ways:

# 1. reverse() method
# Mutates the list in place ⚠️

values = [10, 20, 30]

values.reverse()

print(f"Reverse method: {values} (mutated list)")    # [30, 20, 10]
print()

# 2. Slicing - returns a new reversed list
# Does not mutate original.

values = [10, 20, 30]
rev = values[::-1]      # negative slicing won't change underlying list

print("Reverse slicing:", rev)             # [30, 20, 10]
print(f"Original unchanged: {values}\n")   # original unchanged

print(f"Reversed: {values[::-1]}\nOriginal is still: {values}\n")
print("Type of first element:", type(values[0]))  # <class 'int'>
print()

print(values, end="\n\n")
`````

`````python
# 06.17 Counting & Finding Values
# If the value does not exist, Python raises a **ValueError**.


# .count(value) - Counts how many times a value appears.
letters = ['a', 'b', 'a', 'c', 'a']
print(letters.count('a'))     # 3


# .index(value) - Returns the first index where the value appears.
letters = ['a', 'b', 'a']
print(letters.index('a'))     # 0


# 1. Find where a specific item appears
fruits = ['apple', 'banana', 'cherry', 'apple']

first_apple_pos = fruits.index('apple')        # 0
print(first_apple_pos)

last_apple_pos = fruits.index('apple', 1)     # start searching at index 1 → 3
print(last_apple_pos)


# 2. Reorder columns by moving one column next to another
columns = ['date', 'customer_id', 'product', 'revenue', 'region']

# Move 'region' right after 'product'
pos = columns.index('product')
print(columns.index('product'))  # 2

columns.insert(pos + 1, columns.pop(columns.index('region')))  # ['date', 'customer_id', 'product', 'region', 'revenue']
print(columns)

# ⚠️ **Caution:** ⚠️
# Technically correct, but not recommended in real analytics pipelines.  
# This pattern mutates the list twice (.insert and .pop) and relies on multiple `.index()` lookups.  
# Missing or duplicated column names can cause crashes or silent data corruption.

# Safe Option 1: Reorder columns by building a new list explicitly
columns = ['date', 'customer_id', 'product', 'revenue', 'region']

new_order = [
    'date',
    'customer_id',
    'product',
    'region',
    'revenue'
]

orig_columns = columns[:]  # make a full copy of the original list
columns = new_order 

# Safe Option 2: List comprehension to reorder safely
desired_order = ['date', 'customer_id', 'product', 'region', 'revenue']
columns = [col for col in desired_order if col in columns]
# List comprehension reorder does not mutate — but it replaces which can be dangerous if the original structure matters.


# 🔬 Long Version of list comprehension (What Python Does Internally)
# Python expands the one-liner into:

orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

temp_list = []
for col in desired_order:
    if col in columns:
        temp_list.append(col)

columns = temp_list

# Same logic — the comprehension is just more compact and efficient.


# Safe Option 3: In pandas (safest)
# df = df[['date', 'customer_id', 'product', 'region', 'revenue']]
`````

`````python
# 06.18 Business Example (student-friendly)

# Weekly revenue data (in USD)
revenue = [1200, 1500, 1600, 900, 2000, 1800, 1700]

# 1. Total weekly revenue
total = sum(revenue)
print(f"Total weekly revenue: ${total:,.2f}")

# 2. Best day
best = max(revenue)
print(f"Best daily revenue: ${best:,.2f}")

# 3. Worst day
worst = min(revenue)
print(f"Worst daily revenue: ${worst:,.2f}")

# 4. Average revenue
avg = sum(revenue) / len(revenue)
print(f"Average revenue: ${avg:,.2f}")

# 5. Check if any day was above 1900
above_1900 = any(r > 1900 for r in revenue)      # any() function checks if any element in an iterable evaluates to True.


print(total, best, worst, avg, above_1900)

# This type of list analysis is foundational for:
# - financial analytics  
# - retail metrics  
# - operational dashboards  
# - time series preprocessing  
`````

# 06.15 Sorting Lists: .sort() vs sorted()

This is a **critical** concept in Python.

### **.sort() method**
- Sorts *in place*  
- Permanently changes the list  
- Returns None  

```python
values = [5, 2, 8, 1]
values.sort()

print(values)    # [1, 2, 5, 8]
```

### **sorted() function**
- Returns a *new* sorted list  
- Leaves the original unchanged  

```python
values = [5, 2, 8, 1]
new_values = sorted(values)

print(values)       # [5, 2, 8, 1]
print(new_values)   # [1, 2, 5, 8]
```

### Best practice  
> Do not sort in place until you are sure you no longer need the original data.

This is especially true in analytics where the original ordering matters.

---

# 06.16 Reversing Lists: .reverse() and slicing

Two ways:

### **1. reverse() method**  
Modifies the list in place.

```python
values = [10, 20, 30]
values.reverse()

print(values)    # [30, 20, 10]
```

### **2. slicing (non-destructive)**  
Useful for creating reversed copies.

```python
values = [10, 20, 30]
rev = values[::-1]

print(rev)       # [30, 20, 10]
print(values)    # original unchanged
```

## 🧠 Summary

### ✔ `list.reverse()`  
- Mutates original list  
- Returns `None`  
- Permanent change  

### ✔ `values[::-1]`  
- Does *not* mutate original  
- Returns a new reversed list  
- Safe for analysis  
- Best for beginners and data workflows

# 06.17 Counting & Finding Values

### **.count(value)**  
Counts how many times a value appears.

```python
letters = ['a', 'b', 'a', 'c', 'a']
letters.count('a')   # 3
```

### **.index(value)**  
Returns the **first** index where the value appears.

```python
letters = ['a', 'b', 'a']
letters.index('a')     # 0
```

If the value does not exist, Python raises a **ValueError**.

# 📦 Helper Box: How to Read Any List Comprehension (3-Step Method)

List comprehensions are one of Python’s most powerful and expressive features.  
They compress a full loop + condition + append operation into a clear, readable one-liner.

This helper teaches you how to **decode** any list comprehension using a simple 3-step process, and includes a **Safety Upgrade** showing how data professionals prevent accidental loss of original data.

---

## 🔍 General Pattern

A list comprehension always follows this structure:

```python
[new_item   for item in iterable   if condition]
```

It has **three components**:

1. **Output expression** → what goes into the new list  
2. **Iteration** → where items come from  
3. **Condition (optional)** → keep only items that pass the test  

---

# 🟦 Apply the 3-Step Method to Your Example

Your comprehension:

```python
[col for col in desired_order if col in columns]
```

### Step 1 — “For each…”  
`for col in desired_order`  
→ Loop through each column name in `desired_order`.

### Step 2 — “If…”  
`if col in columns`  
→ Only keep columns that actually exist in the current dataset.

### Step 3 — “Put into the new list…”  
`col`  
→ Place the column name into the new list as-is.

### ✔️ Final meaning (plain English)

> “Rebuild the columns list in the order of `desired_order`, but keep only those columns that appear in the existing `columns` list.”

---

# 🔬 The Long Version (What Python Does Internally)

```python
orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

temp_list = []
for col in desired_order:
    if col in columns:
        temp_list.append(col)

columns = temp_list
```

This is exactly what the comprehension does behind the scenes.

- It **does not mutate** the original list object.  
- It **does overwrite** the variable `columns` so the previous ordering is lost unless backed up.

---

# 🛡️ **Safety Upgrade: How Professionals Protect Original Data**

The long version ends with:

```python
columns = temp_list
```

This **reassigns** the variable `columns`, meaning the original ordering is gone unless saved.

In analytics, ETL, and MSBA contexts, you never want to lose original schema information.

### ✔️ Safe, production-style pattern

```python
orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

columns = [col for col in desired_order if col in orig_columns]
```

### Why this is safer:

- `orig_columns` preserves the original ordering and content  
- You can always restore or compare later  
- The comprehension now filters against the backup, not itself  
- This is how data engineers prevent silent logical bugs  

This pattern becomes essential in:

- schema reconciliation  
- feature engineering  
- pandas pipelines  
- machine learning preprocessing  
- reproducible analytics workflows  

---

# 📘 Additional Examples for Mastery

### Example A: Squaring numbers
```python
squares = [n*n for n in range(5)]
# → [0, 1, 4, 9, 16]
```

### Example B: Filtering even numbers
```python
evens = [n for n in range(10) if n % 2 == 0]
# → [0, 2, 4, 6, 8]
```

### Example C: Cleaning strings
```python
cleaned = [s.strip() for s in raw_strings if s]
```

---

# 🧩 Key Takeaways: Helper Box

- A list comprehension is built from **output → loop → optional filter**.  
- It **creates a new list**; it does **not mutate** the original list.  
- Reassigning a variable (e.g., `columns = temp_list`) can **discard the original data**.  
- In real analytics work, **always back up original data** before reassigning.  
- You should master the **long version first** — training wheels that give you full visibility.  
- Comprehensions become powerful and natural *after* you understand the underlying loop logic.

---

End of Helper Box.

# 06.30 Overview

In the real world, data is often **multi-dimensional**.  
One customer has multiple orders.  
One product has multiple attributes.  
One day has multiple transactions.

Python represents this with **nested lists** (lists inside lists).

In addition, you must understand how **copying lists** works.  
Maven introduces three copy behaviors:

1. **Assignment** (two variables, same object)
2. **shallow copy** via `.copy()`  
3. **deep copy** via `deepcopy()`  

| Operation              | What It Produces                              | Safe for Nested Lists? | Mutates Original? | Notes                                        |
| ---------------------- | --------------------------------------------- | ---------------------- | ----------------- | -------------------------------------------- |
| `b = a`                | **Second reference variable** (same object)   | No                     | Yes               | No copy; `a` and `b` are aliases             |
| `b = a.copy()`         | **Shallow copy** (new list, shared internals) | No                     | No                | Outer list copied; nested lists still shared |
| `b = copy.deepcopy(a)` | **Deep copy** (fully independent data)        | Yes                    | No                | Recursively copies all levels                |


This section explains:

- How nested lists are structured  
- How to access inner elements  
- How methods behave on nested lists  
- What “shallow” vs “deep” truly means  
- Why copying lists incorrectly creates bugs  
- How memory references work  

This is one of the MOST important sections for real analytics work.


`````python
# 06.31 What Are Nested Lists?
# A nested list is simply a list where each element is another list.

orders = [
    [19.99, 5.99],       # Customer 1
    [3.49, 9.99, 14.99], # Customer 2
    [45.00]              # Customer 3
]
`````

`````python
# 06.32 Accessing Nested List Elements
# To access *one entire sublist*, use one index.

orders = [
    [19.99, 5.99],
    [3.49, 9.99, 14.99],
    [45.00]
]

print(orders[1])       # [3.49, 9.99, 14.99]

# To access **one element inside a nested list**, use two indexes:

print(orders[1][2])    # 14.99
`````

`````python
# 06.33 Using List Methods on Nested Lists
# List methods behave normally on nested lists: append() and count() examples

nested = [
    ['a', 'b'],
    ['c', 'd', 'e'],
    ['f']
]

# append a value to the second inner list
nested[1].append('k')


# count how many 'f' values appear in the entire structure
total_f = 0
for sublist in nested:
    total_f += sublist.count('f')

print(nested)
print("Number of 'f' values:", total_f)


# Methods apply to inner lists exactly the same as they do on flat lists.
`````

`````python
# 06.34 Copying Lists: The Three Behaviors - See Helper Box (Below)

# 1. Variable Assignment  
# Creates a second reference variable, NOT a true copy.


list1 = [[10, 20], [30, 40]]
list2 = list1      # assignment, not copying

list2[0][1] = 999

# Both variables point to the same memory location (same object, two names)
# Use this ONLY when intentional
print(list1)  # [[10, 999], [30, 40]]
print(list2)  # [[10, 999], [30, 40]]


# 2. Shallow Copy (.copy())  
# Copies the outer list, but not inner lists.

original = [[1, 2], [3, 4]]
copy1 = original.copy()

copy1[0][1] = 999   # modifying the INNER list

print(original)   # [[1, 999], [3, 4]]
print(copy1)      # [[1, 999], [3, 4]]
# `.copy()` creates a new list, but its elements still reference the same inner lists.

# Important rule  
# Replacing an entire inner list creates separation.
# Modifying an element inside an inner list does not (both lists change)

# Example:

copy1[0] = ['new', 'list']   # replace entire inner list

# Now, original[0] stays unchanged.


# 3. Deep Copy (deepcopy())
# Creates a 100 percent independent copy.

from copy import deepcopy   # import deepcopy

original = [[1, 2], [3, 4]]
copy2 = deepcopy(original)

copy2[0][1] = 999

print(original)   # unchanged, [[1, 2], [3, 4]]
print(copy2)      # changed, [[1, 999], [3, 4]]
`````

`````python
# 06.35 Business Example: Customer Order Copies
# This demonstrates the dramatic difference between shallow and deep behavior.

from copy import deepcopy   # import deepcopy

# Nested data representing customer orders
multi_orders = [
    [15.99, 23.50],
    [9.99, 14.25, 7.50],
    [60.00]
]

# Shallow copy
shallow = multi_orders.copy()

# Deep copy
deep = deepcopy(multi_orders)

# Modify inner structures
shallow[1][0] = 888   # affects both shallow and original
deep[2][0] =    777   # affects only deep copy

print("Original:", multi_orders)
print("Shallow:", shallow)
print("Deep:", deep)
`````

`````python
# 06.36 Assignment — Nested Lists & Copying  


# 🛒 Multi-Order Customers  

orders_C00001 = [179.94, 29.98, 99.99]
orders_C00004 = [15.98, 119.99]
orders_C00006 = [24.99, 24.99]
orders_C00008 = [649.99, 99.99]
orders_C00010 = [59.99, 399.97]


# ⭐ Create a nested list called `vip_list`
# Each element of `vip_list` will be one of the customer transaction lists.

vip_list = [
    orders_C00001,
    orders_C00004,
    orders_C00006,
    orders_C00008,
    orders_C00010
]

print(vip_list)
# [179.94, 29.98, 99.99], [15.98, 119.99], [24.99, 24.99], [649.99, 99.99], [59.99, 399.97]


# 🔎 Part 1 — Get the 2nd and 3rd transactions for the customer with 3 orders  

# Customer **C00001** made three transactions.  
# We will retrieve their second and third orders in two different ways.

# Option A: Direct indexing (when you already know the position)

# vip_list[0] is C00001's list
second_and_third = vip_list[0][1:]   # slice from index 1 to the end
print(second_and_third)              # [29.98, 99.99]


# Option B: Let Python find the right customer dynamically  
# This is more powerful and prepares you for data science work
# where you often want the code to discover the right items.

# Loop through each inner list in vip_list
for orders in vip_list:
    # Look for the customer who made exactly three orders
    if len(orders) == 3:
        print("Second and third orders:", orders[1], orders[2])


# 🧩 Helper Box — How to Search Nested Lists with a Loop  

# General pattern:

# for sublist in nested_list:
    # if CONDITION(sublist):
        # Use sublist here
#         ...

# For this assignment:

# - `nested_list` is `vip_list`  
# - Each `sublist` is a list of orders for one customer  
# - Our condition is `len(sublist) == 3`  
# - Once we find that sublist, we index into it with `[1]` and `[2]`  

# This pattern is very important for:

# - Filtering nested structures  
# - Working with JSON or API responses  
# - Preparing for Pandas and SQL style filtering  


# 💰 Part 2 — Create a Revenue Adjusted Copy  
# We now need a safe copy of `vip_list` that we can modify without touching the original data.  
# Because `vip_list` is a nested list, we must use `deepcopy`.

from copy import deepcopy   # import deepcopy

revenue_adjusted_list = deepcopy(vip_list)

print("Copy before changes:", revenue_adjusted_list)
print("Original vip_list:", vip_list)


# ✏ Adjust C00004's transactions in the copy only  

# Suppose customer C00004 should have their transactions corrected to zero.  
# We will replace their list in the copy and then verify that `vip_list` is unchanged.


# C00004 is at index 1 in both vip_list and revenue_adjusted_list
revenue_adjusted_list[1] = [0.0, 0.0]

print("Adjusted copy:", revenue_adjusted_list)
print("Original vip_list still:", vip_list)

# You should see that:

# - `revenue_adjusted_list` now has `[0.0, 0.0]` for the second customer  
# - `vip_list` still shows `[15.98, 119.99]` for the second customer  


# ✅ Key Takeaways  

# - A **nested list** is a list where each element can itself be a list.  
# - You can access nested values with chained indexing, for example `vip_list[0][1]`.  
# - Loops plus conditions let Python **find the right inner list for you**, rather than hard-coding positions.  
# - Assignment (`b = a`) does not copy; it creates a second reference.  
# - `.copy()` is a **shallow copy** and is **not safe** for nested lists.  
# - `deepcopy()` creates a fully independent copy, so changes in the copy do not affect the original.  
`````

# 06.34 Copying Lists: The Three Behaviors - See Helper Box (Below)

Copying lists in Python is subtle.  
This is where many beginners (and some professionals) make mistakes.

There are **three** main behaviors:

---

## 1. Variable Assignment  
### Creates a **second reference variable**, NOT a true copy.

```python
list1 = [[10, 20], [30, 40]]
list2 = list1      # assignment, not copying

list2[0][1] = 999

print(list1)  # [[10, 999], [30, 40]]
print(list2)  # [[10, 999], [30, 40]]
```

### Why this happens

Both variables point to **the same memory location**.

### Memory diagram

```
list1 ───► [ [10, 20], [30, 40] ]
           ▲
list2 ─────┘  (same object, two names)
```

**Any change to one affects the other.**

Use this ONLY when intentional.

---

## 2. Shallow Copy (.copy())  
### Copies the outer list, but not inner lists.

```python
original = [[1, 2], [3, 4]]
copy1 = original.copy()

copy1[0][1] = 999   # modifying the INNER list

print(original)   # [[1, 999], [3, 4]]
print(copy1)      # [[1, 999], [3, 4]]
```

### Why this happens

`.copy()` creates a new list, **but its elements still reference the same inner lists**.

### Memory diagram

```
original:   [  L1,     L2  ]
               │       │
copy1:      [  L1,     L2  ]
```

Where `L1` and `L2` are the same inner list objects.

### Important rule  
> Replacing an entire inner list creates separation.  
> Modifying an element inside an inner list does not.

Example:

```python
copy1[0] = ['new', 'list']   # replace entire inner list
```

***Now, original[0] stays unchanged.***

---

## 3. Deep Copy (deepcopy())

### Creates a **100 percent independent copy**.

```python
from copy import deepcopy   # import deepcopy

original = [[1, 2], [3, 4]]
copy2 = copy.deepcopy(original)

copy2[0][1] = 999

print(original)   # unchanged, [[1, 2], [3, 4]]
print(copy2)      # changed, [[1, 999], [3, 4]]
```

### Memory diagram

```
original:   [ L1 ,     L2  ]
copy2:      [ L1',     L2' ]
```

Where L1 and L2 are completely separate new lists.

### When to use deepcopy

- Complex nested structures  
- Multi-level JSON-like data  
- When list mutability must not affect the original  
- Data cleaning pipelines that require preservation

# 🧰 Helper Box: **Shallow Copy vs Deep Copy — The Only Rule You Need**

This helper gives you the *correct, professional-level mental model* for copying data structures in Python.  
Most confusion about `.copy()` comes from not understanding how references behave inside nested structures.

This box will stay valid for your entire MSBA program, pandas work, and future data science.

---

## 🟦 1. The One Rule That Solves Everything

### ✅ **`.copy()` is safe only for *flat* lists.**  
### ❌ **`.copy()` is NOT safe for nested data structures.**  
### ⭐ **Use deep copy only when you *must* — and avoid nested structures in analytics whenever possible.**

That’s the rule.

Let’s expand it.

---

## 🟩 2. What `.copy()` Actually Does (Shallow Copy)

```python
new_list = old_list.copy()
```

This copies **only the outer list**.

It does **NOT** copy the objects inside it.

### ✔️ Safe:
```python
[1, 2, 3, 4]
['a', 'b', 'c']
[True, False, True]
```

These contain simple, immutable elements → no shared inner state.

### ❌ Not Safe:
```python
[[1, 2], [3, 4]]
[ ['a', 'b'], ['c', 'd'] ]
[ [1, 2], 99, ['x', 'y'] ]
```

These contain **nested lists**, which are still shared between copies.

---

## 🟥 3. Visual Mental Model You Should Use

### Example List
```python
list1 = [[1, 2], [3, 4]]
list2 = list1.copy()
```

### Reality (shallow copy)
```
list1 ───► [ A , B ]     # where A is [1, 2], and B is [3, 4]  ⚠️
list2 ───► [ A , B ]     # [[1, 2], [3, 4]]  ⚠️

A = [1, 2]   ← shared
B = [3, 4]   ← shared

        ┌───────────────┐
list1 ─►│   index 0 ─────┼──► A ([1,2])
        │   index 1 ─────┼──► B ([3,4])
        └───────────────┘

        ┌───────────────┐
list2 ─►│   index 0 ─────┼──► A ([1,2])
        │   index 1 ─────┼──► B ([3,4])
        └───────────────┘

```

Both lists point to the **same** inner lists.

### Mutation ***inside*** A or B affects BOTH:
```python
list1[0][1] = 999

list1[0] ----> A <---- list2[0]  ⚠️
```
We are mutating A, because list1[0] refers to A.

But since list2[0] ALSO refers to A, we see the mutation in both

Now *both* `list1` and `list2` show:
```
[1, 999]
```

⚠️ This is why shallow copy is unsafe for nested data.

---

## 🟦 4. Why Replacing an Entire Nested List *Doesn't* Affect the Copy

```python
list1[0] = ['x', 'y', 'z']
```

This does *not* mutate the inner list.  
It **rebinds** index 0 to a *new* list.

```
list1 ─► ['x','y','z']   (new)
list2 ─► [1, 2]          (old)

list1 ───► [ NEW_LIST , B ]
list2 ───► [    A     , B ]
```

So the effect does *not* propagate.

This explains why your instructor’s example behaved differently in the two cases.

---

## 🟧 5. The Only Time You Use `deepcopy`

```python
from copy import deepcopy
new_list = deepcopy(old_list)
```

Use `deepcopy` only when:

- You must work with nested Python objects  
- And you must guarantee isolation  
- And flattening is not possible  

**Data analysts rarely use deepcopy because they rarely use nested Python lists at scale.**  
Instead, they use:

- pandas DataFrames  
- NumPy arrays  
- flat lists  
- relational tables  

These structures avoid arbitrary nesting → deepcopy becomes unnecessary.

---

## 🟦 6. Safe Practice Rules (Adopt These)

### ✅ Rule 1: Avoid nested Python lists in analytics  
Flatten early, or convert to pandas.

### ✅ Rule 2: If nesting is unavoidable, assume `.copy()` is unsafe  
Do not trust shallow copies for nested data.

### ✅ Rule 3: Mutation inside a nested object propagates  
Changing elements inside an inner list changes all shallow copies.

### ✅ Rule 4: Rebinding a nested list does NOT propagate  
Replacing the entire inner list with a new one isolates the change.

### ✅ Rule 5: Use `deepcopy` **only** when nested structures cannot be avoided  
But aim to design your code so this is rare.

---

## 🟩 7. Quick Reference Table

| Operation | Copies Outer List? | Copies Inner Lists? | Safe for Nested Data? |
|----------|---------------------|----------------------|------------------------|
| `new = old.copy()` | ✅ Yes | ❌ No | ❌ No |
| `new = old[:]` | ✅ Yes | ❌ No | ❌ No |
| `new = list(old)` | ✅ Yes | ❌ No | ❌ No |
| `new = deepcopy(old)` | ✅ Yes | ✅ Yes | ⭐ Yes |

---

## 🟦 8. Helper Box Summary

- `.copy()` copies the **container**, not the **contents**.  
- Nested objects remain **shared** and therefore **dangerous**.  
- Replace inner lists (rebinding) → safe  
- Modify inside an inner list → unsafe  
- Deep copy solves the problem but should be unnecessary in well-designed analytics code.  
- Flat structures avoid all these issues — which is why pros use them.

---

End of Helper Box.

# 06.40 Overview

A **tuple** is another type of sequence in Python, similar to a list but with one major difference:

# ✔ Tuples are **immutable**  
Once a tuple is created, **it cannot be changed**.  
You cannot add, delete, or replace items.

Tuples are incredibly important in analytics work because they:

- ensure data cannot be accidentally changed  
- take up less memory than lists  
- run faster for some operations  
- appear naturally as function outputs  
- prevent logical errors in pipelines

---

# Tuples vs Lists: When to Choose Which

| Use List When... | Use Tuple When... |
|------------------|-------------------|
| You need to modify data | Data must stay fixed |
| Storing transaction sets | Storing coordinates |
| Cleaning inconsistent values | Returning multiple outputs |
| Building datasets dynamically | Locking configuration values |
| Data is evolving | Data is final and stable |

### Rule of Thumb  
> If you do not need mutability, choose a tuple.  

This protects your logic downstream.

`````python
# 06.41 Creating Tuples
# Tuples use parentheses instead of square brackets.


# Standard tuple
sizes = ('S', 'M', 'L')

# Tuple with mixed types
info = ('Alice', 42, True)

# Empty tuple
empty = ()

# Converting an iterable into a tuple
letters = tuple('abcd')      # ('a', 'b', 'c', 'd')


# Mini-diagram: tuple structure


sizes = ('S', 'M', 'L')

# Index:       0     1     2
# Value:      'S'   'M'   'L'

# Tuples use the same indexing rules as lists.
`````

`````python
# 06.42 One-Element Tuples  

# A tuple with one element **must** include a trailing comma.

one = ('A',)     # This is a tuple
not_tuple = ('A') # This is just the string 'A'

# Python interprets parentheses alone as grouping, not as tuple creation.
`````

`````python
# 06.43 Accessing Tuple Elements
# Tuple indexing works exactly like list indexing:

colors = ('red', 'green', 'blue')

print(colors[0])    # ('red')           → created but not saved
print(colors[-1])   # ('blue')          → created but not saved
print(colors[1:3])  # ('green', 'blue)  → created but not saved

new_colors = colors[1:3]  # saved version
print("\n", new_colors)
print(colors)             # original intact


# - Slicing returns a new tuple object:
# - The original tuple remains unchanged.
# - You can assign the slice to a variable to keep it (just like a list slice).

# Because tuples are immutable:
# - You *cannot* modify the original tuple  
# - You *cannot* modify the slice  
# - You *cannot* use slice assignment  
# - The new tuple is completely isolated and safe  
`````

`````python
# 06.44 Core Concept: Tuple Immutability

# Tuples cannot:

# - append items  
# - remove items  
# - insert items  
# - reassign items  

# Any attempt to modify them will lead to a TypeError.


t = (1, 2, 3)

# t[1] = 999       # TypeError: 'tuple' object does not support item assignment
# t.append(4)      # AttributeError
`````

`````python
# 06.45 Why Use Tuples?


# ✔ 1. Tuples use less memory

import sys

list_sample = [1, 2, 3, 4, 5]
tuple_sample = (1, 2, 3, 4, 5)

print(sys.getsizeof(list_sample))
print(sys.getsizeof(tuple_sample))


# ✔ 2. Tuples are faster than lists  

import timeit

print(timeit.timeit("x = (1,2,3,4,5)", number=1_000_000))
print(timeit.timeit("x = [1,2,3,4,5]", number=1_000_000))


# ✔ 3. Tuples prevent accidental changes  

# Example:

# Coordinates shouldn't change
location = (33.12, -117.28)


# ✔ 4. Functions often return tuples  
# This includes built-in functions and library functions

# Example:
# Define a function and give me a list of numbers
# I am naming that list of numbers "values" inside my function

def get_stats(values):
    return (min(values), max(values), sum(values) / len(values))

stats = get_stats([10, 20, 30])       # parentheses make it a tuple
print("\n", stats)       # (10, 30, 20.0)

# Unpacking the tuple:
minimum, maximum, average = stats
print(minimum, maximum, average)

# Tuple unpacking is the natural way Python returns multiple values cleanly.
`````

`````python
# 06.46 Tuple Unpacking  
# Used heavily in real analytics workflows


transaction1, transaction2, transaction3 = (29.99, 15.99, 9.99)

print(transaction1)    # 29.99
print(transaction2)    # 15.99
print(transaction3)    # 9.99
`````

`````python
# 06.48 Assignment: Sales Tax  

# Customer C00001 made these three transactions
C00001 = [29.99, 15.99, 9.99]  # List with square brackets

# Step 1: Create a tuple from the list
transactions = tuple(C00001)   # tuple() is a built-in command that converts anything into a tuple
print(transactions)            # (29.99, 15.99, 9.99)


# Step 2: Unpack into variables
t1, t2, t3 = transactions
print("Unpack Tuple:", t1, t2, t3)

# Step 3: Apply sales tax (8 percent)
tax_rate = 0.08
tax1 = round(t1 * tax_rate, 2)
tax2 = round(t2 * tax_rate, 2)
tax3 = round(t3 * tax_rate, 2)

print("Sales tax for each transaction:")
print(tax1, tax2, tax3)

# This preserves the logical structure while improving clarity.


# 06.49 Additional Business Example: Immutable Product Records
# Tuples are excellent for data that must not change.

product_record = (
    "SKU12345",       # product ID
    "Winter Gloves",  # name
    29.99,            # price
    True              # in stock
)

# product_record[2] = 35.99   # would raise an error
`````

`````python
# --------------------------------------------------------------
# 🔧 Helper: Tuple Inspector (Index Map + Mutability Checker)
# --------------------------------------------------------------

def show_tuple_info(tpl, show_negative=True, show_types=True):
    """
    Display a structured index map for a tuple, including:
      - positive and negative indices
      - value at each index
      - type of each element
      - mutability warning (e.g., tuple containing lists)
    
    Parameters
    ----------
    tpl : tuple
        The tuple to inspect.
    show_negative : bool, default True
        Whether to show negative indices.
    show_types : bool, default True
        Whether to show type information for each element.
    
    Usage
    -----
    show_tuple_info((1, 2, 3))
    show_tuple_info((1, [2, 3], (4, 5)))
    show_tuple_info(("a", "b", "c"), show_negative=False)
    """

    # ----------------------------------------------------------
    # 1. Validate input
    # ----------------------------------------------------------
    if not isinstance(tpl, tuple):
        raise TypeError(
            f"show_tuple_info expects a tuple, got {type(tpl).__name__}"
        )

    n = len(tpl)
    print(f"Tuple length: {n}")
    print()

    # ----------------------------------------------------------
    # 2. Build header
    # ----------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    header_cols.append("mutable?")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += f"{'value':>12} {'mutable?':>10}"

    print(header)
    print("-" * max(len(header), 40))

    # ----------------------------------------------------------
    # 3. Detect mutability of each element
    # ----------------------------------------------------------
    def is_mutable(obj):
        """Return True if object is a known mutable Python type."""
        return isinstance(obj, (list, dict, set, bytearray))

    # ----------------------------------------------------------
    # 4. Print each element with indices and mutability flag
    # ----------------------------------------------------------
    for i, val in enumerate(tpl):
        neg = i - n if show_negative else None
        type_name = type(val).__name__ if show_types else ""
        mutable_flag = "yes" if is_mutable(val) else "no"

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{repr(val):>12} {mutable_flag:>10}"

        print(row)

    # ----------------------------------------------------------
    # 5. Summary warnings (very helpful in practice)
    # ----------------------------------------------------------
    print("\nSummary:")
    if any(isinstance(v, list) for v in tpl):
        print("⚠ Warning: This tuple contains one or more lists.")
        print("  → The tuple is immutable, but the inner lists ARE mutable.")
        print("  → Changing an inner list will change all references to it.\n")
    else:
        print("✓ No mutable inner objects detected. Tuple is fully immutable.\n")


# ------------------------------
# Example:
# ------------------------------
show_tuple_info((1, 2, (3, 4)))
show_tuple_info((1, [2, 3], 4))
`````

```python
tax2 = round(t2 * tax_rate, 2)
tax3 = round(t3 * tax_rate, 2)

print("Sales tax for each transaction:")
print(tax1, tax2, tax3)
```

This preserves the logical structure while improving clarity.

Tuples are excellent for data that must not change.

```python
    "SKU12345",       # product ID
    "Winter Gloves",  # name

)

# product_record[2] = 35.99   # would raise an error
```

You are protected from unintended modification.

- Tuples are immutable sequences; once created, they cannot be changed.  
- They use **parentheses** and require a trailing comma for one-element tuples.  
- Tuples are more memory-efficient and faster than lists.  
- They prevent accidental changes and improve code safety.  
- Many functions return multiple values packed into tuples.  
- Unpacking enables fast, readable extraction of values.  
- Tuples are ideal for fixed data such as configuration values, coordinates, and stable metadata.

# 🟦 Helper Box — Slicing: Lists vs. Tuples  
Understanding how slicing behaves for **mutable** vs **immutable** sequence types is essential.  
This helper clarifies exactly what happens under the hood — and why tuple slicing is so safe and predictable.

---

## ❗ First, a universal rule (often overlooked)
> **Slicing always *creates* a new sequence object, but it does *not* save it unless you assign it to a variable.**

```python
numbers = [10, 20, 30, 40]
numbers[1:3]        # slice created, used, discarded
subset = numbers[1:3]  # slice saved
```

The same is true for tuples.

This is important: slicing alone never mutates anything — it just produces a temporary result unless you store it.

---

## ✅ 1. Slicing a List  
### What happens?

```python
lst = [10, 20, 30, 40]
slice_lst = lst[1:3]     # [20, 30]
```

- **Slicing always creates a new list object.**
- The original list is unchanged.
- Because lists are **mutable**, you can modify the slice:

```python
slice_lst[0] = 999      # [999, 30]
# lst remains [10, 20, 30, 40] — completely unaffected
```

### Key property  
Even though slicing is safe, **slice assignment** can modify the original list:

```python
lst[1:3] = [777, 888]   # mutates original list
```

This is a powerful feature — but also a source of bugs if used carelessly.

---

## ✅ 2. Slicing a Tuple  
### What happens?

```python
colors = ('red', 'green', 'blue')
subset = colors[1:3]      # ('green', 'blue') → created but not saved

new_colors = colors[1:3]  # saved version

```

- **Slicing creates a new tuple object.**
- The original tuple remains unchanged.
- You can assign the slice to a variable to keep it (just like a list slice).

### But the critical difference:
Because tuples are **immutable**:

- You *cannot* modify the original tuple  
- You *cannot* modify the slice  
- You *cannot* use slice assignment  
- The new tuple is completely isolated and safe  

Slicing tuples is simpler and safer because there is *no* possibility of mutation anywhere.

---

## 🧠 3. Why the Behavior Is Different?

| Operation                    | Lists (Mutable)                      | Tuples (Immutable)                     |
|-----------------------------|---------------------------------------|----------------------------------------|
| Slicing                     | Creates a **new list**                | Creates a **new tuple object**         |
| Assigning back to a slice  | Allowed (mutates original)            | Not allowed (TypeError)                |
| Modifying the slice itself  | Allowed                                | Not allowed                            |
| Risk of unintended mutation | Medium                                 | Zero                                   |

---

## 🏁 4. The Key Concept in One Sentence  
> **Slicing always returns a new sequence object — but only lists allow slice assignment that can mutate the original. Tuples never allow mutation, so slicing is always safe.**

---

## ⭐ 5. The Mental Model (recommended for mastery)

Think of slicing as:

### **“Copying a portion of the sequence into a new container.”**

- For lists → a new *list container*  
- For tuples → a new *tuple object*  

And because tuples cannot change, this new object is **fully independent** and **risk-free**.

---

## ✔ Quick Verification

```python
t = (1, 2, 3, 4)
s = t[1:3]

print(s)       # (2, 3)
print(t)       # (1, 2, 3, 4)
```

Both remain intact. Nothing can mutate either one.

---

## 📌 Final takeaway  
Whether slicing a list or tuple:

- It **never** changes the original sequence  
- It **always** produces a new object  
- It **must be assigned** if you want to keep it  
- Only lists support slice assignment  
- Tuples are 100% mutation-proof  

---

End of Helper Box.

# 06.60 Overview

A **range** is a very special type of sequence in Python.  
Unlike lists or tuples, a range does **not** store all values in memory.  
Instead, it **generates** values on demand using a start, stop, and step.

Ranges are:

- memory-efficient  
- fast  
- commonly used in loops (next chapter!)  
- ideal for generating clean sequences of integers  
- invaluable for indexing, batching, and slicing  

---

# 🟦 Helper Box — Universal Range Index Map (`show_range_index_map`)

This helper is the **range-specific cousin** of your `show_list_index_map`.  
It helps you *see* how `range(start, stop, step)` expands into actual values, including:

- forward ranges  
- backward (negative-step) ranges  
- how indices and negative indices line up with values  

This is especially helpful for:

- understanding `stop` being **excluded**  
- understanding **negative steps**  
- mentally mapping `range()` to list indices

# 06.69 Summary (Part 5 — Ranges)

- `range()` creates a memory-efficient sequence of integers.  
- Ranges are defined by start, stop, and step values.  
- The stop value is **non-inclusive**, just like slicing.  
- Converting to `list()` or `tuple()` reveals the actual generated values.
- Although, any range object can be converted into any other sequence type that accepts an iterable.
- Ranges are ideal for loops, counting, indexing, batching, and generating IDs.  
- Negative step values allow backward ranges.  
- Ranges prepare you directly for Chapter 07 (Loops), where they are used constantly.


`````python
# 06.61 Creating Ranges
# Ranges are created using the built-in `range()` function.

# The full form is: range(start, stop, step)

# - **start** (optional): where the sequence begins  
# - **stop**: where the sequence ends (non-inclusive)  
# - **step** (optional): how much to increment each time  


# Basic range: 0 to 4
r1 = range(5)

# Range with start and stop
r2 = range(2, 7)

# Range with start, stop, and step
r3 = range(0, 20, 5)

print(list(r1))   # [0, 1, 2, 3, 4]
print(list(r2))   # [2, 3, 4, 5, 6]
print(list(r3))   # [0, 5, 10, 15]

# ASCII mini-diagram

# range(2, 7)
# start → 2
# stop  → 7   (not included)
# values → [2, 3, 4, 5, 6]

# This mirrors list slicing behavior, where the stop value is also non-inclusive.
`````

`````python
# ---------------------------------------------------------
# 06.62 Ranges Do Not Directly Display Their Values
# ---------------------------------------------------------

# When printed, a range shows its definition, not the actual integers:
print(range(5), end="\n\n")  # prints object parameters only,"range(0, 5)"


# --------------------------------------------------------------------
# To retrieve and see the actual values, convert to a list or a tuple:
# --------------------------------------------------------------------

print(list(range(5)))    # [0, 1, 2, 3, 4]
print(tuple(range(5)))   # (0, 1, 2, 3, 4)

# -------------------------------------------------------------------------------
# Visualization example: show the index positions of a string
# This is purely for demonstration; not common in real code to turn into a tuple
# -------------------------------------------------------------------------------

print("\n", tuple(range(len('Hello'))))

# Breakdown: generating all valid indices of the string, then turning them into a tuple.
# len('Hello') → 5
# range(5) → range(0, 5)
# tuple(range(5)) → (0, 1, 2, 3, 4)
# print() to display

print("\n", list(range(len('Hello'))))
`````

`````python
# 06.63 Negative Steps and Reversed Ranges

r = range(10, 0, -2)
print(list(r))
# [10, 8, 6, 4, 2]


# ASCII diagram

range(10, 0, -2)
# start = 10
# stop = 0  (excluded)
# step = -2
# values → [10, 8, 6, 4, 2]

# Backward ranges are extremely helpful when:

# - stepping backward through indices  
# - reverse-iterating through sequences  
# - building countdown timers  
# - analyzing time-based data in reverse order
`````

`````python
# 🔧 Helper: `show_range_index_map(rng, show_negative=True)`


def show_range_index_map(rng, show_negative=True):
    """
    Print a table of indices and values for a Python range object.

    Parameters
    ----------
    rng : range
        The range object to inspect.
    show_negative : bool, default True
        If True, show the negative index for each element
        (as if the range were a sequence like a list or tuple).

    Usage
    -----
    show_range_index_map(range(5))
    show_range_index_map(range(10, 0, -2))
    """
    # ---------------------------------------------------------
    # 1. Validate input type
    # ---------------------------------------------------------
    if not isinstance(rng, range):
        raise TypeError(
            f"show_range_index_map expects a range, got {type(rng).__name__}"
        )

    # ---------------------------------------------------------
    # 2. Basic range properties
    # ---------------------------------------------------------
    n = len(rng)
    start = rng.start
    stop = rng.stop
    step = rng.step

    print(f"Range: range({start}, {stop}, {step})")
    print(f"Length: {n}")
    print()

    if n == 0:
        print("This range is empty (no values generated).")
        return

    # ---------------------------------------------------------
    # 3. Build header
    # ---------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    header_cols.append("value")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "value" in header_cols:
        header += f"{'value':>6}"

    print(header)
    print("-" * max(len(header), 20))

    # ---------------------------------------------------------
    # 4. Iterate over the range and print index/value mapping
    # ---------------------------------------------------------
    for i, val in enumerate(rng):
        # Negative index follows standard Python convention:
        # last element → -1, second to last → -2, etc.
        neg_index = i - n if show_negative else None

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        row += f"{val:>6}"
        print(row)


show_range_index_map(range(10, 0, -2))
show_range_index_map(range(20))               # forward range
show_range_index_map(range(50, 20, -5))       # backward range
show_range_index_map(range(6), show_negative=False)

example_range = range(10, 0, -2)
show_range_index_map(example_range)
`````

`````python
# 06.64 Practical Range Patterns
# Common real-world range uses:

# All even numbers from 0 to 20
evens = list(range(0, 21, 2))
print(evens)

# All odd numbers from 1 to 20
odds = list(range(1, 21, 2))
print(odds)

# First 10 integer IDs starting at 100
ids = list(range(100, 111))
print(ids)

# Ranges are the foundation for constructing loops, counters, and index sequences.
`````

`````python
# 06.65 Memory Efficiency Demonstration
# Ranges do NOT store all values in memory; a list does.

import sys

print(sys.getsizeof(range(1_000_000)))
print(sys.getsizeof(list(range(1_000_000))))


# 06.66 Business Example: Batch Processing with Ranges
# Ranges are ideal for chunking data or assigning customer IDs.

# Generate order numbers for the first 50 orders
order_nums = list(range(1, 51))

print(order_nums[:10])   # show the first 10


# Or for selecting specific step intervals:
# Every 5th order should be reviewed for quality control

review_orders = list(range(5, 51, 5))
print(review_orders)
`````

# Chapter 07 — Loops  
## Section 7.01 — Loop Basics  

### Introduction
In Python, a **loop** is a controlled repetition structure: a block of code that executes multiple times until a stopping condition is met.  
For data analytics, loops let you process sequences, clean datasets, transform values, and automate iterative logic.

Python supports two primary loop types:

1. **For Loops**  
   Run a ***specific number of times***, typically once per element in an iterable such as a list, tuple, string, or range object.  
   Use a for loop when the iteration count is known in advance or when you want to examine or transform each element of a dataset.  
   For loops may iterate:
   - **directly over values** (item-based looping),  
   - **over indices** using `range()` when positional control or alignment across multiple sequences is required, or  
   - **over index–value pairs** using `enumerate()` when both position and clarity are needed.  
   *Run a specified number of times.*

2. **While Loops**  
   Run ***as long as a logical condition remains True***.  
   Use a while loop when you do **not** know in advance how many iterations are needed and the loop must respond to changing state.  
   *Run until a logical condition is met.*

This chapter focuses on building mastery for both loop types, integrating your **Helper Toolkit** to visualize execution steps, debug iterative logic, and strengthen intuition for real analytics workflows.

---

# 🟦 Helper Box — Core Loop Visualizers  
These helpers provide foundational insight into how loops actually *iterate* and how internal state evolves during execution.  
They serve the same purpose for Chapter 7 that the **Index Map** and **Slice Visualizer** served for Chapter 6.

Together, they make abstract loop mechanics concrete, predictable, and fully observable.

---

# Preview — Loop Basics

- A **loop** repeats instructions until a **count-based condition** (for loop) or a **logical condition** (while loop) signals completion.  
- **For loops** are used when the number of iterations is known or when the task is “one pass per element.”  
  They may iterate directly over values, over **indices** when positional control is required, or over **index–value pairs** using `enumerate()` for improved clarity and safety.  
- **While loops** provide greater flexibility but must be carefully designed to avoid infinite loops.  
- This section introduces the core loop helpers:  
  - **`loop_trace_for()`** — traces item-by-item for-loop iteration and execution order  
  - **`enumerate_visualizer()`** — makes index–value pairing explicit when using `enumerate()`  
  - **`safe_while_trace()`** — safely traces while-loop state evolution and prevents infinite loops
  - **``while_timeline()``** - advanced helper logs every state value for later visualization or tabular analysis 
  - **`deep_loop_state_visualizer()`** — traces each internal step of a list-processing loop, including variable assignment, calculation execution, each `.append()` operation, and the evolving state of the accumulator container
  - **`validate_alignment`** - Validate iterable alignment before looping. (Length alignment is required for index-based looping)
  - **`sync_lists()`** — synchronized iteration across multiple equal-length iterables 
- Understanding iteration mechanics — including when and how position matters — is essential for dataset traversal, data cleaning, transformation workflows, simulation modeling, and algorithmic problem-solving.


`````python
# ------------------------------------------------------------------
# Understanding Loops Through Examples
# Example 1 — Rewriting a repetitive calculation using loops  
# ------------------------------------------------------------------


# Below is the non-loop approach: writing a separate line for each list element.

exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

# Inefficient repetition (not scalable)
euro_list = [
    round(usd_list[0] * exchange_rate, 2),
    round(usd_list[1] * exchange_rate, 2),
    round(usd_list[2] * exchange_rate, 2),
    round(usd_list[3] * exchange_rate, 2),
    round(usd_list[4] * exchange_rate, 2),
]

euro_list

# Problem if you have hundreds or thousands of prices, manually writing repeated expressions becomes impossible.


# ------------------------------------------------------------------
# Understanding Loops Through Examples
# Example 2 — Applying a for loop
# ------------------------------------------------------------------

# Generic syntax - `for` indicates for loop
iterable = "Dawn"
for i in iterable:
     print(i)

# Given Data:
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

euro_list = []                                    # Start with an empty list container

for price in usd_list:                            # for loop: element in iterable
    euro_value = round(price * exchange_rate, 2)  # assign variable: each element from the iterable is * by the loop-invariant scalar exchange_rate
    euro_list.append(euro_value)                  # .append() empty list container with each converted price
    
print()
print(euro_list, end="\n\n")


# This loop is conceptually equivalent to five function calls, one per list element.  
# But Python handles all iteration mechanics for you.
`````

`````python
# -------------------------------------------------------------
# Looping Mechanics
# How to use a single loop to derive multiple aligned outputs from one input sequence.
# -------------------------------------------------------------
# item-based for loops (no indices yet)
# deterministic iteration (known list length)
# accumulator lists
# stepwise transformation
# parallel output lists that stay aligned by construction

subtotals = [15.98, 899.97, 799.97, 117.96, 5.99, 599.99, 24.99, 1799.94, 99.99]
taxes = []
totals = []

for subtotal in subtotals:
    tax = round(subtotal * 0.08, 2)
    total = round(subtotal + tax, 2)
    taxes.append(tax)
    totals.append(total)

print(taxes)
print(totals)
`````

# Chapter 07 — Loops  
## Section 7.02 — For Loops  

### Overview
A **for loop** iterates over an *iterable* (list, tuple, string, set, dictionary, or range).  
Python automatically handles indexing and sequence progression, allowing analysts to focus on transformation logic rather than iteration mechanics.

Use a for loop when:

- You know **how many iterations** you need.  
- You are iterating directly over each element in a dataset.  
- You want clear, predictable, index-safe iteration.  
- You want to avoid accidental infinite loops.

Syntax describes the standard pattern:

```
for item in iterable:
    do_something_with(item)
```

Python internally executes:

1. `item = iterable[0]`  
2. `item = iterable[1]`  
3. `item = iterable[2]`  
n. until the iterable is exhausted.

---

## 1. Looping Over Items  
This is the **most Pythonic** and most common pattern in data analytics.

### Example — Print each letter in a string  

```python
word = "Hello"

for letter in word:
    print(letter)
```

Internally this becomes:

- `letter = 'H'`  
- `letter = 'e'`  
- `letter = 'l'`  
- `letter = 'l'`  
- `letter = 'o'`


`````python
# -------------------------------------
# 1. Looping Over Items
# -------------------------------------
# This is the most Pythonic and most common pattern in data analytics.
# Use when:
# - you only need the value
# - position does not matter

# Item-Based Loop Pattern:

items = "Hello"
for i in items:
    print(i)
print()


# 🟦 Helper Expansion — `enumerate_items()`

def enumerate_items(iterable):
    """
    Diagnostic helper: prints each item as Python would assign it in a for loop.
    Mirrors the step-by-step iteration shown in Maven pages 7–9.
    """
    print("TRACE: Looping Over Items\n--------------------------")
    for idx, item in enumerate(iterable):
        print(f"Iteration {idx}: item = {item!r}")
        

# Example usage:

enumerate_items("Hello")
`````

`````python
# -------------------------------------
# Exercise: For loops (step-by-step)
# -------------------------------------

# Intentionally broken list (missing comma) to demonstrate a "quiet bug"
customer_ratings = [
    "5 stars", "3 stars", "5 stars", "4 stars", "4 stars"
    "1 star", "4 stars", "5 stars", "2 stars", "5 stars"
]

print("Stage 1 — Inspect the list (notice the quiet bug)")
print("Length:", len(customer_ratings))
print("Element near the bug:", customer_ratings[4])  # likely "4 stars1 star"
print()

# Demonstrate how to extract the first character and convert it to integer
rating_num = int(customer_ratings[0][0])   # works because ratings are 1–5 (not double digits)
print("Stage 2 — Extract numeric rating from the first review")
print("First numeric rating:", rating_num)
print()

print("Stage 3 — Print each rating string (shows data structure)")
for rating_str in customer_ratings:
    print(rating_str)
print()

# Corrected list
customer_ratings = [
    "5 stars", "3 stars", "5 stars", "4 stars", "4 stars",
    "1 star", "4 stars", "5 stars", "2 stars", "5 stars"
]

print("Stage 4 — Corrected list")
print("Length:", len(customer_ratings))
print()

print("Stage 5 — Print numeric rating for each review")
for rating_str in customer_ratings:
    print(int(rating_str[0]))
print()

print("Stage 6 — Build numeric_ratings incrementally")
numeric_ratings = []
for rating_str in customer_ratings:
    numeric_ratings.append(int(rating_str[0]))
    print("Current list:", numeric_ratings)               # print() inside the loop → observe construction

avg_rating = sum(numeric_ratings) / len(numeric_ratings)
print("Average Customer Rating:", round(avg_rating, 2))
print()

# ----------------------------------------------------
# Stage 7 — Production-style version (final output only)
# ----------------------------------------------------
# In practice, we usually build the list silently
# and display only the final result.

numeric_ratings = []
for rating_str in customer_ratings:
    numeric_ratings.append(int(rating_str[0]))

avg_rating = sum(numeric_ratings) / len(numeric_ratings)

print("Numeric ratings:", numeric_ratings)                 # print() after the loop → observe result
print("Average Customer Rating:", round(avg_rating, 2))


# NOTE:
# - print() inside the loop → inspect how data is built step-by-step
# - print() after the loop → inspect the final result only
# During development, step-by-step prints are useful.
# In production or analysis code, they are usually removed.
`````

`````python
# ----------------------------------------------------
# Looping Over Items to Build New Lists
# ----------------------------------------------------
# This approach scales to thousands of values with no change in code size.

# Example — USD → EUR conversion

exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

euro_list = []                     # Start container

for price in usd_list:             # Iterates 5 times
    euro_value = round(price * exchange_rate, 2)
    euro_list.append(euro_value)   # .append() list container

euro_list
`````

`````python
# ---------------------------------------
# 🟦 Helper Expansion - `loop_trace_for()`
# ---------------------------------------

# Given data:
exchange_rate = 0.88                          # exchange_rate → loop-invariant scalar parameter
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]  # usd_list      → iterable dataset

# 🟦 Helper for visualization
def loop_trace_for(iterable):
    print("TRACE: For Loop Execution\n-------------------------")
    for iteration, item in enumerate(iterable):    # for loop: element in dataset, enumerate(usd_list) returns pairs (index, value)
        print(f"Iteration {iteration}: item = {item!r}")

# Call `loop_trace_for()` Helper
loop_trace_for(usd_list)

# Actual loop performing the conversion
euro_list = []                                     # create empty list container
for price in usd_list:                             # for loop: element in iterable
    euro_value = round(price * exchange_rate, 2)   # assign variable: each element from the iterable is * by the loop-invariant scalar exchange_rate
    euro_list.append(euro_value)                   # .append() empty list container

print("\nConverted EUR prices:", euro_list)
`````

`````python
# --------------------------------------------------------
# 🟦 Helper Expansion - `deep_loop_state_visualiser(,)`
# --------------------------------------------------------

# Given data:
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

# 🟦 Helper for visualization
def deep_loop_state_visualizer(usd_list, exchange_rate):
                                                       # usd_list      → iterable dataset
                                                       # exchange_rate → loop-invariant scalar parameter
    """
    Shows detailed iteration state for a currency conversion loop.
    Reveals:
      - iteration number
      - current input value
      - computed euro_value
      - euro_list after each step
    """
    euro_list = []                                    # create empty list container
    
    print("DEEP LOOP STATE VISUALIZER\n--------------------------")
    
    for iteration, price in enumerate(usd_list):      # for loop: element in dataset, enumerate(usd_list) returns pairs (index, value)
        euro_value = round(price * exchange_rate, 2)  # Assign variable: each element from the iterable is * by the loop-invariant scalar exchange_rate
        euro_list.append(euro_value)                  # Append this iteration’s computed euro_value to the accumulator list euro_list
        
        print(f"Iteration {iteration}:")              # iteration index from enumerate()
        print(f"  price       = {price!r}")           # value retrieved from the iterable at this iteration
        print(f"  euro_value  = {euro_value!r}")      # variable holding the results that change with each iteration
        print(f"  euro_list   = {euro_list}\n")       # display the state of the accumulator list after each iteration.
    print("Final Output:", euro_list)
    return euro_list

# Call `deep_loop_state_visualizer()` Helper
deep_loop_state_visualizer(usd_list, exchange_rate)
`````

`````python
# ----------------------------------------------------
# Looping Over Indices
# ----------------------------------------------------
# Index-based iteration is appropriate when:
# - You need to access multiple lists in parallel.  
# - You want both the index and the element.  
# - You are transforming nested lists or multi-dimensional data.


# Example — Access elements by index:

usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

for i in range(len(usd_list)):     # i = 0 → 4
    print(i, usd_list[i])
print()


# 🟦 Helper Expansion — `range_index_map()`
# Integrates your Chapter 6 Range Visualizer into loop logic.


def range_index_map(iterable):
    """
    Shows how Python maps a range to indices for a given iterable.
    Mirrors Maven's step-through on page 10.
    """
    print("RANGE → INDEX MAP\n------------------")
    for idx in range(len(iterable)):
        print(f"range index {idx} → iterable[{idx}] = {iterable[idx]!r}")


# Usage:

range_index_map(usd_list)
`````

`````python
# ---------------------------------------------
# Indexs-based: Looping Over Multiple Iterables  
# ----------------------------------------------
# When two lists share a one-to-one alignment (e.g., item ↔ price), indexing allows synchronized access:

# Example — Print item and its euro price
# This produces 5 synchronized pairs

items = ["Snowboard", "Boots", "Helmet", "Goggles", "Bindings"]
euro_prices = [5.27, 8.79, 17.59, 21.99, 87.99]

for i in range(len(items)):
    print(items[i], euro_prices[i])


# 🟦 Helper — `sync_lists()`
# Synchronized iteration across multiple equal-length iterables

def sync_lists(*lists):
    """
    Prints synchronized items across multiple lists of equal length.
    Excellent for diagnostics in multi-iterable looping.
    """
    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
    length = len(lists[0])
    for iterable in lists:
        if len(iterable) != length:
            raise ValueError("All lists must have the same length for synchronized iteration.")
    
    for idx in range(length):
        row = [lst[idx] for lst in lists]
        print(f"Index {idx}: {row}")

# Usage:
sync_lists(items, euro_prices)
`````

`````python
# ----------------------------------------------------
# Exercise: Index-based looping — step-by-step 
# Why position matters ⚠️
# ----------------------------------------------------
# ⚠️ Caution:
# Index-based loops rely entirely on positional alignment.
# If related lists become misaligned, results may be silently incorrect.
# Index-Based Loop (Positional Control, Higher Risk)

for i in range(len(items)):   # i for indices
    print(i, items[i])

# Use when:
# - you must reference multiple lists by position
# - you need write-access to a list by index


items = ["Snowboard", "Boots", "Helmet"]
prices = [499, 299, 99]

# Intentionally misaligned ratings
ratings = [4, 5, 4]              # should be [5, 4, 4]

print("Index-based loop output (ratings misaligned):\n")

for i in range(len(items)):
    print(items[i], prices[i], ratings[i])

print("\nNote:")
print("Index-based loops assume all lists are aligned by position.")
print("If one list is transposed, results will be silently incorrect.")

# ----------------------------------------------------
# 🟦 Helper - Validate alignment before looping
# ----------------------------------------------------
# Length alignment is required for index-based looping.

if not (len(items) == len(prices) == len(ratings)):
    raise ValueError("Lists are not aligned in length")

print("Lists are aligned by length. Safe to proceed.")
print()

# -------------------------------------------------------------------------
#  🧠 Safer iteration using zip()
# zip() prevents index misuse; but it will still silently accept wrong order
# --------------------------------------------------------------------------

for item, price, rating in zip(items, prices, ratings):
    print(item, price, rating)

# -----------------------------------
# ⚠️ Avoiding Misalignment
# -----------------------------------
# Option 1: Validate assumptions
# Option 2: Use zip() for safer iteration
# Option 3: Use structured data (preferred)

# 3a. Parallel lists (must remain aligned)
items = ["Snowboard", "Boots", "Helmet"]
prices = [499, 299, 99]
ratings = [5, 4, 4]

for i in range(len(items)):
    print(items[i], prices[i], ratings[i])
print()

# Use tuples (alignment enforced by structure)
products = [
    ("Snowboard", 499, 5),
    ("Boots", 299, 4),
    ("Helmet", 99, 4)
]

for item, price, rating in products:
    print(item, price, rating)
print()

# Use dictionaries (most explicit and flexible)
products = [
    {"item": "Snowboard", "price": 499, "rating": 5},
    {"item": "Boots", "price": 299, "rating": 4},
    {"item": "Helmet", "price": 99, "rating": 4},
]

for product in products:
    print(product["item"], product["price"], product["rating"])
`````

`````python
# ----------------------------------------------------
# Exercise: Index-based looping
# ----------------------------------------------------
# ⚠️ Caution:
# Index-Based Loop (Positional Control, Higher Risk)
# Any of the lists can define the range if they’re aligned.
# The list used in range(len(...)) becomes the loop’s authority.

item_list = ["Snowboard", "Boots", "Helmet", "Goggles", "Bindings"]
euro_list = [5.27, 8.79, 17.59, 21.99, 87.99]

# ----------------------------------------------------
# 🟦 Helper - Validate alignment before looping
# ----------------------------------------------------
# Length alignment is required for index-based looping.

if not (len(item_list) == len(euro_list)):
    raise ValueError("Lists are not aligned in length")

print("Lists are aligned by length. Safe to proceed.")


# ----------------------------------------------------
# 🟦 Helper — `sync_lists()`
# Synchronized iteration across multiple equal-length iterables
# ----------------------------------------------------

def sync_lists(*lists):
    """
    Prints synchronized items across multiple lists of equal length.
    Excellent for diagnostics in multi-iterable looping.
    """
    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
    length = len(lists[0])
    for iterable in lists:
        if len(iterable) != length:
            raise ValueError("All lists must have the same length for synchronized iteration.")
    
    for idx in range(length):
        row = [lst[idx] for lst in lists]
        print(f"Index {idx}: {row}")

# Usage:
sync_lists(euro_list, item_list)
print()


item_list = ["Snowboard", "Boots", "Helmet", "Goggles", "Bindings"]
euro_list = [5.27, 8.79, 17.59, 21.99, 87.99]

for i in range(len(euro_list)):
    print(i)
for i in range(len(euro_list)):
    print(f"The {item_list[i].lower()} costs {euro_list[i]} euros.")
print()

for i in range(len(item_list)):
    print(i)
for i in range(len(item_list)):
    print(f"The {item_list[i].lower()} costs {euro_list[i]} euros.")
`````

`````python
# ----------------------------------------------------
# Exercise: Index-based looping
# ----------------------------------------------------
# ⚠️ Caution:
# Index-Based Loop (Positional Control, Higher Risk)
   
inventory = [10, 14, 10, 15, 8, 27]
prices = [20.00, 12.99, 50.00, 40.00, 87.99, 99.99]

# ----------------------------------------------------
# 🟦 Helper - validate_alignment()
# ----------------------------------------------------
# Length alignment is required for index-based looping.

if not (len(inventory) == len(prices)):
    raise ValueError("Lists are not aligned in length")

print("Lists are aligned by length. Safe to proceed.")


for i in range(len(inventory)):
    print(round(inventory[i] * prices[i], 2))                   # Total values each product in inventory  

product_values = []
for i in range(len(prices)):
    product_total = (round(inventory[i] * prices[i], 2))
    product_values.append(product_total)                         # Total values each product in inventory assigned to a variable
product_values

product_values = []
for i in range(len(prices)):
    product_values.append((round(inventory[i] * prices[i], 2)))  # Alternatively
product_values

total = sum(product_values)                                      # Compute the sum

print("\nTotal Value of Inventory:", total)                      # Total value of inventory
# Or with formatting for currency:
print(f"\nTotal Value of Inventory: ${total:,.2f}")              # Total value ($) of inventory
`````

`````python
# -------------------------------------------------------------------------
# looping using `enumerate()
# Clean, safe way to access both the index and the value during iteration
# -------------------------------------------------------------------------
#`enumerate()` Loop (Recommended When Index Is Needed)
# Basic `enumerate()` pattern:

items = ["Snowboard", "Boots", "Helmet"]

for index, element in enumerate(items):
    print(index, element)

# Output:
# 0 Snowboard
# 1 Boots
# 2 Helmet

# Alternatively:
for i, item in enumerate(items):
    print(i, item)

# Use when:
# - you want the index for labeling, tracing, or logic
# - you want clarity and safety
# - you want to avoid manual index bookkeeping


# ----------------------------------------------------
# 🔁 Applied Example — Index-Based Loop 🆚 `enumerate()`
# ----------------------------------------------------
# Index-Based Version:

exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99]
euro_list = []

for price in range(len(usd_list)):
    euro_value = round(usd_list[price] * exchange_rate, 2)
    euro_list.append(euro_value)

print(euro_list)


# `enumerate()` Version (Cleaner)

exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99]
euro_list = []

for i, price in enumerate(usd_list):
    euro_value = round(price * exchange_rate, 2)
    euro_list.append(euro_value)

print(euro_list)
`````

`````python
# -------------------------------------------------------------------------
# 🟦 Helper — Enumerate Visualizer
# This helper makes `enumerate()` behavior explicit and observable
# -------------------------------------------------------------------------

def enumerate_visualizer(iterable):
    """
    Visualizes how enumerate() yields index-value pairs
    during iteration.
    """
    print("ENUMERATE VISUALIZER\n---------------------")
    
    for index, value in enumerate(iterable):
        print(f"index = {index}, value = {value!r}")


# Example Usage:

items = ["Snowboard", "Boots", "Helmet"]
enumerate_visualizer(items)
`````

`````python
# -------------------------------------------
# looping using `enumerate()` )🆚
# 🔁 Comparison using `zip()`
# -------------------------------------------
# Basic `enumerate()` pattern:
# items = ["Snowboard", "Boots", "Helmet", "Bindings", "Goggles"]
# for index, element in enumerate(items):
#     print(index, element)

proceeds_list = (5.27, 8.79, 21.99, 87.99, 99.99)
stock_list = ["Snowboard", "Boots", "Helmet", "Bindings", "Goggles"]

# Demonstrate: enumerate() yields tuples
for pair in enumerate(proceeds_list):
    print(pair, type(pair))

# Example output:
# (0, 5.27) <class 'tuple'>
# (1, 8.79) <class 'tuple'>
# (2, 21.99) <class 'tuple'>
# (3, 87.99) <class 'tuple'>
print()

# ----------------------------------------------------
# 🟦 Helper — Validate alignment before looping
# ----------------------------------------------------

if len(proceeds_list) != len(stock_list):
    raise ValueError("Lists are not aligned in length")

print("Lists are aligned by length. Safe to proceed.")
print()

# -------------------------------------------------------------------------
# 🟦 Helper — Enumerate Visualizer
# This helper makes `enumerate()` behavior explicit and observable
# -------------------------------------------------------------------------

def enumerate_visualizer(iterable):
    """
    Visualizes how enumerate() yields index-value pairs during iteration.
    """
    print("ENUMERATE VISUALIZER\n---------------------")
    for index, value in enumerate(iterable):
        print(f"index = {index}, value = {value!r}")

# Call Enumerate Visualizer():
enumerate_visualizer(proceeds_list)
enumerate_visualizer(stock_list)
print()

# ----------------------------------------------------------------------
# 🧭 Looping using `enumerate()` — defining iteration by one iterable
# ----------------------------------------------------------------------
# ⭐ Best Practice:
# - Equal peers → use zip()
# - One iterable controls iteration → use enumerate()


for index, price in enumerate(proceeds_list):    # proceeds_list defines iteration and index
    print(stock_list[index], price)              # index applied to stock_list; price from proceeds_list
print()

# Alternative Display:
print("Product Price List:")
for index, price in enumerate(proceeds_list):
    product = stock_list[index]
    print(f"{product}: ${price:,.2f}")
print()

# -------------------------------------------------------------
#  🧠 Looping using `zip()`
# -------------------------------------------------------------
# If all iterables are equal peers → use zip()
# If one iterable defines the iteration → use enumerate()

print("Product Price List:")
for stock, price in zip(stock_list, proceeds_list):
    print(f"{stock}: ${price:,.2f}")
`````

`````python
# --------------------------------------------------------
# Exercise — looping using `enumerate()`
# with conditional logic, cross-list alignment via index
# --------------------------------------------------------
# Basic `enumerate()` pattern:
# items = ["Snowboard", "Boots", "Helmet", "Bindings", "Goggles"]
# for index, element in enumerate(items):
#     print(index, element)


# Given data:
# Tax rates: Sun Valley → 8%, Stowe → 6%, Mammoth → 7.75%

# Transaction subtotals
subtotals = [15.98, 899.97, 799.97, 117.96, 5.99, 599.99, 24.99, 1799.94, 99.99]

# Corresponding transaction locations (positionally aligned)
location = [
    "Sun Valley", "Stowe", "Mammoth",
    "Stowe", "Sun Valley", "Mammoth",
    "Sun Valley", "Mammoth", "Stowe"
]


taxes = []
totals = []

for i, element in enumerate(subtotals):
    if location[i] == 'Sun Valley':
        tax = round(subtotal * .08, 2)
    elif location[i] == 'Stowe':
        tax = round(subtotal * .06, 2)
    else:
        tax = round(subtotal * .0775, 2)
    total = round(subtotal + tax, 2)
    taxes.append(tax)
    totals.append(total)

print("Taxes:", taxes)
print("Totals:", totals)

# While this logic works, it will later be replaced by more robust data structures such as:
# records (tuples or dictionaries),
# DataFrames,
# or rule mappings (e.g., dictionaries of tax rates).
`````

- You want both the **index** and **the element**.  
- You are transforming nested lists or multi-dimensional data.

### Example — Access elements by index

```python
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

for i in range(len(usd_list)):     # i = 0 → 4
    print(i, usd_list[i])
```

### Helper Expansion — `range_index_map()`
Integrates your Chapter 6 Range Visualizer into loop logic.

```python
def range_index_map(iterable):
    """
    Shows how Python maps a range to indices for a given iterable.
    Mirrors Maven's step-through on page 10.
    """
    print("RANGE → INDEX MAP\n------------------")
    for idx in range(len(iterable)):
        print(f"range index {idx} → iterable[{idx}] = {iterable[idx]!r}")
```

Usage:

```python
range_index_map(usd_list)
```

---

# 4. Looping Over Multiple Iterables  

When two lists share a one-to-one alignment (e.g., item ↔ price), indexing allows synchronized access:

### Example — Print item and its euro price

```python
items = ["Snowboard", "Boots", "Helmet", "Goggles", "Bindings"]
euro_prices = [5.27, 8.79, 17.59, 21.99, 87.99]

for i in range(len(items)):
    print(items[i], euro_prices[i])
```

This produces 5 synchronized pairs.

### Helper — `sync_lists()`

```python
def sync_lists(*lists):
    """
    Prints synchronized items across multiple lists of equal length.
    Excellent for diagnostics in multi-iterable looping.
    """
    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
    length = len(lists[0])
    for iterable in lists:
        if len(iterable) != length:
            raise ValueError("All lists must have the same length for synchronized iteration.")
    
    for idx in range(length):
        row = [lst[idx] for lst in lists]
        print(f"Index {idx}: {row}")
```

Usage:

```python
sync_lists(items, euro_prices)
```

# 5. `enumerate()`: The Best Practice  

Using `enumerate()` is the preferred method when you want both index and item without constructing `range(len(...))`.

### Example — Using `enumerate()`

```python
euro_list = [5.27, 8.79, 17.59, 21.99, 87.99]
item_list = ["Snowboard", "Boots", "Helmet", "Goggles", "Bindings"]

for idx, value in enumerate(euro_list):
    print(item_list[idx], value)
```

### Helper — `enumerate_visualizer()`

```python
def enumerate_visualizer(iterable):
    """
    Shows exactly what enumerate() produces (index, item).
    """
    print("ENUMERATE VISUALIZER\n---------------------")
    for idx, item in enumerate(iterable):
        print(f"enumerate → (index={idx}, item={item!r})")
```

Example:

```python
enumerate_visualizer(["Snowboard", "Boots", "Helmet"])
```
---

### Overview — Using `enumerate()`
In Python, `enumerate()` provides a clean and safe way to access **both the index and the value** during iteration.  
It resolves the main drawbacks of index-based looping while preserving positional awareness when needed.

Conceptually, `enumerate()` represents the **next evolutionary step** after learning:

1. Item-based for loops (simple, readable)  
2. Index-based for loops (positional control, but fragile)  

`enumerate()` combines the strengths of both approaches.

---

## Why `enumerate()` Exists

Recall the index-based loop pattern:

```python
for i in range(len(items)):
    print(items[i])
```

While correct, this pattern requires you to:
- manually manage indices  
- ensure correct bounds (`len(...)`)  
- mentally track both index and value  

`enumerate()` removes this overhead by yielding **index–value pairs directly**.

---

## Basic `enumerate()` Pattern

```python
items = ["Snowboard", "Boots", "Helmet"]

for index, item in enumerate(items):
    print(index, item)
```

### Output
```
0 Snowboard
1 Boots
2 Helmet
```

### Key Idea
- `index` is automatically generated  
- `item` is the value from the iterable  
- No manual indexing is required  

---

## Comparing Three Looping Styles

### 1. Item-Based Loop (Most Common)

```python
for item in items:
    print(item)
```

Use when:
- you only need the value  
- position does not matter  

---

### 2. Index-Based Loop (Positional Control, Higher Risk)

```python
for i in range(len(items)):
    print(i, items[i])
```

Use when:
- you must reference multiple lists by position  
- you need write-access to a list by index  

⚠️ Requires careful alignment.

---

### 3. `enumerate()` Loop (Recommended When Index Is Needed)

```python
for i, item in enumerate(items):
    print(i, item)
```

Use when:
- you want the index for labeling, tracing, or logic  
- you want clarity and safety  
- you want to avoid manual index bookkeeping  

---

## Applied Example — Index-Based Loop vs `enumerate()`

### Index-Based Version

```python
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99]
euro_list = []

for i in range(len(usd_list)):
    euro_value = round(usd_list[i] * exchange_rate, 2)
    euro_list.append(euro_value)

print(euro_list)
```

---

### `enumerate()` Version (Cleaner)

```python
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99]
euro_list = []

for i, price in enumerate(usd_list):
    euro_value = round(price * exchange_rate, 2)
    euro_list.append(euro_value)

print(euro_list)
```

### Why This Is Better
- `price` is directly available (no `usd_list[i]`)  
- `i` is present if needed (debugging, labeling)  
- Loop intent is clearer and easier to reason about  

---

## 🟦 Helper — Enumerate Visualizer

This helper makes `enumerate()` behavior explicit and observable.

```python
def enumerate_visualizer(iterable):
    """
    Visualizes how enumerate() yields index-value pairs
    during iteration.
    """
    print("ENUMERATE VISUALIZER\n---------------------")
    
    for index, value in enumerate(iterable):
        print(f"index = {index}, value = {value!r}")
```

### Example Usage

```python
items = ["Snowboard", "Boots", "Helmet"]
enumerate_visualizer(items)
```

---

## When to Prefer `enumerate()`

Use `enumerate()` when:
- you need index information for tracing or labeling  
- you want positional awareness without index arithmetic  
- readability and debuggability matter  
- teaching or explaining iteration logic  

Avoid `enumerate()` when:
- you only need the values (simple item-based loop is best)  
- you must mutate a list by index (explicit index-based loop may be required)  

### Related Guidance
- If all iterables are equal peers → use `zip()`  
- If one iterable defines the iteration → use `enumerate()`
- `enumerate()` yields `(index, value)` pairs automatically  
- It eliminates the need for `range(len(...))` in most cases  
- It improves clarity, safety, and debuggability  
- Index-based loops still matter when mutating data or coordinating multiple structures  

---

## How This Fits the Learning Flow

At this point, you have learned:

- Item-based looping (cleanest default)  
- Index-based looping (positional control, but fragile)  
- `enumerate()` (balanced, expressive, safer) → one iterable defines the iteration
- `zip()` pairing two equal-length iterables, avoids manual indexing  

This completes the **for-loop family**.

The next major step is **while loops**, which introduce:
- state-driven iteration  
- termination logic  
- infinite-loop risk  
- simulation-style patterns common in analytics  

---

# 6. Choosing Between Items, Indices, `enumerate()`, and `zip()`

| Pattern                              | Who defines iteration? | What is yielded per loop? |
| ------------------------------------ | ---------------------- | ------------------------- |
| `for item in iterable`               | iterable               | `item`                    |
| `for i in range(len(iterable))`      | iterable               | `i` (int)                 |
| `for i, item in enumerate(iterable)` | iterable               | `(i, item)` tuple         |
| `for a, b in zip(iter1, iter2)`      | **both equally**       | `(a, b)` tuple            |


| Approach                                | When to Use                                                                      | Benefits                            |
| --------------------------------------- | -------------------------------------------------------------------------------- | ----------------------------------- |
| `for item in iterable:`                 | Transforming or inspecting values                                                | Clean, safest, Pythonic             |
| `for i in range(len(iterable)):`        | When **one iterable controls iteration** and index access is explicitly required | Maximum positional control          |
| `for idx, item in enumerate(iterable):` | When you need **both index and value** from a single iterable                    | Best practice, readable, efficient  |
| `for a, b in zip(iter1, iter2):`        | When **multiple iterables are equal peers** and should be processed together     | Safest alignment, avoids index bugs |


---

# Key Takeaways

- **For loops** execute once per element in an iterable; the iteration count is predictable.  
- Use **item-based looping** whenever possible; it is the safest and most readable pattern.  
- Use **indices** when pairing multiple lists or modifying elements in place.  
- Use **enumerate()** for clean, Pythonic access to both **index and item**. If one iterable defines the iteration, use enumerate()
- Use **zip()** if all iterables are equal peers.  

- This section introduces the core loop helpers:  
  - **`loop_trace_for()`** — traces item-by-item for-loop iteration and execution order  
  - **`enumerate_visualizer()`** — makes index–value pairing explicit when using `enumerate()`  
  - **`safe_while_trace()`** — safely traces while-loop state evolution and prevents infinite loops  
  - **`deep_loop_state_visualizer()`** — traces each internal step of a list-processing loop, including variable assignment, calculation execution, each `.append()` operation, and the evolving state of the accumulator container
  - **`validate_alignment()`** - Validate iterable alignment before looping. (Length alignment is required for index-based looping)
  - **`sync_lists()`** — synchronized iteration across multiple equal-length iterables 
- Understanding iteration mechanics — including when and how position matters — is essential for dataset traversal, data cleaning, transformation workflows, simulation modeling, and algorithmic problem-solving.
- These visualizers make loop mechanics traceable, which is essential for debugging, teaching, and analytics work.


# Chapter 07 — Loops
# Section 7.03 — While Loops

### Preview
A **while loop** repeats a block of code *as long as a logical expression evaluates to True*.  
This contrasts with a **for loop**, which repeats a fixed number of times.

Use a while loop when:

- You **do not know** how many iterations will be required.  
- You are simulating a process to completion (e.g., reach a threshold, exhaust inventory).  
- You need iterative refinement until a condition stabilizes (e.g., convergence in algorithms).  

## While loops repeat based on state, not count

Structure can be summarized as:
```Python
while condition_on_state:
    update_state
```

Execution continues until the expression becomes **False**.  
If the condition never turns False, you enter an **infinite loop**, which we will guard against.

This section extends your Helper Toolkit with a **While Loop Safety System**, simulation helpers, and state-tracing tools.

---

# When Should You Use Each Loop Type?

| Scenario | Best Choice | Rationale |
|---------|-------------|-----------|
| You know the dataset length | **For Loop** | One iteration per element; safe, predictable |
| You want to repeat until a condition is false | **While Loop** | Logic-driven iteration |
| You need to prevent infinite loops | **For Loop** (or While Loop with counters) | Counters act as emergency brakes |
| You are iterating over sequences (list, tuple, string) | **For Loop** | Pythonic and safe |
| You need a “repeat until solved” pattern (e.g., simulation) | **While Loop** | Flexible stopping criteria |

---

# 1. Basic While Loop Pattern

A while loop must include **three things**:

1. An **initial state**  
2. A **logical condition** that inspects that state  
3. An **update step** that *moves the state toward termination*  

### Example — Counting upward

```python
counter = 0                     # initial state

while counter < 5:              # logical condition
    print(counter)
    counter += 1                # update step
```

Python evaluates:

- Iteration 0 → counter = 0  
- Iteration 1 → counter = 1  
- Iteration 2 → counter = 2  
- Iteration 3 → counter = 3  
- Iteration 4 → counter = 4  
- Iteration stops when counter becomes 5, making (counter < 5) False.

Without the update (`counter += 1`), the loop would run forever.


`````python
# --------------------------------------------------------
# Demonstrating a While Loop Pattern Safely with Counters
# --------------------------------------------------------
# Simple while loop that counts upward until a threshold is reached
# While loops run until a given logical expression becomes FALSE


counter = 0
while counter < 5:      # Loop continues WHILE this expression is TRUE
    counter += 1        # Without this update, the loop becomes infinite ⚠️
    print("Counter =", counter)  
print()


# --------------------------------------------------------
# While Loop: Inventory until stock runs out
# --------------------------------------------------------

current_inventory = 686
month_counter = 0
monthly_sales = 84

while current_inventory > 0:
    current_inventory -= monthly_sales          # subtraction assignment
    month_counter += 1
    print(
        f"At month-end {month_counter}: "
        f"My inventory balance is {current_inventory}"
    )
print()


# --------------------------------------------------------
# While Loop: Calculating bank balance until money runs out
# --------------------------------------------------------

bank_balance = 7_189
month_counter = 0
spending = 350

while bank_balance > 0:
    bank_balance -= spending          # subtraction assignment
    month_counter += 1
    print(
        f"At month-end {month_counter}: "
        f"My balance is ${round(bank_balance, 2)}"
    )
`````

`````python
# --------------------------------------------------------
# While Loops with Counters
# Counters track loop progress, prevent runaway execution, 
# and can act as secondary safety conditions
# --------------------------------------------------------

# Example — Growing a portfolio until it reaches 1 million
# Loop terminates when the portfolio crosses the threshold.


portfolio = 800_000
year = 0

while portfolio < 1_000_000:            # primary logical condition
    investment_income = portfolio * 0.05
    portfolio += investment_income       # update state
    year += 1                             # update counter
    print(f"End of year {year}: ${round(portfolio, 2)}")



# --------------------------------------------------------
# 🟦 Advanced Helper: State Timeline Recorder
# This helper logs every state value for later visualization or tabular analysis.
# --------------------------------------------------------


def while_timeline(start_state, condition_fn, update_fn, max_iter=100):
    """
    Records each iteration's state—ideal for simulations, 
    convergence testing, and algorithm diagnostics.
    """
    timeline = []
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        timeline.append((iteration, state))
        
        iteration += 1
        if iteration > max_iter:
            timeline.append(("ABORTED", state))
            break
        
        state = update_fn(state)
    
    timeline.append(("FINAL", state))
    return timeline


# Example

timeline = while_timeline(
    start_state=800000,
    condition_fn=lambda s: s < 1_000_000,
    update_fn=lambda s: s * 1.05,
    max_iter=10
)

timeline

# Produces a list of iteration → state pairs, perfect for later DataFrame conversion.
`````

`````python
# --------------------------------------------------------
# 🔁 While Loops — Condition-Driven Iteration 
# 🆚 For Loops - Count-Driven, does not mutate data
# --------------------------------------------------------

# --------------------------------------------------------
# While Loops — Condition-Driven Iteration
# --------------------------------------------------------
# NOTE:
# This example mutates the list in place to demonstrate while-loop logic ⚠️
# In real data analysis workflows, you usually avoid mutating lists directly
# Instead, you would use filtering or list comprehensions to create a new list


customer_ratings = [
    "5 stars", "3 stars", "5 stars", "4 stars", "4 stars",
    "1 star", "4 stars", "5 stars", "2 stars", "5 stars"
]

onestar_counter = 0

while "1 star" in customer_ratings:
    customer_ratings.remove("1 star")
    onestar_counter += 1

print(f"Removed {onestar_counter} one-star review(s)")
print(customer_ratings)

# --------------------------------------------------------
# For Loops — Count-Driven, does not mutate data
# --------------------------------------------------------
# TEACHING TAKEAWAY:
# While loops are useful when iteration depends on changing state.
# For loops are preferred when data can be processed in a single pass.
# In analytics, for loops (or vectorized operations) are usually clearer and safer.


customer_ratings = [
    "5 stars", "3 stars", "5 stars", "4 stars", "4 stars",
    "1 star", "4 stars", "5 stars", "2 stars", "5 stars"
]

onestar_counter = 0
filtered_ratings = []              # empty list container

for rating in customer_ratings:   # loop looks at every element in customer_ratings
    if rating == "1 star":        # if block counts "1 star" and does nothing else with it
        onestar_counter += 1
    else:                         # else block handles everything that is not "1 star"
        filtered_ratings.append(rating)

print(f"Removed {onestar_counter} one-star review(s)")
print(filtered_ratings)
print(customer_ratings)           # does not mutate data in-place
`````

`````python
# ------------------------------------------------------------
# 🟦 Helper: `safe_while_trace()` with Infinite-Loop Protection
# - state value at each iteration  
# - when/why the loop stops  
# - detection of missing or incorrect update logic  
# - protection against accidental infinite loops via `max_iter`
# -------------------------------------------------------------

# ❌ Incorrect logic: update is calculated but not reassigned

counter = 0

# ⚠️ DO NOT RUN
# while counter < 5:
#     counter + 1      # expression evaluated, result discarded
#     print(counter)

# counter += 1   # ✔ correct
# counter + 1    # ❌ does nothing

# -------------------------------------------------------------
# 🟦 Safe demonstration with helper
# -------------------------------------------------------------
safe_while_trace(
    start_state=165_00,
    condition_fn=lambda s: s < 3_500_000,
    update_fn=lambda s: s + s * 0.13,
    max_iter=10
)


# This outputs warnings before an infinite loop occurs.
`````

`````python
# -----------------------------------------------------------
# Exercise — Demonstrating a While Loop
# Example 401(k) - No Contributions, Annual Returns 13%
# ------------------------------------------------------------

# Starting portfolio balance is 165_000
stock_portfolio = 165_000
year_counter = 0

while stock_portfolio < 2_000_000:
    # calculate annual investment return
    investment_return = stock_portfolio * 0.13  # interest rate

    # add income to end of year portfolio balance
    stock_portfolio += investment_return

    # add each year
    year_counter += 1

    print(f"At year-end {year_counter}: "
    + f"My balance is ${round(stock_portfolio, 2)}")
print()


# Alternatively:

stock_portfolio = 165_000
year_counter = 0
annual_return_rate = 0.13

while stock_portfolio < 2_000_000:
    investment_return = stock_portfolio * annual_return_rate
    stock_portfolio += investment_return
    year_counter += 1

    print(f"At year-end {year_counter}: "
          f"My balance is ${round(stock_portfolio, 2)}")

print(f"\nGoal reached in {year_counter} years.")
`````

`````python
# ------------------------------------------------------------
# 🟦 Helper: `safe_while_trace()` with Infinite-Loop Protection
# - state value at each iteration  
# - when/why the loop stops  
# - detection of missing or incorrect update logic  
# - protection against accidental infinite loops via `max_iter`
# -------------------------------------------------------------

stock_portfolio = 165000                        # start state is 165000

while stock_portfolio < 3_500_000:              # condition is s < 3_500_000
    stock_portfolio += stock_portfolio * 0.14   # update step is s * 0.13


def safe_while_trace(start_state, condition_fn, update_fn, max_iter=50):
    """
    Safely traces a while loop with a guard to prevent infinite loops.
    """
    print("SAFE WHILE TRACE\n-----------------")
    
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        print(f"Iteration {iteration}: state = {state!r}")
        
        iteration += 1
        if iteration > max_iter:
            print("ABORTED: Possible infinite loop (max_iter exceeded).")
            break
        
        state = update_fn(state)
    
    print(f"Final state: {state!r}")
    return state
print()

# Function call:
safe_while_trace(
    start_state=165000,
    condition_fn=lambda s: s < 3_500_000,
    update_fn=lambda s: s + (s * 0.14),
    max_iter=50
)
`````

`````python
# --------------------------------------------------------
# Exercise — Demonstrating a While Loop
# Roth IRA Example with monthly contributions
# --------------------------------------------------------

# -----------------------------
# Roth IRA inputs  🧠✨
# -----------------------------
stock_portfolio = 21_077
monthly_contribution = 8_000 / 12      # $666.67 per month
annual_return_rate = 0.14
year_counter = 0

while stock_portfolio < 124_000:
    # add monthly contributions
    stock_portfolio += monthly_contribution * 12  # monthly contributions

    # apply annual investment return
    investment_return = stock_portfolio * annual_return_rate
    stock_portfolio += investment_return

    year_counter += 1

    print(
        f"At year-end {year_counter}: "
        f"My balance is ${round(stock_portfolio, 2)}"
    )

print(f"\nGoal reached in {year_counter} years.")
`````

`````python
# --------------------------------------------------------
# Exercise — Demonstrating a While Loop
# Example Roth IRA — Age + Monthly vs Annual Contribution Timing
# --------------------------------------------------------

from datetime import date

# -----------------------------
# Fixed personal anchors
# -----------------------------
BIRTH_YEAR = 1967
AS_OF = date(2025, 12, 14)   # reporting date each year (Dec 14)

# -----------------------------
# Roth IRA inputs  🧠✨
# -----------------------------
start_balance = 21_077
annual_contribution = 8_000
annual_return_rate = 0.14
goal = 145_000

# -----------------------------
# Simulator with age tracking
# timing = "begin" → contribution first (monthly-style timing)
# timing = "end"   → contribution last (annual-style timing)
# -----------------------------
def simulate_roth_with_age(start_balance, annual_contribution, annual_return_rate, goal, timing="begin"):
    balance = start_balance
    years = 0
    rows = []

    while balance < goal:
        years += 1
        year_end = date(AS_OF.year + years, AS_OF.month, AS_OF.day)
        age = year_end.year - BIRTH_YEAR  # Dec 14 is after May 7

        if timing == "begin":
            balance += annual_contribution
            balance += balance * annual_return_rate
        elif timing == "end":
            balance += balance * annual_return_rate
            balance += annual_contribution
        else:
            raise ValueError("timing must be 'begin' or 'end'")

        rows.append((year_end, age, balance))

    return years, rows

# -----------------------------
# Run both scenarios
# -----------------------------
years_begin, rows_begin = simulate_roth_with_age(
    start_balance, annual_contribution, annual_return_rate, goal, timing="begin"
)

years_end, rows_end = simulate_roth_with_age(
    start_balance, annual_contribution, annual_return_rate, goal, timing="end"
)

# -----------------------------
# Output
# -----------------------------
print(f"As of {AS_OF}: age = {AS_OF.year - BIRTH_YEAR}\n")

print("MONTHLY-style contributions (begin-of-year timing):")
for year_end, age, bal in rows_begin:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_begin} years.\n")

print("ANNUAL contributions (end-of-year timing):")
for year_end, age, bal in rows_end:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_end} years.\n")

diff = years_end - years_begin
print(
    f"Difference: monthly-style timing reaches the goal "
    f"{abs(diff)} year(s) {'sooner' if diff > 0 else 'later' if diff < 0 else 'at the same time'}."
)
`````

`````python
# --------------------------------------------------------
# Exercise — Demonstrating a While Loop
# Example 401(k) with Employer Match - Monthly Contributions
# --------------------------------------------------------

# -----------------------------
# Scenario inputs 🧠✨
# -----------------------------
stock_portfolio = 165_000
monthly_contribution = (32_500 / 12) + (6_900 / 12)
annual_return_rate = 0.14
year_counter = 0

while stock_portfolio < 3_500_000:
    stock_portfolio += monthly_contribution * 12  # add 12 months of contributions
    stock_portfolio += stock_portfolio * annual_return_rate
    year_counter += 1

    print(f"At year-end {year_counter}: "
          f"My balance is ${round(stock_portfolio, 2)}")

print(f"\nGoal reached in {year_counter} years.")
`````

`````python
# --------------------------------------------------------
# Exercise — Demonstrating a While Loop
`````
`````python
# Example 401(k) with Employer Match
# 401(k) — Age + Monthly vs Annual Contribution Timing
# --------------------------------------------------------

from datetime import date

# -----------------------------
# Fixed personal anchors
# -----------------------------
BIRTH_YEAR = 1967
AS_OF = date(2025, 12, 14)   # anchor all "year-end" reporting to Dec 14 each year

# -----------------------------
# Scenario inputs 🧠✨
# -----------------------------
start_balance = 165_000
annual_return_rate = 0.14
annual_contribution = 32_500 + 6_900
goal = 3_500_000

# ----------------------------------------------------------------------
# Simulator (filtered w/steps for every 3 years, prints age + balance) 🧠✨
# timing = "begin"  -> contribution first, then return (your "monthly style" timing)
# timing = "end"    -> return first, then contribution (your "annual style" timing)
# ----------------------------------------------------------------------

def simulate_yearly_with_age(start_balance, annual_contribution, annual_return_rate, goal, timing="begin"):
    balance = start_balance
    years = 0
    rows = []

    while balance < goal:
        years += 1
        year_end = date(AS_OF.year + years, AS_OF.month, AS_OF.day)
        age = year_end.year - BIRTH_YEAR  # Dec 14 is after May 7

        if timing == "begin":
            balance += annual_contribution
            balance += balance * annual_return_rate
        elif timing == "end":
            balance += balance * annual_return_rate
            balance += annual_contribution
        else:
            raise ValueError("timing must be 'begin' or 'end'")

        # ---------------------------------
        # Filtered output (every 3 years + final year)
        # ---------------------------------
        if years % 3 == 0 or balance >= goal:
            rows.append((year_end, age, balance))

    return years, rows

# -----------------------------
# Run both scenarios
# -----------------------------
years_begin, rows_begin = simulate_yearly_with_age(start_balance, annual_contribution, annual_return_rate, goal, timing="begin")
years_end, rows_end = simulate_yearly_with_age(start_balance, annual_contribution, annual_return_rate, goal, timing="end")

# -----------------------------
# Print results
# -----------------------------
print(f"As of {AS_OF}: age = {AS_OF.year - BIRTH_YEAR}")
print("\nBEGIN-OF-YEAR contributions (contrib then return):")
for year_end, age, bal in rows_begin:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_begin} years.\n")

print("END-OF-YEAR contributions (return then contrib):")
for year_end, age, bal in rows_end:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_end} years.\n")

# Quick comparison line
delta = years_end - years_begin
print(f"Difference: begin-of-year timing reaches the goal {abs(delta)} year(s) "
      f"{'sooner' if delta > 0 else 'later' if delta < 0 else 'at the same time'} than end-of-year timing.")
`````

`````python
from datetime import date

# --------------------------------------------------------
# Two-Phase 401(k) Simulator — Monthly Contributions
# Phase 1: high contributions for 8 years (96 months)
# Phase 2: reduced contributions until target reached 🚀🔥
# --------------------------------------------------------

# Personal anchors
BIRTH_YEAR  = 1967
BIRTH_MONTH = 5
BIRTH_DAY   = 7

AS_OF = date(2025, 12, 14)  # anchor reporting to the 14th of each month

def add_months(d, months):
    """Add months to a date (safe here because day=14 exists in every month)."""
    y, m = d.year, d.month
    m0 = (m - 1) + months
    new_y = y + (m0 // 12)
    new_m = (m0 % 12) + 1
    return date(new_y, new_m, d.day)

def age_on(d, birth_year=BIRTH_YEAR, birth_month=BIRTH_MONTH, birth_day=BIRTH_DAY):
    """Compute age at a given date."""
    age = d.year - birth_year
    if (d.month, d.day) < (birth_month, birth_day):
        age -= 1
    return age

# -----------------------------
# Scenario inputs
# -----------------------------
start_balance = 164_658.86
annual_return_rate = 0.14
target_balance = 3_000_000

# Phase 1: through 12/14/2033 (8 years = 96 months)
phase1_months = 96
monthly_contrib_phase1 = (32_500 / 12) + (6_900 / 12)   # employee + match

# Phase 2: after 12/14/2033
monthly_contrib_phase2 = (7_200 / 12) + (7_200 / 12)    # employee + match

# Convert annual to monthly (compound-consistent)
monthly_return_rate = (1 + annual_return_rate) ** (1/12) - 1

# -----------------------------
# Simulation (monthly steps)
# -----------------------------
balance = start_balance
month_counter = 0

print(f"As of {AS_OF}: Age {age_on(AS_OF)} | Start balance ${balance:,.2f}\n")

while balance < target_balance:
    month_counter += 1
    current_date = add_months(AS_OF, month_counter)

    # Switch contributions after 96 months (12/14/2033)
    if month_counter <= phase1_months:
        monthly_contribution = monthly_contrib_phase1
        phase_label = "Phase 1 (high contrib)"
    else:
        monthly_contribution = monthly_contrib_phase2
        phase_label = "Phase 2 (reduced contrib)"

    # Contribution first, then growth (monthly cadence)
    balance += monthly_contribution
    balance += balance * monthly_return_rate

    # Print yearly checkpoints + final month
    if month_counter % 12 == 0 or balance >= target_balance:
        print(
            f"{current_date} | Age {age_on(current_date)} | "
            f"{phase_label} | Balance ${balance:,.2f}"
        )

print(f"\nGoal reached in {month_counter} months (~{month_counter/12:.1f} years) by {current_date}.")
`````

# Chapter 07 — Loops
# Section 7.03 — While Loops

### Preview
A **while loop** repeats a block of code *as long as a logical expression evaluates to True*.  
This contrasts with a **for loop**, which repeats a fixed number of times.

Use a while loop when:

- You **do not know** how many iterations will be required.  
- You are simulating a process to completion (e.g., reach a threshold, exhaust inventory).  
- You need iterative refinement until a condition stabilizes (e.g., convergence in algorithms).  

## While loops repeat based on state, not count

Structure can be summarized as:
```Python
while condition_on_state:
    update_state
```

Execution continues until the expression becomes **False**.  
If the condition never turns False, you enter an **infinite loop**, which we will guard against.

This section extends your Helper Toolkit with a **While Loop Safety System**, simulation helpers, and state-tracing tools.

---

# When Should You Use Each Loop Type?

| Scenario | Best Choice | Rationale |
|---------|-------------|-----------|
| You know the dataset length | **For Loop** | One iteration per element; safe, predictable |
| You want to repeat until a condition is false | **While Loop** | Logic-driven iteration |
| You need to prevent infinite loops | **For Loop** (or While Loop with counters) | Counters act as emergency brakes |
| You are iterating over sequences (list, tuple, string) | **For Loop** | Pythonic and safe |
| You need a “repeat until solved” pattern (e.g., simulation) | **While Loop** | Flexible stopping criteria |

---

# 1. Basic While Loop Pattern  

A while loop must include **three things**:

1. An **initial state**  
2. A **logical condition** that inspects that state  
3. An **update step** that *moves the state toward termination*  

### Example — Counting upward

```python
counter = 0                     # initial state

while counter < 5:              # logical condition
    print(counter)
    counter += 1                # update step
```

Python evaluates:

- Iteration 0 → counter = 0  
- Iteration 1 → counter = 1  
- Iteration 2 → counter = 2  
- Iteration 3 → counter = 3  
- Iteration 4 → counter = 4  
- Iteration stops when counter becomes 5, making (counter < 5) False.

Without the update (`counter += 1`), the loop would run forever.

---

# 2. While Loops with Counters

Counters track loop progress, prevent runaway execution, and can act as **secondary safety conditions**.

### Example — Growing a portfolio until it reaches 1 million

```python
portfolio = 800_000
year = 0

while portfolio < 1_000_000:            # primary logical condition
    investment_income = portfolio * 0.05
    portfolio += investment_income       # update state
    year += 1                             # update counter
    print(f"End of year {year}: ${round(portfolio, 2)}")
```
 
Loop terminates when the portfolio crosses the threshold.

---

# 3. 🟦 Helper Expansion: Safe While Loop Tracer

This upgrades your earlier helper into a reusable diagnostic tool for notebook-based explorations.

```python
def safe_while_trace(start_state, condition_fn, update_fn, max_iter=50):
    """
    A safe, diagnostic tracer for while loops.
    Prevents accidental infinite loops and makes state transitions visible.

    Parameters
    ----------
    start_state : any
        Initial loop state.
    
    condition_fn : callable(current_state) -> bool
        Determines whether the loop should continue.
    
    update_fn : callable(current_state) -> new_state
        Defines how the state changes each iteration.
    
    max_iter : int
        Safety guard to prevent infinite looping.
    """
    
    print("SAFE WHILE TRACE\n-----------------")
    
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        print(f"Iteration {iteration}: state = {state!r}")
        
        iteration += 1
        if iteration > max_iter:
            print("ABORTED: Possible infinite loop (max_iter exceeded).")
            break
        
        state = update_fn(state)
    
    print(f"Final state: {state!r}")
    return state
```

### Example Usage — Portfolio example  

```python
safe_while_trace(
    start_state=800_000,
    condition_fn=lambda s: s < 1_000_000,
    update_fn=lambda s: s + s * 0.05,   # grow by 5%
    max_iter=10
)
```

```python

monthly_sales = 75
month = 0

while current_inventory > 0:
    current_inventory -= monthly_sales      # subtract sales
    month += 1
    print(f"After month {month}, inventory = {current_inventory}")
```

The loop ends when inventory reaches zero or below — the precise moment demand exhausts supply.

This mirrors the structure shown in the inventory projections exercise (Maven pp. 23–24).

# 5. Infinite Loops and Error Prevention  

A while loop becomes infinite when:

- The logical condition is always True  
- The update step never moves state toward termination  
- Logic depends on an impossible or stalled signal  
- The update step is omitted (common beginner error)

### Demonstration of incorrect logic  
*(Do NOT run this pattern without safeguards)*

```python
# while x < 10:
#     print(x)      # x never changes -> infinite loop
```

### Safe demonstration with helper

```python
safe_while_trace(
    start_state=0,
    condition_fn=lambda s: s < 10,
    update_fn=lambda s: s,       # incorrect: state never updates
    max_iter=8
)
```

This outputs warnings before an infinite loop occurs.

*(Maven page 21)*

```python
while bank_balance > 0:
    spending = 1000
    bank_balance -= spending      # or: bank_balance -= 1000
    month += 1
    print(f"End of month {month}: balance = {bank_balance}")
```

This represents a **linear depletion model**, useful in:

- Project budgeting

- Inventory depletion

# 7. 🟦 Advanced Helper: State Timeline Recorder

This helper logs every state value for later visualization or tabular analysis.

```python
def while_timeline(start_state, condition_fn, update_fn, max_iter=100):
    """
    Records each iteration's state—ideal for simulations, 
    convergence testing, and algorithm diagnostics.
    """
    timeline = []
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        timeline.append((iteration, state))
        
        iteration += 1
        if iteration > max_iter:
            timeline.append(("ABORTED", state))
            break
        
        state = update_fn(state)
    
    timeline.append(("FINAL", state))
    return timeline
```

### Example

```python
timeline = while_timeline(
    start_state=800000,
    condition_fn=lambda s: s < 1_000_000,
    update_fn=lambda s: s * 1.05,
    max_iter=10
)

timeline
```

Produces a list of iteration → state pairs, perfect for later DataFrame conversion.

---

# 8. When to Choose a While Loop

Use a **while loop** when:

- Termination is based on *reaching a threshold*, not a known count.  
- You are performing **search**, **waiting**, or **simulation until solved**.  
- You want to protect against infinite loops using counters or safety guards.  
- You need maximum flexibility in defining stopping logic.

Use a **for loop** when:

- You know the dataset length.  
- You want guaranteed finite iteration.  
- You are looping over user-defined collections.

---

# Key Takeaways

- A **while loop** repeats until its logical expression becomes False.  
- It requires **state initialization**, **condition**, and **update** to avoid runaway execution.  
- While loops are ideal for **simulations, threshold-based tasks, and unknown-duration processes**.  
- Infinite loops occur when the condition never becomes False; always include a **state update** or **safety counter**.  
- The new helpers —  
  - `safe_while_trace()`  
  - `while_timeline()`  
provide controlled exploration of while-loop behavior for learning and debugging.

### `deep_loop_state_visualizer()` — Full State Evolution for List-Processing Loops  

Shows:  
- iteration number  
- the current value pulled from the iterable  
- the computed result for that iteration
- the *entire evolving accumulator* (e.g., a list) after each append  
- how data transformations unfold step-by-step

NOTE:
- this helper is useful for learning and debbugging
- it exposes each step of an accumulator-based loop that is normally hidden inside a standard loop
- 

This helper is especially useful for understanding loops that *build new datasets*, such as currency conversion, filtering, ETL steps, or feature engineering.


```python
def deep_loop_state_visualizer(iterable, transform_fn):
    """
    Visualizes the full state evolution of a loop that builds
    an accumulator (e.g., a list) from an iterable.
    
    Parameters
    ----------
    iterable : iterable
        The source data being looped over.
    
    transform_fn : callable
        A function applied to each element of the iterable
        to produce the value appended to the accumulator.
    
    Reveals
    -------
    - iteration number
    - current input value
    - transformed value
    - full accumulator state after each append
    """
    accumulator_list = []
    
    print("DEEP LOOP STATE VISUALIZER\n--------------------------")
    
    for iteration, current_value in enumerate(iterable):
        transformed_value = transform_fn(current_value)
        accumulator_list.append(transformed_value)
        
        print(f"Iteration {iteration}:")
        print(f"  current_value     = {current_value!r}")
        print(f"  transformed_value = {transformed_value!r}")
        print(f"  accumulator_list  = {accumulator_list}\n")
    
    print("Final Output:", accumulator_list)
    return accumulator_list
```
Example usage:
```python
prices = [5.99, 9.99, 19.99]
exchange_rate = 0.88

deep_loop_state_visualizer(
    prices,
    transform_fn=lambda x: round(x * exchange_rate, 2)
)
```

### Helper `validate_alignment()` - Validate alignment before looping

```python

def validate_alignment(*iterables):
    """
    Validates that all provided iterables have the same length.

    This helper is intended to catch silent logic errors
    when using index-based looping across multiple iterables.

    Raises
    ------
    ValueError
        If the iterables are not aligned in length.
    """
    if not iterables:
        raise ValueError("At least one iterable must be provided.")
    
    lengths = [len(it) for it in iterables]
    
    if len(set(lengths)) != 1:
        raise ValueError(
            f"Iterables are not aligned in length: {lengths}"
        )
    
    print("Iterables are aligned by length. Safe to proceed.")
```
### Example Usage:
```python
items = ["Snowboard", "Boots", "Helmet"]
prices = [499, 299, 99]
ratings = [5, 4, 4]

validate_alignment(items, prices, ratings)
```
### NOTE: 
- Length alignment is required for index-based looping.
- This helper prevents silent misalignment bugs that produce incorrect results without raising errors.
### In practice: 
- Prefer zip() when all iterables are equal peers
- Prefer structured data for long-term robustness

### `sync_lists()` — Synchronized iteration across multiple equal-length iterables

```python
def sync_lists(*iterables):
    """
    Prints synchronized elements across multiple iterables
    that are assumed to be positionally aligned.

    This helper is intended for diagnostics and learning,
    not for production workflows.

    Raises
    ------
    ValueError
        If the iterables do not all have the same length.
    """
    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
    
    if not iterables:
        raise ValueError("At least one iterable must be provided.")
    
    length = len(iterables[0])
    
    for it in iterables:
        if len(it) != length:
            raise ValueError(
                "All iterables must have the same length for synchronized iteration."
            )
    
    for idx in range(length):
        row = [it[idx] for it in iterables]
        print(f"Index {idx}: {row}")
```
### Usage:

```python
items   = ["Snowboard", "Boots", "Helmet"]
prices  = [499, 299, 99]
ratings = [5, 4, 4]

sync_lists(items, prices, ratings)
```
### Output conceptually:
```
Index 0: ['Snowboard', 499, 5]
Index 1: ['Boots', 299, 4]
Index 2: ['Helmet', 99, 4]
```

# Summary

- A **loop** repeats instructions until a **count-based condition** (for loop) or a **logical condition** (while loop) signals completion.  
- **For loops** are used when the number of iterations is known or when the task is “one pass per element.”  
  They may iterate directly over values, over **indices** when positional control is required, or over **index–value pairs** using `enumerate()` for improved clarity and safety.  
- **While loops** provide greater flexibility but must be carefully designed to avoid infinite loops.  
- This section introduces the core loop helpers:  
  - **`loop_trace_for()`** — traces item-by-item for-loop iteration and execution order  
  - **`enumerate_visualizer()`** — makes index–value pairing explicit when using `enumerate()`  
  - **`safe_while_trace()`** — safely traces while-loop state evolution and prevents infinite loops
  - **``while_timeline()``** - advanced helper logs every state value for later visualization or tabular analysis 
  - **`deep_loop_state_visualizer()`** — traces each internal step of a list-processing loop, including variable assignment, calculation execution, each `.append()` operation, and the evolving state of the accumulator container
  - **`validate_alignment`** - Validate iterable alignment before looping. (Length alignment is required for index-based looping)
  - **`sync_lists()`** — synchronized iteration across multiple equal-length iterables 
- Understanding iteration mechanics — including when and how position matters — is essential for dataset traversal, data cleaning, transformation workflows, simulation modeling, and algorithmic problem-solving.



# Chapter 07 — Loops  
## Section 7.04 — Nested Loops  

### Preview
A **nested loop** is a loop inside another loop.  
For every single iteration of the **outer loop**, the **inner loop** runs from start to finish.


> Nested loops allow you to process multi-level structures, generate combinations, and traverse data in two dimensions.

Nested loops are essential for:

- Navigating **nested lists**  
- Generating all **combinations** or **pairings**  
- Performing **matrix-like** operations  
- Processing **customer → orders** or **category → items** hierarchies  
- Applying transformations across multi-dimensional datasets  

---

## 🟦 Nested Loop Helpers Expand your Helper Toolkit

- The expanded toolkit includes:  
  - `nested_loop_trace()` visualizer
  - `transform_nested_list()`(general purpose)
  - `cartesian_product()` generator
  - `visualize_matrix()` for 2D structures

These tools give full transparency into multi-dimensional iteration.

---

## 1. Basic Nested Loop Structure

```
for item in outer_list:
    for inner_item in inner_list:
        do_something(item, inner_item)
```

Execution pattern:

- Outer iteration 0 → inner loop runs fully  
- Outer iteration 1 → inner loop runs fully  
- …  
- Outer iteration n → inner loop runs fully  

Total iterations =  
`len(outer_list) × len(inner_list)`


`````python
# --------------------------------------------------------
# Nested List Example — Item × Size Combinations
# --------------------------------------------------------
# Output count:
# - 2 items  
# - 3 sizes  
# - 2 × 3 = 6 combinations

items = ["Snowboard", "Boots"]
sizes = ["small", "medium", "large"]

# OUTER LOOP:
for item in items:
    # INNER LOOP:
    for size in sizes:
        print(f"{item} is available in {size}.")

# The outer loop moves by each list, index [0, 1]
# The inner loop processes every element for each list - [0][0], [0][1], [0][2]; [1][0], [1][1], [1][2]
# Inner loop runs fully before outer loop advances
# This demonstrates the “outer loop controls what, inner loop controls how many” principle


# --------------------------------------------------------
# 🟦 Helper: Nested Loop Trace Visualizer
# --------------------------------------------------------


def nested_loop_trace(outer, inner):
    """
    Visualizes nested loop execution by showing:
      - outer iteration index and value
      - inner iteration index and value
      - the resulting pair

    Mirrors the detailed iteration shown in Maven pp. 25–33.
    """
    print("NESTED LOOP TRACE\n------------------")
    
    for outer_idx, outer_val in enumerate(outer):
        print(f"\nOuter Iteration {outer_idx}: outer_val = {outer_val!r}")
        
        for inner_idx, inner_val in enumerate(inner):
            print(f"    Inner Iteration {inner_idx}: inner_val = {inner_val!r}")
            print(f"        → Pair = ({outer_val!r}, {inner_val!r})")

# Example usage:

nested_loop_trace(items, sizes)


# --------------------------------------------------------
# 🟦 Helper: Cartesian Product Generator
# --------------------------------------------------------
# (Useful for combinations, SKU creation, pairing simulations)

def cartesian_product(a, b):
    """
    Returns the Cartesian product of two iterables as a list of tuples.
    Equivalent to the nested-loop pairings in Maven's examples.
    """
    return [(x, y) for x in a for y in b]


# Example Usage:

cartesian_product(items, sizes)
`````

`````python
# --------------------------------------------------------
# Nested List Demo - Step-by-Step
# --------------------------------------------------------

item_reviews = [
    ["5 stars", "3 stars", "5 stars"],     # Skis
    ["4 stars", "4 stars"],                # Snowboards
    ["1 star", "4 stars"],                 # Goggles
    ["5 stars", "2 stars", "5 stars"]      # Boots
]


# --------------------------------------------------------
# Step 1. Outer list
# --------------------------------------------------------
# OUTER lOOP:
for lst in item_reviews:      # each nested list (4x)
    print(lst)
    
# --------------------------------------------------------
# Step 2. Inner list
# --------------------------------------------------------
# OUTER LOOP:
for lst in item_reviews:    # each nested list (4x)
    
    # INNER LOOP: 
    for review in lst:      # each element in each nested list (10x)
        print(review)

# --------------------------------------------------------
# Step 3. Extract Numeric Ratings Only
# --------------------------------------------------------
# OUTER LOOP:
for lst in item_reviews:             # each nested list (4x)
    
    # INNER LOOP: 
    for review in lst:               # each element in each nested list (10x)
        int_rating = int(review[0])  # extract numeric rating only
        print(int_rating)

# --------------------------------------------------------
# Step 4. Collect Numeric Ratings into Lists
# --------------------------------------------------------
# OUTER LOOP:
for lst in item_reviews:                     # each nested list (4x)
    integer_ratings = []                     # empty list accumulator: ratings in each nested list
    
    # INNER LOOP: 
    for review in lst:                       # each element in each nested list (10x)
        int_rating = int(review[0])          # extract numeric rating only
        integer_ratings.append(int_rating)   # .append() accumulator list
    print(integer_ratings)

# --------------------------------------------------------
# Step 5. Collect Numeric Ratings Back into a Nested List
# --------------------------------------------------------
numeric_item_reviews = []                           # empty list: collect numeric values back into a nested list

# OUTER LOOP:
for lst in item_reviews:                           # each nested list (4x)
    integer_ratings = []                           # empty list accumulator: ratings in each nested list
    
    # INNER LOOP: 
    for review in lst:                             # each element in each nested list (10x)
        int_rating = int(review[0])                # extract numeric rating only
        integer_ratings.append(int_rating)         # .append() accumulator list
    
    numeric_item_reviews.append(integer_ratings)   # Collect back into a nested list

numeric_item_reviews                               # return nested list with numeric values only


# --------------------------------------------------------
# Step 6. Final Pairing with Items
# --------------------------------------------------------
items = ["Skis", "Snowboards", "Goggles", "Boots"]

for item, ratings in zip(items, numeric_item_reviews):
    print(f"{item}: {ratings}")
`````

`````python
# --------------------------------------------------------
# Understand outer vs. inner loop roles
# Exercise: Nested Loops — Processing Reviews by Item
# --------------------------------------------------------
# Practice navigating nested lists
# Transform raw string data into numeric, analyzable structure
# Prepare data for aggregation (averages)


item_reviews = [
    ["5 stars", "3 stars", "5 stars"],     # Skis
    ["4 stars", "4 stars"],                # Snowboards
    ["1 star", "4 stars"],                 # Goggles
    ["5 stars", "2 stars", "5 stars"]      # Boots
]

item_reviews_numeric = []                  # outer loop accumulator

# OUTER LOOP:
for each_nested_lst in item_reviews:       # each nested list in item_reviews (4x)
    
    numeric_ratings = []                   # inner loop accumulator
    
    # INNER LOOP:
    for element in each_nested_lst:         # for element in nested list (10x)
        rating = int(element[0])            # extract numeric rating
        numeric_ratings.append(rating)
    
    # collect processed ratings for this item
    item_reviews_numeric.append(numeric_ratings)
    
    print(item_reviews_numeric)             # shows each iteration step-by-step

item_reviews_numeric

# The outer loop moves by each nested list, indices [0], [1], [2], [3]
# The inner loop processes every element for each nested list - [0][0], [0][1], [0][2]; [1][0], [1][1]; etc.
# The inner loop fully completes before the outer loop advances


# --------------------------------------------------------
# Aggregate results
# --------------------------------------------------------

# Calculate average review score per item
for ratings in item_reviews_numeric:
    average = sum(ratings) / len(ratings)
    print(round(average, 2))

# --------------------------------------------------------
# Pair results with item names
# --------------------------------------------------------

items = ["Skis", "Snowboards", "Goggles", "Boots"]

for i in range(len(items)):
    avg_rating = sum(item_reviews_numeric[i]) / len(item_reviews_numeric[i])
    print(f"{items[i]}: average rating = {avg_rating:.2f}")
`````

`````python
# --------------------------------------------------------
# Nested Loops Over Nested Lists
# Example — Apply 10% discount to each customer’s orders
# --------------------------------------------------------
# This pattern is used when dealing with multi-order customers, grouped transactions, or any list-of-lists structure.
# This preserves the **structure** of the original dataset — a common requirement in analytics workflows.


orders = [
    [120.00, 45.00, 89.99],        # customer 1 orders
    [250.00, 80.00],               # customer 2 orders
    [19.99, 55.00, 150.00, 40.00]  # customer 3 orders
]

discounted_orders = []                        # outer loop accumulator

for transactions in orders:                   # outer loop: each nested list (3x)
    customer_discounted = []                  # inner loop accumulator
    
    for subtotal in transactions:             # inner loop: for element in nested list (9x)
        new_total = round(subtotal * 0.9, 2)  # 10% discount (or 90% of the price)
        customer_discounted.append(new_total)
    
    discounted_orders.append(customer_discounted)

discounted_orders

# --------------------------------------------------------
# 🟦 Helper: Matrix Visualizer  
# --------------------------------------------------------

def visualize_matrix(matrix):
    """
    Prints a clean, aligned representation of a nested list
    to help visualize 2D structures created by nested loops.
    """
    print("MATRIX VIEW\n-----------")
    for row_idx, row in enumerate(matrix):
        print(f"Row {row_idx}: {row}")

# Example Usage:

visualize_matrix(discounted_orders)
print()


# --------------------------------------------------------
# 🟦 Helper: Nested List Transformer (General-Purpose)
# --------------------------------------------------------

orders = [
    [120.00, 45.00, 89.99],        # customer 1 orders
    [250.00, 80.00],               # customer 2 orders
    [19.99, 55.00, 150.00, 40.00]  # customer 3 orders
]

def transform_nested_list(nested, transform_fn):
    """
    Applies a user-defined transformation to each element in a nested list.

    Parameters
    ----------
    nested : list[list]
        A list of lists (e.g., customers → orders).
    
    transform_fn : callable
        Function applied to each inner element.

    Returns
    -------
    list[list]
        A new nested list with the same structure.

    Notes
    -----
    - General version of Maven's multi-order discount logic (pp. 32–33)
    - Works with any transformation: tax, normalization, scaling, flags, etc.
    """
    
    result = []                                  # outer loop accumulator
    
    for i, inner_list in enumerate(nested):      # outer loop
        transformed_inner = []                   # inner loop accumulatore
        
        for j, value in enumerate(inner_list):   # inner loop
            transformed_inner.append(transform_fn(value))
        
        result.append(transformed_inner)
    
    return result


# Example usage:
result = transform_nested_list(orders, lambda x: round(x * 0.9, 2))
# nothing will display unless you print
print(result)

result = transform_nested_list(orders, lambda x: round(x * 0.9, 2))
print("Discounted orders:", result)

# Jupyter Note:
# Only the final expression in a cell is displayed automatically.
# Functions return data silently unless print() is used.
`````

```python
def transform_nested_list(nested, transform_fn):
    """
    Applies a user-defined transformation to each element in a nested list.

    Parameters
    ----------
    nested : list[list]
        A list of lists (e.g., customers → orders).
    
    transform_fn : callable

    Returns
    -------
    list[list]
        A new nested list with the same structure.

    Notes
    -----
    - General version of Maven's multi-order discount logic (pp. 32–33)
    - Works with any transformation: tax, normalization, scaling, flags, etc.
    """
    
    result = []
    
    for i, inner_list in enumerate(nested):      # outer loop
        transformed_inner = []
        
        for j, value in enumerate(inner_list):   # inner loop
            transformed_inner.append(transform_fn(value))
        
        result.append(transformed_inner)
    
    return result
```

# 8. Understanding Nested Loop Complexity  
Nested loops typically have **O(n × m)** time complexity.
Total operations: n × m → O(n × m)

Examples:

- Items (2) × Sizes (3) → 6 operations → O(2 × 3)  
- Customers (3) × Orders (vary 2–4) → total operations = sum of row lengths. (Count how many total elements exist across all inner lists).

- If the inner loop length is fixed → total work = n × m.
- If the inner loop length varies → total work = sum of inner list lengths.
 

Performance matters when:

- Processing large datasets  
- Performing multi-level simulations  
- Building grid-style computations  

---

# 9. When to Use Nested Loops

Use nested loops when:

- You must examine every pairing in a 2D structure  
- Your data is multi-level (e.g., list of lists)  
- You need all combinations of two sets  
- You are generating structured outputs (e.g., matrix, grid, report sections)

Avoid nested loops when:

- A vectorized or built-in approach (NumPy, pandas) is available  
- Complexity becomes too high for large datasets  
- Iteration order does not matter (sometimes flattening is simpler)

---

# Key Takeaways

- A **nested loop** executes the inner loop *once per iteration* of the outer loop.  
- This produces a complete pairing (Cartesian product) of both iterables.  
- Nested loops are essential for processing **nested lists**, **combinations**, and **matrix-like structures**.  
- The expanded toolkit includes:  
  - `nested_loop_trace()`  
  - `transform_nested_list()`  
  - `cartesian_product()`  
  - `visualize_matrix()`  
These tools give full transparency into multi-dimensional iteration.  
- Nested iteration can be powerful but costly; understand when it is necessary and when alternatives exist.



# Chapter 07 — Loops  
## Section 7.05 — Loop Control

### Preview
Loop control statements allow you to **modify the normal flow of iteration**, enabling early exits, selective skipping, placeholder logic, and robust error handling.

Four essential loop control mechanisms include:

- **break** — exit a loop immediately  
- **continue** — skip the rest of the current iteration  
- **pass** — placeholder that does nothing  
- **try / except** — handle errors without stopping the loop  

These tools are critical for:

- Data cleaning  
- Error-tolerant ETL pipelines  
- Complex filtering  
- Avoiding infinite loops  
- Creating maintainable, incremental logic  

This section adds new helper functions for debugging, error tracing, and controlled loop execution.

---

# 🟦 Helper Box: Loop Control Helpers 
- The helper toolkit now includes: 
  - `break_trace()`  
  - `continue_trace()`  
  - `try_except_trace()`  
  - `affordability_filter()`  
These provide full transparency and robustness when working with real-world, messy datasets.

---

# Summary

- Loop control changes the **execution path** inside loops to handle selective skipping, early termination, and error-tolerant iteration.  
- `break` stops the loop immediately.  
- `continue` skips the current iteration only.  
- `pass` does nothing; used as a placeholder.  
- `try`/`except` creates fault-tolerant loops that keep running even when data issues occur.


`````python
# ------------------------------------------------------------------
# 1. The `break` Statement
# `break` stops the loop entirely, regardless of the iteration index
# ------------------------------------------------------------------
# Behavior
# - Loop ends as soon as the condition triggers  
# - Remaining items are not processed  


# Example — Stop when revenue exceeds 2000  

subtotals = [200.98, 350.97, 500, 400, 300, 600, 150, 100, 90]

revenue = 0                          # accumulator to track what the loop produces

for amount in subtotals:
    revenue += amount                # looping through subtotals list and adding revenue
    print(round(revenue,2))          # Visualize each loop
    if revenue > 2000:
        print("Target reached. Stopping early.")
        break

print("Final revenue:", revenue)


# -------------------------------------
# 🟦 Helper — Break Trace Visualizer
# -------------------------------------

def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0
    for idx, value in enumerate(iterable):
        running += value
        print(f"Iteration {idx}: value = {value}, running = {running}")
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break

# Example Usage:
break_trace(subtotals, lambda total: total > 2000)


# Use :.2f formatting in helpers when working with currency or financial data.
# This improves readability and prevents confusion caused by floating-point precision.


# -------------------------------------
# 🟦 Helper — Break Trace Visualizer
# -------------------------------------

def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    Values are formatted to two decimal places for currency-style data.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0.0
    for idx, value in enumerate(iterable):
        running += value
        print(
            f"Iteration {idx}: "
            f"value = ${value:.2f}, "
            f"running = ${running:.2f}"
        )
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break
            
# Example Usage:
break_trace(subtotals, lambda total: total > 2000)
`````

`````python
# ------------------------------------------------------------------
# 1. The `break` Statement — Early Exit from a While Loop
# ------------------------------------------------------------------
# `break` immediately stops a loop, even if the loop condition
# would otherwise continue to evaluate as True.
#
# Use `break` when:
# - you want to stop early once a secondary condition is met
# - continuing the loop no longer makes sense
# - you want to prevent runaway or unrealistic scenarios
# ------------------------------------------------------------------


stock_portfolio = 21_070   # accumulator: tracks the evolving portfolio value (loop state)
year_counter = 0          # counter: tracks how many iterations (years) have passed


# This loop continues WHILE the portfolio is below the target
while stock_portfolio < 150_000:
    
    # Calculate annual investment growth (15% return)
    investment_income = stock_portfolio * 0.15
    
    # Update portfolio value (state change)
    stock_portfolio += investment_income           # accumulator update
    
    # Update loop counter
    year_counter += 1                             # counter update
    
    # Observe the evolving state of the system
    print(
        f"My balance is ${stock_portfolio:,.2f} "
        f"in year {year_counter}"
    )
    
    # SAFETY CHECK:
    # If 10 years have passed and the goal is still not reached,
    # exit the loop early instead of continuing indefinitely.
    if year_counter >= 10:
        print("Goal not reached in 10 years — savings rate may be too low.")
        break

# Teaching takeaway:
# - The while-condition controls whether the loop *can* continue
# - The `break` controls whether the loop *should* continue
`````

`````python
# ------------------------------------------------------------------
# 2. The `continue` Statement  
# `continue` skips the remainder of the current iteration but continues the loop
# ------------------------------------------------------------------
# Behavior
# - `"skis"` and `"skiboots"` are skipped  
# - All other items processed normally
# ------------------------------------------------------------------

# Example — Filter out items containing “ski”


inventory = ["snowboard", "skis", "helmet", "skiboots", "bindings"]
filtered_inventory = []

for item in inventory:
    # filter condition: skip the current item if it contains the substring "ski"
    if "ski" in item:
        # continue means: "do NOT run the rest of the loop body for this item"
        # and move to the next iteration immediately
        continue
    filtered_inventory.append(item)  # .append() empty list accumulator

print(filtered_inventory)
print()


# ---------------------------------------
# 🟦 Helper — Continue Trace Visualizer
# ---------------------------------------

def continue_trace(iterable, skip_fn):
    """
    Prints all iterations and highlights which ones are skipped via 'continue'.
    """
    print("CONTINUE TRACE\n----------------")
    
    for idx, item in enumerate(iterable):
        if skip_fn(item):
            print(f"Iteration {idx}: SKIPPED ({item!r})")
            continue
        
        print(f"Iteration {idx}: processed {item!r}")


# Example Usage:
continue_trace(inventory, lambda x: "ski" in x)
`````

`````python
# ------------------------------------------------------------------
# The `pass` Statement
# `pass` is a placeholder that does NOTHING, but keeps the code valid.
# ------------------------------------------------------------------
# Key idea:
# - `pass` satisfies Python’s requirement that a block cannot be empty
# - It does NOT skip an iteration (that’s `continue`)
# - It does NOT stop a loop (that’s `break`)
# - It simply says: “I’ll fill this in later”
#
# Common use cases:
# - Stubbing logic during development
# - Sketching control flow before implementation
# - Keeping loops and conditionals syntactically valid
# - Incrementally building complex pipelines
# ------------------------------------------------------------------


products = ["Snowboard", "Boots", "Helmet"]

for p in products:
    if p == "Boots":
        pass    # Placeholder: no action yet, loop continues normally
    else:
        print("Processing:", p)


# NOTE:
# `pass` is not control flow.
# It is a structural placeholder.
#

# CONTROL FLOW MENTAL MODEL:
# - break     → exit the loop entirely
# - continue  → skip an iteration
# - pass      → do nothing (for now), but keep structure valid
`````

`````python
# ------------------------------------------------------------------
# Error Handling with `try` / `except`  
# These statements allow the loop to continue running after encountering an error
# ------------------------------------------------------------------
# Notes
# - Strings like `"74.99"` are converted successfully  
# - `0` triggers `ZeroDivisionError`  
# - `None` triggers `TypeError`  
# - Loop does not stop on errors  
# ------------------------------------------------------------------

# ---------------------------------------
# Example 1 — Problematic price list
# ---------------------------------------

price_list = [99.99, "74.99", 0, None, 20]

clean_prices = []

for element in price_list:
    try:
        # Step 1: convert to a numeric value
        numeric_price = float(element)
        
        # Step 2: use the validated value for calculations
        clean_prices.append(round(1 / numeric_price, 4))
    
    except ZeroDivisionError:                                  # zero division error
        print("ZeroDivisionError: price was zero. Skipping.")
    
    except (TypeError, ValueError):                            # multiple error types
        print(f"Invalid price format ({element}). Skipping.")

# ---------------------------------------
# 🟦 Helper — Error Trace Loop Inspector  
# Shows which items cause which errors.
# ---------------------------------------

def try_except_trace(iterable, transform_fn):
    """
    Executes transform_fn on each element, catching errors
    and reporting them with full iteration context.
    """
    print("TRY/EXCEPT TRACE\n-----------------")
    results = []
    
    for idx, value in enumerate(iterable):
        try:
            result = transform_fn(value)
            results.append(result)
            print(f"Iteration {idx}: OK → {result}")
        
        except Exception as e:
            print(f"Iteration {idx}: ERROR → {type(e).__name__}: {e}")
    
    return results


# Example Usage:
try_except_trace(price_list, lambda x: round(1 / float(x), 4))
print()

# ---------------------------------------
# Example 2 — Problematic ratings list
# ---------------------------------------

ratings = ["1 star", "5 star", 
           "3 star", "star 5", 
           "4 star"
]

for rating in ratings:
    try:
        print(int(rating[0]))
    except ValueError:
        print("incorrect data") 



# STUDENT NOTE:
# This helper shows that errors do NOT stop loops by default
# when handled with try/except.
#
# Each iteration is attempted independently:
# - Valid values are transformed and collected
# - Invalid values are reported and skipped
#
# This mirrors real-world data cleaning workflows.
`````

`````python
# ------------------------------------------------------------------
# Error Handling with `try` / `except`  
# These statements allow the loop to continue running after encountering an error
# ------------------------------------------------------------------
# Notes
# - Strings like `"74.99"` are converted successfully  
# - `0` triggers `ZeroDivisionError`  
# - `None` triggers `TypeError`  
# - Loop does not stop on errors  
# ------------------------------------------------------------------

# ---------------------------------------
# Example 3 — Problematic price list
# ---------------------------------------


prices = [5.99, 99.99, "74.99", 0, None, 20, 24.99]
budget = 50

for price in prices:
    if price is None:
        continue

    try:
        numeric_price = float(price)  # convert early (handles strings)

        if numeric_price == 0:
            print("This product is free.")
            continue

        affordable_quantity = budget // numeric_price
        print(f"I can buy {int(affordable_quantity)} of these.")

    except ValueError:
        print("That is not a number.")
    except TypeError:
        # TypeError is unlikely once you skip None, but included for completeness
        print("Invalid type for price.")


# ---------------------------------------
# 🟦 Helper — Robust Affordability Calculator  
# ---------------------------------------

def affordability_filter(prices, budget):
    """
    Returns a list of (idx, raw_price, numeric_price, quantity) for prices that can be evaluated.
    Skips None, non-numeric, and handles zero safely.
    """
    results = []

    for idx, price in enumerate(prices):
        if price is None:
            continue

        try:
            numeric_price = float(price)

            if numeric_price == 0:
                results.append((idx, price, numeric_price, "FREE"))
                continue

            qty = int(budget // numeric_price)
            results.append((idx, price, numeric_price, qty))

        except (ValueError, TypeError):
            results.append((idx, price, None, "INVALID"))

    return results


# Example Usage:
rows = affordability_filter(prices, budget)

for idx, raw, num, qty in rows:
    print(f"Index {idx}: raw={raw!r} | numeric={num!r} | qty={qty}")



# ---------------------------------------
# 🟦 Helper — Error Trace Loop Inspector  
# Shows which items cause which errors.
# ---------------------------------------

def try_except_trace(iterable, transform_fn):
    """
    Executes transform_fn on each element, catching errors
    and reporting them with full iteration context.
    """
    print("TRY/EXCEPT TRACE\n-----------------")
    results = []
    
    for idx, value in enumerate(iterable):
        try:
            result = transform_fn(value)
            results.append(result)
            print(f"Iteration {idx}: OK → {result}")
        
        except Exception as e:
            print(f"Iteration {idx}: ERROR → {type(e).__name__}: {e}")
    
    return results


# Example Usage:
try_except_trace(
    prices,
    lambda p: "FREE" if float(p) == 0 else int(budget // float(p))
)
`````

`````python
# --------------------------------------------------------
# ✅ Analysis-Ready Affordability Toolkit
# - robust parsing (None, strings, bad values)
# - flags FREE / INVALID
# - computes quantity you can buy under a budget
# - returns both detailed rows + a summary dictionary
# --------------------------------------------------------

def affordability_analyze(prices, budget):
    """
    Analyze a mixed-quality price list against a budget.

    Returns
    -------
    rows : list[dict]
        One record per input element with diagnostics:
        idx, raw, numeric_price, status, quantity
    summary : dict
        Counts + quick stats to support analysis decisions.
    """
    rows = []

    # Summary counters
    counts = {
        "OK": 0,
        "FREE": 0,
        "INVALID": 0,
        "SKIPPED_NONE": 0
    }

    valid_prices = []
    valid_quantities = []

    for idx, raw in enumerate(prices):
        # 1) Skip None cleanly (common missing-data pattern)
        if raw is None:
            counts["SKIPPED_NONE"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": None,
                "status": "SKIPPED_NONE",
                "quantity": None
            })
            continue

        # 2) Attempt numeric conversion (handles floats/ints/strings like "74.99")
        try:
            numeric_price = float(raw)
        except (TypeError, ValueError):
            counts["INVALID"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": None,
                "status": "INVALID",
                "quantity": None
            })
            continue

        # 3) Handle zero explicitly (avoid division-by-zero)
        if numeric_price == 0:
            counts["FREE"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": 0.0,
                "status": "FREE",
                "quantity": "FREE"
            })
            continue

        # 4) Negative prices are “valid numbers” but typically invalid business data
        if numeric_price < 0:
            counts["INVALID"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": numeric_price,
                "status": "INVALID_NEGATIVE",
                "quantity": None
            })
            continue

        # 5) OK path: compute how many you can buy (floor division)
        qty = int(budget // numeric_price)

        counts["OK"] += 1
        valid_prices.append(numeric_price)
        valid_quantities.append(qty)

        rows.append({
            "idx": idx,
            "raw": raw,
            "numeric_price": numeric_price,
            "status": "OK",
            "quantity": qty
        })

    # -----------------------------
    # Build summary stats
    # -----------------------------
    summary = {
        "budget": budget,
        "total_inputs": len(prices),
        "counts": counts,
        "valid_price_count": len(valid_prices),
        "min_valid_price": min(valid_prices) if valid_prices else None,
        "max_valid_price": max(valid_prices) if valid_prices else None,
        "avg_valid_price": (sum(valid_prices) / len(valid_prices)) if valid_prices else None,
        "avg_quantity": (sum(valid_quantities) / len(valid_quantities)) if valid_quantities else None,
        "max_quantity": max(valid_quantities) if valid_quantities else None,
    }

    return rows, summary


# --------------------------------------------------------
# ✅ Pretty-printer (optional, keeps your notebook readable)
# --------------------------------------------------------

def print_affordability_report(rows, summary):
    print("AFFORDABILITY REPORT\n--------------------")
    print(f"Budget: ${summary['budget']:,.2f}")
    print(f"Total inputs: {summary['total_inputs']}\n")

    # High-level summary
    c = summary["counts"]
    print("Counts:")
    for k, v in c.items():
        print(f"  {k:>12}: {v}")
    print()

    # Quick stats
    if summary["valid_price_count"] > 0:
        print("Valid price stats:")
        print(f"  min price : ${summary['min_valid_price']:,.2f}")
        print(f"  max price : ${summary['max_valid_price']:,.2f}")
        print(f"  avg price : ${summary['avg_valid_price']:,.2f}")
        print(f"  avg qty   : {summary['avg_quantity']:.2f}")
        print(f"  max qty   : {summary['max_quantity']}")
    else:
        print("No valid prices found.")
    print()

    # Row-by-row detail (traceability)
    print("Row detail:")
    for r in rows:
        idx = r["idx"]
        raw = r["raw"]
        num = r["numeric_price"]
        status = r["status"]
        qty = r["quantity"]

        # consistent formatting
        raw_txt = repr(raw)
        num_txt = "None" if num is None else f"{num:.2f}"
        print(f"  idx {idx:>2}: raw={raw_txt:<10} | numeric={num_txt:>6} | status={status:<15} | qty={qty}")


# --------------------------------------------------------
# ✅ Example usage (your dataset)
# --------------------------------------------------------

prices = [5.99, 99.99, "74.99", 0, None, 20, 24.99]
budget = 50

rows, summary = affordability_analyze(prices, budget)
print_affordability_report(rows, summary)
`````

# Chapter 07 — Loops  
## Section 7.05 — Loop Control

### Overview
Loop control statements allow you to **modify the normal flow of iteration**, enabling early exits, selective skipping, placeholder logic, and robust error handling.

Four essential loop control mechanisms include:

- **break** — exit a loop immediately  
- **continue** — skip the rest of the current iteration  
- **pass** — placeholder that does nothing  
- **try / except** — handle errors without stopping the loop  

These tools are critical for:

- Data cleaning  
- Error-tolerant ETL pipelines  
- Complex filtering  
- Avoiding infinite loops  
- Creating maintainable, incremental logic  

This section adds new helper functions for debugging, error tracing, and controlled loop execution.

---

# 1. The `break` Statement  

`break` **stops the loop entirely**, regardless of the iteration index.

### Example — Stop when revenue exceeds 2000  
(Maven scenario: revenue becomes high enough before list ends)

```python
subtotals = [200, 350, 500, 400, 300, 600, 150, 100, 90]

revenue = 0

for amount in subtotals:
    revenue += amount
    
    if revenue > 2000:
        print("Target reached. Stopping early.")
        break

print("Final revenue:", revenue)
```

### Behavior
- Loop ends as soon as the condition triggers  
- Remaining items are **not** processed  

### 🟦 Helper — Break Trace Visualizer

```python
def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0
    for idx, value in enumerate(iterable):
        running += value
        print(f"Iteration {idx}: value = {value}, running = {running}")
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break
```

Usage:

```python
break_trace(subtotals, lambda total: total > 2000)
```

---

# 2. The `continue` Statement  

`continue` **skips the remainder of the current iteration** but continues the loop.

### Example — Filter out items containing “ski”

```python
inventory = ["snowboard", "skis", "helmet", "skiboots", "bindings"]
snowboards = []

for item in inventory:
    if "ski" in item:      # skip skis and skiboots
        continue
    snowboards.append(item)

snowboards
```

### Behavior

- All other items processed normally

```python
            print(f"Iteration {idx}: SKIPPED ({item!r})")
```

### Example — Outline structure without full logic

```python
products = ["Snowboard", "Boots", "Helmet"]

for p in products:
    if p == "Boots":
        pass    # logic will be added later
    else:
        print("Processing:", p)
```

Use cases:

- Stubbing logic during development  
- Maintaining syntactically valid loops  
- Gradually building complex pipelines

# 4. Error Handling with `try` / `except`  

These statements allow the loop to **continue running after encountering an error**.

### Example — Problematic price list

```python
price_list = [99.99, "74.99", 0, None, 20]

clean_prices = []

for price in price_list:
    try:
        # Convert strings to float and validate nonzero price
        numeric_price = float(price)
        clean_prices.append(round(1 / numeric_price, 4))
    
    except ZeroDivisionError:
        print("ZeroDivisionError: price was zero. Skipping.")
    
    except (TypeError, ValueError):
        print(f"Invalid price format ({price}). Skipping.")
```

### Notes
- Strings like `"74.99"` are converted successfully  
- `0` triggers `ZeroDivisionError`  
- `None` triggers `TypeError`  
- Loop does **not stop** on errors

# 6. 🟦 Helper — Robust Affordability Calculator  


```python
def affordability_filter(prices, budget):
    """
    Robust affordability calculator:
    - skips None values
    - converts string prices
    - handles invalid formats safely
    - avoids division-by-zero issues
    """
    
    affordable = []
    
    for idx, price in enumerate(prices):
        try:
            if price is None:
                continue
            
            numeric_price = float(price)
            
            if numeric_price <= budget:
                affordable.append(numeric_price)
        
        except Exception:
            continue  # safely skip invalid price formats
    
    return affordable
```

---

# When to Use Loop Control Statements

### Use `break` when:
- A target or threshold is reached  
- Further processing is unnecessary  
- You are preventing a runaway or infinite loops early  

### Use `continue` when:
- Filtering datasets  
- Skipping invalid or unwanted entries, i.e., excluding values you don't want to process in a loop
- Differentiating “skippable” vs “processable” values

### Use `pass` when:
- Structuring logic before implementation, i.e., avoiding run errors with incomplete code logic
- Writing placeholder blocks, i.e., serving as a placeholder for future code  
- Maintaining syntactic validity  

### Use `try`/`except` when:
- Data may be incomplete, inconsistent, or malformed  
- You need to avoid interruption during iteration
- You need help with error and exception handling  
- Building robust ETL or cleaning pipelines  

---

# Key Takeaways

- Loop control changes the **execution path** inside loops to handle selective skipping, early termination, and error-tolerant iteration.  
- `break` stops the loop immediately.  
- `continue` skips the current iteration only.  
- `pass` does nothing; used as a placeholder.  
- `try`/`except` creates fault-tolerant loops that keep running even when data issues occur.  
- The helper toolkit now includes:  
  - `break_trace()`  
  - `continue_trace()`  
  - `try_except_trace()`  
  - `affordability_filter()`  
These provide full transparency and robustness when working with real-world, messy datasets.
