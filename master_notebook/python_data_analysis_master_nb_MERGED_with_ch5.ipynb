## 🧠 **Chapter 1 Summary: Data Types in Python**
## **Key Takeaways:**

- Python includes several core data type categories: **numeric, text, sequence, mapping, set, boolean, and NoneType**.  
- The `type()` function identifies an object's data type and is essential for debugging when operations fail due to incompatible types.  
- Type conversion uses the format `datatype(object)` and succeeds only when the value is compatible (e.g., `"123"` → `int`, but `[1,2,3]` → `int` fails).  
- **Iterables** include strings, lists, tuples, dictionaries, ranges, sets, and frozensets—any object you can loop through.  
- **Strings are immutable yet iterable**, meaning characters cannot be changed in place but can be looped over.  
- **Mutable types** (lists, sets, dictionaries) can be changed after creation.  
  **Immutable types** (numbers, strings, tuples, booleans, frozensets, NoneType) cannot.  
- Dictionaries store **key-value pairs**, forming the backbone of JSON-style structured data widely used in analytics.  
- Sets store **unique values only**, removing duplicates automatically during creation or conversion.  
- Curly braces `{}` create:  
  - a **set** when values are comma-separated  
  - a **dictionary** when using key-value pairs  
  - an empty dictionary when used as `{}`  
- An empty set must be created with **`set()`**, since `{}` defaults to a dictionary.  
- Understanding mutability, iterability, and data type behavior is foundational for working with **Pandas, NumPy, data cleaning, merging, grouping, and transformation workflows**.


# 🧭 Chapter 2. Variables: Overview

Variables are containers that store values for later use in a Python program. They can store any data type, including numbers, strings, lists, dictionaries, and even function calls. Understanding how to assign, overwrite, delete, and track variables is essential for writing clear and maintainable code.

A variable assignment uses the format:

```
variable_name = value
```

Variable assignment allows you to label values, reuse them, and perform operations without hard coding every value. Naming conventions and clear organization help make code easier to follow. This section builds on the concepts from the Maven Analytics notes by demonstrating assignment, overwriting, deleting variables, and tracking variables in a Jupyter environment.

---

# 🧪 Naming Rules and Conventions

Variable names:

- Can contain letters, numbers, and underscores  
- Must begin with a letter or underscore  
- Cannot begin with a number  
- Cannot contain spaces or special characters  
- Cannot use reserved keywords such as `del`, `list`, or `for`

Snake case is the preferred Python naming style:

- `price_list`
- `total_revenue`
- `snowboard_inventory_2024`

---


`````python
# -----------------------------------------------------------
# 🔹 Basic Variable Assignment
# -----------------------------------------------------------

# Assigning simple values to variables
price = 5
city = "Los Angeles"
heights = [180, 173, 191]

print(price)          # prints: 5
print(city)           # prints: "Los Angeles"
print(heights)        # prints: [180, 173, 191]

# Using a variable in an operation
print(price + 1)      # prints: 6 (adds 1 to the value stored in 'price')
# adding 1 is hardcoded (e.g., set fee)
# or can it can be a variable for values that can change
`````

`````python
# -----------------------------------------------------------
# 🔹 Assigning a Function Call to a Variable
# -----------------------------------------------------------

price_list = [10.99, 12.49, 8.75]   
result = print(price_list)          
# 'result' stores the return value of print(), which is None

print(result)         # prints: None
# ⚠️ Note: Understanding which functions return values is essential (see markdown)
`````

`````python
# -----------------------------------------------------------
# 🔹 Overwriting a Variable
# -----------------------------------------------------------

# -------------------------------------------------------------
# 1️⃣ Simple overwrite: Python stores the most recent value
# -------------------------------------------------------------
price = 5
price = 6
price = 7           # last assignment "wins"
print("Simple overwrite:", price)   # 7


# -------------------------------------------------------------
# 2️⃣ Keeping both old and new values (good for testing)
# -------------------------------------------------------------
old_price = 5       # original value
price = 6           # updated value
print("Old vs new:", old_price, price)   # 5 6


# -------------------------------------------------------------
# 3️⃣ Overwrite after testing (cleanup pattern)
# -------------------------------------------------------------
price = 5
new_price = 6

# Imagine we run calculations, tests, or print checks here.
print("Before overwrite:", price, new_price)  # 5 6

# Once we are satisfied with the new value, we overwrite:
price = new_price
print("After overwrite:", price)             # 6

# Optional: update again
new_price = 7
price = new_price
print("Final price:", price)                 # 7
`````

`````python
# -----------------------------------------------------------
# 🔹 Assigning One Variable to Another
# -----------------------------------------------------------

price = 5
new_price = 6

# Assign the value of new_price to price
price = new_price

print(price)          # prints: 6
print(new_price)      # prints: 6

# Changing new_price again does not affect price
new_price = 7
print(price)          # still 6
print(new_price)      # now 7
`````

`````python
# -----------------------------------------------------------
# 🔹 Deleting a Variable with del
# -----------------------------------------------------------

value = 25
print(value)          # prints: 25

del value             # removes 'value' from memory

# The following line will raise a NameError
# print(value)
`````

### 📝 Note: Understanding Function Calls Assigned to Variables

In this example, a call to the `print()` function is assigned to the variable `result`. The purpose of this demonstration is to show how Python handles **function return values** during variable assignment.

When Python executes:

```python
result = print(price_list)
```

two separate events occur:

1. **`print(price_list)` runs**, which displays the list in the notebook.  
2. **The return value of `print()` is stored in `result`**.  
   Since `print()` does not return meaningful data, its return value is always `None`.

As a result, printing `result` will display:

```
None
```

This example highlights an important concept:

- Some functions **display information** but **return nothing** (for example, `print()`).
- Other functions **return actual values** that can be stored, reused, or manipulated.  
  Examples include:
  - `len()` → returns the length of a sequence  
  - `sum()` → returns the total of numeric values  
  - `max()` → returns the highest value  
  - `sorted()` → returns a new sorted list  

⚠️ Note: Understanding which functions return values and which only perform actions is essential for writing clear and predictable Python code.


## 🔁 Overwriting and Updating Variables

In Python, a variable always stores the **most recent value** assigned to it.  
When you reassign a variable, the old value is replaced in memory for that name.


### 🧠 Overwriting Variables: Key Takeaways

- A variable name always refers to the **most recent value** assigned to it.  
- Reassigning a variable (for example, `price = 7`) overwrites the previous value.  
- Use an extra variable such as `old_price` or `new_price` when:
  - you want to compare before/after values, or  
  - you are testing a change and do not want to lose the original value.  
- After you are confident the new value is correct, you can overwrite the original
  variable and remove temporary variables to keep the code clean.  
- This pattern is common in data analysis when updating thresholds, prices, or
  configuration values during experimentation.

## 🔢 Numeric Type Conversion

Python allows converting between numeric types using the standard pattern:

```
datatype(object)
```

Examples include:

- `int(3.8)` → rounds down to `3`
- `float(5)` → converts integer to `5.0`
- `float("12.5")` → succeeds because the string contains a valid number
- `int("abc")` → raises `ValueError` (letters cannot be converted)

Numeric conversion is essential when importing data from CSV files, user input, APIs, and JSON, where numbers often arrive as strings.

| Operator | Meaning | Example | Result |

---

## 🧮 Order of Operations (PEMDAS)

Python follows the standard execution priority:

1. **P**arentheses  
2. **E**xponentiation  
3. **M**ultiplication & **D**ivision (including `//` and `%`)  
4. **A**ddition & **S**ubtraction  

Parentheses allow you to control evaluation, especially when expressions become complex.

---

These functions simplify numeric analysis without requiring manual loops.

`````python
# -------------------------------------------------------------
# Len() Exercise: Simple Cryptography
# -------------------------------------------------------------
# Goal:
#   Use the length of different messages as "indexes" into the
#   alphabet string to build a password.
#
# Process:
#   1. Measure the length of each message using len()
#   2. Use that number as an index into the alphabet
#   3. Concatenate the resulting letters into a password
# -------------------------------------------------------------

# A reference alphabet for indexing
alphabet = "abcdefghijklmnopqrstuvwxyz"

# Messages of varying lengths
message1 = "Hello World!"
message2 = ""
message3 = "It snowed a lot today"
message4 = "eeee"
message5 = "I love snow!!"

# -------------------------------------------------------------
# Step-by-step explanation:
#
print(len(message1))  #  → 12
print(alphabet[12])  # → 'm'

print(len(message2))  # → 0
print(alphabet[0])  # → 'a'

print(len(message3))  # → 21
print(alphabet[21])  # → 'v'

print(len(message4))  # → 4
print(alphabet[4])  # → 'e'

print(len(message5))  # → 13
print(alphabet[13])  # → 'n'
#
# Final password: "maven"
# -------------------------------------------------------------

password = (
    alphabet[len(message1)] +
    alphabet[len(message2)] +
    alphabet[len(message3)] +
    alphabet[len(message4)] +
    alphabet[len(message5)]
)

print("Generated password:", password)
`````

### Understanding Why `__doc__` Works for Functions but *not* Variables

Python allows you to attach documentation (called a **docstring**) to functions, classes, and modules by placing a triple-quoted string immediately after the definition. These docstrings are accessible using the `.__doc__` attribute.

However, a **variable does not store docstrings**. If you call `.__doc__` on a string variable, Python returns the documentation for the **`str` class itself**, not the variable’s content.

This section compares how `__doc__` behaves for:
- a triple-quoted *multi-line string variable*, and  
- a properly documented function.

### 📌 Key Takeaways

- Triple-quoted strings assigned to variables are **just strings**, not docstrings.  
- Only **functions, classes, and modules** store real docstrings accessible via `.__doc__`.  
- Calling `.__doc__` on a string gives you the documentation for Python’s built-in `str` class.  
- To see a variable’s contents, print the variable directly.  
- To see a function’s docstring, print `function.__doc__`.

## 🔍 Slicing Strings in Python

Python slicing uses the pattern:

```python
string[start : stop : step]
```

### ✔️ What each argument really means

- **start** — index to begin the slice  
  - default depends on step direction  
  - for positive step, default = `0`  
- **stop** — index to stop *before*  
  - default depends on step direction  
  - for positive step, default = `len(string)`  
- **step** — how much to move between characters  
  - default = `1`  
  - must be **non-zero**

---

### 🔹 Important correction
The defaults for `start` and `stop` are **NOT always 0 and len(string)**.  
Those defaults apply **only when the step is positive**.

We will use **positive-step slicing only** in this section,  
so the defaults here behave normally.

---


---

## 🔄 Reversing Strings and Using Negative Indices

Python allows slicing with **negative indices** and **negative step values**, which gives you powerful ways to read a string from right to left.

### ✔️ Important Rules

- Negative indices count from the right  
  ```
  message = "I hope it snows TONIGHT!"
  Index:    -23 … -1
  ```
- A **negative step (-1)** means slicing moves backwards.
- When step is negative:
  - default `start` becomes the **last index**
  - default `stop` becomes **before the first character**
- Slicing still stops **before** the `stop` index — even for negative steps.

## 🧮 Key Points — `.count()`

- `text.count("a")` → counts how many times the character `"a"` appears  
- `text.count("snow")` → counts how many times the substring `"snow"` appears  
- `text.count(x, start, end)` → counts only within the slice `[start:end]`  
  - `start` is **inclusive**  
  - `end` is **exclusive** (standard Python slicing rule)  
- `.count()` tells you **how many times**, *not where* the substring occurs  
- `.count()` does **not** count overlapping matches  
  - Example: `"aaa".count("aa")` → `1` (not `2`) because `"aa"` overlaps itself  

---

## 📌 When to Use `.count()`

Use `.count()` when answering questions like:

- “How many times does this character appear?”  
- “How many times is this substring repeated?”  
- “How many occurrences appear inside a specific part of the string?”  

Use `.count()` for **frequency**, not for locating positions.

For positions, prefer:

- `.find()`  
- `.rfind()`  
- `.index()`

## 2. If Statements & Core Control Flow

In Part 1 you built a strong foundation in **Boolean logic**:
comparisons, membership tests, `and` / `or` / `not`, precedence, parentheses, and short-circuiting.

Now we use those Boolean expressions to **control the flow of a program** with:

- `if`
- `elif`
- `else`

Control flow lets Python **choose** which block of code to run based on conditions.

---

### 2.1 What Is Control Flow?

Without conditions, a program runs **top to bottom** in a straight line.

With control flow, a program can:

- make decisions  
- follow different paths  
- skip or execute blocks of code based on Boolean results  

You can think of it as a series of **questions**:

> If this is true, do this.  
> Otherwise, do something else.

---

## 2.2 The `if` Statement

### Basic pattern

```python
# General syntax

if condition:
    # This indented block runs only if condition is True
    do_something()
```

Important points:

- `condition` must evaluate to a Boolean (`True` or `False`)
- A colon `:` ends the `if` line
- The following indented block is the **body** of the `if`
- Indentation in Python is not cosmetic. It defines the block.

---

### Example: Basic `if` with a Boolean expression

```python
# --- Example: check if a customer qualifies for free shipping ---

order_total = 120

# Boolean expression from Part 1: comparison operator
if order_total >= 100:
    print("Free shipping applied!")
    
print("Order processed.")
```

What happens:

- If `order_total >= 100` is True, Python prints `"Free shipping applied!"`
- `"Order processed."` prints every time, because it is **outside** the `if` block

Try changing `order_total` to 80 and re-running.

---

## 2.3 `if` + `else`: Two-Way Decision

An `if` statement can stand alone.  
`else` **cannot** stand alone. It always pairs with an `if`.

Use `else` when you want a **default path** if the condition is False.

```python
# General pattern

if condition:
    # runs when condition is True
    do_this()
else:
    # runs when condition is False
    do_that_instead()
```

---

### Example: Free shipping or standard shipping

```python
# --- Example: two-way decision with if/else ---

order_total = 75

if order_total >= 100:
    print("Free shipping applied.")
else:
    print("Standard shipping rate will be used.")

print("Order complete.")
```

- If `order_total` is 120 → prints `Free shipping applied.`
- If `order_total` is 75 → prints `Standard shipping rate will be used.`

Only **one** of the `if` or `else` blocks runs, never both.

---

### Example: Using a helper to show which branch ran

```python

def check_shipping(order_total):
    print(f"\nChecking shipping for order_total = {order_total}")
    
    if order_total >= 100:
        print("Branch: if → Free shipping")
    else:
        print("Branch: else → Standard shipping")

# Try a few values

check_shipping(75)
check_shipping(100)
```

This example makes branching very visible.  
You can change values and see which branch is selected.

## 2.4 `if` + `elif` + `else`: Multi-Way Decision

Sometimes there are **more than two** meaningful cases.  
This is where `elif` (short for "else if") is useful.

```python
# General pattern

if condition_1:
    # runs if condition_1 is True
    block_1
elif condition_2:
    # runs if condition_1 is False and condition_2 is True
    block_2
elif condition_3:
    # runs if all previous conditions are False and condition_3 is True
    block_3
else:
    # runs if none of the above conditions are True
    default_block
```

Rules:

- Conditions are checked **top to bottom**
- As soon as one condition is True, Python runs that block and **skips the rest**
- `if` can exist alone  
- `if` + `else` is a two-branch decision  
- `if` + one or more `elif`s + optional `else` creates a **decision chain**

---

### Example: Customer status and discount level

We will build a simple rule:

- If status is `"VIP"` → 20% discount  
- Else if status is `"LOYAL"` → 10% discount  
- Else if status is `"NEW"` → 5% discount  
- Else → no discount

```python
# --- Example: multi-way decision with if/elif/else ---

def discount_for_status(status):
    print(f"\nChecking discount for status = {status!r}")
    
    if status == "VIP":
        print("20% discount applied.")
    elif status == "LOYAL":
        print("10% discount applied.")
    elif status == "NEW":
        print("5% discount applied.")
    else:
        print("No discount for this status.")

# Try a few different statuses
discount_for_status("VIP")
discount_for_status("LOYAL")
discount_for_status("NEW")
discount_for_status("GUEST")
```

Observe:

- Only one branch runs per call  
- The order of conditions matters

---

### Ordering matters: more specific conditions should appear first

Consider this pattern:

```python
# Example of ordering

score = 95

if score >= 60:
    print("Pass")
elif score >= 90:
    print("High Distinction")
```

For `score = 95`, the result is `"Pass"` because:

- Python checks `score >= 60` first → True  
- It runs that block and **does not** check `elif score >= 90`

Better ordering:

```python
score = 95

if score >= 90:
    print("High Distinction")
elif score >= 60:
    print("Pass")
else:
    print("Fail")
```

Principle:

- Put **stricter** or **higher priority** conditions earlier
- Put more general conditions later in the chain

# 🐍 Chapter 5. Conditional Logic: 
## 🧩 Nested If Statements (Part 3)
 
### Expanding control flow with deeper, multi-layered logic  

Nested `if` statements allow you to evaluate **a second condition only if a first condition is already true**.  
They add an extra “layer” of logic and are useful when your business rules depend on two steps of decision-making.

---

# 1. Why Use Nested If Statements?

Nested logic is appropriate when:

- You need to **check Condition B only when Condition A is true**
- Rules follow a **hierarchy** or **decision tree**
- The second decision would be *nonsensical* unless the first is satisfied  
  (e.g., “If item is out of stock, then check whether you want to offer a display model.”)

### 🔍 Visual Model
A nested structure behaves like this:

```
if condition_A:
    if condition_B:
        # Nested True block
    else:
        # Nested False block
else:
    # Outer False block
```

This is like saying:
> “Only if A is true should I bother checking B.”

---

# 2. Simple Example

```python
customer = "Chris"
inventory = 0

if inventory <= 0:                     # Outer If condition
    if customer == "Chris":            # Nested If condition
        print("You can have the display model, sir.")
    else:
        print("Out of stock.")         # Nested Else code
else:
    print("Item available!")
```

### Why this is nested:
- It makes sense to check the **customer’s name** ONLY after confirming that inventory is zero.
- The business rule is **hierarchical**, not parallel.

---

# 3. The Same Logic *Without* Nesting

From page 3 of the PDF, Maven shows a “without nesting” version.  
Here’s how it looks conceptually:

```
if customer == "Chris" and inventory <= 0:
    ...
```

This works… **but loses hierarchy clarity**.

Nested version makes it obvious:
> “We only ask about Chris because inventory is zero.”

Flat version makes it look like the two conditions are equally important.

Which should you use?  
➡ Use **nesting** when your logic flows like:  
**First A → then maybe check B**

➡ Use **flat logic** when your rule is truly:  
**A AND B must both be true with equal weight**

---

# 4. Real-World Example: Return Authentication

Nested ifs shine in real workflows:

```python
is_logged_in = True
is_admin = False

if is_logged_in:

    else:
        print("Welcome, user.")
else:
    print("Please log in.")
```

This mirrors a real application:
- First confirm login  
- THEN decide admin/user behavior

## 1. The Boolean Data Type (`True` / `False`)

A **Boolean** is a data type with only **two possible values**:

- `True`
- `False`

In arithmetic:

- `True` behaves like `1`
- `False` behaves like `0`
- The `not` keyword **inverts** a Boolean

## 2. Comparison Operators

**Comparison operators** evaluate a relationship and return a **Boolean** (`True` / `False`):

| Operator | Meaning                    | Example        |
|---------:|----------------------------|----------------|
| `==`     | equal to                   | `price == 99`  |
| `!=`     | not equal to               | `item != 'ski'`|
| `<`      | less than                  | `inv < 10`     |
| `>`      | greater than               | `temp > 32`    |
| `<=`     | less than or equal to      | `age <= 18`    |
| `>=`     | greater than or equal to   | `score >= 80`  |

## 🔗 3. Chained Comparisons (Readable Multi-Condition Checks)

Python allows you to “chain” comparisons together, which makes range checks concise and readable.  
Chained comparisons evaluate **left to right**, and Python ensures all comparisons are True.

This is extremely common in pricing logic, validation rules, SLAs, and data quality checks.

## 5. Boolean Operators: `and`, `or`, `not` (+ Short-Circuiting & Precedence)

Boolean operators allow you to combine multiple logical expressions into a single condition:

- **`A and B`** → True only if **both** A and B are True  
- **`A or B`** → True if **at least one** of A or B is True  
- **`not A`** → inverts A (True ↔ False)

---

### 🔄 Short-Circuit Behavior (Efficiency Tip)

Python **stops evaluating early** when the final outcome is already determined:

- For **`A and B`**:  
  - If `A` is **False**, Python **does not evaluate B**
- For **`A or B`**:  
  - If `A` is **True**, Python **does not evaluate B**

This is why we typically place **simple or fast checks first**, and **more expensive operations last**.

---

### 🧠 Operator Precedence (Highest → Lowest)

Python always evaluates Boolean operators in this fixed order:

1. **`not`**  
2. **`and`**  
3. **`or`**

**This means:**
- `not` is always evaluated **before** `and`  
- `and` is always evaluated **before** `or`  
- This happens **even if the operators appear in a different order in the code**

Example mental model:

```
A or B and C
→ A or (B and C)
```

Python will always group `B and C` first because `and` outranks `or`.

---

### 🟦 Parentheses Always Win

Parentheses let you **override** Python’s default precedence rules and intentionally control how the logic is grouped.

Use parentheses to:
- Make your logic explicit  
- Avoid ambiguity in complex conditions  
- Ensure correctness when mixing `and` and `or`  
- Improve readability for yourself and others

Whenever your logic combines different operators, parentheses are a best practice for clarity and safety.

## ⚡ Short-Circuit Evaluation (Efficiency in Action)

Python **stops evaluating early** when a final result is already determined.

- For **`A and B`**:  
  If `A` is False → *Python never evaluates B*
- For **`A or B`**:  
  If `A` is True → *Python never evaluates B*

Short-circuiting is important for performance and safety.

Here is a clear demonstration using “fast” and “slow” checks.

```python
# --- Short-Circuit Demo using clear printed side effects ---

def fast_check():
    print("fast_check() ran → quick evaluation")
    return False  # pretend fast and simple

def slow_check():
    print("slow_check() RAN → expensive evaluation (should NOT run if skipped)")
    return True   # pretend slow and complex


print("\nCase 1: AND short-circuit (first condition is False)")
result_1 = fast_check() and slow_check()
print("Final Result:", result_1)


def fast_true():
    print("fast_true() ran → quick positive result")
    return True

print("\nCase 2: OR short-circuit (first condition is True)")
result_2 = fast_true() or slow_check()
print("Final Result:", result_2)
```

### ⛳ Interpretation
- In the `and` example, `fast_check()` is False → `slow_check()` never runs  
- In the `or` example, `fast_true()` is True → `slow_check()` never runs  

This mirrors real code patterns where we put:
- Cheap validation checks **first**  
- Expensive computations **second**  


## 🧪 `trace_logic()` – Visualize How Your Conditions Run

`trace_logic()` is a small helper that makes it easier to **see** how a complex Boolean expression is evaluated.

You’ll use it together with functions that print when they run (like `item_in_stock()` and `is_discount_item()`).  
This lets you watch:

- which checks run
- which checks are skipped (short-circuited)
- what the final result is

---

```python
# ============================================
# 🔍 trace_logic – Boolean Logic Tracing Helper
# ============================================

def trace_logic(label, func):
    """
    Run a logical expression (wrapped in a zero-argument function),
    print a clear header, and show the final Boolean result.

    Usage:
        trace_logic(
            "Description of the logic",
            lambda: some_check() and other_check() or third_check()
        )

    To see short-circuiting, each inner check should print when it runs.
    """
    print("\n" + "=" * 60)
    print("TRACE:", label)
    print("=" * 60)

    result = func()  # Evaluate the expression once
    print("Final result:", result)
    return result
```

### What you’ll see

def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,

def trace_if(description, condition, if_true, if_false):

# ✅ Key Takeaways — Chapter 5: Conditional Logic

### **1. All conditional logic is built on Boolean expressions**
Everything begins with expressions that evaluate to **True** or **False**:
- Comparisons (`>`, `<`, `>=`, `<=`, `==`, `!=`)
- Membership tests (`in`, `not in`)
- Logical operators (`and`, `or`, `not`)
- Chained comparisons (`0 < x <= 5`)

Understanding how to build Boolean expressions is the foundation of all control flow.

---

### **2. Python follows strict Boolean precedence**
Python evaluates conditions in this order:

1. `not`  
2. `and`  
3. `or`  

Use parentheses when your intent does not match Python’s default order.

---

### **3. Short-circuiting affects program behavior**
- `A and B`: if A is False → Python never evaluates B  
- `A or B`: if A is True → Python never evaluates B  

Short-circuiting controls:
- performance
- side-effects (function calls inside conditions)
- rule behavior in real systems

---

### **4. `if`, `elif`, `else` form a single decision chain**
Important principles:
- Conditions are checked **top to bottom**
- Only **one branch** runs
- When a branch is True, Python **skips** all remaining branches
- Most specific conditions should appear first

Always order conditions from **most restrictive** → **least restrictive**.

---

### **5. Nesting creates multi-layered, real-world decision trees**
Use nested logic when rules follow a natural sequence:

> "If this is true, then check something else."

Nesting is ideal for:
- multi-step approvals  
- customer + inventory combinations  
- authentication + permissions  
- multi-rule eligibility checks  

---

### **6. Combine Boolean logic + control flow for expressive rule systems**
Everything from Part 1 becomes building blocks in Part 2:

- `(status == "VIP") and (order_total >= 100)`
- `"hat" in discount_items`
- `not is_blocked`
- `0 < inventory <= 5`

This enables:
- pricing systems  
- discount logic  
- safety checks  
- menu/option logic  
- workflows and gating rules  

---

### **7. Proper ordering prevents unreachable branches**
Incorrect ordering example:

```
if score >= 60:
    ...
elif score >= 90:   # never runs
```

Correct ordering:

```
if score >= 90:
    ...
elif score >= 60:
    ...
```

Always ask:
- Is this condition more general than the ones below?  
- Should this rule be evaluated earlier?
