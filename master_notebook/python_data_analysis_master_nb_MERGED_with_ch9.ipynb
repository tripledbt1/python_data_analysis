# 🧠 Part 1 — Python Foundations  
## **Chapter 1: Python Data Types**
## ------------------------------------------------------------  

This section introduces the essential data types used in Python and explains how Python classifies, stores, and manages values. Understanding data types is foundational for effective data analysis because it determines how Python interprets variables, executes operations, and handles errors. The lesson also introduces the `type()` function, type conversion, the concept of iterability, and the distinction between mutable and immutable data types.

Python data types fall into key categories:
- **NoneType**  
- **Numeric** (int, float, complex)  
- **Text** (str)  
- **Sequence** (list, tuple, range)  
- **Mapping** (dict)  
- **Set Types** (set, frozenset)  
- **Boolean** (bool)

The goal of this section is to understand what each data type represents, how to identify them, and how to convert between compatible types using built-in functions.

---

## 📘 Python Data Types: Single Value vs Multiple Value

The table below summarizes Python data types based on whether they store a **single value** or **multiple values**. This aligns with the conceptual framework introduced in the lecture notes and your handwritten outline.

| **Single Value** | **Multiple Value** |
|------------------|--------------------|
| **Numeric** <br> • `int` <br> • `float` <br> • `complex` | **Sequence Types** <br> • `list` <br> • `tuple` <br> • `range` |
| **Text** <br> • `str` | **Mapping Type** <br> • `dict` |
| **Boolean** <br> • `bool` (True, False) | **Set Types** <br> • `set` <br> • `frozenset` |
| **NoneType** <br> • `None` (placeholder for no value) | *(No additional categories)* |

---

### Notes:
- **Single-value data types** represent one atomic value, such as a number, text string, or boolean.  
- **Multiple-value data types** store collections of values. These structures allow iteration and support a wide range of data manipulation patterns used in analysis workflows.  
- `NoneType` exists as a placeholder that represents “no value.”
---

# -------------------------------------------------------------
# 🔹 Checking the data type of various Python objects
# -------------------------------------------------------------

empty_value = None
print(type(empty_value))

x_int = 5
x_float = 5.24
x_complex = 5 + 2j
print(type(x_int), type(x_float), type(x_complex))

name = "snowboard"
print(type(name))

numbers_list = [1, 3, 5, 7, 9]
items_tuple = ("snowboard", "skis")
num_range = range(1, 10, 2)
print(type(numbers_list), type(items_tuple), type(num_range))

equipment = {"snowboard": 24, "skis": 17}
print(type(equipment))

flag = True
print(type(flag))

gear_set = {"snowboard", "skis"}
frozen_gear = frozenset({"snowboard", "skis"})
print(type(gear_set), type(frozen_gear))


`````python
# -------------------------------------------------------------
# 🔹 Type Conversion (valid and invalid conversions)
# -------------------------------------------------------------

num_string = "123"
converted_int = int(num_string)
print(converted_int, type(converted_int))

integer_value = 456
int_string = str(integer_value)
print(int_string, type(int_string))

sample_list = [1, 2, 3]
list_tuple = tuple(sample_list)
print(list_tuple, type(list_tuple))

# This will raise a TypeError:
# int(sample_list)
`````

`````python
# -------------------------------------------------------------
# 🔹 Iterable vs Non-Iterable Examples
# -------------------------------------------------------------

# Strings are iterable
for character in "snowboard":
    print(character)

# Lists are iterable
for item in [1, 3, 5]:
    print(item)

# Non-iterable: int
# for i in 5:
#     print(i)    # TypeError if uncommented
`````

`````python
# -------------------------------------------------------------
# 🔹 Mutability Examples
# -------------------------------------------------------------

mutable_list = [1, 2, 3]
mutable_list.append(4)
print(mutable_list)

mutable_set = {"snowboard", "skis"}
mutable_set.add("helmet")
print(mutable_set)

mutable_dict = {"snowboard": 24}
mutable_dict["skis"] = 17
print(mutable_dict)
`````

`````python
# -------------------------------------------------------------
# 🔹 Sets automatically remove duplicates
# -------------------------------------------------------------

numbers = [1, 2, 2, 3, 3, 3]
unique_numbers = set(numbers)
print(unique_numbers)

# Duplicate values inside {} will also be removed
example = {"a", "a", "b"}
print(example)
`````

## 🧩 Syntax Table for Set and Tuple Creation

| Syntax                  | What it Creates        | Why                                                |
| ----------------------- | ---------------------- | -------------------------------------------------- |
| `{"a", "b"}`            | **set**                | curly braces + comma-separated values              |
| `("a", "b")`            | **tuple**              | parentheses + commas                               |
| `"a", "b"`              | **tuple**              | commas alone create a tuple                        |
| `frozenset({"a", "b"})` | **frozen set**         | must call `frozenset()` explicitly                 |
| `{}`                    | **dictionary**         | Python reserves `{}` for empty dict                |
| `set()`                 | **empty set**          | correct way to create an empty set                 |


## 🧩 Conversions to a Set Remove Duplicates

| Conversion      | Removes duplicates? | Explanation                                  |
| --------------- | ------------------- | -------------------------------------------- |
| `set(list)`     | ✔ Yes               | All repeated elements removed                |
| `set(tuple)`    | ✔ Yes               | Adds each unique tuple value once            |
| `set(range)`    | ✔ Yes               | Range contains no duplicates                 |
| `set("string")` | ✔ Yes               | Unique characters only                       |
| `set(dict)`     | ✔ Yes               | Converts dictionary *keys* into a set        |

**Note:** Sets may only contain unique elements.

## 🧩 Mutability Comparison (Mutable vs Immutable)

| **Category**     | **Type**        | **Mutable?** | **Example**                    |
|------------------|------------------|--------------|--------------------------------|
| Numeric          | `int`, `float`, `complex` | No | `5`, `5.24`, `5+2j`            |
| Text             | `str`            | No           | `"snowboard"`                  |

| NoneType         | `NoneType`       | No           | `None`                         |

|                  | `tuple`          | No           | `(1, 2, 3)`                    |
|                  | `range`          | No           | `range(1, 10)`                 |
| Mapping          | `dict`           | Yes          | `{"a": 1}`                     |
| Set Types        | `set`            | Yes          | `{"a", "b"}`                   |
|                  | `frozenset`      | No           | `frozenset({"a", "b"})`        |


## 🔁 Iterable vs Non-Iterable Data Types

| **Type**        | **Iterable?** | **Example**                    | **Notes**                                 |
|-----------------|----------------|--------------------------------|-------------------------------------------|
| `list`          | Yes           | `[1, 2, 3]`                    | Ordered and mutable                       |
| `tuple`         | Yes           | `(1, 2, 3)`                    | Ordered and immutable                     |
| `range`         | Yes           | `range(1, 5)`                  | Generates numbers efficiently             |
| `dict`          | Yes           | `{"a": 1}`                     | Iterates over keys by default             |
| `set`           | Yes           | `{"a", "b"}`                   | Unordered, unique values                  |
| `frozenset`     | Yes           | `frozenset({"a", "b"})`        | Immutable set                             |
| `str`           | Yes           | `"abc"`                        | Character-by-character                    |
| `int`           | No            | `5`                            | Not iterable                              |
| `float`         | No            | `3.14`                         | Not iterable                              |
| `bool`          | No            | `True`                         | Not iterable                              |
| `NoneType`      | No            | `None`                         | Not iterable                              |


## 🧠 **Chapter 1 Summary: Data Types in Python**
## **Key Takeaways:**

- Python includes several core data type categories: **numeric, text, sequence, mapping, set, boolean, and NoneType**.  
- The `type()` function identifies an object's data type and is essential for debugging when operations fail due to incompatible types.  
- Type conversion uses the format `datatype(object)` and succeeds only when the value is compatible (e.g., `"123"` → `int`, but `[1,2,3]` → `int` fails).  
- **Iterables** include strings, lists, tuples, dictionaries, ranges, sets, and frozensets—any object you can loop through.  
- **Strings are immutable yet iterable**, meaning characters cannot be changed in place but can be looped over.  
- **Mutable types** (lists, sets, dictionaries) can be changed after creation.  
  **Immutable types** (numbers, strings, tuples, booleans, frozensets, NoneType) cannot.  
- Dictionaries store **key-value pairs**, forming the backbone of JSON-style structured data widely used in analytics.  
- Sets store **unique values only**, removing duplicates automatically during creation or conversion.  
- Curly braces `{}` create:  
  - a **set** when values are comma-separated  
  - a **dictionary** when using key-value pairs  
  - an empty dictionary when used as `{}`  
- An empty set must be created with **`set()`**, since `{}` defaults to a dictionary.  
- Understanding mutability, iterability, and data type behavior is foundational for working with **Pandas, NumPy, data cleaning, merging, grouping, and transformation workflows**.


# 🧭 Chapter 2. Variables: Overview

Variables are containers that store values for later use in a Python program. They can store any data type, including numbers, strings, lists, dictionaries, and even function calls. Understanding how to assign, overwrite, delete, and track variables is essential for writing clear and maintainable code.

A variable assignment uses the format:

```
variable_name = value
```

Variable assignment allows you to label values, reuse them, and perform operations without hard coding every value. Naming conventions and clear organization help make code easier to follow. This section builds on the concepts from the Maven Analytics notes by demonstrating assignment, overwriting, deleting variables, and tracking variables in a Jupyter environment.

---

# 🧪 Naming Rules and Conventions

Variable names:

- Can contain letters, numbers, and underscores  
- Must begin with a letter or underscore  
- Cannot begin with a number  
- Cannot contain spaces or special characters  
- Cannot use reserved keywords such as `del`, `list`, or `for`

Snake case is the preferred Python naming style:

- `price_list`
- `total_revenue`
- `snowboard_inventory_2024`

---


`````python
# -----------------------------------------------------------
# 🔹 Basic Variable Assignment
# -----------------------------------------------------------

# Assigning simple values to variables
price = 5
city = "Los Angeles"
heights = [180, 173, 191]

print(price)          # prints: 5
print(city)           # prints: "Los Angeles"
print(heights)        # prints: [180, 173, 191]

# Using a variable in an operation
print(price + 1)      # prints: 6 (adds 1 to the value stored in 'price')
# adding 1 is hardcoded (e.g., set fee)
# or can it can be a variable for values that can change
`````

`````python
# -----------------------------------------------------------
# 🔹 Assigning a Function Call to a Variable
# -----------------------------------------------------------

price_list = [10.99, 12.49, 8.75]   
result = print(price_list)          
# 'result' stores the return value of print(), which is None

print(result)         # prints: None
# ⚠️ Note: Understanding which functions return values is essential (see markdown)
`````

`````python
# -----------------------------------------------------------
# 🔹 Overwriting a Variable
# -----------------------------------------------------------

# -------------------------------------------------------------
# 1️⃣ Simple overwrite: Python stores the most recent value
# -------------------------------------------------------------
price = 5
price = 6
price = 7           # last assignment "wins"
print("Simple overwrite:", price)   # 7


# -------------------------------------------------------------
# 2️⃣ Keeping both old and new values (good for testing)
# -------------------------------------------------------------
old_price = 5       # original value
price = 6           # updated value
print("Old vs new:", old_price, price)   # 5 6


# -------------------------------------------------------------
# 3️⃣ Overwrite after testing (cleanup pattern)
# -------------------------------------------------------------
price = 5
new_price = 6

# Imagine we run calculations, tests, or print checks here.
print("Before overwrite:", price, new_price)  # 5 6

# Once we are satisfied with the new value, we overwrite:
price = new_price
print("After overwrite:", price)             # 6

# Optional: update again
new_price = 7
price = new_price
print("Final price:", price)                 # 7
`````

`````python
# -----------------------------------------------------------
# 🔹 Assigning One Variable to Another
# -----------------------------------------------------------

price = 5
new_price = 6

# Assign the value of new_price to price
price = new_price

print(price)          # prints: 6
print(new_price)      # prints: 6

# Changing new_price again does not affect price
new_price = 7
print(price)          # still 6
print(new_price)      # now 7
`````

`````python
# -----------------------------------------------------------
# 🔹 Deleting a Variable with del
# -----------------------------------------------------------

value = 25
print(value)          # prints: 25

del value             # removes 'value' from memory

# The following line will raise a NameError
# print(value)
`````

`````python
# Cleanest version: show ONLY variables created by the user (not Jupyter)
user_vars = [
    var for var, val in globals().items()
    if not var.startswith("_")                    # exclude private + system
    and var not in ('In', 'Out', 'get_ipython')   # exclude Jupyter internals
    and not callable(val)                         # exclude functions
    and not isinstance(val, type(__builtins__))   # exclude modules
]
user_vars
`````

### 📝 Note: Understanding Function Calls Assigned to Variables

In this example, a call to the `print()` function is assigned to the variable `result`. The purpose of this demonstration is to show how Python handles **function return values** during variable assignment.

When Python executes:

```python
result = print(price_list)
```

two separate events occur:

1. **`print(price_list)` runs**, which displays the list in the notebook.  
2. **The return value of `print()` is stored in `result`**.  
   Since `print()` does not return meaningful data, its return value is always `None`.

As a result, printing `result` will display:

```
None
```

This example highlights an important concept:

- Some functions **display information** but **return nothing** (for example, `print()`).
- Other functions **return actual values** that can be stored, reused, or manipulated.  
  Examples include:
  - `len()` → returns the length of a sequence  
  - `sum()` → returns the total of numeric values  
  - `max()` → returns the highest value  
  - `sorted()` → returns a new sorted list  

⚠️ Note: Understanding which functions return values and which only perform actions is essential for writing clear and predictable Python code.


## 🔁 Overwriting and Updating Variables

In Python, a variable always stores the **most recent value** assigned to it.  
When you reassign a variable, the old value is replaced in memory for that name.


### 🧠 Overwriting Variables: Key Takeaways

- A variable name always refers to the **most recent value** assigned to it.  
- Reassigning a variable (for example, `price = 7`) overwrites the previous value.  
- Use an extra variable such as `old_price` or `new_price` when:
  - you want to compare before/after values, or  
  - you are testing a change and do not want to lose the original value.  
- After you are confident the new value is correct, you can overwrite the original
  variable and remove temporary variables to keep the code clean.  
- This pattern is common in data analysis when updating thresholds, prices, or
  configuration values during experimentation.


## 🧾 Tracking Variables in Jupyter with `%who` and `%whos`

IPython (and Jupyter) provide special **magic commands** that help you see which
variables are currently defined in your notebook:

- `%who`  → lists variable **names**  
- `%whos` → shows names **plus** types and extra info  

These commands only work in IPython environments such as **Jupyter, VS Code
Notebooks, and Colab**.

These commands help you check which objects exist and track their data types.

## ✅ When to Use These Tools

Use `%%who` or `%%whos` whenever you need to:

- **Check which variables currently exist** in your notebook  
- **Confirm data types** before passing values into functions  
- **Clean up your namespace** during long exploratory sessions  
- **Debug data transformations**, especially when values change unexpectedly  
- **Prepare clean, reproducible code** before committing to GitHub  
- **Avoid accidental overwriting** by ensuring variable names are unique and intentional  

These commands help you stay organized, prevent logical errors, and maintain a tidy working environment — all essential skills for professional data analysis.


## 🧠 **Chapter 2 Summary: Variables**
## **Key Takeaways**

- Variables store values in memory and allow you to reference them throughout your program.
- Assignment uses the structure `variable_name = value`, with spaces on both sides of the equal sign.
- Overwriting a variable replaces its previous value; that previous value is removed from memory unless stored elsewhere.
- Use intuitive, descriptive variable names following snake case to improve readability.
- Avoid overwriting important values during testing by using names such as `old_price` and `new_price`.
- The `del` keyword removes variables from memory, but in most cases reassigning the variable is sufficient.
- Use `%who` and `%whos` in Jupyter notebooks to track your active variables.

## **Chapter 3: Numeric Data**
## ------------------------------------------------------------  

This chapter introduces Python’s numeric data types and how to work with them in data analysis. You will learn how Python represents numbers, how to convert between types, how arithmetic operators behave, and which built-in numeric functions analysts use most frequently.

Python includes **three numeric types**:

| Numeric Type | Description | Examples |
|--------------|-------------|----------|
| **int** | Whole numbers without decimals | `3`, `42`, `-14` |
| **float** | Real numbers with decimals | `3.14`, `0.088`, `-1.20134` |
| **complex** | Numbers with real + imaginary components | `3 + 2j`, `-2j` |

For data analysis, **int** and **float** are the primary types. Complex numbers appear rarely.

---

## 🔢 Numeric Type Conversion

Python allows converting between numeric types using the standard pattern:

```
datatype(object)
```

Examples include:

- `int(3.8)` → rounds down to `3`
- `float(5)` → converts integer to `5.0`
- `float("12.5")` → succeeds because the string contains a valid number
- `int("abc")` → raises `ValueError` (letters cannot be converted)

Numeric conversion is essential when importing data from CSV files, user input, APIs, and JSON, where numbers often arrive as strings.

---

## ➗ Arithmetic Operators

Python supports all common arithmetic operators:

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division | `5 / 3` | `1.6666666667` |
| `//` | Floor division | `5 // 3` | `1` |
| `%` | Modulo (remainder) | `5 % 3` | `2` |
| `**` | Exponentiation | `5 ** 3` | `125` |

---

## 🧮 Order of Operations (PEMDAS)

Python follows the standard execution priority:

1. **P**arentheses  
2. **E**xponentiation  
3. **M**ultiplication & **D**ivision (including `//` and `%`)  
4. **A**ddition & **S**ubtraction  

Parentheses allow you to control evaluation, especially when expressions become complex.

---

## 🔧 Numeric Functions

Python provides numeric-specific helper functions:

| Function | Purpose | Example | Result |
|----------|----------|---------|--------|
| `round(n, digits)` | Round to a given number of decimal places | `round(3.14159, 2)` | `3.14` |
| `abs(n)` | Absolute value | `abs(-8)` | `8` |
| `sum(iterable)` | Add all numeric elements | `sum([2, 4, 6])` | `12` |
| `min(iterable)` | Smallest value | `min([2, 4, 6])` | `2` |
| `max(iterable)` | Largest value | `max([2, 4, 6])` | `6` |

These functions simplify numeric analysis without requiring manual loops.

---

## 📘 What You Will Learn in the Code Section

The coding portion of this chapter will show:

- How to declare and inspect numeric values  
- How type conversion works (valid and invalid cases)  
- How arithmetic operators behave  
- How Python evaluates expressions with PEMDAS  
- How to apply numeric helper functions to real data  

Code examples will follow the Python Foundations format with detailed inline comments.


`````python
## 🧪 Python Code Examples: Numeric Data

# -------------------------------------------------------------
# 🔹 Declaring numeric variables
# -------------------------------------------------------------

a_int = 10               # integer
b_float = 3.75           # floating-point number
c_complex = 2 + 5j       # complex number (real + imaginary)

print(type(a_int))       # <class 'int'>
print(type(b_float))     # <class 'float'>
print(type(c_complex))   # <class 'complex'>
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Type Conversion: float → integer
# -------------------------------------------------------------

x = int(7.89)             
print(x, type(x))         # truncates decimal, rounds down to 7, <class 'int'>
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Type Conversion: integer → float
# -------------------------------------------------------------

# Convert integer → float
y = float(12)             
print(y, type(y))         # 12.0, <class 'float'>
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Type Conversion: string → float (valid)
# -------------------------------------------------------------

# Convert string → float (valid)
price_str = "19.99"     # <class, 'str'>

price_float = float(price_str)      # string → float, as long as they contain only numeric charaters
print(price_float, type(price_float))
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Type Conversion: string → int (invalid)
# -------------------------------------------------------------

# Invalid conversion (string with letters)
number = "abc123"      # <class, 'str'>
# fixed_number = int(number)       # ❌ ValueError: could not convert string to float
# string → integer, only when they contain only numeric charaters
`````

`````python
# -------------------------------------------------------------
# 🔹 Arithmetic Operators
# -------------------------------------------------------------

x = 10
y = 3

print("Addition:", x + y)           # 13
print("Subtraction:", x - y)        # 7
print("Multiplication:", x * y)     # 30
print("Division:", x / y)           # 3.3333...
print("Floor division:", x // y)    # 3, rounds down
print("Modulo:", x % y)             # 1, returns remainder of a division (not equal to 0 is odd)
print("Exponent:", x ** y)          # 1000
`````

`````python
# -------------------------------------------------------------
# 🔹 Order of Operations (PEMDAS)
# -------------------------------------------------------------

result1 = 10 + 3 * 2         # multiplication first → 10 + 6 = 16
result2 = (10 + 3) * 2       # parentheses override → 13 * 2 = 26
result3 = 10 ** 2 * 3        # exponent first → 100 * 3 = 300

print(result1, result2, result3)
`````

`````python
# -------------------------------------------------------------
# 🔹 Numeric Data Types: Arithmetic demo
# -------------------------------------------------------------

my_int = 20
my_float = 8.33

print(my_int / my_float)          # action between int and float returns float by default
print(int(my_int / my_float))     # wrap in integer function to return int

result = my_int + my_float
print(result)                     # 28.33
print(round(result, 2))           # 28.33

golf_scores = [-3, 0, 2, 30]
print(min(golf_scores))
print(max(golf_scores))
print(sum(golf_scores))
sum(golf_scores, 3)               # sum of golf_scores plus 3
`````

`````python
# 🧮 Financial Calculations Practice (Arithmetic Operations)

# -----------------------------------------------------------
# 1️⃣ Calculate the gross profit on a snowboard
#    Formula: profit = price - cost
# -----------------------------------------------------------
snowboard_price = 499.99
snowboard_cost = 199.99

gross_profit = snowboard_price - snowboard_cost      # subtraction
print("Gross profit:", gross_profit)


# -----------------------------------------------------------
# 2️⃣ Calculate the gross margin on the snowboard
#    Formula: margin = gross_profit / price
# -----------------------------------------------------------
gross_margin = gross_profit / snowboard_price         # division
print("Gross margin (rounded):", round(gross_margin, 2))


# -----------------------------------------------------------
# 3️⃣ Calculate the price needed for a 70% gross margin
#    Formula: price_needed = cost / (1 - desired_margin)
# -----------------------------------------------------------
desired_margin = 0.70

price_needed_for70 = snowboard_cost / (1.0 - desired_margin)
print("Price needed for 70% margin:", round(price_needed_for70, 2))


# -----------------------------------------------------------
# 4️⃣ Calculate sales tax for the snowboard
#    Tax rate: 8%
#    Formula: tax = price * tax_rate
# -----------------------------------------------------------
tax_rate = 0.08

sales_tax = snowboard_price * tax_rate                # multiplication
print("Sales tax (rounded):", round(sales_tax, 2))


# -----------------------------------------------------------
# 5️⃣ Calculate returns on profits invested for one year
#    Assume sale of 5 snowboards and 5% interest rate
#    Formulas:
#      amount_invested = gross_profit * quantity
#      return_after_1yr = amount_invested + (amount_invested * interest_rate)
# -----------------------------------------------------------
quantity_sold = 5
interest_rate = 0.05

amount_invested = gross_profit * quantity_sold        # multiplication
return_after_1yr = amount_invested + (amount_invested * interest_rate)

print("Amount invested:", amount_invested)
print("Value after 1 year:", return_after_1yr)
`````

## **Key Takeaways:**

- Python supports three numeric types: `int`, `float`, and `complex`.  ro
- Type conversion uses the pattern `datatype(object)` and succeeds only when the value is compatible.  
- Converting from `float` to `int` truncates the decimal portion without rounding.

## Other: Escape Sequences
An **escape sequence** is a sequence of characters that Python applies special meaning to when they are encountered in a string. Several common escape sequences are listed below.

| Escape Sequence  | Result |
|:---:|-----|
| **`\'`**  | Prints a single quote. |
| **`\"`**  | Prints a double quote. |
| **`\n`**  | Inserts a newline. |
| **`\t`**  | Inserts a tab. |
| **`\\`**  | Inserts a backslash. |

For an example use case for escape sequences, assume that we want to define a variable containing the following string of characters: 

    He yelled, "I've had enough!" before storming out of the room.
    
As before, the presence of double quotes within the string prohibit us from being able to use double quotes to define the string. Furthermore, since the character for the apostrophe is the same as the character for a single quote, we are now no longer able to use single quotes to define our string. One solution is to use an escape sequence for the apostrophe so that Python knows to interpret it as text.

`````python
# ---------------------------------------------------------------
# Universal String Index Map Helper (Positive & Negative Indices)
# ---------------------------------------------------------------

# Original text:
customer_testimonial = '''I love skiing. It's my favorite hobby.
Some people say, "It's not a hobby, it's life."'''
print(repr(customer_testimonial))  # repr() returns an unambiguous string, often a valid expression

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")  # positive and negative column indices

show_index_map(customer_testimonial)
`````

`````python
# -------------------------------------------------------------
# Creating strings in Python
# -------------------------------------------------------------

# Single quotes
product_single = 'Super Snowboard'
print("Single quotes string:", product_single)

# Double quotes (functionally the same)
product_double = "Super Snowboard"
print("Double quotes string:", product_double)

# Triple quotes: useful for multi-line strings (docstrings, messages)
product_description = """\nThe Super Snowboard is our best-selling model.
It is designed for all-mountain riders and deep powder days."""

# Best practice: use end="\n\n" for adding two lines without extra spaces
print(product_description, end="\n\n")

# docstrings only work inside a function, class, or module — NOT regular variables
# print(product_description.__doc__)  # variable does not store docstrings

# Docstring example
def add(a, b):
    """Add two numbers and return the result.

    Args:
        a (int): The first number.
        b (int): The second number.

    Returns:
        int: The sum of a and b.
    """
    return a + b

print(add.__doc__)   # Safely prints the docstring
`````

`````python
# -------------------------------------------------------------
# String arithmetic: + (concatenation) and * (repetition)
# -------------------------------------------------------------

# Concatenate strings with +
first_name = "Sally"
last_name = "Snow"
full_name = (first_name + " " + last_name)
print(full_name)           # 'Sally Snow'

# Concatenate strings
subject = "My"
noun = "Maven snowboard"
verb = "cost"
price = 22.55

sentence = subject + " " + noun + " " + verb + " $" + str(price) + "! "
print(sentence, end="\n\n")

print("Repeat 3 times:")
print(sentence * 3, end="\n\n")     # * 3 prints three copies in one long line

# range(3) controls the number of repetitions, not the string, using numbers: repeat 3 times (0, 1, 2)
# Loop 3 times - use "_"  or "i" as a throwaway variable since we don't need it
for _ in range(3):
    print("Repeat 3 times:", sentence, end="\n\n")

# Build a marketing headline
product = "Super Snowboard"
headline = "Today only: " + product + " on sale!"
print(headline)

# Repeating strings with *
echo = "Powder day! " * 3
print(echo)                 # repeats the phrase three times

# Concatenating multiple pieces including punctuation
base_message = "Buy now"
urgent_message = base_message + "!!!"
print(urgent_message)
`````

`````python
# --------------------------------------------------------
# 🧵 f-Strings (Formatted String Literals)
# Preferred String Formatting in Modern Python (3.6+)
# --------------------------------------------------------
# f-Strings are the way to embed values and expressions inside strings.
# This section replaces:
# - string concatenation (`+`)
# - old `%` string interpolation
# - most uses of `str.format()`

# -------------------------------
# 🔹 Basic Variable Insertion
# -------------------------------

name = "Al"
age = 4000

# Variables are inserted directly using braces {}
print(f"My name is {name}.")
print(f"I am {age} years old.")


# ------------------------------------------------------------
#🔹 Inline Expressions (Core Advantage of f-Strings)
# You can place expressions, not just variables, inside `{}`
# ------------------------------------------------------------

age = 4000

# Arithmetic inside the string
print(f"Next year I will be {age + 1} years old.")

# Function calls inside the string
print(f"Age as a string: {str(age)}")

# Built-in functions
print(f"Name length: {len('Al')}")


# ------------------------------------
# 🔹 String Methods Inside f-Strings
# ------------------------------------

name = "  Alice  "

# String cleanup and transformation
print(f"Clean name: {name.strip()}")
print(f"Uppercase: {name.strip().upper()}")
print(f"Lowercase: {name.strip().lower()}")


# ------------------------------------------------------
# 🔹 Numeric Formatting (Very Common in Data Analysis)
# ------------------------------------------------------
# Decimal Precision

value = 123.456789

# Format to 2 decimal places
print(f"Rounded value: {value:.2f}")


#Thousands Separator
revenue = 1250000

# Add commas for readability
print(f"Revenue: ${revenue:,}")

# ---------------------------------------
#🔹 Alignment and Width (Tabular Output)
# ---------------------------------------
# Useful for simple reports, logs, and console tables.

item = "Apples"
price = 2.5
quantity = 12

# Left-align text, right-align numbers
print(f"{'Item':<10}{'Qty':>6}{'Price':>10}")
print(f"{item:<10}{quantity:>6}{price:>10.2f}")


# --------------------------------------------
# 🔹 Boolean and Conditional Expressions
# --------------------------------------------

score = 87

# Conditional logic inside an f-string
print(f"Status: {'Pass' if score >= 70 else 'Fail'}")


# -----------------------------------------------------------
#🔹 f-Strings vs Legacy Formatting (For Reference Only)
# -----------------------------------------------------------
name = "Al"
age = 4000

# ❌ Old-style (% formatting)
"My name is %s. I am %s years old." % (name, age)

# ❌ String concatenation
"My name is " + name + ". I am " + str(age) + " years old."

# ✅ Preferred modern approach
f"My name is {name}. I am {age} years old."


# -----------------------------------
# 🔹 Important Rule
# -----------------------------------

# Always include the leading 'f'
print(f"My age is {age}")

# Without 'f', braces are treated as literal text
print("My age is {age}")  # Incorrect


# ✅ Key Takeaways

# - f-Strings are the **standard** for modern Python formatting
# - They improve **readability**, **safety**, and **maintainability**
# - Expressions, functions, and formatting can be embedded directly
# - Ideal for:
#   - Data analysis
#   - Logging and reporting
#   - Notebooks and teaching materials
# - Prefer f-strings over:
#   - `%` formatting
#   - String concatenation
#   - Most uses of `str.format()`
`````

`````python
# -------------------------------------------------------------
# Range() Function Examples
# -------------------------------------------------------------

# Loop through indices in a string
# Helpful when you need both the index and character
name = "Dawn"
for i in range(len(name)):
    print(i, name[i])
print()  # spacing


# Generate a sequence of numbers
numbers = list(range(1, 11))   # 1 through 10
print("Numbers List:", numbers, end="\n\n")


# Loop using indices in a list
# (Not always needed in Python, but useful for understanding range)
for i in range(len(numbers)):
    print(i, numbers[i])
print()  # spacing


# Loop in steps: range(start, stop, step)
# This prints 0, 5, 10, 15 because 20 is not included
for i in range(0, 20, 5):
    print(i)
print()  # spacing


# Simulating repeated events (very common in data science)
# for trial in range(1000):     # run a process 1000 times
#     run_model_once()
`````

`````python
# -------------------------------------------------------------
# Indexing strings with positive and negative indices
# -------------------------------------------------------------

# Indexing: Accessing Individual Characters
# Python is **0-indexed**, so the first character has index `0`.
# message with 0 - 23 characters (including spaces) and length of 24
message = "I hope it snows TONIGHT!"

# Positive indices (from the start)
print(message[0])   # 'I'  (first character)
print(message[2])   # 'h'
print(message[9])   # 't' 

# Negative indices (from the end)
print(message[-1])  # '!'  (last character)
print(message[-2])  # 'T'
print(message[-8])  # 's'

# Length of the string to confirm valid index range
print("Length:", len(message))

# Trying to access an index >= len(message) would raise IndexError
# message[100]  # do not run this line; it will error
`````

`````python
# -------------------------------------------------------------
# Basic slicing examples (positive step only)
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# -------------------------------------------------------------
# 1) Slice from the beginning up to a position (stop not included)
# -------------------------------------------------------------
# Index positions: 0 1 2 3 4 5 6
# Characters:      I _ h o p e _

first_phrase = message[0:6:1]   # indices 0 through 5 (stop not included)
print(first_phrase)

first_phrase = message[0:6]     # step defaults to 1
print(first_phrase)             # Output: 'I hope'


# -------------------------------------------------------------
# 2) Slice from index 2 up to index 6 (stop not included)
# -------------------------------------------------------------
# Index positions: 2 3 4 5
# Characters:      h o p e

middle_slice = message[2:6:1]   # indices 2 through 5 (stop not included)
print(middle_slice)

middle_slice = message[2:6]
print(middle_slice)             # Output: 'hope'


# -------------------------------------------------------------
# 3) Slice from index 6 to the end
# -------------------------------------------------------------
# message[6:] means:
#   start = 6
#   stop = default (len(message))
#   step = 1

after_first_two_words = message[6:None:1]  
# Using None to explicitly slice syntax with start, stop, and step
print(after_first_two_words)

after_first_two_words = message[6:]
print(after_first_two_words)         # Output: ' it snows TONIGHT!'


# -------------------------------------------------------------
# 4) Slice with a step (every other character)
# -------------------------------------------------------------
# message[0:6:2] takes indices: 0, 2, 4

skip_chars = message[0:6:2]
print(skip_chars)               # Output: 'Ihp '
`````

`````python
# -------------------------------------------------------------
# 5) Reverse the entire string
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[::-1] uses the slicing pattern [start : stop : step]
#   - start: omitted → Python begins at the last character (because step = -1)
#   - stop: omitted → Python continues until it moves past the first character
#   - step: -1 → move backward one character at a time
#
# Together, message[::-1] produces a full reverse of the string.

reversed_message = message[::-1]
print(reversed_message)
# This is the most common Pythonic way to reverse a string.


# -------------------------------------------------------------
# 6) Slice from the end using negative indices
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[-8:-1] uses the slicing pattern [start : stop : step]
#   - start at index -8  → the 'T' in 'TONIGHT'
#   - stop at index -1   → but EXCLUDE index -1 (the '!')
#   - step defaults to 1 → moving forward left-to-right

# Negative index mapping for this string:
#    message = "I hope it snows TONIGHT!"
#                      T   O   N   I   G   H   T   !
#            indices: -8  -7  -6  -5  -4  -3  -2  -1

last_word = message[-8:-1]
print(last_word)       # Output: 'TONIGHT' (excludes the '!')


# -------------------------------------------------------------
# 7) Slice the last 8 characters including the final character
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[-8:] uses the slicing pattern [start : stop : step]
#   - start at index -8  → the first 'T' in 'TONIGHT'
#   - stop omitted       → continue to the end of the string
#   - step defaults to 1 → move forward left-to-right

# Negative index mapping for this string:
#    message = "I hope it snows TONIGHT!"
#                      T    O    N    I    G    H    T    !
#            indices: -8  -7   -6   -5   -4   -3   -2   -1

tail_slice = message[-8:]
print(tail_slice)      # Output: 'TONIGHT!'
`````

`````python
# -------------------------------------------------------------
# 6) Slice from the end using negative indices
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[-8:-1] uses the slicing pattern [start : stop : step]
#   - start at index -8  → the 'T' in 'TONIGHT'
#   - stop at index -1   → but EXCLUDE index -1 (the '!')
#   - step defaults to 1 → moving forward left-to-right

# Negative index mapping for this string:
#    message = "I hope it snows TONIGHT!"
#                      T   O   N   I   G   H   T   !
#            indices: -8  -7  -6  -5  -4  -3  -2  -1

last_word = message[-8:-1]
print(last_word)       # Output: 'TONIGHT' (excludes the '!')
`````

`````python
# ---------------------------------------------------------------
`````
`````python
# Universal String Index Map Helper (Positive & Negative Indices)
# ---------------------------------------------------------------

# Original text:
customer_testimonial = '''I love skiing. It's my favorite hobby.
Some people say, "It's not a hobby, it's life."'''
print(repr(customer_testimonial))  # repr() returns an unambiguous string, often a valid expression

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")

show_index_map(customer_testimonial)
`````

`````python
# -------------------------------------------------------------
# Measuring Length with len()
# -------------------------------------------------------------
# len() is a built-in Python function that returns:
#   • the number of characters in a string
#   • the number of items/elements in a list (or other collections)
# -------------------------------------------------------------


# -------------------------------------------------------------
# 1) len() with a STRING
# -------------------------------------------------------------
headline = "Fresh powder on the mountain today!"

# len(string) → counts every character, including:
#   spaces, punctuation, and special characters.
char_count = len(headline)

print("Original headline:", headline)
print("Character count:", char_count)     # total number of characters


# -------------------------------------------------------------
# 2) len() with a LIST
# -------------------------------------------------------------
stages = ["beginner", "enthusiast", "pro"]

# len(list) → counts number of items in the list (3 items here)
element_count = len(stages)

print("\nStages list:", stages)
print("Number of elements in list:", element_count)


# -------------------------------------------------------------
# 3) len() with SPLIT STRINGS (words)
# -------------------------------------------------------------
# headline.split() splits the string into words using spaces
word_list = headline.split()

# Now len() counts number of words in the list
word_count = len(word_list)

print("\nWord list from split():", word_list)
print("Number of words:", word_count)
`````

`````python
# -------------------------------------------------------------
# Len() Exercise: Simple Cryptography
# -------------------------------------------------------------
# Goal:
#   Use the length of different messages as "indexes" into the
#   alphabet string to build a password.
#
# Process:
#   1. Measure the length of each message using len()
#   2. Use that number as an index into the alphabet
#   3. Concatenate the resulting letters into a password
# -------------------------------------------------------------

# A reference alphabet for indexing
alphabet = "abcdefghijklmnopqrstuvwxyz"

# Messages of varying lengths
message1 = "Hello World!"
message2 = ""
message3 = "It snowed a lot today"
message4 = "eeee"
message5 = "I love snow!!"

# -------------------------------------------------------------
# Step-by-step explanation:
#
print(len(message1))  #  → 12
print(alphabet[12])  # → 'm'

print(len(message2))  # → 0
print(alphabet[0])  # → 'a'

print(len(message3))  # → 21
print(alphabet[21])  # → 'v'

print(len(message4))  # → 4
print(alphabet[4])  # → 'e'

print(len(message5))  # → 13
print(alphabet[13])  # → 'n'
#
# Final password: "maven"
# -------------------------------------------------------------

password = (
    alphabet[len(message1)] +
    alphabet[len(message2)] +
    alphabet[len(message3)] +
    alphabet[len(message4)] +
    alphabet[len(message5)]
)

print("Generated password:", password)
`````

`````python
# -------------------------------------------------------------
# String Methods: Demonstrating .strip() to Remove Extra Spaces
# -------------------------------------------------------------
# Methods are functions that apply only to a specific data type or object.
# .strip() removes leading and trailing whitespace automatically - no indexing needed!
# -------------------------------------------------------------

# Example string with intentional leading/trailing spaces
message = "  I hope it snows TONIGHT!  "

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------
# This function helps visualize how .strip() affects the positions
# of characters by printing:
#   • positive indexing (0 → len-1)
#   • negative indexing (-len → -1)
#
# Seeing the spaces in the index map makes it clear
# why .strip() removes ONLY the characters at the edges.
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula for corresponding negative index
        print(f"{i:3} {neg:4} {repr(ch)}")


# Display index map BEFORE stripping spaces
print("Before .strip():")
show_index_map(message)

# -------------------------------------------------------------
# Using .strip() to remove leading and trailing spaces
# -------------------------------------------------------------
# .strip() removes ONLY whitespace at the beginning and end.
# It does NOT remove spaces inside the string.
cleaned = message.strip()

print("\nAfter .strip():")
show_index_map(cleaned)   # show the new indexing after cleaning

# Print both versions to compare clearly
print(f"\nOriginal message: [{message}]")
print(f"Stripped message: [{cleaned}]")
`````

`````python
# -------------------------------------------------------------
`````
`````python
# String Methods: .strip("."), .lstrip(), .rstrip() — remove unwanted whitespace
# -------------------------------------------------------------
# .strip()  → removes BOTH leading + trailing whitespace
# .lstrip() → removes whitespace only on the LEFT
# .rstrip() → removes whitespace only on the RIGHT

message = "   I love skiing and snow.   "   # spaces included for .strip() examples
l_cleaned = message.lstrip()
r_cleaned = message.rstrip()

print(f"[{l_cleaned}]")  # left spaces removed
print(f"[{r_cleaned}]")  # right spaces removed

punctuated = "...powder day!!!..."
# .strip(".") removes ONLY the character "." from both ends (not from the middle)
print(punctuated.strip("."))
`````

`````python
# -------------------------------------------------------------
# String Methods: .strip(chars)
# removes only the characters listed in `chars`,
# and only from the **beginning and end**, *not the middle*
# -------------------------------------------------------------

# Example strings
message = "   I love skiing and snow.   "     # has leading + trailing spaces
punctuated = "...powder day!!!..."            # mix of periods and exclamation marks

# Helper: Show positive & negative index map for any string
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n
        print(f"{i:3} {neg:4} {repr(ch)}")
        
print("\n--- Punctuated String ---")
show_index_map(punctuated)

# Remove only periods from both ends
no_periods = punctuated.strip(".")
print("\n--- After .strip('.') ---")
show_index_map(no_periods)

# Remove periods AND exclamation marks from both ends
cleaned_all = punctuated.strip(".!")
print("\n--- After .strip('.!') (remove . and !) ---")
show_index_map(cleaned_all)

print(f"\nOriginal:   [{punctuated}]")
print(f"strip('.'): [{no_periods}]")
print(f"strip('.!'):[{cleaned_all}]")
`````

`````python
# String Methods
# -------------------------------------------------------------
# 1. .find() — locate the FIRST occurrence of a substring
# -------------------------------------------------------------
# .find("snow") returns:
#   - the index of the FIRST match
#   - OR -1 if the substring is not found

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------

message = "I hope it snows tonight!"

def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")

show_index_map(message)

first_snow_index = message.find("snow")
print("Index of 'snow':", first_snow_index)

missing_index = message.find("rain")        # "rain" does not appear
print("Index of 'rain':", missing_index)    # → -1 (Python’s “not found” indicator)
`````

`````python
# String Methods
# -------------------------------------------------------------
`````
`````python
# .split and .join() — convert between strings and word lists
# -------------------------------------------------------------

sentence = "Fresh powder on the mountain today"

# .split() with no argument splits on whitespace (spaces, tabs, etc.)
words = sentence.split()
print(words)     # → list of individual words

# .join() takes a LIST and assembles it back into a string
rejoined = " ".join(words)    # join using a single space (Pythonic standard)
print(rejoined)

separator = " "
rejoined = separator.join(words)  # join with a manual separated (not preferred)
print(rejoined)

# ----------------------
# Examples of .join()
# ----------------------

words = ['Hello', 'beautiful', 'world']

print(''.join(words))       # without spaces
# Hellobeautifulworld

print(' '.join(words))       # with spaces
# Hello beautiful world

print('-'.join(words))       # dashes without spaces
# Hello-beautiful-world

print(' - '.join(words))       # dashes with spaces
# Hello - beautiful - world

print(', '.join(words))        # comma + space (like CSV)
# Hello, beautiful, world

print('|'.join(words))         # just pipes
# Hello|beautiful|world

print('\n'.join(words))        # new line between each
# Hello
# beautiful
# world
`````

`````python
# String Methods
# -------------------------------------------------------------
# .count()
# -------------------------------------------------------------
message = "I hope it snows tonight because snow is coming."

# -------------------------------------------------------------
# 1. Count single characters
# -------------------------------------------------------------
print(message.count("o"))     # count how many 'o' characters
print(message.count("i"))     # count of 'i' characters

# -------------------------------------------------------------
# 2. Count whole substrings
# -------------------------------------------------------------
print(message.count("snow"))   # number of times "snow" appears

# -------------------------------------------------------------
# 3. Count in a specific range of indexes: .count(sub, start, end)
# -------------------------------------------------------------
# Count only in the first 20 characters
print(message.count("o", 0, 20))

# Count "snow" only in the second half of the sentence
midpoint = len(message) // 2
print(message.count("snow", midpoint, len(message)))
`````

`````python
# String Methods
# -------------------------------------------------------------
# Method Chaining — perform multiple text operations in sequence
# -------------------------------------------------------------

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------

message = "  I HOPE it snows TONIGHT!  "
cleaned_message = message.strip()

def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")

show_index_map(cleaned_message)

# GOAL:
#   Extract everything AFTER "I HOPE "
#   Make the result lowercase
#   Convert to a list of words
#
# Breakdown of what happens:
#   message.strip()      → removed outer space first for cleaned_message
#   [7::]                → skipping [0:6] and slicing on [7:]
#   .lower()             → convert the remaining text to lowercase
#   .split()             → break into list of words

# processed = cleaned_message[7:].lower().split()
processed = (
    cleaned_message
    [7::]         # slice off "I HOPE "
    .lower()      # convert to lowercase
    .split()      # convert to list of words
)

print(processed)                     # ['it', 'snows', 'tonight!']
print("Word count:", len(processed)) # demonstrates chaining + len()
`````

### Understanding Why `__doc__` Works for Functions but *not* Variables

Python allows you to attach documentation (called a **docstring**) to functions, classes, and modules by placing a triple-quoted string immediately after the definition. These docstrings are accessible using the `.__doc__` attribute.

However, a **variable does not store docstrings**. If you call `.__doc__` on a string variable, Python returns the documentation for the **`str` class itself**, not the variable’s content.

This section compares how `__doc__` behaves for:
- a triple-quoted *multi-line string variable*, and  
- a properly documented function.

### 📌 Key Takeaways

- Triple-quoted strings assigned to variables are **just strings**, not docstrings.  
- Only **functions, classes, and modules** store real docstrings accessible via `.__doc__`.  
- Calling `.__doc__` on a string gives you the documentation for Python’s built-in `str` class.  
- To see a variable’s contents, print the variable directly.  
- To see a function’s docstring, print `function.__doc__`.

## 🔍 Slicing Strings in Python

Python slicing uses the pattern:

```python
string[start : stop : step]
```

### ✔️ What each argument really means

- **start** — index to begin the slice  
  - default depends on step direction  
  - for positive step, default = `0`  
- **stop** — index to stop *before*  
  - default depends on step direction  
  - for positive step, default = `len(string)`  
- **step** — how much to move between characters  
  - default = `1`  
  - must be **non-zero**

---

### 🔹 Important correction
The defaults for `start` and `stop` are **NOT always 0 and len(string)**.  
Those defaults apply **only when the step is positive**.

We will use **positive-step slicing only** in this section,  
so the defaults here behave normally.

---


---

## 🔄 Reversing Strings and Using Negative Indices

Python allows slicing with **negative indices** and **negative step values**, which gives you powerful ways to read a string from right to left.

### ✔️ Important Rules

- Negative indices count from the right  
  ```
  message = "I hope it snows TONIGHT!"
  Index:    -23 … -1
  ```
- A **negative step (-1)** means slicing moves backwards.
- When step is negative:
  - default `start` becomes the **last index**
  - default `stop` becomes **before the first character**
- Slicing still stops **before** the `stop` index — even for negative steps.

## ✂️ Cleaning Text with `.strip()` — Visualizing Leading & Trailing Whitespace

In Python, the `.strip()` method removes **only** the whitespace at the *beginning* and *end* of a string.  
It does **not** remove spaces inside the text.  

The helper function below prints a full index map (positive and negative indices) so you can *see*  
how `.strip()` changes the string's length and character positions.

```python
# -------------------------------------------------------------
# Demonstration: .strip() Removes Leading & Trailing Whitespace
# -------------------------------------------------------------
# This example shows exactly which characters are removed and how
# the string's indexing changes after cleaning.
# -------------------------------------------------------------

# Example string with intentional surrounding spaces
message = "  I hope it snows TONIGHT!  "

# -------------------------------------------------------------
# Helper Function: Index Map (Positive + Negative Indices)
# -------------------------------------------------------------
def show_index_map(s):
    """
    Display the positive and negative index for each character in s.
    Useful for understanding slicing and how .strip() affects length.
    """
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n   # convert positive → negative index
        print(f"{i:3} {neg:4} {repr(ch)}")

# -------------------------------------------------------------
# Before .strip(): Visualize All Characters (Including Spaces)
# -------------------------------------------------------------
print("Before .strip():")
show_index_map(message)

# -------------------------------------------------------------
# Apply .strip(): Removes ONLY outside whitespace
# -------------------------------------------------------------
cleaned = message.strip()

print("\nAfter .strip():")
show_index_map(cleaned)

# -------------------------------------------------------------
# Compare Outputs Clearly
# -------------------------------------------------------------
print(f"\nOriginal message: [{message}]")
print(f"Stripped message: [{cleaned}]")
```

---

### 🔑 Key Takeaways

- `.strip()` only removes whitespace at the edges of the string.  
- Interior spaces remain unchanged.  
- Indexing shifts because the string becomes shorter.  
- Negative indices adjust automatically to the new string length.  
- The index map helper makes whitespace behavior visible and intuitive.

## ✂️ Advanced Stripping Methods: `.lstrip()`, `.rstrip()`, `.strip(chars)`

Python provides three related cleaning methods for removing characters from the *edges* of a string:

| Method | Removes | Example |
|--------|----------|----------|
| `.strip()` | Both left + right whitespace (or characters) | `s.strip()` |
| `.lstrip()` | Left-side only | `s.lstrip()` |
| `.rstrip()` | Right-side only | `s.rstrip()` |
| `.strip(chars)` | Removes *only the characters listed* in `chars` from both ends | `s.strip(".!?")` |

Each version affects indexing differently because the *length of the string changes*.  
The helper below visualizes those changes.

---

## 🔍 Example Strings

```python

punctuated = "...powder day!!!..."            # mix of periods and exclamation marks

```

```python
left_stripped = message.lstrip()
```

```python
print(f"\nrstrip(): [{right_stripped}]")
```

---

## 3️⃣ `.strip(chars)` – Remove Specific Characters From Both Ends

`.strip(chars)` removes **only** the characters listed in `chars`,  
and only from the **beginning and end**, *not the middle*.

```python
print("\n--- Punctuated String ---")
show_index_map(punctuated)

show_index_map(no_periods)

print(f"strip('.'): [{no_periods}]")
print(f"strip('.!'):[{cleaned_all}]")
```

---

## 🔑 Key Takeaways

- `.strip()`, `.lstrip()`, and `.rstrip()` do **not** modify the middle of the string—only the edges.
- `.strip(chars)` removes *only the characters listed* in `chars`, not arbitrary whitespace.
- Stripping changes the string length, which shifts both positive and negative indices.
- The index map helper is the most reliable way to understand exactly what was removed.
- These methods are foundational for cleaning raw text and preparing real-world data for analysis.

## 🔍 Searching Within Strings Using `.find()`

The `.find()` method locates the **first occurrence** of a substring inside a string.  
It returns:

- **the index of the first match**, or  
- **–1** if the substring is not found.

Understanding `.find()` becomes much easier when combined with the index-map helper.

---

## Example String

```python
message = "I hope it snows tonight!"
```

---

## 🧭 Helper: Universal Index Map (Positive & Negative Indices)

```python
def show_index_map(s):
```

## 1️⃣ Find a Substring That Exists

```python

print("\nIndex of 'snow':", first_snow_index)
```

This tells you **where the substring begins**, so you can slice it:

```python
snow_word = message[first_snow_index : first_snow_index + len("snow")]
print("Extracted substring:", snow_word)
```

---

## 2️⃣ Find a Substring That Does *Not* Exist

```python
missing_index = message.find("rain")
print("\nIndex of 'rain':", missing_index)   # returns -1 when not found
```

Because `.find()` returns `-1`, it prevents indexing errors:

```python
if missing_index == -1:
    print("Substring 'rain' does not appear in the message.")
```

---

## 🔑 Key Takeaways

- `.find()` returns the starting index of the **first** match.
- If the substring is not found, it returns **–1**, not an error.
- Always check for `-1` before slicing based on `.find()`.
- Visualizing the string with the index map makes `.find()` easier to use in slicing and extraction tasks.
- `.find()` is frequently used in:
  - text parsing  
  - cleaning messy CSV fields  
  - locating delimiters  
  - substring extraction  
  - quick data validation  


---

## 🧮 Key Points — `.count()`

- `text.count("a")` → counts how many times the character `"a"` appears  
- `text.count("snow")` → counts how many times the substring `"snow"` appears  
- `text.count(x, start, end)` → counts only within the slice `[start:end]`  
  - `start` is **inclusive**  
  - `end` is **exclusive** (standard Python slicing rule)  
- `.count()` tells you **how many times**, *not where* the substring occurs  
- `.count()` does **not** count overlapping matches  
  - Example: `"aaa".count("aa")` → `1` (not `2`) because `"aa"` overlaps itself  

---

## 📌 When to Use `.count()`

Use `.count()` when answering questions like:

- “How many times does this character appear?”  
- “How many times is this substring repeated?”  
- “How many occurrences appear inside a specific part of the string?”  

Use `.count()` for **frequency**, not for locating positions.

For positions, prefer:

- `.find()`  
- `.rfind()`  
- `.index()`

# 🧵 Python Strings – Master Reference Section

This section summarizes the **most important string operations** in Python, grouped into five practical categories:

1. Navigating strings (length, indexing, slicing)  
2. Changing appearance (formatting and cleaning)  
3. Splitting and joining (text ↔ lists)  
4. Searching and checking (finding content)  
5. Combining and formatting strings (building output)

It also includes a **universal index map helper** that prints both positive and negative indices for any string.  
This helper makes slicing and character access much easier for longer text.

---

## 1️⃣ Navigating Strings – Length, Indexing, and Slicing

```python
# Basic string for navigation examples
word = "Analytics"

# Length: total number of characters
print("Length of word:", len(word))      # 9

# Positive indexing: 0-based from left to right
print("word[0]:", word[0])              # 'A'
print("word[1]:", word[1])              # 'n'
print("word[8]:", word[8])              # 's' (last character)

# Negative indexing: -1 is last character, -2 is second-to-last, etc.
print("word[-1]:", word[-1])           # 's'
print("word[-2]:", word[-2])           # 'c'

# Slicing: [start:stop] includes start, excludes stop
print("word[0:4]:", word[0:4])         # 'Anal'
print("word[4:9]:", word[4:9])         # 'ytics'

# Slicing with omitted bounds
print("word[:4]:", word[:4])           # 'Anal' (from start up to index 4)
print("word[4:]:", word[4:])           # 'ytics' (from index 4 to end)

# Slicing with a step value: [start:stop:step]
print("word[0:9:2]:", word[0:9:2])     # 'Aayc' (every 2nd character)
print("word[::-1]:", word[::-1])       # 'scitylanA' (reverse string)
```

---

### Universal String Index Map Helper (Positive and Negative Indices)

This helper prints a table that shows **both positive and negative indices** for each character in a string.  
It is useful when you want to design precise slices without counting characters manually.

```python
# ---------------------------------------------------------------
# Universal String Index Map Helper (Positive & Negative Indices)
# ---------------------------------------------------------------

customer_testimonial = '''I love skiing. It's my favorite hobby.
Some people say, "It's not a hobby, it's life."'''

# repr() shows the string with quotes and escape characters clearly
print("Original string (repr):")
print(repr(customer_testimonial), end="\n\n")

def show_index_map(s):
    """
    Print a table showing positive and negative indices
    for every character in the string s.
    """
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n          # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")

# Call the helper to see index mapping for the testimonial
show_index_map(customer_testimonial)
```

---

## 2️⃣ Changing Appearance – Formatting and Cleaning

These methods adjust **how a string looks** without changing its underlying meaning.

```python
text = "  Python Data Analytics  "

# Case transformations
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Title case:", text.title())

# Whitespace cleanup

print("Left strip:", repr(text.lstrip()))      # remove spaces on the left only
print("Right strip:", repr(text.rstrip()))     # remove spaces on the right only

# Replace parts of the string
cleaned = text.strip().replace("Data", "Business")
print("Replaced word:", cleaned)               # 'Python Business Analytics'
```

## 3️⃣ Splitting and Joining – Text to Lists and Back

Splitting and joining are essential for **parsing** and **reconstructing** text, especially when working with CSV fields or logs.

```python
sentence = "I love Python for data analysis"

# Split on a different delimiter
csv_line = "IBM,MSFT,AAPL,GOOG"

# Join a list back into a single string
joined_with_spaces = " ".join(words)
print("Joined with spaces:", joined_with_spaces)

joined_with_dash = "-".join(symbols)
print("Joined with dashes:", joined_with_dash)
```

---

## 4️⃣ Searching and Checking – Finding Content

These operations help you **inspect** whether a string contains certain text and where it occurs.

```python
message = "It is snowing today, and I love the snow."

# Simple membership check using "in"
print('"snow" in message:', "snow" in message)        # True

# Find the index of the first occurrence, or -1 if not found
first_pos = message.find("snow")
print("First 'snow' at index:", first_pos)

# Find another substring
print("Index of 'rain':", message.find("rain"))       # -1 (not found)

# Count occurrences
print("Count of 'snow':", message.count("snow"))      # how many times "snow" appears

# Check the beginning and end of the string
print("Starts with 'It is':", message.startswith("It is"))
print("Ends with 'snow.':", message.endswith("snow."))
```

## 5️⃣ Combining and Formatting Strings – Building Output

These patterns are used constantly for **program output, logs, file names, and messages**.


## 🧵 f-Strings (Formatted String Literals)
### Preferred String Formatting in Modern Python (3.6+)

f-Strings are the **recommended and modern** way to embed values and expressions inside strings.
They are readable, safe, concise, and widely used in data analysis, scientific computing, and production code.

This section replaces:
- string concatenation (`+`)
- old `%` string interpolation
- most uses of `str.format()`

## 🔹 Basic Variable Insertion

## 🔹 f-Strings vs Legacy Formatting (For Reference Only)

- f-Strings are the **standard** for modern Python formatting

## String Formatting: Alternatives for More Control
### Most robust / natural looking solution in most real-world cases:
```python
text = "it's a test of 3d printing in 2026"
result = ' '.join(word.capitalize() for word in text.split())
# → "It's A Test Of 3d Printing In 2026"

# Even better (preserves original spacing):
import re
def title_case(s):
    return re.sub(r"[A-Za-z]+('[A-Za-z]+)?",
                  lambda m: m.group(0).capitalize(),
                  s)

print(title_case("it's o'brien's 3d world"))  
# → "It's O'Brien's 3d World"   ← much nicer!
```

## Strings and Loops – Repeating Actions

Strings often appear inside loops, especially when you need to **repeat actions** or process characters one by one.

```python
subject = "My"
noun = "Maven snowboard"
verb = "cost"
price = 22.55

sentence = subject + " " + noun + " " + verb + " $" + str(price) + "! "
print(sentence, end="\n\n")

# Repeat as one long line using multiplication
print("Repeat 3 times as one long string:")
print(sentence * 3, end="\n\n")

# Repeat line by line using range()
print("Repeat 3 times on separate lines:")
for _ in range(3):   # range(3) → 0, 1, 2, but we do not need the index, so use "_"
    print("Repeat 3 times:", sentence, end="\n\n")
```

## 🔑 Key Takeaways

- Python strings are **sequences of characters**, which means indexing, slicing, and looping apply.
- You create strings with quotes or by converting other data types using `str()`.  
- Use **string arithmetic**:
  - `+` to concatenate text  
  - `*` to repeat text  
- Access single characters with **indexing**:
  - `s[0]` is the first character, `s[-1]` is the last.  
- Use **slicing** `s[start:stop:step]` to extract substrings or to reverse text with `s[::-1]`.  
- `len()` returns the number of characters in a string or the number of elements in a list.  
- Learn core **string methods**:
  - `.find()` to locate substrings; returns the index of the first match, or `-1` if not found
  - `.upper()` and `.lower()` to standardize case; convert entire strings to new casing
  - `.strip()` (and `.lstrip()`, `.rstrip()`) removes whitespace (or a specified character) on both ends
  - `.replace(old, new)` substitutes all occurrences of one substring  
  - `.split()` breaks a string into a list
  - `.join()` combines list elements into one string
  - `.count()` returns the number of non-overlapping occurrences of a substring within a string. It can optionally take start and end argument

### 🧠 The Five Categories of String Operations:

  1. Navigating strings (length, indexing, slicing)
  2. Changing appearance (case changes, trimming, replacing)
  3. Splitting and joining (text ↔ lists)
  4. Searching and checking (membership, find, count, startswith, endswith)
  5. Combining and formatting (concatenation, f-strings, repetition)
- The **index map helper** removes the need to count positions by hand and makes slicing more reliable.
- Methods like `.split()` and `.join()` are critical for data cleaning and preparing text fields for analysis.
- Searching methods (`in`, `.find()`, `.count()`, `.startswith()`, `.endswith()`) are often used for simple text filters and validations.
- You can **chain methods** to build compact data-cleaning pipelines, such as `message.strip().lower().split()`.  
- Use **f-strings** (`f"..."`) to combine variables and text, and format numbers (for example, `price:.2f`) for clean, dynamic output.  
- These core operations build the foundation for more advanced text handling in `pandas` using the `.str` accessor.

These patterns are the foundation for working with categorical fields, cleaning raw text, and building readable reporting messages in real analysis projects.

# 🐍 Chapter 5. Conditional Logic: 
## 🧩 Control Flow Essentials (if / elif / else) (Part 2)


## 2. If Statements & Core Control Flow

In Part 1 you built a strong foundation in **Boolean logic**:
comparisons, membership tests, `and` / `or` / `not`, precedence, parentheses, and short-circuiting.

Now we use those Boolean expressions to **control the flow of a program** with:

- `if`
- `elif`
- `else`

Control flow lets Python **choose** which block of code to run based on conditions.

---

### 2.1 What Is Control Flow?

Without conditions, a program runs **top to bottom** in a straight line.

With control flow, a program can:

- make decisions  
- follow different paths  
- skip or execute blocks of code based on Boolean results  

You can think of it as a series of **questions**:

> If this is true, do this.  
> Otherwise, do something else.

---

## 2.2 The `if` Statement

### Basic pattern

```python
# General syntax

if condition:
    # This indented block runs only if condition is True
    do_something()
```

Important points:

- `condition` must evaluate to a Boolean (`True` or `False`)
- A colon `:` ends the `if` line
- The following indented block is the **body** of the `if`
- Indentation in Python is not cosmetic. It defines the block.

---

### Example: Basic `if` with a Boolean expression

```python
# --- Example: check if a customer qualifies for free shipping ---

order_total = 120

# Boolean expression from Part 1: comparison operator
if order_total >= 100:
    print("Free shipping applied!")
    
print("Order processed.")
```

What happens:

- If `order_total >= 100` is True, Python prints `"Free shipping applied!"`
- `"Order processed."` prints every time, because it is **outside** the `if` block

Try changing `order_total` to 80 and re-running.

---

## 2.3 `if` + `else`: Two-Way Decision

An `if` statement can stand alone.  
`else` **cannot** stand alone. It always pairs with an `if`.

Use `else` when you want a **default path** if the condition is False.

```python
# General pattern

if condition:
    # runs when condition is True
    do_this()
else:
    # runs when condition is False
    do_that_instead()
```

---

### Example: Free shipping or standard shipping

```python
# --- Example: two-way decision with if/else ---

order_total = 75

if order_total >= 100:
    print("Free shipping applied.")
else:
    print("Standard shipping rate will be used.")

print("Order complete.")
```

- If `order_total` is 120 → prints `Free shipping applied.`
- If `order_total` is 75 → prints `Standard shipping rate will be used.`

Only **one** of the `if` or `else` blocks runs, never both.

---

### Example: Using a helper to show which branch ran

```python
# --- Helper: show which branch executed ---

def check_shipping(order_total):
    print(f"\nChecking shipping for order_total = {order_total}")
    
    if order_total >= 100:
        print("Branch: if → Free shipping")
    else:
        print("Branch: else → Standard shipping")

# Try a few values
check_shipping(120)
check_shipping(75)
check_shipping(100)
```

This example makes branching very visible.  
You can change values and see which branch is selected.

---

## 2.4 `if` + `elif` + `else`: Multi-Way Decision

Sometimes there are **more than two** meaningful cases.  
This is where `elif` (short for "else if") is useful.

```python
# General pattern

if condition_1:
    # runs if condition_1 is True
    block_1
elif condition_2:
    # runs if condition_1 is False and condition_2 is True
    block_2
elif condition_3:
    # runs if all previous conditions are False and condition_3 is True
    block_3
else:
    # runs if none of the above conditions are True
    default_block
```

Rules:

- Conditions are checked **top to bottom**
- As soon as one condition is True, Python runs that block and **skips the rest**
- `if` can exist alone  
- `if` + `else` is a two-branch decision  
- `if` + one or more `elif`s + optional `else` creates a **decision chain**

---

### Example: Customer status and discount level

We will build a simple rule:

- If status is `"VIP"` → 20% discount  
- Else if status is `"LOYAL"` → 10% discount  
- Else if status is `"NEW"` → 5% discount  
- Else → no discount

```python
# --- Example: multi-way decision with if/elif/else ---

def discount_for_status(status):
    print(f"\nChecking discount for status = {status!r}")
    
    if status == "VIP":
        print("20% discount applied.")
    elif status == "LOYAL":
        print("10% discount applied.")
    elif status == "NEW":
        print("5% discount applied.")
    else:
        print("No discount for this status.")

# Try a few different statuses
discount_for_status("VIP")
discount_for_status("LOYAL")
discount_for_status("NEW")
discount_for_status("GUEST")
```

Observe:

- Only one branch runs per call  
- The order of conditions matters

---

### Ordering matters: more specific conditions should appear first

Consider this pattern:

```python
# Example of ordering

score = 95

if score >= 60:
    print("Pass")
elif score >= 90:
    print("High Distinction")
```

For `score = 95`, the result is `"Pass"` because:

- Python checks `score >= 60` first → True  
- It runs that block and **does not** check `elif score >= 90`

Better ordering:

```python
score = 95

if score >= 90:
    print("High Distinction")
elif score >= 60:
    print("Pass")
else:
    print("Fail")
```

Principle:

- Put **stricter** or **higher priority** conditions earlier
- Put more general conditions later in the chain

---

## 2.5 Combining Boolean Logic with Control Flow

Everything from Part 1 can be used inside `if`, `elif`, and `else` conditions.

- Comparisons: `price < 100`
- Membership: `"hat" in items`
- Logical operators: `and`, `or`, `not`

### Example: Discount rule with multiple conditions

Rule:

- Apply a discount if:
  - the customer is VIP, or  
  - the order total is at least 100 and the item is in the discount list

```python
# --- Example: combining Boolean logic with control flow ---

def apply_discount(status, order_total, item):
    print(f"\nStatus={status!r}, order_total={order_total}, item={item!r}")
    
    discount_items = ["gloves", "hat"]
    
    # Boolean expression from Part 1:
    # (status == "VIP") or (order_total >= 100 and item in discount_items)
    
    if status == "VIP" or (order_total >= 100 and item in discount_items):
        print("Discount applied.")
    else:
        print("No discount applied.")

apply_discount("VIP", 50, "gloves")          # VIP always qualifies
apply_discount("REGULAR", 120, "gloves")     # Non-VIP but high total and discount item
apply_discount("REGULAR", 80, "jacket")      # Does not qualify
```

This example shows how **Boolean expressions** and **control flow** work together.

---

## 2.6 trace_decision() Helper for Control Flow
### (Advanced Helper: Traces Full If / Elif / Else Structures)

This helper wraps a function that contains your decision logic.  
It prints a clean header and lets the inner function show which branches ran.

```python
# --- Advanced Helper: trace_decision(label, func) ---

def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.
    
    label: A short description of what is being traced
    func:  A zero-argument function containing the decision logic
    """
    print("\n" + "=" * 60)
    print("DECISION TRACE:", label)
    print("=" * 60)
    func()
```

### Example: Shipping logic with rush + order total

```python
# --- Example: using trace_decision with a multi-branch decision ---

def decide_shipping(order_total, rush):
    def inner():                                       # Wrap the trace_decision logic inside an inner function
        print(f"order_total={order_total}, rush={rush}")
        
        if rush:
            print("Branch: if → Express shipping")
        elif order_total >= 100:
            print("Branch: elif → Free standard shipping")
        else:
            print("Branch: else → Paid standard shipping")
    
    trace_decision("Shipping decision", inner)

# --- Run the helper-driven trace ---
decide_shipping(150, rush=True)
decide_shipping(100, rush=False)
decide_shipping(50, rush=False)
```

This helper is excellent for **complex decision trees**.
This pattern encourages you to think about:

- which branch runs  
- which conditions are tested  
- how order affects the outcome
- **Why do we use inner()?**
→ trace_decision() expects a zero-argument function. 
   inner() captures variables (order_total, rush) and runs the logic.
- **Why do we manually pass rush=True/False?**
→ The helper should NEVER define its own inputs.
   We pass the values we want to test, so we can see different branches.

## ✅ Key Takeaways – If, Elif, Else

- `if` runs its block when the condition is **True**.  
- `else` is a fallback block that runs when the `if` condition is **False`.  
- `elif` lets you create multi-branch decisions. Conditions are checked **top to bottom**.  
- Only **one** branch in an `if` / `elif` / `else` chain runs per execution.  
- Order matters. Put **more specific** or higher-priority conditions before more general ones.  
- Boolean logic from Part 1 (`and`, `or`, `not`, comparisons, membership) fits directly inside `if` conditions.  
- Helpers like `check_shipping`, `discount_for_status`, and `trace_decision` can make the flow of decisions easier to understand and debug.

You now have the core tools for **control flow** in Python.  
Next, you will build on this with **nested if statements** and more complex decision structures.


# 🐍 Chapter 5. Conditional Logic: 
## 🧩 Nested If Statements (Part 3)
 
### Expanding control flow with deeper, multi-layered logic  

Nested `if` statements allow you to evaluate **a second condition only if a first condition is already true**.  
They add an extra “layer” of logic and are useful when your business rules depend on two steps of decision-making.

---

# 1. Why Use Nested If Statements?

Nested logic is appropriate when:

- You need to **check Condition B only when Condition A is true**
- Rules follow a **hierarchy** or **decision tree**
- The second decision would be *nonsensical* unless the first is satisfied  
  (e.g., “If item is out of stock, then check whether you want to offer a display model.”)

### 🔍 Visual Model
A nested structure behaves like this:

```
if condition_A:
    if condition_B:
        # Nested True block
    else:
        # Nested False block
else:
    # Outer False block
```

This is like saying:
> “Only if A is true should I bother checking B.”

---

# 2. Simple Example

```python
customer = "Chris"
inventory = 0

if inventory <= 0:                     # Outer If condition
    if customer == "Chris":            # Nested If condition
        print("You can have the display model, sir.")
    else:
        print("Out of stock.")         # Nested Else code
else:
    print("Item available!")
```

### Why this is nested:
- It makes sense to check the **customer’s name** ONLY after confirming that inventory is zero.
- The business rule is **hierarchical**, not parallel.

---

# 3. The Same Logic *Without* Nesting

From page 3 of the PDF, Maven shows a “without nesting” version.  
Here’s how it looks conceptually:

```
if customer == "Chris" and inventory <= 0:
    ...
```

This works… **but loses hierarchy clarity**.

Nested version makes it obvious:
> “We only ask about Chris because inventory is zero.”

Flat version makes it look like the two conditions are equally important.

Which should you use?  
➡ Use **nesting** when your logic flows like:  
**First A → then maybe check B**

➡ Use **flat logic** when your rule is truly:  
**A AND B must both be true with equal weight**

---

# 4. Real-World Example: Return Authentication

Nested ifs shine in real workflows:

```python
is_logged_in = True
is_admin = False

if is_logged_in:
    if is_admin:
        print("Welcome, admin.")
    else:
        print("Welcome, user.")
else:
    print("Please log in.")
```

This mirrors a real application:
- First confirm login  
- THEN decide admin/user behavior

---

So we continue using the tools built in Section 2.8 — especially:

`````python
# --- Combining conditions with 'and' & 'or' ---

inventory = 5
price = 99.0

# Both clauses must be True
print("inventory > 0 and price < 100          →", inventory > 0 and price < 100)

# At least one clause must be True
print("inventory == 0 or price < 100          →", inventory == 0 or price < 100)


# --- Additional Boolean operator examples ---

print("5 > 4 and 7 > 8                        →", 5 > 4 and 7 > 8)      # The `and` operator requires ALL statements to be true
print("5 > 4 or 7 > 8                         →", 5 > 4 or 7 > 8)       # The `or` operator requires ONE statement to be true

print("15 > 10 and 11 > 10 and 12 > 10        →",
      15 > 10 and 11 > 10 and 12 > 10)                                  # Chain together any number of boolean expressions


# Explanation (for reference)
# If the first expression in an AND chain is False,
# Python will NOT evaluate the rest (short-circuit).
# If the first expression in an OR chain is True,
# Python will NOT evaluate the rest.
`````

`````python
# --- Parentheses change the order of evaluation ---

item = "skis"
price = 120

# Example where 'or' is evaluated with parentheses
expr_1 = (item == "skis") or (price <= 100)                 # The `or` operator requires ONE statement to be true
expr_2 = (item == "snowboard") and (price <= 100)           # The `and` operator requires ALL statements to be true

print("item == 'skis' or price <= 100 →", expr_1)
print("item == 'snowboard' and price <= 100 →", expr_2)

# --- Parentheses to control the order of evaluation ---

clothing = "hat"
sale_price = 99

(sale_price >= 100 and clothing == "gloves") or (clothing == "hat")  # The `or` operator requires ONE statement to be true
`````

`````python
# --- Explicit grouping using parentheses ---

clothing = "hat"
sale_price = 99

# Intended rule:
# "Discount applies if (price <= 100 AND item is gloves) OR item is a hat"

with_parentheses = (sale_price <= 100 and clothing == "gloves") or (clothing == "hat")

print("With parentheses →", with_parentheses)


# 🔍 Explanation  
# - `sale_price <= 100` → True  
# - `clothing == "gloves"` → False  
# - `(True AND False)` → False  
# - `clothing == "hat"` → True  
# - Final: `False OR True` → **True**  
# - **Discount applies**


# ❗ Without Parentheses (Python chooses based on precedence)

# --- Same logic, but WITHOUT parentheses ---

without_parentheses = sale_price <= 100 and clothing == "gloves" or clothing == "hat"

print("Without parentheses →", without_parentheses)

#🔍 What happens?
# Python interprets it as:

(sale_price <= 100 and clothing == "gloves") or (clothing == "hat")

# Python is not short-circuiting the string check you expected
# Python is effectively using its own implicit parentheses based on precedence


# 🧪 When Results **Change** (Critical Lesson)

# Let’s change one variable and observe the difference:

# --- A setup where parentheses DO change the result ---

clothing = "jacket"
sale_price = 99

with_parentheses = (sale_price <= 100 and clothing == "gloves") or (clothing == "hat")
without_parentheses = sale_price <= 100 and clothing == "gloves" or clothing == "hat"

print("\nCase where differences appear:")
print("With parentheses →", with_parentheses)
print("Without parentheses →", without_parentheses)

# In this case the final value happens to be the same — but the logic is **ambiguous and error-prone**, especially in real systems with longer rules.
`````

`````python
# ==================================================
# 🔬 Demonstration: How Precedence Affects Results
# ==================================================

# 🧠 Boolean Operator Precedence Demo Helper

def show_precedence(label, expression, expected=None):
    """
    Prints the label, the actual evaluated result,
    and optionally the expected logical interpretation.
    Use this to visualize how Python groups Boolean expressions.
    """
    print(f"{label:45} → {expression}")
    if expected is not None:
        print(f"   Expected (manual reasoning): {expected}")
    print()  # blank line for readability
    

# AND outranks OR
show_precedence(
    "True or False and False",
    True or False and False,
    "True or (False and False) → True"
)

show_precedence(
    "False or True and False",
    False or True and False,
    "False or (True and False) → False"
)

# NOT outranks everything
show_precedence(
    "not True and False",
    not True and False,
    "(not True) and False → False"
)

show_precedence(
    "not False or False",
    not False or False,
    "(not False) or False → True"
)

# Parentheses override default order
show_precedence(
    "not False or False and False",
    not False or False and False,
    "(not False) or (False and False) → True"
)

show_precedence(
    "(not False or False) and False",
    (not False or False) and False,
    "((not False) or False) and False → False"
)
`````

`````python
# =====================================================================
# --- Demonstration: Short-Circuit using clear printed side effects ---
# =====================================================================

def fast_check():
    print("fast_check() ran → quick evaluation")  # pretend fast and simple
    return False  # Note: Returns False

def slow_check():
    print("slow_check() RAN → expensive evaluation (should NOT run if skipped)")  # pretend slow and complex
    return True   # Note: Returns True


print("\nCase 1: AND short-circuit (first condition is False)")
result_1 = fast_check() and slow_check()  # return False AND return True comparison                   
print("Final Result:", result_1)  # AND short-circuit: if A is False, Python NEVER evaluates B

# ⛳ Interpretation
# - In the `and` example, `fast_check()` is False → `slow_check()` never runs  

def fast_true():
    print("fast_true() ran → quick positive result")
    return True

print("\nCase 2: OR short-circuit (first condition is True)")
result_2 = fast_true() or slow_check()
print("Final Result:", result_2)

# ⛳ Interpretation
# - In the `or` example, `fast_true()` is True → `slow_check()` never runs
`````

`````python
# ================================================
# Short-circuit example with realistic conditions
# ================================================

def item_in_stock(inventory):
    print("item_in_stock() ran")
    return inventory > 0

def is_discount_item(item):
    print("is_discount_item() ran")
    return item in ["gloves", "hat"]

inventory = 0
item = "gloves"

print("\nReal short-circuit with AND (inventory = 0):")
result = item_in_stock(inventory) and is_discount_item(item)
print("Final:", result)

# ================================================
# Short-circuit example with realistic conditions
# ================================================

inventory = 1
item = "gloves"

def item_in_stock(inventory):
    print("item_in_stock() ran")
    return inventory > 0

def is_discount_item(item):
    print("is_discount_item() ran")
    return item in ["gloves", "hat"]

print("\nReal short-circuit with AND (inventory = 0):")
result = item_in_stock(inventory) and is_discount_item(item)
print("Final:", result)

# is_discount_item() ran — not short-circuited because inventory was > 0.
`````

`````python
# ============================================
# 🔍 trace_logic – Boolean Logic Tracing Helper
# ============================================

def trace_logic(label, func):
    """
    Run a logical expression (wrapped in a zero-argument function),
    print a clear header, and show the final Boolean result.

    Usage:
        trace_logic(
            "Description of the logic",
            lambda: some_check() and other_check() or third_check()
        )

    To see short-circuiting, each inner check should print when it runs.
    """
    print("\n" + "=" * 60)
    print("TRACE:", label)
    print("=" * 60)

    result = func()  # Evaluate the expression once
    print("Final result:", result)
    return result


# 🧵 Examples: Using `trace_logic()` with Real Checks
# --- Example condition functions with side effects ---
# Scenario 1: inventory = 0 (AND short-circuits, OR still needed)

inventory = 0
item = "gloves"
status = "VIP"

def item_in_stock(inventory):
    print(f"item_in_stock({inventory}) ran")
    return inventory > 0

def is_discount_item(item):
    print(f"is_discount_item({item!r}) ran")
    return item in ["gloves", "hat"]

def is_vip_customer(status):
    print(f"is_vip_customer({status!r}) ran")
    return status == "VIP"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)

# What you’ll see
# - In **Scenario 1** (`inventory = 0`):  
#   - `item_in_stock()` runs, returns False  
#   - `is_discount_item()` is **skipped** (AND short-circuited)  
#   - `is_vip_customer()` runs (needed for OR)


# --- Example condition functions with side effects ---
# Scenario 2: inventory > 0 and discount item (OR short-circuits)

inventory = 5
item = "gloves"
status = "regular"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)

# What you’ll see
# - In **Scenario 2** (`inventory = 5`, `item="gloves"`):  
#   - `item_in_stock()` runs, returns True  
#   - `is_discount_item()` runs, returns True  
#   - `(True and True)` becomes True  
#   - `is_vip_customer()` is **skipped** (OR short-circuited)

# `trace_logic()` gives you a clean header and final result;  
# your inner functions show **which steps actually executed**.
`````

`````python
# ============================================================
# Exercise Using `trace_decision()` and `trace_condition()`
# ============================================================

# ============================================================
# Helper Prerequisite: trace_decision(label, func)
# ============================================================

def trace_condition(name, value):
    print(f"{name:30} → {value}")


inventory = 3

if inventory <= 0:
    print("OUT OF STOCK")
elif inventory <= 5:
    print("LOW STOCK")
else:
    print("IN STOCK")


# ============================================================
# Advanced Helper 1: trace_decision(label, func)
# ============================================================

def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.

    label: A short description of what is being traced
    func:  A zero-argument function containing the decision logic
    """
    print("\n" + "=" * 60)
    print("DECISION TRACE:", label)
    print("=" * 60)
    func()


def classify_inventory(inventory):
    def inner():
        print(f"inventory = {inventory}")
        
        cond_out    = inventory <= 0
        cond_low    = 0 < inventory <= 5
        
        trace_condition("inventory <= 0",     cond_out)
        trace_condition("0 < inventory <= 5", cond_low)
        
        if cond_out:
            print("OUT OF STOCK")
        elif cond_low:
            print("LOW STOCK")
        else:
            print("IN STOCK")
            
    trace_decision("Inventory classification", inner)


# Test it
for inv in [0, 2, 6]:
    classify_inventory(inv)
`````

`````python
# ============================================================
# Exercise Using `trace_decision()` and `trace_condition()`
# ============================================================

# --- Price thresholds, descending order ---
price = 1499.99
luxury_threshold = 1000.00
expert_threshold = 500.00
intermediate_threshold = 300.00

test_prices = [1499.99, 800.00, 450.00, 250.00]

# ============================================================
# Advanced Helper 1: trace_decision(label, func)
# ============================================================

def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.

    label: A short description of what is being traced
    func:  A zero-argument function containing the decision logic
    """
    print("\n" + "=" * 60)
    print("DECISION TRACE:", label)
    print("=" * 60)
    func()

# ============================================================
# Wrap the trace_decision logic inside an inner function
# ============================================================

def classify_snowboard(price):
    def inner():                         # Wrap the trace_decision logic inside an inner function
        print(f"price = {price}")

        if price > intermediate_threshold:
            print("Branch: if → Good for intermediate users.")
        elif price > expert_threshold:
            print("Branch: elif → For experienced users.")
        elif price > luxury_threshold:
            print("Branch: elif → Luxury board (gold doesn't improve the ride!)")
        else:
            print("Branch: else → Suitable for a beginner.")

    trace_decision("Snowboard Skill Classification", inner)

# ============================================================
# --- Run the helper-driven trace ---
# ============================================================

classify_snowboard(price)


# ============================================================
# Helper 2: trace_condition(name, value)
# ============================================================

def trace_condition(name, value):
    """
    Print the name of a condition and its Boolean result.
    Useful for debugging and understanding complex rule logic.
    """
    print(f"{name:30} → {value}")


# =================================================
# 1. Buggy version: ordering is incorrect
# ❌ (demonstrates that elif blocks never run) ❌
# =================================================

def classify_snowboard_buggy(price):
    """
    Buggy snowboard classification example.

    The intent:
      - If price > luxury_threshold         → Luxury board
      - Elif price > expert_threshold       → For experienced users
      - Elif price > intermediate_threshold → For intermediate users
      - Else                                → Beginner board

    But the actual code (below) checks intermediate first, then expert, then luxury.
    Because intermediate_threshold is the lowest, the first condition catches
    almost everything, and the elif branches never run.
    """
    def inner():
        print(f"\n[BUGGY] price = {price}")
        
        # Show how the conditions evaluate
        cond_intermediate = (price > intermediate_threshold)
        cond_expert      = (price > expert_threshold)
        cond_luxury      = (price > luxury_threshold)

        trace_condition("price > intermediate_threshold", cond_intermediate)
        trace_condition("price > expert_threshold",      cond_expert)
        trace_condition("price > luxury_threshold",      cond_luxury)

        # Buggy ordering: intermediate first, then expert, then luxury
        if price > intermediate_threshold:
            print("Branch: if    → Good for intermediate users.")
        elif price > expert_threshold:
            print("Branch: elif1 → For experienced users.")
        elif price > luxury_threshold:
            print("Branch: elif2 → Luxury board (gold looks great).")
        else:
            print("Branch: else  → Suitable for a beginner.")

    trace_decision("Buggy snowboard classification", inner)


# Run buggy classification for several prices
for p in test_prices:
    classify_snowboard_buggy(p)


# ==========================================
# 2. Fixed version: ordering is corrected
#    (highest thresholds checked first)
# ==========================================

def classify_snowboard_fixed(price):
    """
    Correct snowboard classification example.

    Correct logic:
      - If price > luxury_threshold      → Luxury board
      - Elif price > expert_threshold    → For experienced users
      - Elif price > intermediate_threshold → For intermediate users
      - Else                             → Beginner board

    Conditions are checked from highest threshold to lowest.
    """
    def inner():
        print(f"\n[FIXED] price = {price}")
        
        # Boolean evaluation preview
        cond_intermediate = (price > intermediate_threshold)
        cond_expert      = (price > expert_threshold)
        cond_luxury      = (price > luxury_threshold)

        trace_condition("price > luxury_threshold",      cond_luxury)
        trace_condition("price > expert_threshold",      cond_expert)
        trace_condition("price > intermediate_threshold", cond_intermediate)

        # Correct threshold ordering: luxury first, then expert, then intermediate
        if cond_luxury:
            print("Branch: if    → Luxury board.")
        elif cond_expert:
            print("Branch: elif1 → For experienced users.")
        elif cond_intermediate:
            print("Branch: elif2 → Good for intermediate users.")
        else:
            print("Branch: else  → Suitable for a beginner.")

    trace_decision("Correct snowboard classification", inner)


# Run fixed classification for the same prices
for p in test_prices:
    classify_snowboard_fixed(p)
`````

`````python
# ------------------------------------------------------------
# Lesson: Classifying Values (Even vs Odd)
# ------------------------------------------------------------
# This lesson demonstrates:
# - Iteration over a list
# - Conditional classification
# - Clean output formatting
# - Avoiding unnecessary index usage
# ------------------------------------------------------------


# ------------------------------------------------------------
# Input Data
# ------------------------------------------------------------

my_list = [74, -55, -77, 58, 0, 91, 62, 0, -91, 61]


# ------------------------------------------------------------
# 1) Baseline Version — Direct Iteration (Preferred)
# ------------------------------------------------------------

for value in my_list:
    
    if value % 2 == 0:
        label = "even"
    else:
        label = "odd"
    
    print(f"{value} is {label}.")

print()


# ------------------------------------------------------------
# 🟦 Helper: Classification Trace
# ------------------------------------------------------------

def parity_trace(iterable):
    """
    Diagnostic helper: shows how each value is classified as even or odd.
    """
    print("PARITY TRACE\n-------------")
    
    for idx, value in enumerate(iterable):
        parity = "even" if value % 2 == 0 else "odd"
        print(f"Index {idx}: value = {value!r} → classified as {parity!r}")


# Example usage:

parity_trace(my_list)
print()


# ------------------------------------------------------------
# 2) Mechanistic Version — Explicit Index Authority
# (Matches the professor’s structure, but cleaned up)
# ------------------------------------------------------------

for i in range(len(my_list)):
    
    value = my_list[i]
    
    if value % 2 == 0:
        label = "even"
    else:
        label = "odd"
    
    print(f"{value} is {label}.")

print()


# ------------------------------------------------------------
# 🧠 Key Takeaways
# ------------------------------------------------------------
# - Direct iteration is clearer than index-based loops when indices are not needed
# - Avoid str() and string concatenation inside loops
# - Use f-strings for clean formatting
# - Separate classification logic from output formatting
# - Index-based loops should only be used when indices are required
`````

`````python
# ------------------------------------------------------------
# Lesson: Classifying Values (Positive, Negative, Zero)
# ------------------------------------------------------------
# This lesson demonstrates:
# - Iteration over a list
# - Conditional classification into categories
# - Clean output formatting
# - Avoiding unnecessary index usage
# ------------------------------------------------------------


# ------------------------------------------------------------
# Input Data
# ------------------------------------------------------------

my_list = [74, -55, -77, 58, 0, 91, 62, 0, -91, 61]


# ------------------------------------------------------------
# 1) Baseline Version — Direct Iteration (Preferred)
# ------------------------------------------------------------

for value in my_list:
    
    if value < 0:
        label = "negative"
    elif value > 0:
        label = "positive"
    else:
        label = "neither positive nor negative"
    
    print(f"{value} is {label}.")

print()


# ------------------------------------------------------------
# 🟦 Helper: Classification Trace
# ------------------------------------------------------------

def sign_trace(iterable):
    """
    Diagnostic helper: shows how each value is classified
    as negative, positive, or zero.
    """
    print("SIGN TRACE\n-----------")
    
    for idx, value in enumerate(iterable):
        if value < 0:
            label = "negative"
        elif value > 0:
            label = "positive"
        else:
            label = "neither positive nor negative"
        
        print(f"Index {idx}: value = {value!r} → classified as {label!r}")


# Example usage:

sign_trace(my_list)
print()


# ------------------------------------------------------------
# 2) Mechanistic Version — Explicit Index Authority
# (Matches the professor’s structure, but cleaned up)
# ------------------------------------------------------------

for i in range(len(my_list)):
    
    value = my_list[i]
    
    if value < 0:
        label = "negative"
    elif value > 0:
        label = "positive"
    else:
        label = "neither positive nor negative"
    
    print(f"{value} is {label}.")

print()


# ------------------------------------------------------------
# 🧠 Key Takeaways
# ------------------------------------------------------------
# - Direct iteration is clearer than index-based loops when indices are not needed
# - Avoid str() and string concatenation inside loops
# - Use f-strings for clean formatting
# - Separate classification logic from output formatting
# - Conditional logic assigns values into mutually exclusive categories
# - Index-based loops should only be used when indices are required
`````

`````python
# ------------------------------------------------------------
# Threshold-based classification and binning
# Mapping numeric values into discrete categories
# Lesson: Parallel Lists + Threshold Binning (Credit Scores)
# ------------------------------------------------------------
# Assumption (explicit):
# - `names` and `credit` are perfectly aligned (same length).
#
# This lesson demonstrates:
# - Parallel iteration across aligned lists
# - Classifying numeric values using threshold rules (binning)
# - Building a derived list (`credit_group`) from an input list (`credit`)
#
# Key improvement vs. the lecture:
# - Use zip() to express pairing semantics
# - Separate "classification" from "printing"
# - Keep thresholds consistent and readable
# ------------------------------------------------------------


# ------------------------------------------------------------
# Input Data (Perfectly Aligned)
# ------------------------------------------------------------

names = ['Anna', 'Beth', 'Chad', 'Drew', 'Elsa', 'Fred']
credit = [683, 580, 752, 607, 615, 703]


# ------------------------------------------------------------
# 1) Credit Score → Label (Readable Classification Function)
# ------------------------------------------------------------
# We keep this logic in one place so we can reuse it for:
# - printing
# - grouping
# - testing
# - later data work (pandas transforms)

def credit_label(score):
    """
    Returns a credit category label based on score thresholds.
    
    Thresholds used (lecture-style):
    <= 599 → bad
    <= 649 → poor
    <= 699 → fair
    <= 749 → good
    >= 750 → excellent
    """
    if score <= 599:
        return "bad"
    elif score <= 649:
        return "poor"
    elif score <= 699:
        return "fair"
    elif score <= 749:
        return "good"
    else:
        return "excellent"


# ------------------------------------------------------------
# 2) Baseline Version — Safe Parallel Iteration with zip()
# ------------------------------------------------------------
# Note: zip() stops at the shortest list if lengths differ.
# Here we assume perfect alignment.

if len(names) != len(credit):
    raise ValueError("Alignment error: names and credit must have the same length.")

for name, score in zip(names, credit):
    label = credit_label(score)
    print(f"{name} has a {label} credit score.")

print()


# ------------------------------------------------------------
# 3) Credit Score → Group Code (Derived Feature / Binning)
# ------------------------------------------------------------
# Group mapping used (matches lecture intent):
#   750+ → 4
#   700–749 → 3
#   650–699 → 2
#   600–649 → 1
#   < 600 → 0

def credit_group_code(score):
    """Returns an integer bin code based on score thresholds."""
    if score >= 750:
        return 4
    elif score >= 700:
        return 3
    elif score >= 650:
        return 2
    elif score >= 600:
        return 1
    else:
        return 0


# ------------------------------------------------------------
# 4) Build credit_group (Two Teaching Versions)
# ------------------------------------------------------------

# 4a) Explicit accumulator version (mechanistic, teaching-friendly)

credit_group = []

for score in credit:
    credit_group.append(credit_group_code(score))

print("credit:       ", credit)
print("credit_group: ", credit_group)
print()


# 4b) List comprehension version (clean, correct, and intent-revealing)

credit_group_comp = [credit_group_code(score) for score in credit]

print("credit_group (comprehension):", credit_group_comp)
print()


# ------------------------------------------------------------
# 🟦 Helper: Parallel Trace (Optional Diagnostic)
# ------------------------------------------------------------

def credit_trace(names, scores):
    """
    Diagnostic helper: prints aligned (name, score, label, group) records.
    """
    if len(names) != len(scores):
        raise ValueError("Alignment error: names and scores must have the same length.")
    
    print("CREDIT TRACE\n------------")
    for idx, (name, score) in enumerate(zip(names, scores)):
        label = credit_label(score)
        group = credit_group_code(score)
        print(f"Index {idx}: {name!r}, score={score}, label={label!r}, group={group}")


# Example usage:

credit_trace(names, credit)
print()


# ------------------------------------------------------------
# 🧠 Key Takeaways
# ------------------------------------------------------------
# - Threshold logic is a common analytics pattern: classify numeric values into bins
# - Keep classification logic in small functions to avoid duplication
# - zip() expresses pairing intent better than index-based loops
# - Derived lists (like `credit_group`) are "features" computed from raw data
# - When building a new list from an existing list, list comprehensions can be ideal
`````

`````python
# ------------------------------------------------------------
# Lesson: Compound Conditions + Rule Simplification (Exam Results) ⭐
# ------------------------------------------------------------
# Assumption (explicit):
# - `names`, `grades1`, and `grades2` are perfectly aligned (same length).
#
# This lesson demonstrates:
# - Compound boolean logic (and / or)
# - Mutually exclusive classification
# - Threshold logic
# - Rule simplification (removing redundant conditions)
# - Parallel iteration across aligned lists
#
# Conceptual payload: 🧱
# - Classifying records using multiple features (two exam scores) 
# ------------------------------------------------------------


# ------------------------------------------------------------
# Input Data (Perfectly Aligned)
# ------------------------------------------------------------

names = ['Anna', 'Beth', 'Chad', 'Drew', 'Elsa', 'Fred']
grades1 = [56, 92, 87, 43, 75, 62]
grades2 = [81, 95, 72, 21, 58, 64]


# ------------------------------------------------------------
# 1) Pass/Fail Classification Function
# ------------------------------------------------------------
# Centralize business logic so it can be reused, tested, and refactored.

def exam_result_label(score1, score2, threshold=60):
    """
    Returns a classification label based on two exam scores.
    
    Rules:
    - Passed both exams
    - Passed exam 1 only
    - Passed exam 2 only
    - Failed both exams
    """
    passed1 = score1 >= threshold
    passed2 = score2 >= threshold
    
    if passed1 and passed2:
        return "passed both exams"
    elif passed1:
        return "passed exam 1, but not exam 2"
    elif passed2:
        return "passed exam 2, but not exam 1"
    else:
        return "failed both exams"


# ------------------------------------------------------------
# 2) Baseline Version — Safe Parallel Iteration with zip()
# ------------------------------------------------------------
# Note: zip() stops at the shortest list if lengths differ.
# Here we assume perfect alignment.

if not (len(names) == len(grades1) == len(grades2)):
    raise ValueError("Alignment error: all lists must have the same length.")

for name, g1, g2 in zip(names, grades1, grades2):
    label = exam_result_label(g1, g2)
    print(f"{name} {label}.")

print()


# ------------------------------------------------------------
# 🟦 Helper: Parallel Classification Trace
# ------------------------------------------------------------

def exam_trace(names, grades1, grades2, threshold=60):
    """
    Diagnostic helper: shows aligned records and classification results.
    """
    if not (len(names) == len(grades1) == len(grades2)):
        raise ValueError("Alignment error: all lists must have the same length.")
    
    print("EXAM TRACE\n----------")
    for idx, (name, g1, g2) in enumerate(zip(names, grades1, grades2)):
        label = exam_result_label(g1, g2, threshold)
        print(f"Index {idx}: {name!r}, grades=({g1}, {g2}) → {label!r}")


# Example usage:

exam_trace(names, grades1, grades2)
print()


# ------------------------------------------------------------
# 3) Mechanistic Version — Explicit Index Authority
# (Matches the professor’s structure, but cleaned up)
# ------------------------------------------------------------

for i in range(len(names)):
    
    name = names[i]
    g1 = grades1[i]
    g2 = grades2[i]
    
#     label = exam_result_label(g1, g2)  # See alternative
    
#     print(f"{name} {label}.")

# print()

# Alternative without label = exam_result_label() function
    if g1 >= 60 and g2 >= 60:
        print(f"{name} passed both exams.")
    elif g1 >= 60:
        print(f"{name} passed exam 1, but not exam 2.")
    elif g2 >= 60:
        print(f"{name} passed exam 2, but not exam 1.")
    else:
        print(f"{name} failed both exams.")
    
# ------------------------------------------------------------
# 🧠 Key Takeaways
# ------------------------------------------------------------
# - Compound conditions combine multiple boolean tests using and/or
# - Mutually exclusive rules should be ordered from most specific to most general
# - Redundant conditions can often be removed safely
# - Centralizing business logic avoids duplication and inconsistency
# - zip() expresses parallel pairing semantics better than index-based loops
# - This pattern generalizes to multi-feature classification in analytics
`````

`````python
# ==============================================================================
# Helper Version with `trace_decision()`, `trace_condition()`, `trace_branch()`
# ==============================================================================

# --- Helper Toolkit (make sure these are defined first) ---

def trace_decision(label, func):
    print("\n" + "=" * 60)
    print(f"DECISION TRACE: {label}")
    print("=" * 60)
    func()

def trace_condition(name, value):
    print(f"{name:35} → {value}")

def trace_branch(label):
    print(f"Branch: {label}")


# --- Nested purchase logic with floats + helpers ---

def check_purchase(item_price, budget, inventory):
    
    def inner():
        print(f"item_price={item_price}, budget={budget}, inventory={inventory}")

        # These are fine with floats — just comparisons
        cond_can_afford = budget >= item_price
        cond_in_stock   = inventory > 0

        trace_condition("budget >= item_price", cond_can_afford)
        trace_condition("inventory > 0",        cond_in_stock)

        if cond_can_afford:
            trace_branch("outer if → budget >= item_price")
            
            if cond_in_stock:
                trace_branch("nested if → inventory > 0")
                print("You can afford this, and it is in stock!")
            else:
                trace_branch("nested else → out of stock")
                print("You can afford this, but it is out of stock.")
        
        else:
            trace_branch("outer else → too expensive")
            print(f"This item costs ${item_price:.2f}, which is above your budget of ${budget:.2f}.")

    trace_decision("Nested Purchase Logic", inner)


# --- Try a few scenarios ---

check_purchase(599.99, 500.00, 0)   # too expensive
check_purchase(499.99, 500.00, 0)   # can afford, but out of stock
check_purchase(499.99, 500.00, 3)   # can afford, in stock
`````

## 1. The Boolean Data Type (`True` / `False`)

A **Boolean** is a data type with only **two possible values**:

- `True`
- `False`

In arithmetic:

- `True` behaves like `1`
- `False` behaves like `0`
- The `not` keyword **inverts** a Boolean

## 2. Comparison Operators

**Comparison operators** evaluate a relationship and return a **Boolean** (`True` / `False`):

| Operator | Meaning                    | Example        |
|---------:|----------------------------|----------------|
| `==`     | equal to                   | `price == 99`  |
| `!=`     | not equal to               | `item != 'ski'`|
| `<`      | less than                  | `inv < 10`     |
| `>`      | greater than               | `temp > 32`    |
| `<=`     | less than or equal to      | `age <= 18`    |
| `>=`     | greater than or equal to   | `score >= 80`  |

## 🔗 3. Chained Comparisons (Readable Multi-Condition Checks)

Python allows you to “chain” comparisons together, which makes range checks concise and readable.  
Chained comparisons evaluate **left to right**, and Python ensures all comparisons are True.

This is extremely common in pricing logic, validation rules, SLAs, and data quality checks.


This form reads almost like natural language, which is why chained comparisons are widely used in analytics logic, data validation, and business rule engines.


### 🔗 Chained Comparisons Helper Function (Enhanced Clarity)

To make chained comparisons even easier to understand, here is a small helper function that:

- prints the **label** of what you’re testing  
- prints the **expression’s result** (`True` / `False`)  
- keeps your output clean and well-organized  

Use this anywhere in your notebook when experimenting with conditions.

```python
    """
```

## 4. Membership Tests: `in` and `not in`

Membership tests check if a value exists **inside** an iterable (string, list, etc.).

- `value in container` → `True` if found  
- `value not in container` → `True` if **not** found  

They are extremely useful for text searches, product lists, feature flags, and validating user inputs.

---

### 💡What You Learn Here

- `in` checks presence  
- `not in` checks absence  
- Works with strings, lists, tuples, dict keys, etc.  
- Fast, readable, and essential for filtering, validation, and search logic  

This version keeps your notebook fully consistent with your other sections — every logic-related output now uses a clean helper for beautifully aligned results.

## 5. Boolean Operators: `and`, `or`, `not` (+ Short-Circuiting & Precedence)

Boolean operators allow you to combine multiple logical expressions into a single condition:

- **`A and B`** → True only if **both** A and B are True  
- **`A or B`** → True if **at least one** of A or B is True  
- **`not A`** → inverts A (True ↔ False)

---

### 🔄 Short-Circuit Behavior (Efficiency Tip)

Python **stops evaluating early** when the final outcome is already determined:

- For **`A and B`**:  
  - If `A` is **False**, Python **does not evaluate B**
- For **`A or B`**:  
  - If `A` is **True**, Python **does not evaluate B**

This is why we typically place **simple or fast checks first**, and **more expensive operations last**.

---

### 🧠 Operator Precedence (Highest → Lowest)

Python always evaluates Boolean operators in this fixed order:

1. **`not`**  
2. **`and`**  
3. **`or`**

**This means:**
- `not` is always evaluated **before** `and`  
- `and` is always evaluated **before** `or`  
- This happens **even if the operators appear in a different order in the code**

Example mental model:

```
A or B and C
→ A or (B and C)
```

Python will always group `B and C` first because `and` outranks `or`.

---

### 🟦 Parentheses Always Win

Parentheses let you **override** Python’s default precedence rules and intentionally control how the logic is grouped.

Use parentheses to:
- Make your logic explicit  
- Avoid ambiguity in complex conditions  
- Ensure correctness when mixing `and` and `or`  
- Improve readability for yourself and others

Whenever your logic combines different operators, parentheses are a best practice for clarity and safety.

## 🧩 Boolean Logic Truth Tables  
### A Visual Guide to `and`, `or`, and `not`

Truth tables show **every possible combination** of Boolean values and how the operator behaves.  
They are essential for understanding conditional logic.

---

## ✔ Truth Table for `and`
`A and B` is True **only when both** A and B are True.

```
 A      B      A and B
-------------------------
 True   True      True
 True   False     False
 False  True      False
 False  False     False
```

---

## ✔ Truth Table for `or`
`A or B` is True when **at least one** is True.

```
 A      B      A or B
------------------------
 True   True      True
 True   False     True
 False  True      True
 False  False     False
```

---

## ✔ Truth Table for `not`
`not A` flips the value.

```
  A      not A
----------------
 True     False
 False    True
```

## 🧠 Combined Example  
A preview of how precedence (`not` > `and` > `or`) affects evaluation:

```
 A      B      C      A or B and C     A or (B and C)
-------------------------------------------------------
 True   True   True        True              True
 True   True   False       True              True
 True   False  True        True              True
 True   False  False       True              True

 False  True   True        True              True
 False  True   False       False             False
 False  False  True        True              True
 False  False  False       False             False
```

## ⚡ Short-Circuit Evaluation (Efficiency in Action)

Python **stops evaluating early** when a final result is already determined.

- For **`A and B`**:  
  If `A` is False → *Python never evaluates B*
- For **`A or B`**:  
  If `A` is True → *Python never evaluates B*

Short-circuiting is important for performance and safety.

Here is a clear demonstration using “fast” and “slow” checks.

```python
# --- Short-Circuit Demo using clear printed side effects ---

def fast_check():
    print("fast_check() ran → quick evaluation")
    return False  # pretend fast and simple

def slow_check():
    print("slow_check() RAN → expensive evaluation (should NOT run if skipped)")
    return True   # pretend slow and complex


print("\nCase 1: AND short-circuit (first condition is False)")
result_1 = fast_check() and slow_check()
print("Final Result:", result_1)


def fast_true():
    print("fast_true() ran → quick positive result")
    return True

print("\nCase 2: OR short-circuit (first condition is True)")
result_2 = fast_true() or slow_check()
print("Final Result:", result_2)
```

### ⛳ Interpretation
- In the `and` example, `fast_check()` is False → `slow_check()` never runs  
- In the `or` example, `fast_true()` is True → `slow_check()` never runs  

This mirrors real code patterns where we put:
- Cheap validation checks **first**  
- Expensive computations **second**  


## 🧪 `trace_logic()` – Visualize How Your Conditions Run

`trace_logic()` is a small helper that makes it easier to **see** how a complex Boolean expression is evaluated.

You’ll use it together with functions that print when they run (like `item_in_stock()` and `is_discount_item()`).  
This lets you watch:

- which checks run
- which checks are skipped (short-circuited)
- what the final result is

---

```python
# ============================================
# 🔍 trace_logic – Boolean Logic Tracing Helper
# ============================================

def trace_logic(label, func):
    """
    Run a logical expression (wrapped in a zero-argument function),
    print a clear header, and show the final Boolean result.

    Usage:
        trace_logic(
            "Description of the logic",
            lambda: some_check() and other_check() or third_check()
        )

    To see short-circuiting, each inner check should print when it runs.
    """
    print("\n" + "=" * 60)
    print("TRACE:", label)
    print("=" * 60)

    result = func()  # Evaluate the expression once
    print("Final result:", result)
    return result
```

---

## 🧵 Example: Using `trace_logic()` with Real Checks

```python
# --- Example condition functions with side effects ---

def item_in_stock(inventory):
    print(f"item_in_stock({inventory}) ran")
    return inventory > 0

def is_discount_item(item):
    print(f"is_discount_item({item!r}) ran")
    return item in ["gloves", "hat"]

def is_vip_customer(status):
    print(f"is_vip_customer({status!r}) ran")
    return status == "VIP"


# Scenario 1: inventory = 0 (AND short-circuits, OR still needed)
inventory = 0
item = "gloves"
status = "VIP"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)


# Scenario 2: inventory > 0 and discount item (OR short-circuits)
inventory = 5
item = "gloves"
status = "regular"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)
```

### What you’ll see

- In **Scenario 1** (`inventory = 0`):  
  - `item_in_stock()` runs, returns False  
  - `is_discount_item()` is **skipped** (AND short-circuited)  
  - `is_vip_customer()` runs (needed for OR)

- In **Scenario 2** (`inventory = 5`, `item="gloves"`):  
  - `item_in_stock()` runs, returns True  
  - `is_discount_item()` runs, returns True  
  - `(True and True)` becomes True  
  - `is_vip_customer()` is **skipped** (OR short-circuited)

`trace_logic()` gives you a clean header and final result;  
your inner functions show **which steps actually executed**.

---

## ✅ Key Takeaways – `trace_logic()`

- `trace_logic(label, func)` runs a **single Boolean expression** and shows the final result with a clear trace header.
- Wrap your expression in a `lambda` so it **is not evaluated** until `trace_logic()` calls it.
- Put `print()` statements inside your condition functions (like `item_in_stock`) to see:
  - which checks run  
  - where short-circuiting skips later checks  
- Use `trace_logic()` to:
  - experiment with **AND / OR / NOT**
  - confirm **precedence and parentheses**
  - build deep intuition about **how Python really evaluates conditions**

This tool is perfect for “x-ray vision” into your Boolean logic while you’re learning and debugging.


## 2.8 Advanced Decision-Flow Utilities  
These helper tools make it easier to **see** and **understand** which branches execute when using `if`, `elif`, and `else`.  
They are optional, but they give valuable visibility into your control flow logic.

We will start with a simple demonstration helper, then introduce three reusable helpers that work for any decision structure.

---

# 🔹 1. Helper Demo (Simple, In-Function Branch Tracing)

This first example prints the branch **inside the function itself**.  
It is great for beginners and quick tests.

```python
# --- Demonstration Helper: show which branch executes ---

def check_shipping(order_total):
    """
    Simple demonstration function:
    Prints which shipping branch runs based on order_total.
    """
    print(f"\nChecking shipping for order_total = {order_total}")
    
    if order_total >= 100:
        print("Branch: if → Free shipping")
    else:
        print("Branch: else → Standard shipping")

# Try a few values
check_shipping(120)
check_shipping(75)
check_shipping(100)
```

This is a good first step, but the next helpers are more flexible and powerful.

---

# 🔹 2. trace_decision()  
### (Advanced Helper: Traces Full If / Elif / Else Structures)

This helper wraps a function that contains your decision logic.  
It prints a clean header and lets the inner function show which branches ran.

```python
# --- Advanced Helper: trace_decision(label, func) ---

def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.
    
    label: A short description of what is being traced
    func:  A zero-argument function containing the decision logic
    """
    print("\n" + "=" * 60)
    print("DECISION TRACE:", label)
    print("=" * 60)
    func()
```

### Example: Shipping logic with rush + order total

```python
# --- Example: using trace_decision with a multi-branch decision ---

def decide_shipping(order_total, rush):
    def inner():                                       # Wrap the trace_decision logic inside an inner function
        print(f"order_total={order_total}, rush={rush}")
        # if / elif / else chain here...
        if rush:
            print("Branch: if → Express shipping")
        elif order_total >= 100:
            print("Branch: elif → Free standard shipping")
        else:
            print("Branch: else → Paid standard shipping")
    
    trace_decision("Shipping decision", inner)        # label, inner function

# --- Run the helper-driven trace ---
decide_shipping(150, rush=True)
decide_shipping(100, rush=False)
decide_shipping(50, rush=False)
```

This helper is excellent for **complex decision trees**.
This pattern encourages you to think about:

- which branch runs  
- which conditions are tested  
- how order affects the outcome
- **Why do we use inner()?**
→ trace_decision() expects a zero-argument function. 
   inner() captures variables (order_total, rush) and runs the logic.
- **Why do we manually pass rush=True/False?**
→ The helper should NEVER define its own inputs.
   We pass the values we want to test, so we can see different branches.
  
---

# 🔹 3. trace_branch()

```python
# --- Helper: trace_branch(label) ---

def trace_branch(label):

    Helpful when testing individual if/elif/else chains.

```

```python
grade_letter(95)
grade_letter(82)
grade_letter(67)
grade_letter(58)
```

# 🔹 4. trace_condition()  
### (Shows How Each Boolean Condition Evaluates)

This helper prints the **name** of a condition and its Boolean result.  
Very helpful for debugging real-world rules.

```python
# --- Helper: trace_condition(name, value) ---

def trace_condition(name, value):
    """
    Print the name of a condition and its Boolean result.
    Useful for debugging and understanding complex rule logic.
    """
    print(f"{name:30} → {value}")
```

### Example: Debugging a discount rule

```python
# --- Example: inspecting a multi-condition discount rule ---

def debug_discount_rule(status, order_total, item):
    print(f"\nStatus={status!r}, order_total={order_total}, item={item!r}")
    
    discount_items = ["gloves", "hat"]
    
    cond_vip = (status == "VIP")
    cond_high_total = (order_total >= 100)
    cond_discount_item = (item in discount_items)
    
    # Show each condition's result
    trace_condition("status == 'VIP'", cond_vip)
    trace_condition("order_total >= 100", cond_high_total)
    trace_condition("item in discount_items", cond_discount_item)
    
    # Combine the rule
    eligible = cond_vip or (cond_high_total and cond_discount_item)
    trace_condition("Final rule result", eligible)
    
    if eligible:
        print("Decision: Discount applied.")
    else:
        print("Decision: No discount.")

debug_discount_rule("VIP", 50, "gloves")
debug_discount_rule("REGULAR", 120, "gloves")
debug_discount_rule("REGULAR", 80, "jacket")
```

# 🔹 5. trace_if()  
### (Compact Helper for Single If/Else Decisions)

This helper describes the decision, shows the condition result,  
and names the branch selected.

```python
# --- Helper: trace_if(description, condition, if_true, if_false) ---

def trace_if(description, condition, if_true, if_false):
    """
    Describe an if/else style decision in a compact, readable format.
    
    description: description of the decision being checked
    condition:   Boolean expression
    if_true:     message to print if the condition is True
    if_false:    message to print if the condition is False
    """
    print(f"\nChecking: {description}")
    print(f"Condition result: {condition}")
    
    if condition:
        print(f"Branch selected: {if_true}")
    else:
        print(f"Branch selected: {if_false}")
```

### Example: Using trace_if for two related decisions

```python
# --- Example: shipping decisions with trace_if ---

def shipping_decision(order_total, rush):
    print(f"\norder_total={order_total}, rush={rush}")
    
    trace_if(
        description="Rush shipping requested",
        condition=rush,
        if_true="Express shipping",
        if_false="No rush shipping"
    )
    
    trace_if(
        description="Eligible for free standard shipping",
        condition=(order_total >= 100),
        if_true="Free standard shipping",
        if_false="Standard shipping (paid)"
    )

shipping_decision(150, rush=False)
shipping_decision(40, rush=True)
shipping_decision(40, rush=False)
```

        print(f"Branch selected: {if_false}")


### --- Confirmation Message ----------------------------------------------------------------
print("🟢 Helper Toolkit Loaded: trace_decision, trace_condition, trace_branch, trace_if")



# ✅ Key Takeaways — Chapter 5: Conditional Logic

### **1. All conditional logic is built on Boolean expressions**
Everything begins with expressions that evaluate to **True** or **False**:
- Comparisons (`>`, `<`, `>=`, `<=`, `==`, `!=`)
- Membership tests (`in`, `not in`)
- Logical operators (`and`, `or`, `not`)
- Chained comparisons (`0 < x <= 5`)

Understanding how to build Boolean expressions is the foundation of all control flow.

---

### **2. Python follows strict Boolean precedence**
Python evaluates conditions in this order:

1. `not`  
2. `and`  
3. `or`  

Use parentheses when your intent does not match Python’s default order.

---

### **3. Short-circuiting affects program behavior**
- `A and B`: if A is False → Python never evaluates B  
- `A or B`: if A is True → Python never evaluates B  

Short-circuiting controls:
- performance
- side-effects (function calls inside conditions)
- rule behavior in real systems

---

### **4. `if`, `elif`, `else` form a single decision chain**
Important principles:
- Conditions are checked **top to bottom**
- Only **one branch** runs
- When a branch is True, Python **skips** all remaining branches
- Most specific conditions should appear first

Always order conditions from **most restrictive** → **least restrictive**.

---

### **5. Nesting creates multi-layered, real-world decision trees**
Use nested logic when rules follow a natural sequence:

> "If this is true, then check something else."

Nesting is ideal for:
- multi-step approvals  
- customer + inventory combinations  
- authentication + permissions  
- multi-rule eligibility checks  

---

### **6. Combine Boolean logic + control flow for expressive rule systems**
Everything from Part 1 becomes building blocks in Part 2:

- `(status == "VIP") and (order_total >= 100)`
- `"hat" in discount_items`
- `not is_blocked`
- `0 < inventory <= 5`

This enables:
- pricing systems  
- discount logic  
- safety checks  
- menu/option logic  
- workflows and gating rules  

---

### **7. Proper ordering prevents unreachable branches**
Incorrect ordering example:

```
if score >= 60:
    ...
elif score >= 90:   # never runs
```

Correct ordering:

```
if score >= 90:
    ...
elif score >= 60:
    ...
```

Always ask:
- Is this condition more general than the ones below?  
- Should this rule be evaluated earlier?  

---

### **8. Helper tools dramatically increase clarity and reduce bugs**
Your helper suite is a game-changer:

- **trace_decision()**  
  Shows a labeled, full-width decision trace

- **trace_condition()**  
  Displays Boolean results for each rule

- **trace_branch()**  
  Labels which branch executed

- **trace_if()**  
  Compact single-condition inspector

Using these consistently gives you:
- transparency  
- easier debugging  
- better intuition  
- fewer logic errors  

These helpers raise your skill far above the Maven-level exercises.

---

### **9. Maintain readable, intention-driven logic**
Best practices:
- Use descriptive variable names  
- Use parentheses for clarity  
- Keep branches small and readable  
- Avoid unnecessary deep nesting  
- Comment the *why*, not the *what*  

Readable control flow → faster debugging and fewer mistakes.

---

### **10. Conditional logic powers real-world automation**
You can now design systems for:
- pricing rules  
- discount engines  
- inventory logic  
- option/menu trees  
- quality and safety checks  
- multi-step authentication  
- eligibility workflows  

This prepares you for the next big steps:
- loops  
- functions  
- pandas filtering  
- data-cleaning decisions  
- machine-learning preprocessing  

You’ve mastered both the **syntax** and the **thinking** behind decision logic.

`````python
# 06.05 Slicing: Extracting Portions of a List
# Using the List Index Map Helper for a flat list

# Slice syntax: list[start : stop : step]

sizes = ['XS', 'S', 'M', 'L', 'XL', 'XXL']

print(sizes[:3])       # ['XS', 'S', 'M']
print(sizes[:4])       # ['XS', 'S', 'M', 'L']
print(sizes[2:5])      # ['M', 'L', 'XL']
print(sizes[::2])      # ['XS', 'M', 'XL']
print(sizes[2:4])      # ['M', 'L']
print(sizes[-3:])      # ['L', 'XL', 'XXL']
print()

# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input: validate isinstance(obj, type)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
show_list_index_map(sizes)         # flat list
print()
`````

`````python
# --------------------------------------------------------------
# 🔧 Helper: Slice Visualizer (for lists, tuples, strings, ranges)
# --------------------------------------------------------------

def visualize_slice(seq, start=None, stop=None, step=None, show_negative=True, label="seq"):
    """
    Visualize how a slice seq[start:stop:step] selects elements
    from a sequence (list, tuple, string, range, etc.).

    Parameters
    ----------
    seq : sequence
        Any indexable, sliceable Python sequence (e.g., list, tuple, str, range).
    start : int or None
        Slice start index (can be negative or None).
    stop : int or None
        Slice stop index (can be negative or None).
    step : int or None
        Slice step (can be negative or None).
    show_negative : bool, default True
        Whether to display negative indices.
    label : str, default "seq"
        Optional label to show in the output header.

    Usage
    -----
    visualize_slice([10, 20, 30, 40, 50], 1, 4, 2)
    visualize_slice(range(10), 2, 9, 3)
    visualize_slice(('a', 'b', 'c', 'd', 'e'), 0, None, 2)
    """

    # ----------------------------------------------------------
    # 1. Basic validation
    # ----------------------------------------------------------
    try:
        n = len(seq)
    except TypeError:
        raise TypeError(
            f"visualize_slice expects a sequence with len(), got {type(seq).__name__}"
        )

    slice_obj = slice(start, stop, step)

    # Normalize slice parameters using slice.indices()
    norm_start, norm_stop, norm_step = slice_obj.indices(n)

    # Compute selected indices based on normalized values
    selected_indices = list(range(norm_start, norm_stop, norm_step)) if n > 0 else []

    # ----------------------------------------------------------
    # 2. High-level summary
    # ----------------------------------------------------------
    print(f"Sequence label : {label}")
    print(f"Type           : {type(seq).__name__}")
    print(f"Length         : {n}")
    print(f"Slice requested: {label}[{start}:{stop}:{step}]")
    print(f"Normalized to  : {label}[{norm_start}:{norm_stop}:{norm_step}]")
    print()

    if n == 0:
        print("This sequence is empty. No indices to visualize.")
        return

    # ----------------------------------------------------------
    # 3. Build index and value rows
    # ----------------------------------------------------------
    # We'll give each slot a fixed width for alignment
    cell_width = 5

    # Positive indices
    idx_row = "idx : "
    for i in range(n):
        idx_row += f"{i:>{cell_width-1}} "

    # Negative indices (optional)
    if show_negative:
        neg_row = "neg : "
        for i in range(n):
            neg_row += f"{i - n:>{cell_width-1}} "
    else:
        neg_row = None

    # Values (repr truncated for readability)
    val_row = "val : "
    for i in range(n):
        text = repr(seq[i])
        if len(text) > cell_width - 1:
            text = text[:cell_width - 2] + "…"
        val_row += f"{text:>{cell_width-1}} "

    # Selection markers
    sel_row = "sel : "
    selected_set = set(selected_indices)
    for i in range(n):
        marker = "^" if i in selected_set else ""
        sel_row += f"{marker:>{cell_width-1}} "

    # ----------------------------------------------------------
    # 4. Print the visualization
    # ----------------------------------------------------------
    print(idx_row)
    if neg_row is not None:
        print(neg_row)
    print(val_row)
    print(sel_row)
    print()

    # ----------------------------------------------------------
    # 5. Show the actual slice result
    # ----------------------------------------------------------
    result = seq[slice_obj]
    print(f"Result of {label}[{start}:{stop}:{step}] → {repr(result)}")
    print()


# --------------------------------------------------------------
# Example usage (you can comment these out in your notebook):
# --------------------------------------------------------------
visualize_slice(['XS', 'S', 'M', 'L', 'XL', 'XXL'], 1, 5, 2, label="sizes")
visualize_slice(range(10, 0, -2), None, None, None, label="r")
visualize_slice(('r', 'g', 'b', 'y', 'k'), 1, 4, 1, label="colors")
`````

`````python
# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types & Nested Lists)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input: validate isinstance(obj, type)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
        # Optionally expand nested lists one level deep
        if expand_nested and isinstance(value, list):
            inner_len = len(value)
            print(f"      ↳ nested list (len={inner_len})")
            for j, inner_val in enumerate(value):
                inner_neg = j - inner_len if show_negative else None
                inner_type = type(inner_val).__name__ if show_types else ""
                inner_repr = _repr_value(inner_val)
                
                # Example format: [2][1]   -1    int  99
                nested_row = f"      [{i}][{j}] "
                if show_negative:
                    nested_row += f"{inner_neg:>4} "
                if show_types:
                    nested_row += f"{inner_type:>10} "
                nested_row += f"{inner_repr}"
                print(nested_row)


# -------------------------------------------
# ✅ Example Usage (Flat List)
# -------------------------------------------
sizes = ['XS', 'S', 'M', 'L', 'XL']
show_list_index_map(sizes)         # flat list
print()

# -------------------------------------------
# ✅ Example Usage (Nested List, Expanded)
# -------------------------------------------
orders = [
    [19.99, 5.99],
    [3.49, 9.99, 14.99],
    [60.00]
]

show_list_index_map(
    orders,                         # nested list
    show_types=True,
    show_negative=True,
    truncate=50,
)
`````

| **Method**          | **Purpose**                                    | **Mutates list?** | **Example**                               |
|---------------------|------------------------------------------------|-------------------|-------------------------------------------|
| `lst.append(x)`     | Add item to end of list                        | Yes               | `sizes.append("XL")`                      |
| `lst.insert(i, x)`  | Insert item at index `i`                       | Yes               | `sizes.insert(0, "XS")`                   |
| `lst.extend(iter)`  | Add all elements from another iterable         | Yes               | `sizes.extend(["2XL", "3XL"])`            |

| `lst.reverse()`     | Reverse list in place                          | Yes               | `numbers.reverse()`                       |

| `lst.index(x)`      | Return index of first matching value           | No                | `sizes.index("M")`                        |

## 🔍 Notes for Essential Methods

- `append()` vs `insert()`  
  - Use `append()` unless you *truly* need a specific position  
  - `insert()` shifts the entire list (costly with large data)

- `remove()` deletes **only the first** matching item  
  - Useful to know when cleaning data

- `pop()` is the **only list method that returns a value**

- `sort()` is **in-place**  
  - Use `sorted(lst)` when you want to keep the original list intact  
  - This difference becomes crucial for data analysis

## 🧠 Deep-Dive Notes for Advanced Behaviors

### 🔸 Shallow vs Deep Copy (Critical Concept)
- `lst.copy()` copies *only* the outer list  
- nested lists remain linked  
- modifying nested elements affects both lists  
- Use `deepcopy()` when safety is required

### 🔸 Slice Non-Inclusive Behavior
- `lst[a:b]` excludes index `b`  
- Same rule as string slicing  
- Predictable and consistent across Python

### 🔸 Reversing Safely
- `lst[::-1]` → **non-destructive**  
- `lst.reverse()` → **mutates**  
- Choose based on whether you need the original

### 🔸 Slice Assignment

`lst[1:3] = ["A", "B", "C"]`  
- You can replace a slice with a different number of items  
- Useful for editing data in place  
- Python will automatically adjust list size

In Python, **sequences** are ordered collections of items. The three core sequence types in this chapter are:

# 06.03 Membership Tests ("in" and "not in")

Because lists are iterable, you can test whether a value exists inside a list.

```python
sizes_in_stock = ['XS', 'S', 'M', 'L', 'XL']

print('M' in sizes_in_stock)       # True
print('XXL' in sizes_in_stock)     # False

# You can use membership tests in if/elif logic
if 'M' in sizes_in_stock:
    print("Medium is available.")
elif 'S' in sizes_in_stock:
    print("Small is available instead.")
else:
    print("Neither size is available.")
```

This concept is heavily used in:

- Filtering data  
- Validating inputs  
- Checking categories  
- Cleaning data  
- Applying business rules

# 06.04 Indexing (Accessing Individual Elements)

Use square brackets with an index number to access a single item.

```python

```

### Mini-diagram


### Negative indexing  
Python allows negative indexes, which count from the end:

```python

print(colors[-2])   # 'green'
```

Negative indexing is standard in Python and very powerful for last-item access.

# 06.05 Slicing (Extracting Portions of a List)

Slice syntax:

```
list[start : stop : step]
```

- **start** = where the slice begins  
- **stop** = where the slice ends (non-inclusive)  
- **step** = how many to skip  

```python
sizes = ['XS', 'S', 'M', 'L', 'XL', 'XXL']

print(sizes[:3])       # ['XS', 'S', 'M']
print(sizes[2:5])      # ['M', 'L', 'XL']
print(sizes[::2])      # ['XS', 'M', 'XL']
print(sizes[-3:])      # ['L', 'XL', 'XXL']
```

### Mini-diagram: Non-inclusive stop  
For `sizes[2:5]`:

```

Value:   XS    S    M     L    XL   XXL
Slice:              [ M    L    XL ]
Stop index = 5 (excluded)
```

Common uses of slicing:

- Selecting subsets  
- Pulling rows of data  
- Extracting ranges  
- Removing headers  
- Splitting time intervals

# 06.06 Unpacking Lists

Unpacking allows you to assign list elements directly to variables.

```python
sizes = ['S', 'M', 'L']

print(small)    # 'S'
print(medium)   # 'M'
print(large)    # 'L'
```

### Important rule  
> The number of variables must match the number of elements.

Otherwise, Python raises a `ValueError`.

# 🧮 Universal List Index Map Helper (Professional Version – With Parameters)

This helper is the **list equivalent** of your Universal String Index Map Helper.

It is designed to:

- Show **positive and negative indices** for each element in a list  
- Optionally display **data types**  
- Optionally **truncate long values** for readability  
- Optionally **expand nested lists** and show their indices as `[outer][inner]`  
- Provide a **clear, student-friendly view** of how indexing works in real data  

You can use this to:

- Understand and debug indexing  
- Teach indexing to others  
- Explore nested list structures  
- Build confidence with list access patterns before loops and slicing-heavy work  

---

## 🔧 Python Helper: `show_list_index_map`

```python
# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types & Nested Lists)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input validation (student-friendly)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
        # Optionally expand nested lists one level deep
        if expand_nested and isinstance(value, list):
            inner_len = len(value)
            print(f"      ↳ nested list (len={inner_len})")
            for j, inner_val in enumerate(value):
                inner_neg = j - inner_len if show_negative else None
                inner_type = type(inner_val).__name__ if show_types else ""
                inner_repr = _repr_value(inner_val)
                
                # Example format: [2][1]   -1    int  99
                nested_row = f"      [{i}][{j}] "
                if show_negative:
                    nested_row += f"{inner_neg:>4} "
                if show_types:
                    nested_row += f"{inner_type:>10} "
                nested_row += f"{inner_repr}"
                print(nested_row)


# -------------------------------------------
# ✅ Example Usage (Flat List)
# -------------------------------------------
sizes = ['XS', 'S', 'M', 'L', 'XL']
show_list_index_map(sizes)

# -------------------------------------------
# ✅ Example Usage (Nested List, Expanded)
# -------------------------------------------
orders = [
    [19.99, 5.99],
    [3.49, 9.99, 14.99],
    [60.00]
]

show_list_index_map(
    orders,
    show_types=True,
    show_negative=True,
    truncate=50,
    expand_nested=True
)
```

---

## 📝 Key Takeaways for the List Index Helper

- `show_list_index_map(lst, ...)` gives you a **readable table** of:
  - positive index (`idx`)  
  - optional negative index (`neg`)  
  - optional type (`type`)  
  - value (`value`)  

- `expand_nested=True` lets you:
  - see **nested lists** clearly  
  - view indices like `[outer][inner]`  
  - understand how to access them using `lst[i][j]`  

- `truncate` helps keep very long values readable in the console.

- This helper is ideal for:
  - learning list indexing  
  - debugging index errors  
  - understanding nested structures  
  - preparing for loops, slices, and data munging

`````python
# 06.12 Adding Elements: .append() and .insert()

# Two main ways to add new elements:

# 1. append(element) - Adds to the end of the list.
# Mutates the list in place ⚠️

items = ['Hat', 'Gloves', 'Scarf']
items.append('Coat')

print(items)  # display mutated list


# 2. insert(index, element) 
# Inserts at a specific position.
# Mutates the list in place ⚠️ 

items = ['Hat', 'Gloves', 'Scarf']
items.insert(3, 'Coat')   # insert at index 3 (fourth position)

print(items, end="\n\n")   # display mutated list


# ==================================================
# 06.12b Changing List Elements (Direct Assignment)
# ==================================================
# Replace items using indexing
# Mutates the list in place ⚠️

new_items = ["Coat", "Backpack", "Scarf"]

# Replace the second item
new_items[1] = "Gloves"

print(new_items)   # display mutated list
`````

`````python
# 06.13 Combining & Repeating Lists with arithmetic operators

# Lists can be merged (concatenated) with `+` and repeated with `*`

list1 = ['A', 'B']
list2 = ['C', 'D']

combined = list1 + list2
print(combined)      # ['A', 'B', 'C', 'D']

repeated = list1 * 3
print(repeated)      # ['A', 'B', 'A', 'B', 'A', 'B']
`````

`````python
# 06.14 Removing Elements: del and .remove()
# Two main ways to delete elements:

# 1. del removes items by index or slice
# Mutates the list in place ⚠️ 

# del list_name[index:slice]
list_name = ['Hat', 'Gloves', 'Scarf', 'Coat', 'Backpack']
del list_name[0:3]     # deletes indices 0 through 2
print(list_name, "\n")      # ['Coat', 'Backpack']


# 2. remove(value) 
# Removes the first occurrence of a value only ⚠️
# Mutates the list in place ⚠️

colors = ['red', 'blue', 'green', 'blue']
colors.remove('blue')
print(colors)        # ['red', 'green', 'blue']

# 3. remove(value)
t = [1, 2, 3]
removing = 2
t.remove(removing)   # mutates the list in place
print(t)                    # [1, 3]

# 4. remove(value) - Returns None ❌
# t = [1, 2, 3]
# t = t.remove(2)
# print(t)  # Now t points to None instead of your list ❌
`````

`````python
# 06.15 Sorting Lists: .sort() vs sorted()
# This is a **critical** concept in Python.

# .sort() method
# Mutates the list in place ⚠️

values = [5, 2, 8, 1]
values.sort()

print(values, end="\n\n")   # [1, 2, 5, 8]


# sorted() function - returns a new sorted list  
# Does not mutate the original

values = [5, 2, 8, 1]
new_values = sorted(values)

print(new_values, "\n")    # [1, 2, 5, 8]
print(values)              # [5, 2, 8, 1]  # original unchanged
`````

`````python
# 06.16 Reversing Lists: .reverse() and slicing

# Two ways:

# 1. reverse() method
# Mutates the list in place ⚠️

values = [10, 20, 30]

values.reverse()

print(f"Reverse method: {values} (mutated list)")    # [30, 20, 10]
print()

# 2. Slicing - returns a new reversed list
# Does not mutate original.

values = [10, 20, 30]
rev = values[::-1]      # negative slicing won't change underlying list

print("Reverse slicing:", rev)             # [30, 20, 10]
print(f"Original unchanged: {values}\n")   # original unchanged

print(f"Reversed: {values[::-1]}\nOriginal is still: {values}\n")
print("Type of first element:", type(values[0]))  # <class 'int'>
print()

print(values, end="\n\n")
`````

`````python
# 06.17 Counting & Finding Values
# If the value does not exist, Python raises a **ValueError**.


# .count(value) - Counts how many times a value appears.
letters = ['a', 'b', 'a', 'c', 'a']
print(letters.count('a'))     # 3


# .index(value) - Returns the first index where the value appears.
letters = ['a', 'b', 'a']
print(letters.index('a'))     # 0


# 1. Find where a specific item appears
fruits = ['apple', 'banana', 'cherry', 'apple']

first_apple_pos = fruits.index('apple')        # 0
print(first_apple_pos)

last_apple_pos = fruits.index('apple', 1)     # start searching at index 1 → 3
print(last_apple_pos)


# 2. Reorder columns by moving one column next to another
columns = ['date', 'customer_id', 'product', 'revenue', 'region']

# Move 'region' right after 'product'
pos = columns.index('product')
print(columns.index('product'))  # 2

columns.insert(pos + 1, columns.pop(columns.index('region')))  # ['date', 'customer_id', 'product', 'region', 'revenue']
print(columns)

# ⚠️ **Caution:** ⚠️
# Technically correct, but not recommended in real analytics pipelines.  
# This pattern mutates the list twice (.insert and .pop) and relies on multiple `.index()` lookups.  
# Missing or duplicated column names can cause crashes or silent data corruption.

# Safe Option 1: Reorder columns by building a new list explicitly
columns = ['date', 'customer_id', 'product', 'revenue', 'region']

new_order = [
    'date',
    'customer_id',
    'product',
    'region',
    'revenue'
]

orig_columns = columns[:]  # make a full copy of the original list
columns = new_order 

# Safe Option 2: List comprehension to reorder safely
desired_order = ['date', 'customer_id', 'product', 'region', 'revenue']
columns = [col for col in desired_order if col in columns]
# List comprehension reorder does not mutate — but it replaces which can be dangerous if the original structure matters.


# 🔬 Long Version of list comprehension (What Python Does Internally)
# Python expands the one-liner into:

orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

temp_list = []
for col in desired_order:
    if col in columns:
        temp_list.append(col)

columns = temp_list

# Same logic — the comprehension is just more compact and efficient.


# Safe Option 3: In pandas (safest)
# df = df[['date', 'customer_id', 'product', 'region', 'revenue']]
`````

`````python
# -------------------------------
# Mutating inside the function
# List mutation with intent
# -------------------------------

def normalize_scores(scores):
    """
    Mutates the list in place by clamping values to [0, 100].
    This function intentionally modifies the input list.
    """
    for i, value in enumerate(scores):
        scores[i] = max(0, min(100, value))  # Value is clamped into the range [0, 100]
        # min(100, value) → upper bound
        # max(0, result) → lower bound

scores = [110, 85, -5, 92]
normalize_scores(scores)
print(scores)  # [100, 85, 0, 92]


# -------------------------------
# Non-mutating inside the function
# (often preferred)
# -------------------------------

def normalized_scores(scores):
    """Returns a new list without mutating the input."""
    return [max(0, min(100, s)) for s in scores]


scores = [110, 85, -5, 92]
new_scores = normalized_scores(scores)

print(scores)      # unchanged
print(new_scores)  # normalized
`````

# 📦 Lists as Function Arguments (Mutation & Side Effects)

## Concept Overview

When a **mutable object** (such as a list) is passed into a function, the function receives a **reference to the same object**, not a copy.  
As a result, **in-place modifications inside the function affect the original object** outside the function.

This behavior is not inherently good or bad—but it **must be intentional and explicit**, because it introduces **side effects**.

A *side effect* occurs when a function changes something outside its local scope.

---

## Example 1: A Function That MUTATES Its Input (Side Effect)

```python
def pop_first(lst):
    """
    Removes and returns the first element of the list.
    
    ⚠️ Side effect:
    This function MUTATES the original list passed in.
    """
    return lst.pop(0)   # modifies the list in place
```

### Using the function

```python
letters = ['a', 'b', 'c']

first = pop_first(letters)
print(first)    # returned value
print(letters)  # original list after mutation
```

### Output

```python
a
['b', 'c']
```

✔ The function returns `'a'`  
⚠ The original list is permanently changed

---

## Why This Is Risky in Practice

Functions with side effects can:
- Make code harder to reason about
- Cause bugs when the caller does not expect mutation
- Break reuse and composability
- Complicate debugging and testing

For these reasons, **mutating arguments should be a deliberate design choice**, not an accidental consequence.

---

## Example 2: A Safer, NON-MUTATING Version (Preferred)

```python
def pop_first_safe(lst):
    """
    Returns the first element WITHOUT modifying the original list.
    """
    return lst[0]
```

```python
letters = ['a', 'b', 'c']

first = pop_first_safe(letters)
print(first)
print(letters)
```

### Output

```python
a
['a', 'b', 'c']
```

✔ No side effects  
✔ Original list preserved  
✔ Function behavior is predictable

---

## Example 3: Explicit Copying (If Mutation Is Needed Internally)

```python
def pop_first_copy(lst):
    """
    Works on a COPY of the list to avoid side effects.
    """
    temp = lst.copy()
    return temp.pop(0)
```

---

## Key Takeaways

- Lists are **mutable**
- Functions receive **references**, not copies
- Methods like `.pop()` modify objects **in place**
- Mutation introduces **side effects**
- Prefer **pure functions** unless mutation is intentional
- If mutation is required, **document it clearly**
- Copy inputs when safety and clarity matter

This mental model scales cleanly to dictionaries, sets, class instances, and real-world data pipelines.

# 06.15 Sorting Lists: .sort() vs sorted()

This is a **critical** concept in Python.

### **.sort() method**
- Sorts *in place*  
- Permanently changes the list  
- Returns None  

```python
values = [5, 2, 8, 1]
values.sort()

print(values)    # [1, 2, 5, 8]
```

### **sorted() function**
- Returns a *new* sorted list  
- Leaves the original unchanged  

```python
values = [5, 2, 8, 1]
new_values = sorted(values)

print(values)       # [5, 2, 8, 1]
print(new_values)   # [1, 2, 5, 8]
```

### Best practice  
> Do not sort in place until you are sure you no longer need the original data.

This is especially true in analytics where the original ordering matters.

# 06.16 Reversing Lists: .reverse() and slicing

Two ways:

### **1. reverse() method**  
Modifies the list in place.

```python
values = [10, 20, 30]
values.reverse()

print(values)    # [30, 20, 10]
```

### **2. slicing (non-destructive)**  
Useful for creating reversed copies.

```python
values = [10, 20, 30]
rev = values[::-1]

print(rev)       # [30, 20, 10]
print(values)    # original unchanged
```

## 🧠 Summary

### ✔ `list.reverse()`  
- Mutates original list  
- Returns `None`  
- Permanent change  

### ✔ `values[::-1]`  
- Does *not* mutate original  
- Returns a new reversed list  
- Safe for analysis  
- Best for beginners and data workflows

Because the **original list object itself** has been reversed. No copy. No return value.

# 📦 Helper Box: How to Read Any List Comprehension (3-Step Method)

List comprehensions are one of Python’s most powerful and expressive features.  
They compress a full loop + condition + append operation into a clear, readable one-liner.

This helper teaches you how to **decode** any list comprehension using a simple 3-step process, and includes a **Safety Upgrade** showing how data professionals prevent accidental loss of original data.

---

## 🔍 General Pattern

A list comprehension always follows this structure:

```python
[new_item   for item in iterable   if condition]
```

It has **three components**:

1. **Output expression** → what goes into the new list  
2. **Iteration** → where items come from  
3. **Condition (optional)** → keep only items that pass the test  

---

# 🟦 Apply the 3-Step Method to Your Example

Your comprehension:

```python
[col for col in desired_order if col in columns]
```

### Step 1 — “For each…”  
`for col in desired_order`  
→ Loop through each column name in `desired_order`.

### Step 2 — “If…”  
`if col in columns`  
→ Only keep columns that actually exist in the current dataset.

### Step 3 — “Put into the new list…”  
`col`  
→ Place the column name into the new list as-is.

### ✔️ Final meaning (plain English)

> “Rebuild the columns list in the order of `desired_order`, but keep only those columns that appear in the existing `columns` list.”

---

# 🔬 The Long Version (What Python Does Internally)

```python
orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

temp_list = []
for col in desired_order:
    if col in columns:
        temp_list.append(col)

columns = temp_list
```

This is exactly what the comprehension does behind the scenes.

- It **does not mutate** the original list object.  
- It **does overwrite** the variable `columns` so the previous ordering is lost unless backed up.

---

# 🛡️ **Safety Upgrade: How Professionals Protect Original Data**

The long version ends with:

```python
columns = temp_list
```

This **reassigns** the variable `columns`, meaning the original ordering is gone unless saved.

In analytics, ETL, and MSBA contexts, you never want to lose original schema information.

### ✔️ Safe, production-style pattern

```python
orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

columns = [col for col in desired_order if col in orig_columns]
```

### Why this is safer:

- `orig_columns` preserves the original ordering and content  
- You can always restore or compare later  
- The comprehension now filters against the backup, not itself  
- This is how data engineers prevent silent logical bugs  

This pattern becomes essential in:

- schema reconciliation  
- feature engineering  
- pandas pipelines  
- machine learning preprocessing  
- reproducible analytics workflows  

---

# 📘 Additional Examples for Mastery

### Example A: Squaring numbers
```python
squares = [n*n for n in range(5)]
# → [0, 1, 4, 9, 16]
```

### Example B: Filtering even numbers
```python

# → [0, 2, 4, 6, 8]
```

### Example C: Cleaning strings


# 🧩 Key Takeaways: Helper Box

- A list comprehension is built from **output → loop → optional filter**.  
- It **creates a new list**; it does **not mutate** the original list.  
- Reassigning a variable (e.g., `columns = temp_list`) can **discard the original data**.  
- In real analytics work, **always back up original data** before reassigning.  
- You should master the **long version first** — training wheels that give you full visibility.  
- Comprehensions become powerful and natural *after* you understand the underlying loop logic.

---

End of Helper Box.


`````python
# ------------------------------------------------------------
# LIST COMPREHENSIONS — MECHANICS, SHAPE, AND SAFE USAGE
# ------------------------------------------------------------
# A list comprehension is a compact syntax for:
#   1) Iterating over an iterable
#   2) Transforming each element
#   3) Collecting the results into a new list
#
# It is a *syntactic shortcut* for the accumulator pattern:
#
#   result = []
#   for x in iterable:
#       result.append(transform(x))
#
# List comprehensions:
# - Preserve structure (one output per input)
# - Do not mutate the input iterable
# - Are best used for *simple, single-step transformations*
# - Should not contain complex logic or side effects
# ------------------------------------------------------------


# ------------------------------------------------------------
# 1) BASELINE: ACCUMULATOR PATTERN (EXPLICIT MECHANICS)
# ------------------------------------------------------------

numbers = [1, 2, 3, 4, 5]

squares = []                           # accumulator

for n in numbers:                      # iterate input
    squares.append(n ** 2)             # transform + collect

print("Accumulator result:", squares)
print()


# ------------------------------------------------------------
# 2) DIRECT TRANSLATION: LIST COMPREHENSION
# ------------------------------------------------------------

squares_comp = [n ** 2 for n in numbers]

print("List comprehension result:", squares_comp)
print()

# MECHANICAL EQUIVALENCE:
#
# [n ** 2 for n in numbers]
#
# means:
#
# - "for each n in numbers"
# - "compute n ** 2"
# - "collect into a new list"


# ------------------------------------------------------------
# 🟦 Helper: Comprehension Decomposer
# ------------------------------------------------------------

def explain_comprehension(iterable, transform_fn):
    """
    Diagnostic helper: shows what a list comprehension does step by step.
    """
    print("COMPREHENSION TRACE\n--------------------")
    result = []
    
    for idx, value in enumerate(iterable):
        transformed = transform_fn(value)
        result.append(transformed)
        print(f"Iteration {idx}: input = {value!r} → output = {transformed!r}")
    
    print("\nFinal list:", result)
    return result


# Example usage:

explain_comprehension(numbers, lambda x: x ** 2)
print()


# ------------------------------------------------------------
# 3) FILTERING WITH A CONDITION
# ------------------------------------------------------------
# Syntax:
#   [expression for item in iterable if condition]

values = [10, 15, 20, 25, 30]

evens = [x for x in values if x % 2 == 0]

print("Even values:", evens)
print()

# Equivalent accumulator form:

evens_explicit = []

for x in values:
    if x % 2 == 0:
        evens_explicit.append(x)

print("Even values (explicit):", evens_explicit)
print()


# ------------------------------------------------------------
# 4) ELEMENTWISE TRANSFORM OVER PARALLEL LISTS
# ------------------------------------------------------------
# (Zip-based version is safer than indexing)

listA = [13, 18, 23, 42, 27]
listB = [42, 14, 31, 39, 40]

listC = [a + b for a, b in zip(listA, listB)]

print("Elementwise sum:", listC)
print()

# Equivalent accumulator form:

listC_explicit = []

for a, b in zip(listA, listB):
    listC_explicit.append(a + b)

print("Elementwise sum (explicit):", listC_explicit)
print()


# ------------------------------------------------------------
# 5) NESTED LIST COMPREHENSION (STRUCTURE PRESERVATION)
# ------------------------------------------------------------
# Used when transforming nested lists (matrices, grouped records)

orders = [
    [120.00, 45.00, 89.99],
    [250.00, 80.00],
    [19.99, 55.00, 150.00, 40.00]
]

discounted_orders = [
    [round(subtotal * 0.9, 2) for subtotal in customer_orders]
    for customer_orders in orders
]

print("Discounted orders:", discounted_orders)
print()

# Equivalent nested-loop accumulator form:

discounted_orders_explicit = []

for customer_orders in orders:
    new_row = []
    for subtotal in customer_orders:
        new_row.append(round(subtotal * 0.9, 2))
    discounted_orders_explicit.append(new_row)

print("Discounted orders (explicit):", discounted_orders_explicit)
print()


# ------------------------------------------------------------
# ⚠️ 6) WHEN NOT TO USE LIST COMPREHENSIONS
# ------------------------------------------------------------
# Avoid list comprehensions when:
# - The logic requires multiple statements
# - You need intermediate variables
# - You need debugging prints or tracing
# - You are mutating external state
# - The comprehension becomes hard to read

# Bad example (do NOT do this):

# result = [print(x) or x ** 2 for x in range(5)]  # side effects + unreadable


# ------------------------------------------------------------
# 🧠 RULE OF THUMB
# ------------------------------------------------------------
# Use a list comprehension when:
# - One input → one output
# - One clear transformation
# - No side effects
# - Readability improves over a loop
#
# Otherwise, use an explicit loop.
`````

# 06.30 Overview

In the real world, data is often **multi-dimensional**.  
One customer has multiple orders.  
One product has multiple attributes.  
One day has multiple transactions.

Python represents this with **nested lists** (lists inside lists).

In addition, you must understand how **copying lists** works.  
Maven introduces three copy behaviors:

1. **Assignment** (two variables, same object)
2. **shallow copy** via `.copy()`  
3. **deep copy** via `deepcopy()`  

| Operation              | What It Produces                              | Safe for Nested Lists? | Mutates Original? | Notes                                        |
| ---------------------- | --------------------------------------------- | ---------------------- | ----------------- | -------------------------------------------- |
| `b = a`                | **Second reference variable** (same object)   | No                     | Yes               | No copy; `a` and `b` are aliases             |
| `b = a.copy()`         | **Shallow copy** (new list, shared internals) | No                     | No                | Outer list copied; nested lists still shared |
| `b = copy.deepcopy(a)` | **Deep copy** (fully independent data)        | Yes                    | No                | Recursively copies all levels                |


This section explains:

- How nested lists are structured  
- How to access inner elements  
- How methods behave on nested lists  
- What “shallow” vs “deep” truly means  
- Why copying lists incorrectly creates bugs  
- How memory references work  

This is one of the MOST important sections for real analytics work.


`````python
# 06.31 What Are Nested Lists?
# A nested list is simply a list where each element is another list.

orders = [
    [19.99, 5.99],       # Customer 1
    [3.49, 9.99, 14.99], # Customer 2
    [45.00]              # Customer 3
]
`````

`````python
# 06.32 Accessing Nested List Elements
# To access *one entire sublist*, use one index.

orders = [
    [19.99, 5.99],
    [3.49, 9.99, 14.99],
    [45.00]
]

print(orders[1])       # [3.49, 9.99, 14.99]

# To access **one element inside a nested list**, use two indexes:

print(orders[1][2])    # 14.99
`````

`````python
# 06.33 Using List Methods on Nested Lists
# List methods behave normally on nested lists: append() and count() examples

nested = [
    ['a', 'b'],
    ['c', 'd', 'e'],
    ['f']
]

# append a value to the second inner list
nested[1].append('k')


# count how many 'f' values appear in the entire structure
total_f = 0
for sublist in nested:
    total_f += sublist.count('f')

print(nested)
print("Number of 'f' values:", total_f)


# Methods apply to inner lists exactly the same as they do on flat lists.
`````

`````python
# 06.34 Copying Lists: The Three Behaviors - See Helper Box (Below)

# 1. Variable Assignment  
# Creates a second reference variable, NOT a true copy.


list1 = [[10, 20], [30, 40]]
list2 = list1      # assignment, not copying

list2[0][1] = 999

# Both variables point to the same memory location (same object, two names)
# Use this ONLY when intentional
print(list1)  # [[10, 999], [30, 40]]
print(list2)  # [[10, 999], [30, 40]]


# 2. Shallow Copy (.copy())  
# Copies the outer list, but not inner lists.

original = [[1, 2], [3, 4]]
copy1 = original.copy()

copy1[0][1] = 999   # modifying the INNER list

print(original)   # [[1, 999], [3, 4]]
print(copy1)      # [[1, 999], [3, 4]]
# `.copy()` creates a new list, but its elements still reference the same inner lists.

# Important rule  
# Replacing an entire inner list creates separation.
# Modifying an element inside an inner list does not (both lists change)

# Example:

copy1[0] = ['new', 'list']   # replace entire inner list

# Now, original[0] stays unchanged.


# 3. Deep Copy (deepcopy())
# Creates a 100 percent independent copy.

from copy import deepcopy   # import deepcopy

original = [[1, 2], [3, 4]]
copy2 = deepcopy(original)

copy2[0][1] = 999

print(original)   # unchanged, [[1, 2], [3, 4]]
print(copy2)      # changed, [[1, 999], [3, 4]]
`````

`````python
# 06.35 Business Example: Customer Order Copies
# This demonstrates the dramatic difference between shallow and deep behavior.

from copy import deepcopy   # import deepcopy

# Nested data representing customer orders
multi_orders = [
    [15.99, 23.50],
    [9.99, 14.25, 7.50],
    [60.00]
]

# Shallow copy
shallow = multi_orders.copy()

# Deep copy
deep = deepcopy(multi_orders)

# Modify inner structures
shallow[1][0] = 888   # affects both shallow and original
deep[2][0] =    777   # affects only deep copy

print("Original:", multi_orders)
print("Shallow:", shallow)
print("Deep:", deep)
`````

### Mini-diagram

Methods apply to inner lists exactly the same as they do on flat lists.

# 06.34 Copying Lists: The Three Behaviors - See Helper Box (Below)

Copying lists in Python is subtle.  
This is where many beginners (and some professionals) make mistakes.

There are **three** main behaviors:

---

## 1. Variable Assignment  
### Creates a **second reference variable**, NOT a true copy.

```python
list1 = [[10, 20], [30, 40]]
list2 = list1      # assignment, not copying

list2[0][1] = 999

print(list1)  # [[10, 999], [30, 40]]
print(list2)  # [[10, 999], [30, 40]]
```

### Why this happens

Both variables point to **the same memory location**.

### Memory diagram

```
list1 ───► [ [10, 20], [30, 40] ]
           ▲
list2 ─────┘  (same object, two names)
```

**Any change to one affects the other.**

Use this ONLY when intentional.

---

## 2. Shallow Copy (.copy())  
### Copies the outer list, but not inner lists.

```python
original = [[1, 2], [3, 4]]
copy1 = original.copy()

copy1[0][1] = 999   # modifying the INNER list

print(original)   # [[1, 999], [3, 4]]
print(copy1)      # [[1, 999], [3, 4]]
```

### Why this happens

`.copy()` creates a new list, **but its elements still reference the same inner lists**.

### Memory diagram

```
original:   [  L1,     L2  ]
               │       │
copy1:      [  L1,     L2  ]
```

Where `L1` and `L2` are the same inner list objects.

### Important rule  
> Replacing an entire inner list creates separation.  
> Modifying an element inside an inner list does not.

Example:

```python
copy1[0] = ['new', 'list']   # replace entire inner list
```

***Now, original[0] stays unchanged.***

---

## 3. Deep Copy (deepcopy())

### Creates a **100 percent independent copy**.

```python
from copy import deepcopy   # import deepcopy

original = [[1, 2], [3, 4]]
copy2 = copy.deepcopy(original)

copy2[0][1] = 999

print(original)   # unchanged, [[1, 2], [3, 4]]
print(copy2)      # changed, [[1, 999], [3, 4]]
```

### Memory diagram

```
original:   [ L1 ,     L2  ]
copy2:      [ L1',     L2' ]
```

Where L1 and L2 are completely separate new lists.

### When to use deepcopy

- Complex nested structures  
- Multi-level JSON-like data  
- When list mutability must not affect the original  
- Data cleaning pipelines that require preservation  

---

# 🧰 Helper Box: **Shallow Copy vs Deep Copy — The Only Rule You Need**

This helper gives you the *correct, professional-level mental model* for copying data structures in Python.  
Most confusion about `.copy()` comes from not understanding how references behave inside nested structures.

This box will stay valid for your entire MSBA program, pandas work, and future data science.

---

## 🟦 1. The One Rule That Solves Everything

### ✅ **`.copy()` is safe only for *flat* lists.**  
### ❌ **`.copy()` is NOT safe for nested data structures.**  
### ⭐ **Use deep copy only when you *must* — and avoid nested structures in analytics whenever possible.**

That’s the rule.

Let’s expand it.

---

## 🟩 2. What `.copy()` Actually Does (Shallow Copy)

```python
new_list = old_list.copy()
```

This copies **only the outer list**.

It does **NOT** copy the objects inside it.

### ✔️ Safe:
```python
[1, 2, 3, 4]
['a', 'b', 'c']
[True, False, True]
```

These contain simple, immutable elements → no shared inner state.

### ❌ Not Safe:
```python
[[1, 2], [3, 4]]
[ ['a', 'b'], ['c', 'd'] ]
[ [1, 2], 99, ['x', 'y'] ]
```

These contain **nested lists**, which are still shared between copies.

---

## 🟥 3. Visual Mental Model You Should Use

### Example List
```python
list1 = [[1, 2], [3, 4]]
list2 = list1.copy()
```

### Reality (shallow copy)
```
list1 ───► [ A , B ]     # where A is [1, 2], and B is [3, 4]  ⚠️
list2 ───► [ A , B ]     # [[1, 2], [3, 4]]  ⚠️

A = [1, 2]   ← shared
B = [3, 4]   ← shared

        ┌───────────────┐
list1 ─►│   index 0 ─────┼──► A ([1,2])
        │   index 1 ─────┼──► B ([3,4])
        └───────────────┘

        ┌───────────────┐
list2 ─►│   index 0 ─────┼──► A ([1,2])
        │   index 1 ─────┼──► B ([3,4])
        └───────────────┘

```

Both lists point to the **same** inner lists.

### Mutation ***inside*** A or B affects BOTH:
```python
list1[0][1] = 999

list1[0] ----> A <---- list2[0]  ⚠️
```
We are mutating A, because list1[0] refers to A.

But since list2[0] ALSO refers to A, we see the mutation in both

Now *both* `list1` and `list2` show:
```
[1, 999]
```

⚠️ This is why shallow copy is unsafe for nested data.

---

## 🟦 4. Why Replacing an Entire Nested List *Doesn't* Affect the Copy

```python
list1[0] = ['x', 'y', 'z']
```

This does *not* mutate the inner list.  
It **rebinds** index 0 to a *new* list.

```
list1 ─► ['x','y','z']   (new)
list2 ─► [1, 2]          (old)

list1 ───► [ NEW_LIST , B ]
list2 ───► [    A     , B ]
```

So the effect does *not* propagate.

This explains why your instructor’s example behaved differently in the two cases.

---

## 🟧 5. The Only Time You Use `deepcopy`

```python
from copy import deepcopy
new_list = deepcopy(old_list)
```

Use `deepcopy` only when:

- You must work with nested Python objects  
- And you must guarantee isolation  
- And flattening is not possible  

**Data analysts rarely use deepcopy because they rarely use nested Python lists at scale.**  
Instead, they use:

- pandas DataFrames  
- NumPy arrays  
- flat lists  
- relational tables  

These structures avoid arbitrary nesting → deepcopy becomes unnecessary.

---

## 🟦 6. Safe Practice Rules (Adopt These)

### ✅ Rule 1: Avoid nested Python lists in analytics  
Flatten early, or convert to pandas.

### ✅ Rule 2: If nesting is unavoidable, assume `.copy()` is unsafe  
Do not trust shallow copies for nested data.

### ✅ Rule 3: Mutation inside a nested object propagates  
Changing elements inside an inner list changes all shallow copies.

### ✅ Rule 4: Rebinding a nested list does NOT propagate  
Replacing the entire inner list with a new one isolates the change.

### ✅ Rule 5: Use `deepcopy` **only** when nested structures cannot be avoided  
But aim to design your code so this is rare.

---

## 🟩 7. Quick Reference Table

| Operation | Copies Outer List? | Copies Inner Lists? | Safe for Nested Data? |
|----------|---------------------|----------------------|------------------------|
| `new = old.copy()` | ✅ Yes | ❌ No | ❌ No |
| `new = old[:]` | ✅ Yes | ❌ No | ❌ No |
| `new = list(old)` | ✅ Yes | ❌ No | ❌ No |
| `new = deepcopy(old)` | ✅ Yes | ✅ Yes | ⭐ Yes |

---

## 🟦 8. Helper Box Summary

- `.copy()` copies the **container**, not the **contents**.  
- Nested objects remain **shared** and therefore **dangerous**.  
- Replace inner lists (rebinding) → safe  
- Modify inside an inner list → unsafe  
- Deep copy solves the problem but should be unnecessary in well-designed analytics code.  
- Flat structures avoid all these issues — which is why pros use them.

---

End of Helper Box.

# 06.40 Overview

A **tuple** is another type of sequence in Python, similar to a list but with one major difference:

# ✔ Tuples are **immutable**  
Once a tuple is created, **it cannot be changed**.  
You cannot add, delete, or replace items.

Tuples are incredibly important in analytics work because they:

- ensure data cannot be accidentally changed  
- take up less memory than lists  
- run faster for some operations  
- appear naturally as function outputs  
- prevent logical errors in pipelines

---

# Tuples vs Lists: When to Choose Which

| Use List When... | Use Tuple When... |
|------------------|-------------------|
| You need to modify data | Data must stay fixed |
| Storing transaction sets | Storing coordinates |
| Cleaning inconsistent values | Returning multiple outputs |
| Building datasets dynamically | Locking configuration values |
| Data is evolving | Data is final and stable |

### Rule of Thumb  
> If you do not need mutability, choose a tuple.  

This protects your logic downstream.

---

# Summary (Part 4 — Tuples)

- Tuples are immutable sequences; once created, they cannot be changed.  
- They use **parentheses** and require a trailing comma for one-element tuples.  
- Tuples are more memory-efficient and faster than lists.  
- They prevent accidental changes and improve code safety.  
- Many functions return multiple values packed into tuples.  
- Unpacking enables fast, readable extraction of values.  
- Tuples are ideal for fixed data such as configuration values, coordinates, and stable metadata.

---


`````python
# 06.41 Creating Tuples
# Tuples use parentheses instead of square brackets.


# Standard tuple
sizes = ('S', 'M', 'L')

# Tuple with mixed types
info = ('Alice', 42, True)

# Empty tuple
empty = ()

# Converting an iterable into a tuple
letters = tuple('abcd')      # ('a', 'b', 'c', 'd')


# Mini-diagram: tuple structure


sizes = ('S', 'M', 'L')

# Index:       0     1     2
# Value:      'S'   'M'   'L'

# Tuples use the same indexing rules as lists.
`````

`````python
# 06.42 One-Element Tuples  

# A tuple with one element **must** include a trailing comma.

one = ('A',)     # This is a tuple
not_tuple = ('A') # This is just the string 'A'

# Python interprets parentheses alone as grouping, not as tuple creation.
`````

`````python
# 06.43 Accessing Tuple Elements
# Tuple indexing works exactly like list indexing:

colors = ('red', 'green', 'blue')

print(colors[0])    # ('red')           → created but not saved
print(colors[-1])   # ('blue')          → created but not saved
print(colors[1:3])  # ('green', 'blue)  → created but not saved

new_colors = colors[1:3]  # saved version
print("\n", new_colors)
print(colors)             # original intact


# - Slicing returns a new tuple object:
# - The original tuple remains unchanged.
# - You can assign the slice to a variable to keep it (just like a list slice).

# Because tuples are immutable:
# - You *cannot* modify the original tuple  
# - You *cannot* modify the slice  
# - You *cannot* use slice assignment  
# - The new tuple is completely isolated and safe  
`````

`````python
# 06.44 Core Concept: Tuple Immutability

# Tuples cannot:

# - append items  
# - remove items  
# - insert items  
# - reassign items  

# Any attempt to modify them will lead to a TypeError.


t = (1, 2, 3)

# t[1] = 999       # TypeError: 'tuple' object does not support item assignment
# t.append(4)      # AttributeError
`````

`````python
# --------------------------------------------------------------
# 🔧 Helper: Tuple Inspector (Index Map + Mutability Checker)
# --------------------------------------------------------------

def show_tuple_info(tpl, show_negative=True, show_types=True):
    """
    Display a structured index map for a tuple, including:
      - positive and negative indices
      - value at each index
      - type of each element
      - mutability warning (e.g., tuple containing lists)
    
    Parameters
    ----------
    tpl : tuple
        The tuple to inspect.
    show_negative : bool, default True
        Whether to show negative indices.
    show_types : bool, default True
        Whether to show type information for each element.
    
    Usage
    -----
    show_tuple_info((1, 2, 3))
    show_tuple_info((1, [2, 3], (4, 5)))
    show_tuple_info(("a", "b", "c"), show_negative=False)
    """

    # ----------------------------------------------------------
    # 1. Validate input
    # ----------------------------------------------------------
    if not isinstance(tpl, tuple):
        raise TypeError(
            f"show_tuple_info expects a tuple, got {type(tpl).__name__}"
        )

    n = len(tpl)
    print(f"Tuple length: {n}")
    print()

    # ----------------------------------------------------------
    # 2. Build header
    # ----------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    header_cols.append("mutable?")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += f"{'value':>12} {'mutable?':>10}"

    print(header)
    print("-" * max(len(header), 40))

    # ----------------------------------------------------------
    # 3. Detect mutability of each element
    # ----------------------------------------------------------
    def is_mutable(obj):
        """Return True if object is a known mutable Python type."""
        return isinstance(obj, (list, dict, set, bytearray))

    # ----------------------------------------------------------
    # 4. Print each element with indices and mutability flag
    # ----------------------------------------------------------
    for i, val in enumerate(tpl):
        neg = i - n if show_negative else None
        type_name = type(val).__name__ if show_types else ""
        mutable_flag = "yes" if is_mutable(val) else "no"

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{repr(val):>12} {mutable_flag:>10}"

        print(row)

    # ----------------------------------------------------------
    # 5. Summary warnings (very helpful in practice)
    # ----------------------------------------------------------
    print("\nSummary:")
    if any(isinstance(v, list) for v in tpl):
        print("⚠ Warning: This tuple contains one or more lists.")
        print("  → The tuple is immutable, but the inner lists ARE mutable.")
        print("  → Changing an inner list will change all references to it.\n")
    else:
        print("✓ No mutable inner objects detected. Tuple is fully immutable.\n")


# ------------------------------
# Example:
# ------------------------------
show_tuple_info((1, 2, (3, 4)))
show_tuple_info((1, [2, 3], 4))
`````

# 06.51 Tuples - Iterations (Part 4)
## Tuple Iterations Stop at the Shorter Variable. (Extra Elements are Silently Ignored)


### 🟦 Best Solution when 'len()' is Available:

```python
def same_length(a, b):
    return len(a) == len(b)
```

### 🟦 'zip()'-Based Iterator Safety Check. This avoids relying on len() and works for general iterables:

```python
from itertools import zip_longest

def same_length(a, b):
    sentinel = object()
    for x, y in zip_longest(a, b, fillvalue=sentinel):
        if x is sentinel or y is sentinel:
            return False
    return True
```
* ***Note: zip() stops at the shortest iterable.***
* ***zip_longest() continues until all iterables are exhausted, using a sentinel to detect mismatched length.***

# 🟦 Helper Box — Slicing: Lists vs. Tuples  
Understanding how slicing behaves for **mutable** vs **immutable** sequence types is essential.  
This helper clarifies exactly what happens under the hood — and why tuple slicing is so safe and predictable.

---

## ❗ First, a universal rule (often overlooked)
> **Slicing always *creates* a new sequence object, but it does *not* save it unless you assign it to a variable.**

```python
numbers = [10, 20, 30, 40]
numbers[1:3]        # slice created, used, discarded
subset = numbers[1:3]  # slice saved
```

The same is true for tuples.

This is important: slicing alone never mutates anything — it just produces a temporary result unless you store it.

---

## ✅ 1. Slicing a List  
### What happens?

```python
lst = [10, 20, 30, 40]
slice_lst = lst[1:3]     # [20, 30]
```

- **Slicing always creates a new list object.**
- The original list is unchanged.
- Because lists are **mutable**, you can modify the slice:

```python
slice_lst[0] = 999      # [999, 30]
# lst remains [10, 20, 30, 40] — completely unaffected
```

### Key property  
Even though slicing is safe, **slice assignment** can modify the original list:

```python
lst[1:3] = [777, 888]   # mutates original list
```

This is a powerful feature — but also a source of bugs if used carelessly.

---

## ✅ 2. Slicing a Tuple  
### What happens?

```python
colors = ('red', 'green', 'blue')
subset = colors[1:3]      # ('green', 'blue') → created but not saved

new_colors = colors[1:3]  # saved version

```

- **Slicing creates a new tuple object.**
- The original tuple remains unchanged.
- You can assign the slice to a variable to keep it (just like a list slice).

### But the critical difference:
Because tuples are **immutable**:

- You *cannot* modify the original tuple  
- You *cannot* modify the slice  
- You *cannot* use slice assignment  
- The new tuple is completely isolated and safe  

Slicing tuples is simpler and safer because there is *no* possibility of mutation anywhere.

---

## 🧠 3. Why the Behavior Is Different?

| Operation                    | Lists (Mutable)                      | Tuples (Immutable)                     |
|-----------------------------|---------------------------------------|----------------------------------------|
| Slicing                     | Creates a **new list**                | Creates a **new tuple object**         |
| Assigning back to a slice  | Allowed (mutates original)            | Not allowed (TypeError)                |
| Modifying the slice itself  | Allowed                                | Not allowed                            |
| Risk of unintended mutation | Medium                                 | Zero                                   |

---

## 🏁 4. The Key Concept in One Sentence  
> **Slicing always returns a new sequence object — but only lists allow slice assignment that can mutate the original. Tuples never allow mutation, so slicing is always safe.**

---

## ⭐ 5. The Mental Model (recommended for mastery)

Think of slicing as:

### **“Copying a portion of the sequence into a new container.”**

- For lists → a new *list container*  
- For tuples → a new *tuple object*  

And because tuples cannot change, this new object is **fully independent** and **risk-free**.

---

## ✔ Quick Verification

```python
t = (1, 2, 3, 4)
s = t[1:3]

print(s)       # (2, 3)
print(t)       # (1, 2, 3, 4)
```

Both remain intact. Nothing can mutate either one.

---

## 📌 Final takeaway  
Whether slicing a list or tuple:

- It **never** changes the original sequence  
- It **always** produces a new object  
- It **must be assigned** if you want to keep it  
- Only lists support slice assignment  
- Tuples are 100% mutation-proof  

---

End of Helper Box.


# Chapter 06: Sequence Data Types

---

# 06.60 Overview

A **range** is a very special type of sequence in Python.  
Unlike lists or tuples, a range does **not** store all values in memory.  
Instead, it **generates** values on demand using a start, stop, and step.

Ranges are:

- memory-efficient  
- fast  
- commonly used in loops (next chapter!)  
- ideal for generating clean sequences of integers  
- invaluable for indexing, batching, and slicing  

---

# 🟦 Helper Box — Universal Range Index Map (`show_range_index_map`)

This helper is the **range-specific cousin** of your `show_list_index_map`.  
It helps you *see* how `range(start, stop, step)` expands into actual values, including:

- forward ranges  
- backward (negative-step) ranges  
- how indices and negative indices line up with values  

This is especially helpful for:

- understanding `stop` being **excluded**  
- understanding **negative steps**  
- mentally mapping `range()` to list indices  

---

# 06.69 Summary (Part 5 — Ranges)

- `range()` creates a memory-efficient sequence of integers.  
- Ranges are defined by start, stop, and step values.  
- The stop value is **non-inclusive**, just like slicing.  
- Converting to `list()` or `tuple()` reveals the actual generated values.
- Although, any range object can be converted into any other sequence type that accepts an iterable.
- Ranges are ideal for loops, counting, indexing, batching, and generating IDs.  
- Negative step values allow backward ranges.  
- Ranges prepare you directly for Chapter 07 (Loops), where they are used constantly.


`````python
# 06.61 Creating Ranges
# Ranges are created using the built-in `range()` function.

# The full form is: range(start, stop, step)

# - **start** (optional): where the sequence begins  
# - **stop**: where the sequence ends (non-inclusive)  
# - **step** (optional): how much to increment each time  


# Basic range: 0 to 4
r1 = range(5)

# Range with start and stop
r2 = range(2, 7)

# Range with start, stop, and step
r3 = range(0, 20, 5)

print(list(r1))   # [0, 1, 2, 3, 4]
print(list(r2))   # [2, 3, 4, 5, 6]
print(list(r3))   # [0, 5, 10, 15]

# ASCII mini-diagram

# range(2, 7)
# start → 2
# stop  → 7   (not included)
# values → [2, 3, 4, 5, 6]

# This mirrors list slicing behavior, where the stop value is also non-inclusive.
`````

`````python
# ---------------------------------------------------------
# 06.62 Ranges Do Not Directly Display Their Values
# ---------------------------------------------------------

# When printed, a range shows its definition, not the actual integers:
print(range(5), end="\n\n")  # prints object parameters only,"range(0, 5)"


# --------------------------------------------------------------------
# To retrieve and see the actual values, convert to a list or a tuple:
# --------------------------------------------------------------------

print(list(range(5)))    # [0, 1, 2, 3, 4]
print(tuple(range(5)))   # (0, 1, 2, 3, 4)

# -------------------------------------------------------------------------------
# Visualization example: show the index positions of a string
# This is purely for demonstration; not common in real code to turn into a tuple
# -------------------------------------------------------------------------------

print("\n", tuple(range(len('Hello'))))

# Breakdown: generating all valid indices of the string, then turning them into a tuple.
# len('Hello') → 5
# range(5) → range(0, 5)
# tuple(range(5)) → (0, 1, 2, 3, 4)
# print() to display

print("\n", list(range(len('Hello'))))
`````

`````python
# 06.63 Negative Steps and Reversed Ranges

r = range(10, 0, -2)
print(list(r))
# [10, 8, 6, 4, 2]


# ASCII diagram

range(10, 0, -2)
# start = 10
# stop = 0  (excluded)
# step = -2
# values → [10, 8, 6, 4, 2]

# Backward ranges are extremely helpful when:

# - stepping backward through indices  
# - reverse-iterating through sequences  
# - building countdown timers  
# - analyzing time-based data in reverse order
`````

`````python
# 🔧 Helper: `show_range_index_map(rng, show_negative=True)`


def show_range_index_map(rng, show_negative=True):
    """
    Print a table of indices and values for a Python range object.

    Parameters
    ----------
    rng : range
        The range object to inspect.
    show_negative : bool, default True
        If True, show the negative index for each element
        (as if the range were a sequence like a list or tuple).

    Usage
    -----
    show_range_index_map(range(5))
    show_range_index_map(range(10, 0, -2))
    """
    # ---------------------------------------------------------
    # 1. Validate input type
    # ---------------------------------------------------------
    if not isinstance(rng, range):
        raise TypeError(
            f"show_range_index_map expects a range, got {type(rng).__name__}"
        )

    # ---------------------------------------------------------
    # 2. Basic range properties
    # ---------------------------------------------------------
    n = len(rng)
    start = rng.start
    stop = rng.stop
    step = rng.step

    print(f"Range: range({start}, {stop}, {step})")
    print(f"Length: {n}")
    print()

    if n == 0:
        print("This range is empty (no values generated).")
        return

    # ---------------------------------------------------------
    # 3. Build header
    # ---------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    header_cols.append("value")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "value" in header_cols:
        header += f"{'value':>6}"

    print(header)
    print("-" * max(len(header), 20))

    # ---------------------------------------------------------
    # 4. Iterate over the range and print index/value mapping
    # ---------------------------------------------------------
    for i, val in enumerate(rng):
        # Negative index follows standard Python convention:
        # last element → -1, second to last → -2, etc.
        neg_index = i - n if show_negative else None

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        row += f"{val:>6}"
        print(row)


show_range_index_map(range(10, 0, -2))
show_range_index_map(range(20))               # forward range
show_range_index_map(range(50, 20, -5))       # backward range
show_range_index_map(range(6), show_negative=False)

example_range = range(10, 0, -2)
show_range_index_map(example_range)
`````

# 06.61 Creating Ranges

Ranges are created using the built-in `range()` function.

The full form is:

```
range(start, stop, step)
```

### The rules:

- **start** (optional): where the sequence begins  
- **stop**: where the sequence ends (non-inclusive)  
- **step** (optional): how much to increment each time  

```python
# Basic range: 0 to 4
r1 = range(5)

# Range with start and stop
r2 = range(2, 7)

# Range with start, stop, and step
r3 = range(0, 20, 5)

print(list(r1))   # [0, 1, 2, 3, 4]
print(list(r2))   # [2, 3, 4, 5, 6]
print(list(r3))   # [0, 5, 10, 15]
```

### ASCII mini-diagram

```
range(2, 7)
start → 2
stop  → 7   (not included)
values → [2, 3, 4, 5, 6]
```

This mirrors list slicing behavior, where the stop value is also non-inclusive.

---

# 06.62 Ranges Do Not Directly Display Their Values

When printed, a range shows its definition, not its contents:

```python
print(range(5))
# range(0, 5)
```

To see the actual values, convert to:

- list  
- tuple  

```python
print(list(range(5)))   # [0, 1, 2, 3, 4]
print(tuple(range(3)))  # (0, 1, 2)
```

This design saves memory and speeds up looping.

---

# 06.63 Negative Steps and Reversed Ranges

Ranges can count backwards.

```python
r = range(10, 0, -2)
print(list(r))
# [10, 8, 6, 4, 2]
```

### ASCII diagram

```
range(10, 0, -2)
start = 10
stop = 0  (excluded)
step = -2
values → [10, 8, 6, 4, 2]
```

Backward ranges are extremely helpful when:

- stepping backward through indices  
- reverse-iterating through sequences  
- building countdown timers  
- analyzing time-based data in reverse order

# 🟦 Helper Box — Universal Range Index Map (`show_range_index_map`)

This helper is the **range-specific cousin** of your `show_list_index_map`.  
It helps you *see* how `range(start, stop, step)` expands into actual values, including:

- forward ranges  
- backward (negative-step) ranges  
- how indices and negative indices line up with values  

This is especially helpful for:

- understanding `stop` being **excluded**  
- understanding **negative steps**  
- mentally mapping `range()` to list indices  

---

## 🔧 Helper: `show_range_index_map(rng, show_negative=True)`

```python
def show_range_index_map(rng, show_negative=True):
    """
    Print a table of indices and values for a Python range object.

    Parameters
    ----------
    rng : range
        The range object to inspect.
    show_negative : bool, default True
        If True, show the negative index for each element
        (as if the range were a sequence like a list or tuple).

    Usage
    -----
    show_range_index_map(range(5))
    show_range_index_map(range(10, 0, -2))
    """
    # ---------------------------------------------------------
    # 1. Validate input type
    # ---------------------------------------------------------
    if not isinstance(rng, range):
        raise TypeError(
            f"show_range_index_map expects a range, got {type(rng).__name__}"
        )

    # ---------------------------------------------------------
    # 2. Basic range properties
    # ---------------------------------------------------------
    n = len(rng)
    start = rng.start
    stop = rng.stop
    step = rng.step

    print(f"Range: range({start}, {stop}, {step})")
    print(f"Length: {n}")
    print()

    if n == 0:
        print("This range is empty (no values generated).")
        return

    # ---------------------------------------------------------
    # 3. Build header
    # ---------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    header_cols.append("value")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "value" in header_cols:
        header += f"{'value':>6}"

    print(header)
    print("-" * max(len(header), 20))

    # ---------------------------------------------------------
    # 4. Iterate over the range and print index/value mapping
    # ---------------------------------------------------------
    for i, val in enumerate(rng):
        # Negative index follows standard Python convention:
        # last element → -1, second to last → -2, etc.
        neg_index = i - n if show_negative else None

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        row += f"{val:>6}"
        print(row)
```

---

## 📌 Example 1 — Simple Forward Range

```python
r = range(0, 10, 2)
show_range_index_map(r)
```

Expected output (conceptually):

```
Range: range(0, 10, 2)
Length: 5

idx   neg  value
----------------
  0   -5      0
  1   -4      2
  2   -3      4
  3   -2      6
  4   -1      8
```

This shows:

- the **indices** (`0` to `4`)  
- the corresponding **values** (`0, 2, 4, 6, 8`)  
- the **negative indices** (`-5` to `-1`)  

Exactly as if you had:

```python
list(r) == [0, 2, 4, 6, 8]
```

---

## 📌 Example 2 — Backward (Negative Step) Range

```python
r = range(10, 0, -2)
show_range_index_map(r)
```

Conceptual output:

```
Range: range(10, 0, -2)
Length: 5

idx   neg  value
----------------
  0   -5     10
  1   -4      8
  2   -3      6
  3   -2      4
  4   -1      2
```

Key points:

- `start = 10`  
- `stop = 0` is **excluded**  
- `step = -2` walks backward  
- Values are `[10, 8, 6, 4, 2]`  
- Indices still run `0..4`, negative indices `-5..-1`  

This helper makes **negative-step ranges** much easier to reason about.

## 🧠 Why This Helper Matters

- Ranges are **lightweight, lazy sequences**  
- They are used everywhere in loops, indexing, and time-based iteration  
- Off-by-one errors are very common without a clear mental model

`show_range_index_map` lets you:

# Chapter 07 — Loops  
## Section 7.01 — Loop Basics  

### Introduction
In Python, a **loop** is a controlled repetition structure: a block of code that executes multiple times until a stopping condition is met.  
For data analytics, loops let you process sequences, clean datasets, transform values, and automate iterative logic.

Python supports two primary loop types:

1. **For Loops**  
   Run a ***specific number of times***, typically once per element in an iterable such as a list, tuple, string, or range object.  
   Use a for loop when the iteration count is known in advance or when you want to examine or transform each element of a dataset.  
   For loops may iterate:
   - **directly over values** (item-based looping),  
   - **over indices** using `range()` when positional control or alignment across multiple sequences is required, or  
   - **over index–value pairs** using `enumerate()` when both position and clarity are needed.  
   *Run a specified number of times.*

2. **While Loops**  
   Run ***as long as a logical condition remains True***.  
   Use a while loop when you do **not** know in advance how many iterations are needed and the loop must respond to changing state.  
   *Run until a logical condition is met.*

This chapter focuses on building mastery for both loop types, integrating your **Helper Toolkit** to visualize execution steps, debug iterative logic, and strengthen intuition for real analytics workflows.

---

# 🟦 Helper Box — Core Loop Visualizers  
These helpers provide foundational insight into how loops actually *iterate* and how internal state evolves during execution.  
They serve the same purpose for Chapter 7 that the **Index Map** and **Slice Visualizer** served for Chapter 6.

Together, they make abstract loop mechanics concrete, predictable, and fully observable.

---

# Preview — Loop Basics

- A **loop** repeats instructions until a **count-based condition** (for loop) or a **logical condition** (while loop) signals completion.  
- **For loops** are used when the number of iterations is known or when the task is “one pass per element.”  
  They may iterate directly over values, over **indices** when positional control is required, or over **index–value pairs** using `enumerate()` for improved clarity and safety.  
- **While loops** provide greater flexibility but must be carefully designed to avoid infinite loops.  
- This section introduces the core loop helpers:  
  - **`loop_trace_for()`** — traces item-by-item for-loop iteration and execution order  
  - **`enumerate_visualizer()`** — makes index–value pairing explicit when using `enumerate()`  
  - **`safe_while_trace()`** — safely traces while-loop state evolution and prevents infinite loops
  - **``while_timeline()``** - advanced helper logs every state value for later visualization or tabular analysis 
  - **`deep_loop_state_visualizer()`** — traces each internal step of a list-processing loop, including variable assignment, calculation execution, each `.append()` operation, and the evolving state of the accumulator container
  - **`validate_alignment`** - Validate iterable alignment before looping. (Length alignment is required for index-based looping)
  - **`sync_lists()`** — synchronized iteration across multiple equal-length iterables 
- Understanding iteration mechanics — including when and how position matters — is essential for dataset traversal, data cleaning, transformation workflows, simulation modeling, and algorithmic problem-solving.


# Python Loop & List Access Reference Table  
(When working with one list vs. multiple parallel lists)

| Technique / Style                              | Syntax Example                                      | What you get / What it gives you                    | When to use it (best situations)                                                                 | Common in our project? | Notes           |
|------------------------------------------------|-----------------------------------------------------|-----------------------------------------------------|--------------------------------------------------------------------------------------------------|------------------------|-----------------|
| Simple `for` over values                       | `for item in my_list:`                              | `item` = each value directly                        | You only need the elements themselves (no position needed)                                       | Yes (simple cases)     | Cleanest & most readable when index isn't needed |
| `range(len())` for index only                  | `for i in range(len(my_list)):`                     | `i` = 0, 1, 2, ..., len-1                           | Need index to access `my_list[i]`, or multiple lists like `listA[i]`, `listB[i]`                | Very common            | Classic beginner way; works but a bit old-school |
| `enumerate()` (index + value)                  | `for i, item in enumerate(my_list):`                | `i` = index, `item` = value                         | Need both position and value; cleaner than `range(len())`                                        | Recommended upgrade    | Most Pythonic when you want index + value |
| `zip()` to combine parallel lists              | `for a, b, c in zip(listA, listB, listC):`          | `a`, `b`, `c` = values from same position in each list | You have 2–5+ lists of same length and want to process them together without indexes            | Very useful alternative| Stops at shortest list; no index unless combined with enumerate |
| `enumerate()` + `zip()` together               | `for i, (a, b) in enumerate(zip(listA, listB)):`    | `i` = index, `a` & `b` = paired values              | Need index AND values from multiple parallel lists                                               | Great for our project  | Cleanest modern way for parallel lists + index |
| `.index(value)` method                         | `position = my_list.index("asia")`                  | Returns the first index where value appears         | You know a value and want its position (single lookup)                                           | Sometimes              | Only finds first match; raises ValueError if not found |
| List comprehension (not a loop, but related)   | `[x*2 for x in my_list]`                            | New list with transformed values                    | Quick transformations or filtering without full loop                                             | Not yet in project     | Compact; good once comfortable with loops |
| `for i in range(start, stop, step)`            | `for i in range(0, len(my_list), 2):`               | Custom index stepping (every other, etc.)           | Need to skip items or start/end at specific points                                               | Rare                   | Flexible but less common for basic parallel access |

Quick decision tree for choosing:
1. Do I only need the values?  
   → Use `for item in my_list:`
2. Do I need the position (index)?  
   → Use `enumerate()` (best) or `range(len())` (classic)
3. Do I have multiple lists that line up (country, pop, continent, etc.)?  
   → Use `zip()` if you don't need index  
   → Use `enumerate(zip(...))` if you do need index
4. Do I just need to find where one specific value is?  
   → Use `.index(value)`

Examples from DSCI 503 Project 01, problem # 8 context:

```python
# Classic way (see problem #8)
for i in range(len(continent_2018)):
    if continent_2018[i] == "asia":
        temp_pop += population_2018[i]

# More modern / readable alternatives
# Option A: enumerate
for i, continent in enumerate(continent_2018):
    if continent == "asia":
        temp_pop += population_2018[i]

# Option B: zip (no index needed)
for continent, pop in zip(continent_2018, population_2018):
    if continent == "asia":
        temp_pop += pop
```

`````python
# ------------------------------------------------------------------
# Understanding Loops Through Examples
# Example 1 — Rewriting a repetitive calculation using loops  
# ------------------------------------------------------------------


# Below is the non-loop approach: writing a separate line for each list element.

exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

# Inefficient repetition (not scalable)
euro_list = [
    round(usd_list[0] * exchange_rate, 2),
    round(usd_list[1] * exchange_rate, 2),
    round(usd_list[2] * exchange_rate, 2),
    round(usd_list[3] * exchange_rate, 2),
    round(usd_list[4] * exchange_rate, 2),
]

euro_list

# Problem if you have hundreds or thousands of prices, manually writing repeated expressions becomes impossible.


# ------------------------------------------------------------------
# Understanding Loops Through Examples
# Example 2 — Applying a for loop
# ------------------------------------------------------------------

# Generic syntax - `for` indicates for loop
iterable = "Dawn"
for i in iterable:
     print(i)

# Given Data:
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

euro_list = []                                    # Start with an empty list container

for price in usd_list:                            # for loop: element in iterable
    euro_value = round(price * exchange_rate, 2)  # assign variable: each element from the iterable is * by the loop-invariant scalar exchange_rate
    euro_list.append(euro_value)                  # .append() empty list container with each converted price
    
print()
print(euro_list, end="\n\n")


# This loop is conceptually equivalent to five function calls, one per list element.  
# But Python handles all iteration mechanics for you.
`````

`````python
# -------------------------------------------------------------
# Looping Mechanics
# How to use a single loop to derive multiple aligned outputs from one input sequence.
# -------------------------------------------------------------
# item-based for loops (no indices yet)
# deterministic iteration (known list length)
# accumulator lists
# stepwise transformation
# parallel output lists that stay aligned by construction

subtotals = [15.98, 899.97, 799.97, 117.96, 5.99, 599.99, 24.99, 1799.94, 99.99]
taxes = []
totals = []

for subtotal in subtotals:
    tax = round(subtotal * 0.08, 2)
    total = round(subtotal + tax, 2)
    taxes.append(tax)
    totals.append(total)

print(taxes)
print(totals)
`````

# Chapter 07 — Loops  
## Section 7.02 — For Loops  

### Overview
A **for loop** iterates over an *iterable* (list, tuple, string, set, dictionary, or range).  
Python automatically handles indexing and sequence progression, allowing analysts to focus on transformation logic rather than iteration mechanics.

Use a for loop when:

- You know **how many iterations** you need.  
- You are iterating directly over each element in a dataset.  
- You want clear, predictable, index-safe iteration.  
- You want to avoid accidental infinite loops.

Syntax describes the standard pattern:

```
for item in iterable:
    do_something_with(item)
```

Python internally executes:

1. `item = iterable[0]`  
2. `item = iterable[1]`  
3. `item = iterable[2]`  
n. until the iterable is exhausted.

---

## 1. Looping Over Items  
This is the **most Pythonic** and most common pattern in data analytics.

### Example — Print each letter in a string  

```python
word = "Hello"

for letter in word:
    print(letter)
```

Internally this becomes:

- `letter = 'H'`  
- `letter = 'e'`  
- `letter = 'l'`  
- `letter = 'l'`  
- `letter = 'o'`


`````python
# -------------------------------------
# 1. Looping Over Items in a String
# -------------------------------------
# This is the most Pythonic and most common pattern in data analytics.
# Use when:
# - you only need the value
# - position does not matter

# Item-Based Loop Pattern:

items = "Hello"
for i in items:
    print(i)
print()


# 🟦 Helper Expansion — `enumerate_items()`

def enumerate_items(iterable):
    """
    Diagnostic helper: prints each item as Python would assign it in a for loop.
    Mirrors the step-by-step iteration shown in Maven pages 7–9.
    """
    print("TRACE: Looping Over Items\n--------------------------")
    for idx, item in enumerate(iterable):
        print(f"Iteration {idx}: item = {item!r}")
        

# Example usage:
enumerate_items("Hello")
`````

`````python
# -------------------------------------
# 1. Looping Over Items in a range()
# -------------------------------------

# Polished version
for n in range(1, 11):
    print(f"The square of {n} is {n**2}.")
print()

# More explicit version
for n in range(1, 11):
    square = n**2
    message = f"The square of {n} is {square}."
    print(message)


# 🟦 Helper Expansion — `enumerate_items()`

def enumerate_items(iterable):
    """
    Diagnostic helper: prints each item as Python would assign it in a for loop.
    Mirrors the step-by-step iteration shown in Maven pages 7–9.
    """
    print("TRACE: Looping Over Items\n--------------------------")
    for idx, item in enumerate(iterable):
        print(f"Iteration {idx}: item = {item!r}")
        

# Example usage:
enumerate_items(range(1, 11))


# 🟦 Helper Expansion — `range_trace()`

def range_trace(start, stop=None, step=1):
    """
    Diagnostic helper: prints the values produced by range()
    and how Python assigns them during a for-loop.
    
    Mirrors how: for i in range(start, stop, step)
    """
    if stop is None:
        start, stop = 0, start  # allow range(stop) signature
    
    print("TRACE: range() Iteration\n--------------------------")
    print(f"range({start}, {stop}, {step}) produces:")
    
    for idx, value in enumerate(range(start, stop, step)):
        print(f"Iteration {idx}: i = {value!r}")

# Example Usage:
range_trace(1, 11)


# 🟦 Helper Expansion — `range_items()`

def range_items(start, stop=None, step=1):
    """
    Diagnostic helper: treats range() like any other iterable
    and traces what the loop variable actually receives.
    """
    if stop is None:
        start, stop = 0, start
    
    print("TRACE: Looping Over range()\n-----------------------------")
    
    for idx, item in enumerate(range(start, stop, step)):
        print(f"Iteration {idx}: item = {item!r}")

# Example Usage:
range_items(1, 11)
`````

`````python
# -------------------------------------
`````
`````python
# Sum the first 10 positive integers
# Minimalist, hidden version
# -------------------------------------

total = 0
for i in range(1,10):
    total += i**2

print(total)

# ----------------------------------------------------------
# Accumulation Pattern: Running Total (Diagnostic Version)
# Demonstrates how += carries state forward across iterations
# ----------------------------------------------------------

total = 0

print("TRACE: Accumulating Squares\n-----------------------------")

for i in range(1, 10):   # 1 → 9 is enough to see the full pattern
    square = i**2                 # transformation step
    total += square               # accumulation step (state update)

    print(
        f"Iteration {i}: "
        f"i = {i}, "
        f"i^2 = {square}, "
        f"running total = {total}"
    )

print("\nFinal total:", total)
print()

# ------------------------------------
# Alternative Accumulation Pattern
# ------------------------------------

total = 0

print("TRACE: Accumulator State\n--------------------------")

for i in range(1, 10):
    square = i**2
    new_total = total + square
    print(f"Iteration {i}: total({total}) + {square} → {new_total}")
    total = new_total

print("\nFinal total:", total)
print()


# ------------------------
# Stripped Version
# ------------------------

total = 0

for i in range(1, 10):
    total += i**2
    print(f"After adding {i**2:>3}, total = {total}")

print("\nFinal total:", total)

`````

`````python
# ---------------------------------------
# 🟦 Helper Expansion - `loop_trace_for()`
# ---------------------------------------

# Given data:
exchange_rate = 0.88                          # exchange_rate → loop-invariant scalar parameter
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]  # usd_list      → iterable dataset

# 🟦 Helper for visualization
def loop_trace_for(iterable):
    print("TRACE: For Loop Execution\n-------------------------")
    for iteration, item in enumerate(iterable):    # for loop: element in dataset, enumerate(usd_list) returns pairs (index, value)
        print(f"Iteration {iteration}: item = {item!r}")

# Call `loop_trace_for()` Helper
loop_trace_for(usd_list)

# Actual loop performing the conversion
euro_list = []                                     # create empty list container
for price in usd_list:                             # for loop: element in iterable
    euro_value = round(price * exchange_rate, 2)   # assign variable: each element from the iterable is * by the loop-invariant scalar exchange_rate
    euro_list.append(euro_value)                   # .append() empty list container

print("\nConverted EUR prices:", euro_list)
`````

`````python
# --------------------------------------------------------
# 🟦 Helper Expansion - `deep_loop_state_visualiser(,)`
# --------------------------------------------------------

# Given data:
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

# 🟦 Helper for visualization
def deep_loop_state_visualizer(usd_list, exchange_rate):
                                                       # usd_list      → iterable dataset
                                                       # exchange_rate → loop-invariant scalar parameter
    """
    Shows detailed iteration state for a currency conversion loop.
    Reveals:
      - iteration number
      - current input value
      - computed euro_value
      - euro_list after each step
    """
    euro_list = []                                    # create empty list container
    
    print("DEEP LOOP STATE VISUALIZER\n--------------------------")
    
    for iteration, price in enumerate(usd_list):      # for loop: element in dataset, enumerate(usd_list) returns pairs (index, value)
        euro_value = round(price * exchange_rate, 2)  # Assign variable: each element from the iterable is * by the loop-invariant scalar exchange_rate
        euro_list.append(euro_value)                  # Append this iteration’s computed euro_value to the accumulator list euro_list
        
        print(f"Iteration {iteration}:")              # iteration index from enumerate()
        print(f"  price       = {price!r}")           # value retrieved from the iterable at this iteration
        print(f"  euro_value  = {euro_value!r}")      # variable holding the results that change with each iteration
        print(f"  euro_list   = {euro_list}\n")       # display the state of the accumulator list after each iteration.
    print("Final Output:", euro_list)
    return euro_list

# Call `deep_loop_state_visualizer()` Helper
deep_loop_state_visualizer(usd_list, exchange_rate)
`````

`````python
# ---------------------------------------------
# Indexs-based: Looping Over Multiple Iterables  
# ----------------------------------------------
# When two lists share a one-to-one alignment (e.g., item ↔ price), indexing allows synchronized access:

# Example — Print item and its euro price
# This produces 5 synchronized pairs

items = ["Snowboard", "Boots", "Helmet", "Goggles", "Bindings"]
euro_prices = [5.27, 8.79, 17.59, 21.99, 87.99]

for i in range(len(items)):
    print(items[i], euro_prices[i])


# 🟦 Helper — `sync_lists()`
# Synchronized iteration across multiple equal-length iterables

def sync_lists(*lists):
    """
    Prints synchronized items across multiple lists of equal length.
    Excellent for diagnostics in multi-iterable looping.
    """
    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
    length = len(lists[0])
    for iterable in lists:
        if len(iterable) != length:
            raise ValueError("All lists must have the same length for synchronized iteration.")
    
    for idx in range(length):
        row = [lst[idx] for lst in lists]
        print(f"Index {idx}: {row}")

# Usage:
sync_lists(items, euro_prices)
`````

`````python
# ----------------------------------------------------
# Exercise: Index-based looping — step-by-step 
# Why position matters ⚠️
# ----------------------------------------------------
# ⚠️ Caution:
# Index-based loops rely entirely on positional alignment.
# If related lists become misaligned, results may be silently incorrect.
# Index-Based Loop (Positional Control, Higher Risk)

for i in range(len(items)):   # i for indices
    print(i, items[i])

# Use when:
# - you must reference multiple lists by position
# - you need write-access to a list by index


items = ["Snowboard", "Boots", "Helmet"]
prices = [499, 299, 99]

# Intentionally misaligned ratings
ratings = [4, 5, 4]              # should be [5, 4, 4]

print("Index-based loop output (ratings misaligned):\n")

for i in range(len(items)):
    print(items[i], prices[i], ratings[i])

print("\nNote:")
print("Index-based loops assume all lists are aligned by position.")
print("If one list is transposed, results will be silently incorrect.")

# ----------------------------------------------------
# 🟦 Helper - Validate alignment before looping
# ----------------------------------------------------
# Length alignment is required for index-based looping.

if not (len(items) == len(prices) == len(ratings)):
    raise ValueError("Lists are not aligned in length")

print("Lists are aligned by length. Safe to proceed.")
print()

# -------------------------------------------------------------------------
#  🧠 Safer iteration using zip()
# zip() prevents index misuse; but it will still silently accept wrong order
# --------------------------------------------------------------------------

for item, price, rating in zip(items, prices, ratings):
    print(item, price, rating)

# -----------------------------------
# ⚠️ Avoiding Misalignment
# -----------------------------------
# Option 1: Validate assumptions
# Option 2: Use zip() for safer iteration
# Option 3: Use structured data (preferred)

# 3a. Parallel lists (must remain aligned)
items = ["Snowboard", "Boots", "Helmet"]
prices = [499, 299, 99]
ratings = [5, 4, 4]

for i in range(len(items)):
    print(items[i], prices[i], ratings[i])
print()

# Use tuples (alignment enforced by structure)
products = [
    ("Snowboard", 499, 5),
    ("Boots", 299, 4),
    ("Helmet", 99, 4)
]

for item, price, rating in products:
    print(item, price, rating)
print()

# Use dictionaries (most explicit and flexible)
products = [
    {"item": "Snowboard", "price": 499, "rating": 5},
    {"item": "Boots", "price": 299, "rating": 4},
    {"item": "Helmet", "price": 99, "rating": 4},
]

for product in products:
    print(product["item"], product["price"], product["rating"])
`````

# Summary: Chapter 07 — Loops  
## Section 7.02 — For Loops  

### Summary
A **for loop** iterates over an *iterable* (list, tuple, string, set, dictionary, or range).  
Python automatically handles indexing and sequence progression, allowing analysts to focus on transformation logic rather than iteration mechanics.

Use a for loop when:

- You know **how many iterations** you need.  
- You are iterating directly over each element in a dataset.  
- You want clear, predictable, index-safe iteration.  
- You want to avoid accidental infinite loops.

Maven (pp. 5–6) describes the standard pattern:

```
for item in iterable:
    do_something_with(item)
```

Python internally executes:

1. `item = iterable[0]`  
2. `item = iterable[1]`  
3. `item = iterable[2]`  
n. until the iterable is exhausted.

We now expand this with analytics-ready helpers, diagnostics, and deeper reasoning.

---

# 1. Looping Over Items  
This is the **most Pythonic** and most common pattern in data analytics.

### Example — Print each letter in a string  
*(Maven pages 7–8)*

```python
word = "Maven"

for letter in word:
    print(letter)
```

Internally this becomes:

- `letter = 'M'`  
- `letter = 'a'`  
- `letter = 'v'`  
- `letter = 'e'`  
- `letter = 'n'`

### Helper Expansion — `enumerate_items()`

```python
def enumerate_items(iterable):
    """
    Diagnostic helper: prints each item as Python would assign it in a for loop.
    """
    print("TRACE: Looping Over Items\n--------------------------")
    for idx, item in enumerate(iterable):
        print(f"Iteration {idx}: item = {item!r}")
```

Example usage:

```python
enumerate_items("Maven")
```

# 3. Looping Over Indices  

Index-based iteration is appropriate when:

- You need to access **multiple lists** in parallel.  
- You want both the **index** and **the element**.  
- You are transforming nested lists or multi-dimensional data.

### Example — Access elements by index

```python

for i in range(len(usd_list)):     # i = 0 → 4
    print(i, usd_list[i])
```

### Helper Expansion — `range_index_map()`
Integrates your Chapter 6 Range Visualizer into loop logic.

```python
def range_index_map(iterable):
    """
    Shows how Python maps a range to indices for a given iterable.
    Mirrors Maven's step-through on page 10.
    """
    print("RANGE → INDEX MAP\n------------------")
    for idx in range(len(iterable)):
        print(f"range index {idx} → iterable[{idx}] = {iterable[idx]!r}")
```

# 4. Looping Over Multiple Iterables  

When two lists share a one-to-one alignment (e.g., item ↔ price), indexing allows synchronized access:

### Example — Print item and its euro price

```python
items = ["Snowboard", "Boots", "Helmet", "Goggles", "Bindings"]
euro_prices = [5.27, 8.79, 17.59, 21.99, 87.99]

for i in range(len(items)):
    print(items[i], euro_prices[i])
```

This produces 5 synchronized pairs.

### Helper — `sync_lists()`

```python
def sync_lists(*lists):
    """
    Prints synchronized items across multiple lists of equal length.
    Excellent for diagnostics in multi-iterable looping.
    """
    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
    length = len(lists[0])
    for iterable in lists:
        if len(iterable) != length:
            raise ValueError("All lists must have the same length for synchronized iteration.")
    
    for idx in range(length):
        row = [lst[idx] for lst in lists]
        print(f"Index {idx}: {row}")
```

Usage:

```python
sync_lists(items, euro_prices)
```

# 5. `enumerate()`: The Best Practice  

Using `enumerate()` is the preferred method when you want both index and item without constructing `range(len(...))`.

### Example — Using `enumerate()`

```python
euro_list = [5.27, 8.79, 17.59, 21.99, 87.99]
item_list = ["Snowboard", "Boots", "Helmet", "Goggles", "Bindings"]

for idx, value in enumerate(euro_list):
    print(item_list[idx], value)
```

### Helper — `enumerate_visualizer()`

```python
def enumerate_visualizer(iterable):
    """
    Shows exactly what enumerate() produces (index, item).
    """
    print("ENUMERATE VISUALIZER\n---------------------")
    for idx, item in enumerate(iterable):
        print(f"enumerate → (index={idx}, item={item!r})")
```

Example:

```python
enumerate_visualizer(["Snowboard", "Boots", "Helmet"])
```
---

### Overview — Using `enumerate()`
In Python, `enumerate()` provides a clean and safe way to access **both the index and the value** during iteration.  
It resolves the main drawbacks of index-based looping while preserving positional awareness when needed.

Conceptually, `enumerate()` represents the **next evolutionary step** after learning:

1. Item-based for loops (simple, readable)  
2. Index-based for loops (positional control, but fragile)  

`enumerate()` combines the strengths of both approaches.

---

## Why `enumerate()` Exists

Recall the index-based loop pattern:

```python
for i in range(len(items)):
    print(items[i])
```

While correct, this pattern requires you to:
- manually manage indices  
- ensure correct bounds (`len(...)`)  
- mentally track both index and value  

`enumerate()` removes this overhead by yielding **index–value pairs directly**.

---

## Basic `enumerate()` Pattern

```python
items = ["Snowboard", "Boots", "Helmet"]

for index, item in enumerate(items):
    print(index, item)
```

### Output
```
0 Snowboard
1 Boots
2 Helmet
```

### Key Idea
- `index` is automatically generated  
- `item` is the value from the iterable  
- No manual indexing is required  

---

## Comparing Three Looping Styles

### 1. Item-Based Loop (Most Common)

```python
for item in items:
    print(item)
```

Use when:
- you only need the value  
- position does not matter  

---

### 2. Index-Based Loop (Positional Control, Higher Risk)

```python
for i in range(len(items)):
    print(i, items[i])
```

Use when:
- you must reference multiple lists by position  
- you need write-access to a list by index  

⚠️ Requires careful alignment.

---

### 3. `enumerate()` Loop (Recommended When Index Is Needed)

```python
for i, item in enumerate(items):
    print(i, item)
```

Use when:
- you want the index for labeling, tracing, or logic  
- you want clarity and safety  
- you want to avoid manual index bookkeeping  

---

## Applied Example — Index-Based Loop vs `enumerate()`

### Index-Based Version

```python
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99]
euro_list = []

for i in range(len(usd_list)):
    euro_value = round(usd_list[i] * exchange_rate, 2)
    euro_list.append(euro_value)

print(euro_list)
```

---

### `enumerate()` Version (Cleaner)

```python
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99]
euro_list = []

for i, price in enumerate(usd_list):
    euro_value = round(price * exchange_rate, 2)
    euro_list.append(euro_value)

print(euro_list)
```

### Why This Is Better
- `price` is directly available (no `usd_list[i]`)  
- `i` is present if needed (debugging, labeling)  
- Loop intent is clearer and easier to reason about  

---

## 🟦 Helper — Enumerate Visualizer

This helper makes `enumerate()` behavior explicit and observable.

```python
def enumerate_visualizer(iterable):
    """
    Visualizes how enumerate() yields index-value pairs
    during iteration.
    """
    print("ENUMERATE VISUALIZER\n---------------------")
    
    for index, value in enumerate(iterable):
        print(f"index = {index}, value = {value!r}")
```

### Example Usage

```python
items = ["Snowboard", "Boots", "Helmet"]
enumerate_visualizer(items)
```

---

## When to Prefer `enumerate()`

Use `enumerate()` when:
- you need index information for tracing or labeling  
- you want positional awareness without index arithmetic  
- readability and debuggability matter  
- teaching or explaining iteration logic  

Avoid `enumerate()` when:
- you only need the values (simple item-based loop is best)  
- you must mutate a list by index (explicit index-based loop may be required)  

### Related Guidance
- If all iterables are equal peers → use `zip()`  
- If one iterable defines the iteration → use `enumerate()`
- `enumerate()` yields `(index, value)` pairs automatically  
- It eliminates the need for `range(len(...))` in most cases  
- It improves clarity, safety, and debuggability  
- Index-based loops still matter when mutating data or coordinating multiple structures  

---

## How This Fits the Learning Flow

At this point, you have learned:

- Item-based looping (cleanest default)  
- Index-based looping (positional control, but fragile)  
- `enumerate()` (balanced, expressive, safer) → one iterable defines the iteration
- `zip()` pairing two equal-length iterables, avoids manual indexing  

This completes the **for-loop family**.

The next major step is **while loops**, which introduce:
- state-driven iteration  
- termination logic  
- infinite-loop risk  
- simulation-style patterns common in analytics  

---

# 6. Choosing Between Items, Indices, `enumerate()`, and `zip()`

| Pattern                              | Who defines iteration? | What is yielded per loop? |
| ------------------------------------ | ---------------------- | ------------------------- |
| `for item in iterable`               | iterable               | `item`                    |
| `for i in range(len(iterable))`      | iterable               | `i` (int)                 |
| `for i, item in enumerate(iterable)` | iterable               | `(i, item)` tuple         |
| `for a, b in zip(iter1, iter2)`      | **both equally**       | `(a, b)` tuple            |


| Approach                                | When to Use                                                                      | Benefits                            |
| --------------------------------------- | -------------------------------------------------------------------------------- | ----------------------------------- |
| `for item in iterable:`                 | Transforming or inspecting values                                                | Clean, safest, Pythonic             |
| `for i in range(len(iterable)):`        | When **one iterable controls iteration** and index access is explicitly required | Maximum positional control          |
| `for idx, item in enumerate(iterable):` | When you need **both index and value** from a single iterable                    | Best practice, readable, efficient  |
| `for a, b in zip(iter1, iter2):`        | When **multiple iterables are equal peers** and should be processed together     | Safest alignment, avoids index bugs |


---

# Key Takeaways

- **For loops** execute once per element in an iterable; the iteration count is predictable.  
- Use **item-based looping** whenever possible; it is the safest and most readable pattern.  
- Use **indices** when pairing multiple lists or modifying elements in place.  
- Use **enumerate()** for clean, Pythonic access to both **index and item**. If one iterable defines the iteration, use enumerate()
- Use **zip()** if all iterables are equal peers.  

- This section introduces the core loop helpers:  
  - **`loop_trace_for()`** — traces item-by-item for-loop iteration and execution order  
  - **`enumerate_visualizer()`** — makes index–value pairing explicit when using `enumerate()`  
  - **`safe_while_trace()`** — safely traces while-loop state evolution and prevents infinite loops  
  - **`deep_loop_state_visualizer()`** — traces each internal step of a list-processing loop, including variable assignment, calculation execution, each `.append()` operation, and the evolving state of the accumulator container
  - **`validate_alignment()`** - Validate iterable alignment before looping. (Length alignment is required for index-based looping)
  - **`sync_lists()`** — synchronized iteration across multiple equal-length iterables 
- Understanding iteration mechanics — including when and how position matters — is essential for dataset traversal, data cleaning, transformation workflows, simulation modeling, and algorithmic problem-solving.
- These visualizers make loop mechanics traceable, which is essential for debugging, teaching, and analytics work.


# Chapter 07 — Loops
# Section 7.03 — While Loops

### Preview
A **while loop** repeats a block of code *as long as a logical expression evaluates to True*.  
This contrasts with a **for loop**, which repeats a fixed number of times.

Use a while loop when:

- You **do not know** how many iterations will be required.  
- You are simulating a process to completion (e.g., reach a threshold, exhaust inventory).  
- You need iterative refinement until a condition stabilizes (e.g., convergence in algorithms).  

## While loops repeat based on state, not count

Structure can be summarized as:
```Python
while condition_on_state:
    update_state
```

Execution continues until the expression becomes **False**.  
If the condition never turns False, you enter an **infinite loop**, which we will guard against.

This section extends your Helper Toolkit with a **While Loop Safety System**, simulation helpers, and state-tracing tools.

---

# When Should You Use Each Loop Type?

| Scenario | Best Choice | Rationale |
|---------|-------------|-----------|
| You know the dataset length | **For Loop** | One iteration per element; safe, predictable |
| You want to repeat until a condition is false | **While Loop** | Logic-driven iteration |
| You need to prevent infinite loops | **For Loop** (or While Loop with counters) | Counters act as emergency brakes |
| You are iterating over sequences (list, tuple, string) | **For Loop** | Pythonic and safe |
| You need a “repeat until solved” pattern (e.g., simulation) | **While Loop** | Flexible stopping criteria |

---

# 1. Basic While Loop Pattern

A while loop must include **three things**:

1. An **initial state**  
2. A **logical condition** that inspects that state  
3. An **update step** that *moves the state toward termination*  

### Example — Counting upward

```python
counter = 0                     # initial state

while counter < 5:              # logical condition
    print(counter)
    counter += 1                # update step
```

Python evaluates:

- Iteration 0 → counter = 0  
- Iteration 1 → counter = 1  
- Iteration 2 → counter = 2  
- Iteration 3 → counter = 3  
- Iteration 4 → counter = 4  
- Iteration stops when counter becomes 5, making (counter < 5) False.

Without the update (`counter += 1`), the loop would run forever.


`````python
# --------------------------------------------------------
# Demonstrating a While Loop Pattern Safely with Counters
# --------------------------------------------------------
# Simple while loop that counts upward until a threshold is reached
# While loops run until a given logical expression becomes FALSE


counter = 0
while counter < 5:      # Loop continues WHILE this expression is TRUE
    counter += 1        # Without this update, the loop becomes infinite ⚠️
    print("Counter =", counter)  
print()


# --------------------------------------------------------
# While Loop: Inventory until stock runs out
# --------------------------------------------------------

current_inventory = 686
month_counter = 0
monthly_sales = 84

while current_inventory > 0:
    current_inventory -= monthly_sales          # subtraction assignment
    month_counter += 1
    print(
        f"At month-end {month_counter}: "
        f"My inventory balance is {current_inventory}"
    )
print()


# --------------------------------------------------------
# While Loop: Calculating bank balance until money runs out
# --------------------------------------------------------

bank_balance = 7_189
month_counter = 0
spending = 350

while bank_balance > 0:
    bank_balance -= spending          # subtraction assignment
    month_counter += 1
    print(
        f"At month-end {month_counter}: "
        f"My balance is ${round(bank_balance, 2)}"
    )
`````

`````python
# --------------------------------------------------------
# While Loops with Counters
# Counters track loop progress, prevent runaway execution, 
# and can act as secondary safety conditions
# --------------------------------------------------------

# Example — Growing a portfolio until it reaches 1 million
# Loop terminates when the portfolio crosses the threshold.


portfolio = 800_000
year = 0

while portfolio < 1_000_000:            # primary logical condition
    investment_income = portfolio * 0.05
    portfolio += investment_income       # update state
    year += 1                             # update counter
    print(f"End of year {year}: ${round(portfolio, 2)}")



# --------------------------------------------------------
# 🟦 Advanced Helper: State Timeline Recorder
# This helper logs every state value for later visualization or tabular analysis.
# --------------------------------------------------------


def while_timeline(start_state, condition_fn, update_fn, max_iter=100):
    """
    Records each iteration's state—ideal for simulations, 
    convergence testing, and algorithm diagnostics.
    """
    timeline = []
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        timeline.append((iteration, state))
        
        iteration += 1
        if iteration > max_iter:
            timeline.append(("ABORTED", state))
            break
        
        state = update_fn(state)
    
    timeline.append(("FINAL", state))
    return timeline


# Example

timeline = while_timeline(
    start_state=800000,
    condition_fn=lambda s: s < 1_000_000,
    update_fn=lambda s: s * 1.05,
    max_iter=10
)

timeline

# Produces a list of iteration → state pairs, perfect for later DataFrame conversion.
`````

`````python
# --------------------------------------------------------
# 🔁 While Loops — Condition-Driven Iteration 
# 🆚 For Loops - Count-Driven, does not mutate data
# --------------------------------------------------------

# --------------------------------------------------------
# While Loops — Condition-Driven Iteration
# --------------------------------------------------------
# NOTE:
# This example mutates the list in place to demonstrate while-loop logic ⚠️
# In real data analysis workflows, you usually avoid mutating lists directly
# Instead, you would use filtering or list comprehensions to create a new list


customer_ratings = [
    "5 stars", "3 stars", "5 stars", "4 stars", "4 stars",
    "1 star", "4 stars", "5 stars", "2 stars", "5 stars"
]

onestar_counter = 0

while "1 star" in customer_ratings:
    customer_ratings.remove("1 star")
    onestar_counter += 1

print(f"Removed {onestar_counter} one-star review(s)")
print(customer_ratings)

# --------------------------------------------------------
# For Loops — Count-Driven, does not mutate data
# --------------------------------------------------------
# TEACHING TAKEAWAY:
# While loops are useful when iteration depends on changing state.
# For loops are preferred when data can be processed in a single pass.
# In analytics, for loops (or vectorized operations) are usually clearer and safer.


customer_ratings = [
    "5 stars", "3 stars", "5 stars", "4 stars", "4 stars",
    "1 star", "4 stars", "5 stars", "2 stars", "5 stars"
]

onestar_counter = 0
filtered_ratings = []              # empty list container

for rating in customer_ratings:   # loop looks at every element in customer_ratings
    if rating == "1 star":        # if block counts "1 star" and does nothing else with it
        onestar_counter += 1
    else:                         # else block handles everything that is not "1 star"
        filtered_ratings.append(rating)

print(f"Removed {onestar_counter} one-star review(s)")
print(filtered_ratings)
print(customer_ratings)           # does not mutate data in-place
`````

`````python
# ------------------------------------------------------------
# 🟦 Helper: `safe_while_trace()` with Infinite-Loop Protection
# - state value at each iteration  
# - when/why the loop stops  
# - detection of missing or incorrect update logic  
# - protection against accidental infinite loops via `max_iter`
# -------------------------------------------------------------

# ❌ Incorrect logic: update is calculated but not reassigned

counter = 0

# ⚠️ DO NOT RUN
# while counter < 5:
#     counter + 1      # expression evaluated, result discarded
#     print(counter)

# counter += 1   # ✔ correct
# counter + 1    # ❌ does nothing

# -------------------------------------------------------------
# 🟦 Safe demonstration with helper
# -------------------------------------------------------------

def safe_while_trace(start_state, condition_fn, update_fn, max_iter=50):
    """
    Safely traces a while loop with a guard to prevent infinite loops.
    """
    print("SAFE WHILE TRACE\n-----------------")
    
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        print(f"Iteration {iteration}: state = {state!r}")
        
        iteration += 1
        if iteration > max_iter:
            print("ABORTED: Possible infinite loop (max_iter exceeded).")
            break
        
        state = update_fn(state)
    
    print(f"Final state: {state!r}")
    return state
print()

# Function call:
safe_while_trace(
    start_state=165000,
    condition_fn=lambda s: s < 3_500_000,
    update_fn=lambda s: s + (s * 0.14),
    max_iter=50
)

safe_while_trace(
    start_state=165_000,
    condition_fn=lambda s: s < 3_500_000,
    update_fn=lambda s: s + s * 0.13,
    max_iter=10
)


# This outputs warnings before an infinite loop occurs.
`````

`````python
# -----------------------------------------------------------
# Exercise — Demonstrating a While Loop
# Example 401(k) - No Contributions, Annual Returns 13%
# ------------------------------------------------------------

# Starting portfolio balance is 165_000
stock_portfolio = 165_000
year_counter = 0

while stock_portfolio < 2_000_000:
    # calculate annual investment return
    investment_return = stock_portfolio * 0.13  # interest rate

    # add income to end of year portfolio balance
    stock_portfolio += investment_return

    # add each year
    year_counter += 1

    print(f"At year-end {year_counter}: "
    + f"My balance is ${round(stock_portfolio, 2)}")
print()


# Alternatively:

stock_portfolio = 165_000
year_counter = 0
annual_return_rate = 0.13

while stock_portfolio < 2_000_000:
    investment_return = stock_portfolio * annual_return_rate
    stock_portfolio += investment_return
    year_counter += 1

    print(f"At year-end {year_counter}: "
          f"My balance is ${round(stock_portfolio, 2)}")

print(f"\nGoal reached in {year_counter} years.")
`````

`````python
# ------------------------------------------------------------
# 🟦 Helper: `safe_while_trace()` with Infinite-Loop Protection
# - state value at each iteration  
# - when/why the loop stops  
# - detection of missing or incorrect update logic  
# - protection against accidental infinite loops via `max_iter`
# -------------------------------------------------------------

stock_portfolio = 165000                        # start state is 165000

while stock_portfolio < 3_500_000:              # condition is s < 3_500_000
    stock_portfolio += stock_portfolio * 0.14   # update step is s * 0.13


def safe_while_trace(start_state, condition_fn, update_fn, max_iter=50):
    """
    Safely traces a while loop with a guard to prevent infinite loops.
    """
    print("SAFE WHILE TRACE\n-----------------")
    
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        print(f"Iteration {iteration}: state = {state!r}")
        
        iteration += 1
        if iteration > max_iter:
            print("ABORTED: Possible infinite loop (max_iter exceeded).")
            break
        
        state = update_fn(state)
    
    print(f"Final state: {state!r}")
    return state
print()

# Function call:
safe_while_trace(
    start_state=165000,
    condition_fn=lambda s: s < 3_500_000,
    update_fn=lambda s: s + (s * 0.14),
    max_iter=50
)
`````

`````python
# --------------------------------------------------------
# Exercise — Demonstrating a While Loop
# Roth IRA Example with monthly contributions
# --------------------------------------------------------

# -----------------------------
# Roth IRA inputs  🧠✨
# -----------------------------
stock_portfolio = 21_077
monthly_contribution = 8_000 / 12      # $666.67 per month
annual_return_rate = 0.14
year_counter = 0

while stock_portfolio < 124_000:
    # add monthly contributions
    stock_portfolio += monthly_contribution * 12  # monthly contributions

    # apply annual investment return
    investment_return = stock_portfolio * annual_return_rate
    stock_portfolio += investment_return

    year_counter += 1

    print(
        f"At year-end {year_counter}: "
        f"My balance is ${round(stock_portfolio, 2)}"
    )

print(f"\nGoal reached in {year_counter} years.")
`````

`````python
# --------------------------------------------------------
# Exercise — Demonstrating a While Loop
# Example Roth IRA — Age + Monthly vs Annual Contribution Timing
# --------------------------------------------------------

from datetime import date

# -----------------------------
# Fixed personal anchors
# -----------------------------
BIRTH_YEAR = 1967
AS_OF = date(2025, 12, 14)   # reporting date each year (Dec 14)

# -----------------------------
# Roth IRA inputs  🧠✨
# -----------------------------
start_balance = 21_077
annual_contribution = 8_000
annual_return_rate = 0.14
goal = 145_000

# -----------------------------
# Simulator with age tracking
# timing = "begin" → contribution first (monthly-style timing)
# timing = "end"   → contribution last (annual-style timing)
# -----------------------------
def simulate_roth_with_age(start_balance, annual_contribution, annual_return_rate, goal, timing="begin"):
    balance = start_balance
    years = 0
    rows = []

    while balance < goal:
        years += 1
        year_end = date(AS_OF.year + years, AS_OF.month, AS_OF.day)
        age = year_end.year - BIRTH_YEAR  # Dec 14 is after May 7

        if timing == "begin":
            balance += annual_contribution
            balance += balance * annual_return_rate
        elif timing == "end":
            balance += balance * annual_return_rate
            balance += annual_contribution
        else:
            raise ValueError("timing must be 'begin' or 'end'")

        rows.append((year_end, age, balance))

    return years, rows

# -----------------------------
# Run both scenarios
# -----------------------------
years_begin, rows_begin = simulate_roth_with_age(
    start_balance, annual_contribution, annual_return_rate, goal, timing="begin"
)

years_end, rows_end = simulate_roth_with_age(
    start_balance, annual_contribution, annual_return_rate, goal, timing="end"
)

# -----------------------------
# Output
# -----------------------------
print(f"As of {AS_OF}: age = {AS_OF.year - BIRTH_YEAR}\n")

print("MONTHLY-style contributions (begin-of-year timing):")
for year_end, age, bal in rows_begin:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_begin} years.\n")

print("ANNUAL contributions (end-of-year timing):")
for year_end, age, bal in rows_end:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_end} years.\n")

diff = years_end - years_begin
print(
    f"Difference: monthly-style timing reaches the goal "
    f"{abs(diff)} year(s) {'sooner' if diff > 0 else 'later' if diff < 0 else 'at the same time'}."
)
`````

`````python
# --------------------------------------------------------
# Exercise — Demonstrating a While Loop
`````
`````python
# Example 401(k) with Employer Match
# 401(k) — Age + Monthly vs Annual Contribution Timing
# --------------------------------------------------------

from datetime import date

# -----------------------------
# Fixed personal anchors
# -----------------------------
BIRTH_YEAR = 1967
AS_OF = date(2025, 12, 14)   # anchor all "year-end" reporting to Dec 14 each year

# -----------------------------
# Scenario inputs 🧠✨
# -----------------------------
start_balance = 165_000
annual_return_rate = 0.14
annual_contribution = 32_500 + 6_900
goal = 3_500_000

# ----------------------------------------------------------------------
# Simulator (filtered w/steps for every 3 years, prints age + balance) 🧠✨
# timing = "begin"  -> contribution first, then return (your "monthly style" timing)
# timing = "end"    -> return first, then contribution (your "annual style" timing)
# ----------------------------------------------------------------------

def simulate_yearly_with_age(start_balance, annual_contribution, annual_return_rate, goal, timing="begin"):
    balance = start_balance
    years = 0
    rows = []

    while balance < goal:
        years += 1
        year_end = date(AS_OF.year + years, AS_OF.month, AS_OF.day)
        age = year_end.year - BIRTH_YEAR  # Dec 14 is after May 7

        if timing == "begin":
            balance += annual_contribution
            balance += balance * annual_return_rate
        elif timing == "end":
            balance += balance * annual_return_rate
            balance += annual_contribution
        else:
            raise ValueError("timing must be 'begin' or 'end'")

        # ---------------------------------
        # Filtered output (every 3 years + final year)
        # ---------------------------------
        if years % 3 == 0 or balance >= goal:
            rows.append((year_end, age, balance))

    return years, rows

# -----------------------------
# Run both scenarios
# -----------------------------
years_begin, rows_begin = simulate_yearly_with_age(start_balance, annual_contribution, annual_return_rate, goal, timing="begin")
years_end, rows_end = simulate_yearly_with_age(start_balance, annual_contribution, annual_return_rate, goal, timing="end")

# -----------------------------
# Print results
# -----------------------------
print(f"As of {AS_OF}: age = {AS_OF.year - BIRTH_YEAR}")
print("\nBEGIN-OF-YEAR contributions (contrib then return):")
for year_end, age, bal in rows_begin:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_begin} years.\n")

print("END-OF-YEAR contributions (return then contrib):")
for year_end, age, bal in rows_end:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_end} years.\n")

# Quick comparison line
delta = years_end - years_begin
print(f"Difference: begin-of-year timing reaches the goal {abs(delta)} year(s) "
      f"{'sooner' if delta > 0 else 'later' if delta < 0 else 'at the same time'} than end-of-year timing.")
`````

`````python
from datetime import date

# --------------------------------------------------------
# Two-Phase 401(k) Simulator — Monthly Contributions
# Phase 1: high contributions for 8 years (96 months)
# Phase 2: reduced contributions until target reached 🚀🔥
# --------------------------------------------------------

# Personal anchors
BIRTH_YEAR  = 1967
BIRTH_MONTH = 5
BIRTH_DAY   = 7

AS_OF = date(2025, 12, 14)  # anchor reporting to the 14th of each month

def add_months(d, months):
    """Add months to a date (safe here because day=14 exists in every month)."""
    y, m = d.year, d.month
    m0 = (m - 1) + months
    new_y = y + (m0 // 12)
    new_m = (m0 % 12) + 1
    return date(new_y, new_m, d.day)

def age_on(d, birth_year=BIRTH_YEAR, birth_month=BIRTH_MONTH, birth_day=BIRTH_DAY):
    """Compute age at a given date."""
    age = d.year - birth_year
    if (d.month, d.day) < (birth_month, birth_day):
        age -= 1
    return age

# -----------------------------
# Scenario inputs
# -----------------------------
start_balance = 164_658.86
annual_return_rate = 0.14
target_balance = 3_000_000

# Phase 1: through 12/14/2033 (8 years = 96 months)
phase1_months = 96
monthly_contrib_phase1 = (32_500 / 12) + (6_900 / 12)   # employee + match

# Phase 2: after 12/14/2033
monthly_contrib_phase2 = (7_200 / 12) + (7_200 / 12)    # employee + match

# Convert annual to monthly (compound-consistent)
monthly_return_rate = (1 + annual_return_rate) ** (1/12) - 1

# -----------------------------
# Simulation (monthly steps)
# -----------------------------
balance = start_balance
month_counter = 0

print(f"As of {AS_OF}: Age {age_on(AS_OF)} | Start balance ${balance:,.2f}\n")

while balance < target_balance:
    month_counter += 1
    current_date = add_months(AS_OF, month_counter)

    # Switch contributions after 96 months (12/14/2033)
    if month_counter <= phase1_months:
        monthly_contribution = monthly_contrib_phase1
        phase_label = "Phase 1 (high contrib)"
    else:
        monthly_contribution = monthly_contrib_phase2
        phase_label = "Phase 2 (reduced contrib)"

    # Contribution first, then growth (monthly cadence)
    balance += monthly_contribution
    balance += balance * monthly_return_rate

    # Print yearly checkpoints + final month
    if month_counter % 12 == 0 or balance >= target_balance:
        print(
            f"{current_date} | Age {age_on(current_date)} | "
            f"{phase_label} | Balance ${balance:,.2f}"
        )

print(f"\nGoal reached in {month_counter} months (~{month_counter/12:.1f} years) by {current_date}.")
`````

# Chapter 07 — Loops
# Section 7.03 — While Loops

### Preview
A **while loop** repeats a block of code *as long as a logical expression evaluates to True*.  
This contrasts with a **for loop**, which repeats a fixed number of times.

Use a while loop when:

- You **do not know** how many iterations will be required.  
- You are simulating a process to completion (e.g., reach a threshold, exhaust inventory).  
- You need iterative refinement until a condition stabilizes (e.g., convergence in algorithms).  

## While loops repeat based on state, not count

Structure can be summarized as:
```Python
while condition_on_state:
    update_state
```

Execution continues until the expression becomes **False**.  
If the condition never turns False, you enter an **infinite loop**, which we will guard against.

This section extends your Helper Toolkit with a **While Loop Safety System**, simulation helpers, and state-tracing tools.

---

# When Should You Use Each Loop Type?

| Scenario | Best Choice | Rationale |
|---------|-------------|-----------|
| You know the dataset length | **For Loop** | One iteration per element; safe, predictable |
| You want to repeat until a condition is false | **While Loop** | Logic-driven iteration |
| You need to prevent infinite loops | **For Loop** (or While Loop with counters) | Counters act as emergency brakes |
| You are iterating over sequences (list, tuple, string) | **For Loop** | Pythonic and safe |
| You need a “repeat until solved” pattern (e.g., simulation) | **While Loop** | Flexible stopping criteria |

---

# 1. Basic While Loop Pattern  

A while loop must include **three things**:

1. An **initial state**  
2. A **logical condition** that inspects that state  
3. An **update step** that *moves the state toward termination*  

### Example — Counting upward

```python
counter = 0                     # initial state

while counter < 5:              # logical condition
    print(counter)
    counter += 1                # update step
```

Python evaluates:

- Iteration 0 → counter = 0  
- Iteration 1 → counter = 1  
- Iteration 2 → counter = 2  
- Iteration 3 → counter = 3  
- Iteration 4 → counter = 4  
- Iteration stops when counter becomes 5, making (counter < 5) False.

Without the update (`counter += 1`), the loop would run forever.

---

# 2. While Loops with Counters

Counters track loop progress, prevent runaway execution, and can act as **secondary safety conditions**.

### Example — Growing a portfolio until it reaches 1 million

```python
portfolio = 800_000
year = 0

while portfolio < 1_000_000:            # primary logical condition
    investment_income = portfolio * 0.05
    portfolio += investment_income       # update state
    year += 1                             # update counter
    print(f"End of year {year}: ${round(portfolio, 2)}")
```
 
Loop terminates when the portfolio crosses the threshold.

---

# 3. 🟦 Helper Expansion: Safe While Loop Tracer

This upgrades your earlier helper into a reusable diagnostic tool for notebook-based explorations.

```python
def safe_while_trace(start_state, condition_fn, update_fn, max_iter=50):
    """
    A safe, diagnostic tracer for while loops.
    Prevents accidental infinite loops and makes state transitions visible.

    Parameters
    ----------
    start_state : any
        Initial loop state.
    
    condition_fn : callable(current_state) -> bool
        Determines whether the loop should continue.
    
    update_fn : callable(current_state) -> new_state
        Defines how the state changes each iteration.
    
    max_iter : int
        Safety guard to prevent infinite looping.
    """
    
    print("SAFE WHILE TRACE\n-----------------")
    
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        print(f"Iteration {iteration}: state = {state!r}")
        
        iteration += 1
        if iteration > max_iter:
            print("ABORTED: Possible infinite loop (max_iter exceeded).")
            break
        
        state = update_fn(state)
    
    print(f"Final state: {state!r}")
    return state
```

### Example Usage — Portfolio example  

```python
safe_while_trace(
    start_state=800_000,
    condition_fn=lambda s: s < 1_000_000,
    update_fn=lambda s: s + s * 0.05,   # grow by 5%
    max_iter=10
)
```

```python

monthly_sales = 75
month = 0

while current_inventory > 0:
    current_inventory -= monthly_sales      # subtract sales
    month += 1
    print(f"After month {month}, inventory = {current_inventory}")
```

The loop ends when inventory reaches zero or below — the precise moment demand exhausts supply.

This mirrors the structure shown in the inventory projections exercise (Maven pp. 23–24).

# 5. Infinite Loops and Error Prevention  

A while loop becomes infinite when:

- The logical condition is always True  
- The update step never moves state toward termination  
- Logic depends on an impossible or stalled signal  
- The update step is omitted (common beginner error)

### Demonstration of incorrect logic  
*(Do NOT run this pattern without safeguards)*

```python
# while x < 10:
#     print(x)      # x never changes -> infinite loop
```

### Safe demonstration with helper

```python
safe_while_trace(
    start_state=0,
    condition_fn=lambda s: s < 10,
    update_fn=lambda s: s,       # incorrect: state never updates
    max_iter=8
)
```

This outputs warnings before an infinite loop occurs.

*(Maven page 21)*

```python
while bank_balance > 0:
    spending = 1000
    bank_balance -= spending      # or: bank_balance -= 1000
    month += 1
    print(f"End of month {month}: balance = {bank_balance}")
```

This represents a **linear depletion model**, useful in:

- Project budgeting

- Inventory depletion

# 7. 🟦 Advanced Helper: State Timeline Recorder

This helper logs every state value for later visualization or tabular analysis.

```python
def while_timeline(start_state, condition_fn, update_fn, max_iter=100):
    """
    Records each iteration's state—ideal for simulations, 
    convergence testing, and algorithm diagnostics.
    """
    timeline = []
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        timeline.append((iteration, state))
        
        iteration += 1
        if iteration > max_iter:
            timeline.append(("ABORTED", state))
            break
        
        state = update_fn(state)
    
    timeline.append(("FINAL", state))
    return timeline
```

### Example

```python
timeline = while_timeline(
    start_state=800000,
    condition_fn=lambda s: s < 1_000_000,
    update_fn=lambda s: s * 1.05,
    max_iter=10
)

timeline
```

Produces a list of iteration → state pairs, perfect for later DataFrame conversion.

---

# 8. When to Choose a While Loop

Use a **while loop** when:

- Termination is based on *reaching a threshold*, not a known count.  
- You are performing **search**, **waiting**, or **simulation until solved**.  
- You want to protect against infinite loops using counters or safety guards.  
- You need maximum flexibility in defining stopping logic.

Use a **for loop** when:

- You know the dataset length.  
- You want guaranteed finite iteration.  
- You are looping over user-defined collections.

---

# Key Takeaways

- A **while loop** repeats until its logical expression becomes False.  
- It requires **state initialization**, **condition**, and **update** to avoid runaway execution.  
- While loops are ideal for **simulations, threshold-based tasks, and unknown-duration processes**.  
- Infinite loops occur when the condition never becomes False; always include a **state update** or **safety counter**.  
- The new helpers —  
  - `safe_while_trace()`  
  - `while_timeline()`  
provide controlled exploration of while-loop behavior for learning and debugging.

### `deep_loop_state_visualizer()` — Full State Evolution for List-Processing Loops  

Shows:  
- iteration number  
- the current value pulled from the iterable  
- the computed result for that iteration
- the *entire evolving accumulator* (e.g., a list) after each append  
- how data transformations unfold step-by-step

NOTE:
- this helper is useful for learning and debbugging
- it exposes each step of an accumulator-based loop that is normally hidden inside a standard loop
- 

This helper is especially useful for understanding loops that *build new datasets*, such as currency conversion, filtering, ETL steps, or feature engineering.


```python
def deep_loop_state_visualizer(iterable, transform_fn):
    """
    Visualizes the full state evolution of a loop that builds
    an accumulator (e.g., a list) from an iterable.
    
    Parameters
    ----------
    iterable : iterable
        The source data being looped over.
    
    transform_fn : callable
        A function applied to each element of the iterable
        to produce the value appended to the accumulator.
    
    Reveals
    -------
    - iteration number
    - current input value
    - transformed value
    - full accumulator state after each append
    """
    accumulator_list = []
    
    print("DEEP LOOP STATE VISUALIZER\n--------------------------")
    
    for iteration, current_value in enumerate(iterable):
        transformed_value = transform_fn(current_value)
        accumulator_list.append(transformed_value)
        
        print(f"Iteration {iteration}:")
        print(f"  current_value     = {current_value!r}")
        print(f"  transformed_value = {transformed_value!r}")
        print(f"  accumulator_list  = {accumulator_list}\n")
    
    print("Final Output:", accumulator_list)
    return accumulator_list
```
Example usage:
```python
prices = [5.99, 9.99, 19.99]
exchange_rate = 0.88

deep_loop_state_visualizer(
    prices,
    transform_fn=lambda x: round(x * exchange_rate, 2)
)
```

### Helper `validate_alignment()` - Validate alignment before looping

```python

def validate_alignment(*iterables):
    """
    Validates that all provided iterables have the same length.

    This helper is intended to catch silent logic errors
    when using index-based looping across multiple iterables.

    Raises
    ------
    ValueError
        If the iterables are not aligned in length.
    """
    if not iterables:
        raise ValueError("At least one iterable must be provided.")
    
    lengths = [len(it) for it in iterables]
    
    if len(set(lengths)) != 1:
        raise ValueError(
            f"Iterables are not aligned in length: {lengths}"
        )
    
    print("Iterables are aligned by length. Safe to proceed.")
```
### Example Usage:
```python
items = ["Snowboard", "Boots", "Helmet"]
prices = [499, 299, 99]
ratings = [5, 4, 4]

validate_alignment(items, prices, ratings)
```
### NOTE: 
- Length alignment is required for index-based looping.
- This helper prevents silent misalignment bugs that produce incorrect results without raising errors.
### In practice: 
- Prefer zip() when all iterables are equal peers
- Prefer structured data for long-term robustness

### `sync_lists()` — Synchronized iteration across multiple equal-length iterables

```python
def sync_lists(*iterables):
    """
    Prints synchronized elements across multiple iterables
    that are assumed to be positionally aligned.

    This helper is intended for diagnostics and learning,
    not for production workflows.

    Raises
    ------
    ValueError
        If the iterables do not all have the same length.
    """
    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
    
    if not iterables:
        raise ValueError("At least one iterable must be provided.")
    
    length = len(iterables[0])
    
    for it in iterables:
        if len(it) != length:
            raise ValueError(
                "All iterables must have the same length for synchronized iteration."
            )
    
    for idx in range(length):
        row = [it[idx] for it in iterables]
        print(f"Index {idx}: {row}")
```
### Usage:

```python
items   = ["Snowboard", "Boots", "Helmet"]
prices  = [499, 299, 99]
ratings = [5, 4, 4]

sync_lists(items, prices, ratings)
```
### Output conceptually:
```
Index 0: ['Snowboard', 499, 5]
Index 1: ['Boots', 299, 4]
Index 2: ['Helmet', 99, 4]
```

# Summary

- A **loop** repeats instructions until a **count-based condition** (for loop) or a **logical condition** (while loop) signals completion.  
- **For loops** are used when the number of iterations is known or when the task is “one pass per element.”  
  They may iterate directly over values, over **indices** when positional control is required, or over **index–value pairs** using `enumerate()` for improved clarity and safety.  
- **While loops** provide greater flexibility but must be carefully designed to avoid infinite loops.  
- This section introduces the core loop helpers:  
  - **`loop_trace_for()`** — traces item-by-item for-loop iteration and execution order  
  - **`enumerate_visualizer()`** — makes index–value pairing explicit when using `enumerate()`  
  - **`safe_while_trace()`** — safely traces while-loop state evolution and prevents infinite loops
  - **``while_timeline()``** - advanced helper logs every state value for later visualization or tabular analysis 
  - **`deep_loop_state_visualizer()`** — traces each internal step of a list-processing loop, including variable assignment, calculation execution, each `.append()` operation, and the evolving state of the accumulator container
  - **`validate_alignment`** - Validate iterable alignment before looping. (Length alignment is required for index-based looping)
  - **`sync_lists()`** — synchronized iteration across multiple equal-length iterables 
- Understanding iteration mechanics — including when and how position matters — is essential for dataset traversal, data cleaning, transformation workflows, simulation modeling, and algorithmic problem-solving.



# Chapter 07 — Loops  
## Section 7.04 — Nested Loops  

### Preview
A **nested loop** is a loop inside another loop.  
For every single iteration of the **outer loop**, the **inner loop** runs from start to finish.


> Nested loops allow you to process multi-level structures, generate combinations, and traverse data in two dimensions.

Nested loops are essential for:

- Navigating **nested lists**  
- Generating all **combinations** or **pairings**  
- Performing **matrix-like** operations  
- Processing **customer → orders** or **category → items** hierarchies  
- Applying transformations across multi-dimensional datasets  

---

## 🟦 Nested Loop Helpers Expand your Helper Toolkit

- The expanded toolkit includes:  
  - `nested_loop_trace()` visualizer
  - `transform_nested_list()`(general purpose)
  - `cartesian_product()` generator
  - `visualize_matrix()` for 2D structures

These tools give full transparency into multi-dimensional iteration.

---

## 1. Basic Nested Loop Structure

```
for item in outer_list:
    for inner_item in inner_list:
        do_something(item, inner_item)
```

Execution pattern:

- Outer iteration 0 → inner loop runs fully  
- Outer iteration 1 → inner loop runs fully  
- …  
- Outer iteration n → inner loop runs fully  

Total iterations =  
`len(outer_list) × len(inner_list)`


`````python
# --------------------------------------------------------
# Nested List Example — Item × Size Combinations
# --------------------------------------------------------
# Output count:
# - 2 items  
# - 3 sizes  
# - 2 × 3 = 6 combinations

items = ["Snowboard", "Boots"]
sizes = ["small", "medium", "large"]

# OUTER LOOP:
for item in items:
    # INNER LOOP:
    for size in sizes:
        print(f"{item} is available in {size}.")

# The outer loop moves by each list, index [0, 1]
# The inner loop processes every element for each list - [0][0], [0][1], [0][2]; [1][0], [1][1], [1][2]
# Inner loop runs fully before outer loop advances
# This demonstrates the “outer loop controls what, inner loop controls how many” principle


# --------------------------------------------------------
# 🟦 Helper: Nested Loop Trace Visualizer
# --------------------------------------------------------


def nested_loop_trace(outer, inner):
    """
    Visualizes nested loop execution by showing:
      - outer iteration index and value
      - inner iteration index and value
      - the resulting pair

    Mirrors the detailed iteration shown in Maven pp. 25–33.
    """
    print("NESTED LOOP TRACE\n------------------")
    
    for outer_idx, outer_val in enumerate(outer):
        print(f"\nOuter Iteration {outer_idx}: outer_val = {outer_val!r}")
        
        for inner_idx, inner_val in enumerate(inner):
            print(f"    Inner Iteration {inner_idx}: inner_val = {inner_val!r}")
            print(f"        → Pair = ({outer_val!r}, {inner_val!r})")

# Example usage:

nested_loop_trace(items, sizes)


# --------------------------------------------------------
# 🟦 Helper: Cartesian Product Generator
# --------------------------------------------------------
# (Useful for combinations, SKU creation, pairing simulations)

def cartesian_product(a, b):
    """
    Returns the Cartesian product of two iterables as a list of tuples.
    Equivalent to the nested-loop pairings in Maven's examples.
    """
    return [(x, y) for x in a for y in b]


# Example Usage:

cartesian_product(items, sizes)
`````

`````python
# --------------------------------------------------------
# Nested List Demo - Step-by-Step
# --------------------------------------------------------

item_reviews = [
    ["5 stars", "3 stars", "5 stars"],     # Skis
    ["4 stars", "4 stars"],                # Snowboards
    ["1 star", "4 stars"],                 # Goggles
    ["5 stars", "2 stars", "5 stars"]      # Boots
]


# --------------------------------------------------------
# Step 1. Outer list
# --------------------------------------------------------
# OUTER lOOP:
for lst in item_reviews:      # each nested list (4x)
    print(lst)
    
# --------------------------------------------------------
# Step 2. Inner list
# --------------------------------------------------------
# OUTER LOOP:
for lst in item_reviews:    # each nested list (4x)
    
    # INNER LOOP: 
    for review in lst:      # each element in each nested list (10x)
        print(review)

# --------------------------------------------------------
# Step 3. Extract Numeric Ratings Only
# --------------------------------------------------------
# OUTER LOOP:
for lst in item_reviews:             # each nested list (4x)
    
    # INNER LOOP: 
    for review in lst:               # each element in each nested list (10x)
        int_rating = int(review[0])  # extract numeric rating only
        print(int_rating)

# --------------------------------------------------------
# Step 4. Collect Numeric Ratings into Lists
# --------------------------------------------------------
# OUTER LOOP:
for lst in item_reviews:                     # each nested list (4x)
    integer_ratings = []                     # empty list accumulator: ratings in each nested list
    
    # INNER LOOP: 
    for review in lst:                       # each element in each nested list (10x)
        int_rating = int(review[0])          # extract numeric rating only
        integer_ratings.append(int_rating)   # .append() accumulator list
    print(integer_ratings)

# --------------------------------------------------------
# Step 5. Collect Numeric Ratings Back into a Nested List
# --------------------------------------------------------
numeric_item_reviews = []                           # empty list: collect numeric values back into a nested list

# OUTER LOOP:
for lst in item_reviews:                           # each nested list (4x)
    integer_ratings = []                           # empty list accumulator: ratings in each nested list
    
    # INNER LOOP: 
    for review in lst:                             # each element in each nested list (10x)
        int_rating = int(review[0])                # extract numeric rating only
        integer_ratings.append(int_rating)         # .append() accumulator list
    
    numeric_item_reviews.append(integer_ratings)   # Collect back into a nested list

numeric_item_reviews                               # return nested list with numeric values only


# --------------------------------------------------------
# Step 6. Final Pairing with Items
# --------------------------------------------------------
items = ["Skis", "Snowboards", "Goggles", "Boots"]

for item, ratings in zip(items, numeric_item_reviews):
    print(f"{item}: {ratings}")
`````

`````python
# --------------------------------------------------------
# Understand outer vs. inner loop roles
# Exercise: Nested Loops — Processing Reviews by Item
# --------------------------------------------------------
# Practice navigating nested lists
# Transform raw string data into numeric, analyzable structure
# Prepare data for aggregation (averages)


item_reviews = [
    ["5 stars", "3 stars", "5 stars"],     # Skis
    ["4 stars", "4 stars"],                # Snowboards
    ["1 star", "4 stars"],                 # Goggles
    ["5 stars", "2 stars", "5 stars"]      # Boots
]

item_reviews_numeric = []                  # outer loop accumulator

# OUTER LOOP:
for each_nested_lst in item_reviews:       # each nested list in item_reviews (4x)
    
    numeric_ratings = []                   # inner loop accumulator
    
    # INNER LOOP:
    for element in each_nested_lst:         # for element in nested list (10x)
        rating = int(element[0])            # extract numeric rating
        numeric_ratings.append(rating)
    
    # collect processed ratings for this item
    item_reviews_numeric.append(numeric_ratings)
    
    print(item_reviews_numeric)             # shows each iteration step-by-step

item_reviews_numeric

# The outer loop moves by each nested list, indices [0], [1], [2], [3]
# The inner loop processes every element for each nested list - [0][0], [0][1], [0][2]; [1][0], [1][1]; etc.
# The inner loop fully completes before the outer loop advances


# --------------------------------------------------------
# Aggregate results
# --------------------------------------------------------

# Calculate average review score per item
for ratings in item_reviews_numeric:
    average = sum(ratings) / len(ratings)
    print(round(average, 2))

# --------------------------------------------------------
# Pair results with item names
# --------------------------------------------------------

items = ["Skis", "Snowboards", "Goggles", "Boots"]

for i in range(len(items)):
    avg_rating = sum(item_reviews_numeric[i]) / len(item_reviews_numeric[i])
    print(f"{items[i]}: average rating = {avg_rating:.2f}")
`````

`````python
# --------------------------------------------------------
# Nested Loops Over Nested Lists
# Example — Apply 10% discount to each customer’s orders
# --------------------------------------------------------
# This pattern is used when dealing with multi-order customers, grouped transactions, or any list-of-lists structure.
# This preserves the **structure** of the original dataset — a common requirement in analytics workflows.


orders = [
    [120.00, 45.00, 89.99],        # customer 1 orders
    [250.00, 80.00],               # customer 2 orders
    [19.99, 55.00, 150.00, 40.00]  # customer 3 orders
]

discounted_orders = []                        # outer loop accumulator

for transactions in orders:                   # outer loop: each nested list (3x)
    customer_discounted = []                  # inner loop accumulator
    
    for subtotal in transactions:             # inner loop: for element in nested list (9x)
        new_total = round(subtotal * 0.9, 2)  # 10% discount (or 90% of the price)
        customer_discounted.append(new_total)
    
    discounted_orders.append(customer_discounted)

discounted_orders

# --------------------------------------------------------
# 🟦 Helper: Matrix Visualizer  
# --------------------------------------------------------

def visualize_matrix(matrix):
    """
    Prints a clean, aligned representation of a nested list
    to help visualize 2D structures created by nested loops.
    """
    print("MATRIX VIEW\n-----------")
    for row_idx, row in enumerate(matrix):
        print(f"Row {row_idx}: {row}")

# Example Usage:

visualize_matrix(discounted_orders)
print()


# --------------------------------------------------------
# 🟦 Helper: Nested List Transformer (General-Purpose)
# --------------------------------------------------------

orders = [
    [120.00, 45.00, 89.99],        # customer 1 orders
    [250.00, 80.00],               # customer 2 orders
    [19.99, 55.00, 150.00, 40.00]  # customer 3 orders
]

def transform_nested_list(nested, transform_fn):
    """
    Applies a user-defined transformation to each element in a nested list.

    Parameters
    ----------
    nested : list[list]
        A list of lists (e.g., customers → orders).
    
    transform_fn : callable
        Function applied to each inner element.

    Returns
    -------
    list[list]
        A new nested list with the same structure.

    Notes
    -----
    - General version of Maven's multi-order discount logic (pp. 32–33)
    - Works with any transformation: tax, normalization, scaling, flags, etc.
    """
    
    result = []                                  # outer loop accumulator
    
    for i, inner_list in enumerate(nested):      # outer loop
        transformed_inner = []                   # inner loop accumulatore
        
        for j, value in enumerate(inner_list):   # inner loop
            transformed_inner.append(transform_fn(value))
        
        result.append(transformed_inner)
    
    return result


# Example usage:
result = transform_nested_list(orders, lambda x: round(x * 0.9, 2))
# nothing will display unless you print
print(result)

result = transform_nested_list(orders, lambda x: round(x * 0.9, 2))
print("Discounted orders:", result)

# Jupyter Note:
# Only the final expression in a cell is displayed automatically.
# Functions return data silently unless print() is used.
`````

```python
def transform_nested_list(nested, transform_fn):
    """
    Applies a user-defined transformation to each element in a nested list.

    Parameters
    ----------
    nested : list[list]
        A list of lists (e.g., customers → orders).
    
    transform_fn : callable

    Returns
    -------
    list[list]
        A new nested list with the same structure.

    Notes
    -----
    - General version of Maven's multi-order discount logic (pp. 32–33)
    - Works with any transformation: tax, normalization, scaling, flags, etc.
    """
    
    result = []
    
    for i, inner_list in enumerate(nested):      # outer loop
        transformed_inner = []
        
        for j, value in enumerate(inner_list):   # inner loop
            transformed_inner.append(transform_fn(value))
        
        result.append(transformed_inner)
    
    return result
```

# 8. Understanding Nested Loop Complexity  
Nested loops typically have **O(n × m)** time complexity.
Total operations: n × m → O(n × m)

Examples:

- Items (2) × Sizes (3) → 6 operations → O(2 × 3)  
- Customers (3) × Orders (vary 2–4) → total operations = sum of row lengths. (Count how many total elements exist across all inner lists).

- If the inner loop length is fixed → total work = n × m.
- If the inner loop length varies → total work = sum of inner list lengths.
 

Performance matters when:

- Processing large datasets  
- Performing multi-level simulations  
- Building grid-style computations  

---

# 9. When to Use Nested Loops

Use nested loops when:

- You must examine every pairing in a 2D structure  
- Your data is multi-level (e.g., list of lists)  
- You need all combinations of two sets  
- You are generating structured outputs (e.g., matrix, grid, report sections)

Avoid nested loops when:

- A vectorized or built-in approach (NumPy, pandas) is available  
- Complexity becomes too high for large datasets  
- Iteration order does not matter (sometimes flattening is simpler)

---

# Key Takeaways

- A **nested loop** executes the inner loop *once per iteration* of the outer loop.  
- This produces a complete pairing (Cartesian product) of both iterables.  
- Nested loops are essential for processing **nested lists**, **combinations**, and **matrix-like structures**.  
- The expanded toolkit includes:  
  - `nested_loop_trace()`  
  - `transform_nested_list()`  
  - `cartesian_product()`  
  - `visualize_matrix()`  
These tools give full transparency into multi-dimensional iteration.  
- Nested iteration can be powerful but costly; understand when it is necessary and when alternatives exist.



# Chapter 07 — Loops  
## Section 7.05 — Loop Control

### Preview
Loop control statements allow you to **modify the normal flow of iteration**, enabling early exits, selective skipping, placeholder logic, and robust error handling.

Four essential loop control mechanisms include:

- **break** — exit a loop immediately  
- **continue** — skip the rest of the current iteration  
- **pass** — placeholder that does nothing  
- **try / except** — handle errors without stopping the loop  

These tools are critical for:

- Data cleaning  
- Error-tolerant ETL pipelines  
- Complex filtering  
- Avoiding infinite loops  
- Creating maintainable, incremental logic  

This section adds new helper functions for debugging, error tracing, and controlled loop execution.

---

# 🟦 Helper Box: Loop Control Helpers 
- The helper toolkit now includes: 
  - `break_trace()`  
  - `continue_trace()`  
  - `try_except_trace()`  
  - `affordability_filter()`  
These provide full transparency and robustness when working with real-world, messy datasets.

---

# Summary

- Loop control changes the **execution path** inside loops to handle selective skipping, early termination, and error-tolerant iteration.  
- `break` stops the loop immediately.  
- `continue` skips the current iteration only.  
- `pass` does nothing; used as a placeholder.  
- `try`/`except` creates fault-tolerant loops that keep running even when data issues occur.


`````python
# ------------------------------------------------------------------
# 1. The `break` Statement
# `break` stops the loop entirely, regardless of the iteration index
# ------------------------------------------------------------------
# Behavior
# - Loop ends as soon as the condition triggers  
# - Remaining items are not processed  


# Example — Stop when revenue exceeds 2000  

subtotals = [200.98, 350.97, 500, 400, 300, 600, 150, 100, 90]

revenue = 0                          # accumulator to track what the loop produces

for amount in subtotals:
    revenue += amount                # looping through subtotals list and adding revenue
    print(round(revenue,2))          # Visualize each loop
    if revenue > 2000:
        print("Target reached. Stopping early.")
        break

print("Final revenue:", revenue)


# -------------------------------------
# 🟦 Helper — Break Trace Visualizer
# -------------------------------------

def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0
    for idx, value in enumerate(iterable):
        running += value
        print(f"Iteration {idx}: value = {value}, running = {running}")
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break

# Example Usage:
break_trace(subtotals, lambda total: total > 2000)


# Use :.2f formatting in helpers when working with currency or financial data.
# This improves readability and prevents confusion caused by floating-point precision.


# -------------------------------------
# 🟦 Helper — Break Trace Visualizer
# -------------------------------------

def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    Values are formatted to two decimal places for currency-style data.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0.0
    for idx, value in enumerate(iterable):
        running += value
        print(
            f"Iteration {idx}: "
            f"value = ${value:.2f}, "
            f"running = ${running:.2f}"
        )
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break
            
# Example Usage:
break_trace(subtotals, lambda total: total > 2000)
`````

`````python
# ------------------------------------------------------------------
# 1. The `break` Statement — Early Exit from a While Loop
# ------------------------------------------------------------------
# `break` immediately stops a loop, even if the loop condition
# would otherwise continue to evaluate as True.
#
# Use `break` when:
# - you want to stop early once a secondary condition is met
# - continuing the loop no longer makes sense
# - you want to prevent runaway or unrealistic scenarios
# ------------------------------------------------------------------


stock_portfolio = 21_070   # accumulator: tracks the evolving portfolio value (loop state)
year_counter = 0          # counter: tracks how many iterations (years) have passed


# This loop continues WHILE the portfolio is below the target
while stock_portfolio < 150_000:
    
    # Calculate annual investment growth (15% return)
    investment_income = stock_portfolio * 0.15
    
    # Update portfolio value (state change)
    stock_portfolio += investment_income           # accumulator update
    
    # Update loop counter
    year_counter += 1                             # counter update
    
    # Observe the evolving state of the system
    print(
        f"My balance is ${stock_portfolio:,.2f} "
        f"in year {year_counter}"
    )
    
    # SAFETY CHECK:
    # If 10 years have passed and the goal is still not reached,
    # exit the loop early instead of continuing indefinitely.
    if year_counter >= 10:
        print("Goal not reached in 10 years — savings rate may be too low.")
        break

# Teaching takeaway:
# - The while-condition controls whether the loop *can* continue
# - The `break` controls whether the loop *should* continue
`````

`````python
# ------------------------------------------------------------------
# 2. The `continue` Statement  
# `continue` skips the remainder of the current iteration but continues the loop
# ------------------------------------------------------------------
# Behavior
# - `"skis"` and `"skiboots"` are skipped  
# - All other items processed normally
# ------------------------------------------------------------------

# Example — Filter out items containing “ski”


inventory = ["snowboard", "skis", "helmet", "skiboots", "bindings"]
filtered_inventory = []

for item in inventory:
    # filter condition: skip the current item if it contains the substring "ski"
    if "ski" in item:
        # continue means: "do NOT run the rest of the loop body for this item"
        # and move to the next iteration immediately
        continue
    filtered_inventory.append(item)  # .append() empty list accumulator

print(filtered_inventory)
print()


# ---------------------------------------
# 🟦 Helper — Continue Trace Visualizer
# ---------------------------------------

def continue_trace(iterable, skip_fn):
    """
    Prints all iterations and highlights which ones are skipped via 'continue'.
    """
    print("CONTINUE TRACE\n----------------")
    
    for idx, item in enumerate(iterable):
        if skip_fn(item):
            print(f"Iteration {idx}: SKIPPED ({item!r})")
            continue
        
        print(f"Iteration {idx}: processed {item!r}")


# Example Usage:
continue_trace(inventory, lambda x: "ski" in x)
`````

`````python
# ------------------------------------------------------------------
# The `pass` Statement
# `pass` is a placeholder that does NOTHING, but keeps the code valid.
# ------------------------------------------------------------------
# Key idea:
# - `pass` satisfies Python’s requirement that a block cannot be empty
# - It does NOT skip an iteration (that’s `continue`)
# - It does NOT stop a loop (that’s `break`)
# - It simply says: “I’ll fill this in later”
#
# Common use cases:
# - Stubbing logic during development
# - Sketching control flow before implementation
# - Keeping loops and conditionals syntactically valid
# - Incrementally building complex pipelines
# ------------------------------------------------------------------


products = ["Snowboard", "Boots", "Helmet"]

for p in products:
    if p == "Boots":
        pass    # Placeholder: no action yet, loop continues normally
    else:
        print("Processing:", p)


# NOTE:
# `pass` is not control flow.
# It is a structural placeholder.
#

# CONTROL FLOW MENTAL MODEL:
# - break     → exit the loop entirely
# - continue  → skip an iteration
# - pass      → do nothing (for now), but keep structure valid
`````

`````python
# ------------------------------------------------------------------
# Error Handling with `try` / `except`  
# These statements allow the loop to continue running after encountering an error
# ------------------------------------------------------------------
# Notes
# - Strings like `"74.99"` are converted successfully  
# - `0` triggers `ZeroDivisionError`  
# - `None` triggers `TypeError`  
# - Loop does not stop on errors  
# ------------------------------------------------------------------

# ---------------------------------------
# Example 1 — Problematic price list
# ---------------------------------------

price_list = [99.99, "74.99", 0, None, 20]

clean_prices = []

for element in price_list:
    try:
        # Step 1: convert to a numeric value
        numeric_price = float(element)
        
        # Step 2: use the validated value for calculations
        clean_prices.append(round(1 / numeric_price, 4))
    
    except ZeroDivisionError:                                  # zero division error
        print("ZeroDivisionError: price was zero. Skipping.")
    
    except (TypeError, ValueError):                            # multiple error types
        print(f"Invalid price format ({element}). Skipping.")

# ---------------------------------------
# 🟦 Helper — Error Trace Loop Inspector  
# Shows which items cause which errors.
# ---------------------------------------

def try_except_trace(iterable, transform_fn):
    """
    Executes transform_fn on each element, catching errors
    and reporting them with full iteration context.
    """
    print("TRY/EXCEPT TRACE\n-----------------")
    results = []
    
    for idx, value in enumerate(iterable):
        try:
            result = transform_fn(value)
            results.append(result)
            print(f"Iteration {idx}: OK → {result}")
        
        except Exception as e:
            print(f"Iteration {idx}: ERROR → {type(e).__name__}: {e}")
    
    return results


# Example Usage:
try_except_trace(price_list, lambda x: round(1 / float(x), 4))
print()

# ---------------------------------------
# Example 2 — Problematic ratings list
# ---------------------------------------

ratings = ["1 star", "5 star", 
           "3 star", "star 5", 
           "4 star"
]

for rating in ratings:
    try:
        print(int(rating[0]))
    except ValueError:
        print("incorrect data") 



# STUDENT NOTE:
# This helper shows that errors do NOT stop loops by default
# when handled with try/except.
#
# Each iteration is attempted independently:
# - Valid values are transformed and collected
# - Invalid values are reported and skipped
#
# This mirrors real-world data cleaning workflows.
`````

`````python
# ------------------------------------------------------------------
# Error Handling with `try` / `except`  
# These statements allow the loop to continue running after encountering an error
# ------------------------------------------------------------------
# Notes
# - Strings like `"74.99"` are converted successfully  
# - `0` triggers `ZeroDivisionError`  
# - `None` triggers `TypeError`  
# - Loop does not stop on errors  
# ------------------------------------------------------------------

# ---------------------------------------
# Example 3 — Problematic price list
# ---------------------------------------


prices = [5.99, 99.99, "74.99", 0, None, 20, 24.99]
budget = 50

for price in prices:
    if price is None:
        continue

    try:
        numeric_price = float(price)  # convert early (handles strings)

        if numeric_price == 0:
            print("This product is free.")
            continue

        affordable_quantity = budget // numeric_price
        print(f"I can buy {int(affordable_quantity)} of these.")

    except ValueError:
        print("That is not a number.")
    except TypeError:
        # TypeError is unlikely once you skip None, but included for completeness
        print("Invalid type for price.")


# ---------------------------------------
# 🟦 Helper — Robust Affordability Calculator  
# ---------------------------------------

def affordability_filter(prices, budget):
    """
    Returns a list of (idx, raw_price, numeric_price, quantity) for prices that can be evaluated.
    Skips None, non-numeric, and handles zero safely.
    """
    results = []

    for idx, price in enumerate(prices):
        if price is None:
            continue

        try:
            numeric_price = float(price)

            if numeric_price == 0:
                results.append((idx, price, numeric_price, "FREE"))
                continue

            qty = int(budget // numeric_price)
            results.append((idx, price, numeric_price, qty))

        except (ValueError, TypeError):
            results.append((idx, price, None, "INVALID"))

    return results


# Example Usage:
rows = affordability_filter(prices, budget)

for idx, raw, num, qty in rows:
    print(f"Index {idx}: raw={raw!r} | numeric={num!r} | qty={qty}")



# ---------------------------------------
# 🟦 Helper — Error Trace Loop Inspector  
# Shows which items cause which errors.
# ---------------------------------------

def try_except_trace(iterable, transform_fn):
    """
    Executes transform_fn on each element, catching errors
    and reporting them with full iteration context.
    """
    print("TRY/EXCEPT TRACE\n-----------------")
    results = []
    
    for idx, value in enumerate(iterable):
        try:
            result = transform_fn(value)
            results.append(result)
            print(f"Iteration {idx}: OK → {result}")
        
        except Exception as e:
            print(f"Iteration {idx}: ERROR → {type(e).__name__}: {e}")
    
    return results


# Example Usage:
try_except_trace(
    prices,
    lambda p: "FREE" if float(p) == 0 else int(budget // float(p))
)
`````

`````python
# --------------------------------------------------------
# ✅ Analysis-Ready Affordability Toolkit
# - robust parsing (None, strings, bad values)
# - flags FREE / INVALID
# - computes quantity you can buy under a budget
# - returns both detailed rows + a summary dictionary
# --------------------------------------------------------

def affordability_analyze(prices, budget):
    """
    Analyze a mixed-quality price list against a budget.

    Returns
    -------
    rows : list[dict]
        One record per input element with diagnostics:
        idx, raw, numeric_price, status, quantity
    summary : dict
        Counts + quick stats to support analysis decisions.
    """
    rows = []

    # Summary counters
    counts = {
        "OK": 0,
        "FREE": 0,
        "INVALID": 0,
        "SKIPPED_NONE": 0
    }

    valid_prices = []
    valid_quantities = []

    for idx, raw in enumerate(prices):
        # 1) Skip None cleanly (common missing-data pattern)
        if raw is None:
            counts["SKIPPED_NONE"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": None,
                "status": "SKIPPED_NONE",
                "quantity": None
            })
            continue

        # 2) Attempt numeric conversion (handles floats/ints/strings like "74.99")
        try:
            numeric_price = float(raw)
        except (TypeError, ValueError):
            counts["INVALID"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": None,
                "status": "INVALID",
                "quantity": None
            })
            continue

        # 3) Handle zero explicitly (avoid division-by-zero)
        if numeric_price == 0:
            counts["FREE"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": 0.0,
                "status": "FREE",
                "quantity": "FREE"
            })
            continue

        # 4) Negative prices are “valid numbers” but typically invalid business data
        if numeric_price < 0:
            counts["INVALID"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": numeric_price,
                "status": "INVALID_NEGATIVE",
                "quantity": None
            })
            continue

        # 5) OK path: compute how many you can buy (floor division)
        qty = int(budget // numeric_price)

        counts["OK"] += 1
        valid_prices.append(numeric_price)
        valid_quantities.append(qty)

        rows.append({
            "idx": idx,
            "raw": raw,
            "numeric_price": numeric_price,
            "status": "OK",
            "quantity": qty
        })

    # -----------------------------
    # Build summary stats
    # -----------------------------
    summary = {
        "budget": budget,
        "total_inputs": len(prices),
        "counts": counts,
        "valid_price_count": len(valid_prices),
        "min_valid_price": min(valid_prices) if valid_prices else None,
        "max_valid_price": max(valid_prices) if valid_prices else None,
        "avg_valid_price": (sum(valid_prices) / len(valid_prices)) if valid_prices else None,
        "avg_quantity": (sum(valid_quantities) / len(valid_quantities)) if valid_quantities else None,
        "max_quantity": max(valid_quantities) if valid_quantities else None,
    }

    return rows, summary


# --------------------------------------------------------
# ✅ Pretty-printer (optional, keeps your notebook readable)
# --------------------------------------------------------

def print_affordability_report(rows, summary):
    print("AFFORDABILITY REPORT\n--------------------")
    print(f"Budget: ${summary['budget']:,.2f}")
    print(f"Total inputs: {summary['total_inputs']}\n")

    # High-level summary
    c = summary["counts"]
    print("Counts:")
    for k, v in c.items():
        print(f"  {k:>12}: {v}")
    print()

    # Quick stats
    if summary["valid_price_count"] > 0:
        print("Valid price stats:")
        print(f"  min price : ${summary['min_valid_price']:,.2f}")
        print(f"  max price : ${summary['max_valid_price']:,.2f}")
        print(f"  avg price : ${summary['avg_valid_price']:,.2f}")
        print(f"  avg qty   : {summary['avg_quantity']:.2f}")
        print(f"  max qty   : {summary['max_quantity']}")
    else:
        print("No valid prices found.")
    print()

    # Row-by-row detail (traceability)
    print("Row detail:")
    for r in rows:
        idx = r["idx"]
        raw = r["raw"]
        num = r["numeric_price"]
        status = r["status"]
        qty = r["quantity"]

        # consistent formatting
        raw_txt = repr(raw)
        num_txt = "None" if num is None else f"{num:.2f}"
        print(f"  idx {idx:>2}: raw={raw_txt:<10} | numeric={num_txt:>6} | status={status:<15} | qty={qty}")


# --------------------------------------------------------
# ✅ Example usage (your dataset)
# --------------------------------------------------------

prices = [5.99, 99.99, "74.99", 0, None, 20, 24.99]
budget = 50

rows, summary = affordability_analyze(prices, budget)
print_affordability_report(rows, summary)
`````

`````python
# ----------------------------------------------------------
# Control Flow: Truthy & Falsy Values
# Python evaluates conditions by calling bool(value)
# if / while statements rely on this evaluation
# ----------------------------------------------------------

bool(0)
# False   → numeric zero evaluates to False

bool([])
# False   → empty container evaluates to False

bool('hello')
# True    → non-empty string evaluates to True

bool(10)
# True    → non-zero number evaluates to True
`````

# Chapter 07 — Loops  
## Section 7.05 — Loop Control

### Overview
Loop control statements allow you to **modify the normal flow of iteration**, enabling early exits, selective skipping, placeholder logic, and robust error handling.

Four essential loop control mechanisms include:

- **break** — exit a loop immediately  
- **continue** — skip the rest of the current iteration  
- **pass** — placeholder that does nothing  
- **try / except** — handle errors without stopping the loop  

These tools are critical for:

- Data cleaning  
- Error-tolerant ETL pipelines  
- Complex filtering  
- Avoiding infinite loops  
- Creating maintainable, incremental logic  

This section adds new helper functions for debugging, error tracing, and controlled loop execution.

---

# 1. The `break` Statement  

`break` **stops the loop entirely**, regardless of the iteration index.

### Example — Stop when revenue exceeds 2000  
(Maven scenario: revenue becomes high enough before list ends)

```python
subtotals = [200, 350, 500, 400, 300, 600, 150, 100, 90]

revenue = 0

for amount in subtotals:
    revenue += amount
    
    if revenue > 2000:
        print("Target reached. Stopping early.")
        break

print("Final revenue:", revenue)
```

### Behavior
- Loop ends as soon as the condition triggers  
- Remaining items are **not** processed  

### 🟦 Helper — Break Trace Visualizer

```python
def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0
    for idx, value in enumerate(iterable):
        running += value
        print(f"Iteration {idx}: value = {value}, running = {running}")
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break
```

Usage:

```python
break_trace(subtotals, lambda total: total > 2000)
```

---

# 2. The `continue` Statement  

`continue` **skips the remainder of the current iteration** but continues the loop.

### Example — Filter out items containing “ski”

```python
inventory = ["snowboard", "skis", "helmet", "skiboots", "bindings"]
snowboards = []

for item in inventory:
    if "ski" in item:      # skip skis and skiboots
        continue
    snowboards.append(item)

snowboards
```

### Behavior
- `"skis"` and `"skiboots"` are skipped  
- All other items processed normally  

### 🟦 Helper — Continue Trace Visualizer

```python
def continue_trace(iterable, skip_fn):
    """
    Prints all iterations and highlights which ones are skipped via 'continue'.
    """
    print("CONTINUE TRACE\n----------------")
    
    for idx, item in enumerate(iterable):
        if skip_fn(item):
            print(f"Iteration {idx}: SKIPPED ({item!r})")
            continue
        
        print(f"Iteration {idx}: processed {item!r}")
```

Usage:

```python
continue_trace(inventory, lambda x: "ski" in x)
```

---

# 3. The `pass` Statement  

`pass` is a **placeholder** allowing your code to run while you build logic incrementally.

### Example — Outline structure without full logic

```python
products = ["Snowboard", "Boots", "Helmet"]

for p in products:
    if p == "Boots":
        pass    # logic will be added later
    else:
        print("Processing:", p)
```

Use cases:

- Stubbing logic during development  
- Maintaining syntactically valid loops  
- Gradually building complex pipelines  

---

# 4. Error Handling with `try` / `except`  

These statements allow the loop to **continue running after encountering an error**.

### Example — Problematic price list

```python
price_list = [99.99, "74.99", 0, None, 20]

clean_prices = []

for price in price_list:
    try:
        # Convert strings to float and validate nonzero price
        numeric_price = float(price)
        clean_prices.append(round(1 / numeric_price, 4))
    
    except ZeroDivisionError:
        print("ZeroDivisionError: price was zero. Skipping.")
    
    except (TypeError, ValueError):
        print(f"Invalid price format ({price}). Skipping.")
```

### Notes
- Strings like `"74.99"` are converted successfully  
- `0` triggers `ZeroDivisionError`  
- `None` triggers `TypeError`  
- Loop does **not stop** on errors

# 5. 🟦 Helper — Error Trace Loop Inspector  
Shows which items cause which errors.

# 6. 🟦 Helper — Robust Affordability Calculator  


```python
def affordability_filter(prices, budget):
    """
    Robust affordability calculator:
    - skips None values
    - converts string prices
    - handles invalid formats safely
    - avoids division-by-zero issues
    """
    
    affordable = []
    
    for idx, price in enumerate(prices):
        try:
            if price is None:
                continue
            
            numeric_price = float(price)
            
            if numeric_price <= budget:
                affordable.append(numeric_price)
        
        except Exception:
            continue  # safely skip invalid price formats
    
    return affordable
```

---

# When to Use Loop Control Statements

### Use `break` when:
- A target or threshold is reached  
- Further processing is unnecessary  
- You are preventing a runaway or infinite loops early  

### Use `continue` when:
- Filtering datasets  
- Skipping invalid or unwanted entries, i.e., excluding values you don't want to process in a loop
- Differentiating “skippable” vs “processable” values

### Use `pass` when:
- Structuring logic before implementation, i.e., avoiding run errors with incomplete code logic
- Writing placeholder blocks, i.e., serving as a placeholder for future code  
- Maintaining syntactic validity  

### Use `try`/`except` when:
- Data may be incomplete, inconsistent, or malformed  
- You need to avoid interruption during iteration
- You need help with error and exception handling  
- Building robust ETL or cleaning pipelines  

---

# Key Takeaways

- Loop control changes the **execution path** inside loops to handle selective skipping, early termination, and error-tolerant iteration.  
- `break` stops the loop immediately.  
- `continue` skips the current iteration only.  
- `pass` does nothing; used as a placeholder.  
- `try`/`except` creates fault-tolerant loops that keep running even when data issues occur.  
- The helper toolkit now includes:  
  - `break_trace()`  
  - `continue_trace()`  
  - `try_except_trace()`  
  - `affordability_filter()`  
These provide full transparency and robustness when working with real-world, messy datasets.



# 🧭 Chapter 07: Loops. Capstone Reference — Iteration Strategy & Positional Alignment

## When Index-Based Looping Is Required (4 Canonical Cases)
1. In-place modification
2. Neighbor access (i-1, i+1)
3. Controlled stepping / sampling
4. Parallel write-back across aligned lists

## When NOT to Use Index-Based Looping
(Case A, B, C — your section)

## Decision Table

## Alignment Validation & Trace Helpers

# 🧠 When to Use Index-Based Looping (A Practical Reference Sheet)

Goal:
Use index-based looping **only when the problem truly requires positional control**.
Avoid it as a default. Use it deliberately.

This sheet documents the **four canonical cases** where index-based looping is justified,
and what safer alternatives exist in all other situations.

---

## ✏️ Case 1 — In-Place Modification (Write Back by Position)

### Problem
You must **modify elements of a list in place**.

### Example
```python
for i in range(len(A)):
    A[i] = A[i] * 2
```
### 1. You need to CHANGE the list in place  
### ✏️ Case 1 — In-place modification (write back by position)
(You write back to the same spot)

```python
for i in range(len(scores)):
    scores[i] = scores[i] * 1.1   # give everyone a 10% bonus
```

### Why index is required
- You need `A[i]` on the left-hand side.
- You are overwriting the original container.

### Common in
- normalization
- scaling
- in-place preprocessing

### Alternative?
None. This is a legitimate index-based case.

---

## ⏱️ Case 2 — Neighbor / Relative Position Access (i-1, i+1)

### Problem
You need **adjacent elements**.

### Example
```python
for i in range(1, len(A)):
    diff = A[i] - A[i-1]
```
### ⏱️ Case 2 — Neighbor access: compare each day to the previous day
### Goal: compute day-to-day temperature change
* Showing the previous or next position, comparing position i with i-1 (or i+1)

```python
temps = [70, 72, 68, 75, 74, 78]

print("Day-to-day temperature changes:")
print("-------------------------------")

for i in range(1, len(temps)):
    today = temps[i]           # temperature on day i
    yesterday = temps[i - 1]  # temperature on day i-1

    change = today - yesterday

    print(f"Day {i}: {yesterday}° → {today}°   change = {change:+}°")
```

### Why index is required
- You need relative positions.
- You are computing differences, windows, or gradients.

### Common in
- time series
- finite differences
- rolling statistics
- signal processing

### Alternative?
Sometimes `zip(A[1:], A[:-1])`, but index is often clearer.

---

## 🔁 Case 3 — Controlled Stepping / Striding

### Problem
You must **control which indices are visited**.

### Example
```python
for i in range(0, len(A), 2):
    print(A[i])
```

### 🔁 Case 3 — Controlled stepping: sample every k-th element
### Goal: inspect stock prices every 5 trading days (weekly sampling)
```python
prices = [101, 103, 102, 104, 105, 107, 106, 108, 110, 109,
          111, 112, 113, 115, 114, 116, 118, 117, 119, 120]

step = 5   # sample every 5 days

print("Weekly sampled prices:")
print("-----------------------")

for i in range(0, len(prices), step):
    day = i
    price = prices[i]

    print(f"Day {day:>2}: price = {price}")
```

### Why index is required
- You need explicit stride logic.
- You are skipping elements by design.

### Common in
- downsampling
- batching
- block processing

### Alternative?
Occasionally slicing, but index-based is often clearest.

---

## 🔗 Case 4 — Positional Synchronization with an External Structure

### Problem
You must **synchronize by position** with another structure.

### Example
```python
for i in range(len(x)):
    y[i] = f(x[i])
```
### 🔗 Case 4 — Positional synchronization across multiple lists
### Goal: compute discounted prices using matching positions
* One index controls multiple structures:

```python
price = prices[i]
rate = discounts[i]
discounted[i] = final_price

prices = [100, 80, 50, 120, 200]
discounts = [0.10, 0.20, 0.00, 0.15, 0.05]   # per-item discount rates

# Prepare output list (same length as prices)
discounted = [0] * len(prices)

print("Price  Discount  Final Price")
print("-----------------------------")

for i in range(len(prices)):
    price = prices[i]           # price at position i
    rate = discounts[i]        # discount at the same position i

    final_price = price * (1 - rate)

    discounted[i] = final_price   # write back by the same index

    print(f"{price:>5}   {rate:>7.2f}   {final_price:>11.2f}")
```


### Why index is required
- You are writing into a second array by the same index.
- You are implementing an algorithm that depends on position.

### Common in
- dynamic programming
- matrix algorithms
- in-place transforms
- DP tables

### Alternative?
Sometimes `zip()` if no write-back is needed.

---

# 🚫 When NOT to Use Index-Based Looping

These are the cases where `range(len(...))` is a *bad default*.

---

## Case A — You Only Need Values

### Prefer
```python
for name in names:
    print(name)
```

### Not
```python
for i in range(len(names)):
    print(names[i])
```

---

## Case B — You Need Index + Value

### Prefer
```python
for i, grade in enumerate(grades):
    print(f"Student {i}: {grade}")
```

### Not
```python
for i in range(len(grades)):
    print(f"Student {i}: {grades[i]}")
```

---

## Case C — You Have Parallel Lists

### Prefer
```python
for city, pop in zip(cities, populations):
    print(city, pop)
```

or

```python
for i, (city, pop) in enumerate(zip(cities, populations)):
    ...
```

### Not
```python
for i in range(len(cities)):
    print(cities[i], populations[i])
```

---

# 🧭 Decision Table

| Situation                                   | Best Tool             |
|--------------------------------------------|-----------------------|
| Only need values                           | `for value in seq`   |
| Need index + value                        | `enumerate()`        |
| Parallel lists                            | `zip()`              |
| In-place modification                    | `range(len())`       |
| Neighbor access (i-1, i+1)                | `range(len())`       |
| Controlled stepping / stride              | `range(start, stop, step)` |
| Positional write-back / DP               | `range(len())`       |

---

# 🧠 Core Principle

> **Do not use indices because you can.  
> Use indices because the problem demands them.**

And always remember:

- Humans are poor at manual index bookkeeping.
- Instrument index-based loops with:
  - `enumerate()`
  - trace helpers
  - alignment checks
  - assertions

Goal:
- Make index state observable.
- Let the machine show you the truth.

# 📘Chapter 08 — Dictionaries & Sets  
## Section 8.01 — Foundations  

### Preview
Dictionaries and sets introduce **key-based access**, **uniqueness**, and **fast membership testing**.
They replace index-based thinking with **semantic access**, which is foundational for analytics, data validation, and real-world data modeling.

In earlier chapters, lists required you to:
- search linearly
- rely on positional meaning
- manually track relationships between values

## 🟦 Helper — `trace_dict_lookup`
This helper makes dictionary access **visible and safe**.
It shows:
- what key was requested
- whether the key exists
- what value was returned (or why it was not)

This prevents silent failures and reinforces best practices early.

```python
def trace_dict_lookup(data, key):
    """
    Safely traces dictionary access.
    Shows whether the key exists and what value is returned.
    """
    if key in data:
        print(f"Key found: '{key}' → Value: {data[key]}")
        return data[key]
    else:
        print(f"Key NOT found: '{key}'")
        return None
```

### Example Usage
```python
trace_dict_lookup(inventory_status, 'boots')
trace_dict_lookup(inventory_status, 'bindings')
```

**Output**
```text
Key found: 'boots' → Value: in stock
Key NOT found: 'bindings'
```

---

## 🧭 Summary
- Dictionaries replace positional access with **semantic access**
- Keys must be unique and are typically strings or numbers
- Lookup is fast and explicit
- Membership testing (`key in dict`) is foundational
- Dictionaries model real-world data more naturally than lists


`````python
# -----------------------------------------------------------------
# # Section 8.01 — lesson 1. Dictionary Basics
# Introduces dictionaries by contrasting them with lists,
# showing how key-based lookup replaces searching through nested lists
# -----------------------------------------------------------------


# -----------------------------------------------
# List limitations: accessing values via indexing
# -----------------------------------------------

inventory_status_list = [
    ["skis", "in stock"],
    ["snowboard", "sold out"],
    ["goggles", "sold out"],
    ["boots", "in stock"]
]

item_status = "goggles"

# Loop through items in nested list:
for lst in inventory_status_list:
    if lst[0] == item_status:
        print(lst[1])


# -----------------------------------------------
# Dictionary Baseline Example
# -----------------------------------------------
# Keys must be unique and immutable
# Values do not need to be unique and can be any data type


# Inventory status stored as key-value pairs:
inventory_status_dict = {
    'skis': 'in stock',
    'snowboard': 'sold out',
    'goggles': 'sold out',
    'boots': 'in stock'
}


# No loop required: direct key-based lookup
print(inventory_status_dict['goggles'])

# Output: 'sold out'


# -----------------------------------------------
# 🟦 Helper — `trace_dict_lookup`
# -----------------------------------------------
# This helper makes dictionary access visible and safe
# It shows:
# - what key was requested
# - whether the key exists
# - what value was returned (or why it was not)

# This prevents silent failures and reinforces best practices early.


def trace_dict_lookup(data, key):
    """
    Safely traces dictionary access.
    Shows whether the key exists and what value is returned.
    """
    if key in data:
        print(f"Key found: '{key}' -> Value: {data[key]}")
        return data[key]
    else:
        print(f"Key NOT found: '{key}'")
        return None

# Example Usage:
trace_dict_lookup(inventory_status_dict, "goggles")
trace_dict_lookup(inventory_status_dict, "boots")
trace_dict_lookup(inventory_status_dict, "bindings")



# -----------------------------------------------
# Dictionary values can also be complex data structures (like lists)
# -----------------------------------------------
# retrieve the list by looking up its key
# retrieve the list element by using its index


item_details_dict = {"skis": [249.99, 10, "in stock"], 
               "snowboard": [219.99, 0, "sold out"], 
               "goggles": [99.99, 0, "sold out"], 
               "boots": [79.99, 7, "in stock"]}

print(item_details_dict["skis"])      # returns a list [249.99, 10, "in stock"]
print(item_details_dict["skis"][0])   # returns first element in the list
`````

## Section 8.02 — Modifying Dictionaries (Mutation Awareness)

### Focus
Understanding how dictionaries change over time and why mutation must be intentional.

### Concepts Covered
- Adding new key–value pairs
- Overwriting existing keys
- Deleting keys with `del`
- Distinguishing creation vs modification

### Analytics Framing
Dictionaries often represent:
- evolving records
- stateful configurations
- intermediate results in pipelines

Understanding mutation is critical to avoiding silent logic errors.




`````python
# ---------------------------------------------------------------
# Section 8.02 — Modifying Dictionaries
# Referencing a new key and assigning it a value will add a new key-value pair,
# while referencing an existing key will overwrite the existing pair.
# ---------------------------------------------------------------

# Example:

item_details = {"skis": [249.99, 10, "in stock"], 
               "snowboard": [219.99, 0, "sold out"], 
               "goggles": [99.99, 0, "sold out"], 
               "boots": [79.99, 7, "in stock"]}

item_details["bindings"] = [149.99, 4, "in stock"]  # assigning a value to a new key will add the key-value pair to the dictionary
print(item_details)

item_details["bindings"] = [139.99, 0, "out of stock"]  # assigning an existing key a new value will overwrite the existing pair
print(item_details)



# ------------------------------------------------------------
# Use the del keyword to delete key-value pairs
# Removing a key deletes both the key and its associated value
# ------------------------------------------------------------

print(item_details)  # state before deletion

del item_details["boots"]

print(item_details)        # state after deletion
`````

`````python
# ---------------------------------------------------------------
# # Section 8.02 — Fundamental Dictionary Operations Demo
# ---------------------------------------------------------------

product_prices = {"skis": 249.99, 
               "snowboard": 219.99,  
               "goggles": 99.99, 
               "boots": [79.99, 89.99]
}

print(product_prices)


#-----------------------------------
# Lookup value for a given key:
#-----------------------------------
# product_prices["hat"]                   # KeyError for a key that doesn't exist
print(product_prices["boots"][1])         # returns second element in the list, 89.99
print(product_prices["snowboard"])        # returns value, 219.99 


#------------------------------------------------------
# Membership test for key in dictionary: True or False
#------------------------------------------------------
print("skis" in product_prices)
print("hat" in product_prices)


#-----------------------------------
# Loop through keys in dictionary:
#-----------------------------------
# Looping over a dictionary iterates over its keys by default

for product in product_prices:
    print(product)


# ---------------------------------------------------------------
# Increase price by 10% on skis
# ---------------------------------------------------------------
# assigning an existing key a new value will overwrite the existing pair

product_prices["skis"] = round(product_prices["skis"] * 1.1, 2)
print(product_prices["skis"])
# Output: 274.99

# ---------------------------------------------------------------
# Adding a new key-value pair to the dictionary
# ---------------------------------------------------------------
# assigning a value to a new key will add the key-value pair to the dictionary
product_prices["hat"] = 14.99                
print(product_prices)


# ---------------------------------------------
# Adding new lists to the dictionary 
# by creating new key-value pairs
# using 'enumerate()'
# ---------------------------------------------
product_prices = {"skis": 249.99, 
               "snowboard": 219.99, 
               "goggles": 99.99, 
               "boots": [79.99, 89.99]
}

new_products = ["coffee", "helmet"]               # new keys
new_prices = [5.99, 59.99]                        # new values

# Use enumerate() to pair each new product with its corresponding price:
for i, product in enumerate(new_products):        # loop through index and the product in the new_products list
    product_prices[product] = new_prices[i]       # create new key and value pairs

print(product_prices)


# ---------------------------------------------
# Deleting a dictionary key vs mutating a value
# ---------------------------------------------

print(product_prices)          # state before deletion

del product_prices["coffee"]  # deletes key-value pair

print(product_prices)        # state after deletion



print(product_prices)              # state before deletion

del product_prices["boots"][1]     # This deletes an element inside the list stored as the value, not the dictionary key itself 

print(product_prices)             # state after deletion of a value in list


# Alternatively: Assign a new value to key

product_prices["boots"] = 79.99

print(product_prices)
`````

`````python
# ---------------------------------------------
# # Section 8.02 — Adding new lists to a dictionary 
# by creating new key-value pairs using 'enumerate()'
# ---------------------------------------------

inventory_status = {}

# Assumes 'items' and 'inventory' lists are aligned by index
items = ['skis', 'snowboard', 'goggles', 'boots']  # new keys
inventory = [10, 0, 0, 7]                          # new values

# Use enumerate() to pair each new product with its corresponding price:
for i, item in enumerate(items):                  # loop through index and the item in the items list
    inventory_status[item] = inventory[i]

print(inventory_status)


# ---------------------------------------------
# # Section 8.02 — Adding new lists to a dictionary 
# by creating new key-value pairs using 
# 'enumerate()' and conditional logic
# ---------------------------------------------

inventory_status = {}

# Assumes 'items' and 'inventory' lists are aligned by index
items = ['skis', 'snowboard', 'goggles', 'boots']  # new keys
inventory = [10, 0, 0, 7]                          # new values

# Use enumerate() to pair each new product with its corresponding price:
for i, item in enumerate(items):                  # loop through index and the item in the items list
    if inventory[i] == 0:                         # use conditional logic to determine if element is 0 or not
        inventory_status[item] = "sold out"       # if element is 0, create new key-value in pair(s) in dictionary (mark as "sold out")
    else:
        inventory_status[item] = "in stock"       # if element is not 0, create new key-value pairs(s) in dictionary

print(inventory_status)
`````

## Section 8.03 — Dictionary Views: Keys, Values, and Items

### Focus
Learning how Python exposes dictionary contents safely and efficiently.

### Concepts Covered
- `.keys()`
- `.values()`
- `.items()`
- Dictionary view objects vs lists
- Iterating over each view

### Analytics Framing
These views mirror:
- column labels
- feature vectors
- row-level records (key–value pairs)

They are the gateway to transformation and reporting patterns.


`````python
#--------------------------------------------------------------------------------
# View object is a dynamic, live view of the dictionary's keys, values, or items. 
# It updates automatically if the dictionary changes, it's memory-efficient, and it's iterable.
#--------------------------------------------------------------------------------


#-----------------------------
# .keys() method
#-----------------------------
# The .keys() method returns a *view object* that
# represents the keys from a dictionary as a view object (not a list)

item_details = {"skis": [249.99, 10, "in stock"], 
               "snowboard": [219.99, 0, "sold out"], 
               "goggles": [99.99, 0, "sold out"], 
               "boots": [79.99, 7, "in stock"]}

item_details.keys()  
# Output: dict_keys(['skis', 'snowboard', 'goggles', 'boots'])


#---------------------------------------------
# This *view object* can be iterated through,
# which has the same behavior as looping
# through the dictionary keys directly
#---------------------------------------------

for item in item_details.keys():
    print(item)
    
# Ouput:
# skis
# snowboard
# goggles
# boots


#---------------------------------------------
# This *view object* can be converted into
# a list, a tuple, or a set if needed
#---------------------------------------------
# when you need a static copy or need specific
# functionality of a particular container type

key_list = list(item_details.keys())
print(key_list)

key_tuple = tuple(item_details.keys())
print(key_tuple)

key_set = set(item_details.keys())
print(key_set)
`````

`````python
#--------------------------------------------------------------------------------
# View object is a dynamic, live view of the dictionary's keys, values, or items. 
# It updates automatically if the dictionary changes, it's memory-efficient, and it's iterable.
#--------------------------------------------------------------------------------


#-----------------------------
# .values() method
#-----------------------------

# The .values() method returns a *view object* that
# represents the values from a dictionary as a view object (not a list)

item_details = {"skis": [249.99, 10, "in stock"], 
               "snowboard": [219.99, 0, "sold out"], 
               "goggles": [99.99, 0, "sold out"], 
               "boots": [79.99, 7, "in stock"]}

print(item_details.values())
# Output: dict_values([[249.99, 10, 'in stock'], [219.99, 0, 'sold out'], [99.99, 0, 'sold out'], [79.99, 7, 'in stock']])


#---------------------------------------------
# This *view object* can be looped through
#---------------------------------------------
# Extract the price from each value list and append to a new list

price_list = []      # new list container

for element in item_details.values():
    price_list.append(element[0])

print(price_list)
# Output: [249.99, 219.99, 99.99, 79.99]


#-----------------------------------------------------------
# If you want all prices (extract index 0 from every value):
# Iterate (either over the dict itself or over .values())
#-----------------------------------------------------------

# Option 1: Loop over the dictionary (recommended — keeps keys if you need them)
pricing = [item_details[element][0] for element in item_details]
print(pricing)
# Output: [249.99, 219.99, 99.99, 79.99]


# Option 2:
pricings = [value[0] for value in item_details.values()]
print(pricings)
# Output: [249.99, 219.99, 99.99, 79.99]


# Option 3: If you want a set of unique prices
unique_prices = {value[0] for value in item_details.values()}
print(unique_prices)


#---------------------------------------------
# This *view object* can be converted into
# a list, a tuple, or a set if needed
#---------------------------------------------
# when you need a static copy or need specific
# functionality of a particular container type


values_list = list(item_details.values())    # For indexing/slicing/modifying
print(values_list)

values_tuple = tuple(item_details.values())  # For immutability or as a dict key
print(values_tuple)

# values_set = set(item_details.values())    ❌  # TypeError: (mutable) unhashable type 'list'

# If you want a set of unique inner lists → Convert them to tuples first
# Since tuples are immutable and hashable:

values_set = set(tuple(value) for value in item_details.values())  # for uniqueness or set operations
print(values_set)

# If you just want unique set of prices (or some other single element)

unique_prices = {value[0] for value in item_details.values()}
# Result: {249.99, 219.99, 99.99, 79.99}

unique_statuses = {value[2] for value in item_details.values()}
# {'in stock', 'sold out'}
`````

`````python
#--------------------------------------------------------------------------------
# View object is a dynamic, live view of the dictionary's keys, values, or items. 
# It updates automatically if the dictionary changes, it's memory-efficient, and it's iterable.
#--------------------------------------------------------------------------------


#-----------------------------
# .items() method
#-----------------------------

# The .items() returns a *view object* that represents the
# dictionary’s key-value pairs as a *view object* of (key, value) tuples


item_details = {"skis": [249.99, 10, "in stock"], 
               "snowboard": [219.99, 0, "sold out"], 
               "goggles": [99.99, 0, "sold out"], 
               "boots": [79.99, 7, "in stock"]}

item_details.items() 

# Output:
# dict_items([('skis', [249.99, 10, 'in stock']), ('snowboard', [219.99, 0, 'sold out']), etc.


#-----------------------------------
# The *view object* can be unpacked
#-----------------------------------

# Unpack the tuple to retrieve individual keys and values:

for key, value in item_details.items():
    print(f"The {key} costs {value[0]}.")


#-----------------------------------
# Build parallel lists from .items()
#-----------------------------------
# useful for plotting later

items = []
prices = []

for key, value in item_details.items():
    items.append(key)        # item name
    prices.append(value[0])  # price

print(items)
print(prices)



#---------------------------------------------
# This *view object* can be converted into
# a list, a tuple, or a set if needed
#---------------------------------------------
# when you need a static copy or need specific
# functionality of a particular container type

item_details = {"skis": [249.99, 10, "in stock"], 
               "snowboard": [219.99, 0, "sold out"], 
               "goggles": [99.99, 0, "sold out"], 
               "boots": [79.99, 7, "in stock"]}



key_item_list = list(item_details.items())
print(key_item_list)

key_item_tuple = tuple(item_details.items())
print(key_item_tuple)

# key_item_set = set(item_details.items())   ❌ # TypeError: (mutable) unhashable type 'list'
# print(key_item_set)

# Only if you really need a set:
key_item_set = {(k, tuple(v)) for k, v in item_details.items()}
print(key_item_set)  # Works now
`````

`````python
#-----------------------------------------------------
# Safe Value Retrieval: .get() method safely returns 
# the actual value for a given key
#-----------------------------------------------------
# .get() is a lookup, not a *view*

item_details = {"skis": [249.99, 10, "in stock"], 
               "snowboard": [219.99, 0, "sold out"], 
               "goggles": [99.99, 0, "sold out"], 
               "boots": [79.99, 7, "in stock"]}

print(item_details.get("boots")) 

# Output: [79.99, 7, 'in stock']

#------------------------------------------------------------
# If the key is not found, it returns None by default
# or a specified default value if provided
#------------------------------------------------------------
# .get(key, optional value to return if key not found)

print(item_details.get("skis"))

item_details.get("hats")  # returns None; loop continues without breaking, whereas:
# item_details["hats"]  # would raise a KeyError

item_details.get("hats", "Sorry, but we do not carry that item.")
`````

`````python
#-----------------------------------------------------
# .update() method adds or overwrites key-value pairs
#-----------------------------------------------------

# .update() a single pair for a key of "bindings"

item_details = {"skis": [249.99, 10, "in stock"], 
               "snowboard": [219.99, 0, "sold out"], 
               "goggles": [99.99, 0, "sold out"], 
               "boots": [79.99, 7, "in stock"]}

item_details.update({"bindings": [139.99, 0, "out of stock"]})
print(item_details)

#--------------------------------------------------------
# .update() to combine dictionaries or add multiple items
#--------------------------------------------------------

more_item_details = {"scarf": [19.99, 100, "in stock"], "snowpants": "N/A"}  # reminder: values do not need to be the same type
item_details.update(more_item_details)
print(item_details)
`````

`````python
#------------------------------------------------------
# Section 8.03 — Dictionary Methods Demo
#------------------------------------------------------

customer_sales = {
    "C00001": [224.99, 87.36], 
    "C00004": [125.56], 
    "C00014": [99.99, 104.26, 23.57], 
    "C00015": [82.65, 56.88, 201.82, 519.55]
}


#--------------------------------------------
# Extract the keys into a list, tuple, or set
#--------------------------------------------
customer_ids = list(customer_sales.keys())
print(customer_ids)
customer_ids = tuple(customer_sales.keys())
print(customer_ids)
customer_ids = set(customer_sales.keys())
print(customer_ids)


#--------------------------------------------
# Extract the values into a nested list and tuple
#--------------------------------------------
print(list(customer_sales.values()))
print(tuple(customer_sales.values()))

#----------------------------------------------
# Sum the values for each of the customer_sales:
#----------------------------------------------

for sales in customer_sales.values():
    print(sum(sales))
    
# Output:
# 312.35
# 125.56
# 227.82
# 860.9


#-----------------------------------
# Build dictionary with keys and 
# the sum of values using .items()
#-----------------------------------
# This aggregates the values stored
# in a list within a dictionary


customer_sales = {
    "C00001": [224.99, 87.36], 
    "C00004": [125.56], 
    "C00014": [99.99, 104.26, 23.57], 
    "C00015": [82.65, 56.88, 201.82, 519.55]
}

total_sales = {}

for customer_id, sales in customer_sales.items():
    total_sales[customer_id] = sum(sales)
print(total_sales)

# Output: {'C00001': 312.35, 'C00004': 125.56, 'C00014': 227.82, 'C00015': 860.9}


#-------------------------------------------------------
# .get() method safely returns the value for a given key
#-------------------------------------------------------

vip_customers = ["C00001", "C00003", "C00014"]
# total_sales: {'C00001': 312.35, 'C00004': 125.56, 'C00014': 227.82, 'C00015': 860.9}

# for customer in vip_customers:
#     print(total_sales[customer])                      # returns KeyError

for customer in vip_customers:                          #  ["C00001", "C00003", "C00014"]
    print(total_sales.get(customer, "No longer VIP"))

# Output:
# 312.35
# None             # if key is not found, returns None by default or optional message (e.g., "No longer VIP")
# 227.82


#--------------------------------------------------------
# .update() to combine dictionaries or add multiple items
#--------------------------------------------------------
# Adds new customers; existing IDs would be overwritten if duplicated


new_customers = {"C00005": [105.44], 
                 "C00022": [28.99]
}

customer_sales.update(new_customers)
print(customer_sales)
`````

## Section 8.04 — Iterating Over Dictionaries (Patterns That Matter)

### Preview
Dictionaries are most powerful when used inside loops.
However, how you iterate over a dictionary directly affects:
- readability
- correctness
- semantic clarity

This section clarifies **what Python iterates over by default**
and shows the three canonical iteration patterns.

---

### Core Concepts
- Iterating over keys (default behavior)
- Iterating over values explicitly
- Iterating over key–value pairs with `.items()`
- Choosing variable names that preserve meaning

---

### Why This Matters (Analytics Framing)
In analytics, loops are used to:
- aggregate values
- apply transformations
- enforce conditions
- build new structures

Clear iteration patterns prevent subtle logic errors
and make analytical intent obvious.


`````python
# --------------------------------------------------
# # Section 8.04 — Iterating over Dictionaries: Patterns that matter
# --------------------------------------------------

# --------------------------------------------------
# Iteration Pattern 1 — Looping Over Keys (Default) 🔑
# --------------------------------------------------

product_prices = {
    "skis": 249.99,
    "boots": 79.99,
    "goggles": 99.99
}

# Looping over a dictionary iterates over keys by default
for product in product_prices:
    print(product)


# **What this means**
# - Python assumes you want keys
# - This is equivalent to `for product in product_prices.keys()`🔑

# Use this when:
# - keys are the primary identifier 🔑
# - values are accessed conditionally or later


# --------------------------------------------------
# Iteration Pattern 2 — Looping Over Values 🧮
# --------------------------------------------------

# Loop explicitly over values
for price in product_prices.values():
    print(price)
    

# Use this when:
# - keys are irrelevant
# - only the stored values matter 🧮
# - performing numeric analysis or aggregation


# ---------------------------------------------------
# Iteration Pattern 3 — Looping Over Key–Value Pairs 🔗
# ---------------------------------------------------

# Loop over both keys and values using tuple unpacking
for product, price in product_prices.items():
    print(f"{product}: {price}")


# **Why this is powerful**
# - Preserves full context
# - `.items()` is ideal when relationships matter and make them explicit 🔗
# - Common in aggregation and transformation logic

# This is the most common pattern in analytics workflows 🎯


#--------------------------------------------
# # Section 8.04 — Naming Conventions Matter 🔗
#--------------------------------------------

customer_totals = {
    "C00001": [224.99, 87.36], 
    "C00004": [125.56], 
    "C00014": [99.99, 104.26, 23.57], 
    "C00015": [82.65, 56.88, 201.82, 519.55]
}


# Clear, semantic variable names 🔗
for customer_id, total in customer_totals.items():
    print(customer_id, total)

# Avoid:
# for x, y in customer_totals.items():
# Names should reflect meaning, not structure
`````

## Section 8.05 — Safe Access Patterns: `[]` vs `.get()`

### Preview
Dictionaries provide fast, semantic access to data using keys.
However, not all keys can be assumed to exist.

This section introduces **safe access patterns** for dictionaries and explains
when Python should:
- fail loudly (raise an error), or
- fail safely (continue execution)

Understanding this distinction is critical for analytics, where missing or optional
data is normal rather than exceptional.

---

### Core Concepts
- `dict[key]` performs a **direct lookup**
- Missing keys with `[]` raise a `KeyError`
- `.get()` performs a **safe lookup**
- `.get()` returns:
  - the actual value if the key exists
  - `None` by default if it does not
  - or a specified default value if provided

> Important distinction:  
> `.get()` is a **lookup**, not a *view*.  
> It returns the actual stored value, not a dynamic object.

---

### Why This Matters (Analytics Framing)
Real-world datasets are messy:
- columns may be optional
- records may be incomplete
- categories may appear inconsistently

In analytics pipelines:
- crashing on missing data is often undesirable
- silent failure may be acceptable **only if intentional**

Safe access patterns let you choose the right behavior.


`````python
# -----------------------------------------------------
# Section 8.05 — Safe Access Patterns: `[]` vs `.get()`
# -----------------------------------------------------


# Baseline Dictionary

item_details = {
    "skis": [249.99, 10, "in stock"],
    "snowboard": [219.99, 0, "sold out"],
    "goggles": [99.99, 0, "sold out"],
    "boots": [79.99, 7, "in stock"]
}

# --------------------------------------
# Direct Lookup with `[]` (Fail Loudly) ⚠️
# --------------------------------------

# Direct lookup assumes the key exists
print(item_details["boots"])   # returns the value

# Uncommenting the next line would raise a KeyError
# print(item_details["hats"])


# --------------------------------------
# Safe Lookup with `.get()` (Fail Safely) 🚧
# --------------------------------------

print(item_details.get("boots"))    # returns the value
print(item_details.get("hats"))     # returns None

# This is often the preferred behavior in analytics workflows.


# --------------------------------------
# Providing a Default Value
# --------------------------------------

# Provide a fallback value if the key is not found
print(item_details.get("hats", "Item not carried"))

# Defaults can be:
# - strings
# - numbers
# - empty lists
# - sentinel values (e.g., `"UNKNOWN"`)


# ------------------------------------------------
# ⚖️ Choosing Between dict[key] and dict.get(key)
# ------------------------------------------------

# - Decision guideline:
#   - Use dict[key] when the key must exist (missing key indicates a bug or invalid state — let it raise KeyError to fail fast).
#   - Use dict.get(key, default) when the key may be absent (handle gracefully with a default value, e.g., None or sentinel value).

# This reflects code intent: explicit error for invariants vs. safe handling for optional data ⚖️
`````

## Section 8.06 — Building Dictionaries Dynamically

### Preview
In practice, dictionaries are rarely written out by hand.
They are usually **constructed dynamically** from:
- loops
- multiple data sources
- intermediate results

This section focuses on how dictionaries are *built*, not just how they are accessed.

Dynamic construction is foundational for:
- analytics pipelines
- feature engineering
- reshaping raw data into structured formats

---

### Core Concepts
- Using empty dictionaries as accumulators
- Adding key–value pairs inside loops
- Building dictionaries from multiple sequences
- Using `zip()` to align related data safely

---

### Why This Matters (Analytics Framing)
Real-world data often arrives as:
- parallel lists
- IDs and attributes
- raw records that must be reshaped

Being able to **programmatically assemble dictionaries** allows you to:
- map identifiers to values
- construct features
- prepare data for pandas and visualization tools

This is one of the most common dictionary patterns in analytics.


`````python
# ---------------------------------------------
# Section 8.06 — Adding new lists to a dictionary
# by creating new key-value pairs using 'enumerate()'
# ---------------------------------------------

inventory_status = {}  # Initialize an empty dictionary accumulator

# Assumes 'items' and 'inventory' lists are aligned by index 🚧

items = ['skis', 'snowboard', 'goggles', 'boots']  # new keys
inventory = [10, 0, 0, 7]                          # new values

# Use enumerate() to pair each new product with its corresponding price:
for i, item in enumerate(items):                  # loop through index and the item in the items list
    inventory_status[item] = inventory[i]

print(inventory_status)



# ------------------------------------------------
# Section 8.06 — Pattern 1: 'enumerate()'
# Dictionary Accumulator (Loop-Based Construction)
# ------------------------------------------------

inventory_status = {}     # Initialize an empty dictionary accumulator

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

for i, item in enumerate(items):
    if inventory[i] == 0:
        inventory_status[item] = "sold out"
    else:
        inventory_status[item] = "in stock"

print(inventory_status)

# This pattern is common when rules or conditions affect how data is classified 🚧


# ------------------------------------------------
# Section 8.06 — Pattern 2:
# Building from Parallel Lists with `zip()` ↕️
# ------------------------------------------------
# When two lists are already aligned by position, 
# `zip()` is often clearer than `enumerate()`

# Build a dictionary from aligned lists using zip() ↕️

inventory_status = {}     # empty dictionary accumulator

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

for item, qty in zip(items, inventory):
    if qty == 0:
        inventory_status[item] = "sold out"
    else:
        inventory_status[item] = "in stock"

print(inventory_status)

# Use `zip()` when sequences are meant to be paired element-by-element ↕️


# ------------------------------------------------
# Choosing Between `enumerate()` vs `zip()`
# ------------------------------------------------
# - Decision guideline:
#   - Use enumerate(iterable, start=0) when you need the index (position) along with items from a single sequence.
#   - Use zip(iter1, iter2, ...) when you need to pair elements from multiple parallel sequences ↕️

# This choice reflects intent and clarity: 
# tracking position explicitly vs. combining related, parallel data.


# ------------------------------------------------
# Section 8.06 — Pattern 3:
# Building Aggregated Dictionaries
# ------------------------------------------------
# Dictionaries are often built to store computed values, not raw data.

# Aggregate values while building a dictionary
customer_sales = {
    "C00001": [224.99, 87.36],
    "C00004": [125.56],
    "C00014": [99.99, 104.26, 23.57],
    "C00015": [82.65, 56.88, 201.82, 519.55]
}

total_sales = {}  # Initialize an empty dictionary accumulator

for customer_id, sales in customer_sales.items():
    total_sales[customer_id] = sum(sales)

print(total_sales)

# Why this pattern matters:
# - Raw data is transformed into meaningful metrics
# - The dictionary becomes a feature map
# - This mirrors group-by operations in pandas

# Key Takeaways:
# - Dynamic dictionary building is central to analytics workflows 🎯
# - Mastering these patterns prepares you directly for functions, pandas, and real-world data transformations.
`````

## Section 8.06 — Building Dictionaries from Parallel Data with `zip()`

### Overview
`zip()` is the bridge from “parallel lists” (spreadsheet-style columns) to **structured records** (dictionary-style rows).

It pairs elements **by position** across multiple iterables and produces an iterator of tuples.

In analytics terms:
- Parallel lists = columns that are aligned by row index
- `zip()` = row constructor
- Dictionary building = turning rows into keyed records

---

### Core Concepts
- `zip()` pairs items element-by-element
- `zip()` returns an **iterator** (lazy sequence)
- `zip()` stops at the **shortest** iterable
- `zip()` + loop accumulator builds dictionaries explicitly
- `dict(zip(keys, values))` is the “compressed” version (only for 2 iterables)
- For 3+ iterables, you typically store a **tuple** or **nested dict** as the value


`````python
# ------------------------------------------------------------------------
# Section 8.06
# Introduction to zip(): Creating Parallel Tuples from Multiple Iterables
# # Use `zip()` when sequences are meant to be paired element-by-element ↕️
# -----------------------------------------------------------------------

# Part A — Seeing What `zip()` Produces (No Dictionary Yet)

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

z = zip(items, inventory)
print(z)                  # Output: <zip object at 0x...> (some memory address)
print(type(z))            # Output: <class 'zip'>


# zip() returns an iterator (not a list)
# To inspect the pairs (for learning/debugging), we can consume it into a list:
print(list(z))            # Output: [('skis', 10), ('snowboard', 0), ('goggles', 0), ('boots', 7)]

# Key idea
# After you convert a zip object to a list once, it is exhausted 🚧


# Demonstrate exhaustion
z = zip(items, inventory)
print(list(z))   # consumes it
print(list(z))   # now empty, because it was already consumed


# -----------------------------------------------
# Part B — `zip()` Stops at the Shortest Iterable 
# Important “Data Quality” Rule 🚧
# -----------------------------------------------

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0]  # shorter list

paired = list(zip(items, inventory))
print(paired)

# Result: only 3 pairs are produced; "boots" is silently dropped ⚠️


# Analytics framing: this is a silent truncation risk ⚠️
# When data is supposed to align, you either:
# - validate lengths, or
# - deliberately slice to match



# 🟦 Helper 1 — `zip_trace()` (Inspect Zip Safely Without Losing the Iterator)

# Why this helper exists
# When learning, you want to “peek” at zip output without consuming it accidentally.

# This helper:
# - shows the first `n` zipped tuples
# - returns a **fresh** zip iterator so you can keep working


def zip_trace(*iterables, n=5):
    """
    Peek at the first n tuples produced by zip(*iterables) without losing the full sequence.

    Returns a NEW zip iterator you can use afterward.
    """
    z = zip(*iterables)
    preview = []

    # Pull first n tuples safely
    for _ in range(n):
        try:
            preview.append(next(z))
        except StopIteration:
            break

    print("zip_trace preview:", preview)

    # Re-create zip so the caller still has a full iterator available
    return zip(*iterables)


# Example Usage

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

z = zip_trace(items, inventory, n=10)
print(list(z))  # still works, because zip_trace returned a fresh zip iterator


# -----------------------------------------
# Part C — Build a Dictionary with a Loop
# -----------------------------------------
# Goal: create a dictionary mapping:
# - key = item name
# - value = quantity


items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

inventory_map = {}  # dictionary accumulator

for item, qty in zip(items, inventory):
    inventory_map[item] = qty

print(inventory_map)

# Why this matters: 🧱
# This pattern is the foundation of building structured data from raw aligned inputs.


# -----------------------------------------
# Part D — `dict(zip(keys, values))` 
# Compressed Version for Two Iterables
# -----------------------------------------

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

inventory_map_compact = dict(zip(items, inventory))
print(inventory_map_compact)

# Rule: `dict(zip(keys, values))` only makes a dictionary directly when:
# - the first iterable is keys
# - the second iterable is values
# - and you want a 1-to-1 pairing ↕️



# -----------------------------------------
# Part E — `zip()` with 3+ Lists 
# Store Multiple Fields Per Key 🚧
# -----------------------------------------
# When you have multiple “columns,” you usually build:
# - key = ID (or name)
# - value = tuple of fields (positional schema) OR dict of fields (named schema)

# 1️⃣ Option 1 — Tuple schema (positional, fast, but brittle) ⚠️

item_ids = [10001, 10002, 10003]
item_names = ["Coffee", "Beanie", "Gloves"]
prices = [5.27, 8.79, 17.59]

item_data_tuple = {}

for item_id, name, price in zip(item_ids, item_names, prices):
    item_data_tuple[item_id] = (name, price)

print(item_data_tuple)

# Access example:
print(item_data_tuple[10003][0])  # name (index-based)
print(item_data_tuple[10003][1])  # price


# 2️⃣ Option 2 — Named schema (dictionary as the value; clearer, safer) ⚖️
 
item_data_named = {}

for item_id, name, price in zip(item_ids, item_names, prices):
    item_data_named[item_id] = {"name": name, "price": price}

print(item_data_named)

# Access example:
print(item_data_named[10003]["name"])
print(item_data_named[10003]["price"])

# Analytics framing: named schema reduces index mistakes and scales better as fields grow ⚖️



# 🟦 Helper 2 — `validate_parallel_lengths()` (Prevent Silent Truncation)

# Why this helper exists
# `zip()` silently truncates at the shortest iterable. This helper makes alignment explicit ↕️


def validate_parallel_lengths(**named_iterables):
    """
    Validate that multiple parallel iterables have the same length.
    Pass them as keyword arguments so the output is readable.

    Example: validate_parallel_lengths(items=items, inventory=inventory)
    """
    lengths = {name: len(seq) for name, seq in named_iterables.items()}
    print("Lengths:", lengths)

    unique_lengths = set(lengths.values())
    if len(unique_lengths) == 1:
        print("✅ All iterables align (same length).")
        return True
    else:
        print("⚠️ Length mismatch detected. zip() would truncate to the shortest length.")
        shortest = min(lengths.values())
        print(f"Shortest length = {shortest}. Consider slicing or fixing upstream data.")
        return False

# Example Usage

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0]

validate_parallel_lengths(items=items, inventory=inventory)

# If you intentionally want alignment by truncation: 🚧
inventory_map = dict(zip(items[:len(inventory)], inventory))
print(inventory_map)




# Key Takeaways 📌
# - `zip()` pairs aligned elements across iterables and produces tuples.
# - `zip()` is lazy (iterator) and is consumed when iterated.
# - `zip()` stops at the shortest iterable (silent truncation risk).
# - Loop-first dictionary building is the “show your work” approach:
#   - `for a, b in zip(...): dict[a] = b`
# - `dict(zip(keys, values))` is a clean compression for exactly **two** iterables.
# - For 3+ iterables, store a tuple (positional schema) or a dict (named schema) as the value.
# - This section is the direct bridge to the next topic: **nested dictionaries** (two-key lookups).
`````

## Dictionary Comprehension — “Show Your Work” Build-Up

### Preview
A dictionary comprehension is **not** new logic.  
It is a **compression** of logic you already know:

- loop over `(key, value)` pairs
- compute something
- store it in a new dictionary

If you can write it long-form, you can always prove the comprehension is equivalent.

---

## Summary
- A dictionary comprehension is a **compressed loop** that builds a dictionary.
- You can always expand a comprehension into:
  1) an empty accumulator dictionary  
  2) a `for` loop over `.items()`  
  3) a computed value  
  4) an assignment to `new_dict[key]`
- Equivalence can be proven by comparing outputs (`==`).
- For your learning style: use the loop first, then compress only after the logic is fully proven.


`````python
# -----------------------------------------------------
# Dictionary Comprehension — “Show Your Work” Build-Up
# -----------------------------------------------------
# We want a new dictionary where:
# - keys = customer IDs
# - values = total of their sales list


# Raw data: customer → list of transactions
customer_sales = {
    "C00001": [224.99, 87.36],
    "C00004": [125.56],
    "C00014": [99.99, 104.26, 23.57],
    "C00015": [82.65, 56.88, 201.82, 519.55]
}

# -----------------------------------------
# Step 1 — Long-Form Loop with Accumulator
# -----------------------------------------

total_sales = {}       # initialize empty dictionary accumulator

for customer_id, sales in customer_sales.items():
    total_sales[customer_id] = round(sum(sales), 2)  # ← round here, per customer

print(total_sales)


# ------------------------------------------------------
# Step 2 — Show the Loop’s Output Matches a Manual Check
# ------------------------------------------------------

# Manual verification for one customer:

print(sum(customer_sales["C00001"]))   # should match total_sales["C00001"]
print(total_sales["C00001"])


# ------------------------------------------------------------
# Step 3 — The Same Loop, But Written as a Single Mapping Rule
# ------------------------------------------------------------
# Write the rule in English:
# > For each `(customer_id, sales)` in `customer_sales.items()`, map  
# > `customer_id → sum(sales)`.

# 🎯 That sentence is the comprehension.


# Step 4 — Dictionary Comprehension (Compressed Version of the Same Logic)

total_sales_comp = {
    customer_id: round(sum(sales), 2)
    for customer_id, sales in customer_sales.items()
}

print(total_sales_comp)


# ------------------------------------------------------------
# Step 5 — Show Equivalence (Loop Output vs Comprehension Output)
# ------------------------------------------------------------

# Build with the loop
total_sales_loop = {}

for customer_id, sales in customer_sales.items():
    total_sales_loop[customer_id] = sum(sales)


# Build with the comprehension
total_sales_comp = {
    customer_id: sum(sales)
    for customer_id, sales in customer_sales.items()
}

# Proof: these should be identical dictionaries
print(total_sales_loop == total_sales_comp)


# If this prints `True`, the comprehension is proven equivalent.
`````

`````python
# -----------------------------------------------------
# Dictionary Comprehension — “Show Your Work” Build-Up
# -----------------------------------------------------
# Practice Variations (Build Fluency Without Skipping Steps)


# -----------------------------------------
# Variation A — Round totals to 2 decimals
# -----------------------------------------

total_sales_rounded = {}

for customer_id, sales in customer_sales.items():
    total_sales_rounded[customer_id] = round(sum(sales), 2)

print(total_sales_rounded)


#  🎯 Comprehension version

total_sales_rounded_comp = {
    customer_id: round(sum(sales), 2)
    for customer_id, sales in customer_sales.items()
}

print(total_sales_rounded_comp)


# -----------------------------------------------------
# Variation B — Count the number of purchases per customer
# -----------------------------------------------------

purchase_counts = {}

for customer_id, sales in customer_sales.items():
    purchase_counts[customer_id] = len(sales)

print(purchase_counts)


#  🎯 Comprehension version

purchase_counts_comp = {
    customer_id: len(sales)
    for customer_id, sales in customer_sales.items()
}

print(purchase_counts_comp)


# -----------------------------------------------------
# Variation C — Create a nested “feature map” per customer
# -----------------------------------------------------
# Example: total + number of purchases

customer_features = {}

for customer_id, sales in customer_sales.items():
    customer_features[customer_id] = {
        "total": round(sum(sales), 2),
        "num_purchases": len(sales)
    }

print(customer_features)


#  🎯 Comprehension version

customer_features_comp = {
    customer_id: {
        "total": round(sum(sales), 2),
        "num_purchases": len(sales)
    }
    for customer_id, sales in customer_sales.items()
}

print(customer_features_comp)



# Hack version ⚠️
# dict(zip(items, zip(prices, inventory)))
# item_dict = dict(zip(item_list, zip(price_list, inventory)))


# Prefer named fields over positional fields
# ✅ item[item_id]["price"]
# ❌ item[item_id][0]


# Key Takeaways 📌
# - A dictionary comprehension is a compressed loop that builds a dictionary.
# - You can always expand a comprehension into:
#   1) an empty accumulator dictionary  
#   2) a `for` loop over `.items()`  
#   3) a computed value  
#   4) an assignment to `new_dict[key]`
# - Equivalence can be proven by comparing outputs (`==`).
# - For your learning style: use the loop first, then compress only after the logic is fully proven.
`````

`````python
import pandas as pd

names = ['Alice', 'Bob', 'Charlie']
scores = [85, 92, 78]
ages = [25, 30, 22]


records = []

for name, score, age in zip(names, scores, ages):
    records.append({
        "Name": name,
        "Score": score,
        "Age": age
    })

df = pd.DataFrame(records)


# Alternatively, when:
# the zipped data is immediately consumed
# no positional access survives beyond construction

# zip() pairs them lazily – no full list created yet
paired_data = zip(names, scores, ages)

df = pd.DataFrame(paired_data, columns=['Name', 'Score', 'Age'])
print(df)

# ⚠️ Do NOT do this:
# paired_data = list(zip(names, scores, ages))
# paired_data[0][1]  # brittle positional access
`````

## Section 8.07 — Nested Dictionaries (Structured Records & Two-Level Lookup)

### Overview
Nested dictionaries allow you to model **real-world records** where each entity has multiple named attributes.

Instead of:
- one key → one value  

you now have:
- one key → **a dictionary of attributes**

This is the first time Python data starts to resemble:
- database rows
- JSON objects
- pandas DataFrame rows

Conceptually:
- outer key = unique identifier (ID)
- inner dictionary = schema (fields)

---

### Core Concepts
- Dictionaries can store other dictionaries as values
- Access becomes **two-step lookup**
- Named fields reduce index-based errors
- Nested dictionaries scale better than tuple-based schemas
- This pattern is foundational for analytics, APIs, and pandas

`````python
# -------------------------------------
# Section 8.07 — Nested Dictionaries
# Structured Records & Two-Level Lookup
# -------------------------------------


# -------------------------------------------------------------
# 1️⃣ Part A: From Flat to Nested (Why This Exists)
# -------------------------------------------------------------


# Flat structure: meaning depends on position

item_data_flat = {
    10001: ("Coffee", 5.99, "beverage"),
    10002: ("Beanie", 9.99, "clothing")
}

# Access is positional and fragile 
print(item_data_flat[10001][1])  # price, but only if you remember index 1


# Problems: ⚠️
# - You must remember index positions
# - Adding fields breaks old code
# - Hard to read six months later



# Nested dictionary (named schema) 🧱 

item_ids = [10001, 10002, 10003]
names = ["Coffee", "Beanie", "Snowboard"]
prices = [5.99, 9.99, 599.99]
categories = ["beverage", "clothing", "equipment"]

item_data = {}

for item_id, name, price, category in zip(item_ids, names, prices, categories):
    item_data[item_id] = {
        "name": name,
        "price": price,
        "category": category
    }


# Literal version = how data is described
item_data = {
    10001: {
        "name": "Coffee",
        "price": 5.99,
        "category": "beverage"
    },
    10002: {
        "name": "Beanie",
        "price": 9.99,
        "category": "clothing"
    },
    10003: {
        "name": "Snowboard",
        "price": 599.99,
        "category": "equipment"
    }
}

# Access is explicit and readable 🎯
print(item_data[10001])
print(item_data[10001]["price"])
print(item_data[10002])
print(item_data[10002]["category"])
print(item_data[10003])
print(item_data[10003]["name"])


# -------------------------------------------------------------
# 2️⃣ Part B — Visualizing the Structure (Mental Model)
# -------------------------------------------------------------

# item_data
# │
# ├── 10001 ──► {
# │              "name": "Coffee",
# │              "price": 5.99,
# │              "category": "beverage"
# │            }
# │
# └── 10002 ──► {
#                "name": "Beanie",
#                "price": 9.99,
#                "category": "clothing"
#              }


# You are not “looking up two keys at once.”  
# You are doing **sequential lookup**:

# 1. outer dictionary → record
# 2. inner dictionary → field




# -------------------------------------------------------------
# 3️⃣ Part C — Two-Level Lookup (Step-by-Step)
# -------------------------------------------------------------

# Step 1: get the inner dictionary (the record)
record = item_data[10002]
print(record)

# Step 2: get a specific field from the record
price = record["price"]
print(price)


# Condensed (but equivalent): 🎯

print(item_data[10002]["price"])



# 🟦 Helper 1 — `nested_lookup_trace()` (Make Two-Level Access Visible)

# Why this helper exists:
# Nested lookups fail silently until they don’t
# This helper shows each step explicitly


def nested_lookup_trace(data, outer_key, inner_key):
    """
    Trace a two-level dictionary lookup.
    Shows whether the outer key and inner key exist.
    """
    if outer_key not in data:
        print(f"Outer key NOT found: {outer_key}")
        return None

    print(f"Outer key found: {outer_key}")
    record = data[outer_key]

    if inner_key not in record:
        print(f"Inner key NOT found: '{inner_key}'")
        return None

    print(f"Inner key found: '{inner_key}' → {record[inner_key]}")
    return record[inner_key]


# Example Usage:
nested_lookup_trace(item_data, 10001, "price")
nested_lookup_trace(item_data, 10001, "sizes")
nested_lookup_trace(item_data, 99999, "price")

# Output:
# Outer key found: 10001
# Inner key found: 'price' → 5.99
# Outer key found: 10001
# Inner key NOT found: 'sizes'
# Outer key NOT found: 99999


# -------------------------------------------------------------
# 4️⃣ # Part D — Building Nested Dictionaries with Loops
# No Comprehension Yet
# -------------------------------------------------------------
# Goal: Build this structure dynamically from parallel lists.

item_ids = [10001, 10002, 10003]
names = ["Coffee", "Beanie", "Gloves"]
prices = [5.99, 9.99, 19.99]
categories = ["beverage", "clothing", "clothing"]

item_data = {}  # initialize an outer dictionary accumulator

for item_id, name, price, category in zip(item_ids, names, prices, categories):
    item_data[item_id] = {
        "name": name,
        "price": price,
        "category": category
    }

print(item_data)
print()


# Key idea: 🔑
# The loop builds records, not just values.



# -------------------------------------------------------------
# 5️⃣ Part E — Safe Nested Access with `.get()`
# -------------------------------------------------------------

# Safe outer lookup
record = item_data.get(10001)

if record is None:
    print("Item not found")
else:
    # Safe inner lookup
    price = record.get("price", "Price unavailable")
    print(price)
print()

# Analytics framing:
# This mirrors defensive access in:
# - APIs
# - messy datasets
# - optional fields



# 🟦 Helper 2 — `safe_nested_get()` (One Call, Two Levels)


def safe_nested_get(data, outer_key, inner_key, default=None):
    """
    Safely retrieve a nested dictionary value.
    Returns default if any key is missing.
    """
    outer = data.get(outer_key)
    if outer is None:
        return default
    return outer.get(inner_key, default)


# Example Usage:
print(safe_nested_get(item_data, 10002, "price"))
print(safe_nested_get(item_data, 10005, "category", "[] sentinel default"))
print(safe_nested_get(item_data, 99999, "price", "Not found"))



# -------------------------------------------------------------
# 6️⃣ Part F — Why Nested Dictionaries Matter (Analytics Framing)
# -------------------------------------------------------------
#  Key Takeaways
# - Nested dictionaries model structured records with named fields.
# - Access is sequential: outer key → inner key.
# - Named schemas are safer than positional tuples.
# - Loop-first construction builds intuition and correctness.
# - Helpers make multi-level access debuggable.
# - This pattern directly prepares you for:
#   - dictionary comprehensions
#   - JSON handling
#   - pandas DataFrames
`````

`````python
# -------------------------------------
# Exercise: Nested Dictionaries
# ------------------------------------

economic_data = { 
    "India": { 
        "Uttar Pradesh": {"population": 241, "gdp": 220}, 
        "Rajastan": {"population": 200, "gdp": 130}, 
    }, 
    "China": {
        "Guangdong": {"population": 126, "gdp": 1960}, 
        "Fujian": {"population": 42, "gdp": 757},
         },
}

# Level 1: country → Level 2: state → Level 3: field
print(economic_data["China"]["Fujian"]["population"])  # Output: 42


# ---------------------------------------------------------------------------
# Three-level nested lookup: Iterating over inner dictionaries via .values() 🔢
# ---------------------------------------------------------------------------
# Extracting a specific field from a nested record
# Building a derived list from structured data

population = []

for state in economic_data["India"].values():
    population.append(state["population"])

print(population)  # Output: [241, 200]


# ---------------------------------------------------------------------------
# Alternatively: Extract population by state with explicit naming via .items() → (key, value) 🔗
# ---------------------------------------------------------------------------

population = {}

for state_name, state_data in economic_data["India"].items():
    population[state_name] = state_data["population"]

print(population)  # Output: {'Uttar Pradesh': 241, 'Rajastan': 200}


for state_name, state_data in economic_data["India"].items():
    print(f"{state_name}: {state_data['population']}")    

# Output:
# Uttar Pradesh: 241
# Rajastan: 200


# The decision rule (this is the part to remember):
# ✔️ Use .values() when:
# You only need the data
# The key has no analytical value
# You’re computing aggregates
# You want minimal noise

# ✔️ Use .items() when:
# The key identifies the record
# You may need it later
# You are transforming or labeling data
# You want clarity over brevity

# Additionally:
for state_name, state_data in economic_data["India"].items():
    print(f"{state_name}: GDP per capita = {state_data['gdp'] / state_data['population']}")


# Clean decision: 🤔 
# List → unlabeled sequence of values
# Set → unique unlabeled values
# Dictionary → labeled measurements (mapping)
`````

`````python
# ----------------------------------
# Assignment: Nested Dictionaries
# ----------------------------------


item_ids = [
    10001, 10002, 10003, 10004, 10005, 
    10006, 10007, 10008, 10009
]

item_names = [
    "Coffee", "Beanie", "Gloves", "Sweatshirt", "Helmet",
    "Snow Pants", "Coat", "Ski Poles", "Ski Boots"
]

euro_prices = [
    5.27, 8.79, 17.59, 21.99, 87.99, 
    70.39, 105.59, 87.99, 175.99
]

item_category = [
    "beverage", "clothing", "clothing", "clothing", "safety",
    "clothing", "clothing", "hardware", "hardware",
]

sizes = [
    ["250mL"],
    ["Child", "Adult"],
    ["Child", "Adult"],
    ["XS", "S", "M", "L", "XL", "XXL"],
    ["Child", "Adult"],
    ["XS", "S", "M", "L", "XL", "XXL"],
    ["S", "M", "L"],
    ["S", "M", "L"],
    [5, 6, 7, 8, 9, 10, 11],
]


# 📘 Validate lengths before building to prevent silent truncation: 🔍
lengths = [len(item_ids), len(item_names), len(euro_prices), len(item_category), len(sizes)]
if len(set(lengths)) != 1:
    raise ValueError(f"Parallel lists must be same length. Lengths: {lengths}")


# Dictionary Comprehension (compressed form of the loop-based build)
euro_data_comp = {
    item_id: {"name": name, "price": price, "category": category, "sizes": size_list}
    for item_id, name, price, category, size_list in zip(
        item_ids, item_names, euro_prices, item_category, sizes
    )
}
    

# Build dictionary with for loop and zip()
euro_data = {}  # Initialize empty dictionary accumulator

for item_id, name, price, category, size_list in zip(
    item_ids, item_names, euro_prices, item_category, sizes
):
    euro_data[item_id] = {
        "name": name,
        "price": price,
        "category": category,
        "sizes": size_list
    }


# Inspect the nested structure with .items() → (key, record)
# Each value is a dictionary (named schema) 🧱
for item_id, details in euro_data.items():
    print(item_id, "→", details)

# Level 1: item_id → Level 2: price
print(euro_data[10009]["price"])

# Build dictionary: Map product name → available sizes 🏗️
product_sizes = {}

for details in euro_data.values():
    product_sizes[details["name"]] = details["sizes"]

print(product_sizes)

# Alternatively, dictionary comprehension: 🧠
product_sizes = {
    details["name"]: details["sizes"]
    for details in euro_data.values()
}
`````

## Section 8.08–8.09 — Sets: Uniqueness, Membership, and Relationships

### Overview
Sets solve a very specific class of problems:
- **uniqueness**
- **fast membership testing**
- **relationships between groups**

They are not about order.
They are not about indexing.
They are about **presence vs. absence**.

In analytics terms, sets answer questions like:
- “Which categories exist?”
- “Which IDs are duplicated?”
- “Which customers appear in both groups?”
- “What is new, missing, or overlapping?”

---

## Core Concepts
- Set literals and construction
- Automatic uniqueness (**duplicates removed**)
- Fast membership testing (`in`)
- Converting lists to sets
- Set relationships:
  - union
  - intersection
  - difference
  - symmetric difference


`````python
# -----------------------------------------------------
# Part A — ♻️ Why Sets Exist — eliminating duplication
# -----------------------------------------------------
# If you ask: “What categories do we carry?”
# A list is the wrong structure.

# Raw data with duplicates
categories = [
    "clothing", "hardware", "clothing", "beverage",
    "clothing", "hardware", "clothing"
]

print(categories)

# Output: ['clothing', 'hardware', 'clothing', 'beverage', 'clothing', 'hardware', 'clothing']


# --------------------------------------------------------------
# Part B - 🧹 Converting to a Set — automatic deduplication
# --------------------------------------------------------------

unique_categories = set(categories)
print(unique_categories)

# Output: {'hardware', 'beverage', 'clothing'}

# Key observations:
# - Duplicates are removed automatically
# - Sets are unordered (do not rely on display order)


# Analytics framing: 
# This mirrors extracting distinct values from a column.


# ---------------------------------------------------------------
# 🟦 Helper 1 — `set_build_trace()` (See Deduplication Happen) 🎯
# ---------------------------------------------------------------
# Why this helper exists
# Sets hide work by design. This helper makes the transition visible.

def set_build_trace(iterable):
    """
    Trace how a set grows as items are added.
    """
    result = set()
    for item in iterable:
        before = result.copy()
        result.add(item)
        if before != result:
            print(f"Added: {item} → {result}")
        else:
            print(f"Ignored duplicate: {item}")
    return result


# Example Usage:
categories_demo = ["clothing", "hardware", "clothing", "beverage"]
final_set = set_build_trace(categories_demo)
print("Final set:", final_set)

# Output:
# Added: clothing → {'clothing'}
# Added: hardware → {'hardware', 'clothing'}
# Ignored duplicate: clothing
# Added: beverage → {'hardware', 'beverage', 'clothing'}
# Final set: {'hardware', 'beverage', 'clothing'}


# --------------------------------------------------
# Part C —⚡ Membership Testing — fast `in` checks
# --------------------------------------------------
# Why this matters:
# - `in` on a list → linear search
# - `in` on a set → hash-based lookup (very fast)


categories_set = set(categories)

print("clothing" in categories_set)   # True
print("electronics" in categories_set)  # False

# Analytics framing: 
# Sets are ideal for:
# - validation rules
# - filters
# - allowed-value checks


# -----------------------------------------------------------------
# 🟦 Helper 2 — `set_membership_trace()` (Explain Membership Logic) 🎯
# -----------------------------------------------------------------
 
def set_membership_trace(value, reference_set):
    """
    Explain membership testing against a set.
    """
    if value in reference_set:
        print(f"'{value}' FOUND in set.")
        return True
    else:
        print(f"'{value}' NOT found in set.")
        return False


# Example Usage:
# Schema-defined allowed values
allowed_categories = set(categories)
print(allowed_categories)   # {"clothing", "hardware", "beverage"}

set_membership_trace("clothing", allowed_categories)
set_membership_trace("electronics", allowed_categories)

# Output:
# 'clothing' FOUND in set.
# 'electronics' NOT found in set.
# False

# --------------------------
# Filter Values Using a Set
# --------------------------
# What this demonstrates:
# - Sets act as filters (fast membership checks)
# - Lists preserve event history (order + frequency)
# - Deduplication is a separate, explicit step

expected_products = {"skis", "boots", "goggles"}

sales_log = ["skis", "boots", "skis", "snowboard", "goggles"]

# Step 1: Filter valid products (preserve events)
filtered_sales = []

for product in sales_log:
    if product in expected_products:
        filtered_sales.append(product)

print(filtered_sales)
# ['skis', 'boots', 'skis', 'goggles']

# Step 2: Deduplicate when uniqueness is required
unique_sales = set(filtered_sales)
print(unique_sales)
# {'boots', 'skis', 'goggles'}


# --------------------------------------------------
# Part D — # 🔗 Set Relationships — comparing groups
# --------------------------------------------------
# Think of sets as groups of IDs or labels

store_a = {"skis", "boots", "helmet", "gloves"}
store_b = {"boots", "helmet", "goggles"}


# ---------------------------------------------
# Union — Everything in Either Group
# ➕ Union — elements in A or B (or both)
# ---------------------------------------------
# returns all unique elements that appear in either set (or both)
# Use case:
# - full catalog
# - combined coverage

all_items = store_a | store_b
print(all_items)

# Output: {'skis', 'goggles', 'boots', 'gloves', 'helmet'}

# Equivalent to: store_a.union(store_b)
# Could also be: store_a | store_b | store_c | store_d


# ---------------------------------------------
# Intersection — Only What Overlaps
# 🤝 Intersection = elements present in ALL sets
# ---------------------------------------------
# Use case:
# - shared customers
# - overlapping products
# - cohort intersection
    
common_items = store_a & store_b
print(common_items)

# Output: {'boots', 'helmet'}

# Equivalent to: store_a.intersection(store_b)
# Could also be: store_a & store_b & store_c & store_d

# ---------------------------------------------
# Difference — What’s Missing from One Group
# ➖ A - B — elements only in A
# ➖ B - A — elements only in B
# ---------------------------------------------
# Use case:
# - exclusives
# - gaps
# - churn analysis
    
only_a = store_a - store_b
print(only_a)

# Output:
# {'gloves', 'skis'}

only_b = store_b - store_a
print(only_b)

# Output:
# {'goggles'}

# Equivalent to: store_a.difference(store_b)

# ------------------------------------------------
# Symmetric Difference — What’s Different Overall
# 🔀 Symmetric Difference — in A or B, but not both
# ------------------------------------------------
# Use case:
# - change detection
# - mismatched inventories

exclusive_items = store_a ^ store_b
print(exclusive_items)

# Output: {'skis', 'goggles', 'gloves'}

# Equivalent to: store_a.Symmetric_difference(store_b)

# ---------------------------------------------------------------------
# 🟦 Helper 3 — `set_relationship_trace()` (Explain Set Logic Clearly) 🎯
# ---------------------------------------------------------------------

def set_relationship_trace(a, b):
    """
    Show all major set relationships between two sets.
    """
    print("Set A:", a)
    print("Set B:", b)
    print("In A or B (Union):", a | b)
    print("In both A and B (Intersection):", a & b)
    print("Only in A:", a - b)
    print("Only in B:", b - a)
    print("In exactly one set (Symmetric Difference):", a ^ b)

# Example Usage
set_relationship_trace(store_a, store_b)

# Output:
# Set A: {'skis', 'boots', 'gloves', 'helmet'}
# Set B: {'boots', 'goggles', 'helmet'}
# In A or B (Union): {'skis', 'goggles', 'boots', 'gloves', 'helmet'}
# In both A and B (Intersection): {'boots', 'helmet'}
# Only in A: {'gloves', 'skis'}
# Only in B: {'goggles'}
# In exactly one set (Symmetric Difference): {'skis', 'goggles', 'gloves'}


# --------------------------------------------------
# 🧠 Schema vs Observed Values (Validation Context)
# --------------------------------------------------

# Observed values extracted from the data
observed_categories = set(categories)

# Schema-defined allowed values (policy / contract)
allowed_categories = {"clothing", "hardware", "beverage"}

# Values observed in data but NOT allowed by schema
invalid_categories = observed_categories - allowed_categories

print("Observed:", observed_categories)
print("Allowed:", allowed_categories)
print("Invalid:", invalid_categories)


# -----------------------------------------------------------------
# 🟦 Helper 4 — `validate_allowed()` (Validate Expected Categories)
# -----------------------------------------------------------------

def validate_allowed(value, allowed_set):
    """
    Validate a value against an allowed set.
    Returns True/False and prints a clear message.
    """
    if value not in allowed_set:
        print(f"⚠️ Invalid value: '{value}'. Allowed: {allowed_set}")
        return False
    print(f"✅ Valid: '{value}'")
    return True


# Example Usage:
validate_allowed("clothing", allowed_categories)
validate_allowed("electronics", allowed_categories)

# Output:
# ✅ Valid: 'clothing'
# ⚠️ Invalid value: 'electronics'. Allowed: {'hardware', 'clothing', 'beverage'}
# False


# ---------------------------------------------------
# Part E — When to Use Sets vs Lists vs Dictionaries 🤔
# ---------------------------------------------------
# Use a list when:
# - order matters
# - duplicates matter
# - indexing matters

# Use a set when:
# - uniqueness matters
# - membership testing matters
# - relationships between groups matter

# Use a dictionary when:
# - you need keys → values
# - you are modeling records
# - you need structured data
`````

`````python
# --------------------------------------------------
# Mini Exercise — Category Validation Summary 🧠
# --------------------------------------------------

# Observed data (from transactions or catalog)
observed_categories = {"clothing", "hardware", "beverage", "electronics"}

# Schema / policy
allowed_categories = {"clothing", "hardware", "beverage"}

# Diagnostics
invalid_categories = observed_categories - allowed_categories
missing_categories = allowed_categories - observed_categories

print("Observed:", observed_categories)
print("Allowed:", allowed_categories)
print("Invalid:", invalid_categories)
print("Missing:", missing_categories)
`````

`````python
# --------------------------------------
# Assignment: Nested Dictionaries & Sets
# --------------------------------------


item_ids = [
    10001, 10002, 10003, 10004, 10005, 
    10006, 10007, 10008, 10009
]

item_names = [
    "Coffee", "Beanie", "Gloves", "Sweatshirt", "Helmet",
    "Snow Pants", "Coat", "Ski Poles", "Ski Boots"
]

euro_prices = [
    5.27, 8.79, 17.59, 21.99, 87.99, 
    70.39, 105.59, 87.99, 175.99
]

item_category = [
    "beverage", "clothing", "clothing", "clothing", "safety",
    "clothing", "clothing", "hardware", "hardware",
]

sizes = [
    ["250mL"],
    ["Child", "Adult"],
    ["Child", "Adult"],
    ["XS", "S", "M", "L", "XL", "XXL"],
    ["Child", "Adult"],
    ["XS", "S", "M", "L", "XL", "XXL"],
    ["S", "M", "L"],
    ["S", "M", "L"],
    [5, 6, 7, 8, 9, 10, 11],
]


# 📘 Validate lengths before building to prevent silent truncation: 🔍
lengths = [len(item_ids), len(item_names), len(euro_prices), len(item_category), len(sizes)]
if len(set(lengths)) != 1:
    raise ValueError(f"Parallel lists must be same length. Lengths: {lengths}")


# Build dictionary with for loop and zip()
euro_dicts = {}  # Initialize empty dictionary accumulator

for item_id, name, price, category, size_list in zip(
    item_ids, item_names, euro_prices, item_category, sizes
):
    euro_dicts[item_id] = {
        "name": name,
        "price": price,
        "category": category,
        "sizes": size_list
    }


# Inspect the nested structure with .items() → (key, record)
# Each value is a dictionary (named schema) 🧱
for item_id, details in euro_dicts.items():
    print(item_id, "→", details)



# --------------------------------------------------------
# Sets: Collect unique item categories from the dictionary
# --------------------------------------------------------

unique_categories = set()

for record in euro_dicts.values():
    unique_categories.add(record["category"])

print(unique_categories)
`````

### Overview
Dictionaries and sets introduce **key-based access**, **uniqueness**, and **fast membership testing**.
They replace index-based thinking with **semantic access**, which is foundational for analytics, data validation, and real-world data modeling.

In earlier chapters, lists required you to:
- search linearly
- rely on positional meaning
- manually track relationships between values

Dictionaries and sets eliminate those limitations by allowing Python to manage relationships directly.

This chapter builds on:
- `for` loops
- `while` loops
- conditionals
- accumulators

and prepares you for:
- structured data
- functions
- pandas-style thinking (later, not now)

---

## Core Concepts Introduced
- Dictionary literals and access
- Key–value pairs
- Membership testing (`in`)

## Maven Baseline Example
The Maven lesson introduces dictionaries by contrasting them with lists and showing how **key-based lookup** replaces searching through nested lists :contentReference[oaicite:0]{index=0}.

```python
# Inventory status stored as key-value pairs
inventory_status = {
    'skis': 'in stock',
    'snowboard': 'sold out',
    'goggles': 'sold out',
    'boots': 'in stock'
}

# Direct lookup using a key
inventory_status['goggles']
```

**Output**
```text

```

This lookup is:
- direct

## Enhanced Explanation
This example demonstrates a fundamental shift:

> We no longer ask *where* the value lives.

With lists, you previously had to:
- loop through items
- compare values manually
- extract meaning from position

With dictionaries:
- keys represent **real-world entities**

## 🟦 Helper — `trace_dict_lookup`
This helper makes dictionary access **visible and safe**.
It shows:
- what key was requested
- whether the key exists
- what value was returned (or why it was not)

This prevents silent failures and reinforces best practices early.

```python
def trace_dict_lookup(data, key):
    """
    Safely traces dictionary access.
    Shows whether the key exists and what value is returned.
    """
    if key in data:
        print(f"Key found: '{key}' → Value: {data[key]}")
        return data[key]
    else:
        print(f"Key NOT found: '{key}'")
        return None
```

### Example Usage
```python
trace_dict_lookup(inventory_status, 'boots')
trace_dict_lookup(inventory_status, 'bindings')
```

**Output**
```text
Key found: 'boots' → Value: in stock
Key NOT found: 'bindings'
```

This reinforces:
- membership testing with `in`
- safe access patterns
- why `.get()` exists (coming shortly)

---

## Key Takeaways
- Dictionaries replace positional access with **semantic access**
- Keys must be unique and are typically strings or numbers
- Lookup is fast and explicit
- Membership testing (`key in dict`) is foundational
- Dictionaries model real-world data more naturally than lists

### Common Pitfalls
- Assuming order matters (it does not conceptually)
- Accessing missing keys without checking
- Treating dictionaries like lists

At this stage, focus on:
**reading**, **accessing**, and **thinking in keys**, not memorizing methods.

## Section 8.04 — Iterating Over Dictionaries (Patterns That Matter)

### Overview
Dictionaries are most powerful when used inside loops.
However, how you iterate over a dictionary directly affects:
- readability
- correctness
- semantic clarity

This section clarifies **what Python iterates over by default**
and shows the three canonical iteration patterns.

---

### Core Concepts
- Iterating over keys (default behavior)
- Iterating over values explicitly
- Iterating over key–value pairs with `.items()`
- Choosing variable names that preserve meaning

---

### Why This Matters (Analytics Framing)
In analytics, loops are used to:
- aggregate values
- apply transformations
- enforce conditions
- build new structures

Clear iteration patterns prevent subtle logic errors
and make analytical intent obvious.

---

### Iteration Pattern 1 — Looping Over Keys (Default)

```python
product_prices = {
    "skis": 249.99,
    "boots": 79.99,
    "goggles": 99.99
}

# Looping over a dictionary iterates over keys by default
for product in product_prices:
    print(product)
```

**What this means**
- Python assumes you want keys
- This is equivalent to `for product in product_prices.keys()`

Use this when:
- keys are the primary identifier
- values are accessed conditionally or later

---

### Iteration Pattern 2 — Looping Over Values

```python
# Loop explicitly over values
for price in product_prices.values():
    print(price)
```

Use this when:
- keys are irrelevant
- only the stored values matter
- performing numeric analysis or aggregation

---

### Iteration Pattern 3 — Looping Over Key–Value Pairs

```python
# Loop over both keys and values using tuple unpacking
for product, price in product_prices.items():
    print(f"{product}: {price}")
```

**Why this is powerful**
- Preserves full context
- Makes relationships explicit
- Common in aggregation and transformation logic

This is the most common pattern in analytics workflows.

---

### Naming Conventions Matter

```python
# Clear, semantic variable names
for customer_id, total in customer_totals.items():
    print(customer_id, total)
```

Avoid:
```python
for x, y in customer_totals.items():
    ...
```

Names should reflect **meaning**, not structure.

---

### Common Beginner Mistakes
- Forgetting that `for x in dict` iterates over keys
- Looping over values when keys are needed
- Using unclear variable names
- Rebuilding dictionaries when `.items()` would suffice

---

### Key Takeaways
- Dictionaries iterate over keys by default
- `.values()` and `.items()` make intent explicit
- `.items()` is ideal when relationships matter
- Clear naming preserves semantic meaning
- Most dictionary logic lives inside loops

Understanding these patterns ensures your code
remains readable, correct, and analytically sound.


## Section 8.05 — Safe Access Patterns: `[]` vs `.get()`

### Overview
Dictionaries provide fast, semantic access to data using keys.
However, not all keys can be assumed to exist.

This section introduces **safe access patterns** for dictionaries and explains
when Python should:
- fail loudly (raise an error), or
- fail safely (continue execution)

Understanding this distinction is critical for analytics, where missing or optional
data is normal rather than exceptional.

---

### Core Concepts
- `dict[key]` performs a **direct lookup**
- Missing keys with `[]` raise a `KeyError`
- `.get()` performs a **safe lookup**
- `.get()` returns:
  - the actual value if the key exists
  - `None` by default if it does not
  - or a specified default value if provided

> Important distinction:  
> `.get()` is a **lookup**, not a *view*.  
> It returns the actual stored value, not a dynamic object.

---

### Why This Matters (Analytics Framing)
Real-world datasets are messy:
- columns may be optional
- records may be incomplete
- categories may appear inconsistently

In analytics pipelines:
- crashing on missing data is often undesirable
- silent failure may be acceptable **only if intentional**

Safe access patterns let you choose the right behavior.

---

### Baseline Dictionary
```python
item_details = {
    "skis": [249.99, 10, "in stock"],
    "snowboard": [219.99, 0, "sold out"],
    "goggles": [99.99, 0, "sold out"],
    "boots": [79.99, 7, "in stock"]
}
```

---

### Direct Lookup with `[]` (Fail Loudly)
```python
# Direct lookup assumes the key exists
print(item_details["boots"])   # returns the value

# Uncommenting the next line would raise a KeyError
# print(item_details["hats"])
```

**Behavior**
- Fast and explicit
- Raises `KeyError` if the key is missing
- Appropriate when missing data indicates a serious problem

---

### Safe Lookup with `.get()` (Fail Safely)
```python
# Safe lookup using .get()
print(item_details.get("boots"))    # returns the value
print(item_details.get("hats"))     # returns None
```

**Behavior**
- No error is raised
- Execution continues
- Missing data is handled gracefully

This is often the preferred behavior in analytics workflows.

---

### Providing a Default Value
```python
# Provide a fallback value if the key is not found
print(item_details.get("hats", "Item not carried"))
```

**Why this is useful**
- Prevents downstream errors
- Makes intent explicit
- Keeps pipelines robust

Defaults can be:
- strings
- numbers
- empty lists
- sentinel values (e.g., `"UNKNOWN"`)

---

### Choosing Between `[]` and `.get()`
Use this decision rule:

- Use `dict[key]` when:
  - the key **must** exist
  - missing data indicates a bug or logic error

- Use `.get()` when:
  - the key may or may not exist
  - missing data is expected
  - graceful handling is required

This choice is about **intent**, not convenience.

---

### Common Beginner Mistakes
- Assuming `.get()` returns a view (it does not)
- Forgetting that `.get()` returns `None` by default
- Using `.get()` when failure should actually stop execution
- Treating missing data as equivalent to valid data without checking

---

### Key Takeaways
- `dict[key]` performs direct lookup and raises errors if keys are missing
- `.get()` performs safe lookup and never raises `KeyError`
- `.get()` returns the **actual value**, not a view
- Default values make missing data handling explicit
- Safe access patterns are essential for analytics and data pipelines

Mastering this distinction prevents subtle bugs and makes your code resilient.


## Section 8.05 — Applied Dictionary Patterns

### Overview
Up to this point, dictionaries have been introduced as:
- containers for key–value pairs
- objects that support safe access
- structures that can be built dynamically

In this section, dictionaries become **problem-solving tools**.

These patterns show how dictionaries are used in realistic scenarios such as:
- counting occurrences
- mapping one set of values to another
- incrementally updating results
- preparing data for analysis

This is where dictionaries stop being “a data type” and start becoming **analytical instruments**.

---

### Core Patterns Introduced
- Counting and frequency tables
- Mapping transformations
- Incremental updates
- Business-style aggregations

---

### Pattern 1 — Counting with Dictionaries (Frequency Tables)

Counting occurrences is one of the most common dictionary use cases.

```python
# Count how many times each product appears
sales_log = [
    "skis", "boots", "skis", "goggles",
    "boots", "skis", "snowboard", "boots"
]

product_counts = {}

for product in sales_log:
    # Use .get() to safely retrieve the current count
    product_counts[product] = product_counts.get(product, 0) + 1

print(product_counts)
```

**What this demonstrates**
- Dictionaries act as counters
- `.get()` enables safe incremental updates
- Missing keys are handled intentionally

This pattern appears constantly in analytics, logging, and feature engineering.

---

### Pattern 2 — Mapping Transformations

Dictionaries are often used to translate values from one domain to another.

```python
# Map product codes to descriptive names
product_codes = ["S", "B", "G", "S", "B"]
code_map = {
    "S": "skis",
    "B": "boots",
    "G": "goggles"
}

translated_products = []

for code in product_codes:
    translated_products.append(code_map.get(code, "UNKNOWN"))

print(translated_products)
```

**Why this matters**
- Separates raw data from meaning
- Allows controlled defaults
- Mirrors categorical encoding in analytics

---

### Pattern 3 — Incremental Updates (Running Totals)

Dictionaries can store values that evolve over time.

```python
# Track total sales per customer
transactions = [
    ("C001", 120.50),
    ("C002", 75.00),
    ("C001", 40.25),
    ("C003", 210.10),
    ("C002", 15.75)
]

customer_totals = {}

for customer_id, amount in transactions:
    customer_totals[customer_id] = customer_totals.get(customer_id, 0) + amount

print(customer_totals)
```

**What this mirrors**
- Grouped aggregation
- Running totals
- Precursor to `groupby` logic in pandas

---

### Pattern 4 — Business-Style Classification

Dictionaries are often used to classify entities based on rules.

```python
# Classify customers by total spend
customer_status = {}

for customer_id, total in customer_totals.items():
    if total >= 200:
        customer_status[customer_id] = "VIP"
    else:
        customer_status[customer_id] = "Standard"

print(customer_status)
```

**Why this is important**
- Encodes business rules explicitly
- Produces structured outputs
- Supports downstream reporting and analysis

---

### Common Beginner Mistakes
- Forgetting to initialize dictionary accumulators
- Using direct lookup (`[]`) instead of `.get()` for incremental updates
- Mixing raw data and transformed data unintentionally
- Overwriting values when accumulation is required

---

### Key Takeaways
- Dictionaries are powerful tools for applied problem solving
- Counting and aggregation are natural dictionary use cases
- `.get()` enables safe incremental updates
- Mapping transformations separate codes from meaning
- These patterns directly prepare you for DataFrames and analytics workflows

At this stage, dictionaries are no longer just containers —
they are **structures for reasoning about data**.


## Section 8.06 — Building Dictionaries Dynamically

### Overview
In practice, dictionaries are rarely written out by hand.
They are usually **constructed dynamically** from:
- loops
- multiple data sources
- intermediate results

This section focuses on how dictionaries are *built*, not just how they are accessed.

Dynamic construction is foundational for:
- analytics pipelines
- feature engineering
- reshaping raw data into structured formats

---

### Core Concepts
- Using empty dictionaries as accumulators
- Adding key–value pairs inside loops
- Building dictionaries from multiple sequences
- Using `zip()` to align related data safely

---

### Why This Matters (Analytics Framing)
Real-world data often arrives as:
- parallel lists
- IDs and attributes
- raw records that must be reshaped

Being able to **programmatically assemble dictionaries** allows you to:
- map identifiers to values
- construct features
- prepare data for pandas and visualization tools

This is one of the most common dictionary patterns in analytics.

---

### Pattern 1 — Dictionary Accumulator (Loop-Based Construction)

Start with an empty dictionary and add entries incrementally.

```python
# Initialize an empty dictionary accumulator
inventory_status = {}

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

for i, item in enumerate(items):
    if inventory[i] == 0:
        inventory_status[item] = "sold out"
    else:
        inventory_status[item] = "in stock"

print(inventory_status)
```

**What this demonstrates**
- Dictionaries can be built one entry at a time
- Keys and values are added dynamically
- Logic determines the stored value

This pattern is common when rules or conditions affect how data is classified.

---

### Pattern 2 — Building from Parallel Lists with `zip()`

When two lists are already aligned by position, `zip()` is often clearer than `enumerate()`.

```python
# Build a dictionary from aligned lists using zip()
inventory_status = {}

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

for item, qty in zip(items, inventory):
    if qty == 0:
        inventory_status[item] = "sold out"
    else:
        inventory_status[item] = "in stock"

print(inventory_status)
```

**Why `zip()` is useful**
- Eliminates manual index tracking
- Makes relationships explicit
- Reduces indexing errors

Use `zip()` when sequences are meant to be paired element-by-element.

---

### `enumerate()` vs `zip()`
Both approaches are valid, but they serve different purposes.

- Use `enumerate()` when:
  - you need the index
  - you are tracking positions explicitly

- Use `zip()` when:
  - lists represent parallel data
  - values should be processed together

Choosing between them is about **clarity**, not performance.

---

### Pattern 3 — Building Aggregated Dictionaries

Dictionaries are often built to store computed values, not raw data.

```python
# Aggregate values while building a dictionary
customer_sales = {
    "C00001": [224.99, 87.36],
    "C00004": [125.56],
    "C00014": [99.99, 104.26, 23.57],
    "C00015": [82.65, 56.88, 201.82, 519.55]
}

total_sales = {}

for customer_id, sales in customer_sales.items():
    total_sales[customer_id] = sum(sales)

print(total_sales)
```

**Why this pattern matters**
- Raw data is transformed into meaningful metrics
- The dictionary becomes a feature map
- This mirrors group-by operations in pandas

---

### Common Beginner Mistakes
- Predefining dictionary keys when they should be dynamic
- Misusing `enumerate()` when `zip()` is clearer
- Forgetting to initialize the dictionary accumulator
- Assuming lists will always stay aligned

---

### Key Takeaways
- Dictionaries are commonly built dynamically, not manually
- Empty dictionaries act as accumulators in loops
- `enumerate()` and `zip()` both support dynamic construction
- `zip()` is ideal for pairing parallel sequences
- Dynamic dictionary building is central to analytics workflows

Mastering these patterns prepares you directly for functions, pandas,
and real-world data transformations.


## Section 8.06 — Building Dictionaries from Parallel Data with `zip()`

### Overview
`zip()` is the bridge from “parallel lists” (spreadsheet-style columns) to **structured records** (dictionary-style rows).

It pairs elements **by position** across multiple iterables and produces an iterator of tuples.

In analytics terms:
- Parallel lists = columns that are aligned by row index
- `zip()` = row constructor
- Dictionary building = turning rows into keyed records

---

### Core Concepts
- `zip()` pairs items element-by-element
- `zip()` returns an **iterator** (lazy sequence)
- `zip()` stops at the **shortest** iterable
- `zip()` + loop accumulator builds dictionaries explicitly
- `dict(zip(keys, values))` is the “compressed” version (only for 2 iterables)
- For 3+ iterables, you typically store a **tuple** or **nested dict** as the value

---

## Part A — Seeing What `zip()` Produces (No Dictionary Yet)

```python
# zip() pairs elements by position
items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

z = zip(items, inventory)

# zip() returns an iterator (not a list)
print(z)

# Convert to list ONLY for inspection / learning (consumes the iterator)
print(list(z))
```

### Key idea
After you convert a zip object to a list once, it is **exhausted**.

```python
# Demonstrate exhaustion
z = zip(items, inventory)
print(list(z))   # consumes it
print(list(z))   # now empty, because it was already consumed
```

---

## Part B — `zip()` Stops at the Shortest Iterable (Important “Data Quality” Rule)

```python
items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0]  # shorter list

paired = list(zip(items, inventory))
print(paired)

# Result: only 3 pairs are produced; "boots" is silently dropped
```

**Analytics framing:** this is a silent truncation risk.  
When data is supposed to align, you either:
- validate lengths, or
- deliberately slice to match

---

## 🟦 Helper 1 — `zip_trace()` (Inspect Zip Safely Without Losing the Iterator)

### Why this helper exists
When learning, you want to “peek” at zip output without consuming it accidentally.

This helper:
- shows the first `n` zipped tuples
- returns a **fresh** zip iterator so you can keep working

```python
def zip_trace(*iterables, n=5):
    """
    Peek at the first n tuples produced by zip(*iterables) without losing the full sequence.

    Returns a NEW zip iterator you can use afterward.
    """
    z = zip(*iterables)
    preview = []

    # Pull first n tuples safely
    for _ in range(n):
        try:
            preview.append(next(z))
        except StopIteration:
            break

    print("zip_trace preview:", preview)

    # Re-create zip so the caller still has a full iterator available
    return zip(*iterables)
```

### Example Usage
```python
items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

z = zip_trace(items, inventory, n=10)
print(list(z))  # still works, because zip_trace returned a fresh zip iterator
```

---

## Part C — Build a Dictionary with a Loop (Loop-First “Show Your Work” Version)

### Goal
Create a dictionary mapping:
- key = item name
- value = quantity

```python
items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

inventory_map = {}  # dictionary accumulator

for item, qty in zip(items, inventory):
    inventory_map[item] = qty

print(inventory_map)
```

**Why this matters:** This pattern is the foundation of building structured data from raw aligned inputs.

---

## Part D — `dict(zip(keys, values))` (Compressed Version for Two Iterables)

```python
items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

inventory_map_compact = dict(zip(items, inventory))
print(inventory_map_compact)
```

**Rule:** `dict(zip(keys, values))` only makes a dictionary directly when:
- the first iterable is keys
- the second iterable is values
- and you want a 1-to-1 pairing

---

## Part E — `zip()` with 3+ Lists (Store Multiple Fields Per Key)

When you have multiple “columns,” you usually build:
- key = ID (or name)
- value = tuple of fields (positional schema) OR dict of fields (named schema)

### Option 1 — Tuple schema (positional, fast, but brittle)
```python
item_ids = [10001, 10002, 10003]
item_names = ["Coffee", "Beanie", "Gloves"]
prices = [5.27, 8.79, 17.59]

item_data_tuple = {}

for item_id, name, price in zip(item_ids, item_names, prices):
    item_data_tuple[item_id] = (name, price)

print(item_data_tuple)

# Access example:
print(item_data_tuple[10003][0])  # name (index-based)
print(item_data_tuple[10003][1])  # price
```

### Option 2 — Named schema (dictionary as the value; clearer, safer)
```python
item_data_named = {}

for item_id, name, price in zip(item_ids, item_names, prices):
    item_data_named[item_id] = {"name": name, "price": price}

print(item_data_named)

# Access example:
print(item_data_named[10003]["name"])
print(item_data_named[10003]["price"])
```

**Analytics framing:** named schema reduces index mistakes and scales better as fields grow.

This is the direct bridge to **nested dictionaries** (your next section).

---

## 🟦 Helper 2 — `validate_parallel_lengths()` (Prevent Silent Truncation)

### Why this helper exists
`zip()` silently truncates at the shortest iterable. This helper makes alignment explicit.

```python
def validate_parallel_lengths(**named_iterables):
    """
    Validate that multiple parallel iterables have the same length.
    Pass them as keyword arguments so the output is readable.

    Example: validate_parallel_lengths(items=items, inventory=inventory)
    """
    lengths = {name: len(seq) for name, seq in named_iterables.items()}
    print("Lengths:", lengths)

    unique_lengths = set(lengths.values())
    if len(unique_lengths) == 1:
        print("✅ All iterables align (same length).")
        return True
    else:
        print("⚠️ Length mismatch detected. zip() would truncate to the shortest length.")
        shortest = min(lengths.values())
        print(f"Shortest length = {shortest}. Consider slicing or fixing upstream data.")
        return False
```

### Example Usage
```python
items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0]

validate_parallel_lengths(items=items, inventory=inventory)

# If you intentionally want alignment by truncation:
inventory_map = dict(zip(items[:len(inventory)], inventory))
print(inventory_map)
```

---

## Key Takeaways
- `zip()` pairs aligned elements across iterables and produces tuples.
- `zip()` is lazy (iterator) and is consumed when iterated.
- `zip()` stops at the shortest iterable (silent truncation risk).
- Loop-first dictionary building is the “show your work” approach:
  - `for a, b in zip(...): dict[a] = b`
- `dict(zip(keys, values))` is a clean compression for exactly **two** iterables.
- For 3+ iterables, store a tuple (positional schema) or a dict (named schema) as the value.
- This section is the direct bridge to the next topic: **nested dictionaries** (two-key lookups).


## Comprehensions from First Principles — “Show Your Work” Build-Up (Dictionary Comprehension)

### Overview
A dictionary comprehension is **not** new logic.  
It is a **compression** of logic you already know:

- loop over `(key, value)` pairs
- compute something
- store it in a new dictionary

If you can write it long-form, you can always prove the comprehension is equivalent.

---

## Given Data (Raw Inputs)
```python
# Raw data: customer → list of transactions
customer_sales = {
    "C00001": [224.99, 87.36],
    "C00004": [125.56],
    "C00014": [99.99, 104.26, 23.57],
    "C00015": [82.65, 56.88, 201.82, 519.55]
}
```

---

## Step 0 — The Goal (State It Explicitly)
We want a new dictionary where:
- keys = customer IDs
- values = total of their sales list

So:
```python
# Desired shape:
# {
#   "C00001": 312.35,
#   "C00004": 125.56,
#   ...
# }
```

---

## Step 1 — Long-Form Loop with Accumulator (Full “Show Your Work”)
```python
total_sales = {}  # empty dictionary accumulator

for customer_id, sales in customer_sales.items():
    # Compute the total for this customer
    customer_total = sum(sales)
    
    # Store it using the same key (customer_id)
    total_sales[customer_id] = customer_total

print(total_sales)
```

**What you proved**
- `.items()` gives `(customer_id, sales_list)`
- `sum(sales_list)` produces a single numeric total
- that total is stored under the same key

---

## Step 2 — Make the Assignment Explicit (Highlight the Pattern)
Here we isolate the “template”:

```python
total_sales = {}

for customer_id, sales in customer_sales.items():
    # (A) compute a new value from the old value
    new_value = sum(sales)
    
    # (B) store it under the same key
    total_sales[customer_id] = new_value

print(total_sales)
```

**Pattern recognized**
> “For each key, compute a transformed value, then store it.”

---

## Step 3 — Prove the Loop’s Output Matches a Manual Check
This is your “math proof” moment: verify at least one row.

```python
# Manual verification for one customer:
print(sum(customer_sales["C00001"]))   # should match total_sales["C00001"]
print(total_sales["C00001"])
```

If those match, your logic is correct.

---

## Step 4 — The Same Loop, But Written as a Single Mapping Rule
Now we write the rule in English:

> For each `(customer_id, sales)` in `customer_sales.items()`, map  
> `customer_id → sum(sales)`.

That sentence is the comprehension.

---

## Step 5 — Dictionary Comprehension (Compressed Version of the Same Logic)
```python
total_sales_comp = {
    customer_id: sum(sales)
    for customer_id, sales in customer_sales.items()
}

print(total_sales_comp)
```

---

## Step 6 — Prove Equivalence (Loop Output vs Comprehension Output)
This is the “show your work” equivalence check:

```python
# Build with the loop
total_sales_loop = {}

for customer_id, sales in customer_sales.items():
    total_sales_loop[customer_id] = sum(sales)

# Build with the comprehension
total_sales_comp = {
    customer_id: sum(sales)
    for customer_id, sales in customer_sales.items()
}

# Proof: these should be identical dictionaries
print(total_sales_loop == total_sales_comp)
```

If this prints `True`, the comprehension is proven equivalent.

---

## Practice Variations (Build Fluency Without Skipping Steps)

### Variation A — Round totals to 2 decimals
```python
total_sales_rounded = {}

for customer_id, sales in customer_sales.items():
    total_sales_rounded[customer_id] = round(sum(sales), 2)

print(total_sales_rounded)
```

**Comprehension version**
```python
total_sales_rounded_comp = {
    customer_id: round(sum(sales), 2)
    for customer_id, sales in customer_sales.items()
}

print(total_sales_rounded_comp)
```

---

### Variation B — Count the number of purchases per customer
```python
purchase_counts = {}

for customer_id, sales in customer_sales.items():
    purchase_counts[customer_id] = len(sales)

print(purchase_counts)
```

**Comprehension version**
```python
purchase_counts_comp = {
    customer_id: len(sales)
    for customer_id, sales in customer_sales.items()
}

print(purchase_counts_comp)
```

---

### Variation C — Create a nested “feature map” per customer
(Example: total + number of purchases)

```python
customer_features = {}

for customer_id, sales in customer_sales.items():
    customer_features[customer_id] = {
        "total": round(sum(sales), 2),
        "num_purchases": len(sales)
    }

print(customer_features)
```

**Comprehension version**
```python
customer_features_comp = {
    customer_id: {
        "total": round(sum(sales), 2),
        "num_purchases": len(sales)
    }
    for customer_id, sales in customer_sales.items()
}

print(customer_features_comp)
```

---

## Key Takeaways
- A dictionary comprehension is a **compressed loop** that builds a dictionary.
- You can always expand a comprehension into:
  1) an empty accumulator dictionary  
  2) a `for` loop over `.items()`  
  3) a computed value  
  4) an assignment to `new_dict[key]`
- Equivalence can be proven by comparing outputs (`==`).
- For your learning style: use the loop first, then compress only after the logic is fully proven.


## Section 8.07 — Nested Dictionaries (Structured Records & Two-Level Lookup)

### Overview
Nested dictionaries allow you to model **real-world records** where each entity has multiple named attributes.

Instead of:
- one key → one value  

you now have:
- one key → **a dictionary of attributes**

This is the first time Python data starts to resemble:
- database rows
- JSON objects
- pandas DataFrame rows

Conceptually:
- outer key = unique identifier (ID)
- inner dictionary = schema (fields)

---

### Core Concepts
- Dictionaries can store other dictionaries as values
- Access becomes **two-step lookup**
- Named fields reduce index-based errors
- Nested dictionaries scale better than tuple-based schemas
- This pattern is foundational for analytics, APIs, and pandas

---

## Part A — From Flat to Nested (Why This Exists)

### Flat dictionary (limited meaning)
```python
# Flat structure: meaning depends on position
item_data_flat = {
    10001: ("Coffee", 5.99, "beverage"),
    10002: ("Beanie", 9.99, "clothing")
}

# Access is positional and fragile
print(item_data_flat[10001][1])  # price, but only if you remember index 1
```

Problems:
- You must remember index positions
- Adding fields breaks old code
- Hard to read six months later

---

### Nested dictionary (named schema)
```python
item_data = {
    10001: {
        "name": "Coffee",
        "price": 5.99,
        "category": "beverage"
    },
    10002: {
        "name": "Beanie",
        "price": 9.99,
        "category": "clothing"
    }
}

# Access is explicit and readable
print(item_data[10001]["price"])
```

---

## Part B — Visualizing the Structure (Mental Model)

```
item_data
│
├── 10001 ──► {
│              "name": "Coffee",
│              "price": 5.99,
│              "category": "beverage"
│            }
│
└── 10002 ──► {
               "name": "Beanie",
               "price": 9.99,
               "category": "clothing"
             }
```

You are **not** “looking up two keys at once.”  
You are doing **sequential lookup**:

1. outer dictionary → record
2. inner dictionary → field

---

## Part C — Two-Level Lookup (Step-by-Step)

```python
# Step 1: get the inner dictionary (the record)
record = item_data[10002]
print(record)

# Step 2: get a specific field from the record
price = record["price"]
print(price)
```

Condensed (but equivalent):
```python
print(item_data[10002]["price"])
```

---

## 🟦 Helper 1 — `nested_lookup_trace()` (Make Two-Level Access Visible)

### Why this helper exists
Nested lookups fail silently *until they don’t*.
This helper shows each step explicitly.

```python
def nested_lookup_trace(data, outer_key, inner_key):
    """
    Trace a two-level dictionary lookup.
    Shows whether the outer key and inner key exist.
    """
    if outer_key not in data:
        print(f"Outer key NOT found: {outer_key}")
        return None

    print(f"Outer key found: {outer_key}")
    record = data[outer_key]

    if inner_key not in record:
        print(f"Inner key NOT found: '{inner_key}'")
        return None

    print(f"Inner key found: '{inner_key}' → {record[inner_key]}")
    return record[inner_key]
```

### Example Usage
```python
nested_lookup_trace(item_data, 10001, "price")
nested_lookup_trace(item_data, 10001, "sizes")
nested_lookup_trace(item_data, 99999, "price")
```

---

## Part D — Building Nested Dictionaries with Loops (No Comprehension Yet)

### Goal
Build this structure dynamically from parallel lists.

```python
item_ids = [10001, 10002, 10003]
names = ["Coffee", "Beanie", "Gloves"]
prices = [5.99, 9.99, 19.99]
categories = ["beverage", "clothing", "clothing"]

item_data = {}  # outer dictionary accumulator

for item_id, name, price, category in zip(item_ids, names, prices, categories):
    item_data[item_id] = {
        "name": name,
        "price": price,
        "category": category
    }

print(item_data)
```

**Key idea:**  
The loop builds **records**, not just values.

---

## Part E — Safe Nested Access with `.get()`

```python
# Safe outer lookup
record = item_data.get(10004)

if record is None:
    print("Item not found")
else:
    # Safe inner lookup
    price = record.get("price", "Price unavailable")
    print(price)
```

**Analytics framing:**  
This mirrors defensive access in:
- APIs
- messy datasets
- optional fields

---

## 🟦 Helper 2 — `safe_nested_get()` (One Call, Two Levels)

```python
def safe_nested_get(data, outer_key, inner_key, default=None):
    """
    Safely retrieve a nested dictionary value.
    Returns default if any key is missing.
    """
    outer = data.get(outer_key)
    if outer is None:
        return default
    return outer.get(inner_key, default)
```

### Example Usage
```python
print(safe_nested_get(item_data, 10002, "price"))
print(safe_nested_get(item_data, 10002, "sizes", []))
print(safe_nested_get(item_data, 99999, "price", "Not found"))
```

---

## Part F — Why Nested Dictionaries Matter (Analytics Framing)

Nested dictionaries map directly to:
- JSON payloads
- API responses
- row-based records
- pandas DataFrame rows (before tabularization)

They allow:
- explicit schemas
- safer transformations
- scalable feature construction

This is **not** a detour — it is the on-ramp to:
- comprehensions
- functions
- pandas
- real analytics pipelines

---

## Key Takeaways
- Nested dictionaries model structured records with named fields.
- Access is sequential: outer key → inner key.
- Named schemas are safer than positional tuples.
- Loop-first construction builds intuition and correctness.
- Helpers make multi-level access debuggable.
- This pattern directly prepares you for:
  - dictionary comprehensions
  - JSON handling
  - pandas DataFrames

Next up: **Sets — uniqueness, membership, and why they are fast.**


## Section 8.08–8.09 — Sets: Uniqueness, Membership, and Relationships

### Overview
Sets solve a very specific class of problems:
- **uniqueness**
- **fast membership testing**
- **relationships between groups**

They are not about order.
They are not about indexing.
They are about **presence vs. absence**.

In analytics terms, sets answer questions like:
- “Which categories exist?”
- “Which IDs are duplicated?”
- “Which customers appear in both groups?”
- “What is new, missing, or overlapping?”

---

## Core Concepts
- Set literals and construction
- Automatic uniqueness (duplicates removed)
- Fast membership testing (`in`)
- Converting lists to sets
- Set relationships:
  - union
  - intersection
  - difference
  - symmetric difference

---

## Part A — Why Sets Exist (The Duplication Problem)

```python
# Raw data with duplicates
categories = [
    "clothing", "hardware", "clothing", "beverage",
    "clothing", "hardware", "clothing"
]

print(categories)
```

If you ask:
> “What categories do we carry?”

A list is the **wrong structure**.

## Part B — Converting a List to a Set (Automatic Deduplication)

```python
unique_categories = set(categories)
print(unique_categories)
```

Key observations:
- Duplicates are removed automatically
- Sets are unordered (do not rely on display order)
- This operation is intentional and meaningful

**Analytics framing:**  
This mirrors extracting **distinct values** from a column.

---

## 🟦 Helper 1 — `set_build_trace()` (See Deduplication Happen)

### Why this helper exists
Sets hide work by design. This helper makes the transition visible.

```python
def set_build_trace(iterable):
    """
    Trace how a set grows as items are added.
    """
    result = set()
    for item in iterable:
        before = result.copy()
        result.add(item)
        if before != result:
            print(f"Added: {item} → {result}")
        else:
            print(f"Ignored duplicate: {item}")
    return result
```

### Example Usage
```python
categories = ["clothing", "hardware", "clothing", "beverage"]
final_set = set_build_trace(categories)
print("Final set:", final_set)
```

## Part C — Membership Testing (Why Sets Are Fast)

```python
categories_set = set(categories)

print("clothing" in categories_set)   # True
print("electronics" in categories_set)  # False
```

Why this matters:
- `in` on a list → linear search
- `in` on a set → hash-based lookup (very fast)

**Analytics framing:**  
Sets are ideal for:
- validation rules
- filters
- allowed-value checks

## 🟦 Helper 2 — `set_membership_trace()` (Explain Membership Logic)

```python
def set_membership_trace(value, reference_set):
    """
    Explain membership testing against a set.
    """
    if value in reference_set:
        print(f"'{value}' FOUND in set.")
        return True
    else:
        print(f"'{value}' NOT found in set.")
        return False
```

### Example Usage
```python
allowed_categories = {"clothing", "hardware", "beverage"}

```

## Section 8.09 — Set Operations: Relationships Between Groups

### Overview
Up to this point, sets have been used to answer a simple question:

> “Does this value exist?”

Set operations extend that idea to a more powerful question:

> “How do groups of values relate to one another?”

These operations allow you to:
- compare datasets
- detect overlap
- identify exclusions
- reason about categories and cohorts

This section introduces **set relationships**, not mathematics.

---

### Core Concepts
- Union: values in *either* group
- Intersection: values in *both* groups
- Difference: values in *one group but not another*
- Symmetric difference: values in *either group, but not both*

---

### Why This Matters (Analytics Framing)
In analytics, set operations appear everywhere:
- cohort analysis
- customer overlap
- segmentation
- rule-based filtering

These operations help answer questions like:
- Which users appear in both datasets?
- Which categories are unique to one source?
- What changed between two time periods?

---

### Baseline Sets
```python
# Two example datasets
online_customers = {"C001", "C002", "C003", "C005"}
store_customers = {"C002", "C003", "C004"}
```

---

### Union — Combine All Unique Values
```python
# Customers who purchased online OR in-store
all_customers = online_customers | store_customers
print(all_customers)
```

**Interpretation**
- Includes anyone who appears in either group
- Removes duplicates automatically

This is useful when combining datasets.

---

### Intersection — Shared Values Only
```python
# Customers who purchased both online AND in-store
shared_customers = online_customers & store_customers
print(shared_customers)
```

**Interpretation**
- Identifies overlap
- Commonly used for cohort analysis

This answers the question:
> “Who appears in both groups?”

---

### Difference — Values in One Group but Not the Other
```python
# Customers who purchased online but not in-store
online_only = online_customers - store_customers
print(online_only)
```

**Interpretation**
- Highlights exclusivity
- Useful for targeting or exclusion rules

Order matters for difference operations.

---

### Symmetric Difference — Values in Either Group, But Not Both
```python
# Customers who purchased in exactly one channel
exclusive_customers = online_customers ^ store_customers
print(exclusive_customers)
```

**Interpretation**
- Removes shared values
- Identifies what is unique across groups

This is often used for change detection.

---

### When to Use Each Operation

- **Union (`|`)**
  - Combine datasets
  - Build complete populations

- **Intersection (`&`)**
  - Find overlap
  - Analyze shared behavior

- **Difference (`-`)**
  - Exclude categories
  - Identify one-sided membership

- **Symmetric Difference (`^`)**
  - Detect changes
  - Compare mutually exclusive groups

---

### Common Beginner Mistakes
- Expecting order to be preserved
- Using lists when uniqueness is required
- Forgetting that difference is directional
- Treating set operations as mathematical abstractions instead of data tools

---

### Key Takeaways
- Set operations describe relationships between groups
- These patterns map directly to analytics use cases
- Sets make overlap and exclusion explicit
- Operations are expressive, readable, and intentional

At this stage, sets are no longer just containers —
they are **tools for reasoning about data relationships**.

---

## Chapter 08 — Closing Perspective

Dictionaries and sets are not endpoints.

They are:
- building blocks for structured data
- tools for modeling relationships
- foundations for analytical thinking in Python

In this chapter, you learned how to:
- access data safely and intentionally
- build dictionaries dynamically
- apply dictionaries to real problems
- enforce uniqueness with sets
- analyze relationships between groups

With these skills:
- functions will feel more natural
- pandas concepts will be intuitive
- analytics workflows will be easier to reason about

You are now prepared to move from **structures** to **abstractions** —
from containers to **computation**.

## Brief Intro (What these helpers are for)
These helpers are your “debugging-by-design” toolkit for Chapter 08.

Use them when you want debugging to be **diagnostic, not desperate**:
- You want to **inspect structure** (without guessing).
- You want to **validate assumptions** (alignment, schema, keys).
- You want to **trace logic** step-by-step (nested lookups, membership tests).
- You want reusable patterns that will transfer directly into pandas and real data pipelines.

---

## Helper Groups (What you have in this pack)

### 1) 🧩 Alignment & Construction Helpers (zip / parallel data)
These prevent brittle “parallel list” mistakes and help you safely inspect zipped data:
- `zip_trace()` — preview zipped tuples *without consuming the iterator*
- `validate_parallel_lengths()` — ensure iterables truly align before `zip()` silently truncates

### 2) 🧱 Nested Dictionary Access Helpers (schema + two-level lookup)
These make nested lookups explicit and safe:
- `nested_lookup_trace()` — step-by-step trace of outer key → inner key
- `safe_nested_get()` — one-call safe retrieval with a default if anything is missing

### 3) ♻️ Set Mechanics & Validation Helpers (dedupe, membership, relationships)
These helpers clarify what sets are doing and why they are useful:
- `set_build_trace()` — watch deduplication happen as the set grows
- `set_membership_trace()` — explain membership testing (`in`) clearly
- `set_relationship_trace()` — show union/intersection/differences in one place
- `validate_allowed()` — schema-style validation (allowed/expected vs invalid/unexpected)

---

## How to Use This Pack (recommended workflow)
1) Start with **alignment**: validate lengths before building records (`validate_parallel_lengths`).
2) Build structures (dicts/nested dicts) and **trace access** if anything feels unclear (`nested_lookup_trace`).
3) When shifting to sets:
   - trace deduplication (`set_build_trace`)
   - validate expected values (`validate_allowed`)
   - compare groups (`set_relationship_trace`)

---

## Guiding Principle
These helpers are **optional**—use them when they reduce uncertainty, clarify structure, or prevent silent failure.  
When the code is stable and obvious, you can remove them and keep the clean version.


`````python
# -----------------------------------------------
# Dictionary Baseline Example
# -----------------------------------------------
# Keys must be unique and immutable
# Values do not need to be unique and can be any data type


# Inventory status stored as key-value pairs:
inventory_status_dict = {
    'skis': 'in stock',
    'snowboard': 'sold out',
    'goggles': 'sold out',
    'boots': 'in stock'
}


# No loop required: direct key-based lookup
print(inventory_status_dict['goggles'])

# Output: 'sold out'


# -----------------------------------------------
# 🟦 Helper — `trace_dict_lookup`
# -----------------------------------------------
# This helper makes dictionary access visible and safe
# It shows:
# - what key was requested
# - whether the key exists
# - what value was returned (or why it was not)

# This prevents silent failures and reinforces best practices early.


def trace_dict_lookup(data, key):
    """
    Safely traces dictionary access.
    Shows whether the key exists and what value is returned.
    """
    if key in data:
        print(f"Key found: '{key}' -> Value: {data[key]}")
        return data[key]
    else:
        print(f"Key NOT found: '{key}'")
        return None

# Example Usage:
trace_dict_lookup(inventory_status_dict, "goggles")
trace_dict_lookup(inventory_status_dict, "boots")
trace_dict_lookup(inventory_status_dict, "bindings")

# Output:
# sold out
# Key found: 'goggles' -> Value: sold out
# Key found: 'boots' -> Value: in stock
# Key NOT found: 'bindings'



#----------------------------------------------------------------------------
# 🟦 Helper — `zip_trace()` (Inspect Zip Safely Without Losing the Iterator)
#----------------------------------------------------------------------------
# Why this helper exists
# When learning, you want to “peek” at zip output without consuming it accidentally.

# This helper:
# - shows the first `n` zipped tuples
# - returns a **fresh** zip iterator so you can keep working


def zip_trace(*iterables, n=5):
    """
    Peek at the first n tuples produced by zip(*iterables) without losing the full sequence.

    Returns a NEW zip iterator you can use afterward.
    """
    z = zip(*iterables)
    preview = []

    # Pull first n tuples safely
    for _ in range(n):
        try:
            preview.append(next(z))
        except StopIteration:
            break

    print("zip_trace preview:", preview)

    # Re-create zip so the caller still has a full iterator available
    return zip(*iterables)


# Example Usage

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0, 7]

z = zip_trace(items, inventory, n=10)
print(list(z))  # still works, because zip_trace returned a fresh zip iterator


#---------------------------------------------------------------------
# 🟦 Helper — `validate_parallel_lengths()` (Prevent Silent Truncation)
#---------------------------------------------------------------------
# Why this helper exists
# `zip()` silently truncates at the shortest iterable. This helper makes alignment explicit ↕️


def validate_parallel_lengths(**named_iterables):
    """
    Validate that multiple parallel iterables have the same length.
    Pass them as keyword arguments so the output is readable.

    Example: validate_parallel_lengths(items=items, inventory=inventory)
    """
    lengths = {name: len(seq) for name, seq in named_iterables.items()}
    print("Lengths:", lengths)

    unique_lengths = set(lengths.values())
    if len(unique_lengths) == 1:
        print("✅ All iterables align (same length).")
        return True
    else:
        print("⚠️ Length mismatch detected. zip() would truncate to the shortest length.")
        shortest = min(lengths.values())
        print(f"Shortest length = {shortest}. Consider slicing or fixing upstream data.")
        return False

# Example Usage

items = ["skis", "snowboard", "goggles", "boots"]
inventory = [10, 0, 0]

validate_parallel_lengths(items=items, inventory=inventory)

# If you intentionally want alignment by truncation: 🚧
inventory_map = dict(zip(items[:len(inventory)], inventory))
print(inventory_map)


#---------------------------------------------------------------------
# 🟦 Helper — `nested_lookup_trace()` (Make Two-Level Access Visible)
#---------------------------------------------------------------------
# Why this helper exists:
# Nested lookups fail silently until they don’t
# This helper shows each step explicitly


# Nested dictionary (named schema) 🧱 

item_ids = [10001, 10002, 10003]
names = ["Coffee", "Beanie", "Snowboard"]
prices = [5.99, 9.99, 599.99]
categories = ["beverage", "clothing", "equipment"]

item_data = {}

for item_id, name, price, category in zip(item_ids, names, prices, categories):
    item_data[item_id] = {
        "name": name,
        "price": price,
        "category": category
    }


def nested_lookup_trace(data, outer_key, inner_key):
    """
    Trace a two-level dictionary lookup.
    Shows whether the outer key and inner key exist.
    """
    if outer_key not in data:
        print(f"Outer key NOT found: {outer_key}")
        return None

    print(f"Outer key found: {outer_key}")
    record = data[outer_key]

    if inner_key not in record:
        print(f"Inner key NOT found: '{inner_key}'")
        return None

    print(f"Inner key found: '{inner_key}' → {record[inner_key]}")
    return record[inner_key]


# Example Usage:
nested_lookup_trace(item_data, 10001, "price")
nested_lookup_trace(item_data, 10001, "sizes")
nested_lookup_trace(item_data, 99999, "price")

# Output:
# Outer key found: 10001
# Inner key found: 'price' → 5.99
# Outer key found: 10001
# Inner key NOT found: 'sizes'
# Outer key NOT found: 99999



# 🟦 Helper — `safe_nested_get()` (One Call, Two Levels)


def safe_nested_get(data, outer_key, inner_key, default=None):
    """
    Safely retrieve a nested dictionary value.
    Returns default if any key is missing.
    """
    outer = data.get(outer_key)
    if outer is None:
        return default
    return outer.get(inner_key, default)


# Example Usage:
print(safe_nested_get(item_data, 10002, "price"))
print(safe_nested_get(item_data, 10005, "category", "[] sentinel default"))
print(safe_nested_get(item_data, 99999, "price", "Not found"))



# ---------------------------------------------------------------
# 🟦 Helper — `set_build_trace()` (See Deduplication Happen) 🎯
# ---------------------------------------------------------------
# Why this helper exists
# Sets hide work by design. This helper makes the transition visible.

def set_build_trace(iterable):
    """
    Trace how a set grows as items are added.
    """
    result = set()
    for item in iterable:
        before = result.copy()
        result.add(item)
        if before != result:
            print(f"Added: {item} → {result}")
        else:
            print(f"Ignored duplicate: {item}")
    return result


# Example Usage:
categories_demo = ["clothing", "hardware", "clothing", "beverage"]
final_set = set_build_trace(categories_demo)
print("Final set:", final_set)

# Output:
# Added: clothing → {'clothing'}
# Added: hardware → {'hardware', 'clothing'}
# Ignored duplicate: clothing
# Added: beverage → {'hardware', 'beverage', 'clothing'}
# Final set: {'hardware', 'beverage', 'clothing'}



# -----------------------------------------------------------------
# 🟦 Helper — `set_membership_trace()` (Explain Membership Logic) 🎯
# -----------------------------------------------------------------
 
def set_membership_trace(value, reference_set):
    """
    Explain membership testing against a set.
    """
    if value in reference_set:
        print(f"'{value}' FOUND in set.")
        return True
    else:
        print(f"'{value}' NOT found in set.")
        return False


# Example Usage:
# Schema-defined allowed values
allowed_categories = set(categories)
print(allowed_categories)   # {"clothing", "hardware", "beverage"}

set_membership_trace("clothing", allowed_categories)
set_membership_trace("electronics", allowed_categories)

# Output:
# 'clothing' FOUND in set.
# 'electronics' NOT found in set.
# False



# ---------------------------------------------------------------------
# 🟦 Helper — `set_relationship_trace()` (Explain Set Logic Clearly) 🎯
# ---------------------------------------------------------------------

# Think of sets as groups of IDs or labels

store_a = {"skis", "boots", "helmet", "gloves"}
store_b = {"boots", "helmet", "goggles"}



def set_relationship_trace(a, b):
    """
    Show all major set relationships between two sets.
    """
    print("Set A:", a)
    print("Set B:", b)
    print("In A or B (Union):", a | b)
    print("In both A and B (Intersection):", a & b)
    print("Only in A:", a - b)
    print("Only in B:", b - a)
    print("In exactly one set (Symmetric Difference):", a ^ b)

# Example Usage
set_relationship_trace(store_a, store_b)

# Output:
# Set A: {'skis', 'boots', 'gloves', 'helmet'}
# Set B: {'boots', 'goggles', 'helmet'}
# In A or B (Union): {'skis', 'goggles', 'boots', 'gloves', 'helmet'}
# In both A and B (Intersection): {'boots', 'helmet'}
# Only in A: {'gloves', 'skis'}
# Only in B: {'goggles'}
# In exactly one set (Symmetric Difference): {'skis', 'goggles', 'gloves'}



# -----------------------------------------------------------------
# 🟦 Helper — `validate_allowed()` (Validate Expected Categories)
# -----------------------------------------------------------------

def validate_allowed(value, allowed_set):
    """
    Validate a value against an allowed set.
    Returns True/False and prints a clear message.
    """
    if value not in allowed_set:
        print(f"⚠️ Invalid value: '{value}'. Allowed: {allowed_set}")
        return False
    print(f"✅ Valid: '{value}'")
    return True


# Example Usage:
validate_allowed("clothing", allowed_categories)
validate_allowed("electronics", allowed_categories)

# Output:
# ✅ Valid: 'clothing'
# ⚠️ Invalid value: 'electronics'. Allowed: {'hardware', 'clothing', 'beverage'}
# False
`````

# 🧠 SECTION 1 — Introduction to Functions
## Chapter 9. Functions as Analytical Abstractions

## Why Functions Exist (Beyond “Reusability”)

In analytics, functions are not just a way to avoid copy-paste code.  
They are **abstractions** — named, trusted operations that convert inputs into outputs in a predictable way.

Think of a function as a **mini-pipeline**:

Input data  
→ validation  
→ transformation  
→ output  

This mirrors how real analytics systems work.

Built-in functions like `max()`, `len()`, and `sum()` already follow this pattern.  
Our goal in this chapter is to learn how to **design our own** functions with the same discipline.

---

## The Anatomy of a Function Call

A function has four conceptual parts:

1. **Inputs (Arguments)**  
   Data and configuration values passed into the function

2. **Processing Logic**  
   The work performed on the inputs

3. **Outputs (Return Values)**  
   The result(s) sent back to the caller

4. **Side Effects (Optional)**  
   Actions taken that are not returned (printing, saving files, logging)

Let’s ground this with a familiar built-in example.

---

## Example: `max()` as a Mental Model

```python
# Input data: a list of prices
price_list = [5.99, 19.99, 24.99, 0, 74.99, 99.99]

# Function call
highest_price = max(price_list)

# Output value: 99.99
```

What happened conceptually:

- **Input**: `price_list`
- **Processing**: Python compares all values
- **Output**: `99.99`
- **Side effects**: None

This exact structure is what we will replicate with custom functions.

---

## Why This Matters in Data Analysis

In real analytics work, you routinely:

- Clean data the same way across projects
- Apply consistent business rules
- Transform raw values into standardized outputs
- Validate inputs before analysis

Functions allow you to:

- Encode **business logic once**
- Reuse it safely
- Improve readability of notebooks
- Reduce error rates

A notebook full of functions reads like **documentation**, not just code.

---

## Preview: From Script Thinking to Function Thinking

Compare these two mindsets:

**Script mindset:**  
“Run these steps top to bottom.”

**Function mindset:**  
“Define reliable operations, then compose them.”

The second mindset scales.  
The first does not.

Everything else in this chapter builds on this shift.

---

## Key Takeaways

- Functions are analytical abstractions, not just shortcuts
- Every function follows input → processing → output
- Built-in functions model best practices we should emulate
- Strong function design leads to cleaner, safer analytics workflows
- This chapter is about **thinking in functions**, not memorizing syntax


# 🧠 SECTION 2 — Functions
## Chapter 9. Preview: Defining Custom Functions (Foundations)

## The Purpose of `def`

When you define a function, you are doing three things simultaneously:

1. **Naming a transformation**
2. **Declaring inputs the function expects**
3. **Guaranteeing what the function will return**

In analytics, a function definition is a **contract**:
> “If you give me X (in the correct form), I will reliably give you Y.”

This mindset prevents fragile notebooks and silent errors.

---

## Basic Function Structure

Every function definition follows this structure:

```python
def function_name(parameters):
    """
    Docstring (optional but strongly recommended)
    """
    # processing logic
    return output
```

Key rules:
- `def` starts the definition
- Parameters live inside parentheses
- The colon (`:`) signals a new code block
- `return` explicitly sends a value back to the caller

Without `return`, the function returns `None` by default.

---

## A Minimal, Correct Function

Let’s begin with a deliberately simple function so we can focus on structure.

```python
def add_numbers(a, b):
    """
    Adds two numeric values together.

    Args:
        a (int or float): First number
        b (int or float): Second number

    Returns:
        int or float: Sum of a and b
    """
    total = a + b
    return total
```

Function call:

```python
result = add_numbers(5, 7)
result

# Output: 12
```

What matters here is not the math.  
What matters is **clarity of intent**.

---

# Parameters vs Arguments (Precision Matters)

- **Parameters** are defined in the function signature  
- **Arguments** are the actual values passed during a function call

```python
# parameters: a, b
def multiply(a, b):     # ← Here, 'a' and 'b' are the PARAMETERS
    return a * b

# arguments: 4, 10
result = multiply(4, 10)   # ← Here, 4 and 10 are the ARGUMENTS

# Example:
def greet(name, age):           # parameters: name, age
    print(f"Hello {name}, you are {age} years old.")

greet("Alice", 25)              # arguments: "Alice", 25

# Output: Hello Alice, you are 25 years old.
```
In analytics documentation and team discussions, using these terms correctly matters.

---

## Chapter 9. Section 2: Summary

- `def` creates a named, reusable transformation
- Parameters define expected inputs
- Arguments are supplied at call time
- `return` is mandatory for reusable analytics logic
- Docstrings are part of professional-quality code
- Functions should return data, not print it


`````python
# --------------------------------------------
# Defining a Function
# Parameters vs Arguments
# --------------------------------------------
# - **Parameters** are defined in the function signature  
# - **Arguments** are the actual values passed during a function call

def build_greeting(name, age):           # parameters: name, age
    """
    Builds a greeting message.

    Args:
        name (str): Person's name
        age (int): Person's age

    Returns:
        str: Greeting message
    """
    return f"Hello {name}, you are {age} years old."

# Usage:
message = build_greeting("Alice", 25)    # arguments: "Alice", 25
print(message)

# Output: Hello Alice, you are 25 years old.


# --------------------------------------------
# Example 1: Defining a function
# Concatenate two words separated by a space
# --------------------------------------------

def concatenator(string1, string2):
    combined_string = string1 + " " + string2
    return combined_string

# Call function:
concatenator("Hello", "World!")

# Output: 'Hello World!'

# --------------------------------------------
# Example 2: Defining a function
# Tax calculator that takes a subtotal and 
# a tax rate and returns the total invoice amount
# --------------------------------------------
sub_total = 100.00
sales_tax = 0.06

def tax_calc(sub_total, sales_tax):
    """Calculates sales tax for a given subtotal.

    Args:
        sub_total (float): Pre-tax amount
        sales_tax (float): Tax rate as a decimal
    
    Returns:
        float: Tax amount rounded to 2 decimal places
    """
    total = sub_total + (sub_total * sales_tax)
    return round(total, 2)    # ← This rounds to 2 decimal places

# Call function:
total = tax_calc(sub_total, sales_tax)
print(total)  # Output: 106.0

# --------------------------------------------
# Return Values Are the Output Boundary
# A function should return data, not print data
# --------------------------------------------
# Why this matters:
# - Returned values can be reused
# - Printed values cannot
# - Functions should be composable

# Bad analytics practice: ⚠️
def calculate_tax_bad(subtotal, tax_rate):
    print(subtotal * tax_rate)
    
# Correct analytics practice:
def calculate_tax(subtotal, tax_rate):
    """
    Calculates sales tax for a given subtotal.

    Args:
        subtotal (float): Pre-tax amount
        tax_rate (float): Tax rate as a decimal
    
    Returns:
        float: Tax amount rounded to 2 decimal places
    """
    tax = subtotal * tax_rate
    return round(tax, 2)    # ← This rounds to 2 decimal places


# Example usage:
subtotal = 100.00
tax_rate = 0.08
result = calculate_tax(subtotal, tax_rate)
print(result)  # Output: 8.0
print(type(result)) # <class 'float'> There is no difference between 8.00, 8.0, and 8 as a float

# Additional examples:
subtotal = 100.00
tax_rate = 0.08
tax = calculate_tax(subtotal, tax_rate)  # tax is 8.0 (float)
total = subtotal + tax               # Works perfectly: 108.0
print(total)                         # Output: 108.0

# Now you can reuse it in math
total = subtotal + tax                # → 108.0

# And when you want to DISPLAY it nicely:
print(f"Tax: ${tax:.2f}")             # → Tax: $8.00
print(f"Total: ${total:.2f}")         # → Total: $108.00

# ------------------------------------------------------
# Realistic Analytics Usage Example
# ------------------------------------------------------
def calculate_tax_float(subtotal, tax_rate):
    """Simple version using regular floats — easy to understand."""
    tax = subtotal * tax_rate
    return round(tax, 2)          # ← Return a number (float or Decimal)

# Usage:
subtotal = 100.00
tax_rate = 0.08

# Step 1: Compute — keep everything as numbers for reuse
tax = calculate_tax_float(subtotal, tax_rate)   # → 8.0 (float)
total = subtotal + tax                          # → 108.0 (float) — perfect math!

# You can now safely reuse these in further analytics:
average_tax_rate = tax / subtotal               # → 0.08
grand_total_with_discount = total * 0.9         # Apply 10% off, etc.

# Step 2: # And when you want to DISPLAY it nicely:
print(f"Tax: ${tax:.2f}")             # → Tax: $8.00
print(f"Total: ${total:.2f}")         # → Total: $108.00
print(f"Subtotal:  ${subtotal:.2f}")           # → Subtotal:  $100.00
print(f"Tax (8%):   ${tax:.2f}")               # → Tax (8%):   $8.00
print(f"Total:      ${total:.2f}")             # → Total:      $108.00
print("-" * 30)
print(f"Final amount: ${grand_total_with_discount:.2f}")  # → reusable in reports


# ------------------------------------------------------
# Docstrings are embedded documentation
# This is professional discipline, not academic formality
# ------------------------------------------------------
# They are how future you and other analysts understand your work.
# Good docstrings answer:
# - What does this function do?
# - What inputs does it expect?
# - What does it return?

# Realistic Analytics Example:

def usd_to_eur(amount_usd, exchange_rate=0.88):
    """
    Converts a USD amount to EUR using a given exchange rate.

    Args:
        amount_usd (float): Amount in US dollars
        exchange_rate (float, optional): USD to EUR rate. Defaults to 0.88.

    Returns:
        float: Amount converted to euros
    """
    eur_amount = amount_usd * exchange_rate
    return round(eur_amount, 2)

# Usage:
usd_to_eur(150)
usd_to_eur(150, exchange_rate=0.91)


# Key Takeaways

# - `def` creates a named, reusable transformation
# - Parameters define expected inputs
# - Arguments are supplied at call time
# - `return` is mandatory for reusable analytics logic
# - Docstrings are part of professional-quality code
# - Functions should return data, not print it
`````

`````python
# ------------------------------------------------------
# Currency in Analytics, using the built-in decimal module
# ------------------------------------------------------
# it stores exact decimal values and remembers precision

# Version A — Strict Decimal API (preferred for analytics)

from decimal import Decimal, ROUND_HALF_UP

def calculate_tax_dec(subtotal: Decimal, tax_rate: Decimal = Decimal("0.08")):
    """
    Calculate sales tax using Decimal arithmetic only.

    Contract:
      - Caller MUST pass Decimal values.
      - This avoids all float precision issues by design.
    """
    tax = subtotal * tax_rate
    return tax.quantize(Decimal("0.00"), rounding=ROUND_HALF_UP)

# Example usage:
subtotal = Decimal("100.00")
tax = calculate_tax_dec(subtotal)
total = subtotal + tax

print(tax)     # 8.00
print(total)   # 108.00

# -------------------------------------------------------------------
# Version B - Adapter version (when accepting floats is intentional)
# -------------------------------------------------------------------

from decimal import Decimal, ROUND_HALF_UP

def calculate_tax_safe(subtotal, tax_rate=0.08):
    """
    Calculate sales tax safely when inputs may be float or int.

    Strategy:
      - Convert through str() to discard binary float error.
      - Then use Decimal for exact arithmetic.
    """
    subtotal_dec = Decimal(str(subtotal))
    tax_rate_dec = Decimal(str(tax_rate))

    tax = subtotal_dec * tax_rate_dec
    return tax.quantize(Decimal("0.00"), rounding=ROUND_HALF_UP)

# Easy to use!
tax = calculate_tax_safe(100.00)          # Uses default 8%
print(tax)                               # 8.00

# Usage with Decimal
subtotal = Decimal('100.00')
tax_rate = Decimal('0.08')
tax = calculate_tax_safe(subtotal, tax_rate)
total = subtotal + tax

print(tax)         # → 8.00
print(total)       # → 108.00
print(type(tax))   # <class 'decimal.Decimal'>
print(tax == Decimal('8.00'))  # True



# Key teaching point:
# ⚠️ Why convert through str()?
# Decimal(0.1) preserves the binary float error
# Decimal("0.1") creates the exact decimal value
# Decimal(str(x)) recovers the human-intended number
`````

# 🧠 SECTION 3 — Functions
## Chapter 9. Defining Custom Functions (Foundations)

## Why Argument Types Matter in Analytics

In analytics, functions rarely operate on *one fixed input*.

They must handle:
- Different data sources
- Optional configuration values
- Varying levels of detail
- Evolving business rules

Argument types are how we design **flexibility without chaos**.

Poorly designed arguments lead to:
- Confusing function calls
- Silent bugs
- Fragile notebooks

Well-designed arguments lead to:
- Self-documenting code
- Safe reuse
- Clean pipelines

---
## Summary:

- Positional arguments are matched by order
- Keyword arguments improve clarity and safety
- Default arguments support flexibility
- `*args` captures variable positional inputs
- `**kwargs` captures variable named inputs
- Argument design is a **core analytics skill**, not just syntax


`````python
# --------------------------------------------
# 1️⃣ Positional Arguments (The Default)
# Positional arguments are matched by order
# --------------------------------------------
# Analytics Guidance:
# - Use positional arguments for primary data
# - Keep them few; order them logically

def revenue(price, quantity):
    """
    Calculates total revenue.

    Args:
        price (float): Unit price
        quantity (int): Number of units sold

    Returns:
        float: Total revenue
    """
    return price * quantity


# Usage:
total = revenue(19.99, 3)
print(total)

# -----------------------------
# Positional argument example:
# -----------------------------

def unit_price(total_revenue, quantity):
    """
    Calculates unit price.

    Swapping positional arguments will produce an incorrect result
    without raising an error.

    Args:
        total_revenue (float): Total sales revenue
        quantity (int): Units sold

    Returns:
        float: Unit price
    """
    return total_revenue / quantity   # division, non-communicative

# Correct usage:
price = unit_price(100, 4)
print(price)   # 25.0

# ⚠️ Risk: Order matters:
# This will run, but produces the wrong result
bad_price = unit_price(4, 100)
print(bad_price)  # 0.04 ❌ clearly wrong


# -----------------------------
# Positional argument example:
# -----------------------------

def apply_discount(amount, discount_rate):
    """
    Applies a percentage discount to a monetary amount.

    This function demonstrates asymmetric business logic:
    swapping positional arguments will produce an incorrect result
    without raising an error.

    Args:
        amount (float): Original monetary amount.
        discount_rate (float): Discount rate as a decimal (e.g., 0.10 for 10%).

    Returns:
        float: Discounted amount.
    """
    return amount * (1 - discount_rate)

# Correct usage:
apply_discount(100, 0.10)  # 90.0

# ⚠️ Risk: Order matters:
# This will run, but produces the wrong result
apply_discount(0.10, 100)  # -9.9 ❌ obviously wrong


# --------------------------------------------
# 2️⃣ Keyword Arguments (Clarity Over Brevity)
# Keyword arguments explicitly name each input
# --------------------------------------------
# Analytics Guidance:
# - Prefer keyword arguments for configuration
# - Especially useful when revisiting notebooks months later


def apply_discount(amount, discount_rate):
    """
    Applies a percentage discount to a monetary amount.

    Args:
        amount (float): Original amount.
        discount_rate (float): Discount rate as a decimal (e.g., 0.10 for 10%).

    Returns:
        float: Discounted amount.
    """
    return amount * (1 - discount_rate)

# Correct, readable, order-independent:
total_1 = apply_discount(amount=100, discount_rate=0.10)
total_2 = apply_discount(discount_rate=0.10, amount=100)

print(total_1)  # 90.0
print(total_2)  # 90.0

# Why this matters:
# - Order no longer matters
# - Code reads like documentation
# - Safer for complex functions

# -----------------------------
# Keyword argument example:
# -----------------------------

def gross_revenue(price, quantity):
    """
    Calculates gross revenue.

    Args:
        price (float): Unit price
        quantity (int): Number of units sold

    Returns:
        float: Gross revenue
    """
    return price * quantity

# Correct, readable, order-independent:
gross_revenue(price=19.99, quantity=3)
gross_revenue(quantity=3, price=19.99)

# Note: Because gross_revenue uses multiplication, the numeric result is the same
# either way. The point here is that keyword arguments bind by NAME, improving
# clarity and preventing order-related bugs in general.


# --------------------------------------------
# 3️⃣ Default Arguments (Safe Flexibility)
# --------------------------------------------
# Default arguments allow functions to work out of the box, while still supporting overrides


def usd_to_euro(amount_usd, exchange_rate=0.88):
    """
    Converts USD to EUR.

    Args:
        amount_usd (float): Amount in USD
        exchange_rate (float, optional): Conversion rate

    Returns:
        float: Amount in EUR
    """
    return round(amount_usd * exchange_rate, 2)

# Usage:
usd_to_euro(100)                       # Output: 88.0
usd_to_euro(100, exchange_rate=0.91)   # Output: 91.0


# Rules You Must Remember
# - Default arguments must come last ⚠️
# - Defaults should represent the most common case

# ❌ Invalid
# def bad_example(rate=0.88, amount):
    # pass


# --------------------------------------------
# 4️⃣ `*args` — Variable Positional Arguments
# `*args` allows a function to accept any number 
# of positional inputs (without having to define them), 
# collected as a tuple
# --------------------------------------------
# Analytics Guidance:
# - Use `*args` when the number of inputs is unknown
# - Avoid it when inputs have distinct meanings

# When This Is Useful
# - Aggregations
# - Concatenation
# - Flexible calculations


def average(*values):
    """
    Calculates the average of one or more numeric values.

    Raises:
        ValueError: If no values are provided
    """
    if not values:
        raise ValueError("At least one value is required")
    return sum(values) / len(values)


# Usage:
avg_value = average(10, 20, 30)
print(avg_value)

# What’s happening internally:
# - `*values` is a tuple
# - We can loop, sum, or unpack it


# --------------------------------------------
# Example: *args — Variable Positional Arguments
# --------------------------------------------

def concatenating(*args):
    new_string = ''                 # initialize an empty string to build up over time
    for arg in args:                # start a loop: args == ('Hello', 'world!', 'How', 'are', 'you?')
        new_string += (arg + ' ')   # iterations: new_string += (arg + ' ') appends each word plus a space to the accumulator
    return new_string.rstrip()      # .rstrip() removes trailing whitespace

# Usage:
concatenating("Hello,", "world!", "How", "are", "you?")
# Output: 'Hello, world! How are you?'


# 🟢 Alternatively:🟢 
def pref_concatenate(*words):
    return ' '.join(words)  # joins all strings using a single space as the separator
    
# Usage:    
memo = pref_concatenate("Hello,", "world!", "How", "are", "you?")
print(memo)
# Output: 'Hello, world! How are you?'



# --------------------------------------------
# Mixing normal values with *args
# --------------------------------------------
# Data flow:
#   Call-time inputs  → tuple (*names)
#   Function output   → list[str]
#   Presentation loop → individual strings
#
# Purpose:
#   Separate computation (function) from presentation (printing)

def build_greetings(greeting, *names):
    return [f"{greeting}, {name}!" for name in names]

# Usage:
messages = build_greetings("Hello", "Alice", "Bob", "Charlie")
print(messages)  
# Output: ['Hello, Alice!', 'Hello, Bob!', 'Hello, Charlie!']

for msg in messages:     # for presentation only, not computation
    print(msg)

# Output:
# Hello, Alice!
# Hello, Bob!
# Hello, Charlie!


# --------------------------------------------
# 5️⃣ `**kwargs` — Variable Keyword Arguments
# `**kwargs` accepts any number of named arguments, 
# collected as a dictionary
# --------------------------------------------
# This is extremely powerful in analytics pipelines.
# def concat(**kwargs):
#     """
#     Concatenates keyword argument values into a single string.

#     Returns:
#         str
#     """


def concat(**kwargs):
    new_string = ''                      # initialize an empty string accumulator
    for value in kwargs.values():        # kwargs is a dict: {'a': 'Hello', 'b': 'there!', ...}
        new_string += (value + ' ')      # append each VALUE plus a space
    return new_string.rstrip()           # remove trailing whitespace


# Usage:
lines = concat(a="Hello", b="there!", c="What's", d="up?")
print(lines)
# Output: "Hello there! What's up?"


# 🟢 Alternatively: 🟢 
def pref_concat(**words):
    return ' '.join(words.values())  # joins all strings using a single space as the separator
    
# Usage:    
post = pref_concat(a="Hello", b="there!", c="What's", d="up?")
print(post)
# Output: "Hello there! What's up?"


# ----------------------------------
# Example: `**kwargs` — Named Inputs
# ----------------------------------
# `**kwargs` collects keyword arguments into a DICTIONARY.
# Each key is the argument name, and
# Each value is the argument value.

def summarize_metrics(**metrics):
    """
    Summarizes named metrics.

    Args:
        **metrics (float): Named metric values passed as keyword arguments

    Returns:
        dict: Metrics with values rounded to 2 decimal places
    """
    summary = {}  
    # Initialize an empty dictionary to store the cleaned results

    for key, value in metrics.items():
        # metrics.items() returns (key, value) pairs from the kwargs dictionary
        # Example iteration:
        # key = 'revenue', value = 12345.678

        summary[key] = round(value, 2)
        # Round each metric value and store it under the same key
        # This preserves the original metric names while cleaning the data

    return summary
    # Return structured data (not printed output) for reuse in analytics


# Usage:

metrics = summarize_metrics(
    revenue=12345.678,
    cost=9876.543,
    margin=2469.135
)
# Internally, Python builds this dictionary before the function runs:
# metrics = {
#     'revenue': 12345.678,
#     'cost': 9876.543,
#     'margin': 2469.135
# }

print(metrics)
# Output:
# {'revenue': 12345.68, 'cost': 9876.54, 'margin': 2469.14}

# Why This Is Important:
# - Accepts dynamic inputs
# - Works naturally with dictionaries
# - Ideal for reporting outputs



# ----------------------------------
# Example: Unpacking a dictionary with `**`
# ----------------------------------
# 📦➡️ The `**` operator UNPACKS a dictionary into keyword arguments.
# 🎯 Dictionary KEYS must match the function's parameter names exactly.


def exponentiator(constant, base, exponent):
    """
    Calculates an exponential expression with a scaling constant.

    Args:
        constant (int or float): Multiplier applied to the result
        base (int or float): Base number
        exponent (int): Power to raise the base to

    Returns:
        int or float: Result of constant * (base ** exponent)
    """
    return constant * (base ** exponent)


# Usage:
parameter_dict = {
    "constant": 2,
    "base": 3,
    "exponent": 2
}

# The `**` operator tells Python to:
# 1️⃣ Take each key-value pair in the dictionary
# 2️⃣ Treat each key as a parameter name
# 3️⃣ Pass each value as the corresponding argument
#
# 🔁 This call is equivalent to writing:
# exponentiator(constant=2, base=3, exponent=2)

result = exponentiator(**parameter_dict)
print(result)

# Output:
# 18


#----------------------------------------------
# 🚨 Pro Insight: Argument Ordering Discipline
#-----------------------------------------------
# The correct argument order in professional code is:

# def function_name(
#     required_positional,
#     optional_positional=default,
#     *args,
#     keyword_only=default,
#     **kwargs
# ):
#     pass

# You won’t always use all of these, but you must recognize the pattern.

# Analytics design pattern example
# Demonstrates correct signature ordering discipline:
def tax_calculator(
    subtotal,
    tax_rate=0.06,
    *discounts,
    **metadata
):
    """
    Calculates final total with optional discounts and metadata.

    Args:
        subtotal (float): Base transaction amount
        tax_rate (float, optional): Sales tax rate
        *discounts (float): Any discount amounts
        **metadata: Additional transaction info

    Returns:
        dict: Transaction summary
    """
    # Execution order:
    discounted_total = subtotal - sum(discounts)
    tax = discounted_total * tax_rate
    total = discounted_total + tax
    
    # Return structure order (presentation, not computation):
    return {
    "subtotal": subtotal,
    "discounts": discounts,
    "discounted_total": discounted_total,
    "tax": round(tax, 2),
    "total": round(total, 2),
    "metadata": metadata
}


# Usage:
tax_calculator(
    200,
    0.07,
    10,
    5,
    customer_id="C0042",
    region="Stowe"
)

# This mirrors real ETL-style analytics logic
# metadata allows arbitrary contextual information
# without changing the function signature


# Key Takeaways:
# - Positional arguments are matched by order
# - Keyword arguments improve clarity and safety
# - Default arguments support flexibility
# - `*args` captures variable positional inputs
# - `**kwargs` captures variable named inputs
# - Argument design is a core analytics skill, not just syntax
`````

# 🧠 SECTION 4 — Functions
## Chapter 9. Return Strategies (Designing Outputs for Reuse)

## Why Return Strategy Matters More Than You Think

In analytics, **what you return** from a function determines:

- How easily results can be reused
- Whether outputs compose cleanly into pipelines
- How readable downstream code becomes
- Whether your logic scales beyond a single notebook cell

A function’s return value is its **public interface**.

Bad return design forces callers to:
- Recompute values
- Rely on globals
- Parse printed output (❌)
- Rewrite logic

Good return design makes functions **lego bricks**.

---

## Rule Zero: Functions Should Return Data, Not Behavior

A function should:
- Return values
- Avoid printing (except debugging)
- Avoid mutating external state

This keeps functions:
- Testable
- Reusable
- Predictable

---

## Design Heuristic: Choosing the Right Return Type 🧭

Ask yourself:

- One value? → scalar
- Multiple values, fixed order, internal use? → tuple
- Iterable output? → list
- Named, reusable analytics output? → dictionary

This decision is **architectural**, not cosmetic.


`````python
# 🧠 Lesson: Understanding Return Values

# A function's main job is to **compute** 
# something and **give it back** to the caller using `return`. 
# This lesson shows what return values are, how to use them, and why they matter.

# ============================================
# 1. What is a return value?
# ============================================

# A return value is the result a function sends back to the caller.
# Think of a function like a machine: you put inputs in, it gives an output back.

def square(n):
    """Return the square of n."""
    result = n ** 2
    return result      # <-- this is the return value


# Calling the function and capturing the return value:
answer = square(5)     # answer now holds 25
print(answer)          # 25

# You can also use the return value directly:
print(square(7))       # 49


# ============================================
# 2. ⚠️ print() vs return — they are NOT the same
# ============================================

# ❌ BAD: printing inside the function
def square_bad(n):
    print(n ** 2)      # displays on screen, but returns nothing

output = square_bad(5)  # prints 25
print(output)           # None  ← can't reuse the result!


# ✅ GOOD: returning from the function
def square_good(n):
    return n ** 2       # sends value back to caller

output = square_good(5)
print(output)           # 25 ← now you can use it!

# Why returning is better:
total = square_good(3) + square_good(4)   # 9 + 16 = 25
print(total)                               # 25


# ============================================
# 3. Using return values in expressions
# ============================================

def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

# Chain function calls together:
result = multiply(add(2, 3), add(4, 1))   # multiply(5, 5) → 25
print(result)                              # 25

# Store intermediate results for clarity:
sum1 = add(2, 3)       # 5
sum2 = add(4, 1)       # 5
product = multiply(sum1, sum2)
print(product)         # 25


# ============================================
# 4. Returning multiple values (tuple)
# ============================================

def div_and_remainder(num, divisor):
    """
    Returns BOTH the quotient and remainder when dividing num by divisor.
    Returns a tuple: (quotient, remainder)
    """
    quotient  = num // divisor     # whole number part
    remainder = num % divisor      # what's left over
    return quotient, remainder     # comma = tuple packing (parentheses optional)

# Calling the function and capturing the result
result = div_and_remainder(17, 5)
print("Raw returned value:", result)          # (3, 2)  ← it's a tuple!
print("Type of result:", type(result))        # <class 'tuple'>

# Better: unpack directly into separate variables (very common!)
q, r = div_and_remainder(17, 5)
print("Quotient:", q)                         # 3
print("Remainder:", r)                        # 2


# ============================================
# 5. Early return (exit function immediately)
# ============================================

def first_positive(numbers):
    """Return the first positive number, or None if there are none."""
    for num in numbers:
        if num > 0:
            return num   # exits the function immediately
    return None          # only reached if no positive found


print(first_positive([-3, -1, 0, 4, 7]))   # 4  (first positive)
print(first_positive([-5, -2, 0]))         # None (no positives)


# ============================================
# 6. Practical example: reusable calculations
# ============================================

def tip_amount(bill, tip_percent=15):
    """Calculate tip based on bill total and tip percentage."""
    return bill * (tip_percent / 100)


def total_with_tip(bill, tip_percent=15):
    """Calculate total bill including tip."""
    tip = tip_amount(bill, tip_percent)   # reuse the other function!
    return bill + tip


# Example calls:
my_bill = 45.00

print(f"Tip (15%): ${tip_amount(my_bill):.2f}")              # $6.75
print(f"Tip (20%): ${tip_amount(my_bill, 20):.2f}")          # $9.00
print(f"Total with 15% tip: ${total_with_tip(my_bill):.2f}") # $51.75
print(f"Total with 20% tip: ${total_with_tip(my_bill, 20):.2f}") # $54.00


# **Key takeaways**

# - `return` sends a value **back to the caller**; `print()` just displays something on screen.  
# - If you don't use `return`, the function returns `None` by default, which is usually not useful.  
# - Return values can be stored in variables, used in expressions, or passed to other functions.  
# - Functions can return **multiple values** by packing them into a tuple; the caller can unpack them.  
# - A `return` inside a loop causes an **early exit**, which is useful for "find first" or "stop when condition met" patterns.  
# - Keeping computation in functions (with `return`) and printing outside makes your code **reusable** and easier to test.
`````

`````python
# ✅ Return Strategies: Multiple Outputs Without the Mess

# - **clean computation**
# - **explicit return shapes**
# - **caller-controlled presentation**


# 1️⃣ Multiple return values = one tuple (mechanics)

# Python can “return multiple values” because it actually returns **one tuple**.

def sentence_and_last_word(*words):
    """
    Builds a sentence and returns the sentence plus the last word.

    Returns:
        tuple[str, str]: (sentence, last_word)
    """
    if not words:
        raise ValueError("At least one word is required")

    sentence = " ".join(words)     # clean, scalable string building
    last_word = words[-1]          # last item from the tuple of inputs

    return sentence, last_word     # tuple packing

# Usage (tuple unpacking)

sentence, last_word = sentence_and_last_word("Hello", "world!", "How", "are", "you?")
sentence
last_word

## 2️⃣ Better for analytics: return a dictionary (explicit meaning)

# Tuples are fine, but in analytics **meaning should be named**, not positional.

def sentence_summary(*words):
    """
    Returns a structured summary of a sentence-building step.

    Returns:
        dict: {sentence, last_word, word_count}
    """
    if not words:
        raise ValueError("At least one word is required")

    sentence = " ".join(words)     # clean, scalable string building

    return {
        "sentence": sentence,
        "last_word": words[-1],
        "word_count": len(words)
    }

# Usage (robust + readable)

summary = sentence_summary("Hello", "world!", "How", "are", "you?")
summary["sentence"]
summary["last_word"]
summary["word_count"]

# 3️⃣ Presentation happens outside the function (your Rule #1)

# Presentation only (not computation)
print(summary["sentence"])
print(summary["last_word"])
print(summary["word_count"])

# Key Takeaways

# - “Multiple returns” in Python = returning **one tuple**
# - Use `" ".join(...)` instead of manual `+=` in loops
# - Tuple unpacking is fine, but tuples are **order-dependent**
# - For analytics outputs, **dictionaries are preferred** (explicit meaning)
# - Functions return data; callers handle presentation
`````

`````python
# 🧠 SECTION — Return Values (Structured Returns & Unpacking) ⭐

# In Python, a function can return **any object**, including:
# - tuples
# - lists
# - dictionaries
# - custom objects

# When you write:
# return a, b
# Python is actually doing:
# return (a, b)   # tuple packing

# This enables a powerful pattern called **multiple return values + unpacking**.

# Think in two phases:
# inside function:   values are PACKED into one object
# outside function: values are UNPACKED into separate variables


# 1️⃣ Returning Multiple Values (Tuple Packing)

def rating_cleaner(rating):
    """
    Extracts the numeric portion from a text rating.

    Args:
        rating (str): Rating like "5 stars"

    Returns:
        tuple: (original_rating, numeric_rating)
    """
    numeric_rating = int(rating[0])      # extract first character and convert to int
    return numeric_rating, rating        # ← tuple packing return 📦 (e.g., return a, b, c)

    
# Usage: capture the returned tuple

website_ratings = ["5 stars", "4 stars", "3 stars"]

result = rating_cleaner("5 stars")

# Internally:
result == (5, "5 stars")

print(result)
print(type(result))     # <class 'tuple'>
print()

# 🧠 Notes:
# - The comma creates a **tuple**
# - Parentheses are optional
# - This is how Python returns “multiple values”


# 2️⃣ Unpacking Returned Tuples (The Real Power)

# Instead of working with the tuple directly, you usually **unpack** it.

rating_text, numeric_rating = rating_cleaner("5 stars")  # tuple unpacking 📤

# Now:
rating_text     == "5 stars"
numeric_rating  == 5

print(rating_text)
print(numeric_rating)


# 🧠 Key ideas:
# - The number of variables on the left must match the tuple length
# - Unpacking is **positional** (order matters) 📤
# - This is extremely common in real code


# 3️⃣ Returning Lists Instead of Tuples

# You can explicitly return a list if mutability or list semantics are desired.

def rating_cleaner_list(rating):
    """
    Returns rating info as a list.

    Returns:
        list: [original_rating, numeric_rating]
    """
    numeric_rating = int(rating[0])
    return [rating, numeric_rating]   #  ← explicit list 📋


# Usage:

data = rating_cleaner_list("4 stars")

print(data)           # ["4 stars", 4]
print(type(data))    # <class 'list'>

# 🧠 Notes:
# - Lists are mutable
# - Meaning is still positional
# - Less common than tuples for fixed-size returns


# 4️⃣ Returning Dictionaries (Preferred for Analytics)

# For analytics, **named fields beat positional order**.

def rating_cleaner_dict(rating):
    """
    Returns rating info as a structured dictionary.

    Returns:
        dict: { "rating": str, "numeric": int }
    """
    numeric_rating = int(rating[0])

    return {
        "rating": rating,              #  ← explicit dictionary
        "numeric": numeric_rating
    }                                  # named fields beat positional order

# Usage:

info = rating_cleaner_dict("3 stars")

print(info)
print(info["rating"])     #  ← safe, explicit access
print(info["numeric"])


# 🧠 Why this is better for analytics:
# - No order dependence
# - Self-documenting
# - Robust to future changes
# - Natural for DataFrames, JSON, APIs


# 5️⃣ Unpacking With Dictionaries vs Tuples
# Tuple unpacking (positional) ⚠️
# Continued in next cell block
`````

`````python
# 5️⃣ Unpacking With Dictionaries vs Tuples 📌
# Tuple unpacking (positional) ⚠️

# This mini‑lesson shows **how unpacking works with tuples**, 
# why **position (order) matters**, and how dictionaries avoid 
# that problem by using **names** instead of positions.

# --------------------------------------------
# The function that returns multiple values
# --------------------------------------------

def rating_cleaner_2(rating):
    """
    Extracts the numeric portion from a text rating.

    Args:
        rating (str): Rating like "5 stars"

    Returns:
        tuple: (numeric_rating, original_rating)
    """
    numeric_rating = int(rating[0])      # take first character, e.g. "5" → 5
    return numeric_rating, rating     # tuple packing return 📦

# --------------------------------------------
# ⚠️ Tuple unpacking — ORDER matters (positional)
# --------------------------------------------
# The function returns: (numeric_rating, original_rating)
# So we MUST unpack in that same order:

website_ratings = ["5 stars", "4 stars", "3 stars"]

numeric_rating, rating_text = rating_cleaner_2("5 stars")
# numeric_rating → 5
# rating_text    → "5 stars"

print("numeric_rating:", numeric_rating)
print("rating_text:   ", rating_text)

# ❌ If we reverse the order, we silently get nonsense:
rating_text_bad, numeric_rating_bad = rating_cleaner_2("4 stars")

print("\nBAD unpacking example:")
print("rating_text_bad:   ", rating_text_bad)    # actually 4 (int), not text
print("numeric_rating_bad:", numeric_rating_bad) # actually "4 stars"

# --------------------------------------------
# Dictionary return — names instead of positions ⭐
# --------------------------------------------

def rating_cleaner_dict(rating):
    """
    Returns rating info as a structured dictionary.

    Returns:
        dict: { "rating": str, "numeric": int }
    """
    numeric_rating = int(rating[0])
    return {
        "rating": rating,          # name: "rating"
        "numeric": numeric_rating  # name: "numeric"
    }  # dict return 🗂


info = rating_cleaner_dict("5 stars")

# Access by key (name), not by position:
rating_text_dict = info["rating"]
numeric_rating_dict = info["numeric"]

print("\nDictionary version:")
print("numeric_rating_dict:", numeric_rating_dict)
print("rating_text_dict:   ", rating_text_dict)

# If we swap these two assignments, the values are still correct,
# - Order-independent
# - Safer for pipelines
# - Preferred in analytics code


# 6️⃣ General Rule: What You Can Return

# A `return` statement can return:

# return 5                        # scalar
# return a, b                     # tuple
# return [a, b]                   # list
# return {"a": a, "b": b}         # dictionary
# return some_object              # any Python object


# The caller receives **exactly one object**.

# If it looks like “multiple values”, it is actually:
# - one tuple
# - being unpacked by the caller


# Key Takeaways

# - `return a, b` returns **one tuple** containing two values  
# - Tuple unpacking lets you bind those values to separate variables  
# - Lists and dictionaries can also be returned  
# - Tuples are positional and order-dependent  
# - Dictionaries are preferred for analytics outputs  
# - Returning structured data makes pipelines safer and clearer  

# > **Return values are not limited to scalars.  
# They define the data interface between functions.**
`````

# 🧠 SECTION 5 — Functions
## Chapter 9. Variable Scope & Lifetime (Data Safety by Design)

## Why Variable Scope Matters in Analytics

Variable scope determines **where a variable exists** and **who can see it**.

In analytics work, poor understanding of scope leads to:
- Silent overwrites
- Hard-to-debug errors
- Fragile notebooks
- Functions that behave differently depending on execution order

Good scope discipline leads to:
- Predictable behavior
- Reusable functions
- Safe pipelines
- Code that scales beyond a single notebook

Scope is not a Python quirk.  
It is a **data safety concept**.

---

## The Two Core Scopes You Must Master

Python introduces many scopes, but for now we focus on the two that matter most:

1. **Local scope** — variables created *inside* a function  
2. **Global scope** — variables created *outside* all functions  


`````python
# 🧠 Variable Scope: Practical Examples

# ============================================
# 1️⃣ Local scope: variables live inside functions
# ============================================

def calculate_total(price, quantity):
    total = price * quantity      # total is LOCAL to this function
    return total

print("Order total:", calculate_total(10, 3))   # 30

# This will raise a NameError, because total is not defined here:
# print(total)   # ❌ Uncomment to see the error

# ============================================
# 2️⃣ Global scope: works, but can be fragile
# ============================================

tax_rate = 0.06   # global variable (shared everywhere)

def calculate_tax(subtotal):
    """
    ❗ Fragile version: depends on a global variable `tax_rate`.

    The result can change even if this function's code never changes,
    because it reads tax_rate from the outer/global scope.
    """
    # Uses whatever tax_rate happens to be in the global scope
    return subtotal * tax_rate                 # ⚠️ tax_rate is a global variable


print("\nTax with 6%:", calculate_tax(100))    # 6.0

# Later in the notebook, someone (maybe in a different cell) changes the global:
tax_rate = 0.09    # same name, new value

print("Tax with 9%:", calculate_tax(100))      # 9.0

# 🔍 Same function, different result, just because the GLOBAL tax_rate changed.
# A person reading only calculate_tax() might NEVER see where tax_rate is changed.
# This is how subtle analytics bugs are born!


# ============================================
# ✅ Safer pattern: pass everything in as parameters
# ============================================

def calculate_tax_safe(subtotal, tax_rate):     # ✅ tax_rate is a local variable
    """
    Safe version: tax_rate is an EXPLICIT input.
    Same inputs → same output, no hidden dependency on globals.
    """
    return subtotal * tax_rate


print("\nSafe version (explicit tax_rate):")
print("Tax with 6%:", calculate_tax_safe(100, 0.06))   # 6.0
print("Tax with 9%:", calculate_tax_safe(100, 0.09))   # 9.0

# Anyone reading this call site can SEE which rate is being used.
# No surprises from hidden global variables.


# ============================================
# 3️⃣ global keyword: powerful but dangerous
# ============================================

counter = 0   # global

def increment():
    global counter      # tells Python we mean the global counter ⚠️
    counter += 1        # modifies global state

increment()
increment()
print("\nCounter after two increments:", counter)  # 2

# This kind of hidden mutation makes behavior depend on call order,
# which is risky in analytics notebooks.

# ===========================================================
# Counting INSIDE a function (classic loop counter, SAFE) ✅
# ===========================================================

def count_positives(values):
    """
    Count how many numbers in the list are > 0.
    The counter is local to the function.
    """
    count = 0                  # local counter
    for value in values:
        if value > 0:
            count += 1         # += used safely inside the loop
    return count               # return final count

nums = [-3, 0, 5, 2, -1, 7]
print("Positive count:", count_positives(nums))  # 3
# No globals, no hidden state. Same input → same output.


# ===========================================================
# Increment as a pure helper (SAFE, like add_to_total) ✅
# ===========================================================

def increment_count(current_count, step=1):
    """
    Pure-style increment:
    takes the current count and a step, returns a NEW count.
    """
    return current_count + step   # conceptually the same as current_count += step

count = 0
count = increment_count(count)        # 1
count = increment_count(count)        # 2
count = increment_count(count, 5)     # 7
print("Incremented count:", count)

# ===========================================================
# Example: Count how many ratings are "5 stars" (or higher) in a dataset
# ===========================================================

def count_top_ratings(ratings, top_value="5"):
    """
    Count how many ratings start with a given top_value, e.g. "5" in "5 stars".
    """
    count = 0
    for text in ratings:
        if text.startswith(top_value):
            count += 1          # local counter, safe
    return count


website_ratings = ["5 stars", "4 stars", "5 stars", "3 stars", "5 stars"]

num_fives = count_top_ratings(website_ratings, top_value="5")
print("Number of '5 star' ratings:", num_fives)    # 3

# ===========================================================
# Example where increment_count is realistic: stepping through pages / batches
# ===========================================================

def increment_count(current_index, step=1):
    """
    Pure-style increment:
    takes the current index and a step, returns a NEW index.
    """
    return current_index + step


# Imagine paging through batches of 100 rows in a dataset:
page_index = 0          # start at batch 0
page_index = increment_count(page_index)        # move to batch 1
page_index = increment_count(page_index)        # move to batch 2

print("Current batch index:", page_index)       # 2


# ============================================
# 4️⃣ Name shadowing: local vs global
# ============================================

value = 100   # global

def double_value():
    value = 50          # local variable with the SAME NAME
    return value * 2    # uses local value (50), not global (100)

result = double_value()
print("\nResult from double_value():", result)  # 100 (50 * 2)
print("Global value still:", value)             # 100

# Inside the function, 'value' refers to the local one.
# The global 'value' is unchanged.


# ============================================
# 5️⃣ Safe analytics-style function (no leaks)
# ============================================

def compute_metrics(values):
    """
    Computes summary metrics from a list of values.

    Args:
        values (list of float): Input values

    Returns:
        dict: Summary statistics
    """
    count = len(values)
    total = sum(values)
    average = total / count if count > 0 else 0.0

    return {
        "count": count,
        "total": total,
        "average": round(average, 2),
    }

data = [1, 2, 3, 4]

metrics = compute_metrics(data)
print("\nMetrics:", metrics)

# No globals modified, no hidden state:
# - If you call compute_metrics(data) again, you get the same result.
# - Safe to reuse in other notebooks and pipelines.

# **Key patterns illustrated**

# - Local variables (like `total` inside `calculate_total`) do **not** exist outside their function.  
# - Globals (like `tax_rate`, `counter`, `value`) can be read from functions, but relying on them makes behavior depend on external state.  
# - The `global` keyword lets a function modify a global, but this creates hidden side effects and order‑dependent bugs.  
# - The preferred analytics pattern is: **functions take inputs and return outputs**, with no mutation of external variables.  
# - Name shadowing shows why reusing global names inside functions is confusing—descriptive, local names are safer.  
# - A function like `compute_metrics` is “analytics‑grade”: predictable, testable, and ready to drop into a pipeline.
`````

## Creating Modules & Reuse (From Notebook to Asset)

## Why Modules Matter in Analytics

In real analytics work, notebooks are often:
- Exploratory
- Iterative
- Messy by design

Modules, on the other hand, are:
- Stable
- Reusable
- Shareable
- Testable

A **module** is simply a `.py` file that contains functions (and sometimes constants).
The moment you move logic into a module, you are declaring:

> “This logic is trusted and reusable.”

This is a major professional shift.

---

## Mental Model: Notebook vs Module

Think of it this way:

- **Notebook** → analysis, experimentation, storytelling
- **Module** → utilities, helpers, business logic

Not everything belongs in a module.
But anything you expect to reuse **does**.

---

## Lifecycle of a Module

* Create / overwrite module with %%writefile
* Import module
* Reload after edits with importlib.reload
* Call functions
* Observe outputs

## Keep

* Each step in its own cell
* Each step documented in Markdown
* No mixed concerns

# 🧠 SECTION 8 — Functions
## Chapter 9. `map()` & Function Application (Scaling Clean Logic)

## Why This Section Exists

In analytics, you rarely apply logic to a single value.

You apply the **same transformation** to:
- Every row in a dataset
- Every transaction in a list
- Every value in a column

The question is not *whether* to scale logic —  
it is **how to do so clearly and safely**.

`map()` is one such mechanism.

---

## What `map()` Actually Does

The `map()` function applies a function to **each element** in an iterable.

Syntax:
```python
map(function, iterable)
```

Important:
- `map()` does **not** immediately return a list
- It returns a **map object** (lazy evaluation)

`````python
# ------------------------------------------------------------------
# Lecture: Understanding map() – Applying a Function to Every Item
# ------------------------------------------------------------------
# **Core idea:**  
# `map(function, iterable)` runs the same function on **every item**
# for any iterable and returns a map object containing the results.

# This is super useful for transforming data without writing explicit loops

# Basic Pattern:

# Define a function that transforms ONE item
def currency_formatter(number):
    """
    Convert a number to a dollar-string format.
    Example: 19.99 → '$19.99'
    """
    return '$' + str(number)

# A list of prices (iterable)
price_list = [5.99, 19.99, 24.99, 0, 74.99, 99.99]

# map() applies the function to EVERY element
# It returns a map object (lazy / iterator – doesn't compute until needed)
formatted_map = map(currency_formatter, price_list)

# See the map object (not very useful by itself)
print(formatted_map)          # <map object at 0x...>

# Convert to a list to see the results
formatted_prices = list(formatted_map)
print(formatted_prices)
# Output: ['$5.99', '$19.99', '$24.99', '$0', '$74.99', '$99.99']


# Alternative Options:

# 1. Using a lambda (anonymous function) – very typical
formatted_prices_lambda = list(map(lambda n: f"${n:.2f}", price_list))
print(formatted_prices_lambda)
# ['$5.99', '$19.99', '$24.99', '$0.00', '$74.99', '$99.99']

# 2. Combining with other functions (real pipeline example)
discounted = list(map(lambda p: round(p * 0.9, 2), price_list))  # 10% off
print(discounted)
# [5.39, 17.99, 22.49, 0.0, 67.49, 89.99]

# 3. Preferred method: list comprehension (cleaner than map + lambda) ⭐
formatted = [f"${p:.2f}" for p in price_list]
print(formatted)

# ---------------------------------------------------
# Comparison: `map()` version vs. list comprehension
# ---------------------------------------------------

# map() version
def currency_format(p):
    return f"${p:.2f}"

map_version = list(map(currency_format, price_list))

# list comprehension version (same result)
comp_version = [f"${p:.2f}" for p in price_list]

print("map() result:   ", map_version)
print("comp result:    ", comp_version)
# Both give: ['$5.99', '$19.99', '$24.99', '$0.00', '$74.99', '$99.99']
`````

`````python
# -----------------------------------------------------
# Lambda Functions
# -----------------------------------------------------
# Lambda functions are **anonymous** (unnamed) functions, typically used for short, throwaway logic.  
# You’ll most often see them when manipulating lists, Series, or columns in pandas DataFrames.

# Mental model:

# General lambda pattern
# lambda x: expression

# is equivalent to:

# def f(x):
#     return expression

# Key idea: a lambda is just a tiny function defined *in place*, usually without giving it a name
# lambdas return the value of the expression automatically
# Model: expression → returned value

# ------------------------------------------------
# Basic lambda usage
# ------------------------------------------------

# Single-argument lambda, called immediately
(lambda x: x**2)(3)          # 9

# ❌ NOT allowed: multiple expressions inside a lambda
# (lambda x: x**2, x**3)(3) # Invalid syntax: a lambda must contain exactly ONE expression


# Multiple arguments + conditional logic
(lambda x, y: x * y if y > 5 else x / y)(6, 5)   # 1.2

# Notes:
# - A lambda can take multiple parameters.
# - It can use conditional expressions (`a if condition else b`).
# - It must always evaluate to a **single** expression (no assignments, loops, or multiple return values inside)

# ------------------------------------------------
# Lambda with map(): currency conversion example
# ------------------------------------------------

price_list = [5.99, 19.99, 24.99, 0, 74.99, 99.99]
ex_rate = 0.88  # example exchange rate

# Apply the exchange rate and round to 2 decimals
converted = map(lambda x: round(x * ex_rate, 2), price_list)

converted_prices = list(converted)
print(converted_prices)

# Here:
# - `map()` applies the lambda to every element in `price_list`.
# - The lambda multiplies each price by `ex_rate` and rounds it.
# - `map()` returns an iterator, so we wrap it in `list()` to see the results

# ------------------------------------------------
# Exercise: Lambda + map() + conditional logic
# Apply a 10% discount to customers who spent > $500
# ------------------------------------------------

subtotaling = [15.98, 899.97, 799.97, 117.97, 5.99, 599.99]

# Standalone lambda line is just to show the pattern;
# by itself it does nothing until we call it

lambda s: round(s * 0.9, 2) if s > 500 else s

# Use map() to apply the lambda to every value in the list
discounted_subtotals = list(map(lambda s: round(s * 0.9, 2) if s > 500 else s,subtotaling))
print(discounted_subtotals)

# Customers who qualify for the discount (based on original spend)
for subtotal in subtotaling:
    if subtotal > 500:
        print(subtotal)

# Optional: Track BOTH Orignial and Discounted Together:
subtotaling = [15.98, 899.97, 799.97, 117.97, 5.99, 599.99]

discounted_subtotals = list(
    map(lambda s: round(s * 0.9, 2) if s > 500 else s, subtotaling)
)

# Zip original and discounted so you can see both
for original, discounted in zip(subtotaling, discounted_subtotals):
    if original > 500:   # qualification is based on original spend
        print(f"Original: {original}, Discounted: {discounted}")

# -----------------------------------------------------
# Equivalent transformation using a list comprehension:
# -----------------------------------------------------
converted_prices_lc = [round(x * ex_rate, 2) for x in price_list]

# map() + lambda and list comprehensions are often interchangeable;
# choice is mostly about readability and style.
        
# Key takeaways
# - Lambda functions are short, **anonymous** functions defined with `lambda`.
# - They are perfect for one-off logic passed into tools like `map()`, `filter()`, or pandas `.apply()`.
# - A lambda must be a **single expression** (no statements like `for`, `while`, or assignments).
# - For anything complex or reusable, prefer a regular `def` function with a clear name.
`````

# 🧠 SECTION 9 — Functions
## Chapter 9. Comprehensions (Readable, Scalable Function Consumption)

## Why Comprehensions Matter More Than `map()`

Comprehensions are not just syntactic sugar.

They are:
- Explicit
- Readable
- Flexible
- Pipeline-friendly

In analytics, comprehension syntax often mirrors how we *think*:
> “For each value, apply this transformation, optionally filter, and collect the result.”

This clarity is why comprehensions are usually preferred over `map()`.

---

## The Core Comprehension Pattern

### List Comprehension Syntax

```python
[expression for item in iterable]
```

Think of it as:
- **expression** → what to produce
- **item** → each element
- **iterable** → the data source


`````python
# -------------------------------------------
# Comprehensions
# Generate sequences from other sequences
# -------------------------------------------
# Syntax: new_list = [expression for item in iterable (if condition)]


# Example: Creating a list of Euro prices from USD prices
xchange_rate = 0.88 
us_list = [5.99, 9.99, 24.99, 99.99, 0]

uro_list = [round(price * xchange_rate, 2) for price in us_list]
uro_list

# Contrast with using a for loop to create a new list
eur_list = []

for price in us_list:
    eur_list.append(round(price * xchange_rate, 2))
eur_list

# Comprehension with conditional logic and adding if filtering 
uro_list = [round(price * xchange_rate, 2) for price in us_list if price <= 9.99]
uro_list
`````

`````python
import numpy as np
`````

`````python
star_ratings = ["5 star", "1 star", "3 star", "4 star", "3 star", "5 star"]
`````

`````python
num_rating = [int(rating[0]) for rating in star_ratings]  # for every rating in ratings, store rating index in list
num_rating
`````

`````python
np.mean(num_rating)
`````

`````python
np.mean(lower_ratings)
`````

`````python
# ❌ Anti-pattern: recomputes extraction twice per element (avoid)
lower_ratings = [int(rating[0]) for rating in star_ratings if int(rating[0]) < 4]


# ✅ Good pattern: compute once, then filter (generator inside comprehension)
lower_ratings = [
    r
    for r in (int(rating[0]) for rating in star_ratings)
    if r < 4
]


# ✅ Best for teaching and debugging: explicit loop with intermediate variable
lower_ratings = []
for rating in star_ratings:
    r = int(rating[0])      # extract once
    if r < 4:              # then filter
        lower_ratings.append(r)

# Teaching note:
# - All three produce the same result.
# - Prefer the explicit loop for teaching and debugging.
# - Prefer comprehensions when the logic is simple and clear.
`````

## Chapter 9. Analytics Workflow Patterns (Composing Real Pipelines)

## Why This Section Exists

- Defining functions

## The Analytics Pipeline Mental Model

Most analytics workflows follow this shape:

1. **Ingest** raw inputs  
2. **Clean** and normalize data  
3. **Validate** assumptions  
4. **Transform** values  
5. **Aggregate / summarize**  
6. **Expose structured outputs**

Each step should be:
- Small
- Testable
- Reusable
- Explicit

Functions are the units that make this possible.

`````python
# -------------------------------------------
# Dictionary Comprehensions
# Creating a dictionary from parallel iterables
# -------------------------------------------

items = ["skis", "snowboard", "goggles", "boots"]  # keys
status = [[5, 249.99], [0, 219.99], [0, 99.99], [12, 79.99]]  # [quantity, price]


# ------------------------------------------------
# Basic dictionary comprehension (no filtering)
# ------------------------------------------------

# Compute inventory cost = quantity * price for each item
inventory_costs = {
    k: round(v[0] * v[1], 2)
    for k, v in zip(items, status)
}

inventory_costs


# ------------------------------------------------
# Dictionary comprehension with filtering
# Keep only items with quantity > 0
# ------------------------------------------------

# ✅ Good pattern: compute once, then filter
inv_cost_filtered = {
    k: round(v[0] * v[1], 2)
    for k, v in zip(items, status)
    if v[0] > 0
}

inv_cost_filtered


# ------------------------------------------------
# Best for teaching and debugging: explicit loop
# ------------------------------------------------

inv_cost_loop = {}

for k, v in zip(items, status):
    qty = v[0]
    price = v[1]
    cost = round(qty * price, 2)   # compute once

    if qty > 0:                   # then filter
        inv_cost_loop[k] = cost

inv_cost_loop


# Teaching note:
# - All three produce the same dictionary.
# - The comprehension is concise and idiomatic.
# - The explicit loop is best for teaching and debugging.
# - Avoid overwriting loop variables (like reusing 'v' for a float).
`````

`````python
from tax_calculations import tax_breakdown


# -------------------------------------------
# Dictionary comprehensions with functions
# -------------------------------------------

customer_ids = ['C00004', 'C00007', 'C00015', 'C00016', 'C00020', 'C00010']
subtotals = [15.98, 899.97, 799.97, 117.96, 5.99, 599.99]


# Using zip(): parallel iteration over two lists
cust_dict_zip = {
    customer_id: tax_breakdown(subtotal, tax_rate=0.08)
    for customer_id, subtotal in zip(customer_ids, subtotals)
}

cust_dict_zip


# Using enumerate(): index-based pairing
cust_dict_enum = {
    customer_ids[i]: tax_breakdown(subtotal, tax_rate=0.08)
    for i, subtotal in enumerate(subtotals)
}

cust_dict_enum


# -------------------------------------------
# Encapsulating the logic in a reusable function
# -------------------------------------------

def transaction_dict_creator(customer_ids, subtotals, tax_rate):
    """
    Create a dictionary mapping customer IDs to tax breakdowns.

    Args:
        customer_ids (list[str]): List of customer identifiers.
        subtotals (list[float]): List of transaction subtotals.
        tax_rate (float): Sales tax rate to apply.

    Returns:
        dict: {customer_id: [subtotal, tax, total]}
    """
    if len(customer_ids) != len(subtotals):
        raise ValueError("customer_ids and subtotals must have the same length")

    cust_dict = {
        customer_id: tax_breakdown(subtotal, tax_rate=tax_rate)
        for customer_id, subtotal in zip(customer_ids, subtotals)
    }
    return cust_dict


transaction_dict_creator(customer_ids, subtotals, tax_rate=0.08)
`````

`````python
# ---------------------------------------------------------------
# Comprehensions vs. map()
# Both provide an efficient way to apply a function to an iterable
# ---------------------------------------------------------------

def currency_converter(price, exchange_rate=0.88):
    return round(float(price) * exchange_rate, 2)


# Example data
price_list = [5.99, 19.99, 24.99, 0, 74.99, 99.99]


# ------------------------------------------------
# Using a list comprehension
# ------------------------------------------------

# Clear, readable, and flexible
converted_comp = [currency_converter(price) for price in price_list]
converted_comp


# ------------------------------------------------
# Using map()
# ------------------------------------------------

# map() returns a lazy iterator; wrap in list() to materialize results
converted_map = list(map(currency_converter, price_list))
converted_map


# ------------------------------------------------
# Changing function parameters
# ------------------------------------------------

# Comprehension with a different exchange rate
converted_comp_85 = [currency_converter(price, exchange_rate=0.85) for price in price_list]
converted_comp_85


# Using map() with a fixed parameter via functools.partial
import functools

converted_map_85 = list(
    map(functools.partial(currency_converter, exchange_rate=0.85), price_list)
)
converted_map_85


# Teaching notes:
# - Comprehensions are often preferred for readability and flexibility.
# - map() applies a function to an iterable without the use of a for loop.
# - map() is useful when passing existing functions into pipelines 
# - (e.g., map() with lambda for pandas and dataframes).
# - For complex or reusable logic, define a named function (with def),
#   then apply it with either a comprehension or map().
`````

# 🧠 SECTION 1 — Introduction to Functions
## Chapter 9. Functions as Analytical Abstractions

## Why Functions Exist (Beyond “Reusability”)

In analytics, functions are not just a way to avoid copy-paste code.  
They are **abstractions** — named, trusted operations that convert inputs into outputs in a predictable way.

Think of a function as a **mini-pipeline**:

Input data  
→ validation  
→ transformation  
→ output  

This mirrors how real analytics systems work.

Built-in functions like `max()`, `len()`, and `sum()` already follow this pattern.  
Our goal in this chapter is to learn how to **design our own** functions with the same discipline.

---

## The Anatomy of a Function Call

A function has four conceptual parts:

1. **Inputs (Arguments)**  
   Data and configuration values passed into the function

2. **Processing Logic**  
   The work performed on the inputs

3. **Outputs (Return Values)**  
   The result(s) sent back to the caller

4. **Side Effects (Optional)**  
   Actions taken that are not returned (printing, saving files, logging)

Let’s ground this with a familiar built-in example.

---

## Example: `max()` as a Mental Model

```python
# Input data: a list of prices
price_list = [5.99, 19.99, 24.99, 0, 74.99, 99.99]

# Function call
highest_price = max(price_list)

# Output value
highest_price
```

What happened conceptually:

- **Input**: `price_list`
- **Processing**: Python compares all values
- **Output**: `99.99`
- **Side effects**: None

This exact structure is what we will replicate with custom functions.

---

## Why This Matters in Data Analysis

In real analytics work, you routinely:

- Clean data the same way across projects
- Apply consistent business rules
- Transform raw values into standardized outputs
- Validate inputs before analysis

Functions allow you to:

- Encode **business logic once**
- Reuse it safely
- Improve readability of notebooks
- Reduce error rates

A notebook full of functions reads like **documentation**, not just code.

---

## Preview: From Script Thinking to Function Thinking

Compare these two mindsets:

**Script mindset:**  
“Run these steps top to bottom.”

**Function mindset:**  
“Define reliable operations, then compose them.”

The second mindset scales.  
The first does not.

Everything else in this chapter builds on this shift.

---

## Key Takeaways

- Functions are analytical abstractions, not just shortcuts
- Every function follows input → processing → output
- Built-in functions model best practices we should emulate
- Strong function design leads to cleaner, safer analytics workflows
- This chapter is about **thinking in functions**, not memorizing syntax


# 🧠 SECTION 2 — Functions
## Chapter 9. Defining Custom Functions (Foundations)

## The Purpose of `def`

When you define a function, you are doing three things simultaneously:

1. **Naming a transformation**
2. **Declaring inputs the function expects**
3. **Guaranteeing what the function will return**

In analytics, a function definition is a **contract**:
> “If you give me X (in the correct form), I will reliably give you Y.”

This mindset prevents fragile notebooks and silent errors.

---

## Basic Function Structure

Every function definition follows this structure:

```python
def function_name(parameters):
    """
    Docstring (optional but strongly recommended)
    """
    # processing logic
    return output
```

Key rules:
- `def` starts the definition
- Parameters live inside parentheses
- The colon (`:`) signals a new code block
- `return` explicitly sends a value back to the caller

Without `return`, the function returns `None` by default.

---

## A Minimal, Correct Function

Let’s begin with a deliberately simple function so we can focus on structure.

```python
def add_numbers(a, b):
    """
    Adds two numeric values together.

    Args:
        a (int or float): First number
        b (int or float): Second number

    Returns:
        int or float: Sum of a and b
    """
    total = a + b
    return total
```

Function call:

```python
result = add_numbers(5, 7)
result

# Output: 12
```

What matters here is not the math.  
What matters is **clarity of intent**.

---

## Parameters vs Arguments (Precision Matters)

- **Parameters** are defined in the function signature  
- **Arguments** are the actual values passed during a function call

```python
# parameters: a, b
def multiply(a, b):     # ← Here, 'a' and 'b' are the PARAMETERS
    return a * b

# arguments: 4, 10
result = multiply(4, 10)   # ← Here, 4 and 10 are the ARGUMENTS

# Example:
def greet(name, age):           # parameters: name, age
    print(f"Hello {name}, you are {age} years old.")

greet("Alice", 25)              # arguments: "Alice", 25

# Output: Hello Alice, you are 25 years old.
```
In analytics documentation and team discussions, using these terms correctly matters.

---

## Return Values Are the Output Boundary

A function should **return data**, not *print* data.

Bad analytics practice:
```python
def calculate_tax(subtotal, tax_rate):
    print(subtotal * tax_rate)
```

Correct analytics practice:
```python
def calculate_tax(subtotal, tax_rate):
    """
    Calculates sales tax for a given subtotal.

    Args:
        subtotal (float): Pre-tax amount
        tax_rate (float): Tax rate as a decimal

    Returns:
        float: Tax amount
    """
    tax = subtotal * tax_rate
    return tax
```

Why this matters:
- Returned values can be reused
- Printed values cannot
- Functions should be composable

---

## Docstrings Are Not Optional in Real Analytics

Docstrings are embedded documentation.  
They are how *future you* and *other analysts* understand your work.

Python automatically associates the docstring with the function object.

```python
calculate_tax?
```

Good docstrings answer:
- What does this function do?
- What inputs does it expect?
- What does it return?

This is **professional discipline**, not academic formality.

---

## A Realistic Analytics Example

Let’s formalize something you would absolutely reuse.

```python
def usd_to_eur(amount_usd, exchange_rate=0.88):
    """
    Converts a USD amount to EUR using a given exchange rate.

    Args:
        amount_usd (float): Amount in US dollars
        exchange_rate (float, optional): USD to EUR rate. Defaults to 0.88.

    Returns:
        float: Amount converted to euros
    """
    eur_amount = amount_usd * exchange_rate
    return round(eur_amount, 2)
```

Usage:

```python
usd_to_eur(150)
usd_to_eur(150, exchange_rate=0.91)
```

Notice:
- Default arguments support flexibility
- Rounding is handled inside the function
- Callers don’t need to remember conversion logic

---

## What We Are NOT Doing Yet (On Purpose)

We are *not* yet:
- Adding validation logic
- Handling edge cases
- Raising exceptions
- Writing helpers

Those come **after** the mental model is stable.

Right now, the goal is:
> **Write clean, predictable, reusable function definitions.**

---

## Key Takeaways

- `def` creates a named, reusable transformation
- Parameters define expected inputs
- Arguments are supplied at call time
- `return` is mandatory for reusable analytics logic
- Docstrings are part of professional-quality code
- Functions should return data, not print it

(positional, keyword, defaults, `*args`, `**kwargs`)

# 🧠 SECTION 3 — Functions
## Chapter 9. Defining Custom Functions (Foundations)

## Why Argument Types Matter in Analytics

In analytics, functions rarely operate on *one fixed input*.

They must handle:
- Different data sources
- Optional configuration values
- Varying levels of detail
- Evolving business rules

Argument types are how we design **flexibility without chaos**.

Poorly designed arguments lead to:
- Confusing function calls
- Silent bugs
- Fragile notebooks

Well-designed arguments lead to:
- Self-documenting code
- Safe reuse
- Clean pipelines

---

## 1️⃣ Positional Arguments (The Default)

Positional arguments are matched **by order**.

```python
def revenue(price, quantity):
    """
    Calculates total revenue.

    Args:
        price (float): Unit price
        quantity (int): Number of units sold

    Returns:
        float: Total revenue
    """
    return price * quantity
```

Usage:

```python
revenue(19.99, 3)
```

⚠️ **Risk**: Order matters.

```python
# This will run, but produces the wrong result
revenue(3, 19.99)
```

### Analytics Guidance
- Use positional arguments for **primary data**
- Keep them few
- Order them logically

---

## 2️⃣ Keyword Arguments (Clarity Over Brevity)

Keyword arguments explicitly name each input.

```python
revenue(price=19.99, quantity=3)
```

Why this matters:
- Order no longer matters
- Code reads like documentation
- Safer for complex functions

```python
revenue(quantity=3, price=19.99)
```

### Analytics Guidance
- Prefer keyword arguments for **configuration**
- Especially useful when revisiting notebooks months later

---

## 3️⃣ Default Arguments (Safe Flexibility)

Default arguments allow functions to work **out of the box**, while still supporting overrides.

```python
def usd_to_eur(amount_usd, exchange_rate=0.88):
    """
    Converts USD to EUR.

    Args:
        amount_usd (float): Amount in USD
        exchange_rate (float, optional): Conversion rate

    Returns:
        float: Amount in EUR
    """
    return round(amount_usd * exchange_rate, 2)
```

Usage:

```python
usd_to_eur(100)
usd_to_eur(100, exchange_rate=0.91)
```

### Rules You Must Remember
- Default arguments **must come last**
- Defaults should represent the **most common case**

```python
# ❌ Invalid
def bad_example(rate=0.88, amount):
    pass
```

---

## 4️⃣ `*args` — Variable Positional Arguments

`*args` allows a function to accept **any number of positional inputs**, collected as a tuple.

### When This Is Useful
- Aggregations
- Concatenation
- Flexible calculations

```python
def average(*values):
    """
    Calculates the average of any number of numeric values.

    Args:
        *values (float): One or more numeric values

    Returns:
        float: Average of values
    """
    return sum(values) / len(values)
```

Usage:

```python
average(10, 20, 30)
average(5, 15)
```

What’s happening internally:
- `values` is a tuple
- We can loop, sum, or unpack it

### Analytics Guidance
- Use `*args` when the **number of inputs is unknown**
- Avoid it when inputs have distinct meanings

---

## 5️⃣ `**kwargs` — Variable Keyword Arguments

`**kwargs` accepts **any number of named arguments**, collected as a dictionary.

This is extremely powerful in analytics pipelines.

```python
def summarize_metrics(**metrics):
    """
    Summarizes named metrics.

    Args:
        **metrics (float): Named metric values

    Returns:
        dict: Metrics with rounded values
    """
    summary = {}
    for key, value in metrics.items():
        summary[key] = round(value, 2)
    return summary
```

Usage:

```python
summarize_metrics(
    revenue=12345.678,
    cost=9876.543,
    margin=2469.135
)
```

### Why This Is Important
- Accepts dynamic inputs
- Works naturally with dictionaries
- Ideal for reporting outputs

---

## 🚨 Pro Insight: Argument Ordering Discipline

The correct argument order in professional code is:

```python
def function_name(
    required_positional,
    optional_positional=default,
    *args,
    keyword_only=default,
    **kwargs
):
    pass
```

You won’t always use all of these, but you **must recognize the pattern**.

---

## Analytics Design Pattern Example

Let’s combine everything in a realistic function.

```python
def tax_calculator(
    subtotal,
    tax_rate=0.06,
    *discounts,
    **metadata
):
    """
    Calculates final total with optional discounts and metadata.

    Args:
        subtotal (float): Base transaction amount
        tax_rate (float, optional): Sales tax rate
        *discounts (float): Any discount amounts
        **metadata: Additional transaction info

    Returns:
        dict: Transaction summary
    """
    discounted_total = subtotal - sum(discounts)
    tax = discounted_total * tax_rate
    total = discounted_total + tax

    return {
        "subtotal": subtotal,
        "discounts": discounts,
        "tax": round(tax, 2),
        "total": round(total, 2),
        "metadata": metadata
    }
```

Usage:

```python
tax_calculator(
    200,
    0.07,
    10,
    5,
    customer_id="C0042",
    region="Stowe"
)
```

This mirrors **real ETL-style analytics logic**.

---

## What We Are Intentionally Deferring

We are still **not**:
- Validating types
- Raising exceptions
- Using decorators

Those belong after scope and return strategies are fully internalized.

---

## Key Takeaways

- Positional arguments are matched by order
- Keyword arguments improve clarity and safety
- Default arguments support flexibility
- `*args` captures variable positional inputs
- `**kwargs` captures variable named inputs
- Argument design is a **core analytics skill**, not just syntax

# 🧠 SECTION 4 — Functions
## Chapter 9. Return Strategies (Designing Outputs for Reuse)

## Why Return Strategy Matters More Than You Think

In analytics, **what you return** from a function determines:

- How easily results can be reused
- Whether outputs compose cleanly into pipelines
- How readable downstream code becomes
- Whether your logic scales beyond a single notebook cell

A function’s return value is its **public interface**.

Bad return design forces callers to:
- Recompute values
- Rely on globals
- Parse printed output (❌)
- Rewrite logic

Good return design makes functions **lego bricks**.

---

## Rule Zero: Functions Should Return Data, Not Behavior

A function should:
- Return values
- Avoid printing (except debugging)
- Avoid mutating external state

This keeps functions:
- Testable
- Reusable
- Predictable

---

## 1️⃣ Returning a Single Value (Scalar)

This is the simplest and most common case.

```python
def profit(revenue, cost):
    """
    Calculates profit.

    Args:
        revenue (float): Total revenue
        cost (float): Total cost

    Returns:
        float: Profit
    """
    return revenue - cost
```

Usage:

```python
profit(12000, 8700)
```

### Analytics Guidance
- Use scalar returns when the function answers **one clear question**
- Ideal for metrics and KPIs

---

## 2️⃣ Returning Multiple Values (Tuple)

Python allows returning multiple values separated by commas.
Internally, this creates a **tuple**.

```python
def revenue_breakdown(price, quantity):
    """
    Calculates revenue and tax-free subtotal.

    Args:
        price (float): Unit price
        quantity (int): Units sold

    Returns:
        tuple: (subtotal, revenue)
    """
    subtotal = price * quantity
    revenue = subtotal * 1.06
    return subtotal, revenue
```

Usage:

```python
subtotal, revenue = revenue_breakdown(19.99, 5)
subtotal
revenue
```

### Why Tuples Are Useful
- Lightweight
- Ordered
- Easy to unpack

### When NOT to Use Tuples
- When outputs have **semantic meaning**
- When order confusion is likely

If order matters for understanding, tuples are fragile.

---

## 3️⃣ Returning a List (Ordered, Mutable)

Lists are useful when:
- Order matters
- Outputs are homogeneous
- Values may be iterated or transformed

```python
def apply_tax(subtotal, tax_rate=0.06):
    """
    Returns subtotal, tax amount, and total.

    Args:
        subtotal (float): Base amount
        tax_rate (float): Tax rate

    Returns:
        list: [subtotal, tax, total]
    """
    tax = subtotal * tax_rate
    total = subtotal + tax
    return [subtotal, tax, total]
```

Usage:

```python
transaction = apply_tax(100)
transaction
```

### Analytics Guidance
- Lists are acceptable for **pipeline-style iteration**
- But meaning is implicit, not explicit

This is often a stepping stone, not a final design.

---

## 4️⃣ Returning a Dictionary (Preferred for Analytics)

Dictionaries are the **gold standard** for analytical outputs.

They provide:
- Explicit meaning
- Named fields
- Safe downstream access
- Natural compatibility with DataFrames later

```python
def transaction_summary(subtotal, tax_rate=0.06):
    """
    Builds a transaction summary.

    Args:
        subtotal (float): Base amount
        tax_rate (float): Tax rate

    Returns:
        dict: Transaction details
    """
    tax = subtotal * tax_rate
    total = subtotal + tax

    return {
        "subtotal": round(subtotal, 2),
        "tax": round(tax, 2),
        "total": round(total, 2)
    }
```

Usage:

```python
summary = transaction_summary(100)
summary["total"]
```

### Why Dictionaries Win
- Self-documenting
- Order-independent
- Robust to future changes
- Ideal for ETL and reporting

If you are unsure which return type to use — **choose a dictionary**.

---

## 5️⃣ Returning Structured Data for Pipelines

Let’s design a function meant to feed another function.

```python
def clean_price(raw_price):
    """
    Cleans and standardizes price input.

    Args:
        raw_price (str or float): Raw price input

    Returns:
        dict: Cleaned price data
    """
    cleaned = float(str(raw_price).replace("$", "").strip())
    return {
        "raw": raw_price,
        "cleaned": cleaned,
        "valid": cleaned >= 0
    }
```

Usage:

```python
price_info = clean_price("$19.99")
price_info
```

Downstream code can now safely rely on:
- Validation status
- Cleaned value
- Original input

This is **pipeline-ready design**.

---

## 🚨 Anti-Pattern: Returning Nothing (Implicit `None`)

```python
def bad_function(x):
    y = x * 2
```

```python
result = bad_function(10)
result  # None
```

This silently breaks pipelines.

### Rule
If a function does work that matters, **it must return something**.

---

## Design Heuristic: Choosing the Right Return Type

Ask yourself:

- One value? → scalar
- Multiple values, fixed order, internal use? → tuple
- Iterable output? → list
- Named, reusable analytics output? → dictionary

This decision is **architectural**, not cosmetic.

---

## Key Takeaways

- Return values define how functions are reused
- Tuples are lightweight but order-dependent
- Lists support iteration but lack meaning
- Dictionaries are the preferred analytics return type
- Never rely on printing instead of returning
- Good return design enables clean pipelines

---

# 🧠 SECTION 5 — Functions
## Chapter 9. Variable Scope & Lifetime (Data Safety by Design)

## Why Variable Scope Matters in Analytics

Variable scope determines **where a variable exists** and **who can see it**.

In analytics work, poor understanding of scope leads to:
- Silent overwrites
- Hard-to-debug errors
- Fragile notebooks
- Functions that behave differently depending on execution order

Good scope discipline leads to:
- Predictable behavior
- Reusable functions
- Safe pipelines
- Code that scales beyond a single notebook

Scope is not a Python quirk.  
It is a **data safety concept**.

---

## The Two Core Scopes You Must Master

Python introduces many scopes, but for now we focus on the two that matter most:

1. **Local scope** — variables created *inside* a function  
2. **Global scope** — variables created *outside* all functions  

---

## 1️⃣ Local Scope (Default and Preferred)

Variables created inside a function live **only inside that function**.

```python
def calculate_total(price, quantity):
    total = price * quantity
    return total
```

```python
calculate_total(10, 3)
```

```python
# This will fail
total
```

Why?
- `total` was created inside the function
- It does not exist outside the function

This is **intentional and good**.

### Analytics Insight
Local scope:
- Prevents accidental overwrites
- Forces explicit data flow
- Makes functions reliable building blocks

---

## 2️⃣ Global Scope (Powerful but Dangerous)

Variables defined outside functions are global.

```python
tax_rate = 0.06

def calculate_tax(subtotal):
    return subtotal * tax_rate
```

```python
calculate_tax(100)
```

This works — but introduces risk.

### The Hidden Problem

```python
tax_rate = 0.09
calculate_tax(100)
```

Same function.  
Different result.

Nothing inside the function changed.

This is how analytics bugs are born.

---

## 🚨 The `global` Keyword (Why We Avoid It)

Python allows functions to modify global variables using `global`.

```python
counter = 0

def increment():
    global counter
    counter += 1
```

```python
increment()
increment()
counter
```

This *works* — and is almost always a **bad idea**.

### Why This Is Dangerous
- Hidden side effects
- Execution-order dependence
- Impossible-to-debug pipelines
- Breaks reuse across notebooks

### Professional Rule
> **Analytics functions should return values, not mutate globals.**

---

## Correct Pattern: Return Values, Then Assign

Instead of this:

```python
total = 0

def add_to_total(x):
    global total
    total += x
```

Do this:

```python
def add_to_total(current_total, x):
    return current_total + x
```

```python
total = 0
total = add_to_total(total, 10)
total = add_to_total(total, 5)
total
```

This is:
- Explicit
- Testable
- Predictable

---

## Scope and Name Shadowing (A Subtle Trap)

Python allows local variables to have the same name as globals.

```python
value = 100

def double_value():
    value = 50
    return value * 2
```

```python
double_value()
value
```

What happened?
- Inside the function, `value` refers to the local variable
- The global `value` remains unchanged

This is called **name shadowing**.

### Analytics Guidance
- Avoid reusing global variable names inside functions
- Choose descriptive parameter names

---

## Lifetime: Variables Exist Only While Needed

Variable **lifetime** refers to how long a variable exists in memory.

- Local variables exist only while the function runs
- Global variables persist for the entire session

Why this matters:
- Memory efficiency
- Preventing stale state
- Avoiding accidental reuse

Functions naturally clean up after themselves.  
Globals do not.

---

## Analytics Design Rule (Memorize This)

> **Functions should behave the same way every time they are called with the same inputs.**

This property is called **referential transparency**.

Global state breaks this property.

---

## A Safe, Analytics-Grade Pattern

```python
def compute_metrics(values):
    """
    Computes summary metrics from a list of values.

    Args:
        values (list of float): Input values

    Returns:
        dict: Summary statistics
    """
    count = len(values)
    total = sum(values)
    average = total / count

    return {
        "count": count,
        "total": total,
        "average": round(average, 2)
    }
```

Nothing leaks.
Nothing mutates.
Everything is explicit.

This function is:
- Safe
- Reusable
- Pipeline-ready

---

## What We Are Deliberately Avoiding

At this stage, we are **not**:
- Using closures
- Using nonlocal variables
- Introducing decorators

Those come later, if needed.

Right now, the goal is **discipline, not cleverness**.

---

## Key Takeaways

- Local scope is safe and preferred
- Global scope introduces hidden risk
- `global` should almost never be used in analytics code
- Functions should return values, not mutate external state
- Predictability beats convenience
- Scope discipline is foundational for reusable analytics

---

➡️ **Functions as Reusable Assets**  
Creating modules, saving helpers, and importing cleanly across notebooks

When ready, say: **“Proceed to Creating Modules & Reuse.”**


# 🧠 SECTION 6 — Functions
## Creating Modules & Reuse (From Notebook to Asset)

## Why Modules Matter in Analytics

In real analytics work, notebooks are often:
- Exploratory
- Iterative
- Messy by design

Modules, on the other hand, are:
- Stable
- Reusable
- Shareable
- Testable

A **module** is simply a `.py` file that contains functions (and sometimes constants).
The moment you move logic into a module, you are declaring:

> “This logic is trusted and reusable.”

This is a major professional shift.

---

## Mental Model: Notebook vs Module

Think of it this way:

- **Notebook** → analysis, experimentation, storytelling
- **Module** → utilities, helpers, business logic

Not everything belongs in a module.
But anything you expect to reuse **does**.

---

## Step 1️⃣ — Creating a Module from a Notebook

In Jupyter, we use the `%%writefile` magic command to create a `.py` file.

```python
%%writefile pricing_utils.py

def apply_discount(price, discount_rate=0.10):
    """
    Applies a discount to a price.

    Args:
        price (float): Original price
        discount_rate (float): Discount rate as a decimal

    Returns:
        float: Discounted price
    """
    discounted_price = price * (1 - discount_rate)
    return round(discounted_price, 2)
```

What just happened:
- A file named `pricing_utils.py` was created
- The function is now **persisted on disk**
- It can be imported into any notebook in the same directory

This is the bridge from **exploration to engineering**.

---

## Step 2️⃣ — Importing a Module

### Option A: Import the Entire Module (Preferred)

```python
import pricing_utils
```

Usage:

```python
pricing_utils.apply_discount(100)
```

Why this is good practice:
- Prevents naming conflicts
- Makes function origin explicit
- Scales well as modules grow

---

### Option B: Import Specific Functions (Use Carefully)

```python
from pricing_utils import apply_discount
```

Usage:

```python
apply_discount(100)
```

⚠️ Risk:
- Easy to lose track of where functions come from
- Can shadow other names accidentally

### Analytics Rule of Thumb
- Use `import module_name` for shared or growing code
- Use `from module import function` sparingly

---

## Step 3️⃣ — Multiple Functions per Module (Best Practice)

Modules should group **related logic**.

```python
%%writefile pricing_utils.py

def apply_discount(price, discount_rate=0.10):
    """
    Applies a discount to a price.
    """
    return round(price * (1 - discount_rate), 2)


def add_tax(price, tax_rate=0.06):
    """
    Adds tax to a price.
    """
    return round(price * (1 + tax_rate), 2)
```

Usage:

```python
import pricing_utils

pricing_utils.add_tax(100)
pricing_utils.apply_discount(100, 0.15)
```

This is how **helper libraries** begin to form.

---

## Step 4️⃣ — Designing Modules for Analytics Pipelines

Let’s design a module that returns **structured outputs**.

```python
%%writefile transaction_utils.py

def process_transaction(subtotal, tax_rate=0.06, discount_rate=0.0):
    """
    Processes a transaction and returns a structured summary.

    Args:
        subtotal (float): Base transaction amount
        tax_rate (float): Sales tax rate
        discount_rate (float): Discount rate

    Returns:
        dict: Transaction summary
    """
    discounted = subtotal * (1 - discount_rate)
    tax = discounted * tax_rate
    total = discounted + tax

    return {
        "subtotal": round(subtotal, 2),
        "discounted": round(discounted, 2),
        "tax": round(tax, 2),
        "total": round(total, 2)
    }
```

Usage:

```python
import transaction_utils

transaction_utils.process_transaction(200, discount_rate=0.1)
```

This is **ETL-style thinking**:
- Input → transformation → structured output

---

## Naming Discipline (Critical)

### File Names
- lowercase
- underscores
- descriptive

Good:
```text
pricing_utils.py
date_helpers.py
validation_tools.py
```

Bad:
```text
Stuff.py
MyFunctionsFINAL_v3.py
```

### Function Names
- verbs for actions
- nouns for data builders

Good:
```python
calculate_margin
build_summary
clean_price
```

---

## Where Modules Live (Important for Later)

For now:
- Modules live in the **same folder** as your notebook

Later (future you):
- You’ll organize helpers into folders
- You’ll manage environments
- You’ll version-control these files

But the *discipline* starts now.

---

## Common Anti-Patterns to Avoid

❌ Saving logic only in notebooks  
❌ Copy-pasting helper code across notebooks  
❌ Using globals inside modules  
❌ Treating modules as “advanced”  

Modules are not advanced.
They are **necessary**.

---

## Key Takeaways

- Modules turn notebook logic into reusable assets
- `%%writefile` creates persistent `.py` files
- Importing modules makes code explicit and safe
- Group related helpers into the same module
- Return structured data from module functions
- Modules are foundational for scalable analytics

---

# 🧠 SECTION 7 — Functions
## Chapter 9. External Imports & Packages (Using the Ecosystem Safely)

## Why External Packages Matter in Analytics

Python is powerful not because of the core language alone, but because of its **ecosystem**.

Analytics work routinely depends on:
- Mathematical functions
- Date/time utilities
- File I/O helpers
- Data manipulation libraries

These capabilities live in **external modules and packages**, not in your notebook.

Your responsibility as an analyst is not to memorize them —  
it is to **import them correctly, safely, and transparently**.

---

## Terminology (Precise but Practical)

- **Module** → a single `.py` file  
- **Package** → a collection of modules  
- **Library** → a collection of packages (term used loosely)

Example:
- `math` → module
- `datetime` → package
- `pandas` → library (contains many packages and modules)

---

## 1️⃣ Importing External Modules (Baseline Pattern)

The most common and safest pattern:

```python
import math
```

Usage:

```python
math.sqrt(25)
math.log10(100)
```

Why this pattern is preferred:
- Prevents naming conflicts
- Makes function origin explicit
- Scales well in large notebooks

---

## 2️⃣ Importing with Aliases (Professional Standard)

Aliases reduce verbosity while preserving clarity.

```python
import math as m
```

Usage:

```python
m.sqrt(16)
m.pi
```

### Analytics Rule
- Use **standard aliases** when they exist
- Be consistent across notebooks

Examples you will see later:
```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Consistency is more important than personal preference.

---

## 3️⃣ Importing Specific Functions (Use Sparingly)

```python
from math import sqrt
```

Usage:

```python
sqrt(36)
```

⚠️ This looks clean — but introduces risk.

### The Naming Conflict Problem

```python
def sqrt(x):
    return x ** 0.5
```

```python
from math import sqrt
sqrt(25)  # Which sqrt is this?
```

The **most recent definition wins**, silently.

### Professional Guidance
- Avoid `from module import function` in shared analytics code
- Prefer explicit namespaces (`math.sqrt`)

---

## 4️⃣ Inspecting a Package with `dir()`

When you don’t know what a package contains:

```python
import math
dir(math)
```

This reveals:
- Available functions
- Constants
- Attributes

This is especially useful when:
- Exploring new libraries
- Debugging unexpected behavior

---

## 5️⃣ Reading Documentation (Non-Negotiable Skill)

Good analysts read docs.

For example, the official `math` documentation explains:
- Expected inputs
- Return types
- Edge cases

This habit prevents misuse and incorrect assumptions.

---

## 6️⃣ External Packages in Analytics Context

Let’s simulate a realistic workflow.

```python
import math

def safe_square_root(value):
    """
    Computes square root of a non-negative number.

    Args:
        value (float): Input value

    Returns:
        float or None: Square root if valid
    """
    if value < 0:
        return None
    return math.sqrt(value)
```

Usage:

```python
safe_square_root(25)
safe_square_root(-4)
```

This demonstrates:
- External function reuse
- Local validation
- Safe return behavior

---

## 7️⃣ Installing Packages (Environment Awareness)

In analytics environments (especially Anaconda), **where** you install matters.

⚠️ This is often misunderstood.

### What You Will See (But Should Avoid Blindly)

```python
!pip install some_package
```

or

```python
!conda install some_package
```

These commands talk to your **system shell**, not Python itself.

### Correct Mental Model
- Python runs inside an **environment**
- Packages must be installed into the **active environment**
- Jupyter kernels must match that environment

For now:
- Prefer installing via **Anaconda Navigator**
- Or ensure the correct environment is active

We will revisit this when pandas enters the picture.

---

## 8️⃣ Package Versions & Compatibility

Different packages depend on different versions of other packages.

This means:
- Updating one package can break another
- Reproducibility matters

Tools you should know exist:
```python
conda list
```

And Anaconda’s **Environments** tab.

This is not optional knowledge for analytics professionals.

---

## 9️⃣ External Imports + Your Modules (Together)

You can freely mix:
- Your custom modules
- External packages

```python
import math
import transaction_utils

def transaction_with_fee(subtotal, fee_rate=0.02):
    summary = transaction_utils.process_transaction(subtotal)
    fee = math.ceil(summary["total"] * fee_rate)
    summary["fee"] = fee
    summary["final_total"] = summary["total"] + fee
    return summary
```

This is **composable analytics design**.

---

## Anti-Patterns to Avoid

❌ Importing everything at the top without using it  
❌ Wildcard imports (`from math import *`)  
❌ Installing packages inside notebooks without environment awareness  
❌ Ignoring version conflicts  
❌ Overwriting imported function names  

---

## Key Takeaways

- External packages extend Python’s analytical power
- Prefer `import module` with explicit namespaces
- Use aliases consistently
- Avoid function-level imports in shared code
- Inspect packages with `dir()`
- Be environment-aware when installing packages
- Combine external packages with your own modules thoughtfully

# 🧠 SECTION 8 — Functions
## Chapter 9. `map()` & Function Application (Scaling Clean Logic)

## Why This Section Exists

In analytics, you rarely apply logic to a single value.

You apply the **same transformation** to:
- Every row in a dataset
- Every transaction in a list
- Every value in a column

The question is not *whether* to scale logic —  
it is **how to do so clearly and safely**.

`map()` is one such mechanism.

---

## What `map()` Actually Does

The `map()` function applies a function to **each element** in an iterable.

Syntax:
```python
map(function, iterable)
```

Important:
- `map()` does **not** immediately return a list
- It returns a **map object** (lazy evaluation)

---

## A Simple, Grounded Example

```python
def square(x):
    """
    Squares a numeric value.

    Args:
        x (int or float): Input value

    Returns:
        int or float: Squared value
    """
    return x ** 2
```

```python
values = [1, 2, 3, 4]
```

Applying `map()`:

```python
mapped_values = map(square, values)
mapped_values
```

Nothing visible yet.

Convert to a list:

```python
list(mapped_values)
```

### Mental Model
- `map()` describes *what to do*
- Conversion (`list`, `tuple`, etc.) forces execution

---

## Why `map()` Is Memory-Efficient

`map()` evaluates values **one at a time**, not all at once.

This matters when:
- Working with large datasets
- Streaming data
- Avoiding unnecessary memory usage

However…

---

## Analytics Reality Check

In most **modern analytics workflows**, especially notebooks:

- Readability matters more than micro-optimizations
- Pipelines must be easily understood
- Debugging clarity beats cleverness

This is why `map()` is **not always preferred**, even though it is efficient.

---

## `map()` with Functions You Wrote (Common Case)

```python
def usd_to_eur(amount_usd, exchange_rate=0.88):
    return round(amount_usd * exchange_rate, 2)
```

```python
usd_prices = [10, 25, 40, 100]
```

Using `map()`:

```python
eur_prices = list(map(usd_to_eur, usd_prices))
eur_prices
```

This works cleanly **because**:
- The function accepts one primary argument
- Defaults handle configuration

---

## `map()` with Multiple Iterables

```python
def revenue(price, quantity):
    return price * quantity
```

```python
prices = [10, 20, 30]
quantities = [1, 3, 2]
```

```python
list(map(revenue, prices, quantities))
```

`map()` pairs values **by position**.

⚠️ If lengths differ, extra values are silently ignored.

---

## 🚨 When `map()` Starts to Strain

Configuration arguments become awkward.

```python
list(map(lambda x: usd_to_eur(x, exchange_rate=0.91), usd_prices))
```

This works — but readability declines quickly.

This is your first hint that **another tool may be better**.

---

## `map()` + Lambda (One-Off Logic)

Lambda functions shine when:
- Logic is trivial
- Reuse is unnecessary
- Context is obvious

```python
orders = [120, 450, 600, 80]
```

Apply a one-time discount:

```python
discounted = list(
    map(lambda x: x * 0.9 if x > 500 else x, orders)
)
discounted
```

### Analytics Guidance
- Lambdas should be short and obvious
- If you need a docstring, don’t use lambda
- If logic grows, extract a function

---

## Comparing `map()` to a For Loop

```python
results = []
for value in values:
    results.append(square(value))
```

Versus:

```python
list(map(square, values))
```

Both are valid.

### Decision Rule
- For learning and debugging → for loop
- For concise transformation → `map()` or comprehension
- For pipelines → comprehension (next section)

---

## `map()` in Modern Analytics (Important Truth)

In isolation, `map()` is fine.

In **real analytics**, you will see it most often:
- Inside libraries (pandas, NumPy)
- Combined with lambdas
- Applied to Series objects

Standalone `map()` is less common than comprehensions, but still worth understanding deeply.

---

## Design Heuristic (Memorize This)

Use `map()` when:
- Applying a function to one iterable
- Logic is simple
- Memory efficiency matters
- Configuration is minimal

Avoid `map()` when:
- Logic is complex
- Multiple parameters are needed
- Readability suffers

---

## Key Takeaways

- `map()` applies a function across an iterable
- It returns a lazy map object
- Convert to a list to materialize results
- Works best with simple, single-argument functions
- Lambdas are appropriate for short-lived logic
- Readability matters more than cleverness in analytics
- `map()` is a tool, not a default

---

### Next Section

➡️ **Comprehensions as Function Consumers**  
Why comprehensions often replace `map()` and how to use them cleanly in analytics pipelines


# 🧠 SECTION 9 — Functions
## Chapter 9. Comprehensions (Readable, Scalable Function Consumption)

## Why Comprehensions Matter More Than `map()`

Comprehensions are not just syntactic sugar.

They are:
- Explicit
- Readable
- Flexible
- Pipeline-friendly

In analytics, comprehension syntax often mirrors how we *think*:
> “For each value, apply this transformation, optionally filter, and collect the result.”

This clarity is why comprehensions are usually preferred over `map()`.

---

## The Core Comprehension Pattern

### List Comprehension Syntax

```python
[expression for item in iterable]
```

Think of it as:
- **expression** → what to produce
- **item** → each element
- **iterable** → the data source

---

## A Direct Comparison to `map()`

Using `map()`:

```python
list(map(square, values))
```

Using a comprehension:

```python
[square(x) for x in values]
```

Why the comprehension is better:
- No lazy object to reason about
- Reads left to right
- Expression-first emphasizes the transformation

---

## Comprehensions with Custom Functions

```python
def usd_to_eur(amount_usd, exchange_rate=0.88):
    return round(amount_usd * exchange_rate, 2)
```

```python
usd_prices = [10, 25, 40, 100]
```

Using a comprehension:

```python
eur_prices = [usd_to_eur(price) for price in usd_prices]
eur_prices
```

This is **canonical analytics Python**.

---

## Adding Conditional Logic (Filtering)

Comprehensions support inline filtering.

```python
high_value_orders = [x for x in orders if x > 500]
```

With transformation:

```python
discounted_orders = [
    x * 0.9
    for x in orders
    if x > 500
]
```

### Mental Model
- The `if` clause filters
- The expression transforms

---

## Conditional Logic Inside the Expression

```python
adjusted_orders = [
    x * 0.9 if x > 500 else x
    for x in orders
]
```

This reads as:
> “For each order, apply a discount if it qualifies; otherwise leave it unchanged.”

Be careful:
- This is **not** a filter
- All items remain in the output

---

## Dictionary Comprehensions (Analytics Favorite)

Dictionary comprehensions are ideal for:
- Mapping IDs to results
- Building structured outputs
- ETL-style transformations

### Syntax

```python
{key_expression: value_expression for item in iterable}
```

---

## Example: Transaction Mapping

```python
transactions = {
    "C001": 120,
    "C002": 560,
    "C003": 80,
    "C004": 900
}
```

```python
def apply_discount(amount):
    return round(amount * 0.9, 2)
```

```python
discounted_transactions = {
    customer_id: apply_discount(amount)
    for customer_id, amount in transactions.items()
    if amount > 500
}
```

This produces:
- Explicit keys
- Filtered values
- Clean transformation

---

## Comprehensions + Functions = Pipelines

Let’s chain functions together cleanly.

```python
def clean_price(raw):
    return float(str(raw).replace("$", "").strip())
```

```python
raw_prices = ["$10.00", " $25 ", "$40.50"]
```

```python
cleaned_prices = [clean_price(p) for p in raw_prices]
cleaned_prices
```

This is the beginning of **functional pipelines**.

---

## When NOT to Use Comprehensions

Comprehensions lose value when:
- Logic becomes nested
- Multiple iterables are involved
- Complex branching is required
- Debugging becomes difficult

Bad example:

```python
[
    complex_logic(a, b, c)
    for a in data1
    for b in data2
    if condition(a, b)
]
```

This is a readability red flag.

### Rule
> If you can’t explain the comprehension in one sentence, refactor.

---

## Comprehensions vs `map()` (Final Verdict)

Use comprehensions when:
- Readability matters
- Configuration is needed
- Conditional logic exists
- Output structure matters

Use `map()` when:
- Logic is trivial
- You are inside a library context
- Laziness or memory efficiency matters

In notebooks, comprehensions usually win.

---

## Key Takeaways

- Comprehensions are the preferred way to apply functions in analytics
- They read naturally and left-to-right
- Support transformation and filtering
- Dictionary comprehensions enable clean ETL mappings
- Avoid overcomplicated comprehensions
- Readability is the primary optimization

---

# 🧠 SECTION 10 — Functions
## Chapter 9. Analytics Workflow Patterns (Composing Real Pipelines)

## Why This Section Exists

Up to this point, we have learned *mechanics*:
- Defining functions
- Designing arguments
- Choosing return strategies
- Managing scope
- Creating modules
- Applying functions with `map()` and comprehensions

This section answers the real question:

> **How do analysts actually use all of this together?**

The answer is **composition**.

Good analytics code is not one big function.
It is a sequence of **small, trusted transformations**.

---

## The Analytics Pipeline Mental Model

Most analytics workflows follow this shape:

1. **Ingest** raw inputs  
2. **Clean** and normalize data  
3. **Validate** assumptions  
4. **Transform** values  
5. **Aggregate / summarize**  
6. **Expose structured outputs**

Each step should be:
- Small
- Testable
- Reusable
- Explicit

Functions are the units that make this possible.

---

## Pattern 1️⃣ — One Responsibility per Function

Each function should do **one thing well**.

```python
def clean_price(raw_price):
    """
    Cleans raw price input and converts to float.

    Args:
        raw_price (str or float): Raw price value

    Returns:
        float: Cleaned price
    """
    return float(str(raw_price).replace("$", "").strip())
```

```python
def apply_tax(amount, tax_rate=0.06):
    """
    Applies sales tax to an amount.

    Args:
        amount (float): Base amount
        tax_rate (float): Tax rate

    Returns:
        float: Amount with tax
    """
    return round(amount * (1 + tax_rate), 2)
```

Each function:
- Has a single responsibility
- Is independently testable
- Can be reused elsewhere

---

## Pattern 2️⃣ — Compose Functions, Don’t Nest Them

Instead of this:

```python
apply_tax(clean_price("$19.99"))
```

Prefer this:

```python
price = clean_price("$19.99")
final_price = apply_tax(price)
final_price
```

Why?
- Easier to debug
- Easier to log or inspect
- Easier to modify later

Clarity beats cleverness.

---

## Pattern 3️⃣ — Pipelines with Comprehensions

Let’s scale the workflow.

```python
raw_prices = ["$10.00", " $25 ", "$40.50", "$5"]
```

Pipeline:

```python
cleaned_prices = [clean_price(p) for p in raw_prices]
final_prices = [apply_tax(p) for p in cleaned_prices]
final_prices
```

This reads like a story:
1. Clean the prices
2. Apply tax

Each step is explicit.

---

## Pattern 4️⃣ — Returning Structured Outputs for Downstream Use

When workflows grow, return **dictionaries**.

```python
def process_price(raw_price, tax_rate=0.06):
    """
    Cleans a price and applies tax.

    Args:
        raw_price (str or float): Raw price
        tax_rate (float): Tax rate

    Returns:
        dict: Price processing summary
    """
    cleaned = clean_price(raw_price)
    total = apply_tax(cleaned, tax_rate)

    return {
        "raw": raw_price,
        "cleaned": cleaned,
        "total": total
    }
```

Usage:

```python
summaries = [process_price(p) for p in raw_prices]
summaries
```

This structure:
- Preserves provenance
- Supports auditing
- Feeds reporting layers cleanly

---

## Pattern 5️⃣ — Dictionary Pipelines (ETL Style)

```python
transactions = {
    "C001": "$120.00",
    "C002": "$560.50",
    "C003": "$80.00"
}
```

```python
processed_transactions = {
    customer_id: process_price(amount)
    for customer_id, amount in transactions.items()
}
```

This is **ETL in miniature**:
- Extract → raw input
- Transform → functions
- Load → structured dictionary

---

## Pattern 6️⃣ — Promote Stable Logic to Modules

Once logic is trusted:

- Move it to a `.py` file
- Import it into notebooks
- Treat it as infrastructure

```python
import pricing_utils
```

```python
pricing_utils.apply_discount(100)
```

This is how analytics code becomes **organizational memory**.

---

You are no longer:
- Writing linear scripts
- Copy-pasting logic
- Treating notebooks as disposable

You are now:
- Designing interfaces
- Controlling data flow
- Building reusable assets
- Thinking in pipelines

This is the difference between *using Python* and *doing analytics with Python*.

---

## Final Design Principles (Bookmark These)

- Small functions beat large ones
- Explicit data flow beats hidden state
- Dictionaries beat positional tuples
- Comprehensions beat `map()` for clarity
- Modules beat copy-paste
- Readability beats cleverness
- Predictability beats convenience

---

## Chapter-Level Key Takeaways

- Functions are analytical building blocks
- Argument and return design is architectural
- Scope discipline prevents bugs
- Modules enable reuse and scale
- Comprehensions enable readable pipelines
- Analytics code should read like logic, not tricks
