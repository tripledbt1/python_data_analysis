# Chapter 07 — Loops  
## Section 7.04 — Nested Loops  

### Preview
A **nested loop** is a loop inside another loop.  
For every single iteration of the **outer loop**, the **inner loop** runs from start to finish.


> Nested loops allow you to process multi-level structures, generate combinations, and traverse data in two dimensions.

Nested loops are essential for:

- Navigating **nested lists**  
- Generating all **combinations** or **pairings**  
- Performing **matrix-like** operations  
- Processing **customer → orders** or **category → items** hierarchies  
- Applying transformations across multi-dimensional datasets  

---

- The expanded toolkit includes:  
  - `nested_loop_trace()` visualizer
  - `transform_nested_list()`(general purpose)
  - `cartesian_product()` generator
  - `visualize_matrix()` for 2D structures

These tools give full transparency into multi-dimensional iteration.

## 1. Basic Nested Loop Structure

```
for item in outer_list:
    for inner_item in inner_list:
        do_something(item, inner_item)
```

Execution pattern:

- Outer iteration 0 → inner loop runs fully  
- Outer iteration 1 → inner loop runs fully  
- …  
- Outer iteration n → inner loop runs fully  

Total iterations =  
`len(outer_list) × len(inner_list)`


````python
# --------------------------------------------------------
# Nested List Example — Item × Size Combinations
# --------------------------------------------------------
# Output count:
# - 2 items  
# - 3 sizes  
# - 2 × 3 = 6 combinations

items = ["Snowboard", "Boots"]
sizes = ["small", "medium", "large"]

# OUTER LOOP:
for item in items:
    # INNER LOOP:
    for size in sizes:
        print(f"{item} is available in {size}.")

# The outer loop moves by each list, index [0, 1]
# The inner loop processes every element for each list - [0][0], [0][1], [0][2]; [1][0], [1][1], [1][2]
# Inner loop runs fully before outer loop advances
# This demonstrates the “outer loop controls what, inner loop controls how many” principle


# --------------------------------------------------------
# 🟦 Helper: Nested Loop Trace Visualizer
# --------------------------------------------------------


def nested_loop_trace(outer, inner):
    """
    Visualizes nested loop execution by showing:
      - outer iteration index and value
      - inner iteration index and value
      - the resulting pair

    Mirrors the detailed iteration shown in Maven pp. 25–33.
    """
    print("NESTED LOOP TRACE\n------------------")
    
    for outer_idx, outer_val in enumerate(outer):
        print(f"\nOuter Iteration {outer_idx}: outer_val = {outer_val!r}")
        
        for inner_idx, inner_val in enumerate(inner):
            print(f"    Inner Iteration {inner_idx}: inner_val = {inner_val!r}")
            print(f"        → Pair = ({outer_val!r}, {inner_val!r})")

# Example usage:

nested_loop_trace(items, sizes)


# --------------------------------------------------------
# 🟦 Helper: Cartesian Product Generator
# --------------------------------------------------------
# (Useful for combinations, SKU creation, pairing simulations)

def cartesian_product(a, b):
    """
    Returns the Cartesian product of two iterables as a list of tuples.
    Equivalent to the nested-loop pairings in Maven's examples.
    """
    return [(x, y) for x in a for y in b]


# Example Usage:

cartesian_product(items, sizes)
````

````python
# --------------------------------------------------------
# Nested List Demo - Step-by-Step
# --------------------------------------------------------

item_reviews = [
    ["5 stars", "3 stars", "5 stars"],     # Skis
    ["4 stars", "4 stars"],                # Snowboards
    ["1 star", "4 stars"],                 # Goggles
    ["5 stars", "2 stars", "5 stars"]      # Boots
]


# --------------------------------------------------------
# Step 1. Outer list
# --------------------------------------------------------
# OUTER lOOP:
for lst in item_reviews:      # each nested list (4x)
    print(lst)
    
# --------------------------------------------------------
# Step 2. Inner list
# --------------------------------------------------------
# OUTER LOOP:
for lst in item_reviews:    # each nested list (4x)
    
    # INNER LOOP: 
    for review in lst:      # each element in each nested list (10x)
        print(review)

# --------------------------------------------------------
# Step 3. Extract Numeric Ratings Only
# --------------------------------------------------------
# OUTER LOOP:
for lst in item_reviews:             # each nested list (4x)
    
    # INNER LOOP: 
    for review in lst:               # each element in each nested list (10x)
        int_rating = int(review[0])  # extract numeric rating only
        print(int_rating)

# --------------------------------------------------------
# Step 4. Collect Numeric Ratings into Lists
# --------------------------------------------------------
# OUTER LOOP:
for lst in item_reviews:                     # each nested list (4x)
    integer_ratings = []                     # empty list accumulator: ratings in each nested list
    
    # INNER LOOP: 
    for review in lst:                       # each element in each nested list (10x)
        int_rating = int(review[0])          # extract numeric rating only
        integer_ratings.append(int_rating)   # .append() accumulator list
    print(integer_ratings)

# --------------------------------------------------------
# Step 5. Collect Numeric Ratings Back into a Nested List
# --------------------------------------------------------
numeric_item_reviews = []                           # empty list: collect numeric values back into a nested list

# OUTER LOOP:
for lst in item_reviews:                           # each nested list (4x)
    integer_ratings = []                           # empty list accumulator: ratings in each nested list
    
    # INNER LOOP: 
    for review in lst:                             # each element in each nested list (10x)
        int_rating = int(review[0])                # extract numeric rating only
        integer_ratings.append(int_rating)         # .append() accumulator list
    
    numeric_item_reviews.append(integer_ratings)   # Collect back into a nested list

numeric_item_reviews                               # return nested list with numeric values only


# --------------------------------------------------------
# Step 6. Final Pairing with Items
# --------------------------------------------------------
items = ["Skis", "Snowboards", "Goggles", "Boots"]

for item, ratings in zip(items, numeric_item_reviews):
    print(f"{item}: {ratings}")
````

````python
# --------------------------------------------------------
# Understand outer vs. inner loop roles
# Exercise: Nested Loops — Processing Reviews by Item
# --------------------------------------------------------
# Practice navigating nested lists
# Transform raw string data into numeric, analyzable structure
# Prepare data for aggregation (averages)


item_reviews = [
    ["5 stars", "3 stars", "5 stars"],     # Skis
    ["4 stars", "4 stars"],                # Snowboards
    ["1 star", "4 stars"],                 # Goggles
    ["5 stars", "2 stars", "5 stars"]      # Boots
]

item_reviews_numeric = []                  # outer loop accumulator

# OUTER LOOP:
for each_nested_lst in item_reviews:       # each nested list in item_reviews (4x)
    
    numeric_ratings = []                   # inner loop accumulator
    
    # INNER LOOP:
    for element in each_nested_lst:         # for element in nested list (10x)
        rating = int(element[0])            # extract numeric rating
        numeric_ratings.append(rating)
    
    # collect processed ratings for this item
    item_reviews_numeric.append(numeric_ratings)
    
    print(item_reviews_numeric)             # shows each iteration step-by-step

item_reviews_numeric

# The outer loop moves by each nested list, indices [0], [1], [2], [3]
# The inner loop processes every element for each nested list - [0][0], [0][1], [0][2]; [1][0], [1][1]; etc.
# The inner loop fully completes before the outer loop advances


# --------------------------------------------------------
# Aggregate results
# --------------------------------------------------------

# Calculate average review score per item
for ratings in item_reviews_numeric:
    average = sum(ratings) / len(ratings)
    print(round(average, 2))

# --------------------------------------------------------
# Pair results with item names
# --------------------------------------------------------

items = ["Skis", "Snowboards", "Goggles", "Boots"]

for i in range(len(items)):
    avg_rating = sum(item_reviews_numeric[i]) / len(item_reviews_numeric[i])
    print(f"{items[i]}: average rating = {avg_rating:.2f}")
````

````python
# --------------------------------------------------------
# Nested Loops Over Nested Lists
# Example — Apply 10% discount to each customer’s orders
# --------------------------------------------------------
# This pattern is used when dealing with multi-order customers, grouped transactions, or any list-of-lists structure.
# This preserves the **structure** of the original dataset — a common requirement in analytics workflows.


orders = [
    [120.00, 45.00, 89.99],        # customer 1 orders
    [250.00, 80.00],               # customer 2 orders
    [19.99, 55.00, 150.00, 40.00]  # customer 3 orders
]

discounted_orders = []                        # outer loop accumulator

for transactions in orders:                   # outer loop: each nested list (3x)
    customer_discounted = []                  # inner loop accumulator
    
    for subtotal in transactions:             # inner loop: for element in nested list (9x)
        new_total = round(subtotal * 0.9, 2)  # 10% discount (or 90% of the price)
        customer_discounted.append(new_total)
    
    discounted_orders.append(customer_discounted)

discounted_orders

# --------------------------------------------------------
# 🟦 Helper: Matrix Visualizer  
# --------------------------------------------------------

def visualize_matrix(matrix):
    """
    Prints a clean, aligned representation of a nested list
    to help visualize 2D structures created by nested loops.
    """
    print("MATRIX VIEW\n-----------")
    for row_idx, row in enumerate(matrix):
        print(f"Row {row_idx}: {row}")

# Example Usage:

visualize_matrix(discounted_orders)
print()


# --------------------------------------------------------
# 🟦 Helper: Nested List Transformer (General-Purpose)
# --------------------------------------------------------

orders = [
    [120.00, 45.00, 89.99],        # customer 1 orders
    [250.00, 80.00],               # customer 2 orders
    [19.99, 55.00, 150.00, 40.00]  # customer 3 orders
]

def transform_nested_list(nested, transform_fn):
    """
    Applies a user-defined transformation to each element in a nested list.

    Parameters
    ----------
    nested : list[list]
        A list of lists (e.g., customers → orders).
    
    transform_fn : callable
        Function applied to each inner element.

    Returns
    -------
    list[list]
        A new nested list with the same structure.

    Notes
    -----
    - General version of Maven's multi-order discount logic (pp. 32–33)
    - Works with any transformation: tax, normalization, scaling, flags, etc.
    """
    
    result = []                                  # outer loop accumulator
    
    for i, inner_list in enumerate(nested):      # outer loop
        transformed_inner = []                   # inner loop accumulatore
        
        for j, value in enumerate(inner_list):   # inner loop
            transformed_inner.append(transform_fn(value))
        
        result.append(transformed_inner)
    
    return result


# Example usage:
result = transform_nested_list(orders, lambda x: round(x * 0.9, 2))
# nothing will display unless you print
print(result)

result = transform_nested_list(orders, lambda x: round(x * 0.9, 2))
print("Discounted orders:", result)

# Jupyter Note:
# Only the final expression in a cell is displayed automatically.
# Functions return data silently unless print() is used.
````

```python
def transform_nested_list(nested, transform_fn):
    """
    Applies a user-defined transformation to each element in a nested list.

    Parameters
    ----------
    nested : list[list]
        A list of lists (e.g., customers → orders).
    
    transform_fn : callable
        Function applied to each inner element.

    Returns
    -------
    list[list]
        A new nested list with the same structure.

    Notes
    -----
    - General version of Maven's multi-order discount logic (pp. 32–33)
    - Works with any transformation: tax, normalization, scaling, flags, etc.
    """
    
    result = []
    
    for i, inner_list in enumerate(nested):      # outer loop
        transformed_inner = []
        
        for j, value in enumerate(inner_list):   # inner loop
            transformed_inner.append(transform_fn(value))
        
        result.append(transformed_inner)
    
    return result
```

# 8. Understanding Nested Loop Complexity  
Nested loops typically have **O(n × m)** time complexity.
Total operations: n × m → O(n × m)

Examples:

- Items (2) × Sizes (3) → 6 operations → O(2 × 3)  
- Customers (3) × Orders (vary 2–4) → total operations = sum of row lengths. (Count how many total elements exist across all inner lists).

- If the inner loop length is fixed → total work = n × m.
- If the inner loop length varies → total work = sum of inner list lengths.
 

Performance matters when:

- Processing large datasets  
- Performing multi-level simulations  
- Building grid-style computations  

---

# 9. When to Use Nested Loops

Use nested loops when:

- You must examine every pairing in a 2D structure  
- Your data is multi-level (e.g., list of lists)  
- You need all combinations of two sets  
- You are generating structured outputs (e.g., matrix, grid, report sections)

Avoid nested loops when:

- A vectorized or built-in approach (NumPy, pandas) is available  
- Complexity becomes too high for large datasets  
- Iteration order does not matter (sometimes flattening is simpler)

# Key Takeaways

- A **nested loop** executes the inner loop *once per iteration* of the outer loop.  
- This produces a complete pairing (Cartesian product) of both iterables.  
- Nested loops are essential for processing **nested lists**, **combinations**, and **matrix-like structures**.  
- The expanded toolkit includes:  
  - `nested_loop_trace()`  
  - `transform_nested_list()`

These tools give full transparency into multi-dimensional iteration.  
- Nested iteration can be powerful but costly; understand when it is necessary and when alternatives exist.
