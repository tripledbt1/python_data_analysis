## Section 7.01 — Loop Basics  

### Introduction
In Python, a **loop** is a controlled repetition structure: a block of code that executes multiple times until a stopping condition is met.  
For data analytics, loops let you process sequences, clean datasets, transform values, and automate iterative logic.

Python supports two primary loop types:

1. **For Loops**  
   Run a ***specific number of times***, typically once per element in an iterable such as a list, tuple, string, or range object.  
   Use a for loop when the iteration count is known in advance or when you want to examine or transform each element of a dataset.  
   For loops may iterate:
   - **directly over values** (item-based looping),  
   - **over indices** using `range()` when positional control or alignment across multiple sequences is required, or  
   - **over index–value pairs** using `enumerate()` when both position and clarity are needed.  
   *Run a specified number of times.*

2. **While Loops**  
   Run ***as long as a logical condition remains True***.  
   Use a while loop when you do **not** know in advance how many iterations are needed and the loop must respond to changing state.  
   *Run until a logical condition is met.*

This chapter focuses on building mastery for both loop types, integrating your **Helper Toolkit** to visualize execution steps, debug iterative logic, and strengthen intuition for real analytics workflows.

````python
# ------------------------------------------------------------------
# Understanding Loops Through Examples
# Example 1 — Rewriting a repetitive calculation using loops  
# ------------------------------------------------------------------


# Below is the non-loop approach: writing a separate line for each list element.

exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

# Inefficient repetition (not scalable)
euro_list = [
    round(usd_list[0] * exchange_rate, 2),
    round(usd_list[1] * exchange_rate, 2),
    round(usd_list[2] * exchange_rate, 2),
    round(usd_list[3] * exchange_rate, 2),
    round(usd_list[4] * exchange_rate, 2),
]

euro_list

# Problem if you have hundreds or thousands of prices, manually writing repeated expressions becomes impossible.


# ------------------------------------------------------------------
# Understanding Loops Through Examples
# Example 2 — Applying a for loop
# ------------------------------------------------------------------

# Generic syntax - `for` indicates for loop
iterable = "Dawn"
for i in iterable:
     print(i)

# Given Data:
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

euro_list = []                                    # Start with an empty list container

for price in usd_list:                            # for loop: element in iterable
    euro_value = round(price * exchange_rate, 2)  # assign variable: each element from the iterable is * by the loop-invariant scalar exchange_rate
    euro_list.append(euro_value)                  # .append() empty list container with each converted price
    
print()
print(euro_list, end="\n\n")


# This loop is conceptually equivalent to five function calls, one per list element.  
# But Python handles all iteration mechanics for you.
````

````python
# -------------------------------------------------------------
# Looping Mechanics
# How to use a single loop to derive multiple aligned outputs from one input sequence.
# -------------------------------------------------------------
# item-based for loops (no indices yet)
# deterministic iteration (known list length)
# accumulator lists
# stepwise transformation
# parallel output lists that stay aligned by construction

subtotals = [15.98, 899.97, 799.97, 117.96, 5.99, 599.99, 24.99, 1799.94, 99.99]
taxes = []
totals = []

for subtotal in subtotals:
    tax = round(subtotal * 0.08, 2)
    total = round(subtotal + tax, 2)
    taxes.append(tax)
    totals.append(total)

print(taxes)
print(totals)
````

## Section 7.02 — For Loops  

### Overview
A **for loop** iterates over an *iterable* (list, tuple, string, set, dictionary, or range).  
Python automatically handles indexing and sequence progression, allowing analysts to focus on transformation logic rather than iteration mechanics.

Use a for loop when:

- You know **how many iterations** you need.  
- You are iterating directly over each element in a dataset.  
- You want clear, predictable, index-safe iteration.  
- You want to avoid accidental infinite loops.

Syntax describes the standard pattern:

```
for item in iterable:
    do_something_with(item)
```

Python internally executes:

1. `item = iterable[0]`  
2. `item = iterable[1]`  
3. `item = iterable[2]`  
n. until the iterable is exhausted.

---

## 1. Looping Over Items  
This is the **most Pythonic** and most common pattern in data analytics.

### Example — Print each letter in a string  

```python
word = "Hello"

for letter in word:
    print(letter)
```

Internally this becomes:

- `letter = 'H'`  
- `letter = 'e'`  
- `letter = 'l'`  
- `letter = 'l'`  
- `letter = 'o'`


````python
# -------------------------------------
# 1. Looping Over Items
# -------------------------------------
# This is the most Pythonic and most common pattern in data analytics.
# Use when:
# - you only need the value
# - position does not matter

# Item-Based Loop Pattern:

items = "Hello"
for i in items:
    print(i)
print()


# 🟦 Helper Expansion — `enumerate_items()`

def enumerate_items(iterable):
    """
    Diagnostic helper: prints each item as Python would assign it in a for loop.
    Mirrors the step-by-step iteration shown in Maven pages 7–9.
    """
    print("TRACE: Looping Over Items\n--------------------------")
    for idx, item in enumerate(iterable):
        print(f"Iteration {idx}: item = {item!r}")
        

# Example usage:

enumerate_items("Hello")
````

````python
# ----------------------------------------------------
# Looping Over Items to Build New Lists
# ----------------------------------------------------
# This approach scales to thousands of values with no change in code size.

# Example — USD → EUR conversion

exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

euro_list = []                     # Start container

for price in usd_list:             # Iterates 5 times
    euro_value = round(price * exchange_rate, 2)
    euro_list.append(euro_value)   # .append() list container

euro_list
````

````python
# ---------------------------------------
# 🟦 Helper Expansion - `loop_trace_for()`
# ---------------------------------------

# Given data:
exchange_rate = 0.88                          # exchange_rate → loop-invariant scalar parameter
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]  # usd_list      → iterable dataset

# 🟦 Helper for visualization
def loop_trace_for(iterable):
    print("TRACE: For Loop Execution\n-------------------------")
    for iteration, item in enumerate(iterable):    # for loop: element in dataset, enumerate(usd_list) returns pairs (index, value)
        print(f"Iteration {iteration}: item = {item!r}")

# Call `loop_trace_for()` Helper
loop_trace_for(usd_list)

# Actual loop performing the conversion
euro_list = []                                     # create empty list container
for price in usd_list:                             # for loop: element in iterable
    euro_value = round(price * exchange_rate, 2)   # assign variable: each element from the iterable is * by the loop-invariant scalar exchange_rate
    euro_list.append(euro_value)                   # .append() empty list container

print("\nConverted EUR prices:", euro_list)
````

````python
# --------------------------------------------------------
# 🟦 Helper Expansion - `deep_loop_state_visualiser(,)`
# --------------------------------------------------------

# Given data:
exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

# 🟦 Helper for visualization
def deep_loop_state_visualizer(usd_list, exchange_rate):
                                                       # usd_list      → iterable dataset
                                                       # exchange_rate → loop-invariant scalar parameter
    """
    Shows detailed iteration state for a currency conversion loop.
    Reveals:
      - iteration number
      - current input value
      - computed euro_value
      - euro_list after each step
    """
    euro_list = []                                    # create empty list container
    
    print("DEEP LOOP STATE VISUALIZER\n--------------------------")
    
    for iteration, price in enumerate(usd_list):      # for loop: element in dataset, enumerate(usd_list) returns pairs (index, value)
        euro_value = round(price * exchange_rate, 2)  # Assign variable: each element from the iterable is * by the loop-invariant scalar exchange_rate
        euro_list.append(euro_value)                  # Append this iteration’s computed euro_value to the accumulator list euro_list
        
        print(f"Iteration {iteration}:")              # iteration index from enumerate()
        print(f"  price       = {price!r}")           # value retrieved from the iterable at this iteration
        print(f"  euro_value  = {euro_value!r}")      # variable holding the results that change with each iteration
        print(f"  euro_list   = {euro_list}\n")       # display the state of the accumulator list after each iteration.
    print("Final Output:", euro_list)
    return euro_list

# Call `deep_loop_state_visualizer()` Helper
deep_loop_state_visualizer(usd_list, exchange_rate)
````

````python
# ----------------------------------------------------
# Looping Over Indices
# ----------------------------------------------------
# Index-based iteration is appropriate when:
# - You need to access multiple lists in parallel.  
# - You want both the index and the element.  
# - You are transforming nested lists or multi-dimensional data.


# Example — Access elements by index:

usd_list = [5.99, 9.99, 19.99, 24.99, 99.99]

for i in range(len(usd_list)):     # i = 0 → 4
    print(i, usd_list[i])
print()


# 🟦 Helper Expansion — `range_index_map()`
# Integrates your Chapter 6 Range Visualizer into loop logic.


def range_index_map(iterable):
    """
    Shows how Python maps a range to indices for a given iterable.
    Mirrors Maven's step-through on page 10.
    """
    print("RANGE → INDEX MAP\n------------------")
    for idx in range(len(iterable)):
        print(f"range index {idx} → iterable[{idx}] = {iterable[idx]!r}")


# Usage:

range_index_map(usd_list)
````

````python
# ---------------------------------------------
# Indexs-based: Looping Over Multiple Iterables  
# ----------------------------------------------
# When two lists share a one-to-one alignment (e.g., item ↔ price), indexing allows synchronized access:

# Example — Print item and its euro price
# This produces 5 synchronized pairs

items = ["Snowboard", "Boots", "Helmet", "Goggles", "Bindings"]
euro_prices = [5.27, 8.79, 17.59, 21.99, 87.99]

for i in range(len(items)):
    print(items[i], euro_prices[i])


# 🟦 Helper — `sync_lists()`
# Synchronized iteration across multiple equal-length iterables

def sync_lists(*lists):
    """
    Prints synchronized items across multiple lists of equal length.
    Excellent for diagnostics in multi-iterable looping.
    """
    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
    length = len(lists[0])
    for iterable in lists:
        if len(iterable) != length:
            raise ValueError("All lists must have the same length for synchronized iteration.")
    
    for idx in range(length):
        row = [lst[idx] for lst in lists]
        print(f"Index {idx}: {row}")

# Usage:
sync_lists(items, euro_prices)
````

````python
# -------------------------------------------------------------------------
# looping using `enumerate()
# Clean, safe way to access both the index and the value during iteration
# -------------------------------------------------------------------------
#`enumerate()` Loop (Recommended When Index Is Needed)
# Basic `enumerate()` pattern:

items = ["Snowboard", "Boots", "Helmet"]

for index, element in enumerate(items):
    print(index, element)

# Output:
# 0 Snowboard
# 1 Boots
# 2 Helmet

# Alternatively:
for i, item in enumerate(items):
    print(i, item)

# Use when:
# - you want the index for labeling, tracing, or logic
# - you want clarity and safety
# - you want to avoid manual index bookkeeping


# ----------------------------------------------------
# 🔁 Applied Example — Index-Based Loop 🆚 `enumerate()`
# ----------------------------------------------------
# Index-Based Version:

exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99]
euro_list = []

for price in range(len(usd_list)):
    euro_value = round(usd_list[price] * exchange_rate, 2)
    euro_list.append(euro_value)

print(euro_list)


# `enumerate()` Version (Cleaner)

exchange_rate = 0.88
usd_list = [5.99, 9.99, 19.99]
euro_list = []

for i, price in enumerate(usd_list):
    euro_value = round(price * exchange_rate, 2)
    euro_list.append(euro_value)

print(euro_list)
````

```python
def sync_lists(*lists):

    Excellent for diagnostics in multi-iterable looping.

    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
```

### Overview — Using `enumerate()`
In Python, `enumerate()` provides a clean and safe way to access **both the index and the value** during iteration.  
It resolves the main drawbacks of index-based looping while preserving positional awareness when needed.

Conceptually, `enumerate()` represents the **next evolutionary step** after learning:

1. Item-based for loops (simple, readable)  
2. Index-based for loops (positional control, but fragile)  

`enumerate()` combines the strengths of both approaches.

---

## Why `enumerate()` Exists

Recall the index-based loop pattern:

```python
for i in range(len(items)):
    print(items[i])
```

While correct, this pattern requires you to:
- manually manage indices  
- ensure correct bounds (`len(...)`)  
- mentally track both index and value  

`enumerate()` removes this overhead by yielding **index–value pairs directly**.

---

## Basic `enumerate()` Pattern

```python
items = ["Snowboard", "Boots", "Helmet"]

for index, item in enumerate(items):
    print(index, item)
```

### Output
```
0 Snowboard
1 Boots
2 Helmet
```

### Key Idea
- `index` is automatically generated  
- `item` is the value from the iterable  
- No manual indexing is required

## Comparing Three Looping Styles

### 1. Item-Based Loop (Most Common)

```python
for item in items:
    print(item)
```

Use when:
- you only need the value  
- position does not matter  

---

### 2. Index-Based Loop (Positional Control, Higher Risk)

```python
for i in range(len(items)):
    print(i, items[i])
```

Use when:
- you must reference multiple lists by position  
- you need write-access to a list by index  

⚠️ Requires careful alignment.

---

### 3. `enumerate()` Loop (Recommended When Index Is Needed)

```python
for i, item in enumerate(items):
    print(i, item)
```

Use when:
- you want the index for labeling, tracing, or logic  
- you want clarity and safety  
- you want to avoid manual index bookkeeping

## When to Prefer `enumerate()`

Use `enumerate()` when:
- you need index information for tracing or labeling  
- you want positional awareness without index arithmetic  
- readability and debuggability matter  
- teaching or explaining iteration logic  

Avoid `enumerate()` when:
- you only need the values (simple item-based loop is best)  
- you must mutate a list by index (explicit index-based loop may be required)  

### Related Guidance
- If all iterables are equal peers → use `zip()`  
- If one iterable defines the iteration → use `enumerate()`
- `enumerate()` yields `(index, value)` pairs automatically  
- It eliminates the need for `range(len(...))` in most cases  
- It improves clarity, safety, and debuggability  
- Index-based loops still matter when mutating data or coordinating multiple structures

# 6. Choosing Between Items, Indices, `enumerate()`, and `zip()`

| Pattern                              | Who defines iteration? | What is yielded per loop? |
| ------------------------------------ | ---------------------- | ------------------------- |
| `for item in iterable`               | iterable               | `item`                    |
| `for i in range(len(iterable))`      | iterable               | `i` (int)                 |
| `for i, item in enumerate(iterable)` | iterable               | `(i, item)` tuple         |
| `for a, b in zip(iter1, iter2)`      | **both equally**       | `(a, b)` tuple            |


| Approach                                | When to Use                                                                      | Benefits                            |
| --------------------------------------- | -------------------------------------------------------------------------------- | ----------------------------------- |
| `for item in iterable:`                 | Transforming or inspecting values                                                | Clean, safest, Pythonic             |
| `for i in range(len(iterable)):`        | When **one iterable controls iteration** and index access is explicitly required | Maximum positional control          |
| `for idx, item in enumerate(iterable):` | When you need **both index and value** from a single iterable                    | Best practice, readable, efficient  |
| `for a, b in zip(iter1, iter2):`        | When **multiple iterables are equal peers** and should be processed together     | Safest alignment, avoids index bugs |

- **For loops** execute once per element in an iterable; the iteration count is predictable.

# Chapter 07 — Loops
# Section 7.03 — While Loops

### Preview
A **while loop** repeats a block of code *as long as a logical expression evaluates to True*.  
This contrasts with a **for loop**, which repeats a fixed number of times.

Use a while loop when:

- You **do not know** how many iterations will be required.  
- You are simulating a process to completion (e.g., reach a threshold, exhaust inventory).  
- You need iterative refinement until a condition stabilizes (e.g., convergence in algorithms).  

## While loops repeat based on state, not count

Structure can be summarized as:
```Python
while condition_on_state:
    update_state
```

Execution continues until the expression becomes **False**.  
If the condition never turns False, you enter an **infinite loop**, which we will guard against.

This section extends your Helper Toolkit with a **While Loop Safety System**, simulation helpers, and state-tracing tools.

---

# When Should You Use Each Loop Type?

| Scenario | Best Choice | Rationale |
|---------|-------------|-----------|
| You know the dataset length | **For Loop** | One iteration per element; safe, predictable |
| You want to repeat until a condition is false | **While Loop** | Logic-driven iteration |
| You need to prevent infinite loops | **For Loop** (or While Loop with counters) | Counters act as emergency brakes |
| You are iterating over sequences (list, tuple, string) | **For Loop** | Pythonic and safe |
| You need a “repeat until solved” pattern (e.g., simulation) | **While Loop** | Flexible stopping criteria |

---

# 1. Basic While Loop Pattern

A while loop must include **three things**:

1. An **initial state**  
2. A **logical condition** that inspects that state  
3. An **update step** that *moves the state toward termination*  

### Example — Counting upward

```python
counter = 0                     # initial state

while counter < 5:              # logical condition
    print(counter)
    counter += 1                # update step
```

Python evaluates:

- Iteration 0 → counter = 0  
- Iteration 1 → counter = 1  
- Iteration 2 → counter = 2  
- Iteration 3 → counter = 3  
- Iteration 4 → counter = 4  
- Iteration stops when counter becomes 5, making (counter < 5) False.

Without the update (`counter += 1`), the loop would run forever.


````python
# --------------------------------------------------------
# Demonstrating a While Loop Pattern Safely with Counters
# --------------------------------------------------------
# Simple while loop that counts upward until a threshold is reached
# While loops run until a given logical expression becomes FALSE


counter = 0
while counter < 5:      # Loop continues WHILE this expression is TRUE
    counter += 1        # Without this update, the loop becomes infinite ⚠️
    print("Counter =", counter)  
print()


# --------------------------------------------------------
# While Loop: Inventory until stock runs out
# --------------------------------------------------------

current_inventory = 686
month_counter = 0
monthly_sales = 84

while current_inventory > 0:
    current_inventory -= monthly_sales          # subtraction assignment
    month_counter += 1
    print(
        f"At month-end {month_counter}: "
        f"My inventory balance is {current_inventory}"
    )
print()


# --------------------------------------------------------
# While Loop: Calculating bank balance until money runs out
# --------------------------------------------------------

bank_balance = 7_189
month_counter = 0
spending = 350

while bank_balance > 0:
    bank_balance -= spending          # subtraction assignment
    month_counter += 1
    print(
        f"At month-end {month_counter}: "
        f"My balance is ${round(bank_balance, 2)}"
    )
````

````python
# --------------------------------------------------------
# While Loops with Counters
# Counters track loop progress, prevent runaway execution, 
# and can act as secondary safety conditions
# --------------------------------------------------------

# Example — Growing a portfolio until it reaches 1 million
# Loop terminates when the portfolio crosses the threshold.


portfolio = 800_000
year = 0

while portfolio < 1_000_000:            # primary logical condition
    investment_income = portfolio * 0.05
    portfolio += investment_income       # update state
    year += 1                             # update counter
    print(f"End of year {year}: ${round(portfolio, 2)}")



# --------------------------------------------------------
# 🟦 Advanced Helper: State Timeline Recorder
# This helper logs every state value for later visualization or tabular analysis.
# --------------------------------------------------------


def while_timeline(start_state, condition_fn, update_fn, max_iter=100):
    """
    Records each iteration's state—ideal for simulations, 
    convergence testing, and algorithm diagnostics.
    """
    timeline = []
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        timeline.append((iteration, state))
        
        iteration += 1
        if iteration > max_iter:
            timeline.append(("ABORTED", state))
            break
        
        state = update_fn(state)
    
    timeline.append(("FINAL", state))
    return timeline


# Example

timeline = while_timeline(
    start_state=800000,
    condition_fn=lambda s: s < 1_000_000,
    update_fn=lambda s: s * 1.05,
    max_iter=10
)

timeline

# Produces a list of iteration → state pairs, perfect for later DataFrame conversion.
````

````python
# --------------------------------------------------------
# 🔁 While Loops — Condition-Driven Iteration 
# 🆚 For Loops - Count-Driven, does not mutate data
# --------------------------------------------------------

# --------------------------------------------------------
# While Loops — Condition-Driven Iteration
# --------------------------------------------------------
# NOTE:
# This example mutates the list in place to demonstrate while-loop logic ⚠️
# In real data analysis workflows, you usually avoid mutating lists directly
# Instead, you would use filtering or list comprehensions to create a new list


customer_ratings = [
    "5 stars", "3 stars", "5 stars", "4 stars", "4 stars",
    "1 star", "4 stars", "5 stars", "2 stars", "5 stars"
]

onestar_counter = 0

while "1 star" in customer_ratings:
    customer_ratings.remove("1 star")
    onestar_counter += 1

print(f"Removed {onestar_counter} one-star review(s)")
print(customer_ratings)

# --------------------------------------------------------
# For Loops — Count-Driven, does not mutate data
# --------------------------------------------------------
# TEACHING TAKEAWAY:
# While loops are useful when iteration depends on changing state.
# For loops are preferred when data can be processed in a single pass.
# In analytics, for loops (or vectorized operations) are usually clearer and safer.


customer_ratings = [
    "5 stars", "3 stars", "5 stars", "4 stars", "4 stars",
    "1 star", "4 stars", "5 stars", "2 stars", "5 stars"
]

onestar_counter = 0
filtered_ratings = []              # empty list container

for rating in customer_ratings:   # loop looks at every element in customer_ratings
    if rating == "1 star":        # if block counts "1 star" and does nothing else with it
        onestar_counter += 1
    else:                         # else block handles everything that is not "1 star"
        filtered_ratings.append(rating)

print(f"Removed {onestar_counter} one-star review(s)")
print(filtered_ratings)
print(customer_ratings)           # does not mutate data in-place
````

````python
# ------------------------------------------------------------
# 🟦 Helper: `safe_while_trace()` with Infinite-Loop Protection
# - state value at each iteration  
# - when/why the loop stops  
# - detection of missing or incorrect update logic  
# - protection against accidental infinite loops via `max_iter`
# -------------------------------------------------------------
````

````python
stock_portfolio = 165000                        # start state is 165000

while stock_portfolio < 3_500_000:              # condition is s < 3_500_000
    stock_portfolio += stock_portfolio * 0.14   # update step is s * 0.13


def safe_while_trace(start_state, condition_fn, update_fn, max_iter=50):
    """
    Safely traces a while loop with a guard to prevent infinite loops.
    """
    print("SAFE WHILE TRACE\n-----------------")
    
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        print(f"Iteration {iteration}: state = {state!r}")
        
        iteration += 1
        if iteration > max_iter:
            print("ABORTED: Possible infinite loop (max_iter exceeded).")
            break
        
        state = update_fn(state)
    
    print(f"Final state: {state!r}")
    return state
print()

# Function call:
safe_while_trace(
    start_state=165000,
    condition_fn=lambda s: s < 3_500_000,
    update_fn=lambda s: s + (s * 0.14),
    max_iter=50
)
````

````python
# --------------------------------------------------------
# Exercise — Demonstrating a While Loop
````
````python
# Example Roth IRA — Age + Monthly vs Annual Contribution Timing
# --------------------------------------------------------

from datetime import date

# -----------------------------
# Fixed personal anchors
# -----------------------------
BIRTH_YEAR = 1967
AS_OF = date(2025, 12, 14)   # reporting date each year (Dec 14)

# -----------------------------
# Roth IRA inputs  🧠✨
# -----------------------------
start_balance = 21_077
annual_contribution = 8_000
annual_return_rate = 0.14
goal = 145_000

# -----------------------------
# Simulator with age tracking
# timing = "begin" → contribution first (monthly-style timing)
# timing = "end"   → contribution last (annual-style timing)
# -----------------------------
def simulate_roth_with_age(start_balance, annual_contribution, annual_return_rate, goal, timing="begin"):
    balance = start_balance
    years = 0
    rows = []

    while balance < goal:
        years += 1
        year_end = date(AS_OF.year + years, AS_OF.month, AS_OF.day)
        age = year_end.year - BIRTH_YEAR  # Dec 14 is after May 7

        if timing == "begin":
            balance += annual_contribution
            balance += balance * annual_return_rate
        elif timing == "end":
            balance += balance * annual_return_rate
            balance += annual_contribution
        else:
            raise ValueError("timing must be 'begin' or 'end'")

        rows.append((year_end, age, balance))

    return years, rows

# -----------------------------
# Run both scenarios
# -----------------------------
years_begin, rows_begin = simulate_roth_with_age(
    start_balance, annual_contribution, annual_return_rate, goal, timing="begin"
)

years_end, rows_end = simulate_roth_with_age(
    start_balance, annual_contribution, annual_return_rate, goal, timing="end"
)

# -----------------------------
# Output
# -----------------------------
print(f"As of {AS_OF}: age = {AS_OF.year - BIRTH_YEAR}\n")

print("MONTHLY-style contributions (begin-of-year timing):")
for year_end, age, bal in rows_begin:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_begin} years.\n")

print("ANNUAL contributions (end-of-year timing):")
for year_end, age, bal in rows_end:
    print(f"{year_end} | Age {age} | Balance ${bal:,.2f}")
print(f"\nGoal reached in {years_end} years.\n")

diff = years_end - years_begin
print(
    f"Difference: monthly-style timing reaches the goal "
    f"{abs(diff)} year(s) {'sooner' if diff > 0 else 'later' if diff < 0 else 'at the same time'}."
)
````

````python
from datetime import date

# --------------------------------------------------------
# Two-Phase 401(k) Simulator — Monthly Contributions
# Phase 1: high contributions for 8 years (96 months)
# Phase 2: reduced contributions until target reached 🚀🔥
# --------------------------------------------------------

# Personal anchors
BIRTH_YEAR  = 1967
BIRTH_MONTH = 5
BIRTH_DAY   = 7

AS_OF = date(2025, 12, 14)  # anchor reporting to the 14th of each month

def add_months(d, months):
    """Add months to a date (safe here because day=14 exists in every month)."""
    y, m = d.year, d.month
    m0 = (m - 1) + months
    new_y = y + (m0 // 12)
    new_m = (m0 % 12) + 1
    return date(new_y, new_m, d.day)

def age_on(d, birth_year=BIRTH_YEAR, birth_month=BIRTH_MONTH, birth_day=BIRTH_DAY):
    """Compute age at a given date."""
    age = d.year - birth_year
    if (d.month, d.day) < (birth_month, birth_day):
        age -= 1
    return age

# -----------------------------
# Scenario inputs
# -----------------------------
start_balance = 164_658.86
annual_return_rate = 0.14
target_balance = 3_000_000

# Phase 1: through 12/14/2033 (8 years = 96 months)
phase1_months = 96
monthly_contrib_phase1 = (32_500 / 12) + (6_900 / 12)   # employee + match

# Phase 2: after 12/14/2033
monthly_contrib_phase2 = (7_200 / 12) + (7_200 / 12)    # employee + match

# Convert annual to monthly (compound-consistent)
monthly_return_rate = (1 + annual_return_rate) ** (1/12) - 1

# -----------------------------
# Simulation (monthly steps)
# -----------------------------
balance = start_balance
month_counter = 0

print(f"As of {AS_OF}: Age {age_on(AS_OF)} | Start balance ${balance:,.2f}\n")

while balance < target_balance:
    month_counter += 1
    current_date = add_months(AS_OF, month_counter)

    # Switch contributions after 96 months (12/14/2033)
    if month_counter <= phase1_months:
        monthly_contribution = monthly_contrib_phase1
        phase_label = "Phase 1 (high contrib)"
    else:
        monthly_contribution = monthly_contrib_phase2
        phase_label = "Phase 2 (reduced contrib)"

    # Contribution first, then growth (monthly cadence)
    balance += monthly_contribution
    balance += balance * monthly_return_rate

    # Print yearly checkpoints + final month
    if month_counter % 12 == 0 or balance >= target_balance:
        print(
            f"{current_date} | Age {age_on(current_date)} | "
            f"{phase_label} | Balance ${balance:,.2f}"
        )

print(f"\nGoal reached in {month_counter} months (~{month_counter/12:.1f} years) by {current_date}.")
````

# Chapter 07 — Loops
# Section 7.03 — While Loops

### Preview
A **while loop** repeats a block of code *as long as a logical expression evaluates to True*.  
This contrasts with a **for loop**, which repeats a fixed number of times.

Use a while loop when:

- You **do not know** how many iterations will be required.  
- You are simulating a process to completion (e.g., reach a threshold, exhaust inventory).  
- You need iterative refinement until a condition stabilizes (e.g., convergence in algorithms).  

## While loops repeat based on state, not count

Structure can be summarized as:
```Python
while condition_on_state:
    update_state
```

Execution continues until the expression becomes **False**.  
If the condition never turns False, you enter an **infinite loop**, which we will guard against.

This section extends your Helper Toolkit with a **While Loop Safety System**, simulation helpers, and state-tracing tools.

---

# When Should You Use Each Loop Type?

| Scenario | Best Choice | Rationale |
|---------|-------------|-----------|
| You know the dataset length | **For Loop** | One iteration per element; safe, predictable |
| You want to repeat until a condition is false | **While Loop** | Logic-driven iteration |
| You need to prevent infinite loops | **For Loop** (or While Loop with counters) | Counters act as emergency brakes |
| You are iterating over sequences (list, tuple, string) | **For Loop** | Pythonic and safe |
| You need a “repeat until solved” pattern (e.g., simulation) | **While Loop** | Flexible stopping criteria |

---

# 1. Basic While Loop Pattern  

A while loop must include **three things**:

1. An **initial state**  
2. A **logical condition** that inspects that state  
3. An **update step** that *moves the state toward termination*  

### Example — Counting upward

```python
counter = 0                     # initial state

while counter < 5:              # logical condition
    print(counter)
    counter += 1                # update step
```

Python evaluates:

- Iteration 0 → counter = 0  
- Iteration 1 → counter = 1  
- Iteration 2 → counter = 2  
- Iteration 3 → counter = 3  
- Iteration 4 → counter = 4  
- Iteration stops when counter becomes 5, making (counter < 5) False.

Without the update (`counter += 1`), the loop would run forever.

---

# 2. While Loops with Counters

Counters track loop progress, prevent runaway execution, and can act as **secondary safety conditions**.

### Example — Growing a portfolio until it reaches 1 million

```python
portfolio = 800_000
year = 0

while portfolio < 1_000_000:            # primary logical condition
    investment_income = portfolio * 0.05
    portfolio += investment_income       # update state
    year += 1                             # update counter
    print(f"End of year {year}: ${round(portfolio, 2)}")
```
 
Loop terminates when the portfolio crosses the threshold.

# 3. 🟦 Helper Expansion: Safe While Loop Tracer

This upgrades your earlier helper into a reusable diagnostic tool for notebook-based explorations.

```python
def safe_while_trace(start_state, condition_fn, update_fn, max_iter=50):
    """
    A safe, diagnostic tracer for while loops.
    Prevents accidental infinite loops and makes state transitions visible.

    Parameters
    ----------
    start_state : any
        Initial loop state.
    
    condition_fn : callable(current_state) -> bool
        Determines whether the loop should continue.
    
    update_fn : callable(current_state) -> new_state
        Defines how the state changes each iteration.
    
    max_iter : int
        Safety guard to prevent infinite looping.
    """
    
    print("SAFE WHILE TRACE\n-----------------")
    
    state = start_state
    iteration = 0
    
    while condition_fn(state):
        print(f"Iteration {iteration}: state = {state!r}")
        
        iteration += 1
        if iteration > max_iter:
            print("ABORTED: Possible infinite loop (max_iter exceeded).")
            break
        
        state = update_fn(state)
    
    print(f"Final state: {state!r}")
    return state
```

```python
    update_fn=lambda s: s + s * 0.05,   # grow by 5%
    max_iter=10
)
```

# 5. Infinite Loops and Error Prevention  

A while loop becomes infinite when:

- The logical condition is always True  
- The update step never moves state toward termination  
- Logic depends on an impossible or stalled signal  
- The update step is omitted (common beginner error)

### Demonstration of incorrect logic  
*(Do NOT run this pattern without safeguards)*

```python

#     print(x)      # x never changes -> infinite loop
```

# 8. When to Choose a While Loop

Use a **while loop** when:

- Termination is based on *reaching a threshold*, not a known count.  
- You are performing **search**, **waiting**, or **simulation until solved**.  
- You want to protect against infinite loops using counters or safety guards.  
- You need maximum flexibility in defining stopping logic.

Use a **for loop** when:

- You know the dataset length.  
- You want guaranteed finite iteration.  
- You are looping over user-defined collections.

---

# Key Takeaways

- A **while loop** repeats until its logical expression becomes False.  
- It requires **state initialization**, **condition**, and **update** to avoid runaway execution.  
- While loops are ideal for **simulations, threshold-based tasks, and unknown-duration processes**.  
- Infinite loops occur when the condition never becomes False; always include a **state update** or **safety counter**.  
- The new helpers —  
  - `safe_while_trace()`  
  - `while_timeline()`  
provide controlled exploration of while-loop behavior for learning and debugging.

```python
def validate_alignment(*iterables):
    """
    Validates that all provided iterables have the same length.

    This helper is intended to catch silent logic errors
    when using index-based looping across multiple iterables.

    Raises
    ------
    ValueError
        If the iterables are not aligned in length.
    """
    if not iterables:

    if len(set(lengths)) != 1:
```

```python
def sync_lists(*iterables):
    """
    Prints synchronized elements across multiple iterables
    that are assumed to be positionally aligned.

    This helper is intended for diagnostics and learning,
    not for production workflows.

    Raises
    ------
    ValueError
        If the iterables do not all have the same length.
    """
    print("SYNCHRONIZED ITERABLE TRACE\n---------------------------")
    
    if not iterables:

    length = len(iterables[0])
    
    for it in iterables:
        if len(it) != length:

    for idx in range(length):
```

## 1. Basic Nested Loop Structure

```
for item in outer_list:
    for inner_item in inner_list:
        do_something(item, inner_item)
```

Execution pattern:

- Outer iteration 0 → inner loop runs fully  
- Outer iteration 1 → inner loop runs fully  
- …  
- Outer iteration n → inner loop runs fully  

Total iterations =

````python
# --------------------------------------------------------
# Nested List Example — Item × Size Combinations
# --------------------------------------------------------
# Output count:
# - 2 items  
# - 3 sizes  
# - 2 × 3 = 6 combinations

items = ["Snowboard", "Boots"]
sizes = ["small", "medium", "large"]

# OUTER LOOP:
for item in items:
    # INNER LOOP:
    for size in sizes:
        print(f"{item} is available in {size}.")

# The outer loop moves by each list, index [0, 1]
# The inner loop processes every element for each list - [0][0], [0][1], [0][2]; [1][0], [1][1], [1][2]
# Inner loop runs fully before outer loop advances
# This demonstrates the “outer loop controls what, inner loop controls how many” principle


# --------------------------------------------------------
# 🟦 Helper: Nested Loop Trace Visualizer
# --------------------------------------------------------


def nested_loop_trace(outer, inner):
    """
    Visualizes nested loop execution by showing:
      - outer iteration index and value
      - inner iteration index and value
      - the resulting pair

    Mirrors the detailed iteration shown in Maven pp. 25–33.
    """
    print("NESTED LOOP TRACE\n------------------")
    
    for outer_idx, outer_val in enumerate(outer):
        print(f"\nOuter Iteration {outer_idx}: outer_val = {outer_val!r}")
        
        for inner_idx, inner_val in enumerate(inner):
            print(f"    Inner Iteration {inner_idx}: inner_val = {inner_val!r}")
            print(f"        → Pair = ({outer_val!r}, {inner_val!r})")

# Example usage:

nested_loop_trace(items, sizes)


# --------------------------------------------------------
# 🟦 Helper: Cartesian Product Generator
# --------------------------------------------------------
# (Useful for combinations, SKU creation, pairing simulations)

def cartesian_product(a, b):
    """
    Returns the Cartesian product of two iterables as a list of tuples.
    Equivalent to the nested-loop pairings in Maven's examples.
    """
    return [(x, y) for x in a for y in b]


# Example Usage:

cartesian_product(items, sizes)
````

# 8. Understanding Nested Loop Complexity  
Nested loops typically have **O(n × m)** time complexity.
Total operations: n × m → O(n × m)

Examples:

- Items (2) × Sizes (3) → 6 operations → O(2 × 3)  
- Customers (3) × Orders (vary 2–4) → total operations = sum of row lengths. (Count how many total elements exist across all inner lists).

- If the inner loop length is fixed → total work = n × m.
- If the inner loop length varies → total work = sum of inner list lengths.
 

Performance matters when:

- Processing large datasets  
- Performing multi-level simulations  
- Building grid-style computations

# 9. When to Use Nested Loops

Use nested loops when:

- You must examine every pairing in a 2D structure  
- Your data is multi-level (e.g., list of lists)  
- You need all combinations of two sets  
- You are generating structured outputs (e.g., matrix, grid, report sections)

Avoid nested loops when:

- A vectorized or built-in approach (NumPy, pandas) is available  
- Complexity becomes too high for large datasets  
- Iteration order does not matter (sometimes flattening is simpler)

````python
# ------------------------------------------------------------------
# 1. The `break` Statement
# `break` stops the loop entirely, regardless of the iteration index
# ------------------------------------------------------------------
# Behavior
# - Loop ends as soon as the condition triggers  
# - Remaining items are not processed  


# Example — Stop when revenue exceeds 2000  

subtotals = [200.98, 350.97, 500, 400, 300, 600, 150, 100, 90]

revenue = 0                          # accumulator to track what the loop produces

for amount in subtotals:
    revenue += amount                # looping through subtotals list and adding revenue
    print(round(revenue,2))          # Visualize each loop
    if revenue > 2000:
        print("Target reached. Stopping early.")
        break

print("Final revenue:", revenue)


# -------------------------------------
# 🟦 Helper — Break Trace Visualizer
# -------------------------------------

def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0
    for idx, value in enumerate(iterable):
        running += value
        print(f"Iteration {idx}: value = {value}, running = {running}")
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break

# Example Usage:
break_trace(subtotals, lambda total: total > 2000)


# Use :.2f formatting in helpers when working with currency or financial data.
# This improves readability and prevents confusion caused by floating-point precision.


# -------------------------------------
# 🟦 Helper — Break Trace Visualizer
# -------------------------------------

def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    Values are formatted to two decimal places for currency-style data.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0.0
    for idx, value in enumerate(iterable):
        running += value
        print(
            f"Iteration {idx}: "
            f"value = ${value:.2f}, "
            f"running = ${running:.2f}"
        )
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break
            
# Example Usage:
break_trace(subtotals, lambda total: total > 2000)
````

````python
# ------------------------------------------------------------------
# 1. The `break` Statement — Early Exit from a While Loop
# ------------------------------------------------------------------
# `break` immediately stops a loop, even if the loop condition
# would otherwise continue to evaluate as True.
#
# Use `break` when:
# - you want to stop early once a secondary condition is met
# - continuing the loop no longer makes sense
# - you want to prevent runaway or unrealistic scenarios
# ------------------------------------------------------------------


stock_portfolio = 21_070   # accumulator: tracks the evolving portfolio value (loop state)
year_counter = 0          # counter: tracks how many iterations (years) have passed


# This loop continues WHILE the portfolio is below the target
while stock_portfolio < 150_000:
    
    # Calculate annual investment growth (15% return)
    investment_income = stock_portfolio * 0.15
    
    # Update portfolio value (state change)
    stock_portfolio += investment_income           # accumulator update
    
    # Update loop counter
    year_counter += 1                             # counter update
    
    # Observe the evolving state of the system
    print(
        f"My balance is ${stock_portfolio:,.2f} "
        f"in year {year_counter}"
    )
    
    # SAFETY CHECK:
    # If 10 years have passed and the goal is still not reached,
    # exit the loop early instead of continuing indefinitely.
    if year_counter >= 10:
        print("Goal not reached in 10 years — savings rate may be too low.")
        break

# Teaching takeaway:
# - The while-condition controls whether the loop *can* continue
# - The `break` controls whether the loop *should* continue
````

````python
# ------------------------------------------------------------------
# 2. The `continue` Statement  
# `continue` skips the remainder of the current iteration but continues the loop
# ------------------------------------------------------------------
# Behavior
# - `"skis"` and `"skiboots"` are skipped  
# - All other items processed normally
# ------------------------------------------------------------------

# Example — Filter out items containing “ski”


inventory = ["snowboard", "skis", "helmet", "skiboots", "bindings"]
filtered_inventory = []

for item in inventory:
    # filter condition: skip the current item if it contains the substring "ski"
    if "ski" in item:
        # continue means: "do NOT run the rest of the loop body for this item"
        # and move to the next iteration immediately
        continue
    filtered_inventory.append(item)  # .append() empty list accumulator

print(filtered_inventory)
print()


# ---------------------------------------
# 🟦 Helper — Continue Trace Visualizer
# ---------------------------------------

def continue_trace(iterable, skip_fn):
    """
    Prints all iterations and highlights which ones are skipped via 'continue'.
    """
    print("CONTINUE TRACE\n----------------")
    
    for idx, item in enumerate(iterable):
        if skip_fn(item):
            print(f"Iteration {idx}: SKIPPED ({item!r})")
            continue
        
        print(f"Iteration {idx}: processed {item!r}")


# Example Usage:
continue_trace(inventory, lambda x: "ski" in x)
````

````python
# ------------------------------------------------------------------
# The `pass` Statement
# `pass` is a placeholder that does NOTHING, but keeps the code valid.
# ------------------------------------------------------------------
# Key idea:
# - `pass` satisfies Python’s requirement that a block cannot be empty
# - It does NOT skip an iteration (that’s `continue`)
# - It does NOT stop a loop (that’s `break`)
# - It simply says: “I’ll fill this in later”
#
# Common use cases:
# - Stubbing logic during development
# - Sketching control flow before implementation
# - Keeping loops and conditionals syntactically valid
# - Incrementally building complex pipelines
# ------------------------------------------------------------------


products = ["Snowboard", "Boots", "Helmet"]

for p in products:
    if p == "Boots":
        pass    # Placeholder: no action yet, loop continues normally
    else:
        print("Processing:", p)


# NOTE:
# `pass` is not control flow.
# It is a structural placeholder.
#

# CONTROL FLOW MENTAL MODEL:
# - break     → exit the loop entirely
# - continue  → skip an iteration
# - pass      → do nothing (for now), but keep structure valid
````

````python
# ------------------------------------------------------------------
# Error Handling with `try` / `except`  
# These statements allow the loop to continue running after encountering an error
# ------------------------------------------------------------------
# Notes
# - Strings like `"74.99"` are converted successfully  
# - `0` triggers `ZeroDivisionError`  
# - `None` triggers `TypeError`  
# - Loop does not stop on errors  
# ------------------------------------------------------------------

# ---------------------------------------
# Example 1 — Problematic price list
# ---------------------------------------

price_list = [99.99, "74.99", 0, None, 20]

clean_prices = []

for element in price_list:
    try:
        # Step 1: convert to a numeric value
        numeric_price = float(element)
        
        # Step 2: use the validated value for calculations
        clean_prices.append(round(1 / numeric_price, 4))
    
    except ZeroDivisionError:                                  # zero division error
        print("ZeroDivisionError: price was zero. Skipping.")
    
    except (TypeError, ValueError):                            # multiple error types
        print(f"Invalid price format ({element}). Skipping.")

# ---------------------------------------
# 🟦 Helper — Error Trace Loop Inspector  
# Shows which items cause which errors.
# ---------------------------------------

def try_except_trace(iterable, transform_fn):
    """
    Executes transform_fn on each element, catching errors
    and reporting them with full iteration context.
    """
    print("TRY/EXCEPT TRACE\n-----------------")
    results = []
    
    for idx, value in enumerate(iterable):
        try:
            result = transform_fn(value)
            results.append(result)
            print(f"Iteration {idx}: OK → {result}")
        
        except Exception as e:
            print(f"Iteration {idx}: ERROR → {type(e).__name__}: {e}")
    
    return results


# Example Usage:
try_except_trace(price_list, lambda x: round(1 / float(x), 4))
print()

# ---------------------------------------
# Example 2 — Problematic ratings list
# ---------------------------------------

ratings = ["1 star", "5 star", 
           "3 star", "star 5", 
           "4 star"
]

for rating in ratings:
    try:
        print(int(rating[0]))
    except ValueError:
        print("incorrect data") 



# STUDENT NOTE:
# This helper shows that errors do NOT stop loops by default
# when handled with try/except.
#
# Each iteration is attempted independently:
# - Valid values are transformed and collected
# - Invalid values are reported and skipped
#
# This mirrors real-world data cleaning workflows.
````

````python
# ------------------------------------------------------------------
# Error Handling with `try` / `except`  
# These statements allow the loop to continue running after encountering an error
# ------------------------------------------------------------------
# Notes
# - Strings like `"74.99"` are converted successfully  
# - `0` triggers `ZeroDivisionError`  
# - `None` triggers `TypeError`  
# - Loop does not stop on errors  
# ------------------------------------------------------------------

# ---------------------------------------
# Example 3 — Problematic price list
# ---------------------------------------


prices = [5.99, 99.99, "74.99", 0, None, 20, 24.99]
budget = 50

for price in prices:
    if price is None:
        continue

    try:
        numeric_price = float(price)  # convert early (handles strings)

        if numeric_price == 0:
            print("This product is free.")
            continue

        affordable_quantity = budget // numeric_price
        print(f"I can buy {int(affordable_quantity)} of these.")

    except ValueError:
        print("That is not a number.")
    except TypeError:
        # TypeError is unlikely once you skip None, but included for completeness
        print("Invalid type for price.")


# ---------------------------------------
# 🟦 Helper — Robust Affordability Calculator  
# ---------------------------------------

def affordability_filter(prices, budget):
    """
    Returns a list of (idx, raw_price, numeric_price, quantity) for prices that can be evaluated.
    Skips None, non-numeric, and handles zero safely.
    """
    results = []

    for idx, price in enumerate(prices):
        if price is None:
            continue

        try:
            numeric_price = float(price)

            if numeric_price == 0:
                results.append((idx, price, numeric_price, "FREE"))
                continue

            qty = int(budget // numeric_price)
            results.append((idx, price, numeric_price, qty))

        except (ValueError, TypeError):
            results.append((idx, price, None, "INVALID"))

    return results


# Example Usage:
rows = affordability_filter(prices, budget)

for idx, raw, num, qty in rows:
    print(f"Index {idx}: raw={raw!r} | numeric={num!r} | qty={qty}")



# ---------------------------------------
# 🟦 Helper — Error Trace Loop Inspector  
# Shows which items cause which errors.
# ---------------------------------------

def try_except_trace(iterable, transform_fn):
    """
    Executes transform_fn on each element, catching errors
    and reporting them with full iteration context.
    """
    print("TRY/EXCEPT TRACE\n-----------------")
    results = []
    
    for idx, value in enumerate(iterable):
        try:
            result = transform_fn(value)
            results.append(result)
            print(f"Iteration {idx}: OK → {result}")
        
        except Exception as e:
            print(f"Iteration {idx}: ERROR → {type(e).__name__}: {e}")
    
    return results


# Example Usage:
try_except_trace(
    prices,
    lambda p: "FREE" if float(p) == 0 else int(budget // float(p))
)
````

````python
# --------------------------------------------------------
# ✅ Analysis-Ready Affordability Toolkit
# - robust parsing (None, strings, bad values)
# - flags FREE / INVALID
# - computes quantity you can buy under a budget
# - returns both detailed rows + a summary dictionary
# --------------------------------------------------------

def affordability_analyze(prices, budget):
    """
    Analyze a mixed-quality price list against a budget.

    Returns
    -------
    rows : list[dict]
        One record per input element with diagnostics:
        idx, raw, numeric_price, status, quantity
    summary : dict
        Counts + quick stats to support analysis decisions.
    """
    rows = []

    # Summary counters
    counts = {
        "OK": 0,
        "FREE": 0,
        "INVALID": 0,
        "SKIPPED_NONE": 0
    }

    valid_prices = []
    valid_quantities = []

    for idx, raw in enumerate(prices):
        # 1) Skip None cleanly (common missing-data pattern)
        if raw is None:
            counts["SKIPPED_NONE"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": None,
                "status": "SKIPPED_NONE",
                "quantity": None
            })
            continue

        # 2) Attempt numeric conversion (handles floats/ints/strings like "74.99")
        try:
            numeric_price = float(raw)
        except (TypeError, ValueError):
            counts["INVALID"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": None,
                "status": "INVALID",
                "quantity": None
            })
            continue

        # 3) Handle zero explicitly (avoid division-by-zero)
        if numeric_price == 0:
            counts["FREE"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": 0.0,
                "status": "FREE",
                "quantity": "FREE"
            })
            continue

        # 4) Negative prices are “valid numbers” but typically invalid business data
        if numeric_price < 0:
            counts["INVALID"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": numeric_price,
                "status": "INVALID_NEGATIVE",
                "quantity": None
            })
            continue

        # 5) OK path: compute how many you can buy (floor division)
        qty = int(budget // numeric_price)

        counts["OK"] += 1
        valid_prices.append(numeric_price)
        valid_quantities.append(qty)

        rows.append({
            "idx": idx,
            "raw": raw,
            "numeric_price": numeric_price,
            "status": "OK",
            "quantity": qty
        })

    # -----------------------------
    # Build summary stats
    # -----------------------------
    summary = {
        "budget": budget,
        "total_inputs": len(prices),
        "counts": counts,
        "valid_price_count": len(valid_prices),
        "min_valid_price": min(valid_prices) if valid_prices else None,
        "max_valid_price": max(valid_prices) if valid_prices else None,
        "avg_valid_price": (sum(valid_prices) / len(valid_prices)) if valid_prices else None,
        "avg_quantity": (sum(valid_quantities) / len(valid_quantities)) if valid_quantities else None,
        "max_quantity": max(valid_quantities) if valid_quantities else None,
    }

    return rows, summary


# --------------------------------------------------------
# ✅ Pretty-printer (optional, keeps your notebook readable)
# --------------------------------------------------------

def print_affordability_report(rows, summary):
    print("AFFORDABILITY REPORT\n--------------------")
    print(f"Budget: ${summary['budget']:,.2f}")
    print(f"Total inputs: {summary['total_inputs']}\n")

    # High-level summary
    c = summary["counts"]
    print("Counts:")
    for k, v in c.items():
        print(f"  {k:>12}: {v}")
    print()

    # Quick stats
    if summary["valid_price_count"] > 0:
        print("Valid price stats:")
        print(f"  min price : ${summary['min_valid_price']:,.2f}")
        print(f"  max price : ${summary['max_valid_price']:,.2f}")
        print(f"  avg price : ${summary['avg_valid_price']:,.2f}")
        print(f"  avg qty   : {summary['avg_quantity']:.2f}")
        print(f"  max qty   : {summary['max_quantity']}")
    else:
        print("No valid prices found.")
    print()

    # Row-by-row detail (traceability)
    print("Row detail:")
    for r in rows:
        idx = r["idx"]
        raw = r["raw"]
        num = r["numeric_price"]
        status = r["status"]
        qty = r["quantity"]

        # consistent formatting
        raw_txt = repr(raw)
        num_txt = "None" if num is None else f"{num:.2f}"
        print(f"  idx {idx:>2}: raw={raw_txt:<10} | numeric={num_txt:>6} | status={status:<15} | qty={qty}")


# --------------------------------------------------------
# ✅ Example usage (your dataset)
# --------------------------------------------------------

prices = [5.99, 99.99, "74.99", 0, None, 20, 24.99]
budget = 50

rows, summary = affordability_analyze(prices, budget)
print_affordability_report(rows, summary)
````

# Chapter 07 — Loops  
## Section 7.05 — Loop Control

### Overview
Loop control statements allow you to **modify the normal flow of iteration**, enabling early exits, selective skipping, placeholder logic, and robust error handling.

Four essential loop control mechanisms include:

- **break** — exit a loop immediately  
- **continue** — skip the rest of the current iteration  
- **pass** — placeholder that does nothing  
- **try / except** — handle errors without stopping the loop  

These tools are critical for:

- Data cleaning  
- Error-tolerant ETL pipelines  
- Complex filtering  
- Avoiding infinite loops  
- Creating maintainable, incremental logic  

This section adds new helper functions for debugging, error tracing, and controlled loop execution.

---

# 1. The `break` Statement  

`break` **stops the loop entirely**, regardless of the iteration index.

### Example — Stop when revenue exceeds 2000  
(Maven scenario: revenue becomes high enough before list ends)

```python
subtotals = [200, 350, 500, 400, 300, 600, 150, 100, 90]

revenue = 0

for amount in subtotals:
    revenue += amount
    
    if revenue > 2000:
        print("Target reached. Stopping early.")
        break

print("Final revenue:", revenue)
```

### Behavior
- Loop ends as soon as the condition triggers  
- Remaining items are **not** processed  

### 🟦 Helper — Break Trace Visualizer

```python
def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.

        running += value
        print(f"Iteration {idx}: value = {value}, running = {running}")
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break
```

# When to Use Loop Control Statements

### Use `break` when:
- A target or threshold is reached  
- Further processing is unnecessary  
- You are preventing a runaway or infinite loops early  

### Use `continue` when:
- Filtering datasets  
- Skipping invalid or unwanted entries, i.e., excluding values you don't want to process in a loop
- Differentiating “skippable” vs “processable” values

### Use `pass` when:
- Structuring logic before implementation, i.e., avoiding run errors with incomplete code logic
- Writing placeholder blocks, i.e., serving as a placeholder for future code  
- Maintaining syntactic validity  

### Use `try`/`except` when:
- Data may be incomplete, inconsistent, or malformed  
- You need to avoid interruption during iteration
- You need help with error and exception handling  
- Building robust ETL or cleaning pipelines

# Key Takeaways

- `break` stops the loop immediately.  
- `continue` skips the current iteration only.  
- `pass` does nothing; used as a placeholder.  
- `try`/`except` creates fault-tolerant loops that keep running even when data issues occur.  
- The helper toolkit now includes:  
  - `break_trace()`  
  - `continue_trace()`  
  - `try_except_trace()`  
  - `affordability_filter()`  
These provide full transparency and robustness when working with real-world, messy datasets.
