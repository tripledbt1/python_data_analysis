# 06.60 Overview

A **range** is a very special type of sequence in Python.  
Unlike lists or tuples, a range does **not** store all values in memory.  
Instead, it **generates** values on demand using a start, stop, and step.

Ranges are:

- memory-efficient  
- fast  
- commonly used in loops (next chapter!)  
- ideal for generating clean sequences of integers  
- invaluable for indexing, batching, and slicing

# 06.69 Summary (Part 5 — Ranges)

- `range()` creates a memory-efficient sequence of integers.  
- Ranges are defined by start, stop, and step values.  
- The stop value is **non-inclusive**, just like slicing.  
- Converting to `list()` or `tuple()` reveals the actual generated values.
- Although, any range object can be converted into any other sequence type that accepts an iterable.
- Ranges are ideal for loops, counting, indexing, batching, and generating IDs.  
- Negative step values allow backward ranges.  
- Ranges prepare you directly for Chapter 07 (Loops), where they are used constantly.


````python
# 06.61 Creating Ranges
# Ranges are created using the built-in `range()` function.

# The full form is: range(start, stop, step)

# - **start** (optional): where the sequence begins  
# - **stop**: where the sequence ends (non-inclusive)  
# - **step** (optional): how much to increment each time  


# Basic range: 0 to 4
r1 = range(5)

# Range with start and stop
r2 = range(2, 7)

# Range with start, stop, and step
r3 = range(0, 20, 5)

print(list(r1))   # [0, 1, 2, 3, 4]
print(list(r2))   # [2, 3, 4, 5, 6]
print(list(r3))   # [0, 5, 10, 15]

# ASCII mini-diagram

# range(2, 7)
# start → 2
# stop  → 7   (not included)
# values → [2, 3, 4, 5, 6]

# This mirrors list slicing behavior, where the stop value is also non-inclusive.
````

````python
# ---------------------------------------------------------
# 06.62 Ranges Do Not Directly Display Their Values
# ---------------------------------------------------------

# When printed, a range shows its definition, not the actual integers:
print(range(5), end="\n\n")  # prints object parameters only,"range(0, 5)"


# --------------------------------------------------------------------
# To retrieve and see the actual values, convert to a list or a tuple:
# --------------------------------------------------------------------

print(list(range(5)))    # [0, 1, 2, 3, 4]
print(tuple(range(5)))   # (0, 1, 2, 3, 4)

# -------------------------------------------------------------------------------
# Visualization example: show the index positions of a string
# This is purely for demonstration; not common in real code to turn into a tuple
# -------------------------------------------------------------------------------

print("\n", tuple(range(len('Hello'))))

# Breakdown: generating all valid indices of the string, then turning them into a tuple.
# len('Hello') → 5
# range(5) → range(0, 5)
# tuple(range(5)) → (0, 1, 2, 3, 4)
# print() to display

print("\n", list(range(len('Hello'))))
````

````python
# 06.63 Negative Steps and Reversed Ranges

r = range(10, 0, -2)
print(list(r))
# [10, 8, 6, 4, 2]


# ASCII diagram

range(10, 0, -2)
# start = 10
# stop = 0  (excluded)
# step = -2
# values → [10, 8, 6, 4, 2]

# Backward ranges are extremely helpful when:

# - stepping backward through indices  
# - reverse-iterating through sequences  
# - building countdown timers  
# - analyzing time-based data in reverse order
````

````python
# 🔧 Helper: `show_range_index_map(rng, show_negative=True)`


def show_range_index_map(rng, show_negative=True):
    """
    Print a table of indices and values for a Python range object.

    Parameters
    ----------
    rng : range
        The range object to inspect.
    show_negative : bool, default True
        If True, show the negative index for each element
        (as if the range were a sequence like a list or tuple).

    Usage
    -----
    show_range_index_map(range(5))
    show_range_index_map(range(10, 0, -2))
    """
    # ---------------------------------------------------------
    # 1. Validate input type
    # ---------------------------------------------------------
    if not isinstance(rng, range):
        raise TypeError(
            f"show_range_index_map expects a range, got {type(rng).__name__}"
        )

    # ---------------------------------------------------------
    # 2. Basic range properties
    # ---------------------------------------------------------
    n = len(rng)
    start = rng.start
    stop = rng.stop
    step = rng.step

    print(f"Range: range({start}, {stop}, {step})")
    print(f"Length: {n}")
    print()

    if n == 0:
        print("This range is empty (no values generated).")
        return

    # ---------------------------------------------------------
    # 3. Build header
    # ---------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    header_cols.append("value")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "value" in header_cols:
        header += f"{'value':>6}"

    print(header)
    print("-" * max(len(header), 20))

    # ---------------------------------------------------------
    # 4. Iterate over the range and print index/value mapping
    # ---------------------------------------------------------
    for i, val in enumerate(rng):
        # Negative index follows standard Python convention:
        # last element → -1, second to last → -2, etc.
        neg_index = i - n if show_negative else None

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        row += f"{val:>6}"
        print(row)


show_range_index_map(range(10, 0, -2))
show_range_index_map(range(20))               # forward range
show_range_index_map(range(50, 20, -5))       # backward range
show_range_index_map(range(6), show_negative=False)

example_range = range(10, 0, -2)
show_range_index_map(example_range)
````

````python
# 06.64 Practical Range Patterns
# Common real-world range uses:

# All even numbers from 0 to 20
evens = list(range(0, 21, 2))
print(evens)

# All odd numbers from 1 to 20
odds = list(range(1, 21, 2))
print(odds)

# First 10 integer IDs starting at 100
ids = list(range(100, 111))
print(ids)

# Ranges are the foundation for constructing loops, counters, and index sequences.
````

## 📌 Example 1 — Simple Forward Range

```python
r = range(0, 10, 2)
show_range_index_map(r)
```

Expected output (conceptually):

```
Range: range(0, 10, 2)
Length: 5

idx   neg  value
----------------
  0   -5      0
  1   -4      2
  2   -3      4
  3   -2      6
  4   -1      8
```

This shows:

- the **indices** (`0` to `4`)  
- the corresponding **values** (`0, 2, 4, 6, 8`)  
- the **negative indices** (`-5` to `-1`)  

Exactly as if you had:

```python
list(r) == [0, 2, 4, 6, 8]
```

## 📌 Example 2 — Backward (Negative Step) Range

```python
r = range(10, 0, -2)
show_range_index_map(r)
```

Conceptual output:

```
Range: range(10, 0, -2)
Length: 5

idx   neg  value
----------------
  0   -5     10
  1   -4      8
  2   -3      6
  3   -2      4
  4   -1      2
```

Key points:

- `start = 10`  
- `stop = 0` is **excluded**  
- `step = -2` walks backward  
- Values are `[10, 8, 6, 4, 2]`  
- Indices still run `0..4`, negative indices `-5..-1`  

This helper makes **negative-step ranges** much easier to reason about.
