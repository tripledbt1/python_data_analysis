# 🐍 Chapter 5. Conditional Logic: 
## 🧩 Nested If Statements (Part 3)
 
### Expanding control flow with deeper, multi-layered logic  

Nested `if` statements allow you to evaluate **a second condition only if a first condition is already true**.  
They add an extra “layer” of logic and are useful when your business rules depend on two steps of decision-making.

---

# 1. Why Use Nested If Statements?

Nested logic is appropriate when:

- You need to **check Condition B only when Condition A is true**
- Rules follow a **hierarchy** or **decision tree**
- The second decision would be *nonsensical* unless the first is satisfied  
  (e.g., “If item is out of stock, then check whether you want to offer a display model.”)

### 🔍 Visual Model
A nested structure behaves like this:

```
if condition_A:
    if condition_B:
        # Nested True block
    else:
        # Nested False block
else:
    # Outer False block
```

This is like saying:
> “Only if A is true should I bother checking B.”

---

# 2. Simple Example

```python
customer = "Chris"
inventory = 0

if inventory <= 0:                     # Outer If condition
    if customer == "Chris":            # Nested If condition
        print("You can have the display model, sir.")
    else:
        print("Out of stock.")         # Nested Else code
else:
    print("Item available!")
```

### Why this is nested:
- It makes sense to check the **customer’s name** ONLY after confirming that inventory is zero.
- The business rule is **hierarchical**, not parallel.

---

# 3. The Same Logic *Without* Nesting

From page 3 of the PDF, Maven shows a “without nesting” version.  
Here’s how it looks conceptually:

```
if customer == "Chris" and inventory <= 0:
    ...
```

This works… **but loses hierarchy clarity**.

Nested version makes it obvious:
> “We only ask about Chris because inventory is zero.”

Flat version makes it look like the two conditions are equally important.

Which should you use?  
➡ Use **nesting** when your logic flows like:  
**First A → then maybe check B**

➡ Use **flat logic** when your rule is truly:  
**A AND B must both be true with equal weight**

---

# 4. Real-World Example: Return Authentication

Nested ifs shine in real workflows:

```python
is_logged_in = True
is_admin = False

if is_logged_in:
    if is_admin:
        print("Welcome, admin.")
    else:
        print("Welcome, user.")
else:
    print("Please log in.")
```

This mirrors a real application:
- First confirm login  
- THEN decide admin/user behavior

## 10. Key Takeaways

- **Boolean values (`True` / `False`)** are the foundation of all conditional logic.
- **Nested conditional logic** creates multi-layered decision trees for real-world rules.
- Use **nesting** when conditions naturally follow a **“first this, then that”** hierarchy.
- **Boolean operators** (`and`, `or`, `not`), **membership tests**, and **comparisons** help you construct expressive rule systems.
- **Good ordering** and **clear hierarchy** prevent bugs — especially in menu logic, pricing rules, and authentication workflows.
- Debugging helpers like **`trace_decision()`** and **`trace_condition()`** offer deep visibility into execution flow.

## 🧭 Helper Selection Guide (Updated with Power Combo)

| Your Goal                                        | Best Helper(s)                                  |
|--------------------------------------------------|--------------------------------------------------|
| See **which branch** executed                    | `trace_branch` or `trace_decision`               |
| Understand **why** a branch executed             | `trace_condition` + `trace_decision`             |
| Debug incorrect **branch ordering**              | `trace_condition` + `trace_decision`             |
| Quick, simple **if/else** check                  | `trace_if`                                       |
| Analyze **multi-branch** logic                   | `trace_decision`                                 |
| Teach or inspect **Boolean logic**               | `trace_condition`                               |
| **Full visibility for complex decisions**        | ⭐ `trace_branch` + `trace_condition` + `trace_decision` ⭐ |

# --- Helper 1: trace_decision --------------------------------
def trace_decision(label, func):
    """
    Run a decision function that contains if/elif/else,
    print a clear header, and let the function's own prints
    explain which branch executed.
    
    label: Short description of what is being traced.
    func:  Zero-argument function containing the decision logic.
    """
    print("\n" + "=" * 60)
    print(f"DECISION TRACE: {label}")
    print("=" * 60)
    func()


# --- Helper 2: trace_condition -------------------------------
def trace_condition(name, value):
    """
    Print the name of a Boolean condition and its result (True/False).
    Useful for debugging and teaching how conditions evaluate.
    
    name:  String label for the condition being checked.
    value: Boolean expression result.
    """
    print(f"{name:30} → {value}")


# --- Helper 3: trace_branch ----------------------------------
def trace_branch(label):
    """
    Print a simple message showing which branch was taken.
    Use inside if / elif / else blocks.
    
    label: Text describing the branch.
    """
    print(f"→ Branch taken: {label}")


# --- Helper 4: trace_if --------------------------------------
def trace_if(description, condition, if_true, if_false):
    """
    Compact helper for simple if/else decisions.
    
    description: Description of the decision.
    condition:   Boolean expression.
    if_true:     Message printed when condition is True.
    if_false:    Message printed when condition is False.
    """
    print(f"\nChecking: {description}")
    print(f"Condition result: {condition}")
    
    if condition:
        print(f"Branch selected: {if_true}")
    else:
        print(f"Branch selected: {if_false}")
