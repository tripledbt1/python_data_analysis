````python
# 06.11 Built-In Functions That Work With Lists

# Python includes general-purpose functions that work on many iterable types, including lists.

# Example list
scores = [88, 92, 79, 93, 85]

print(len(scores))   # number of elements
print(sum(scores))   # total of the elements
print(min(scores))   # smallest value
print(max(scores))   # largest value

# Why these matter
# - `len()` is used constantly in loops, pandas, and analytics.  
# - `sum()` is fundamental for aggregations.  
# - `min()` and `max()` help validate data and find outliers.
# - `min()` and `max()` help validate data and find outliers.
# - dividing `sum()` and `len() gives average, e.g., avg = round(sum(revenue) / len(revenue), 2)
````

````python
# 06.12 Adding Elements: append() and insert()

# Two main ways to add new elements:

# 1. append(element) - Adds to the end of the list.
# Mutates the list in place ⚠️

items = ['Hat', 'Gloves', 'Scarf']
items.append('Coat')

print(items)  # display mutated list


# 2. insert(index, element) 
# Inserts at a specific position.
# Mutates the list in place ⚠️ 

items = ['Hat', 'Gloves', 'Scarf']
items.insert(3, 'Coat')   # insert at index 3 (fourth position)

print(items, end="\n\n")   # display mutated list


# ==================================================
# 06.12b Changing List Elements (Direct Assignment)
# ==================================================
# Replace items using indexing
# Mutates the list in place ⚠️

new_items = ["Coat", "Backpack", "Scarf"]

# Replace the second item
new_items[1] = "Gloves"

print(new_items)   # display mutated list
````

````python
# 06.13 Combining & Repeating Lists with arithmetic operators

# Lists can be merged (concatenated) with `+` and repeated with `*`

list1 = ['A', 'B']
list2 = ['C', 'D']

combined = list1 + list2
print(combined)      # ['A', 'B', 'C', 'D']

repeated = list1 * 3
print(repeated)      # ['A', 'B', 'A', 'B', 'A', 'B']
````

````python
# 06.14 Removing Elements: del and remove()
# Two main ways to delete elements:

# 1. del removes items by index or slice
# Mutates the list in place ⚠️ 

# del list_name[index:slice]
list_name = ['Hat', 'Gloves', 'Scarf', 'Coat', 'Backpack']
del list_name[0:3]     # deletes indices 0 through 2

print(list_name, "\n")      # ['Coat', 'Backpack']


# 2. remove(value) 
# Removes the first occurrence of a value only ⚠️
# Mutates the list in place ⚠️

colors = ['red', 'blue', 'green', 'blue']
colors.remove('blue')

print(colors)        # ['red', 'green', 'blue']
````

````python
# 06.15 Sorting Lists: .sort() vs sorted()
# This is a **critical** concept in Python.

# .sort() method
# Mutates the list in place ⚠️

values = [5, 2, 8, 1]
values.sort()

print(values, end="\n\n")   # [1, 2, 5, 8]


# sorted() function - returns a new sorted list  
# Does not mutate the original

values = [5, 2, 8, 1]
new_values = sorted(values)

print(new_values, "\n")    # [1, 2, 5, 8]
print(values)              # [5, 2, 8, 1]  # original unchanged
````

````python
# 06.16 Reversing Lists: .reverse() and slicing

# Two ways:

# 1. reverse() method
# Mutates the list in place ⚠️

values = [10, 20, 30]

values.reverse()

print(f"Reverse method: {values} (mutated list)")    # [30, 20, 10]
print()

# 2. Slicing - returns a new reversed list
# Does not mutate original.

values = [10, 20, 30]
rev = values[::-1]      # negative slicing won't change underlying list

print("Reverse slicing:", rev)             # [30, 20, 10]
print(f"Original unchanged: {values}\n")   # original unchanged

print(f"Reversed: {values[::-1]}\nOriginal is still: {values}\n")
print("Type of first element:", type(values[0]))  # <class 'int'>
print()

print(values, end="\n\n")
````

````python
# 06.17 Counting & Finding Values
# If the value does not exist, Python raises a **ValueError**.


# .count(value) - Counts how many times a value appears.
letters = ['a', 'b', 'a', 'c', 'a']
print(letters.count('a'))     # 3


# .index(value) - Returns the first index where the value appears.
letters = ['a', 'b', 'a']
print(letters.index('a'))     # 0


# 1. Find where a specific item appears
fruits = ['apple', 'banana', 'cherry', 'apple']

first_apple_pos = fruits.index('apple')        # 0
print(first_apple_pos)

last_apple_pos = fruits.index('apple', 1)     # start searching at index 1 → 3
print(last_apple_pos)


# 2. Reorder columns by moving one column next to another
columns = ['date', 'customer_id', 'product', 'revenue', 'region']

# Move 'region' right after 'product'
pos = columns.index('product')
print(columns.index('product'))  # 2

columns.insert(pos + 1, columns.pop(columns.index('region')))  # ['date', 'customer_id', 'product', 'region', 'revenue']
print(columns)

# ⚠️ **Caution:** ⚠️
# Technically correct, but not recommended in real analytics pipelines.  
# This pattern mutates the list twice (.insert and .pop) and relies on multiple `.index()` lookups.  
# Missing or duplicated column names can cause crashes or silent data corruption.

# Safe Option 1: Reorder columns by building a new list explicitly
columns = ['date', 'customer_id', 'product', 'revenue', 'region']

new_order = [
    'date',
    'customer_id',
    'product',
    'region',
    'revenue'
]

orig_columns = columns[:]  # make a full copy of the original list
columns = new_order 

# Safe Option 2: List comprehension to reorder safely
desired_order = ['date', 'customer_id', 'product', 'region', 'revenue']
columns = [col for col in desired_order if col in columns]
# List comprehension reorder does not mutate — but it replaces which can be dangerous if the original structure matters.


# 🔬 Long Version of list comprehension (What Python Does Internally)
# Python expands the one-liner into:

orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

temp_list = []
for col in desired_order:
    if col in columns:
        temp_list.append(col)

columns = temp_list

# Same logic — the comprehension is just more compact and efficient.


# Safe Option 3: In pandas (safest)
# df = df[['date', 'customer_id', 'product', 'region', 'revenue']]
````

````python
# 06.18 Business Example (student-friendly)

# Weekly revenue data (in USD)
revenue = [1200, 1500, 1600, 900, 2000, 1800, 1700]

# 1. Total weekly revenue
total = sum(revenue)
print(f"Total weekly revenue: ${total:,.2f}")

# 2. Best day
best = max(revenue)
print(f"Best daily revenue: ${best:,.2f}")

# 3. Worst day
worst = min(revenue)
print(f"Worst daily revenue: ${worst:,.2f}")

# 4. Average revenue
avg = sum(revenue) / len(revenue)
print(f"Average revenue: ${avg:,.2f}")

# 5. Check if any day was above 1900
above_1900 = any(r > 1900 for r in revenue)      # any() function checks if any element in an iterable evaluates to True.


print(total, best, worst, avg, above_1900)

# This type of list analysis is foundational for:
# - financial analytics  
# - retail metrics  
# - operational dashboards  
# - time series preprocessing  
````

# 06.15 Sorting Lists: .sort() vs sorted()

This is a **critical** concept in Python.

### **.sort() method**
- Sorts *in place*  
- Permanently changes the list  
- Returns None  

```python
values = [5, 2, 8, 1]
values.sort()

print(values)    # [1, 2, 5, 8]
```

### **sorted() function**
- Returns a *new* sorted list  
- Leaves the original unchanged  

```python
values = [5, 2, 8, 1]
new_values = sorted(values)

print(values)       # [5, 2, 8, 1]
print(new_values)   # [1, 2, 5, 8]
```

### Best practice  
> Do not sort in place until you are sure you no longer need the original data.

This is especially true in analytics where the original ordering matters.

values = [10, 20, 30]

## 🧠 Summary

### ✔ `values[::-1]`  
- Does *not* mutate original  
- Returns a new reversed list  
- Safe for analysis  
- Best for beginners and data workflows

# 📦 Helper Box: How to Read Any List Comprehension (3-Step Method)

List comprehensions are one of Python’s most powerful and expressive features.  
They compress a full loop + condition + append operation into a clear, readable one-liner.

This helper teaches you how to **decode** any list comprehension using a simple 3-step process, and includes a **Safety Upgrade** showing how data professionals prevent accidental loss of original data.

---

## 🔍 General Pattern

A list comprehension always follows this structure:

```python
[new_item   for item in iterable   if condition]
```

It has **three components**:

1. **Output expression** → what goes into the new list  
2. **Iteration** → where items come from  
3. **Condition (optional)** → keep only items that pass the test  

---

# 🟦 Apply the 3-Step Method to Your Example

Your comprehension:

```python
[col for col in desired_order if col in columns]
```

### Step 1 — “For each…”  
`for col in desired_order`  
→ Loop through each column name in `desired_order`.

### Step 2 — “If…”  
`if col in columns`  
→ Only keep columns that actually exist in the current dataset.

### Step 3 — “Put into the new list…”  
`col`  
→ Place the column name into the new list as-is.

### ✔️ Final meaning (plain English)

> “Rebuild the columns list in the order of `desired_order`, but keep only those columns that appear in the existing `columns` list.”

---

# 🔬 The Long Version (What Python Does Internally)

```python
orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

temp_list = []
for col in desired_order:
    if col in columns:
        temp_list.append(col)

columns = temp_list
```

This is exactly what the comprehension does behind the scenes.

- It **does not mutate** the original list object.  
- It **does overwrite** the variable `columns` so the previous ordering is lost unless backed up.

---

# 🛡️ **Safety Upgrade: How Professionals Protect Original Data**

The long version ends with:

```python
columns = temp_list
```

This **reassigns** the variable `columns`, meaning the original ordering is gone unless saved.

In analytics, ETL, and MSBA contexts, you never want to lose original schema information.

### ✔️ Safe, production-style pattern

```python
orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

columns = [col for col in desired_order if col in orig_columns]
```

### Why this is safer:

- `orig_columns` preserves the original ordering and content  
- You can always restore or compare later  
- The comprehension now filters against the backup, not itself  
- This is how data engineers prevent silent logical bugs  

This pattern becomes essential in:

- schema reconciliation  
- feature engineering  
- pandas pipelines  
- machine learning preprocessing  
- reproducible analytics workflows

# 🧩 Key Takeaways: Helper Box

- A list comprehension is built from **output → loop → optional filter**.  
- It **creates a new list**; it does **not mutate** the original list.  
- Reassigning a variable (e.g., `columns = temp_list`) can **discard the original data**.  
- In real analytics work, **always back up original data** before reassigning.  
- You should master the **long version first** — training wheels that give you full visibility.  
- Comprehensions become powerful and natural *after* you understand the underlying loop logic.
