# 06.40 Overview

A **tuple** is another type of sequence in Python, similar to a list but with one major difference:

# ✔ Tuples are **immutable**  
Once a tuple is created, **it cannot be changed**.  
You cannot add, delete, or replace items.

Tuples are incredibly important in analytics work because they:

- ensure data cannot be accidentally changed  
- take up less memory than lists  
- run faster for some operations  
- appear naturally as function outputs  
- prevent logical errors in pipelines

---

# Tuples vs Lists: When to Choose Which

| Use List When... | Use Tuple When... |
|------------------|-------------------|
| You need to modify data | Data must stay fixed |
| Storing transaction sets | Storing coordinates |
| Cleaning inconsistent values | Returning multiple outputs |
| Building datasets dynamically | Locking configuration values |
| Data is evolving | Data is final and stable |

### Rule of Thumb  
> If you do not need mutability, choose a tuple.  

This protects your logic downstream.

# Summary (Part 4 — Tuples)

- Tuples are immutable sequences; once created, they cannot be changed.  
- They use **parentheses** and require a trailing comma for one-element tuples.  
- Tuples are more memory-efficient and faster than lists.  
- They prevent accidental changes and improve code safety.  
- Many functions return multiple values packed into tuples.  
- Unpacking enables fast, readable extraction of values.

````python
# 06.41 Creating Tuples
# Tuples use parentheses instead of square brackets.


# Standard tuple
sizes = ('S', 'M', 'L')

# Tuple with mixed types
info = ('Alice', 42, True)

# Empty tuple
empty = ()

# Converting an iterable into a tuple
letters = tuple('abcd')      # ('a', 'b', 'c', 'd')


# Mini-diagram: tuple structure


sizes = ('S', 'M', 'L')

# Index:       0     1     2
# Value:      'S'   'M'   'L'

# Tuples use the same indexing rules as lists.
````

````python
# 06.42 One-Element Tuples  

# A tuple with one element **must** include a trailing comma.

one = ('A',)     # This is a tuple
not_tuple = ('A') # This is just the string 'A'

# Python interprets parentheses alone as grouping, not as tuple creation.
````

````python
# 06.43 Accessing Tuple Elements
# Tuple indexing works exactly like list indexing:

colors = ('red', 'green', 'blue')

print(colors[0])    # ('red')           → created but not saved
print(colors[-1])   # ('blue')          → created but not saved
print(colors[1:3])  # ('green', 'blue)  → created but not saved

new_colors = colors[1:3]  # saved version
print("\n", new_colors)
print(colors)             # original intact


# - Slicing returns a new tuple object:
# - The original tuple remains unchanged.
# - You can assign the slice to a variable to keep it (just like a list slice).

# Because tuples are immutable:
# - You *cannot* modify the original tuple  
# - You *cannot* modify the slice  
# - You *cannot* use slice assignment  
# - The new tuple is completely isolated and safe  
````

````python
# 06.44 Core Concept: Tuple Immutability

# Tuples cannot:

# - append items  
# - remove items  
# - insert items  
# - reassign items  

# Any attempt to modify them will lead to a TypeError.


t = (1, 2, 3)

# t[1] = 999       # TypeError: 'tuple' object does not support item assignment
# t.append(4)      # AttributeError
````

````python
# 06.45 Why Use Tuples?


# ✔ 1. Tuples use less memory

import sys

list_sample = [1, 2, 3, 4, 5]
tuple_sample = (1, 2, 3, 4, 5)

print(sys.getsizeof(list_sample))
print(sys.getsizeof(tuple_sample))


# ✔ 2. Tuples are faster than lists  

import timeit

print(timeit.timeit("x = (1,2,3,4,5)", number=1_000_000))
print(timeit.timeit("x = [1,2,3,4,5]", number=1_000_000))


# ✔ 3. Tuples prevent accidental changes  

# Example:

# Coordinates shouldn't change
location = (33.12, -117.28)


# ✔ 4. Functions often return tuples  
# This includes built-in functions and library functions

# Example:
# Define a function and give me a list of numbers
# I am naming that list of numbers "values" inside my function

def get_stats(values):
    return (min(values), max(values), sum(values) / len(values))

stats = get_stats([10, 20, 30])       # parentheses make it a tuple
print("\n", stats)       # (10, 30, 20.0)

# Unpacking the tuple:
minimum, maximum, average = stats
print(minimum, maximum, average)

# Tuple unpacking is the natural way Python returns multiple values cleanly.
````

````python
# 06.46 Tuple Unpacking  
# Used heavily in real analytics workflows


transaction1, transaction2, transaction3 = (29.99, 15.99, 9.99)

print(transaction1)    # 29.99
print(transaction2)    # 15.99
print(transaction3)    # 9.99
````

````python
# 06.48 Assignment: Sales Tax  

# Customer C00001 made these three transactions
C00001 = [29.99, 15.99, 9.99]  # List with square brackets

# Step 1: Create a tuple from the list
transactions = tuple(C00001)   # tuple() is a built-in command that converts anything into a tuple
print(transactions)            # (29.99, 15.99, 9.99)


# Step 2: Unpack into variables
t1, t2, t3 = transactions
print("Unpack Tuple:", t1, t2, t3)

# Step 3: Apply sales tax (8 percent)
tax_rate = 0.08
tax1 = round(t1 * tax_rate, 2)
tax2 = round(t2 * tax_rate, 2)
tax3 = round(t3 * tax_rate, 2)

print("Sales tax for each transaction:")
print(tax1, tax2, tax3)

# This preserves the logical structure while improving clarity.


# 06.49 Additional Business Example: Immutable Product Records
# Tuples are excellent for data that must not change.

product_record = (
    "SKU12345",       # product ID
    "Winter Gloves",  # name
    29.99,            # price
    True              # in stock
)

# product_record[2] = 35.99   # would raise an error
````

````python
# --------------------------------------------------------------
# 🔧 Helper: Tuple Inspector (Index Map + Mutability Checker)
# --------------------------------------------------------------

def show_tuple_info(tpl, show_negative=True, show_types=True):
    """
    Display a structured index map for a tuple, including:
      - positive and negative indices
      - value at each index
      - type of each element
      - mutability warning (e.g., tuple containing lists)
    
    Parameters
    ----------
    tpl : tuple
        The tuple to inspect.
    show_negative : bool, default True
        Whether to show negative indices.
    show_types : bool, default True
        Whether to show type information for each element.
    
    Usage
    -----
    show_tuple_info((1, 2, 3))
    show_tuple_info((1, [2, 3], (4, 5)))
    show_tuple_info(("a", "b", "c"), show_negative=False)
    """

    # ----------------------------------------------------------
    # 1. Validate input
    # ----------------------------------------------------------
    if not isinstance(tpl, tuple):
        raise TypeError(
            f"show_tuple_info expects a tuple, got {type(tpl).__name__}"
        )

    n = len(tpl)
    print(f"Tuple length: {n}")
    print()

    # ----------------------------------------------------------
    # 2. Build header
    # ----------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    header_cols.append("mutable?")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += f"{'value':>12} {'mutable?':>10}"

    print(header)
    print("-" * max(len(header), 40))

    # ----------------------------------------------------------
    # 3. Detect mutability of each element
    # ----------------------------------------------------------
    def is_mutable(obj):
        """Return True if object is a known mutable Python type."""
        return isinstance(obj, (list, dict, set, bytearray))

    # ----------------------------------------------------------
    # 4. Print each element with indices and mutability flag
    # ----------------------------------------------------------
    for i, val in enumerate(tpl):
        neg = i - n if show_negative else None
        type_name = type(val).__name__ if show_types else ""
        mutable_flag = "yes" if is_mutable(val) else "no"

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{repr(val):>12} {mutable_flag:>10}"

        print(row)

    # ----------------------------------------------------------
    # 5. Summary warnings (very helpful in practice)
    # ----------------------------------------------------------
    print("\nSummary:")
    if any(isinstance(v, list) for v in tpl):
        print("⚠ Warning: This tuple contains one or more lists.")
        print("  → The tuple is immutable, but the inner lists ARE mutable.")
        print("  → Changing an inner list will change all references to it.\n")
    else:
        print("✓ No mutable inner objects detected. Tuple is fully immutable.\n")


# ------------------------------
# Example:
# ------------------------------
show_tuple_info((1, 2, (3, 4)))
show_tuple_info((1, [2, 3], 4))
````

# 🟦 Helper Box — Slicing: Lists vs. Tuples  
Understanding how slicing behaves for **mutable** vs **immutable** sequence types is essential.  
This helper clarifies exactly what happens under the hood — and why tuple slicing is so safe and predictable.

---

## ❗ First, a universal rule (often overlooked)
> **Slicing always *creates* a new sequence object, but it does *not* save it unless you assign it to a variable.**

```python
numbers = [10, 20, 30, 40]
numbers[1:3]        # slice created, used, discarded
subset = numbers[1:3]  # slice saved
```

The same is true for tuples.

This is important: slicing alone never mutates anything — it just produces a temporary result unless you store it.

---

## ✅ 1. Slicing a List  
### What happens?

```python
lst = [10, 20, 30, 40]
slice_lst = lst[1:3]     # [20, 30]
```

- **Slicing always creates a new list object.**
- The original list is unchanged.
- Because lists are **mutable**, you can modify the slice:

```python
slice_lst[0] = 999      # [999, 30]
# lst remains [10, 20, 30, 40] — completely unaffected
```

### Key property  
Even though slicing is safe, **slice assignment** can modify the original list:

```python
lst[1:3] = [777, 888]   # mutates original list
```

This is a powerful feature — but also a source of bugs if used carelessly.

---

## ✅ 2. Slicing a Tuple  
### What happens?

```python
colors = ('red', 'green', 'blue')
subset = colors[1:3]      # ('green', 'blue') → created but not saved

new_colors = colors[1:3]  # saved version

```

- **Slicing creates a new tuple object.**
- The original tuple remains unchanged.
- You can assign the slice to a variable to keep it (just like a list slice).

### But the critical difference:
Because tuples are **immutable**:

- You *cannot* modify the original tuple  
- You *cannot* modify the slice  
- You *cannot* use slice assignment  
- The new tuple is completely isolated and safe  

Slicing tuples is simpler and safer because there is *no* possibility of mutation anywhere.

---

## 🧠 3. Why the Behavior Is Different?

| Operation                    | Lists (Mutable)                      | Tuples (Immutable)                     |
|-----------------------------|---------------------------------------|----------------------------------------|
| Slicing                     | Creates a **new list**                | Creates a **new tuple object**         |
| Assigning back to a slice  | Allowed (mutates original)            | Not allowed (TypeError)                |
| Modifying the slice itself  | Allowed                                | Not allowed                            |
| Risk of unintended mutation | Medium                                 | Zero                                   |

---

## 🏁 4. The Key Concept in One Sentence  
> **Slicing always returns a new sequence object — but only lists allow slice assignment that can mutate the original. Tuples never allow mutation, so slicing is always safe.**

---

## ⭐ 5. The Mental Model (recommended for mastery)

Think of slicing as:

### **“Copying a portion of the sequence into a new container.”**

- For lists → a new *list container*  
- For tuples → a new *tuple object*  

And because tuples cannot change, this new object is **fully independent** and **risk-free**.

---

## ✔ Quick Verification

```python
t = (1, 2, 3, 4)
s = t[1:3]

print(s)       # (2, 3)
print(t)       # (1, 2, 3, 4)
```

Both remain intact. Nothing can mutate either one.

---

## 📌 Final takeaway  
Whether slicing a list or tuple:

- It **never** changes the original sequence  
- It **always** produces a new object  
- It **must be assigned** if you want to keep it  
- Only lists support slice assignment  
- Tuples are 100% mutation-proof
