## 1. The Boolean Data Type (`True` / `False`)

A **Boolean** is a data type with only **two possible values**:

- `True`
- `False`

In arithmetic:

- `True` behaves like `1`
- `False` behaves like `0`
- The `not` keyword **inverts** a Boolean


````python
# --- Boolean basics: True & False as values and numbers ---

is_ski_season = True
is_store_closed = False

print("is_ski_season:", is_ski_season)
print("is_store_closed:", is_store_closed)

# Booleans act like 1 and 0 in arithmetic
print("True + 1  →", True + 1)   # 1 + 1 = 2
print("False + 1 →", False + 1)  # 0 + 1 = 1


# --- Using 'not' to invert Boolean values ---

print("not True  →", not True)   # becomes False
print("not False →", not False)  # becomes True

on_sale = False
print("on_sale:", on_sale)
print("not on_sale:", not on_sale)  # flips False → True
````

## 2. Comparison Operators

**Comparison operators** evaluate a relationship and return a **Boolean** (`True` / `False`):

| Operator | Meaning                    | Example        |
|---------:|----------------------------|----------------|
| `==`     | equal to                   | `price == 99`  |
| `!=`     | not equal to               | `item != 'ski'`|
| `<`      | less than                  | `inv < 10`     |
| `>`      | greater than               | `temp > 32`    |
| `<=`     | less than or equal to      | `age <= 18`    |
| `>=`     | greater than or equal to   | `score >= 80`  |



````python
# --- Comparison Operators: Numbers ---

inventory_count = 5

print("inventory_count == 0  →", inventory_count == 0)
print("inventory_count != 0  →", inventory_count != 0)
print("inventory_count < 10  →", inventory_count < 10)
print("inventory_count > 10  →", inventory_count > 10)
print("inventory_count <= 5  →", inventory_count <= 5)
print("inventory_count >= 5  →", inventory_count >= 5)
````

## 5. Boolean Operators: `and`, `or`, `not` (+ Short-Circuiting & Precedence)

Boolean operators allow you to combine multiple logical expressions into a single condition:

- **`A and B`** → True only if **both** A and B are True  
- **`A or B`** → True if **at least one** of A or B is True  
- **`not A`** → inverts A (True ↔ False)

---

### 🔄 Short-Circuit Behavior (Efficiency Tip)

Python **stops evaluating early** when the final outcome is already determined:

- For **`A and B`**:  
  - If `A` is **False**, Python **does not evaluate B**
- For **`A or B`**:  
  - If `A` is **True**, Python **does not evaluate B**

This is why we typically place **simple or fast checks first**, and **more expensive operations last**.

---

### 🧠 Operator Precedence (Highest → Lowest)

Python always evaluates Boolean operators in this fixed order:

1. **`not`**  
2. **`and`**  
3. **`or`**

**This means:**
- `not` is always evaluated **before** `and`  
- `and` is always evaluated **before** `or`  
- This happens **even if the operators appear in a different order in the code**

Example mental model:

```
A or B and C
→ A or (B and C)
```

Python will always group `B and C` first because `and` outranks `or`.

---

### 🟦 Parentheses Always Win

Parentheses let you **override** Python’s default precedence rules and intentionally control how the logic is grouped.

Use parentheses to:
- Make your logic explicit  
- Avoid ambiguity in complex conditions  
- Ensure correctness when mixing `and` and `or`  
- Improve readability for yourself and others

Whenever your logic combines different operators, parentheses are a best practice for clarity and safety.



````python
# --- Combining conditions with 'and' & 'or' ---

inventory = 5
price = 99.0

# Both clauses must be True
print("inventory > 0 and price < 100          →", inventory > 0 and price < 100)

# At least one clause must be True
print("inventory == 0 or price < 100          →", inventory == 0 or price < 100)


# --- Additional Boolean operator examples ---

print("5 > 4 and 7 > 8                        →", 5 > 4 and 7 > 8)      # The `and` operator requires ALL statements to be true
print("5 > 4 or 7 > 8                         →", 5 > 4 or 7 > 8)       # The `or` operator requires ONE statement to be true

print("15 > 10 and 11 > 10 and 12 > 10        →",
      15 > 10 and 11 > 10 and 12 > 10)                                  # Chain together any number of boolean expressions


# Explanation (for reference)
# If the first expression in an AND chain is False,
# Python will NOT evaluate the rest (short-circuit).
# If the first expression in an OR chain is True,
# Python will NOT evaluate the rest.
````

````python
# --- Parentheses change the order of evaluation ---

item = "skis"
price = 120

# Example where 'or' is evaluated with parentheses
expr_1 = (item == "skis") or (price <= 100)                 # The `or` operator requires ONE statement to be true
expr_2 = (item == "snowboard") and (price <= 100)           # The `and` operator requires ALL statements to be true

print("item == 'skis' or price <= 100 →", expr_1)
print("item == 'snowboard' and price <= 100 →", expr_2)

# --- Parentheses to control the order of evaluation ---

clothing = "hat"
sale_price = 99

(sale_price >= 100 and clothing == "gloves") or (clothing == "hat")  # The `or` operator requires ONE statement to be true
````

## 🎯 Parentheses & Order of Evaluation  
### How to Control Your Logic Explicitly  

You learned in the previous section that Python evaluates Boolean operators in this order:

1. `not`
2. `and`
3. `or`

This next part **demonstrates** how parentheses let you override that default behavior to ensure your logic is interpreted exactly as intended.

---

## 👕 Example Scenario  
**Discount Rule:**  
A customer receives a discount if:  
- the item is inexpensive **and** it is a pair of gloves,  
**OR**  
- the item is a hat.

Parentheses make this rule explicit.


## ✔ Why Parentheses Matter (Key Takeaways)

- They make your logic **explicit**, not assumed  
- They remove ambiguity when combining `and` and `or`  
- They prevent bugs in business logic and conditional rules  
- They keep your conditions readable and maintainable  



````python
# --- Explicit grouping using parentheses ---

clothing = "hat"
sale_price = 99

# Intended rule:
# "Discount applies if (price <= 100 AND item is gloves) OR item is a hat"

with_parentheses = (sale_price <= 100 and clothing == "gloves") or (clothing == "hat")

print("With parentheses →", with_parentheses)


# 🔍 Explanation  
# - `sale_price <= 100` → True  
# - `clothing == "gloves"` → False  
# - `(True AND False)` → False  
# - `clothing == "hat"` → True  
# - Final: `False OR True` → **True**  
# - **Discount applies**


# ❗ Without Parentheses (Python chooses based on precedence)

# --- Same logic, but WITHOUT parentheses ---

without_parentheses = sale_price <= 100 and clothing == "gloves" or clothing == "hat"

print("Without parentheses →", without_parentheses)

#🔍 What happens?
# Python interprets it as:

(sale_price <= 100 and clothing == "gloves") or (clothing == "hat")

# Python is not short-circuiting the string check you expected
# Python is effectively using its own implicit parentheses based on precedence


# 🧪 When Results **Change** (Critical Lesson)

# Let’s change one variable and observe the difference:

# --- A setup where parentheses DO change the result ---

clothing = "jacket"
sale_price = 99

with_parentheses = (sale_price <= 100 and clothing == "gloves") or (clothing == "hat")
without_parentheses = sale_price <= 100 and clothing == "gloves" or clothing == "hat"

print("\nCase where differences appear:")
print("With parentheses →", with_parentheses)
print("Without parentheses →", without_parentheses)

# In this case the final value happens to be the same — but the logic is **ambiguous and error-prone**, especially in real systems with longer rules.
````

## ⚡ Short-Circuit Evaluation (Efficiency in Action)

Python **stops evaluating early** when a final result is already determined.

- For **`A and B`**:  
  If `A` is False → *Python never evaluates B*
- For **`A or B`**:  
  If `A` is True → *Python never evaluates B*

Short-circuiting is important for performance and safety.

Here is a clear demonstration using “fast” and “slow” checks.

```python
# --- Short-Circuit Demo using clear printed side effects ---

def fast_check():
    print("fast_check() ran → quick evaluation")
    return False  # pretend fast and simple

def slow_check():
    print("slow_check() RAN → expensive evaluation (should NOT run if skipped)")
    return True   # pretend slow and complex


print("\nCase 1: AND short-circuit (first condition is False)")
result_1 = fast_check() and slow_check()
print("Final Result:", result_1)


def fast_true():
    print("fast_true() ran → quick positive result")
    return True

print("\nCase 2: OR short-circuit (first condition is True)")
result_2 = fast_true() or slow_check()
print("Final Result:", result_2)
```

### ⛳ Interpretation
- In the `and` example, `fast_check()` is False → `slow_check()` never runs  
- In the `or` example, `fast_true()` is True → `slow_check()` never runs  

This mirrors real code patterns where we put:
- Cheap validation checks **first**  
- Expensive computations **second**  


````python
# =====================================================================
# --- Demonstration: Short-Circuit using clear printed side effects ---
# =====================================================================

def fast_check():
    print("fast_check() ran → quick evaluation")  # pretend fast and simple
    return False  # Note: Returns False

def slow_check():
    print("slow_check() RAN → expensive evaluation (should NOT run if skipped)")  # pretend slow and complex
    return True   # Note: Returns True


print("\nCase 1: AND short-circuit (first condition is False)")
result_1 = fast_check() and slow_check()  # return False AND return True comparison                   
print("Final Result:", result_1)  # AND short-circuit: if A is False, Python NEVER evaluates B

# ⛳ Interpretation
# - In the `and` example, `fast_check()` is False → `slow_check()` never runs  

def fast_true():
    print("fast_true() ran → quick positive result")
    return True

print("\nCase 2: OR short-circuit (first condition is True)")
result_2 = fast_true() or slow_check()
print("Final Result:", result_2)

# ⛳ Interpretation
# - In the `or` example, `fast_true()` is True → `slow_check()` never runs
````

````python
# ================================================
# Short-circuit example with realistic conditions
# ================================================

def item_in_stock(inventory):
    print("item_in_stock() ran")
    return inventory > 0

def is_discount_item(item):
    print("is_discount_item() ran")
    return item in ["gloves", "hat"]

inventory = 0
item = "gloves"

print("\nReal short-circuit with AND (inventory = 0):")
result = item_in_stock(inventory) and is_discount_item(item)
print("Final:", result)

# ================================================
# Short-circuit example with realistic conditions
# ================================================

inventory = 1
item = "gloves"

def item_in_stock(inventory):
    print("item_in_stock() ran")
    return inventory > 0

def is_discount_item(item):
    print("is_discount_item() ran")
    return item in ["gloves", "hat"]

print("\nReal short-circuit with AND (inventory = 0):")
result = item_in_stock(inventory) and is_discount_item(item)
print("Final:", result)

# is_discount_item() ran — not short-circuited because inventory was > 0.
````

````python
# ============================================
# 🔍 trace_logic – Boolean Logic Tracing Helper
# ============================================

def trace_logic(label, func):
    """
    Run a logical expression (wrapped in a zero-argument function),
    print a clear header, and show the final Boolean result.

    Usage:
        trace_logic(
            "Description of the logic",
            lambda: some_check() and other_check() or third_check()
        )

    To see short-circuiting, each inner check should print when it runs.
    """
    print("\n" + "=" * 60)
    print("TRACE:", label)
    print("=" * 60)

    result = func()  # Evaluate the expression once
    print("Final result:", result)
    return result


# 🧵 Examples: Using `trace_logic()` with Real Checks
# --- Example condition functions with side effects ---
# Scenario 1: inventory = 0 (AND short-circuits, OR still needed)

inventory = 0
item = "gloves"
status = "VIP"

def item_in_stock(inventory):
    print(f"item_in_stock({inventory}) ran")
    return inventory > 0

def is_discount_item(item):
    print(f"is_discount_item({item!r}) ran")
    return item in ["gloves", "hat"]

def is_vip_customer(status):
    print(f"is_vip_customer({status!r}) ran")
    return status == "VIP"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)

# What you’ll see
# - In **Scenario 1** (`inventory = 0`):  
#   - `item_in_stock()` runs, returns False  
#   - `is_discount_item()` is **skipped** (AND short-circuited)  
#   - `is_vip_customer()` runs (needed for OR)


# --- Example condition functions with side effects ---
# Scenario 2: inventory > 0 and discount item (OR short-circuits)

inventory = 5
item = "gloves"
status = "regular"

trace_logic(
    "inventory > 0 and discount item or VIP",
    lambda: item_in_stock(inventory) and is_discount_item(item) or is_vip_customer(status)
)

# What you’ll see
# - In **Scenario 2** (`inventory = 5`, `item="gloves"`):  
#   - `item_in_stock()` runs, returns True  
#   - `is_discount_item()` runs, returns True  
#   - `(True and True)` becomes True  
#   - `is_vip_customer()` is **skipped** (OR short-circuited)

# `trace_logic()` gives you a clean header and final result;  
# your inner functions show **which steps actually executed**.
````

## ✅ Key Takeaways – `trace_logic()`

- `trace_logic(label, func)` runs a **single Boolean expression** and shows the final result with a clear trace header.
- Wrap your expression in a `lambda` so it **is not evaluated** until `trace_logic()` calls it.
- Put `print()` statements inside your condition functions (like `item_in_stock`) to see:
  - which checks run  
  - where short-circuiting skips later checks  
- Use `trace_logic()` to:
  - experiment with **AND / OR / NOT**
  - confirm **precedence and parentheses**
  - build deep intuition about **how Python really evaluates conditions**

This tool is perfect for “x-ray vision” into your Boolean logic while you’re learning and debugging.
