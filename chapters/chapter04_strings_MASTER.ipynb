# 🧵 Python Strings — The 5 Core Categories of String Operations

This reference summarizes the essential operations you will use when working with Python strings.  
Instead of memorizing many separate methods, it is more effective to remember them in categories.  
Each category reflects a specific type of task you will perform regularly in data cleaning, parsing,  
and text manipulation.

---

## 1️⃣ Navigating Strings (Length, Indexing, Slicing)

These operations help identify positions or extract specific parts of a string.

```python
s = "Analytics"

len(s)          # count of characters → 9
s[0]            # first character → 'A'
s[-1]           # last character → 's'
s[2:6]          # slicing → 'alyt'
s[::-1]         # reverse string → 'scitylanA'
```

---

## 2️⃣ Changing Appearance (Formatting & Cleaning)

These methods modify the presentation of the text.

```python
s = "  Python Data  "

s.upper()       # '  PYTHON DATA  '
s.lower()       # '  python data  '
s.title()       # '  Python Data  '
s.strip()       # 'Python Data'
s.replace("Data", "Analysis")   # '  Python Analysis  '
```

---

## 3️⃣ Splitting & Joining (Parsing & Reconstruction)

These are essential for working with CSVs, logs, and text-based datasets.

```python
line = "A,B,C"

line.split(",")          # ['A', 'B', 'C']

items = ["A", "B", "C"]
",".join(items)          # 'A,B,C'
```

---

## 4️⃣ Searching, Finding, Checking (Pattern Detection)

These help determine what the string contains.

```python
s = "machine learning"

"learn" in s              # True
s.startswith("mach")      # True
s.endswith("ing")         # True
s.count("n")              # count occurrences → 2
s.find("learn")           # starting index → 8 (or -1 if not found)
```

---

## 5️⃣ Combining Strings (Building Output)

Building readable strings for printing, logging, file names, and messages.

```python
name = "Dawn"
price = 22.55

"Hello " + name           # concatenation
f"The price is ${price}"  # f-string formatting
"*" * 5                   # repetition → '*****'
```

````python
# ---------------------------------------------------------------
# Universal String Index Map Helper (Positive & Negative Indices)
# ---------------------------------------------------------------

# Original text:
customer_testimonial = '''I love skiing. It's my favorite hobby.
Some people say, "It's not a hobby, it's life."'''
print(repr(customer_testimonial))  # repr() returns an unambiguous string, often a valid expression

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")  # positive and negative column indices

show_index_map(customer_testimonial)
````

````python
# -------------------------------------------------------------
# Basic slicing examples (positive step only)
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# -------------------------------------------------------------
# 1) Slice from the beginning up to a position (stop not included)
# -------------------------------------------------------------
# Index positions: 0 1 2 3 4 5 6
# Characters:      I _ h o p e _

first_phrase = message[0:6:1]   # indices 0 through 5 (stop not included)
print(first_phrase)

first_phrase = message[0:6]     # step defaults to 1
print(first_phrase)             # Output: 'I hope'


# -------------------------------------------------------------
# 2) Slice from index 2 up to index 6 (stop not included)
# -------------------------------------------------------------
# Index positions: 2 3 4 5
# Characters:      h o p e

middle_slice = message[2:6:1]   # indices 2 through 5 (stop not included)
print(middle_slice)

middle_slice = message[2:6]
print(middle_slice)             # Output: 'hope'


# -------------------------------------------------------------
# 3) Slice from index 6 to the end
# -------------------------------------------------------------
# message[6:] means:
#   start = 6
#   stop = default (len(message))
#   step = 1

after_first_two_words = message[6:None:1]  
# Using None to explicitly slice syntax with start, stop, and step
print(after_first_two_words)

after_first_two_words = message[6:]
print(after_first_two_words)         # Output: ' it snows TONIGHT!'


# -------------------------------------------------------------
# 4) Slice with a step (every other character)
# -------------------------------------------------------------
# message[0:6:2] takes indices: 0, 2, 4

skip_chars = message[0:6:2]
print(skip_chars)               # Output: 'Ihp '
````

````python
# -------------------------------------------------------------
# 5) Reverse the entire string
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[::-1] uses the slicing pattern [start : stop : step]
#   - start: omitted → Python begins at the last character (because step = -1)
#   - stop: omitted → Python continues until it moves past the first character
#   - step: -1 → move backward one character at a time
#
# Together, message[::-1] produces a full reverse of the string.

reversed_message = message[::-1]
print(reversed_message)
# This is the most common Pythonic way to reverse a string.


# -------------------------------------------------------------
# 6) Slice from the end using negative indices
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[-8:-1] uses the slicing pattern [start : stop : step]
#   - start at index -8  → the 'T' in 'TONIGHT'
#   - stop at index -1   → but EXCLUDE index -1 (the '!')
#   - step defaults to 1 → moving forward left-to-right

# Negative index mapping for this string:
#    message = "I hope it snows TONIGHT!"
#                      T   O   N   I   G   H   T   !
#            indices: -8  -7  -6  -5  -4  -3  -2  -1

last_word = message[-8:-1]
print(last_word)       # Output: 'TONIGHT' (excludes the '!')


# -------------------------------------------------------------
# 7) Slice the last 8 characters including the final character
# -------------------------------------------------------------

message = "I hope it snows TONIGHT!"

# message[-8:] uses the slicing pattern [start : stop : step]
#   - start at index -8  → the first 'T' in 'TONIGHT'
#   - stop omitted       → continue to the end of the string
#   - step defaults to 1 → move forward left-to-right

# Negative index mapping for this string:
#    message = "I hope it snows TONIGHT!"
#                      T    O    N    I    G    H    T    !
#            indices: -8  -7   -6   -5   -4   -3   -2   -1

tail_slice = message[-8:]
print(tail_slice)      # Output: 'TONIGHT!'
````

````python
# -------------------------------------------------------------
# Len() Exercise: Simple Cryptography
# -------------------------------------------------------------
# Goal:
#   Use the length of different messages as "indexes" into the
#   alphabet string to build a password.
#
# Process:
#   1. Measure the length of each message using len()
#   2. Use that number as an index into the alphabet
#   3. Concatenate the resulting letters into a password
# -------------------------------------------------------------

# A reference alphabet for indexing
alphabet = "abcdefghijklmnopqrstuvwxyz"

# Messages of varying lengths
message1 = "Hello World!"
message2 = ""
message3 = "It snowed a lot today"
message4 = "eeee"
message5 = "I love snow!!"

# -------------------------------------------------------------
# Step-by-step explanation:
#
print(len(message1))  #  → 12
print(alphabet[12])  # → 'm'

print(len(message2))  # → 0
print(alphabet[0])  # → 'a'

print(len(message3))  # → 21
print(alphabet[21])  # → 'v'

print(len(message4))  # → 4
print(alphabet[4])  # → 'e'

print(len(message5))  # → 13
print(alphabet[13])  # → 'n'
#
# Final password: "maven"
# -------------------------------------------------------------

password = (
    alphabet[len(message1)] +
    alphabet[len(message2)] +
    alphabet[len(message3)] +
    alphabet[len(message4)] +
    alphabet[len(message5)]
)

print("Generated password:", password)
````

````python
# -------------------------------------------------------------
# String Methods: Demonstrating .strip() to Remove Extra Spaces
# -------------------------------------------------------------
# Methods are functions that apply only to a specific data type or object.
# .strip() removes leading and trailing whitespace automatically - no indexing needed!
# -------------------------------------------------------------

# Example string with intentional leading/trailing spaces
message = "  I hope it snows TONIGHT!  "

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------
# This function helps visualize how .strip() affects the positions
# of characters by printing:
#   • positive indexing (0 → len-1)
#   • negative indexing (-len → -1)
#
# Seeing the spaces in the index map makes it clear
# why .strip() removes ONLY the characters at the edges.
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula for corresponding negative index
        print(f"{i:3} {neg:4} {repr(ch)}")


# Display index map BEFORE stripping spaces
print("Before .strip():")
show_index_map(message)

# -------------------------------------------------------------
# Using .strip() to remove leading and trailing spaces
# -------------------------------------------------------------
# .strip() removes ONLY whitespace at the beginning and end.
# It does NOT remove spaces inside the string.
cleaned = message.strip()

print("\nAfter .strip():")
show_index_map(cleaned)   # show the new indexing after cleaning

# Print both versions to compare clearly
print(f"\nOriginal message: [{message}]")
print(f"Stripped message: [{cleaned}]")
````

````python
# -------------------------------------------------------------
# String Methods: .strip("."), .lstrip(), .rstrip() — remove unwanted whitespace
# -------------------------------------------------------------
# .strip()  → removes BOTH leading + trailing whitespace
# .lstrip() → removes whitespace only on the LEFT
# .rstrip() → removes whitespace only on the RIGHT

message = "   I love skiing and snow.   "   # spaces included for .strip() examples
l_cleaned = message.lstrip()
r_cleaned = message.rstrip()

print(f"[{l_cleaned}]")  # left spaces removed
print(f"[{r_cleaned}]")  # right spaces removed

punctuated = "...powder day!!!..."
# .strip(".") removes ONLY the character "." from both ends (not from the middle)
print(punctuated.strip("."))
````

````python
# -------------------------------------------------------------
# String Methods: .strip(chars)
# removes only the characters listed in `chars`,
# and only from the **beginning and end**, *not the middle*
# -------------------------------------------------------------

# Example strings
message = "   I love skiing and snow.   "     # has leading + trailing spaces
punctuated = "...powder day!!!..."            # mix of periods and exclamation marks

# Helper: Show positive & negative index map for any string
def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n
        print(f"{i:3} {neg:4} {repr(ch)}")
        
print("\n--- Punctuated String ---")
show_index_map(punctuated)

# Remove only periods from both ends
no_periods = punctuated.strip(".")
print("\n--- After .strip('.') ---")
show_index_map(no_periods)

# Remove periods AND exclamation marks from both ends
cleaned_all = punctuated.strip(".!")
print("\n--- After .strip('.!') (remove . and !) ---")
show_index_map(cleaned_all)

print(f"\nOriginal:   [{punctuated}]")
print(f"strip('.'): [{no_periods}]")
print(f"strip('.!'):[{cleaned_all}]")
````

````python
# String Methods
# -------------------------------------------------------------
# 1. .find() — locate the FIRST occurrence of a substring
# -------------------------------------------------------------
# .find("snow") returns:
#   - the index of the FIRST match
#   - OR -1 if the substring is not found

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------

message = "I hope it snows tonight!"

def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")

show_index_map(message)

first_snow_index = message.find("snow")
print("Index of 'snow':", first_snow_index)

missing_index = message.find("rain")        # "rain" does not appear
print("Index of 'rain':", missing_index)    # → -1 (Python’s “not found” indicator)
````

````python
# String Methods
# -------------------------------------------------------------
# .count()
# -------------------------------------------------------------
message = "I hope it snows tonight because snow is coming."

# -------------------------------------------------------------
# 1. Count single characters
# -------------------------------------------------------------
print(message.count("o"))     # count how many 'o' characters
print(message.count("i"))     # count of 'i' characters

# -------------------------------------------------------------
# 2. Count whole substrings
# -------------------------------------------------------------
print(message.count("snow"))   # number of times "snow" appears

# -------------------------------------------------------------
# 3. Count in a specific range of indexes: .count(sub, start, end)
# -------------------------------------------------------------
# Count only in the first 20 characters
print(message.count("o", 0, 20))

# Count "snow" only in the second half of the sentence
midpoint = len(message) // 2
print(message.count("snow", midpoint, len(message)))
````

````python
# String Methods
# -------------------------------------------------------------
# Method Chaining — perform multiple text operations in sequence
# -------------------------------------------------------------

# -------------------------------------------------------------
# Helper: Show positive & negative index for each character
# -------------------------------------------------------------

message = "  I HOPE it snows TONIGHT!  "
cleaned_message = message.strip()

def show_index_map(s):
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n      # formula: negative index for this position
        print(f"{i:3} {neg:4} {repr(ch)}")

show_index_map(cleaned_message)

# GOAL:
#   Extract everything AFTER "I HOPE "
#   Make the result lowercase
#   Convert to a list of words
#
# Breakdown of what happens:
#   message.strip()      → removed outer space first for cleaned_message
#   [7::]                → skipping [0:6] and slicing on [7:]
#   .lower()             → convert the remaining text to lowercase
#   .split()             → break into list of words

# processed = cleaned_message[7:].lower().split()
processed = (
    cleaned_message
    [7::]         # slice off "I HOPE "
    .lower()      # convert to lowercase
    .split()      # convert to list of words
)

print(processed)                     # ['it', 'snows', 'tonight!']
print("Word count:", len(processed)) # demonstrates chaining + len()
````

### Understanding Why `__doc__` Works for Functions but *not* Variables

Python allows you to attach documentation (called a **docstring**) to functions, classes, and modules by placing a triple-quoted string immediately after the definition. These docstrings are accessible using the `.__doc__` attribute.

However, a **variable does not store docstrings**. If you call `.__doc__` on a string variable, Python returns the documentation for the **`str` class itself**, not the variable’s content.

This section compares how `__doc__` behaves for:
- a triple-quoted *multi-line string variable*, and  
- a properly documented function.

### 📌 Key Takeaways

- Triple-quoted strings assigned to variables are **just strings**, not docstrings.  
- Only **functions, classes, and modules** store real docstrings accessible via `.__doc__`.  
- Calling `.__doc__` on a string gives you the documentation for Python’s built-in `str` class.  
- To see a variable’s contents, print the variable directly.  
- To see a function’s docstring, print `function.__doc__`.

## 🔍 Slicing Strings in Python

Python slicing uses the pattern:

```python
string[start : stop : step]
```

### ✔️ What each argument really means

- **start** — index to begin the slice  
  - default depends on step direction  
  - for positive step, default = `0`  
- **stop** — index to stop *before*  
  - default depends on step direction  
  - for positive step, default = `len(string)`  
- **step** — how much to move between characters  
  - default = `1`  
  - must be **non-zero**

---

### 🔹 Important correction
The defaults for `start` and `stop` are **NOT always 0 and len(string)**.  
Those defaults apply **only when the step is positive**.

We will use **positive-step slicing only** in this section,  
so the defaults here behave normally.

---


---

## 🔄 Reversing Strings and Using Negative Indices

Python allows slicing with **negative indices** and **negative step values**, which gives you powerful ways to read a string from right to left.

### ✔️ Important Rules

- Negative indices count from the right  
  ```
  message = "I hope it snows TONIGHT!"
  Index:    -23 … -1
  ```
- A **negative step (-1)** means slicing moves backwards.
- When step is negative:
  - default `start` becomes the **last index**
  - default `stop` becomes **before the first character**
- Slicing still stops **before** the `stop` index — even for negative steps.

## ✂️ Cleaning Text with `.strip()` — Visualizing Leading & Trailing Whitespace

In Python, the `.strip()` method removes **only** the whitespace at the *beginning* and *end* of a string.  
It does **not** remove spaces inside the text.  

The helper function below prints a full index map (positive and negative indices) so you can *see*  
how `.strip()` changes the string's length and character positions.

```python
# -------------------------------------------------------------
# Demonstration: .strip() Removes Leading & Trailing Whitespace
# -------------------------------------------------------------
# This example shows exactly which characters are removed and how
# the string's indexing changes after cleaning.
# -------------------------------------------------------------

# Example string with intentional surrounding spaces
message = "  I hope it snows TONIGHT!  "

# -------------------------------------------------------------
# Helper Function: Index Map (Positive + Negative Indices)
# -------------------------------------------------------------
def show_index_map(s):
    """
    Display the positive and negative index for each character in s.
    Useful for understanding slicing and how .strip() affects length.
    """
    n = len(s)
    print(f"Length: {n}\n")
    print(" idx  neg  char")
    print("----------------")
    for i, ch in enumerate(s):
        neg = i - n   # convert positive → negative index
        print(f"{i:3} {neg:4} {repr(ch)}")

# -------------------------------------------------------------
# Before .strip(): Visualize All Characters (Including Spaces)
# -------------------------------------------------------------
print("Before .strip():")
show_index_map(message)

# -------------------------------------------------------------
# Apply .strip(): Removes ONLY outside whitespace
# -------------------------------------------------------------
cleaned = message.strip()

print("\nAfter .strip():")
show_index_map(cleaned)

# -------------------------------------------------------------
# Compare Outputs Clearly
# -------------------------------------------------------------
print(f"\nOriginal message: [{message}]")
print(f"Stripped message: [{cleaned}]")
```

---

### 🔑 Key Takeaways

- `.strip()` only removes whitespace at the edges of the string.  
- Interior spaces remain unchanged.  
- Indexing shifts because the string becomes shorter.  
- Negative indices adjust automatically to the new string length.  
- The index map helper makes whitespace behavior visible and intuitive.

## ✂️ Advanced Stripping Methods: `.lstrip()`, `.rstrip()`, `.strip(chars)`

Python provides three related cleaning methods for removing characters from the *edges* of a string:

| Method | Removes | Example |
|--------|----------|----------|
| `.strip()` | Both left + right whitespace (or characters) | `s.strip()` |
| `.lstrip()` | Left-side only | `s.lstrip()` |
| `.rstrip()` | Right-side only | `s.rstrip()` |
| `.strip(chars)` | Removes *only the characters listed* in `chars` from both ends | `s.strip(".!?")` |

Each version affects indexing differently because the *length of the string changes*.  
The helper below visualizes those changes.

```

## 3️⃣ `.strip(chars)` – Remove Specific Characters From Both Ends

`.strip(chars)` removes **only** the characters listed in `chars`,  
and only from the **beginning and end**, *not the middle*.

```

## 🔑 Key Takeaways

- `.strip()`, `.lstrip()`, and `.rstrip()` do **not** modify the middle of the string—only the edges.
- `.strip(chars)` removes *only the characters listed* in `chars`, not arbitrary whitespace.
- Stripping changes the string length, which shifts both positive and negative indices.
- The index map helper is the most reliable way to understand exactly what was removed.
- These methods are foundational for cleaning raw text and preparing real-world data for analysis.

## 🔍 Searching Within Strings Using `.find()`

The `.find()` method locates the **first occurrence** of a substring inside a string.  
It returns:

- **the index of the first match**, or  
- **–1** if the substring is not found.

Understanding `.find()` becomes much easier when combined with the index-map helper.

- If the substring is not found, it returns **–1**, not an error.

- Visualizing the string with the index map makes `.find()` easier to use in slicing and extraction tasks.
- `.find()` is frequently used in:
  - text parsing  
  - cleaning messy CSV fields

## 🧮 Key Points — `.count()`

- `text.count("a")` → counts how many times the character `"a"` appears  
- `text.count("snow")` → counts how many times the substring `"snow"` appears  
- `text.count(x, start, end)` → counts only within the slice `[start:end]`  
  - `start` is **inclusive**  
  - `end` is **exclusive** (standard Python slicing rule)  
- `.count()` tells you **how many times**, *not where* the substring occurs  
- `.count()` does **not** count overlapping matches  
  - Example: `"aaa".count("aa")` → `1` (not `2`) because `"aa"` overlaps itself  

---

## 📌 When to Use `.count()`

Use `.count()` when answering questions like:

- “How many times does this character appear?”  
- “How many times is this substring repeated?”  
- “How many occurrences appear inside a specific part of the string?”  

Use `.count()` for **frequency**, not for locating positions.

For positions, prefer:

- `.find()`  
- `.rfind()`  
- `.index()`

## 1️⃣ Navigating Strings – Length, Indexing, and Slicing

```python
# Basic string for navigation examples
word = "Analytics"

# Length: total number of characters
print("Length of word:", len(word))      # 9

# Positive indexing: 0-based from left to right
print("word[0]:", word[0])              # 'A'
print("word[1]:", word[1])              # 'n'
print("word[8]:", word[8])              # 's' (last character)

# Negative indexing: -1 is last character, -2 is second-to-last, etc.
print("word[-1]:", word[-1])           # 's'
print("word[-2]:", word[-2])           # 'c'

# Slicing: [start:stop] includes start, excludes stop
print("word[0:4]:", word[0:4])         # 'Anal'
print("word[4:9]:", word[4:9])         # 'ytics'

# Slicing with omitted bounds
print("word[:4]:", word[:4])           # 'Anal' (from start up to index 4)
print("word[4:]:", word[4:])           # 'ytics' (from index 4 to end)

# Slicing with a step value: [start:stop:step]
print("word[0:9:2]:", word[0:9:2])     # 'Aayc' (every 2nd character)
print("word[::-1]:", word[::-1])       # 'scitylanA' (reverse string)
```
