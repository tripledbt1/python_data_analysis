````python
# 06.05 Slicing: Extracting Portions of a List
# Using the List Index Map Helper for a flat list

# Slice syntax: list[start : stop : step]

sizes = ['XS', 'S', 'M', 'L', 'XL', 'XXL']

print(sizes[:3])       # ['XS', 'S', 'M']
print(sizes[:4])       # ['XS', 'S', 'M', 'L']
print(sizes[2:5])      # ['M', 'L', 'XL']
print(sizes[::2])      # ['XS', 'M', 'XL']
print(sizes[2:4])      # ['M', 'L']
print(sizes[-3:])      # ['L', 'XL', 'XXL']
print()

# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input: validate isinstance(obj, type)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
show_list_index_map(sizes)         # flat list
print()
````

````python
# --------------------------------------------------------------
# 🔧 Helper: Slice Visualizer (for lists, tuples, strings, ranges)
# --------------------------------------------------------------

def visualize_slice(seq, start=None, stop=None, step=None, show_negative=True, label="seq"):
    """
    Visualize how a slice seq[start:stop:step] selects elements
    from a sequence (list, tuple, string, range, etc.).

    Parameters
    ----------
    seq : sequence
        Any indexable, sliceable Python sequence (e.g., list, tuple, str, range).
    start : int or None
        Slice start index (can be negative or None).
    stop : int or None
        Slice stop index (can be negative or None).
    step : int or None
        Slice step (can be negative or None).
    show_negative : bool, default True
        Whether to display negative indices.
    label : str, default "seq"
        Optional label to show in the output header.

    Usage
    -----
    visualize_slice([10, 20, 30, 40, 50], 1, 4, 2)
    visualize_slice(range(10), 2, 9, 3)
    visualize_slice(('a', 'b', 'c', 'd', 'e'), 0, None, 2)
    """

    # ----------------------------------------------------------
    # 1. Basic validation
    # ----------------------------------------------------------
    try:
        n = len(seq)
    except TypeError:
        raise TypeError(
            f"visualize_slice expects a sequence with len(), got {type(seq).__name__}"
        )

    slice_obj = slice(start, stop, step)

    # Normalize slice parameters using slice.indices()
    norm_start, norm_stop, norm_step = slice_obj.indices(n)

    # Compute selected indices based on normalized values
    selected_indices = list(range(norm_start, norm_stop, norm_step)) if n > 0 else []

    # ----------------------------------------------------------
    # 2. High-level summary
    # ----------------------------------------------------------
    print(f"Sequence label : {label}")
    print(f"Type           : {type(seq).__name__}")
    print(f"Length         : {n}")
    print(f"Slice requested: {label}[{start}:{stop}:{step}]")
    print(f"Normalized to  : {label}[{norm_start}:{norm_stop}:{norm_step}]")
    print()

    if n == 0:
        print("This sequence is empty. No indices to visualize.")
        return

    # ----------------------------------------------------------
    # 3. Build index and value rows
    # ----------------------------------------------------------
    # We'll give each slot a fixed width for alignment
    cell_width = 5

    # Positive indices
    idx_row = "idx : "
    for i in range(n):
        idx_row += f"{i:>{cell_width-1}} "

    # Negative indices (optional)
    if show_negative:
        neg_row = "neg : "
        for i in range(n):
            neg_row += f"{i - n:>{cell_width-1}} "
    else:
        neg_row = None

    # Values (repr truncated for readability)
    val_row = "val : "
    for i in range(n):
        text = repr(seq[i])
        if len(text) > cell_width - 1:
            text = text[:cell_width - 2] + "…"
        val_row += f"{text:>{cell_width-1}} "

    # Selection markers
    sel_row = "sel : "
    selected_set = set(selected_indices)
    for i in range(n):
        marker = "^" if i in selected_set else ""
        sel_row += f"{marker:>{cell_width-1}} "

    # ----------------------------------------------------------
    # 4. Print the visualization
    # ----------------------------------------------------------
    print(idx_row)
    if neg_row is not None:
        print(neg_row)
    print(val_row)
    print(sel_row)
    print()

    # ----------------------------------------------------------
    # 5. Show the actual slice result
    # ----------------------------------------------------------
    result = seq[slice_obj]
    print(f"Result of {label}[{start}:{stop}:{step}] → {repr(result)}")
    print()


# --------------------------------------------------------------
# Example usage (you can comment these out in your notebook):
# --------------------------------------------------------------
visualize_slice(['XS', 'S', 'M', 'L', 'XL', 'XXL'], 1, 5, 2, label="sizes")
visualize_slice(range(10, 0, -2), None, None, None, label="r")
visualize_slice(('r', 'g', 'b', 'y', 'k'), 1, 4, 1, label="colors")
````

````python
# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types & Nested Lists)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input: validate isinstance(obj, type)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
        # Optionally expand nested lists one level deep
        if expand_nested and isinstance(value, list):
            inner_len = len(value)
            print(f"      ↳ nested list (len={inner_len})")
            for j, inner_val in enumerate(value):
                inner_neg = j - inner_len if show_negative else None
                inner_type = type(inner_val).__name__ if show_types else ""
                inner_repr = _repr_value(inner_val)
                
                # Example format: [2][1]   -1    int  99
                nested_row = f"      [{i}][{j}] "
                if show_negative:
                    nested_row += f"{inner_neg:>4} "
                if show_types:
                    nested_row += f"{inner_type:>10} "
                nested_row += f"{inner_repr}"
                print(nested_row)


# -------------------------------------------
# ✅ Example Usage (Flat List)
# -------------------------------------------
sizes = ['XS', 'S', 'M', 'L', 'XL']
show_list_index_map(sizes)         # flat list
print()

# -------------------------------------------
# ✅ Example Usage (Nested List, Expanded)
# -------------------------------------------
orders = [
    [19.99, 5.99],
    [3.49, 9.99, 14.99],
    [60.00]
]

show_list_index_map(
    orders,                         # nested list
    show_types=True,
    show_negative=True,
    truncate=50,
)
````

## 🔍 Notes for Essential Methods

- `append()` vs `insert()`  
  - Use `append()` unless you *truly* need a specific position  
  - `insert()` shifts the entire list (costly with large data)

- `remove()` deletes **only the first** matching item  
  - Useful to know when cleaning data

- `pop()` is the **only list method that returns a value**

- `sort()` is **in-place**  
  - Use `sorted(lst)` when you want to keep the original list intact  
  - This difference becomes crucial for data analysis

# 06.05 Slicing (Extracting Portions of a List)

Slice syntax:

```
list[start : stop : step]
```

- **start** = where the slice begins  
- **stop** = where the slice ends (non-inclusive)  
- **step** = how many to skip  

```python
sizes = ['XS', 'S', 'M', 'L', 'XL', 'XXL']

print(sizes[:3])       # ['XS', 'S', 'M']
print(sizes[2:5])      # ['M', 'L', 'XL']
print(sizes[::2])      # ['XS', 'M', 'XL']
print(sizes[-3:])      # ['L', 'XL', 'XXL']
```


- Selecting subsets

# 06.06 Unpacking Lists

Unpacking allows you to assign list elements directly to variables.

```python

print(small)    # 'S'
print(medium)   # 'M'
print(large)    # 'L'
```

# 🧮 Universal List Index Map Helper (Professional Version – With Parameters)

This helper is the **list equivalent** of your Universal String Index Map Helper.

It is designed to:

- Show **positive and negative indices** for each element in a list  
- Optionally display **data types**  
- Optionally **truncate long values** for readability  
- Optionally **expand nested lists** and show their indices as `[outer][inner]`  
- Provide a **clear, student-friendly view** of how indexing works in real data  

You can use this to:

- Teach indexing to others

## 🔧 Python Helper: `show_list_index_map`

```python
# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types & Nested Lists)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input validation (student-friendly)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
        # Optionally expand nested lists one level deep
        if expand_nested and isinstance(value, list):
            inner_len = len(value)
            print(f"      ↳ nested list (len={inner_len})")
            for j, inner_val in enumerate(value):
                inner_neg = j - inner_len if show_negative else None
                inner_type = type(inner_val).__name__ if show_types else ""
                inner_repr = _repr_value(inner_val)
                
                # Example format: [2][1]   -1    int  99
                nested_row = f"      [{i}][{j}] "
                if show_negative:
                    nested_row += f"{inner_neg:>4} "
                if show_types:
                    nested_row += f"{inner_type:>10} "
                nested_row += f"{inner_repr}"
                print(nested_row)


# -------------------------------------------
# ✅ Example Usage (Flat List)
# -------------------------------------------
sizes = ['XS', 'S', 'M', 'L', 'XL']
show_list_index_map(sizes)

# -------------------------------------------
# ✅ Example Usage (Nested List, Expanded)
# -------------------------------------------
orders = [
    [19.99, 5.99],
    [3.49, 9.99, 14.99],
    [60.00]
]

show_list_index_map(
    orders,
    show_types=True,
    show_negative=True,
    truncate=50,
    expand_nested=True
)
```

# 📦 Helper Box: How to Read Any List Comprehension (3-Step Method)

List comprehensions are one of Python’s most powerful and expressive features.  
They compress a full loop + condition + append operation into a clear, readable one-liner.

This helper teaches you how to **decode** any list comprehension using a simple 3-step process, and includes a **Safety Upgrade** showing how data professionals prevent accidental loss of original data.

---

## 🔍 General Pattern

A list comprehension always follows this structure:

```python
[new_item   for item in iterable   if condition]
```

It has **three components**:

1. **Output expression** → what goes into the new list  
2. **Iteration** → where items come from  
3. **Condition (optional)** → keep only items that pass the test  

---

# 🟦 Apply the 3-Step Method to Your Example

Your comprehension:

```python
[col for col in desired_order if col in columns]
```

### Step 1 — “For each…”  
`for col in desired_order`  
→ Loop through each column name in `desired_order`.

### Step 2 — “If…”  
`if col in columns`  
→ Only keep columns that actually exist in the current dataset.

### Step 3 — “Put into the new list…”  
`col`  
→ Place the column name into the new list as-is.

### ✔️ Final meaning (plain English)

> “Rebuild the columns list in the order of `desired_order`, but keep only those columns that appear in the existing `columns` list.”

---

# 🔬 The Long Version (What Python Does Internally)

```python
orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

temp_list = []
for col in desired_order:
    if col in columns:
        temp_list.append(col)

columns = temp_list
```

This is exactly what the comprehension does behind the scenes.

- It **does not mutate** the original list object.  
- It **does overwrite** the variable `columns` so the previous ordering is lost unless backed up.

---

# 🛡️ **Safety Upgrade: How Professionals Protect Original Data**

The long version ends with:

```python
columns = temp_list
```

This **reassigns** the variable `columns`, meaning the original ordering is gone unless saved.

In analytics, ETL, and MSBA contexts, you never want to lose original schema information.

### ✔️ Safe, production-style pattern

```python
orig_columns = columns[:]   # make a full copy of the original list
orig = columns.copy()
orig = list(columns)

columns = [col for col in desired_order if col in orig_columns]
```

### Why this is safer:

- `orig_columns` preserves the original ordering and content  
- You can always restore or compare later  
- The comprehension now filters against the backup, not itself  
- This is how data engineers prevent silent logical bugs  

This pattern becomes essential in:

- schema reconciliation  
- feature engineering  
- pandas pipelines  
- machine learning preprocessing  
- reproducible analytics workflows

# 06.34 Copying Lists: The Three Behaviors - See Helper Box (Below)

Copying lists in Python is subtle.  
This is where many beginners (and some professionals) make mistakes.

There are **three** main behaviors:

---

## 1. Variable Assignment  
### Creates a **second reference variable**, NOT a true copy.

```python
list1 = [[10, 20], [30, 40]]
list2 = list1      # assignment, not copying

list2[0][1] = 999

print(list1)  # [[10, 999], [30, 40]]
print(list2)  # [[10, 999], [30, 40]]
```

### Why this happens

Both variables point to **the same memory location**.

### Memory diagram

```
list1 ───► [ [10, 20], [30, 40] ]
           ▲
list2 ─────┘  (same object, two names)
```

**Any change to one affects the other.**

Use this ONLY when intentional.

---

## 2. Shallow Copy (.copy())  
### Copies the outer list, but not inner lists.

```python
original = [[1, 2], [3, 4]]
copy1 = original.copy()

copy1[0][1] = 999   # modifying the INNER list

print(original)   # [[1, 999], [3, 4]]
print(copy1)      # [[1, 999], [3, 4]]
```

### Why this happens

`.copy()` creates a new list, **but its elements still reference the same inner lists**.

### Memory diagram

```
original:   [  L1,     L2  ]
               │       │
copy1:      [  L1,     L2  ]
```

Where `L1` and `L2` are the same inner list objects.

### Important rule  
> Replacing an entire inner list creates separation.  
> Modifying an element inside an inner list does not.

Example:

```python
copy1[0] = ['new', 'list']   # replace entire inner list
```

***Now, original[0] stays unchanged.***

---

## 3. Deep Copy (deepcopy())

### Creates a **100 percent independent copy**.

```python
from copy import deepcopy   # import deepcopy

original = [[1, 2], [3, 4]]
copy2 = copy.deepcopy(original)

copy2[0][1] = 999

print(original)   # unchanged, [[1, 2], [3, 4]]
print(copy2)      # changed, [[1, 999], [3, 4]]
```

### Memory diagram

```
original:   [ L1 ,     L2  ]
copy2:      [ L1',     L2' ]
```

Where L1 and L2 are completely separate new lists.

### When to use deepcopy

- Complex nested structures  
- Multi-level JSON-like data  
- When list mutability must not affect the original  
- Data cleaning pipelines that require preservation

# 🧰 Helper Box: **Shallow Copy vs Deep Copy — The Only Rule You Need**

This helper gives you the *correct, professional-level mental model* for copying data structures in Python.  
Most confusion about `.copy()` comes from not understanding how references behave inside nested structures.

This box will stay valid for your entire MSBA program, pandas work, and future data science.

---

## 🟦 1. The One Rule That Solves Everything

### ✅ **`.copy()` is safe only for *flat* lists.**  
### ❌ **`.copy()` is NOT safe for nested data structures.**  
### ⭐ **Use deep copy only when you *must* — and avoid nested structures in analytics whenever possible.**

That’s the rule.

Let’s expand it.

---

## 🟩 2. What `.copy()` Actually Does (Shallow Copy)

```python
new_list = old_list.copy()
```

This copies **only the outer list**.

It does **NOT** copy the objects inside it.

### ✔️ Safe:
```python
[1, 2, 3, 4]
['a', 'b', 'c']
[True, False, True]
```

These contain simple, immutable elements → no shared inner state.

### ❌ Not Safe:
```python
[[1, 2], [3, 4]]
[ ['a', 'b'], ['c', 'd'] ]
[ [1, 2], 99, ['x', 'y'] ]
```

These contain **nested lists**, which are still shared between copies.

---

## 🟥 3. Visual Mental Model You Should Use

### Example List
```python
list1 = [[1, 2], [3, 4]]
list2 = list1.copy()
```

### Reality (shallow copy)
```
list1 ───► [ A , B ]     # where A is [1, 2], and B is [3, 4]  ⚠️
list2 ───► [ A , B ]     # [[1, 2], [3, 4]]  ⚠️

A = [1, 2]   ← shared
B = [3, 4]   ← shared

        ┌───────────────┐
list1 ─►│   index 0 ─────┼──► A ([1,2])
        │   index 1 ─────┼──► B ([3,4])
        └───────────────┘

        ┌───────────────┐
list2 ─►│   index 0 ─────┼──► A ([1,2])
        │   index 1 ─────┼──► B ([3,4])
        └───────────────┘

```

Both lists point to the **same** inner lists.

### Mutation ***inside*** A or B affects BOTH:
```python
list1[0][1] = 999

list1[0] ----> A <---- list2[0]  ⚠️
```
We are mutating A, because list1[0] refers to A.

But since list2[0] ALSO refers to A, we see the mutation in both

Now *both* `list1` and `list2` show:
```
[1, 999]
```

⚠️ This is why shallow copy is unsafe for nested data.

---

## 🟦 4. Why Replacing an Entire Nested List *Doesn't* Affect the Copy

```python
list1[0] = ['x', 'y', 'z']
```

This does *not* mutate the inner list.  
It **rebinds** index 0 to a *new* list.

```
list1 ─► ['x','y','z']   (new)
list2 ─► [1, 2]          (old)

list1 ───► [ NEW_LIST , B ]
list2 ───► [    A     , B ]
```

So the effect does *not* propagate.

This explains why your instructor’s example behaved differently in the two cases.

---

## 🟧 5. The Only Time You Use `deepcopy`

```python
from copy import deepcopy
new_list = deepcopy(old_list)
```

Use `deepcopy` only when:

- You must work with nested Python objects  
- And you must guarantee isolation  
- And flattening is not possible  

**Data analysts rarely use deepcopy because they rarely use nested Python lists at scale.**  
Instead, they use:

- pandas DataFrames  
- NumPy arrays  
- flat lists  
- relational tables  

These structures avoid arbitrary nesting → deepcopy becomes unnecessary.

---

## 🟦 6. Safe Practice Rules (Adopt These)

### ✅ Rule 1: Avoid nested Python lists in analytics  
Flatten early, or convert to pandas.

### ✅ Rule 2: If nesting is unavoidable, assume `.copy()` is unsafe  
Do not trust shallow copies for nested data.

### ✅ Rule 3: Mutation inside a nested object propagates  
Changing elements inside an inner list changes all shallow copies.

### ✅ Rule 4: Rebinding a nested list does NOT propagate  
Replacing the entire inner list with a new one isolates the change.

### ✅ Rule 5: Use `deepcopy` **only** when nested structures cannot be avoided  
But aim to design your code so this is rare.

---

## 🟩 7. Quick Reference Table

| Operation | Copies Outer List? | Copies Inner Lists? | Safe for Nested Data? |
|----------|---------------------|----------------------|------------------------|
| `new = old.copy()` | ✅ Yes | ❌ No | ❌ No |
| `new = old[:]` | ✅ Yes | ❌ No | ❌ No |
| `new = list(old)` | ✅ Yes | ❌ No | ❌ No |
| `new = deepcopy(old)` | ✅ Yes | ✅ Yes | ⭐ Yes |

---

## 🟦 8. Helper Box Summary

- `.copy()` copies the **container**, not the **contents**.  
- Nested objects remain **shared** and therefore **dangerous**.  
- Replace inner lists (rebinding) → safe  
- Modify inside an inner list → unsafe  
- Deep copy solves the problem but should be unnecessary in well-designed analytics code.  
- Flat structures avoid all these issues — which is why pros use them.

````python
# --------------------------------------------------------------
# 🔧 Helper: Tuple Inspector (Index Map + Mutability Checker)
# --------------------------------------------------------------

def show_tuple_info(tpl, show_negative=True, show_types=True):
    """
    Display a structured index map for a tuple, including:
      - positive and negative indices
      - value at each index
      - type of each element
      - mutability warning (e.g., tuple containing lists)
    
    Parameters
    ----------
    tpl : tuple
        The tuple to inspect.
    show_negative : bool, default True
        Whether to show negative indices.
    show_types : bool, default True
        Whether to show type information for each element.
    
    Usage
    -----
    show_tuple_info((1, 2, 3))
    show_tuple_info((1, [2, 3], (4, 5)))
    show_tuple_info(("a", "b", "c"), show_negative=False)
    """

    # ----------------------------------------------------------
    # 1. Validate input
    # ----------------------------------------------------------
    if not isinstance(tpl, tuple):
        raise TypeError(
            f"show_tuple_info expects a tuple, got {type(tpl).__name__}"
        )

    n = len(tpl)
    print(f"Tuple length: {n}")
    print()

    # ----------------------------------------------------------
    # 2. Build header
    # ----------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    header_cols.append("mutable?")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += f"{'value':>12} {'mutable?':>10}"

    print(header)
    print("-" * max(len(header), 40))

    # ----------------------------------------------------------
    # 3. Detect mutability of each element
    # ----------------------------------------------------------
    def is_mutable(obj):
        """Return True if object is a known mutable Python type."""
        return isinstance(obj, (list, dict, set, bytearray))

    # ----------------------------------------------------------
    # 4. Print each element with indices and mutability flag
    # ----------------------------------------------------------
    for i, val in enumerate(tpl):
        neg = i - n if show_negative else None
        type_name = type(val).__name__ if show_types else ""
        mutable_flag = "yes" if is_mutable(val) else "no"

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{repr(val):>12} {mutable_flag:>10}"

        print(row)

    # ----------------------------------------------------------
    # 5. Summary warnings (very helpful in practice)
    # ----------------------------------------------------------
    print("\nSummary:")
    if any(isinstance(v, list) for v in tpl):
        print("⚠ Warning: This tuple contains one or more lists.")
        print("  → The tuple is immutable, but the inner lists ARE mutable.")
        print("  → Changing an inner list will change all references to it.\n")
    else:
        print("✓ No mutable inner objects detected. Tuple is fully immutable.\n")


# ------------------------------
# Example:
# ------------------------------
show_tuple_info((1, 2, (3, 4)))
show_tuple_info((1, [2, 3], 4))
````

# 🟦 Helper Box — Slicing: Lists vs. Tuples  
Understanding how slicing behaves for **mutable** vs **immutable** sequence types is essential.  
This helper clarifies exactly what happens under the hood — and why tuple slicing is so safe and predictable.

---

## ❗ First, a universal rule (often overlooked)
> **Slicing always *creates* a new sequence object, but it does *not* save it unless you assign it to a variable.**

```python
numbers = [10, 20, 30, 40]
numbers[1:3]        # slice created, used, discarded
subset = numbers[1:3]  # slice saved
```

The same is true for tuples.

This is important: slicing alone never mutates anything — it just produces a temporary result unless you store it.

---

## ✅ 1. Slicing a List  
### What happens?

```python
lst = [10, 20, 30, 40]
slice_lst = lst[1:3]     # [20, 30]
```

- **Slicing always creates a new list object.**
- The original list is unchanged.
- Because lists are **mutable**, you can modify the slice:

```python
slice_lst[0] = 999      # [999, 30]
# lst remains [10, 20, 30, 40] — completely unaffected
```

### Key property  
Even though slicing is safe, **slice assignment** can modify the original list:

```python
lst[1:3] = [777, 888]   # mutates original list
```

This is a powerful feature — but also a source of bugs if used carelessly.

---

## ✅ 2. Slicing a Tuple  
### What happens?

```python
colors = ('red', 'green', 'blue')
subset = colors[1:3]      # ('green', 'blue') → created but not saved

new_colors = colors[1:3]  # saved version

```

- **Slicing creates a new tuple object.**
- The original tuple remains unchanged.
- You can assign the slice to a variable to keep it (just like a list slice).

### But the critical difference:
Because tuples are **immutable**:

- You *cannot* modify the original tuple  
- You *cannot* modify the slice  
- You *cannot* use slice assignment  
- The new tuple is completely isolated and safe  

Slicing tuples is simpler and safer because there is *no* possibility of mutation anywhere.

---

## 🧠 3. Why the Behavior Is Different?

| Operation                    | Lists (Mutable)                      | Tuples (Immutable)                     |
|-----------------------------|---------------------------------------|----------------------------------------|
| Slicing                     | Creates a **new list**                | Creates a **new tuple object**         |
| Assigning back to a slice  | Allowed (mutates original)            | Not allowed (TypeError)                |
| Modifying the slice itself  | Allowed                                | Not allowed                            |
| Risk of unintended mutation | Medium                                 | Zero                                   |

---

## 🏁 4. The Key Concept in One Sentence  
> **Slicing always returns a new sequence object — but only lists allow slice assignment that can mutate the original. Tuples never allow mutation, so slicing is always safe.**

---

## ⭐ 5. The Mental Model (recommended for mastery)

Think of slicing as:

### **“Copying a portion of the sequence into a new container.”**

- For lists → a new *list container*  
- For tuples → a new *tuple object*  

And because tuples cannot change, this new object is **fully independent** and **risk-free**.

---

## ✔ Quick Verification

```python
t = (1, 2, 3, 4)
s = t[1:3]

print(s)       # (2, 3)
print(t)       # (1, 2, 3, 4)
```

Both remain intact. Nothing can mutate either one.

---

## 📌 Final takeaway  
Whether slicing a list or tuple:

- It **never** changes the original sequence  
- It **always** produces a new object  
- It **must be assigned** if you want to keep it  
- Only lists support slice assignment  
- Tuples are 100% mutation-proof

````python
# 🔧 Helper: `show_range_index_map(rng, show_negative=True)`


def show_range_index_map(rng, show_negative=True):
    """
    Print a table of indices and values for a Python range object.

    Parameters
    ----------
    rng : range
        The range object to inspect.
    show_negative : bool, default True
        If True, show the negative index for each element
        (as if the range were a sequence like a list or tuple).

    Usage
    -----
    show_range_index_map(range(5))
    show_range_index_map(range(10, 0, -2))
    """
    # ---------------------------------------------------------
    # 1. Validate input type
    # ---------------------------------------------------------
    if not isinstance(rng, range):
        raise TypeError(
            f"show_range_index_map expects a range, got {type(rng).__name__}"
        )

    # ---------------------------------------------------------
    # 2. Basic range properties
    # ---------------------------------------------------------
    n = len(rng)
    start = rng.start
    stop = rng.stop
    step = rng.step

    print(f"Range: range({start}, {stop}, {step})")
    print(f"Length: {n}")
    print()

    if n == 0:
        print("This range is empty (no values generated).")
        return

    # ---------------------------------------------------------
    # 3. Build header
    # ---------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    header_cols.append("value")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "value" in header_cols:
        header += f"{'value':>6}"

    print(header)
    print("-" * max(len(header), 20))

    # ---------------------------------------------------------
    # 4. Iterate over the range and print index/value mapping
    # ---------------------------------------------------------
    for i, val in enumerate(rng):
        # Negative index follows standard Python convention:
        # last element → -1, second to last → -2, etc.
        neg_index = i - n if show_negative else None

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        row += f"{val:>6}"
        print(row)


show_range_index_map(range(10, 0, -2))
show_range_index_map(range(20))               # forward range
show_range_index_map(range(50, 20, -5))       # backward range
show_range_index_map(range(6), show_negative=False)

example_range = range(10, 0, -2)
show_range_index_map(example_range)
````
