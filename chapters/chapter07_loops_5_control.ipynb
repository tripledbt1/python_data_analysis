# Chapter 07 — Loops  
## Section 7.05 — Loop Control

### Preview
Loop control statements allow you to **modify the normal flow of iteration**, enabling early exits, selective skipping, placeholder logic, and robust error handling.

Four essential loop control mechanisms include:

- **break** — exit a loop immediately  
- **continue** — skip the rest of the current iteration  
- **pass** — placeholder that does nothing  
- **try / except** — handle errors without stopping the loop  

These tools are critical for:

- Data cleaning  
- Error-tolerant ETL pipelines  
- Complex filtering  
- Avoiding infinite loops  
- Creating maintainable, incremental logic  

This section adds new helper functions for debugging, error tracing, and controlled loop execution.

---

  - `break_trace()`

  - `try_except_trace()`  
  - `affordability_filter()`  
These provide full transparency and robustness when working with real-world, messy datasets.

# Summary

- Loop control changes the **execution path** inside loops to handle selective skipping, early termination, and error-tolerant iteration.  
- `break` stops the loop immediately.  
- `continue` skips the current iteration only.  
- `pass` does nothing; used as a placeholder.  
- `try`/`except` creates fault-tolerant loops that keep running even when data issues occur.


````python
# ------------------------------------------------------------------
# 1. The `break` Statement
# `break` stops the loop entirely, regardless of the iteration index
# ------------------------------------------------------------------
# Behavior
# - Loop ends as soon as the condition triggers  
# - Remaining items are not processed  


# Example — Stop when revenue exceeds 2000  

subtotals = [200.98, 350.97, 500, 400, 300, 600, 150, 100, 90]

revenue = 0                          # accumulator to track what the loop produces

for amount in subtotals:
    revenue += amount                # looping through subtotals list and adding revenue
    print(round(revenue,2))          # Visualize each loop
    if revenue > 2000:
        print("Target reached. Stopping early.")
        break

print("Final revenue:", revenue)


# -------------------------------------
# 🟦 Helper — Break Trace Visualizer
# -------------------------------------

def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0
    for idx, value in enumerate(iterable):
        running += value
        print(f"Iteration {idx}: value = {value}, running = {running}")
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break

# Example Usage:
break_trace(subtotals, lambda total: total > 2000)


# Use :.2f formatting in helpers when working with currency or financial data.
# This improves readability and prevents confusion caused by floating-point precision.


# -------------------------------------
# 🟦 Helper — Break Trace Visualizer
# -------------------------------------

def break_trace(iterable, condition_fn):
    """
    Shows exactly when 'break' would trigger inside a loop.
    Values are formatted to two decimal places for currency-style data.
    """
    print("BREAK TRACE\n-----------")
    
    running = 0.0
    for idx, value in enumerate(iterable):
        running += value
        print(
            f"Iteration {idx}: "
            f"value = ${value:.2f}, "
            f"running = ${running:.2f}"
        )
        
        if condition_fn(running):
            print(f"→ BREAK triggered at iteration {idx}")
            break
            
# Example Usage:
break_trace(subtotals, lambda total: total > 2000)
````

````python
# ------------------------------------------------------------------
# 1. The `break` Statement — Early Exit from a While Loop
# ------------------------------------------------------------------
# `break` immediately stops a loop, even if the loop condition
# would otherwise continue to evaluate as True.
#
# Use `break` when:
# - you want to stop early once a secondary condition is met
# - continuing the loop no longer makes sense
# - you want to prevent runaway or unrealistic scenarios
# ------------------------------------------------------------------


stock_portfolio = 21_070   # accumulator: tracks the evolving portfolio value (loop state)
year_counter = 0          # counter: tracks how many iterations (years) have passed


# This loop continues WHILE the portfolio is below the target
while stock_portfolio < 150_000:
    
    # Calculate annual investment growth (15% return)
    investment_income = stock_portfolio * 0.15
    
    # Update portfolio value (state change)
    stock_portfolio += investment_income           # accumulator update
    
    # Update loop counter
    year_counter += 1                             # counter update
    
    # Observe the evolving state of the system
    print(
        f"My balance is ${stock_portfolio:,.2f} "
        f"in year {year_counter}"
    )
    
    # SAFETY CHECK:
    # If 10 years have passed and the goal is still not reached,
    # exit the loop early instead of continuing indefinitely.
    if year_counter >= 10:
        print("Goal not reached in 10 years — savings rate may be too low.")
        break

# Teaching takeaway:
# - The while-condition controls whether the loop *can* continue
# - The `break` controls whether the loop *should* continue
````

````python
# ------------------------------------------------------------------
# 2. The `continue` Statement  
# `continue` skips the remainder of the current iteration but continues the loop
# ------------------------------------------------------------------
# Behavior
# - `"skis"` and `"skiboots"` are skipped  
# - All other items processed normally
# ------------------------------------------------------------------

# Example — Filter out items containing “ski”


inventory = ["snowboard", "skis", "helmet", "skiboots", "bindings"]
filtered_inventory = []

for item in inventory:
    # filter condition: skip the current item if it contains the substring "ski"
    if "ski" in item:
        # continue means: "do NOT run the rest of the loop body for this item"
        # and move to the next iteration immediately
        continue
    filtered_inventory.append(item)  # .append() empty list accumulator

print(filtered_inventory)
print()


# ---------------------------------------
# 🟦 Helper — Continue Trace Visualizer
# ---------------------------------------

def continue_trace(iterable, skip_fn):
    """
    Prints all iterations and highlights which ones are skipped via 'continue'.
    """
    print("CONTINUE TRACE\n----------------")
    
    for idx, item in enumerate(iterable):
        if skip_fn(item):
            print(f"Iteration {idx}: SKIPPED ({item!r})")
            continue
        
        print(f"Iteration {idx}: processed {item!r}")


# Example Usage:
continue_trace(inventory, lambda x: "ski" in x)
````

````python
# ------------------------------------------------------------------
# The `pass` Statement
# `pass` is a placeholder that does NOTHING, but keeps the code valid.
# ------------------------------------------------------------------
# Key idea:
# - `pass` satisfies Python’s requirement that a block cannot be empty
# - It does NOT skip an iteration (that’s `continue`)
# - It does NOT stop a loop (that’s `break`)
# - It simply says: “I’ll fill this in later”
#
# Common use cases:
# - Stubbing logic during development
# - Sketching control flow before implementation
# - Keeping loops and conditionals syntactically valid
# - Incrementally building complex pipelines
# ------------------------------------------------------------------


products = ["Snowboard", "Boots", "Helmet"]

for p in products:
    if p == "Boots":
        pass    # Placeholder: no action yet, loop continues normally
    else:
        print("Processing:", p)


# NOTE:
# `pass` is not control flow.
# It is a structural placeholder.
#

# CONTROL FLOW MENTAL MODEL:
# - break     → exit the loop entirely
# - continue  → skip an iteration
# - pass      → do nothing (for now), but keep structure valid
````

````python
# ------------------------------------------------------------------
# Error Handling with `try` / `except`  
# These statements allow the loop to continue running after encountering an error
# ------------------------------------------------------------------
# Notes
# - Strings like `"74.99"` are converted successfully  
# - `0` triggers `ZeroDivisionError`  
# - `None` triggers `TypeError`  
# - Loop does not stop on errors  
# ------------------------------------------------------------------

# ---------------------------------------
# Example 1 — Problematic price list
# ---------------------------------------

price_list = [99.99, "74.99", 0, None, 20]

clean_prices = []

for element in price_list:
    try:
        # Step 1: convert to a numeric value
        numeric_price = float(element)
        
        # Step 2: use the validated value for calculations
        clean_prices.append(round(1 / numeric_price, 4))
    
    except ZeroDivisionError:                                  # zero division error
        print("ZeroDivisionError: price was zero. Skipping.")
    
    except (TypeError, ValueError):                            # multiple error types
        print(f"Invalid price format ({element}). Skipping.")

# ---------------------------------------
# 🟦 Helper — Error Trace Loop Inspector  
# Shows which items cause which errors.
# ---------------------------------------

def try_except_trace(iterable, transform_fn):
    """
    Executes transform_fn on each element, catching errors
    and reporting them with full iteration context.
    """
    print("TRY/EXCEPT TRACE\n-----------------")
    results = []
    
    for idx, value in enumerate(iterable):
        try:
            result = transform_fn(value)
            results.append(result)
            print(f"Iteration {idx}: OK → {result}")
        
        except Exception as e:
            print(f"Iteration {idx}: ERROR → {type(e).__name__}: {e}")
    
    return results


# Example Usage:
try_except_trace(price_list, lambda x: round(1 / float(x), 4))
print()

# ---------------------------------------
# Example 2 — Problematic ratings list
# ---------------------------------------

ratings = ["1 star", "5 star", 
           "3 star", "star 5", 
           "4 star"
]

for rating in ratings:
    try:
        print(int(rating[0]))
    except ValueError:
        print("incorrect data") 



# STUDENT NOTE:
# This helper shows that errors do NOT stop loops by default
# when handled with try/except.
#
# Each iteration is attempted independently:
# - Valid values are transformed and collected
# - Invalid values are reported and skipped
#
# This mirrors real-world data cleaning workflows.
````

````python
# ------------------------------------------------------------------
# Error Handling with `try` / `except`  
# These statements allow the loop to continue running after encountering an error
# ------------------------------------------------------------------
# Notes
# - Strings like `"74.99"` are converted successfully  
# - `0` triggers `ZeroDivisionError`  
# - `None` triggers `TypeError`  
# - Loop does not stop on errors  
# ------------------------------------------------------------------

# ---------------------------------------
# Example 3 — Problematic price list
# ---------------------------------------


prices = [5.99, 99.99, "74.99", 0, None, 20, 24.99]
budget = 50

for price in prices:
    if price is None:
        continue

    try:
        numeric_price = float(price)  # convert early (handles strings)

        if numeric_price == 0:
            print("This product is free.")
            continue

        affordable_quantity = budget // numeric_price
        print(f"I can buy {int(affordable_quantity)} of these.")

    except ValueError:
        print("That is not a number.")
    except TypeError:
        # TypeError is unlikely once you skip None, but included for completeness
        print("Invalid type for price.")


# ---------------------------------------
# 🟦 Helper — Robust Affordability Calculator  
# ---------------------------------------

def affordability_filter(prices, budget):
    """
    Returns a list of (idx, raw_price, numeric_price, quantity) for prices that can be evaluated.
    Skips None, non-numeric, and handles zero safely.
    """
    results = []

    for idx, price in enumerate(prices):
        if price is None:
            continue

        try:
            numeric_price = float(price)

            if numeric_price == 0:
                results.append((idx, price, numeric_price, "FREE"))
                continue

            qty = int(budget // numeric_price)
            results.append((idx, price, numeric_price, qty))

        except (ValueError, TypeError):
            results.append((idx, price, None, "INVALID"))

    return results


# Example Usage:
rows = affordability_filter(prices, budget)

for idx, raw, num, qty in rows:
    print(f"Index {idx}: raw={raw!r} | numeric={num!r} | qty={qty}")



# ---------------------------------------
# 🟦 Helper — Error Trace Loop Inspector  
# Shows which items cause which errors.
# ---------------------------------------

def try_except_trace(iterable, transform_fn):
    """
    Executes transform_fn on each element, catching errors
    and reporting them with full iteration context.
    """
    print("TRY/EXCEPT TRACE\n-----------------")
    results = []
    
    for idx, value in enumerate(iterable):
        try:
            result = transform_fn(value)
            results.append(result)
            print(f"Iteration {idx}: OK → {result}")
        
        except Exception as e:
            print(f"Iteration {idx}: ERROR → {type(e).__name__}: {e}")
    
    return results


# Example Usage:
try_except_trace(
    prices,
    lambda p: "FREE" if float(p) == 0 else int(budget // float(p))
)

````

````python
# --------------------------------------------------------
# ✅ Analysis-Ready Affordability Toolkit
# - robust parsing (None, strings, bad values)
# - flags FREE / INVALID
# - computes quantity you can buy under a budget
# - returns both detailed rows + a summary dictionary
# --------------------------------------------------------

def affordability_analyze(prices, budget):
    """
    Analyze a mixed-quality price list against a budget.

    Returns
    -------
    rows : list[dict]
        One record per input element with diagnostics:
        idx, raw, numeric_price, status, quantity
    summary : dict
        Counts + quick stats to support analysis decisions.
    """
    rows = []

    # Summary counters
    counts = {
        "OK": 0,
        "FREE": 0,
        "INVALID": 0,
        "SKIPPED_NONE": 0
    }

    valid_prices = []
    valid_quantities = []

    for idx, raw in enumerate(prices):
        # 1) Skip None cleanly (common missing-data pattern)
        if raw is None:
            counts["SKIPPED_NONE"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": None,
                "status": "SKIPPED_NONE",
                "quantity": None
            })
            continue

        # 2) Attempt numeric conversion (handles floats/ints/strings like "74.99")
        try:
            numeric_price = float(raw)
        except (TypeError, ValueError):
            counts["INVALID"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": None,
                "status": "INVALID",
                "quantity": None
            })
            continue

        # 3) Handle zero explicitly (avoid division-by-zero)
        if numeric_price == 0:
            counts["FREE"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": 0.0,
                "status": "FREE",
                "quantity": "FREE"
            })
            continue

        # 4) Negative prices are “valid numbers” but typically invalid business data
        if numeric_price < 0:
            counts["INVALID"] += 1
            rows.append({
                "idx": idx,
                "raw": raw,
                "numeric_price": numeric_price,
                "status": "INVALID_NEGATIVE",
                "quantity": None
            })
            continue

        # 5) OK path: compute how many you can buy (floor division)
        qty = int(budget // numeric_price)

        counts["OK"] += 1
        valid_prices.append(numeric_price)
        valid_quantities.append(qty)

        rows.append({
            "idx": idx,
            "raw": raw,
            "numeric_price": numeric_price,
            "status": "OK",
            "quantity": qty
        })

    # -----------------------------
    # Build summary stats
    # -----------------------------
    summary = {
        "budget": budget,
        "total_inputs": len(prices),
        "counts": counts,
        "valid_price_count": len(valid_prices),
        "min_valid_price": min(valid_prices) if valid_prices else None,
        "max_valid_price": max(valid_prices) if valid_prices else None,
        "avg_valid_price": (sum(valid_prices) / len(valid_prices)) if valid_prices else None,
        "avg_quantity": (sum(valid_quantities) / len(valid_quantities)) if valid_quantities else None,
        "max_quantity": max(valid_quantities) if valid_quantities else None,
    }

    return rows, summary


# --------------------------------------------------------
# ✅ Pretty-printer (optional, keeps your notebook readable)
# --------------------------------------------------------

def print_affordability_report(rows, summary):
    print("AFFORDABILITY REPORT\n--------------------")
    print(f"Budget: ${summary['budget']:,.2f}")
    print(f"Total inputs: {summary['total_inputs']}\n")

    # High-level summary
    c = summary["counts"]
    print("Counts:")
    for k, v in c.items():
        print(f"  {k:>12}: {v}")
    print()

    # Quick stats
    if summary["valid_price_count"] > 0:
        print("Valid price stats:")
        print(f"  min price : ${summary['min_valid_price']:,.2f}")
        print(f"  max price : ${summary['max_valid_price']:,.2f}")
        print(f"  avg price : ${summary['avg_valid_price']:,.2f}")
        print(f"  avg qty   : {summary['avg_quantity']:.2f}")
        print(f"  max qty   : {summary['max_quantity']}")
    else:
        print("No valid prices found.")
    print()

    # Row-by-row detail (traceability)
    print("Row detail:")
    for r in rows:
        idx = r["idx"]
        raw = r["raw"]
        num = r["numeric_price"]
        status = r["status"]
        qty = r["quantity"]

        # consistent formatting
        raw_txt = repr(raw)
        num_txt = "None" if num is None else f"{num:.2f}"
        print(f"  idx {idx:>2}: raw={raw_txt:<10} | numeric={num_txt:>6} | status={status:<15} | qty={qty}")


# --------------------------------------------------------
# ✅ Example usage (your dataset)
# --------------------------------------------------------

prices = [5.99, 99.99, "74.99", 0, None, 20, 24.99]
budget = 50

rows, summary = affordability_analyze(prices, budget)
print_affordability_report(rows, summary)
````

# 6. 🟦 Helper — Robust Affordability Calculator

```python
def affordability_filter(prices, budget):
    """
    Robust affordability calculator:
    - skips None values
    - converts string prices
    - handles invalid formats safely
    - avoids division-by-zero issues
    """
    
    affordable = []
    
    for idx, price in enumerate(prices):
        try:
            if price is None:
                continue
            
            numeric_price = float(price)
            
            if numeric_price <= budget:
                affordable.append(numeric_price)
        
        except Exception:
            continue  # safely skip invalid price formats
    
    return affordable
```

# When to Use Loop Control Statements

### Use `break` when:
- A target or threshold is reached  
- Further processing is unnecessary  
- You are preventing a runaway or infinite loops early  

### Use `continue` when:
- Filtering datasets  
- Skipping invalid or unwanted entries, i.e., excluding values you don't want to process in a loop
- Differentiating “skippable” vs “processable” values

### Use `pass` when:
- Structuring logic before implementation, i.e., avoiding run errors with incomplete code logic
- Writing placeholder blocks, i.e., serving as a placeholder for future code  
- Maintaining syntactic validity  

### Use `try`/`except` when:
- Data may be incomplete, inconsistent, or malformed  
- You need to avoid interruption during iteration
- You need help with error and exception handling  
- Building robust ETL or cleaning pipelines
