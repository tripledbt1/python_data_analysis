## 📘 Python Data Types: Single Value vs Multiple Value

The table below summarizes Python data types based on whether they store a **single value** or **multiple values**. This aligns with the conceptual framework introduced in the lecture notes and your handwritten outline.

| **Single Value** | **Multiple Value** |
|------------------|--------------------|
| **Numeric** <br> • `int` <br> • `float` <br> • `complex` | **Sequence Types** <br> • `list` <br> • `tuple` <br> • `range` |
| **Text** <br> • `str` | **Mapping Type** <br> • `dict` |
| **Boolean** <br> • `bool` (True, False) | **Set Types** <br> • `set` <br> • `frozenset` |
| **NoneType** <br> • `None` (placeholder for no value) | *(No additional categories)* |

---

### Notes:
- **Single-value data types** represent one atomic value, such as a number, text string, or boolean.  
- **Multiple-value data types** store collections of values. These structures allow iteration and support a wide range of data manipulation patterns used in analysis workflows.  
- `NoneType` exists as a placeholder that represents “no value.”
---

```python
# -------------------------------------------------------------
# Checking the data type of various Python objects
# -------------------------------------------------------------

# NoneType
empty_value = None
print(type(empty_value))        # <class 'NoneType'>

# Numeric Types
x_int = 5
x_float = 5.24
x_complex = 5 + 2j
print(type(x_int))              # <class 'int'>
print(type(x_float))            # <class 'float'>
print(type(x_complex))          # <class 'complex'>

# Text Type
name = "snowboard"
print(type(name))               # <class 'str'>

# Sequence Types
numbers_list = [1, 3, 5, 7, 9]
print(type(numbers_list))       # <class 'list'>

items_tuple = ("snowboard", "skis")
print(type(items_tuple))        # <class 'tuple'>

num_range = range(1, 10, 2)
print(type(num_range))          # <class 'range'>

# Mapping Type. Maps keys to values for efficient information retrieval
equipment = {"snowboard": 24, "skis": 17}
print(type(equipment))          # <class 'dict'>

# Boolean Type
flag = True
print(type(flag))               # <class 'bool'>
banner = False
print(type(banner))               # <class 'bool'>

# Set Types. Represents a collection of unique, non-duplicate values.
# A set may contain only unique elements.
gear_set = {"snowboard", "skis"}  # {} + values = set. This is an empty set()
print(type(gear_set))           # <class 'set'>

frozen_gear = frozenset({"snowboard", "skis"})  # must call frozenset({}) function
print(type(frozen_gear))        # <class 'frozenset'>
```

```python
# -------------------------------------------------------------
# 🔹 Sets automatically remove duplicates
# -------------------------------------------------------------

numbers = [1, 2, 2, 3, 3, 3]
unique_numbers = set(numbers)
print(unique_numbers)

# Duplicate values inside {} will also be removed
example = {"a", "a", "b"}
print(example)
```

### 🧩 Syntax Table (“{ }”, “set()”, tuple rules, frozenset”)

| Syntax                  | What it Creates        | Why                                                |
| ----------------------- | ---------------------- | -------------------------------------------------- |
| `{"a", "b"}`            | **set**                | curly braces + values = set                        |
| `("a", "b")`            | **tuple**              | commas inside parentheses = tuple                  |
| `"a", "b"`              | **tuple**              | commas alone create tuple                          |
| `frozenset({"a", "b"})` | **frozen set**         | must call the frozenset() function                 |
| `{}`                    | **dictionary** (empty) | Python reserves `{}` for empty dict, not empty set |
| `set()`                 | **empty set**          | must call set() to create an empty set             |


- **{}** This is not a set. It is an empty dictionary.
- To make an empty set: **set()** 


```python
# -------------------------------------------------------------
# Type Conversion (valid and invalid conversions)
# -------------------------------------------------------------

# Valid type conversion: converting a string to an integer
num_string = "123"
converted_int = int(num_string)      # succeeds because '123' is numeric text
print(converted_int, type(converted_int))

# Valid type conversion: converting an integer to a string
integer = (456)
int_string = str(integer)
print (int_string, type(int_string))

# Valid type conversion: converting an list to a tuple
list = [1, 2, 3]
list_tuple = tuple([1, 2, 3])
print(list_tuple, type(list_tuple))


# Invalid type conversion: converting a list to an integer
sample_list = [1, 2, 3]

# This will raise a TypeError because a list has multiple values.
# Uncomment to observe the error:
# int(sample_list)
```

### 🧩 Conversions to a set removes duplicates

| Conversion      | Removes duplicates? | Explanation                                  |
| --------------- | ------------------- | -------------------------------------------- |
| `set(list)`     | ✔ Yes               | All repeats removed                          |
| `set(tuple)`    | ✔ Yes               | Tuple values added once                      |
| `set(range)`    | ✔ Yes               | Range has no duplicates, so unchanged        |
| `set("string")` | ✔ Yes               | Converts characters into unique values       |
| `set(dict)`     | ✔ Yes               | Converts dictionary *keys* into a unique set |

⚠️ Note: Creating a set directly with braces will also drop duplicates.

```python
# -------------------------------------------------------------
# Iterable vs Non-Iterable examples
# -------------------------------------------------------------

# Text strings are iterable because you can loop through characters
for character in "snowboard":
    print(character)

# Lists, tuples, ranges, sets, and dictionaries are iterable
for item in [1, 3, 5]:
    print(item)

# Non-iterable example: integers cannot be looped through
# The following will raise a TypeError:
# for i in 5:
#     print(i)
```

```python
# -------------------------------------------------------------
# 🔹 Mutability Examples
# -------------------------------------------------------------

# Mutable: lists, sets, dictionaries can be modified directly
mutable_list = [1, 2, 3]
mutable_list.append(4)          # modifies the original list
print(mutable_list)

mutable_set = {"snowboard", "skis"}
mutable_set.add("helmet")       # adds a new item
print(mutable_set)

mutable_dict = {"snowboard": 24}
mutable_dict["skis"] = 17       # adds a new key-value pair
print(mutable_dict)

# Immutable: integers, floats, strings, booleans, tuples, frozensets
immutable_string = "hello"
# immutable_string[0] = 'H'     # would raise a TypeError (cannot change characters)

immutable_tuple = (1, 2, 3)
# immutable_tuple[0] = 5        # TypeError if uncommented
```

## 🔍 Key Points to Remember

- **`type()` is essential for debugging** because it identifies the precise data type of any object.  
- **Numeric types (int, float, complex)** behave differently in arithmetic and analysis tasks.  
- **Strings (`str`) are iterable**, even though they are considered single-value data types.  
- **Lists, tuples, and ranges** are examples of sequence types. They differ in mutability, which will matter later in the course.  
- **Dictionaries (`dict`) store key-value pairs**, which is fundamental for structured data and JSON processing.  
- **Booleans (`bool`) evaluate to `True` or `False`** and often appear as the output of comparisons or logical expressions.  
- **Sets** store unique values with no duplicates, while **frozen sets** are the immutable (read-only) version.  
- Curly braces `{}` create:  
  - a **set** if the values are comma-separated  
  - a **dictionary** if the values are key-value pairs  
- An empty set must be created with `set()` because `{}` defaults to an empty dictionary.

## 🧩 Mutability Comparison (Mutable vs Immutable)

Understanding mutability is important because it determines whether an object can be changed after creation. Mutable objects allow direct updates, while immutable objects must be replaced entirely when modified.

| **Category**     | **Type**        | **Mutable?** | **Example**                    |
|------------------|------------------|--------------|--------------------------------|
| Numeric          | `int`            | No           | `5`                            |
|                  | `float`          | No           | `5.24`                         |
|                  | `complex`        | No           | `5 + 2j`                       |
| Text             | `str`            | No           | `"snowboard"`                  |
| Boolean          | `bool`           | No           | `True`, `False`                |
| NoneType         | `NoneType`       | No           | `None`                         |
| Sequence         | `tuple`          | No           | `("a", "b")`                   |
|                  | `range`          | No           | `range(1, 10)`                 |
|                  | `list`           | Yes          | `[1, 2, 3]`                    |
| Mapping          | `dict`           | Yes          | `{"a": 1}`                     |
| Set Types        | `set`            | Yes          | `{"a", "b"}`                   |
|                  | `frozenset`      | No           | `frozenset({"a", "b"})`        |

**Key idea:**  
Mutable types can be modifed after their creation, while immutable data types cannot be changed without being overwritten.

## 🔁 Iterable vs Non-Iterable Data Types

A data type is iterable if you can loop through its elements using a `for` loop. Most multi-value data types are iterable.

| **Type**        | **Iterable?** | **Example**                    | **Notes**                                 |
|-----------------|----------------|--------------------------------|-------------------------------------------|
| `list`          | Yes           | `[1, 2, 3]`                    | Ordered and mutable                       |
| `tuple`         | Yes           | `(1, 2, 3)`                    | Ordered and immutable                     |
| `range`         | Yes           | `range(1, 5)`                  | Efficient sequence of numbers             |
| `dict`          | Yes           | `{"a": 1}`                     | Iterates over keys by default             |
| `set`           | Yes           | `{"a", "b"}`                   | Unordered, no duplicates                  |
| `frozenset`     | Yes           | `frozenset({"a", "b"})`        | Immutable version of a set                |
| `str`           | Yes           | `"abc"`                        | Loops character-by-character              |
| `int`           | No            | `5`                            | Not iterable                              |
| `float`         | No            | `3.14`                         | Not iterable                              |
| `bool`          | No            | `True`                         | Not iterable                              |
| `NoneType`      | No            | `None`                         | Not iterable                              |

**Key idea:**  
Iterability is independent of mutability. Strings are immutable but still iterable.

# 🧠 Key Takeaways

- Python includes several categories of data types, such as numeric, text, sequence, mapping, boolean, set, and NoneType.  
- The `type()` function identifies the data type of an object and is useful when debugging TypeErrors.  
- Type conversion uses the format `datatype(object)` and will only succeed when the underlying value is compatible.  
- Iterables are data types that can be looped through, including sequences, dictionaries, sets, and text strings.  
- Strings are considered to contain a single value, but they remain iterable because you can iterate through their characters.  
- Mutable data types (lists, sets, dictionaries) can be modified after creation. Immutable types (integers, floats, strings, tuples, booleans, frozensets) cannot be changed without recreating them.  
- Errors caused by incompatible type operations will stop program execution, which reinforces the value of validating data types early in a workflow.
