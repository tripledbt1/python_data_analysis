### 2.1 `show_indices(text)`: Visualize String Index Positions

This helper prints **positive** and **negative** indices under each character in a string.

```python
def show_indices(text):
    """
    Print a string with its positive and negative indices aligned under each character.
    
    Useful for:
    - Understanding and teaching indexing
    - Debugging slicing off-by-one errors
    """
    # Convert the string to a list of characters so we can align them
    chars = list(text)
    length = len(chars)
    
    # Build positive index row: 0   1   2   ...
    pos_indices = [str(i).rjust(3) for i in range(length)]
    
    # Build negative index row: -len ... -1
    neg_indices = [str(i - length).rjust(3) for i in range(length)]
    
    # Build the characters row (use repr-like display for clarity)
    char_row = []
    for ch in chars:
        # Represent spaces visibly
        if ch == " ":
            char_row.append(" ␣ ".center(3))
        elif ch == "\t":
            char_row.append("\\t".center(3))
        elif ch == "\n":
            char_row.append("\\n".center(3))
        else:
            char_row.append(ch.center(3))
    
    print("Text: ", repr(text))
    print("Chars:", "".join(char_row))
    print("Index:", "".join(pos_indices))
    print("Neg:  ", "".join(neg_indices))


# Example usage:
show_indices("super snowboard")
show_indices("Hello world")
```

### 2.2 `show_hidden(text)`: Reveal Whitespace and Non-Printable Characters

This helper focuses on **what is really in the string**, including spaces, tabs, and newlines.

```python
def show_hidden(text):
    """
    Reveal hidden characters (spaces, tabs, newlines) in a string.
    
    Prints:
    - The raw repr() of the string
    - A second line where spaces and special characters are replaced
      with visible markers.
    """
    # First, show the standard Python repr
    print("repr(text):", repr(text))
    
    # Build a "visualized" version
    visual = []
    for ch in text:
        if ch == " ":
            visual.append("␣")      # visible space symbol
        elif ch == "\t":
            visual.append("⭾")      # visible tab symbol
        elif ch == "\n":
            visual.append("⤶")      # visible newline symbol
        else:
            visual.append(ch)
    
    print("visual:   ", "".join(visual))


# Example usage:
show_hidden("Hello world")
show_hidden("Line 1\nLine 2\tTabbed")
show_hidden("  leading and  multiple   spaces  ")
```

These two helpers make indexing and whitespace **visible**, which is especially helpful when debugging slices or text cleaning.

### 3.1 `trace_decision(label, func)`: Wrap a Full Decision Tree

```python
def trace_decision(label, func):
    """
    Wrap a decision-making function with a labeled header.
    
    label: short description of the decision being traced
    func:  zero-argument function containing if/elif/else logic
    
    Usage pattern:
      def some_logic(...):
          def inner():
              # prints + if/elif/else live here
              ...
          trace_decision("Description", inner)
    """
    print("\n" + "=" * 60)
    print(f"DECISION TRACE: {label}")
    print("=" * 60)
    func()
```

**Example:**

```python
def decide_shipping(order_total, rush):
    def inner():
        print(f"order_total={order_total}, rush={rush}")
        
        if rush:
            print("Branch: if    → Express shipping")
        elif order_total >= 100:
            print("Branch: elif  → Free standard shipping")
        else:
            print("Branch: else  → Paid standard shipping")
    
    trace_decision("Shipping decision", inner)


# Try it:
decide_shipping(150, rush=True)
decide_shipping(100, rush=False)
decide_shipping(40,  rush=False)
```

### 3.2 `trace_condition(name, value)`: Show Condition Results

```python
def trace_condition(name, value):
    """
    Print the name of a condition and its Boolean result.
    
    Useful for:
    - Debugging multi-condition rules
    - Checking which conditions are True vs False
    - Explaining logic in teaching / walkthroughs
    """
    print(f"{name:30} → {value}")
```

**Example:**

```python
def debug_discount_rule(status, order_total, item):
    print(f"\nStatus={status!r}, order_total={order_total}, item={item!r}")
    discount_items = ["gloves", "hat"]
    
    cond_vip          = (status == "VIP")
    cond_high_total   = (order_total >= 100)
    cond_discount_item = (item in discount_items)
    
    trace_condition("status == 'VIP'",            cond_vip)
    trace_condition("order_total >= 100",         cond_high_total)
    trace_condition("item in discount_items",     cond_discount_item)
    
    eligible = cond_vip or (cond_high_total and cond_discount_item)
    trace_condition("Final rule result",          eligible)
    
    if eligible:
        print("Decision: Discount applied.")
    else:
        print("Decision: No discount.")


# Try it:

debug_discount_rule("REGULAR", 120, "gloves")
```

### 3.4 `trace_if(description, condition, if_true, if_false)`: Compact Single-Decision Helper

```python
def trace_if(description, condition, if_true, if_false):
    """
    Describe and evaluate a single if/else-style decision in a compact format.
    
    description: text explaining what is being checked
    condition:   Boolean expression to evaluate
    if_true:     label or message for the True branch
    if_false:    label or message for the False branch
    """
    print(f"\nChecking: {description}")
    print(f"Condition result: {condition}")
    
    if condition:
        print(f"Branch selected: {if_true}")
    else:
        print(f"Branch selected: {if_false}")
```

**Example:**

```python
def shipping_decision(order_total, rush):
    print(f"\norder_total={order_total}, rush={rush}")
    
    trace_if(
        description="Rush shipping requested",
        condition=rush,
        if_true="Express shipping",
        if_false="No rush shipping"
    )
    
    trace_if(
        description="Eligible for free standard shipping",
        condition=(order_total >= 100),
        if_true="Free standard shipping",
        if_false="Standard shipping (paid)"
    )


# Try it:
```

# 🔹 4. Slice Visualizer Helper (Included in This Notebook)  
This notebook also includes a custom helper tool, **visualize_slice()**, which provides a clear visual map of how slicing works for:  
- lists  
- tuples  
- strings  
- ranges  

The helper shows:  
- positive and negative indices  
- which elements a slice selects  
- how start/stop/step behave under the hood  
- normalized slice parameters  

This tool will support your intuition throughout the rest of the course and will be reused in later chapters involving loops, strings, dictionaries, and pandas.

---

Together, these skills build a strong foundation for the rest of Python’s sequence operations and prepare you for loops, data cleaning, and more advanced list manipulation later in the course.


````python
# --------------------------------------------------------------
# 🔧 Helper: Slice Visualizer (for lists, tuples, strings, ranges)
# --------------------------------------------------------------

def visualize_slice(seq, start=None, stop=None, step=None, show_negative=True, label="seq"):
    """
    Visualize how a slice seq[start:stop:step] selects elements
    from a sequence (list, tuple, string, range, etc.).

    Parameters
    ----------
    seq : sequence
        Any indexable, sliceable Python sequence (e.g., list, tuple, str, range).
    start : int or None
        Slice start index (can be negative or None).
    stop : int or None
        Slice stop index (can be negative or None).
    step : int or None
        Slice step (can be negative or None).
    show_negative : bool, default True
        Whether to display negative indices.
    label : str, default "seq"
        Optional label to show in the output header.

    Usage
    -----
    visualize_slice([10, 20, 30, 40, 50], 1, 4, 2)
    visualize_slice(range(10), 2, 9, 3)
    visualize_slice(('a', 'b', 'c', 'd', 'e'), 0, None, 2)
    """

    # ----------------------------------------------------------
    # 1. Basic validation
    # ----------------------------------------------------------
    try:
        n = len(seq)
    except TypeError:
        raise TypeError(
            f"visualize_slice expects a sequence with len(), got {type(seq).__name__}"
        )

    slice_obj = slice(start, stop, step)

    # Normalize slice parameters using slice.indices()
    norm_start, norm_stop, norm_step = slice_obj.indices(n)

    # Compute selected indices based on normalized values
    selected_indices = list(range(norm_start, norm_stop, norm_step)) if n > 0 else []

    # ----------------------------------------------------------
    # 2. High-level summary
    # ----------------------------------------------------------
    print(f"Sequence label : {label}")
    print(f"Type           : {type(seq).__name__}")
    print(f"Length         : {n}")
    print(f"Slice requested: {label}[{start}:{stop}:{step}]")
    print(f"Normalized to  : {label}[{norm_start}:{norm_stop}:{norm_step}]")
    print()

    if n == 0:
        print("This sequence is empty. No indices to visualize.")
        return

    # ----------------------------------------------------------
    # 3. Build index and value rows
    # ----------------------------------------------------------
    # We'll give each slot a fixed width for alignment
    cell_width = 5

    # Positive indices
    idx_row = "idx : "
    for i in range(n):
        idx_row += f"{i:>{cell_width-1}} "

    # Negative indices (optional)
    if show_negative:
        neg_row = "neg : "
        for i in range(n):
            neg_row += f"{i - n:>{cell_width-1}} "
    else:
        neg_row = None

    # Values (repr truncated for readability)
    val_row = "val : "
    for i in range(n):
        text = repr(seq[i])
        if len(text) > cell_width - 1:
            text = text[:cell_width - 2] + "…"
        val_row += f"{text:>{cell_width-1}} "

    # Selection markers
    sel_row = "sel : "
    selected_set = set(selected_indices)
    for i in range(n):
        marker = "^" if i in selected_set else ""
        sel_row += f"{marker:>{cell_width-1}} "

    # ----------------------------------------------------------
    # 4. Print the visualization
    # ----------------------------------------------------------
    print(idx_row)
    if neg_row is not None:
        print(neg_row)
    print(val_row)
    print(sel_row)
    print()

    # ----------------------------------------------------------
    # 5. Show the actual slice result
    # ----------------------------------------------------------
    result = seq[slice_obj]
    print(f"Result of {label}[{start}:{stop}:{step}] → {repr(result)}")
    print()


# --------------------------------------------------------------
# Example usage (you can comment these out in your notebook):
# --------------------------------------------------------------
# visualize_slice(['XS', 'S', 'M', 'L', 'XL', 'XXL'], 1, 5, 2, label="sizes")
# visualize_slice(range(10, 0, -2), None, None, None, label="r")
# visualize_slice(('r', 'g', 'b', 'y', 'k'), 1, 4, 1, label="colors")
````

## 🔧 Python Helper: `show_list_index_map`

```python
# --------------------------------------------------------------
# Universal List Index Map Helper (Supports Types & Nested Lists)
# --------------------------------------------------------------

def show_list_index_map(
    lst,
    show_types=True,
    show_negative=True,
    truncate=40,
    expand_nested=False
):
    """
    Print a table of indices and values for a Python list.
    
    Parameters
    ----------
    lst : list
        The list to inspect.
    show_types : bool, default True
        If True, show the type of each element.
    show_negative : bool, default True
        If True, show the negative index for each element.
    truncate : int or None, default 40
        If an integer, long repr() strings will be truncated to this length.
        If None, no truncation is applied.
    expand_nested : bool, default False
        If True, nested lists will be expanded one level and shown with
        indices of the form [outer][inner].
    
    Usage
    -----
    show_list_index_map(my_list)
    show_list_index_map(my_list, expand_nested=True)
    """
    # Basic input validation (student-friendly)
    if not isinstance(lst, list):
        raise TypeError(
            f"show_list_index_map expects a list, got {type(lst).__name__}"
        )
    
    n = len(lst)
    print(f"List length: {n}\n")
    
    # Build header dynamically based on settings
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    
    # Format header
    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += "value"
    
    print(header)
    print("-" * max(len(header), 20))
    
    # Helper function to truncate representation if needed
    def _repr_value(value):
        text = repr(value)
        if truncate is not None and len(text) > truncate:
            return text[:truncate - 3] + "..."
        return text
    
    # Iterate over the outer list
    for i, value in enumerate(lst):
        neg_index = i - n if show_negative else None
        type_name = type(value).__name__ if show_types else ""
        value_repr = _repr_value(value)
        
        # Build row string
        row = f"{i:>3} "
        if show_negative:
            row += f"{neg_index:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{value_repr}"
        
        print(row)
        
        # Optionally expand nested lists one level deep
        if expand_nested and isinstance(value, list):
            inner_len = len(value)
            print(f"      ↳ nested list (len={inner_len})")
            for j, inner_val in enumerate(value):
                inner_neg = j - inner_len if show_negative else None
                inner_type = type(inner_val).__name__ if show_types else ""
                inner_repr = _repr_value(inner_val)
                
                # Example format: [2][1]   -1    int  99
                nested_row = f"      [{i}][{j}] "
                if show_negative:
                    nested_row += f"{inner_neg:>4} "
                if show_types:
                    nested_row += f"{inner_type:>10} "
                nested_row += f"{inner_repr}"
                print(nested_row)


# -------------------------------------------
# ✅ Example Usage (Flat List)
# -------------------------------------------
sizes = ['XS', 'S', 'M', 'L', 'XL']
show_list_index_map(sizes)

# -------------------------------------------
# ✅ Example Usage (Nested List, Expanded)
# -------------------------------------------
orders = [
    [19.99, 5.99],
    [3.49, 9.99, 14.99],
    [60.00]
]

show_list_index_map(
    orders,
    show_types=True,
    show_negative=True,
    truncate=50,
    expand_nested=True
)
```

````python
# --------------------------------------------------------------
# 🔧 Helper: Tuple Inspector (Index Map + Mutability Checker)
# --------------------------------------------------------------

def show_tuple_info(tpl, show_negative=True, show_types=True):
    """
    Display a structured index map for a tuple, including:
      - positive and negative indices
      - value at each index
      - type of each element
      - mutability warning (e.g., tuple containing lists)
    
    Parameters
    ----------
    tpl : tuple
        The tuple to inspect.
    show_negative : bool, default True
        Whether to show negative indices.
    show_types : bool, default True
        Whether to show type information for each element.
    
    Usage
    -----
    show_tuple_info((1, 2, 3))
    show_tuple_info((1, [2, 3], (4, 5)))
    show_tuple_info(("a", "b", "c"), show_negative=False)
    """

    # ----------------------------------------------------------
    # 1. Validate input
    # ----------------------------------------------------------
    if not isinstance(tpl, tuple):
        raise TypeError(
            f"show_tuple_info expects a tuple, got {type(tpl).__name__}"
        )

    n = len(tpl)
    print(f"Tuple length: {n}")
    print()

    # ----------------------------------------------------------
    # 2. Build header
    # ----------------------------------------------------------
    header_cols = ["idx"]
    if show_negative:
        header_cols.append("neg")
    if show_types:
        header_cols.append("type")
    header_cols.append("value")
    header_cols.append("mutable?")

    header = ""
    if "idx" in header_cols:
        header += f"{'idx':>3} "
    if "neg" in header_cols:
        header += f"{'neg':>4} "
    if "type" in header_cols:
        header += f"{'type':>10} "
    header += f"{'value':>12} {'mutable?':>10}"

    print(header)
    print("-" * max(len(header), 40))

    # ----------------------------------------------------------
    # 3. Detect mutability of each element
    # ----------------------------------------------------------
    def is_mutable(obj):
        """Return True if object is a known mutable Python type."""
        return isinstance(obj, (list, dict, set, bytearray))

    # ----------------------------------------------------------
    # 4. Print each element with indices and mutability flag
    # ----------------------------------------------------------
    for i, val in enumerate(tpl):
        neg = i - n if show_negative else None
        type_name = type(val).__name__ if show_types else ""
        mutable_flag = "yes" if is_mutable(val) else "no"

        row = f"{i:>3} "
        if show_negative:
            row += f"{neg:>4} "
        if show_types:
            row += f"{type_name:>10} "
        row += f"{repr(val):>12} {mutable_flag:>10}"

        print(row)

    # ----------------------------------------------------------
    # 5. Summary warnings (very helpful in practice)
    # ----------------------------------------------------------
    print("\nSummary:")
    if any(isinstance(v, list) for v in tpl):
        print("⚠ Warning: This tuple contains one or more lists.")
        print("  → The tuple is immutable, but the inner lists ARE mutable.")
        print("  → Changing an inner list will change all references to it.\n")
    else:
        print("✓ No mutable inner objects detected. Tuple is fully immutable.\n")


# ------------------------------
# Example:
# ------------------------------
show_tuple_info((1, 2, (3, 4)))
show_tuple_info((1, [2, 3], 4))
````

```python
def transform_nested_list(nested, transform_fn):
    """
    Applies a user-defined transformation to each element in a nested list.

    Parameters
    ----------
    nested : list[list]
        A list of lists (e.g., customers → orders).
    
    transform_fn : callable
        Function applied to each inner element.

    Returns
    -------

    Notes
    -----
    - General version of Maven's multi-order discount logic (pp. 32–33)
    - Works with any transformation: tax, normalization, scaling, flags, etc.
    """
    
    result = []
    
    for i, inner_list in enumerate(nested):      # outer loop
        transformed_inner = []
        
        for j, value in enumerate(inner_list):   # inner loop
            transformed_inner.append(transform_fn(value))
        
        result.append(transformed_inner)
    
    return result
```

# 🟦 Helper: Cartesian Product Generator  
*(Useful for combinations, SKU creation, pairing simulations)*

```python
def cartesian_product(a, b):
    """
    Returns the Cartesian product of two iterables as a list of tuples.
    Equivalent to the nested-loop pairings in Maven's examples.
    """
    return [(x, y) for x in a for y in b]
```

### Example:

```python
cartesian_product(items, sizes)
```

```python
def visualize_matrix(matrix):
```
